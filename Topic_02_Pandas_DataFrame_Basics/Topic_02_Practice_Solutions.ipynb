## ✅ Solution — Practice 38, 39 & 40 — Cumulative, `shift`, rolling

````python
sales = pd.DataFrame({"Date": pd.date_range("2026-01-01", periods=7, freq="D"),
                      "Sales": [100, 150, 80, 200, 170, 90, 220]})
sales["CumulativeSales"] = sales["Sales"].cumsum()
sales["RunningMax"] = sales["Sales"].cummax()
sales["PrevDay"] = sales["Sales"].shift(1)
sales["Change"] = sales["Sales"] - sales["PrevDay"]            # same as .diff()
sales["PctChange"] = (sales["Sales"].pct_change() * 100).round(1)
sales["MA3"] = sales["Sales"].rolling(3).mean().round(1)       # first 2 rows are NaN (not enough history)
sales
````

## Capstone — full solution

````python
cap = pd.DataFrame({
    "EmployeeID": range(1, 11),
    "Name": [" Ali ", "Ahmed", "Sara", "Hamza", "Ayesha", "Bilal", "Zara", "Usman", "Hina", "Omar"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "IT", "HR", "Finance"],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Lahore", "Islamabad", "Lahore", "Karachi", "Islamabad"],
    "Age": [25, 32, 28, 41, 24, 35, 29, 38, 27, 34],
    "Salary": [80000, 65000, 95000, 120000, 70000, 110000, 90000, 105000, 72000, 115000],
    "Experience": [2, 6, 4, 12, 2, 8, 5, 10, 3, 9],
    "JoinDate": ["2024-01-10", "2020-05-12", "2022-03-15", "2014-07-01", "2024-02-20",
                 "2018-09-10", "2021-01-05", "2016-11-20", "2023-06-10", "2017-04-18"],
})

# Part A — inspect
print(cap.shape, cap.columns.tolist(), sep="\n")
print(cap.dtypes)
cap.info()
print(cap.describe())

# Part B — clean
cap["Name"] = cap["Name"].str.strip()
cap["JoinDate"] = pd.to_datetime(cap["JoinDate"])
print("missing:", cap.isna().sum().sum(), "| duplicates:", cap.duplicated().sum())

# Part C — new columns
cap["AnnualSalary"] = cap["Salary"] * 12
cap["Bonus"] = cap["Salary"] * 0.10
cap["ExperienceLevel"] = np.select([cap["Experience"] < 3, cap["Experience"] <= 7],
                                   ["Junior", "Mid"], default="Senior")
cap["JoinYear"] = cap["JoinDate"].dt.year
cap.head()
````

````python
# Part D — filter
print(cap[cap["Department"] == "IT"]["Name"].tolist())
print(cap[cap["Salary"] > 100000]["Name"].tolist())
print(cap[(cap["City"] == "Lahore") & (cap["Experience"] > 5)]["Name"].tolist())
print(cap[(cap["ExperienceLevel"] == "Senior") & (cap["Salary"] > 100000)]["Name"].tolist(), "\n")

# Part E — analysis
print(cap.groupby("Department")["Salary"].agg(["mean", "sum"]), "\n")
print(cap["City"].value_counts(), "\n")
print("Highest paid:", cap.loc[cap["Salary"].idxmax(), "Name"])
cap["SalaryRank"] = cap["Salary"].rank(ascending=False).astype(int)

# Part F — reshaping
print(cap.pivot_table(values="Salary", index="Department", columns="City", aggfunc="mean"), "\n")
print(cap.pivot_table(values="EmployeeID", index="Department", columns="City", aggfunc="count", fill_value=0), "\n")

# Part G — final report
report = cap.sort_values("Salary", ascending=False)
print(report[["Name", "Department", "Salary", "SalaryRank"]].head(3))
report.to_csv(os.path.join(OUT, "employees_clean.csv"), index=False)
print("saved to", os.path.join(OUT, "employees_clean.csv"))
````
