### Problem 2 — Variance and standard deviation

Data: `[10, 12, 14, 16, 18]`.
1. Calculate the population and sample variance and std **by hand** (show the squared deviations).
2. Verify with `np.var`, `np.std` and Pandas. Explain why NumPy and Pandas give different defaults.

```python
# ✍️ Problem 2 — your attempt
```

### Solution 2
Mean = 70 / 5 = 14. Deviations: −4, −2, 0, 2, 4 → squares: 16, 4, 0, 4, 16 → sum = **40**
- Population variance = 40 / 5 = **8** → std = √8 = **2.828**
- Sample variance = 40 / 4 = **10** → std = √10 = **3.162**

```python
d = np.array([10, 12, 14, 16, 18])
print('np.var (ddof=0):', np.var(d), '| np.std:', round(np.std(d), 3))
print('np.var (ddof=1):', np.var(d, ddof=1), '| np.std(ddof=1):', round(np.std(d, ddof=1), 3))
print('pandas var/std :', pd.Series(d).var(), '/', round(pd.Series(d).std(), 3))
```

**Explanation:** NumPy defaults to the **population** formula (÷ n, `ddof=0`); Pandas defaults to the **sample** formula (÷ n−1, `ddof=1`), because in data analysis we usually have a sample and want an unbiased estimate of the population variance.

**Explanation:** random and stratified samples match the population well (stratified matches the city mix almost exactly). The convenience sample contains only top earners → its mean salary is far too high and Quetta is almost absent. This is **selection bias**.

### Problem 8 — Central Limit Theorem

Population: `rng_p.uniform(0, 100, 100_000)` (flat, not bell-shaped), `rng_p = np.random.default_rng(8)`.
1. For n = 2, 10 and 40, draw 3,000 samples and plot the sample means.
2. Compare the std of the sample means with σ/√n.

```python
# ✍️ Problem 8 — your attempt
```

### Solution 8

```python
rng_p = np.random.default_rng(8)
pop8 = rng_p.uniform(0, 100, 100_000)
sigma8 = pop8.std()

fig, axes = plt.subplots(1, 4, figsize=(19, 3.6))
sns.histplot(pop8, bins=40, ax=axes[0]); axes[0].set_title('Population (uniform)')
for ax, n in zip(axes[1:], [2, 10, 40]):
    means = rng_p.choice(pop8, size=(3000, n)).mean(axis=1)
    sns.histplot(means, bins=40, ax=ax)
    ax.set_title(f'n = {n}: std of means {means.std():.2f} vs σ/√n {sigma8 / np.sqrt(n):.2f}')
plt.tight_layout(); plt.show()
```

**Explanation:** even with a flat population, the sample means become bell-shaped as n grows (n = 2 gives a triangle, n = 10 is already close to normal), and their spread matches σ/√n.

### Problem 9 — Confidence interval

A sample of 25 phone batteries has mean life 18.4 hours and sample std 2.5 hours.
1. Compute the 90%, 95% and 99% CIs using the t-distribution.
2. Write a **correct** one-sentence interpretation of the 95% CI.
3. How would the 95% CI width change with n = 100 (same mean and std)?

```python
# ✍️ Problem 9 — your attempt
```

### Solution 9

```python
n9, mean9, s9 = 25, 18.4, 2.5
se9 = s9 / np.sqrt(n9)
for level in [0.90, 0.95, 0.99]:
    lo, hi = stats.t.interval(level, df=n9 - 1, loc=mean9, scale=se9)
    print(f'{int(level * 100)}% CI: ({lo:.2f}, {hi:.2f})  width = {hi - lo:.2f}')

lo100, hi100 = stats.t.interval(0.95, df=99, loc=mean9, scale=s9 / np.sqrt(100))
print(f'\n95% CI with n = 100: ({lo100:.2f}, {hi100:.2f})  width = {hi100 - lo100:.2f}')
```

**Explanation:**
- Correct interpretation: *"We are 95% confident that the true mean battery life is between about 17.4 and 19.4 hours — meaning this method of building intervals captures the true mean in 95% of repeated samples."*
- With 4× more data the width roughly **halves** (SE ∝ 1/√n), and slightly more because the t critical value also shrinks.

### Problem 10 — p-values

A website claims 10% of visitors buy something. In a sample of 400 visitors, 52 bought.
1. Test H0: p = 0.10 vs H1: p ≠ 0.10 with `stats.binomtest`.
2. Interpret the p-value correctly.
3. List two things this p-value does **not** tell you.

```python
# ✍️ Problem 10 — your attempt
```

### Solution 10

```python
res10 = stats.binomtest(52, n=400, p=0.10)
print(f'Observed rate = {52 / 400:.3f}, p-value = {res10.pvalue:.4f}')
ci10 = res10.proportion_ci(confidence_level=0.95)
print(f'95% CI for the true purchase rate: ({ci10.low:.3f}, {ci10.high:.3f})')
```

**Explanation:** the p-value is the probability of seeing a result at least as far from 10% as 13% (52/400) **if the true rate really were 10%**. Here it is around 0.05 — borderline evidence.

It does **not** tell us: (1) the probability that the true rate is 10%; (2) how large or important the difference is (look at the CI); (3) whether the sample was collected without bias.

### Problem 11 — Hypothesis test (one-sample)

A café claims its coffee has 150 mg caffeine per cup. Lab data: `caffeine = rng_p.normal(156, 10, 30)` with `rng_p = np.random.default_rng(11)`.
State H0/H1, run the right test at α = 0.05, report the 95% CI, and write a plain-English conclusion.

```python
# ✍️ Problem 11 — your attempt
```

### Solution 11

```python
rng_p = np.random.default_rng(11)
caffeine = rng_p.normal(156, 10, 30)
res11 = stats.ttest_1samp(caffeine, 150)
ci11 = res11.confidence_interval(0.95)
print('H0: μ = 150 mg | H1: μ ≠ 150 mg')
print(f'Mean = {caffeine.mean():.1f}, t = {res11.statistic:.2f}, p = {res11.pvalue:.4f}')
print(f'95% CI: ({ci11.low:.1f}, {ci11.high:.1f})')
```

**Explanation:** a one-sample t-test is right (one numerical sample vs a claimed value; n = 30). If p < 0.05 and 150 lies outside the CI, we conclude the average caffeine content is **higher** than the café claims — by roughly the amount shown in the CI.

### Problem 12 — Type I / Type II errors

A fraud-detection system flags transactions. H0: the transaction is legitimate.
1. Describe a Type I and a Type II error in this context. Which is more costly for a bank? For a customer?
2. By simulation, estimate the power of a two-sample t-test (α = 0.05) to detect a difference of 5 points (sd 15) with n = 30 and n = 150 per group.

```python
# ✍️ Problem 12 — your attempt
```

### Solution 12

```python
rng_p = np.random.default_rng(12)
def power(n, effect=5, sd=15, sims=3000):
    return np.mean([stats.ttest_ind(rng_p.normal(0, sd, n), rng_p.normal(effect, sd, n)).pvalue < 0.05
                    for _ in range(sims)])
print('Power with n = 30 per group :', power(30))
print('Power with n = 150 per group:', power(150))
```

**Explanation:**
- **Type I (false positive):** a legitimate transaction is flagged as fraud → annoyed customer, blocked card. Costly for the **customer** experience.
- **Type II (false negative):** a fraudulent transaction is **not** flagged → direct money loss. Usually more costly for the **bank**.
- Power: with 30 per group a 5-point difference is detected only about a quarter of the time; with 150 per group, most of the time. Small studies miss real effects.

### Problem 13 — t-tests

(a) Independent: `team_a = rng_p.normal(72, 10, 40)`, `team_b = rng_p.normal(77, 12, 45)` — do the teams' mean scores differ? Use Welch's test and compute Cohen's d.
(b) Paired: `before = rng_p.normal(80, 8, 20)`, `after = before - rng_p.normal(2, 3, 20)` (blood pressure after medication) — did it change? Compare a paired and an (incorrect) independent test.
Use `rng_p = np.random.default_rng(13)`.

```python
# ✍️ Problem 13 — your attempt
```

### Solution 13

```python
rng_p = np.random.default_rng(13)
team_a = rng_p.normal(72, 10, 40)
team_b = rng_p.normal(77, 12, 45)
w13 = stats.ttest_ind(team_a, team_b, equal_var=False)
d13 = (team_b.mean() - team_a.mean()) / np.sqrt((team_a.var(ddof=1) + team_b.var(ddof=1)) / 2)
print(f"(a) Welch t = {w13.statistic:.2f}, p = {w13.pvalue:.4f}, Cohen's d = {d13:.2f}")

before = rng_p.normal(80, 8, 20)
after = before - rng_p.normal(2, 3, 20)
print(f'(b) Paired t-test      p = {stats.ttest_rel(before, after).pvalue:.5f}')
print(f'    Independent (wrong) p = {stats.ttest_ind(before, after).pvalue:.5f}')
```

**Explanation:**
- (a) Welch's test handles the different group sizes and variances. Here the observed effect is **small** (Cohen's d ≈ 0.2) and p > 0.05, so we **fail to reject H0**. The data was generated with a true 5-point difference, but ~40 people per group gives **low power** for an effect this size. "Not significant" ≠ "no difference" (Part 16/17).
- (b) The paired test detects the ~2-point drop clearly because it compares each person with **themselves**; the independent test is swamped by the large differences **between** people and misses it.

### Problem 14 — Chi-square

A survey of 300 people:

| | Tea | Coffee | Juice |
|---|---|---|---|
| Age < 30 | 40 | 70 | 40 |
| Age ≥ 30 | 75 | 50 | 25 |

1. Test whether drink preference is independent of age group.
2. Show the expected counts, compute Cramér's V and interpret.

```python
# ✍️ Problem 14 — your attempt
```

### Solution 14

```python
survey = pd.DataFrame({'Tea': [40, 75], 'Coffee': [70, 50], 'Juice': [40, 25]}, index=['Age < 30', 'Age ≥ 30'])
chi2_14, p14, dof14, exp14 = stats.chi2_contingency(survey)
print('Expected counts:\n', pd.DataFrame(exp14, index=survey.index, columns=survey.columns).round(1))
print(f'\nχ² = {chi2_14:.2f}, df = {dof14}, p = {p14:.6f}')
v14 = np.sqrt(chi2_14 / (survey.to_numpy().sum() * (min(survey.shape) - 1)))
print(f"Cramér's V = {v14:.3f}")
```

**Explanation:** the very small p-value means drink preference is **not independent** of age: younger people choose coffee and juice more, older people choose tea more. Cramér's V ≈ 0.25 → a weak-to-moderate association.

# Part 25 — Final Project: Employee Salary Statistical Analysis

**Questions we want to answer:**
1. What is a typical salary, and how spread out are salaries?
2. What does the salary distribution look like? Is it skewed? Are there outliers?
3. How are salary, experience and age related?
4. What is a plausible range for the company's true average salary?
5. Do men and women earn different average salaries?
6. Do departments differ in salary?
7. Is department related to education level?

## Step 0 — The dataset (80 employees)

```python
rproj = np.random.default_rng(2025)
n = 80
department = rproj.choice(['IT', 'Finance', 'HR', 'Sales'], n, p=[0.35, 0.25, 0.15, 0.25])
edu_probs = {'IT': [0.3, 0.55, 0.15], 'Finance': [0.35, 0.55, 0.10], 'HR': [0.7, 0.3, 0.0], 'Sales': [0.75, 0.25, 0.0]}
education = np.array([rproj.choice(['Bachelor', 'Master', 'PhD'], p=edu_probs[d]) for d in department])
experience = rproj.gamma(2.2, 3.5, n).clip(0, 35).round(1)
age = (23 + experience + rproj.normal(2, 2.5, n)).round(0)
gender = rproj.choice(['Male', 'Female'], n, p=[0.6, 0.4])

dept_base = pd.Series(department).map({'IT': 110000, 'Finance': 100000, 'HR': 80000, 'Sales': 85000}).to_numpy()
edu_add = pd.Series(education).map({'Bachelor': 0, 'Master': 12000, 'PhD': 30000}).to_numpy()
salary = (dept_base + edu_add + 5000 * experience) * rproj.lognormal(0, 0.08, n)
salary[rproj.choice(n, 2, replace=False)] *= 2.2          # two senior executives → high outliers

employees = pd.DataFrame({
    'Employee_ID': [f'E{101 + i}' for i in range(n)],
    'Age': age, 'Gender': gender, 'Department': department, 'Education': education,
    'Experience': experience, 'Salary': salary.round(-2),
})
employees.head(10)
```

## Step 1 — Central tendency

```python
sal = employees['Salary']
central = pd.Series({'Mean': sal.mean(), 'Median': sal.median(), 'Mode (rounded to 10k)': (sal.round(-4)).mode()[0]})
print(central.round(0))
```

## Step 2 — Dispersion

```python
q1, q3 = np.percentile(sal, [25, 75])
dispersion = pd.Series({
    'Range': sal.max() - sal.min(),
    'Variance (sample)': sal.var(),
    'Std (sample)': sal.std(),
    'Coefficient of variation': sal.std() / sal.mean(),       # std relative to the mean
    'Q1': q1, 'Q3': q3, 'IQR': q3 - q1,
    'MAD (median abs. deviation)': stats.median_abs_deviation(sal),
})
dispersion.map('{:,.2f}'.format)      # readable formatting instead of scientific notation
```

## Step 3 — Distribution, skewness and outliers

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 4))
sns.histplot(sal, bins=20, kde=True, ax=axes[0])
axes[0].axvline(sal.mean(), color='red', label='Mean'); axes[0].axvline(sal.median(), color='green', ls='--', label='Median')
axes[0].legend(); axes[0].set_title(f'Salary distribution (skew {sal.skew():.2f}, kurtosis {sal.kurt():.2f})')
sns.boxplot(x=sal, ax=axes[1]); axes[1].set_title('Boxplot')
stats.probplot(sal, dist='norm', plot=axes[2]); axes[2].set_title('Q-Q plot vs normal')
plt.tight_layout(); plt.show()

low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
outliers = employees[(sal < low) | (sal > high)]
print(f'IQR fences: ({low:,.0f}, {high:,.0f})')
print(f'Shapiro-Wilk p = {stats.shapiro(sal).pvalue:.5f}')
outliers[['Employee_ID', 'Department', 'Education', 'Experience', 'Salary']]
```

## Step 4 — Correlation

```python
num = employees[['Age', 'Experience', 'Salary']]
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
sns.heatmap(num.corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, ax=axes[0])
axes[0].set_title('Pearson correlation')
sns.scatterplot(data=employees, x='Experience', y='Salary', hue='Department', ax=axes[1])
axes[1].set_title('Experience vs salary'); plt.tight_layout(); plt.show()

print('Pearson :\n', num.corr().round(3), '\n')
print('Spearman:\n', num.corr(method='spearman').round(3))
r, p = stats.pearsonr(employees['Experience'], employees['Salary'])
print(f'\nExperience–Salary: r = {r:.3f}, p = {p:.2g}')
```

## Step 5 — Confidence intervals

```python
def mean_ci(x, level=0.95):
    return stats.t.interval(level, df=len(x) - 1, loc=np.mean(x), scale=stats.sem(x))

lo, hi = mean_ci(sal)
print(f'Average salary: {sal.mean():,.0f}  (95% CI {lo:,.0f} – {hi:,.0f})')

# Bootstrap CI for the MEDIAN (no formula needed: resample with replacement many times)
boot_medians = [np.median(rproj.choice(sal, len(sal), replace=True)) for _ in range(5000)]
b_lo, b_hi = np.percentile(boot_medians, [2.5, 97.5])
print(f'Median salary : {sal.median():,.0f}  (95% bootstrap CI {b_lo:,.0f} – {b_hi:,.0f})')

dept_ci = employees.groupby('Department')['Salary'].apply(
    lambda x: pd.Series({'mean': x.mean(), 'ci_low': mean_ci(x)[0], 'ci_high': mean_ci(x)[1], 'n': len(x)})).unstack()
dept_ci.round(0)
```

(**Bootstrap** = resample your own data with replacement thousands of times to see how much a statistic varies — useful for medians and other statistics without a simple formula.)

## Step 6 — Hypothesis test: do men and women earn different averages?
- H0: μ_male = μ_female
- H1: μ_male ≠ μ_female

```python
male = employees.loc[employees['Gender'] == 'Male', 'Salary']
female = employees.loc[employees['Gender'] == 'Female', 'Salary']

welch = stats.ttest_ind(male, female, equal_var=False)
mw = stats.mannwhitneyu(male, female)
print(f'Mean male = {male.mean():,.0f} (n={len(male)}), mean female = {female.mean():,.0f} (n={len(female)})')
print(f'Welch t-test : t = {welch.statistic:.2f}, p = {welch.pvalue:.4f}')
print(f'Mann-Whitney : p = {mw.pvalue:.4f}  (robust check, because salary is skewed)')
print('95% CI for the difference (male − female):', np.round(welch.confidence_interval(0.95), 0))
```

## Step 7 — Department differences: ANOVA + post-hoc

```python
groups = {d: g['Salary'].to_numpy() for d, g in employees.groupby('Department')}
print('Levene (equal variances) p =', round(stats.levene(*groups.values()).pvalue, 4))

f, p = stats.f_oneway(*groups.values())
k = stats.kruskal(*groups.values())
print(f'One-way ANOVA : F = {f:.2f}, p = {p:.2g}')
print(f'Kruskal-Wallis: H = {k.statistic:.2f}, p = {k.pvalue:.2g}  (non-parametric check)')

names = list(groups)
tk = stats.tukey_hsd(*groups.values())
pairs = [{'Pair': f'{names[i]} vs {names[j]}', 'Mean diff': groups[names[i]].mean() - groups[names[j]].mean(),
          'Adjusted p': tk.pvalue[i, j]} for i in range(len(names)) for j in range(i + 1, len(names))]
print()
print(pd.DataFrame(pairs).round(4).to_string(index=False))

plt.figure(figsize=(8, 4))
sns.boxplot(data=employees, x='Department', y='Salary', order=['IT', 'Finance', 'Sales', 'HR'])
plt.title(f'Salary by department (ANOVA p = {p:.2g})'); plt.show()
```

## Step 8 — Categorical relationship: department vs education (chi-square)

```python
ct = pd.crosstab(employees['Department'], employees['Education'])
chi2, p_chi, dof, expected = stats.chi2_contingency(ct)
v = np.sqrt(chi2 / (ct.to_numpy().sum() * (min(ct.shape) - 1)))
print(ct, '\n')
print(f'χ² = {chi2:.2f}, df = {dof}, p = {p_chi:.4f}, Cramér\'s V = {v:.2f}')
print('Cells with expected count < 5:', int((expected < 5).sum()), 'of', expected.size)
```

## Step 9 — Final statistical interpretation (in simple English)

1. **Typical salary.** The mean (≈ 146k) is higher than the median (≈ 142k) and the skewness is clearly positive (≈ 1.7) → salaries are **right-skewed**. The **median** is the better "typical salary". Three very high earners (≈ 240k–300k) are flagged as **outliers** by the 1.5×IQR rule; two of them are senior executives. They are real people, so we keep them — but we report the median and IQR alongside the mean and std.

2. **Spread.** The std is ≈ 37k (coefficient of variation ≈ 0.26 → salaries typically differ from the mean by about a quarter of it). The middle 50% of employees earn between ≈ 122k and 162k (IQR ≈ 40k).

3. **Distribution.** Positive skew, high kurtosis (heavy right tail), a Q-Q plot that curves away at the top, and a Shapiro-Wilk p < 0.001 → salaries are **not normal**. With n = 80 the CLT still makes tests on **means** reasonable, but we back every test up with a **non-parametric** version.

4. **Relationships.** Experience is **moderately positively correlated** with salary (Pearson r ≈ 0.45, p < 0.001; Spearman ≈ 0.62 — higher because Spearman is less affected by the executive outliers). Age is also correlated with salary, but mostly because age and experience are strongly related (r ≈ 0.87). Correlation is not causation, and department and education also drive salary.

5. **Average salary.** We are 95% confident that the company's true average salary is between ≈ 137k and 154k. The bootstrap 95% CI for the **median** is ≈ 135k – 150k. (These CIs describe how reliable the *method* is; see Part 14.)

6. **Gender.** Women's average salary in this sample is slightly higher, but the Welch t-test (p ≈ 0.27) and Mann-Whitney test (p ≈ 0.82) show **no evidence** of a difference. The 95% CI for the difference (≈ −26k to +8k) is **wide** and includes 0 → a real difference cannot be ruled out, and it could go either way. A serious pay-equity study would need more data and should control for department, education and experience.

7. **Departments.** ANOVA (F ≈ 10.5, p < 0.001) and Kruskal-Wallis (p < 0.001) agree: average salaries are **not equal** across departments. Tukey's HSD shows **IT pays significantly more than HR and Sales**, and **Finance pays more than Sales**. The other pairs are not significant — partly because HR has only 6 employees, so its mean is very uncertain (its CI is very wide). Levene's test (p ≈ 0.96) supports the equal-variance assumption.

8. **Department vs education.** The chi-square test (p ≈ 0.001, Cramér's V ≈ 0.37) suggests a **moderate association**: IT and Finance employ more Master's/PhD holders, while Sales and HR are mostly Bachelor's. **But** 6 of 12 cells have expected counts below 5, so this result is only **approximate** — merging Master + PhD into one "Postgraduate" group would make it more reliable.

### Overall conclusion
Salaries in this company depend mainly on **department**, **education** and **experience**. They are right-skewed, with a few very high earners, so **medians, IQRs and robust tests** matter. The data gives **no evidence of a gender pay difference**, but the sample is too small to rule one out.
