# Practice Problem 16 — Leakage Detection

Suppose you have 10,000 rows.

You calculate target encoding using all 10,000 rows and only afterward split the data into train and test.

Answer:

1. Is this safe?
2. Why or why not?
3. What should you do instead?

````python
# ✍️ Your answer here
````

## Solution 16 — Leakage detection

1. **No, it is not safe.**
2. The encoding for each category was computed using the targets of the *test* rows too. The training features
   now contain information about the test answers, so the test score is optimistic — the model has "seen" part of the test.
   The effect is largest for rare categories (a category with 1 row gets exactly its own target!).
3. **Split first.** Compute the category means on the training set only, map them onto both train and test,
   fill unseen categories with the global training mean, and ideally use out-of-fold (cross-validated) means for the
   training rows themselves (this is what `sklearn.preprocessing.TargetEncoder` does).

# Practice Problem 21

Explain in your own words:

1. What is hash encoding?
2. Why is it useful for very high-cardinality data?
3. What is a hash collision?

````python
# ✍️ Your answer here
````

## Solution 19 — Unknown categories

The encoder only knows the categories it saw during `fit` (Lahore, Karachi, Islamabad). When `Multan` appears at
prediction time there is no column for it. With the default `handle_unknown='error'`, `transform` raises an error and the
production system crashes.

`handle_unknown='ignore'` makes the encoder output **all zeros** for that row's City columns, so the row is still
processed — the model simply gets "none of the known cities".

## Solution 20 — Binary encoding arithmetic

1. One-hot: **16 columns** (15 with `drop_first`).
2. Binary: 2⁴ = 16, so **4 bits** (in practice libraries often need 5 because code 0 is reserved — still tiny).
3. Binary encoding needs about log₂(k) columns instead of k, so dimensionality grows very slowly as categories increase.

## Solution 21 — Hash encoding

1. A hash function turns each category string into a number, and that number (modulo *n*) chooses one of *n* fixed output columns.
2. The number of columns is fixed in advance, no matter how many categories exist (even ones never seen before), and no
   category dictionary has to be stored.
3. A **collision** is when two different categories land in the same column, so the model cannot tell them apart.
   More columns → fewer collisions.

## Solution 30 — Train/test encoding with a full pipeline

````python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

rng = np.random.default_rng(0)
n = 60
d30 = pd.DataFrame({
    'Age': rng.integers(20, 60, n),
    'Income': rng.normal(60, 20, n).round(1),
    'Gender': rng.choice(['Male', 'Female'], n),
    'City': rng.choice(['Lahore', 'Karachi', 'Islamabad', 'Multan'], n),
})
# target depends on Income and City so the model has something to learn
score = 0.05 * (d30['Income'] - 60) + (d30['City'] == 'Lahore') * 1.0
d30['Purchased'] = (score + rng.normal(0, 0.5, n) > 0).astype(int)

X = d30.drop(columns='Purchased')
y = d30['Purchased']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

num_cols = X.select_dtypes(include='number').columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'string', 'category']).columns.tolist()
print('numeric:', num_cols, '| categorical:', cat_cols)

pipe30 = Pipeline([
    ('pre', ColumnTransformer([
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ])),
    ('model', LogisticRegression())
])
pipe30.fit(X_train, y_train)                 # everything is learned from TRAIN only
pred = pipe30.predict(X_test)
print('Test accuracy:', round(accuracy_score(y_test, pred), 3))
print(classification_report(y_test, pred, zero_division=0))
````

## Solution 33 — Final Project: Customer Purchase Prediction

````python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.metrics import confusion_matrix

rng = np.random.default_rng(7)
n = 200
proj = pd.DataFrame({
    'Age': rng.integers(18, 65, n),
    'Income': rng.normal(70, 25, n).clip(15).round(1),
    'Gender': rng.choice(['Male', 'Female'], n),
    'Education': rng.choice(['School', 'Bachelor', 'Master', 'PhD'], n, p=[0.2, 0.4, 0.3, 0.1]),
    'City': rng.choice(['Lahore', 'Karachi', 'Islamabad', 'Multan'], n),
    'Browser': rng.choice(['Chrome', 'Safari', 'Firefox'], n, p=[0.6, 0.25, 0.15]),
    'Device': rng.choice(['Mobile', 'Desktop', 'Tablet'], n, p=[0.55, 0.35, 0.10]),
})
edu_rank = proj['Education'].map({'School': 0, 'Bachelor': 1, 'Master': 2, 'PhD': 3})
logit = 0.04 * (proj['Income'] - 70) + 0.5 * edu_rank + 0.8 * (proj['Device'] == 'Desktop') - 0.8
proj['Purchased'] = (rng.random(n) < 1 / (1 + np.exp(-logit))).astype(int)

# Part 1 — inspection
print(proj.head(), '\n')
print('shape:', proj.shape)
print(proj.dtypes, '\n')
X = proj.drop(columns='Purchased')
y = proj['Purchased']
cat_cols = X.select_dtypes(include=['object', 'string', 'category']).columns.tolist()
num_cols = X.select_dtypes(include='number').columns.tolist()
print('categorical:', cat_cols)
print('numerical  :', num_cols)
````

````python
# Part 5 — split BEFORE fitting anything
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# Parts 2-4 + 6 — one ColumnTransformer does binary, ordinal and one-hot encoding
pre33 = ColumnTransformer([
    ('num', StandardScaler(), ['Age', 'Income']),
    ('gender', OneHotEncoder(drop='if_binary'), ['Gender']),                                       # binary 0/1
    ('edu', OrdinalEncoder(categories=[['School', 'Bachelor', 'Master', 'PhD']]), ['Education']),  # ordered
    ('nominal', OneHotEncoder(handle_unknown='ignore'), ['City', 'Browser', 'Device'])              # one-hot
])

# Parts 7-8 — pipeline + model
model33 = Pipeline([('pre', pre33), ('model', LogisticRegression(max_iter=1000))])
model33.fit(X_train, y_train)

# Part 9 — evaluation
pred = model33.predict(X_test)
print('Accuracy:', round(accuracy_score(y_test, pred), 3))
print('Confusion matrix:\n', confusion_matrix(y_test, pred))
print(classification_report(y_test, pred, zero_division=0))
print('Features after encoding:', len(model33.named_steps['pre'].get_feature_names_out()))
````

**Part 10 — explanation**

- **Gender** has two values, so one 0/1 column is enough (`drop='if_binary'`).
- **Education** has a real order, so `OrdinalEncoder` with an explicit category order keeps School < Bachelor < Master < PhD.
- **City** has no order — label encoding (0,1,2,3) would invent one, so it is one-hot encoded.
- **Leakage:** the scaler and encoders are fitted inside the pipeline on the training split only; the test set is only transformed.
- **Unknown categories:** `handle_unknown='ignore'` means a new city/browser/device in production produces zeros instead of an error.
