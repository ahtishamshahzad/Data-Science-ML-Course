# 1. What Is Encoding?

Suppose we have:

| Name | Gender | City |
|---|---|---|
| Ali | Male | Lahore |
| Sara | Female | Karachi |
| Ahmed | Male | Islamabad |

The values `Male`, `Female`, `Lahore`, `Karachi`, and `Islamabad` are categorical values.

A machine-learning model usually needs numerical input.

Encoding converts categorical information into numerical representation.

Conceptually:

```text
Categorical data
       ↓
    Encoding
       ↓
Numerical representation
       ↓
Machine-learning model
```

Encoding does NOT mean that the original category itself becomes mathematically ordered. The correct encoding technique depends on the type of categorical variable.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'John'],
    'Gender': ['Male', 'Female', 'Male', 'Male'],
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

df
````

<a id="sec-2"></a>
# 2. What Is Categorical Data?

A categorical column contains values that represent groups or categories.

Examples:

```text
Gender: Male, Female
City: Lahore, Karachi, Islamabad
Color: Red, Blue, Green
Browser: Chrome, Safari, Firefox
```

There are two important types:

## Nominal categorical data

There is no meaningful order.

Example:

```text
Lahore
Karachi
Islamabad
```

There is no mathematical meaning to saying one city is greater than another.

## Ordinal categorical data

There is a meaningful order.

Example:

```text
School < Bachelor < Master < PhD
```

Another example:

```text
Low < Medium < High
```

This distinction is one of the most important ideas when choosing an encoding method.

<a id="sec-3"></a>
# 3. Binary Categories

A binary categorical variable has only two possible categories.

Examples:

```text
Male / Female
Yes / No
True / False
Pass / Fail
```

A simple representation is:

```text
No  → 0
Yes → 1
```

For binary variables, mapping to 0 and 1 is often straightforward.

````python
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male']
})

df['Gender_encoded'] = df['Gender'].map({
    'Female': 0,
    'Male': 1
})

df
````

# 4. Encoding with `map()`

`map()` is very useful when you explicitly know the mapping you want.

Example:

```python
df['Gender'].map({'Female': 0, 'Male': 1})
```

Advantages:

- Easy to understand
- Explicit mapping
- Good for binary categories
- Good for ordinal categories when you define the correct order

Be careful: if a value does not exist in the mapping dictionary, `map()` can produce `NaN`.

````python
df = pd.DataFrame({
    'Status': ['Active', 'Inactive', 'Active', 'Active']
})

df['Status_encoded'] = df['Status'].map({
    'Inactive': 0,
    'Active': 1
})

df
````

# 5. Encoding with `replace()`

`replace()` can also replace category names with numerical values.

Example:

```python
df['Gender'] = df['Gender'].replace({
    'Male': 1,
    'Female': 0
})
```

Difference in idea:

- `map()` is commonly used to create a transformed Series from a mapping.
- `replace()` is a more general replacement operation and can work across DataFrame values.

````python
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female']
})

df['Gender_encoded'] = df['Gender'].replace({
    'Male': 1,
    'Female': 0
})

df
````

# 6. Pandas `category` Data Type

Pandas provides a special categorical data type.

You can convert a column using:

```python
df['City'] = df['City'].astype('category')
```

This is useful when a column contains repeated categorical values.

Categorical dtype can also provide a structured list of categories and can save memory in appropriate datasets.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Lahore', 'Islamabad', 'Karachi']
})

df['City'] = df['City'].astype('category')

print(df)
print(df['City'].cat.categories)
````

<a id="sec-7"></a>
# 7. `cat.codes`

After converting a Series to categorical dtype, Pandas provides category codes.

Example:

```python
df['City'].astype('category').cat.codes
```

Important warning:

`cat.codes` assigns integer codes to categories, but those codes should not automatically be interpreted as meaningful numerical order.

For example:

```text
Lahore     → 1
Karachi    → 0
Islamabad  → 2
```

does NOT mean:

```text
Islamabad > Lahore > Karachi
```

For nominal categories, one-hot encoding is often safer.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

df['City_category'] = df['City'].astype('category')
df['City_code'] = df['City_category'].cat.codes

df
````

<a id="sec-8"></a>
# 8. Ordinal Encoding

Ordinal encoding is appropriate when categories have a meaningful order.

Example:

```text
School
Bachelor
Master
PhD
```

We can encode:

```text
School   → 0
Bachelor → 1
Master   → 2
PhD      → 3
```

The numerical order represents the real-world order.

````python
df = pd.DataFrame({
    'Education': [
        'Bachelor',
        'Master',
        'School',
        'PhD',
        'Master'
    ]
})

education_map = {
    'School': 0,
    'Bachelor': 1,
    'Master': 2,
    'PhD': 3
}

df['Education_encoded'] = df['Education'].map(education_map)

df
````

# 9. One-Hot Encoding

One-Hot Encoding creates a separate binary column for each category.

Suppose:

```text
City
Lahore
Karachi
Islamabad
```

One-hot representation:

```text
City        Lahore  Karachi  Islamabad
Lahore         1       0         0
Karachi        0       1         0
Islamabad      0       0         1
```

This avoids creating an artificial numerical order between nominal categories.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

encoded = pd.get_dummies(
    df,
    columns=['City'],
    dtype=int
)

encoded
````

# 10. `drop_first=True`

Pandas allows you to remove one dummy column:

```python
pd.get_dummies(
    df,
    columns=['City'],
    drop_first=True,
    dtype=int
)
```

If there are three categories, two dummy columns can be enough for many linear-model setups because the omitted category can be represented by all zeros.

However, `drop_first=True` should not be treated as a universal rule. The appropriate representation depends on the model and preprocessing design.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

encoded = pd.get_dummies(
    df,
    columns=['City'],
    drop_first=True,
    dtype=int
)

encoded
````

# 13. Frequency Encoding

Frequency encoding replaces each category with how frequently it appears.

Example:

```text
Lahore      appears 3 times
Karachi     appears 2 times
Islamabad   appears 1 time
```

Frequency encoding:

```text
Lahore     → 3
Karachi    → 2
Islamabad  → 1
```

This can be useful for high-cardinality categorical variables.

````python
df = pd.DataFrame({
    'City': [
        'Lahore',
        'Lahore',
        'Lahore',
        'Karachi',
        'Karachi',
        'Islamabad'
    ]
})

frequency = df['City'].value_counts()

df['City_frequency'] = df['City'].map(frequency)

df
````

# 15. Count Encoding

Count encoding is closely related to frequency encoding.

Instead of representing the category itself, we represent how many times that category appears.

Example:

```text
Chrome → 500
Safari → 300
Firefox → 200
```

The key idea is:

```python
counts = df['Browser'].value_counts()
df['Browser_count'] = df['Browser'].map(counts)
```

````python
df = pd.DataFrame({
    'Browser': [
        'Chrome', 'Chrome', 'Safari',
        'Firefox', 'Chrome', 'Safari'
    ]
})

counts = df['Browser'].value_counts()

df['Browser_count'] = df['Browser'].map(counts)

df
````

# 16. Target Encoding

Target encoding replaces each category with a statistic calculated from the target variable, commonly the mean target value for that category.

Example:

| City | Purchased |
|---|---:|
| Lahore | 1 |
| Lahore | 1 |
| Lahore | 0 |
| Karachi | 0 |
| Karachi | 0 |
| Islamabad | 1 |

Mean purchase rate:

```text
Lahore     → (1 + 1 + 0) / 3 = 0.67
Karachi    → (0 + 0) / 2 = 0.00
Islamabad  → 1 / 1 = 1.00
```

The category is therefore represented using information about the target.

````python
df = pd.DataFrame({
    'City': [
        'Lahore', 'Lahore', 'Lahore',
        'Karachi', 'Karachi', 'Islamabad'
    ],
    'Purchased': [1, 1, 0, 0, 0, 1]
})

target_mean = df.groupby('City')['Purchased'].mean()

df['City_target_encoded'] = df['City'].map(target_mean)

df
````

<a id="sec-17"></a>
# 17. Data Leakage in Target Encoding

Target encoding has an important risk: **data leakage**.

If you calculate target statistics using the entire dataset before splitting into training and test data, information from the test set can influence the features.

This is especially dangerous when the category has few observations.

Correct high-level workflow:

```text
Raw data
   ↓
Train/Test split
   ↓
Learn encoding from training data
   ↓
Transform training data
   ↓
Transform test data using training mapping
```

Never allow test-target information to influence training preprocessing.

# 19. Rare Categories

Imagine a dataset contains:

```text
Lahore      5000
Karachi     4000
Islamabad   3000
Multan        20
Quetta        10
Peshawar       5
```

If there are hundreds or thousands of rare categories, one-hot encoding may create many columns.

A common strategy is to group rare categories into `Other`.

For example:

```text
Multan   → Other
Quetta   → Other
Peshawar → Other
```

````python
df = pd.DataFrame({
    'City': [
        'Lahore', 'Lahore', 'Lahore', 'Lahore',
        'Karachi', 'Karachi',
        'Islamabad',
        'Multan',
        'Quetta'
    ]
})

counts = df['City'].value_counts()

threshold = 2

rare_categories = counts[counts < threshold].index

# Create a copy so the original category column is preserved.
df['City_grouped'] = df['City'].where(
    ~df['City'].isin(rare_categories),
    'Other'
)

print(df)
````

# 21. Binary Encoding

Binary encoding represents categories using binary digits.

Suppose there are eight categories:

```text
A → 000
B → 001
C → 010
D → 011
E → 100
F → 101
G → 110
H → 111
```

One-hot encoding would need eight columns.

Binary encoding needs only three binary columns for eight categories.

Binary encoding is especially useful when categorical cardinality is large.

For practical implementation, libraries such as `category_encoders` can be used.

# 22. Hash Encoding

Hash encoding maps categories to a fixed number of columns using a hashing function.

Conceptually:

```text
100,000 unique categories
          ↓
     Hash encoding
          ↓
      100 columns
```

The main advantage is controlling the number of output features.

The trade-off is that different categories can sometimes map to the same location. This is called a hash collision.

# 23. Scikit-learn `OneHotEncoder`

For production machine-learning pipelines, Scikit-learn's `OneHotEncoder` is often more suitable than manually encoding train and test data independently.

Important parameters include:

- `handle_unknown='ignore'`
- `sparse_output=False`

The encoder learns categories from the training data and then transforms new data using the learned representation.

````python
from sklearn.preprocessing import OneHotEncoder

train = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

train_encoded = encoder.fit_transform(train[['City']])

train_encoded
````

<a id="sec-24"></a>
# 24. `fit()` vs `transform()`

This distinction is critical in machine learning.

### `fit()`

The encoder learns information from data.

For one-hot encoding, it learns the available categories.

### `transform()`

The encoder applies what it already learned to new data.

Correct workflow:

```python
encoder.fit(X_train)
X_train_encoded = encoder.transform(X_train)
X_test_encoded = encoder.transform(X_test)
```

Shortcut for training data:

```python
X_train_encoded = encoder.fit_transform(X_train)
```

Do NOT fit a separate encoder on the test data.

````python
train = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

test = pd.DataFrame({
    'City': ['Lahore', 'Multan']
})

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

X_train = encoder.fit_transform(train[['City']])
X_test = encoder.transform(test[['City']])

print('Train:')
print(X_train)

print('\nTest:')
print(X_test)
````

# 25. Encoding With `ColumnTransformer`

Real datasets contain both numerical and categorical columns.

Example:

```text
Age       → numerical
Salary    → numerical
Gender    → categorical
City      → categorical
```

We can apply different preprocessing to different columns using `ColumnTransformer`.

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

X = pd.DataFrame({
    'Age': [25, 30, 28, 35],
    'Salary': [50000, 70000, 60000, 90000],
    'Gender': ['Male', 'Female', 'Male', 'Female'],
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

numeric_features = ['Age', 'Salary']
categorical_features = ['Gender', 'City']

preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numeric_features
        ),
        (
            'cat',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        )
    ]
)

X_transformed = preprocessor.fit_transform(X)

print(X_transformed)
````

Machine-learning model

This reduces the chance of accidentally applying inconsistent preprocessing.

# 28. Choosing the Correct Encoding Technique

Use this mental decision process.

```text
Categorical variable
        |
        v
Does it have a meaningful order?
        |
   +----+----+
   |         |
  Yes        No
   |         |
Ordinal    How many categories?
Encoding       |
          +----+----+
          |         |
         Few       Many
          |         |
       One-Hot   Consider frequency,
                 target, binary, or hash
```

Examples:

| Feature | Possible technique |
|---|---|
| Yes/No | Binary |
| Gender | Binary or one-hot |
| City | One-hot |
| Education level | Ordinal |
| Satisfaction level | Ordinal |
| Browser | One-hot |
| Very high-cardinality ID-like feature | Frequency/hash/other domain-specific strategy |
| Multiple skills per user | Multi-hot |
| Category with target relationship | Target encoding with leakage controls |

# 35. Final Revision Cheat Sheet

## Binary

```text
Yes → 1
No  → 0
```

Use for two-category variables when a binary representation makes sense.

## Ordinal

```text
Low → 0
Medium → 1
High → 2
```

Use when there is a meaningful order.

## One-Hot

```text
Lahore → [1,0,0]
Karachi → [0,1,0]
Islamabad → [0,0,1]
```

Use commonly for nominal categories with manageable cardinality.

## Frequency

```text
Lahore → 500
Karachi → 300
```

Represent categories using their occurrence frequency.

## Target

```text
Category → target statistic
```

Powerful but requires careful leakage prevention.

## Multi-Hot

One row can contain multiple active categories.

## Binary / Hash Encoding

Useful techniques to investigate when cardinality becomes very large.

---

### Most important rule

```text
Train data
    ↓
fit preprocessing
    ↓
transform train
    ↓
transform test using the SAME learned preprocessing
```

Never independently learn preprocessing from the test set.
