# 1. Introduction — What Is Correlation?

### 1. What it means

> **Analogy:** When it's hotter outside, ice-cream shops sell more ice cream. You notice the two things **move together**. That "moving together" is correlation.

**Correlation** is a number that tells us **how strongly two numerical variables move together** and **in which direction**.

### 2. Why is it important?
- To **understand data** (EDA): which things are related?
- To **find useful features**: a feature related to the target can help predict it.
- To **find redundant features**: two features that move together almost perfectly may carry the same information (multicollinearity — Section 10).

### Feature vs target (quick reminder)

```text
| Hours_Studied | Sleep_Hours | Exam_Score |
|---------------|-------------|------------|
|      5        |      7      |     78     |
 \_______ features (X) _______/   target (y)
```

- **Feature** = an input we use to predict.
- **Target** = the output we want to predict.
- **Feature ↔ target correlation** → "is this feature useful?"
- **Feature ↔ feature correlation** → "are these features repeating each other?"

### 3. How it works — the three directions

| Relationship | Meaning | Real example |
|---|---|---|
| **Positive** | One goes up → the other tends to go up | Hours studied ↑ → exam score ↑ |
| **Negative** | One goes up → the other tends to go down | Car age ↑ → car price ↓ |
| **No relationship** | Knowing one tells you nothing about the other | Shoe size vs exam score |

### Strong vs weak

- **Strong:** the points lie close to a line — you can predict one from the other quite well.
- **Weak:** there is a trend, but with lots of scatter.

### ⚠️ Correlation does NOT mean causation
Ice-cream sales and drowning accidents are correlated. Ice cream does not cause drowning — **hot weather** causes both. (Full explanation in Section 8.)

````python
n = 30

# Hours studied vs exam score → strong POSITIVE
hours = rng.uniform(0, 10, n)
score = 40 + 5.5 * hours + rng.normal(0, 4, n)

# Temperature vs ice cream sales → strong POSITIVE
temperature = rng.uniform(15, 40, n)
ice_cream = 20 + 8 * temperature + rng.normal(0, 25, n)

# Age vs salary → moderate POSITIVE (lots of other factors affect salary)
age = rng.uniform(22, 60, n)
salary = 20000 + 1500 * age + rng.normal(0, 18000, n)

# Car age vs price → strong NEGATIVE
car_age = rng.uniform(0, 15, n)
car_price = 3_000_000 - 170_000 * car_age + rng.normal(0, 200_000, n)

# Shoe size vs exam score → NO relationship
shoe_size = rng.uniform(36, 46, n)
score_random = rng.uniform(40, 95, n)

# Advertising budget vs sales → strong POSITIVE
ad_budget = rng.uniform(1, 50, n)          # thousand rupees
sales = 100 + 12 * ad_budget + rng.normal(0, 60, n)

examples = [
    (hours, score, 'Hours studied', 'Exam score'),
    (temperature, ice_cream, 'Temperature (°C)', 'Ice cream sales'),
    (age, salary, 'Age', 'Salary'),
    (ad_budget, sales, 'Ad budget (k)', 'Sales'),
    (car_age, car_price, 'Car age (years)', 'Car price'),
    (shoe_size, score_random, 'Shoe size', 'Exam score'),
]

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, (x, y, xl, yl) in zip(axes.ravel(), examples):
    r = np.corrcoef(x, y)[0, 1]              # Pearson correlation (explained in Section 3)
    ax.scatter(x, y, alpha=0.8)
    ax.set_xlabel(xl); ax.set_ylabel(yl)
    ax.set_title(f'{xl} vs {yl}\ncorrelation r = {r:.2f}')
plt.tight_layout(); plt.show()
````

### 8. Expected output — how to read the plots
- **Hours vs score, temperature vs ice cream, ad budget vs sales:** points climb from bottom-left to top-right → **strong positive** (r close to +1).
- **Age vs salary:** upward trend but more scattered → **moderate positive**.
- **Car age vs price:** points fall from top-left to bottom-right → **strong negative** (r close to −1).
- **Shoe size vs score:** a random cloud → **no relationship** (r close to 0).

### 4. When to use correlation
- During EDA, to discover relationships.
- Before modelling, to find useful and redundant features.

### 5. When NOT to use it
- To prove that one thing **causes** another.
- On categorical data like `City` (needs other tools — Section 7).
- Without looking at a plot (curved relationships and outliers can fool the number — Sections 4 and 9).

### 9. Common mistakes
- Thinking "strong correlation = one causes the other".
- Thinking "correlation 0 = no relationship at all" (there could be a **curved** relationship!).

# 2. Covariance

### 1. What it means

> **Analogy:** Two friends walking in a park. If whenever one walks **above** their usual speed the other also walks **above** theirs, they "vary together" → positive covariance.

**Covariance** measures whether two variables tend to be **above/below their averages at the same time**.

| Covariance | Meaning |
|---|---|
| **Positive** | When X is above its mean, Y tends to be above its mean |
| **Negative** | When X is above its mean, Y tends to be below its mean |
| **≈ 0** | No consistent linear pattern |

### 2. Why is it important?
It is the **building block** of correlation, PCA, and many statistical methods.

### 3. How it works — the formula (step by step)

$$\text{Cov}(X, Y) = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{n - 1}$$

In plain English:
1. Find the mean of X and the mean of Y.
2. For each row, compute how far X is from its mean and how far Y is from its mean.
3. **Multiply** those two distances.
   - both above mean → (+)(+) = **+**
   - both below mean → (−)(−) = **+**
   - one above, one below → (+)(−) = **−**
4. Add up all the products and divide by `n − 1`.

(Why `n − 1` instead of `n`? It's the "sample" version, which corrects a small bias when we only have a sample. Pandas uses `n − 1` by default.)

### 6. A simple numerical example (by hand)

| Student | Hours (x) | Score (y) | x − x̄ | y − ȳ | product |
|---|---|---|---|---|---|
| A | 1 | 50 | −2 | −14 | 28 |
| B | 2 | 55 | −1 | −9 | 9 |
| C | 3 | 65 | 0 | 1 | 0 |
| D | 4 | 70 | 1 | 6 | 6 |
| E | 5 | 80 | 2 | 16 | 32 |
| | x̄ = 3 | ȳ = 64 | | | **sum = 75** |

$$\text{Cov} = \frac{75}{5 - 1} = 18.75$$

Positive → more hours go with higher scores.

````python
tiny = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5],
    'Score': [50, 55, 65, 70, 80]
})

# Step-by-step manual calculation
x_mean = tiny['Hours'].mean()                      # 3
y_mean = tiny['Score'].mean()                      # 64
tiny['x_dev'] = tiny['Hours'] - x_mean             # distance from mean
tiny['y_dev'] = tiny['Score'] - y_mean
tiny['product'] = tiny['x_dev'] * tiny['y_dev']    # multiply the distances

manual_cov = tiny['product'].sum() / (len(tiny) - 1)
print(tiny)
print('\nSum of products:', tiny['product'].sum())
print('Manual covariance:', manual_cov)

# Same thing with Pandas
print('Pandas covariance:', tiny['Hours'].cov(tiny['Score']))
````

### Covariance matrix with `df.cov()`

````python
cov_df = pd.DataFrame({
    'Hours':      [1, 2, 3, 4, 5],
    'Score':      [50, 55, 65, 70, 80],
    'TV_Hours':   [5, 4, 4, 2, 1],        # more TV → fewer study hours
    'Shoe_Size':  [42, 38, 44, 39, 41]    # unrelated
})
cov_df.cov().round(2)
````

### 8. Expected output — reading `df.cov()`

```text
           Hours  Score  TV_Hours  Shoe_Size
Hours       2.50  18.75     -2.50      -0.25
Score      18.75 142.50    -18.50       1.00
TV_Hours   -2.50 -18.50      2.70       1.05
Shoe_Size  -0.25   1.00      1.05       5.70
```

How to read it:
- It is a **square, symmetric** table: `Cov(Hours, Score) = Cov(Score, Hours)`.
- The **diagonal** is each variable's covariance with itself = its **variance** (`Hours` variance = 2.5).
- `Hours`–`Score` = **+18.75** → positive relationship.
- `Hours`–`TV_Hours` = **negative** → more TV, fewer study hours.
- `Shoe_Size` values are **near zero** → no clear pattern.

### ⚠️ Why covariance depends on scale (its big weakness)

````python
scale_demo = pd.DataFrame({
    'Height_m':  [1.60, 1.65, 1.70, 1.75, 1.80],
    'Weight_kg': [55, 60, 66, 72, 78]
})
scale_demo['Height_cm'] = scale_demo['Height_m'] * 100      # SAME information, different unit

print('Cov(Height_m,  Weight):', round(scale_demo['Height_m'].cov(scale_demo['Weight_kg']), 3))
print('Cov(Height_cm, Weight):', round(scale_demo['Height_cm'].cov(scale_demo['Weight_kg']), 3))
print()
print('Corr(Height_m,  Weight):', round(scale_demo['Height_m'].corr(scale_demo['Weight_kg']), 3))
print('Corr(Height_cm, Weight):', round(scale_demo['Height_cm'].corr(scale_demo['Weight_kg']), 3))
````

The **relationship is identical**, but covariance became **100× bigger** just because we changed metres to centimetres. So:
- The **sign** of covariance (+/−) is meaningful.
- The **size** of covariance is **not** easy to interpret or compare.

Correlation fixes this by dividing covariance by the standard deviations → always between −1 and +1, **unit-free**.

### Covariance vs correlation

| | Covariance | Correlation |
|---|---|---|
| Range | −∞ to +∞ | −1 to +1 |
| Depends on units? | **Yes** | **No** |
| Tells direction? | Yes | Yes |
| Tells strength? | Not directly | **Yes** |
| Easy to compare? | No | Yes |

### 4. When to use covariance
Inside math (PCA, portfolio risk in finance, Gaussian models). For **interpreting** relationships, use correlation.

### 5. When NOT to use it
To compare the strength of different pairs of variables (units differ!).

### 9. Common mistakes
- "Covariance of 5000 is stronger than covariance of 2" — wrong; units could be different.
- Forgetting that pandas uses `n − 1` (NumPy's `np.cov` also does by default; `ddof=0` would use `n`).

# 3. Pearson Correlation

### 1. What it means
The **Pearson correlation coefficient (r)** measures the strength and direction of a **linear (straight-line)** relationship between two numerical variables.

### 3. How it works — mathematical intuition

$$r = \frac{\text{Cov}(X, Y)}{\sigma_X \cdot \sigma_Y}$$

- Covariance tells the direction but depends on units.
- Dividing by the standard deviations (σ) of X and Y **removes the units**.
- The result is always between **−1 and +1**.

Another way to think of it: convert both columns to **z-scores** (Topic 4: standardization), then Pearson r ≈ the average of `z_x × z_y`.

### Numerical example (same tiny data)
From Section 2: Cov = 18.75, σ_Hours = √2.5 = 1.581, σ_Score = √142.5 = 11.937

$$r = \frac{18.75}{1.581 \times 11.937} = \frac{18.75}{18.875} \approx 0.993$$

Very close to +1 → a very strong positive linear relationship.

### What the values mean

| r | Meaning | Picture |
|---|---|---|
| **+1** | Perfect positive line | All points exactly on a rising line |
| **close to +1** (0.8, 0.9) | Strong positive | Points tightly around a rising line |
| **0** | No **linear** relationship | Random cloud (or a curve!) |
| **close to −1** | Strong negative | Points tightly around a falling line |
| **−1** | Perfect negative line | All points exactly on a falling line |

### Interpretation guide (a common rule of thumb — not a law)

| \|r\| | Strength |
|---|---|
| 0.00 – 0.19 | Very weak / none |
| 0.20 – 0.39 | Weak |
| 0.40 – 0.59 | Moderate |
| 0.60 – 0.79 | Strong |
| 0.80 – 1.00 | Very strong |

The sign (+/−) gives the **direction**; the absolute value gives the **strength**.

````python
tiny = pd.DataFrame({'Hours': [1, 2, 3, 4, 5], 'Score': [50, 55, 65, 70, 80]})

cov = tiny['Hours'].cov(tiny['Score'])
std_x = tiny['Hours'].std()
std_y = tiny['Score'].std()
r_manual = cov / (std_x * std_y)

print('Covariance        :', cov)
print('Std Hours, Score  :', round(std_x, 3), round(std_y, 3))
print('Pearson r (manual):', round(r_manual, 4))
print('Pearson r (pandas):', round(tiny['Hours'].corr(tiny['Score']), 4))
````

### Examples of every strength

````python
n = 200
x = rng.normal(0, 1, n)

corr_examples = pd.DataFrame({
    'X': x,
    'Strong_Positive': x + rng.normal(0, 0.3, n),
    'Weak_Positive':   x + rng.normal(0, 2.5, n),
    'Strong_Negative': -x + rng.normal(0, 0.3, n),
    'Weak_Negative':   -x + rng.normal(0, 2.5, n),
    'Zero':            rng.normal(0, 1, n),
})

# Correlation of each column with X
print(corr_examples.corr()['X'].round(2))
````

````python
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for ax, col in zip(axes, corr_examples.columns[1:]):
    r = corr_examples['X'].corr(corr_examples[col])
    ax.scatter(corr_examples['X'], corr_examples[col], alpha=0.5, s=15)
    ax.set_title(f'{col}\nr = {r:.2f}')
    ax.set_xlabel('X')
plt.tight_layout(); plt.show()
````

### Pandas functions
- `df.corr(numeric_only=True)` → a full correlation **matrix** of every numeric column. `numeric_only=True` skips text columns (without it, text columns raise an error).
- `df['A'].corr(df['B'])` → a **single number** for one pair.

````python
people = pd.DataFrame({
    'Name':       ['Ali', 'Sara', 'Ahmed', 'Hina', 'Omar', 'Zara', 'Bilal', 'Ayesha'],
    'Age':        [23, 27, 31, 35, 40, 45, 50, 58],
    'Experience': [1, 4, 7, 10, 15, 20, 24, 30],
    'Salary':     [45000, 60000, 72000, 85000, 105000, 120000, 125000, 150000],
    'Commute_km': [12, 5, 20, 8, 15, 3, 18, 10],
    'City':       ['Lahore', 'Karachi', 'Lahore', 'Islamabad', 'Karachi', 'Lahore', 'Multan', 'Lahore']
})

print('Single pair: Age vs Salary =', round(people['Age'].corr(people['Salary']), 3))
people.corr(numeric_only=True).round(2)
````

### 8. Expected output
- `Age`, `Experience`, `Salary` all correlate **very strongly** with each other (≈ 0.98–1.00).
- `Commute_km` has weak correlations with everything.
- `Name` and `City` are ignored because of `numeric_only=True`.
- The diagonal is always **1.00** (every variable is perfectly correlated with itself).

### 4. When to use Pearson
- Both variables are numerical.
- The relationship looks roughly **linear** in a scatter plot.
- There are no extreme outliers.

### 5. When NOT to use Pearson
- Curved (non-linear) relationships → use Spearman or look at the plot.
- Strong outliers → Spearman is more robust.
- Ordinal data (ranks, ratings 1–5) → Spearman/Kendall.
- Categorical data → Section 7.

### 9. Common mistakes
- Forgetting `numeric_only=True` when the DataFrame has text columns (error in modern pandas).
- Reading r = 0.3 as "30% related" — r is **not** a percentage. (r² = 0.09 means ~9% of the variance is explained by a linear fit.)
- Concluding "no relationship" from r ≈ 0 when the relationship is curved.

````python
# r ≈ 0 does NOT always mean "no relationship": a perfect U-shape
u = pd.DataFrame({'X': np.linspace(-3, 3, 50)})
u['Y'] = u['X'] ** 2

print('Pearson r for Y = X²:', round(u['X'].corr(u['Y']), 3))
plt.figure(figsize=(5, 3.5))
plt.scatter(u['X'], u['Y'])
plt.title('Y = X²  (perfect relationship, yet r ≈ 0)')
plt.show()
````

# 4. Visualizing Correlation

### 1. What it means
Three main plots:

| Plot | Shows | Best for |
|---|---|---|
| **Scatter plot** | Every point for **one pair** of variables | Shape of the relationship, outliers, curves |
| **Heatmap** | The whole correlation **matrix** as coloured squares | Quick overview of many variables |
| **Pair plot** | Scatter plots for **every pair** + distributions | Detailed EDA of a few variables |

### 2. Why is it important?
A single number can hide the truth. **Anscombe's quartet** (below) shows four datasets with **the same correlation** but completely different shapes.

### How to visually identify relationships in a scatter plot
- Points rising left → right: **positive**
- Points falling left → right: **negative**
- Round/random cloud: **none**
- Tight band: **strong**; wide band: **weak**
- A curve: **non-linear** (Pearson may mislead)
- A few points far away: **outliers**

````python
# Scatter plot with seaborn, including a regression line
study = pd.DataFrame({'Hours': hours, 'Score': score})

plt.figure(figsize=(6, 4))
sns.scatterplot(data=study, x='Hours', y='Score')
sns.regplot(data=study, x='Hours', y='Score', scatter=False, color='red')   # best-fit line only
plt.title(f"Hours vs Score (r = {study['Hours'].corr(study['Score']):.2f})")
plt.show()
````

````python
# Positive, negative and no correlation side by side
x = rng.uniform(0, 10, 100)
vis = pd.DataFrame({
    'x': np.tile(x, 3),
    'y': np.concatenate([2 * x + rng.normal(0, 2, 100),
                         -2 * x + rng.normal(0, 2, 100),
                         rng.normal(10, 5, 100)]),
    'type': ['Positive'] * 100 + ['Negative'] * 100 + ['No correlation'] * 100
})

g = sns.relplot(data=vis, x='x', y='y', col='type', kind='scatter', height=3.5,
                facet_kws={'sharey': False})
for ax, t in zip(g.axes.flat, ['Positive', 'Negative', 'No correlation']):
    sub = vis[vis['type'] == t]
    ax.set_title(f"{t} (r = {sub['x'].corr(sub['y']):.2f})")
plt.show()
````

### Heatmap

````python
corr_matrix = people.corr(numeric_only=True)

plt.figure(figsize=(6, 4.5))
sns.heatmap(
    corr_matrix,
    annot=True,          # write the numbers inside the squares
    fmt='.2f',           # 2 decimal places
    cmap='coolwarm',     # blue = negative, red = positive
    vmin=-1, vmax=1,     # fix the colour scale to the full correlation range
    square=True
)
plt.title('Correlation heatmap')
plt.show()
````

**Reading the heatmap:** dark red = strong positive, dark blue = strong negative, white/pale = near zero. The diagonal is always 1.

**Tip:** for many columns, show only the lower triangle (the upper triangle is a mirror copy):

````python
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))    # True for the upper triangle → hidden
plt.figure(figsize=(6, 4.5))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, square=True)
plt.title('Lower-triangle heatmap')
plt.show()
````

### Pair plot

````python
sns.pairplot(people[['Age', 'Experience', 'Salary', 'Commute_km']], height=2)
plt.show()
````

**Reading the pair plot:** each off-diagonal square is a scatter plot of two variables. The diagonal shows each variable's distribution (histogram). Age/Experience/Salary show clear straight lines; Commute_km looks random.

### Why scatter plots matter even when correlation is high or low — Anscombe's quartet

````python
# Anscombe's quartet (built manually so no internet download is needed)
x_a = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
anscombe = pd.DataFrame({
    'I_x': x_a,  'I_y':  [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68],
    'II_x': x_a, 'II_y': [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74],
    'III_x': x_a,'III_y':[7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73],
    'IV_x': [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8],
    'IV_y': [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89],
})

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, name in zip(axes, ['I', 'II', 'III', 'IV']):
    xa, ya = anscombe[f'{name}_x'], anscombe[f'{name}_y']
    ax.scatter(xa, ya)
    ax.set_title(f'Dataset {name}: r = {xa.corr(ya):.3f}')
plt.tight_layout(); plt.show()
````

All four datasets have **r ≈ 0.816**, but:
- **I:** a normal linear relationship ✅
- **II:** a perfect **curve** — Pearson is the wrong tool
- **III:** a perfect line ruined by **one outlier**
- **IV:** no relationship at all — **one extreme point** creates the entire correlation

**Lesson:** always plot your data before trusting a correlation number.

### 9. Common mistakes
- Heatmap without `vmin=-1, vmax=1` → colours rescale and a weak 0.3 can look dark red.
- Pair plots with 30 columns → 900 tiny plots; use them for ≤ 6–8 variables.
- Only looking at the heatmap and never at scatter plots.

# 5. Spearman Correlation

### 1. What it means

> **Analogy:** In a race, you don't care *how many seconds* each runner took — only **who came 1st, 2nd, 3rd**. Spearman looks only at the **ranks (order)** of values.

**Spearman's rank correlation (ρ, "rho")** = Pearson correlation calculated on the **ranks** of the values instead of the values themselves.

### Linear vs monotonic

| Relationship | Meaning | Example |
|---|---|---|
| **Linear** | Rises (or falls) at a **constant** rate — a straight line | Price × 2 = cost of 2 items |
| **Monotonic** | Always rises (or always falls), but the rate can **change** — a curve that never turns back | Experience vs skill: learns fast at first, then slower |
| **Non-monotonic** | Goes up then down (or vice versa) | Y = X² (U-shape) |

- Pearson measures **linear** relationships.
- Spearman measures **monotonic** relationships.
- Neither measures U-shaped (non-monotonic) relationships well.

### 3. How it works
1. Replace each value with its rank (smallest = 1).
2. Compute Pearson correlation on the ranks.

### 6. Numerical example

| X | Y = X³ | rank X | rank Y |
|---|---|---|---|
| 1 | 1 | 1 | 1 |
| 2 | 8 | 2 | 2 |
| 3 | 27 | 3 | 3 |
| 4 | 64 | 4 | 4 |
| 5 | 125 | 5 | 5 |

The ranks match perfectly → **Spearman = 1.0**. Pearson is lower (≈ 0.94) because the points form a curve, not a line.

````python
cube = pd.DataFrame({'X': [1, 2, 3, 4, 5]})
cube['Y'] = cube['X'] ** 3
cube['rank_X'] = cube['X'].rank()
cube['rank_Y'] = cube['Y'].rank()
print(cube)

print('\nPearson :', round(cube['X'].corr(cube['Y']), 3))
print('Spearman:', round(cube['X'].corr(cube['Y'], method='spearman'), 3))
print('Pearson on ranks:', round(cube['rank_X'].corr(cube['rank_Y']), 3), '← same as Spearman!')
````

### A dataset where Pearson and Spearman behave differently

````python
n = 100
exp = rng.uniform(0, 20, n)
skill = np.log1p(exp) * 30 + rng.normal(0, 3, n)             # monotonic curve: fast early, slow later
income = np.exp(exp / 4) * 1000 + rng.normal(0, 500, n)     # monotonic, grows explosively

mono = pd.DataFrame({'Experience': exp, 'Skill': skill, 'Income': income})

print('Pearson :')
print(mono.corr(method='pearson').round(3), '\n')
print('Spearman:')
print(mono.corr(method='spearman').round(3))
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, col in zip(axes, ['Skill', 'Income']):
    p = mono['Experience'].corr(mono[col])
    s = mono['Experience'].corr(mono[col], method='spearman')
    ax.scatter(mono['Experience'], mono[col], alpha=0.6)
    ax.set_title(f'Experience vs {col}\nPearson = {p:.2f}   Spearman = {s:.2f}')
    ax.set_xlabel('Experience')
plt.tight_layout(); plt.show()
````

### 8. Expected output
- **Experience vs Income** (explosive curve): Pearson is noticeably lower (≈ 0.86) because the points are not on a straight line; Spearman is very close to 1 because income **always increases** with experience.
- **Experience vs Skill**: both are high, Spearman slightly higher.

### 4. When is Spearman useful?
- Monotonic but curved relationships.
- **Ordinal** data (satisfaction 1–5, rankings, grades A/B/C coded as numbers).
- Data with **outliers** (ranks limit their influence — Section 9).
- Skewed data (income, prices).

### 5. When is Pearson better?
- The relationship is truly linear and data is clean — Pearson uses the actual distances, so it is more precise.
- When you care about **linear** models (linear regression assumes linear relationships).

### 9. Common mistakes
- Using Spearman for a U-shape — it will also be ≈ 0.
- Forgetting `method='spearman'` (the default is Pearson).

# 6. Kendall Correlation (Kendall's Tau, τ)

### 1. What it means

> **Analogy:** Take every possible **pair of people**. If the person who studied more also scored more, that pair "agrees" (concordant). If they studied more but scored less, the pair "disagrees" (discordant). Kendall's tau counts agreements vs disagreements.

### 3. How it works (basic intuition)

$$\tau = \frac{\text{concordant pairs} - \text{discordant pairs}}{\text{total pairs}}$$

(The version used by SciPy, "tau-b", also adjusts for ties.)

### 6. Numerical example
Students: A (hours 1, score 50), B (2, 60), C (3, 55)

| Pair | Hours order | Score order | Result |
|---|---|---|---|
| A–B | A < B | 50 < 60 | Concordant ✅ |
| A–C | A < C | 50 < 55 | Concordant ✅ |
| B–C | B < C | 60 > 55 | Discordant ❌ |

τ = (2 − 1) / 3 = **0.333**

### 4. When is Kendall useful?
- **Small datasets**.
- Data with **many ties** (e.g. ratings 1–5).
- When you want a value with a clear probability meaning: τ relates to "the chance that a random pair is in the same order".
- It is a bit more robust than Spearman to errors/outliers.

### 5. When NOT to use it
- Large datasets (slower to compute — it looks at every pair).
- When you need a linear measure (use Pearson).

Kendall values are usually **smaller in size** than Spearman for the same data (e.g. Spearman 0.9 ↔ Kendall 0.75). That is normal — they are on different scales, so don't compare a Kendall value to a Spearman threshold.

````python
from scipy import stats

small = pd.DataFrame({'Hours': [1, 2, 3], 'Score': [50, 60, 55]})
tau, p_value = stats.kendalltau(small['Hours'], small['Score'])
print('Kendall tau (tiny example):', round(tau, 3))

# SciPy also gives Pearson and Spearman, plus a p-value
r_p, p_p = stats.pearsonr(mono['Experience'], mono['Income'])
r_s, p_s = stats.spearmanr(mono['Experience'], mono['Income'])
r_k, p_k = stats.kendalltau(mono['Experience'], mono['Income'])

print(f'\nExperience vs Income')
print(f'Pearson : {r_p:.3f}  (p = {p_p:.2e})')
print(f'Spearman: {r_s:.3f}  (p = {p_s:.2e})')
print(f'Kendall : {r_k:.3f}  (p = {p_k:.2e})')

# Pandas can do Kendall too
print('\nPandas Kendall:', round(mono['Experience'].corr(mono['Income'], method='kendall'), 3))
````

**What is the p-value?** It answers: *"If there were truly no relationship, how surprising would this correlation be?"* A small p-value (commonly < 0.05) suggests the correlation is unlikely to be just random chance. It does **not** tell you how strong or important the relationship is.

### Comparison: Pearson vs Spearman vs Kendall

| | Pearson (r) | Spearman (ρ) | Kendall (τ) |
|---|---|---|---|
| Measures | **Linear** relationship | **Monotonic** relationship | **Monotonic** (pair agreement) |
| Uses | Actual values | Ranks | Pairs of ranks |
| Range | −1 to +1 | −1 to +1 | −1 to +1 |
| Outlier-sensitive? | **Very** | Less | Least |
| Good for ordinal data? | No | Yes | Yes (esp. with ties) |
| Best dataset size | Any | Any | Small–medium |
| Speed | Fast | Fast | Slower on big data |
| Pandas | `corr()` | `corr(method='spearman')` | `corr(method='kendall')` |
| SciPy | `stats.pearsonr` | `stats.spearmanr` | `stats.kendalltau` |

### 9. Common mistakes
- Comparing a Kendall value directly with a Pearson value ("0.7 is weaker than 0.9") — different scales.
- Believing a tiny p-value means a strong relationship. With 1 million rows even r = 0.01 can have p < 0.05.

# 7. Correlation with Categorical Variables

### 1. What it means
Pearson, Spearman and Kendall need **numbers that have an order**. `City = Lahore / Karachi` has no numeric meaning — if you code Lahore = 1, Karachi = 2, Islamabad = 3, the "correlation" depends on your arbitrary coding. So we need **different tools** depending on the variable types.

| Variable pair | Example | Tool |
|---|---|---|
| **Numerical ↔ numerical** | Age vs Salary | Pearson / Spearman / Kendall |
| **Binary categorical ↔ numerical** | Gender (M/F) vs Salary | **Point-biserial correlation** |
| **Multi-class categorical ↔ numerical** | City vs Salary | Compare group means (`groupby`), ANOVA, correlation ratio (η) |
| **Categorical ↔ categorical** | Gender vs Purchased (Yes/No) | **Chi-square test** + **Cramér's V** |

---

<a id="sec-8"></a>
## 7.1 Point-biserial correlation (binary ↔ numerical)

### What it means
A special case of Pearson where one variable is **binary (0/1)**. It asks: *"Is the numeric value generally higher in group 1 than in group 0?"*

### How it works
Code the binary variable as 0/1 and compute Pearson. Mathematically it compares the two group means.

### When to use / not use
- ✅ Two groups (Yes/No, Churned/Not, Male/Female) vs a numeric variable.
- ❌ More than two categories.

````python
from scipy import stats

n = 60
purchase_df = pd.DataFrame({'Time_on_Site_min': rng.normal(8, 3, n).round(1).clip(0.5)})
# Customers who spend more time are more likely to buy
prob = 1 / (1 + np.exp(-(purchase_df['Time_on_Site_min'] - 8)))
purchase_df['Purchased'] = (rng.random(n) < prob).astype(int)

r_pb, p = stats.pointbiserialr(purchase_df['Purchased'], purchase_df['Time_on_Site_min'])
print(f'Point-biserial r = {r_pb:.3f}  (p = {p:.4f})')
print('Same as Pearson with 0/1 :', round(purchase_df['Purchased'].corr(purchase_df['Time_on_Site_min']), 3))
print('\nAverage time by group:')
print(purchase_df.groupby('Purchased')['Time_on_Site_min'].mean().round(2))
````

````python
plt.figure(figsize=(5, 4))
sns.boxplot(data=purchase_df, x='Purchased', y='Time_on_Site_min')
plt.title(f'Time on site by purchase (r_pb = {r_pb:.2f})')
plt.show()
````

**Reading it:** a positive point-biserial r means group **1** (purchased) has higher time on site on average. The box plot confirms it visually.

## 7.2 Chi-square test (categorical ↔ categorical)

### What it means

> **Analogy:** If gender had nothing to do with which product people buy, you'd expect the same buying proportions for men and women. The chi-square test measures **how far the real counts are from those "no relationship" expected counts**.

### How it works
1. Build a **contingency table** (a count table) with `pd.crosstab`.
2. Calculate what the counts **would be** if the variables were independent.
3. Compare actual vs expected. A big difference → large chi-square → small p-value → probably related.

### When to use / not use
- ✅ Two categorical variables.
- ❌ Expected counts smaller than ~5 in many cells (results unreliable).
- ❌ Measuring **strength** — chi-square grows with sample size. Use Cramér's V for strength.

````python
n = 300
survey = pd.DataFrame({'Gender': rng.choice(['Male', 'Female'], n)})
# Make product preference depend a bit on gender
survey['Product'] = np.where(
    survey['Gender'] == 'Male',
    rng.choice(['Laptop', 'Phone', 'Watch'], n, p=[0.5, 0.3, 0.2]),
    rng.choice(['Laptop', 'Phone', 'Watch'], n, p=[0.25, 0.35, 0.40])
)

table = pd.crosstab(survey['Gender'], survey['Product'])
print('Observed counts:')
print(table, '\n')

chi2, p, dof, expected = stats.chi2_contingency(table)
print('Expected counts if NO relationship:')
print(pd.DataFrame(expected, index=table.index, columns=table.columns).round(1), '\n')
print(f'Chi-square = {chi2:.2f}, degrees of freedom = {dof}, p-value = {p:.5f}')
````

**Reading it:** Men buy more laptops than expected, women more watches. The p-value is very small → gender and product choice are **probably related** (not independent).

## 7.3 Cramér's V (strength for categorical ↔ categorical)

### What it means
Cramér's V turns chi-square into a **0 to 1** strength score (like a correlation, but without direction, because categories have no order).

$$V = \sqrt{\frac{\chi^2}{n \cdot (\min(\text{rows}, \text{cols}) - 1)}}$$

| V | Strength (rough guide) |
|---|---|
| 0 – 0.1 | Negligible |
| 0.1 – 0.3 | Weak |
| 0.3 – 0.5 | Moderate |
| > 0.5 | Strong |

````python
def cramers_v(x, y):
    '''Cramér's V between two categorical Series (0 = no association, 1 = perfect).'''
    table = pd.crosstab(x, y)                              # count table
    chi2 = stats.chi2_contingency(table, correction=False)[0]
    n = table.to_numpy().sum()                             # total number of observations
    k = min(table.shape) - 1                               # smaller dimension minus 1
    return np.sqrt(chi2 / (n * k))

print("Cramér's V (Gender vs Product):", round(cramers_v(survey['Gender'], survey['Product']), 3))

# Compare with a completely unrelated random column
survey['Random_Color'] = rng.choice(['Red', 'Blue', 'Green'], n)
print("Cramér's V (Gender vs Random_Color):", round(cramers_v(survey['Gender'], survey['Random_Color']), 3))
````

### 9. Common mistakes
- Label-encoding `City` (1, 2, 3...) and running Pearson → meaningless number that changes if you change the coding.
- Using the chi-square value (or p-value) as a strength measure → use Cramér's V.
- Running chi-square on continuous data without binning it first.

# 8. Correlation Does NOT Mean Causation

### 1. What it means
**Correlation:** two things move together.
**Causation:** one thing **makes** the other happen.

Correlation is **necessary** evidence for many causal claims, but it is **never sufficient** by itself.

### Why two variables can be correlated without causing each other

| Reason | Meaning | Example |
|---|---|---|
| **Confounding variable** | A third variable causes both | Hot weather → more ice cream **and** more drownings |
| **Hidden (lurking) variable** | An unmeasured cause | Cities with more police have more crime — because **big population** causes both |
| **Spurious correlation** | Pure coincidence, often two things that both trend over time | Number of films an actor made per year vs pool drownings |
| **Reverse causation** | The direction is opposite of what you think | Hospitals have many sick people — hospitals don't make people sick |

### 2. Why is it important?
- Businesses make bad decisions ("let's ban ice cream to stop drowning").
- In ML, a model may use a correlated feature that **stops working** when conditions change.

### Example — a confounder in action

````python
n = 120
temperature = rng.uniform(10, 42, n)                                    # the CONFOUNDER
ice_cream_sales = 50 + 10 * temperature + rng.normal(0, 40, n)          # caused by temperature
drownings = 0.3 * temperature + rng.normal(0, 1.5, n)                   # also caused by temperature

conf = pd.DataFrame({'Temperature': temperature, 'Ice_Cream_Sales': ice_cream_sales, 'Drownings': drownings})
print('Correlation matrix:')
print(conf.corr().round(2))
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(conf['Ice_Cream_Sales'], conf['Drownings'], alpha=0.7)
axes[0].set_title(f"Ice cream vs drownings (r = {conf['Ice_Cream_Sales'].corr(conf['Drownings']):.2f})")
axes[0].set_xlabel('Ice cream sales'); axes[0].set_ylabel('Drownings')

sc = axes[1].scatter(conf['Ice_Cream_Sales'], conf['Drownings'], c=conf['Temperature'], cmap='coolwarm')
plt.colorbar(sc, ax=axes[1], label='Temperature')
axes[1].set_title('Same plot coloured by temperature (the hidden cause)')
axes[1].set_xlabel('Ice cream sales')
plt.tight_layout(); plt.show()
````

The colours show hot days (red) on the top-right and cold days (blue) on the bottom-left. **Temperature drives both.**

### Removing the confounder's effect
If we look **within similar temperatures**, the ice cream–drowning relationship mostly disappears. A simple way: remove the part of each variable explained by temperature (the "residuals") and correlate what is left. This is called **partial correlation**.

````python
# Fit a straight line of each variable on temperature and keep the leftover (residual) part
ice_resid = conf['Ice_Cream_Sales'] - np.polyval(np.polyfit(conf['Temperature'], conf['Ice_Cream_Sales'], 1), conf['Temperature'])
drown_resid = conf['Drownings'] - np.polyval(np.polyfit(conf['Temperature'], conf['Drownings'], 1), conf['Temperature'])

print('Correlation BEFORE controlling for temperature:', round(conf['Ice_Cream_Sales'].corr(conf['Drownings']), 3))
print('Correlation AFTER controlling for temperature :', round(np.corrcoef(ice_resid, drown_resid)[0, 1], 3))
````

### Spurious correlation — two unrelated things that both grow over time

````python
years = np.arange(2005, 2025)
spur = pd.DataFrame({
    'Year': years,
    'Smartphone_Users_M': 5 + 3.5 * (years - 2005) + rng.normal(0, 2, len(years)),    # grows over time
    'Coffee_Shops': 200 + 25 * (years - 2005) + rng.normal(0, 15, len(years))          # also grows over time
})
print('Correlation:', round(spur['Smartphone_Users_M'].corr(spur['Coffee_Shops']), 3))
print('Does smartphone use cause coffee shops to open? No — both simply grow with TIME.')
````

### How to get closer to causation
- **Randomized controlled experiments** (A/B tests) — the gold standard.
- Control for confounders (partial correlation, regression with the confounder included).
- Domain knowledge and a plausible mechanism.
- Time order (the cause must happen before the effect).

### 9. Common mistakes
- "Students who eat breakfast get better grades, so breakfast causes good grades" — family income could cause both.
- Trusting a correlation between two time-trending series.

# 9. Outliers and Correlation

### 1. What it means
An **outlier** is a value far away from the rest (Topic 5). Because Pearson uses the actual distances from the mean — and **squares** them inside the standard deviation — a single extreme point can **create, destroy, or even reverse** a correlation.

### 3. How it works
Pearson multiplies `(x − x̄)(y − ȳ)`. For a normal point these distances are small; for an extreme outlier they are huge, so that single product can dominate the whole sum.

Spearman uses **ranks**. An outlier is just "rank 31 out of 31" — it cannot be *more* extreme than the largest rank, so its influence is limited.

````python
n = 30
x_norm = rng.uniform(0, 10, n)
y_norm = rng.normal(50, 5, n)                  # NO relationship with x

out_df = pd.DataFrame({'X': x_norm, 'Y': y_norm})
with_outlier = pd.concat([out_df, pd.DataFrame({'X': [60], 'Y': [200]})], ignore_index=True)

print('WITHOUT outlier → Pearson:', round(out_df['X'].corr(out_df['Y']), 3),
      '| Spearman:', round(out_df['X'].corr(out_df['Y'], method='spearman'), 3))
print('WITH outlier    → Pearson:', round(with_outlier['X'].corr(with_outlier['Y']), 3),
      '| Spearman:', round(with_outlier['X'].corr(with_outlier['Y'], method='spearman'), 3))
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(out_df['X'], out_df['Y'])
axes[0].set_title(f"Without outlier: Pearson r = {out_df['X'].corr(out_df['Y']):.2f}")

axes[1].scatter(with_outlier['X'][:-1], with_outlier['Y'][:-1], label='normal points')
axes[1].scatter([60], [200], color='red', s=120, label='ONE outlier')
axes[1].set_title(f"With outlier: Pearson r = {with_outlier['X'].corr(with_outlier['Y']):.2f}")
axes[1].legend()
plt.tight_layout(); plt.show()
````

### 8. Expected output
- Without the outlier: Pearson ≈ 0 (no relationship — correct).
- With **one** outlier out of 31 points: Pearson jumps to a **strong** positive value (≈ 0.9). A fake relationship was created by a single row!
- Spearman barely changes, because the outlier only moves one rank.

### Can an outlier also destroy a real correlation? Yes:

````python
x_line = np.arange(1, 21)
y_line = 3 * x_line + rng.normal(0, 2, 20)            # strong positive
line_df = pd.DataFrame({'X': x_line, 'Y': y_line})
line_bad = pd.concat([line_df, pd.DataFrame({'X': [25], 'Y': [-150]})], ignore_index=True)   # typo/data error

print('Real data     → Pearson:', round(line_df['X'].corr(line_df['Y']), 3),
      '| Spearman:', round(line_df['X'].corr(line_df['Y'], method='spearman'), 3))
print('+ one bad row → Pearson:', round(line_bad['X'].corr(line_bad['Y']), 3),
      '| Spearman:', round(line_bad['X'].corr(line_bad['Y'], method='spearman'), 3))
````

### Does Spearman help?
**Yes, partly.** Spearman is much more **robust** to outliers. But it is not a cure:
- It changes the question (monotonic instead of linear).
- The outlier might be a **data error** that should be fixed or removed anyway.

### What to do with outliers when measuring correlation
1. **Always plot** (scatter plot).
2. Investigate outliers: error or real?
3. Compare Pearson with Spearman — a big difference is a warning sign.
4. Consider removing errors, capping (winsorizing), or log-transforming skewed data (Topics 5 and 11).

### 9. Common mistakes
- Reporting a correlation without checking for outliers.
- Automatically deleting every outlier — sometimes they are the most important data (fraud, rare diseases).

# 10. Multicollinearity — What Is It?

### 1. What it means

> **Analogy:** You ask three witnesses what happened. Witness 1 tells the story in English, Witness 2 tells the **same story** in Urdu, Witness 3 tells it again in Punjabi. You have three witnesses, but only **one piece of information**. And if the judge must decide *which witness deserves credit*, it becomes impossible.

**Multicollinearity** happens when **two or more features (inputs)** in a model are **highly correlated with each other**, so they carry **overlapping (redundant) information**.

Important: multicollinearity is about **features vs features**, not features vs target.

```text
Age            → 25
Age_in_months  → 300      ← same information × 12
Age_in_days    → 9125     ← same information × 365
```

These three columns contain **the same information** in different units. Their correlation is exactly 1.0. This is **perfect multicollinearity**.

### Types
| Type | Meaning | Example |
|---|---|---|
| **Perfect** | One feature is an exact linear function of others | `Age_months = 12 × Age` ; `Total = A + B` with A, B, Total all included |
| **High / strong** | Features are strongly (not perfectly) related | Age and Years_of_Experience (r ≈ 0.95) |
| **Multi-feature** | One feature is predictable from a **combination** of others, even if no single pair is highly correlated | `Savings ≈ Income − Expenses` |

### Why does it happen?
- Same information in different units (cm & m, kg & lb).
- Derived features included together with their parts (`Total_Price`, `Price`, `Quantity`).
- Naturally related measurements (age, experience, years in company).
- Dummy variable trap (Section 15).
- Too many polynomial/interaction features (Topic 11).

### 2. Why does it matter in machine learning?
- The model cannot decide **how to split credit** between the redundant features → unstable, strange coefficients.
- Wasted features: more columns, no new information.
- Harder to explain the model.
(Details in Section 11.)

````python
n = 50
age = rng.integers(20, 60, n)
redundant = pd.DataFrame({
    'Age': age,
    'Age_in_months': age * 12,
    'Age_in_days': age * 365,
    'Height_cm': rng.normal(170, 8, n).round(1),
})
redundant['Height_m'] = redundant['Height_cm'] / 100
redundant['Monthly_Spend'] = rng.normal(40000, 8000, n).round(0)

print(redundant.corr().round(3))
````

````python
plt.figure(figsize=(6.5, 5))
sns.heatmap(redundant.corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Redundant features show up as blocks of 1.00')
plt.show()
````

### 8. Expected output
Blocks of **1.00**: {Age, Age_in_months, Age_in_days} and {Height_cm, Height_m}. Each block is really **one** feature.

### 9. Common mistakes
- Confusing multicollinearity (feature ↔ feature) with a strong feature ↔ target correlation (which is **good**!).
- Thinking only *pairs* matter — a feature can be predictable from a **combination** of several others (VIF catches this, Section 13).

# 11. Why Multicollinearity Is a Problem

### 1. What it means — in Linear Regression
Linear regression learns one **coefficient** per feature:

```text
Salary = b0 + b1 × Age + b2 × Experience
```

`b1` is meant to be *"how much salary changes when Age increases by 1, **keeping Experience fixed**."*
But if Age and Experience always move together, "keep Experience fixed while Age changes" almost **never happens in the data**. The model has very little information to separate their effects.

### Impact on each area

| Area | What goes wrong |
|---|---|
| **Coefficients** | Can become very large, very small, or even have the **wrong sign** (e.g. more experience → *lower* salary) |
| **Standard errors** | Become **inflated** (large) → the model is uncertain about each coefficient |
| **p-values / significance** | Important features can look "not significant" |
| **Interpretability** | You can't trust "the effect of X" statements |
| **Feature importance** | Importance gets split randomly between the correlated features |
| **Model stability** | Small changes in data → big changes in coefficients |
| **Predictions** | Often still **fine** on similar data! Multicollinearity mostly hurts *interpretation*, not always *accuracy* |

### Logistic Regression
The same problem: its coefficients (log-odds) become unstable and hard to interpret, and in extreme cases the solver may struggle to converge.

### When does it matter most?
- When you **interpret coefficients** (economics, medicine, business reports: "each extra year of experience adds X rupees").
- When you need **stable** models over time.
- Linear models without regularization.

### Tree-based models (Decision Tree, Random Forest, XGBoost)
- They don't compute coefficients, so the "unstable coefficient" problem does not apply.
- **Predictions** are usually fine with correlated features.
- **But** feature importance is still split between correlated features — e.g. Age and Age_in_months each look "half as important" — so interpretation can still be misleading.

### Demonstration: unstable coefficients
We fit the same model on **different random samples** of the same population and watch the coefficients.

````python
from sklearn.linear_model import LinearRegression

def make_sample(seed, n=60):
    r = np.random.default_rng(seed)
    age = r.uniform(22, 60, n)
    experience = age - 22 + r.normal(0, 1.0, n)          # almost the same information as age
    salary = 20000 + 2000 * experience + r.normal(0, 8000, n)
    return pd.DataFrame({'Age': age, 'Experience': experience, 'Salary': salary})

rows = []
for seed in range(8):
    d = make_sample(seed)
    both = LinearRegression().fit(d[['Age', 'Experience']], d['Salary'])
    single = LinearRegression().fit(d[['Experience']], d['Salary'])
    rows.append({'sample': seed,
                 'coef_Age (both)': both.coef_[0],
                 'coef_Experience (both)': both.coef_[1],
                 'coef_Experience (alone)': single.coef_[0]})

coef_table = pd.DataFrame(rows).round(0)
print('Correlation Age vs Experience in one sample:', round(make_sample(0).corr().loc['Age', 'Experience'], 3))
coef_table
````

````python
print('Standard deviation of coefficients across samples (lower = more stable):')
print(coef_table.drop(columns='sample').std().round(0))
````

### 8. Expected output — what you should see
- True effect: +2000 rupees per year of experience.
- **With both features:** `coef_Age` and `coef_Experience` jump around a lot between samples, sometimes one is **negative**. Their *sum* stays around 2000 — the model just can't decide how to split it.
- **Experience alone:** the coefficient is stable and close to 2000.

### Inflated standard errors (with statsmodels)
`statsmodels` prints a regression summary with **standard errors** (`std err`) — how uncertain each coefficient is.

````python
import statsmodels.api as sm

d = make_sample(0)
X_both = sm.add_constant(d[['Age', 'Experience']])      # add_constant adds the intercept column
X_one = sm.add_constant(d[['Experience']])

res_both = sm.OLS(d['Salary'], X_both).fit()
res_one = sm.OLS(d['Salary'], X_one).fit()

compare = pd.DataFrame({
    'coef (both)': res_both.params, 'std err (both)': res_both.bse,
    'coef (Exp only)': res_one.params, 'std err (Exp only)': res_one.bse
}).round(1)
compare
````

The standard error of `Experience` is **many times larger** when `Age` is also in the model → the coefficient is far less certain. That is exactly what VIF (Section 13) measures: VIF ≈ (how much the variance of the coefficient is inflated).

### 9. Common mistakes
- Thinking multicollinearity always ruins predictions — often it doesn't; it ruins **interpretation and stability**.
- Ignoring it in a model whose coefficients will be reported to managers.

# 12. Detecting Multicollinearity Using Correlation

### 1. What it means
The simplest first check: compute the correlation matrix of the **features** and look for pairs with a high absolute correlation (a common threshold is |r| > 0.8 or 0.9).

### 4. When to use
As a **quick first screen** during EDA.

### 5. Why correlation alone is not always enough
- It only looks at **pairs**. A feature can be an almost exact combination of **several** others while every pairwise correlation looks moderate.
- Example: `Savings = Income − Expenses`. Savings might only have r ≈ 0.6 with Income and r ≈ −0.3 with Expenses, but together Income and Expenses predict Savings **perfectly**.
- → That's why we also need **VIF** (Section 13).

````python
n = 60
house = pd.DataFrame({'Area_sqft': rng.normal(1800, 400, n).round(0)})
house['Rooms'] = (house['Area_sqft'] / 350 + rng.normal(0, 0.5, n)).round(0)
house['Bathrooms'] = (house['Rooms'] / 2 + rng.normal(0, 0.5, n)).round(0).clip(1)
house['Area_sqm'] = (house['Area_sqft'] * 0.0929).round(1)            # same info, different unit
house['Age_years'] = rng.integers(0, 40, n)
house['Distance_km'] = rng.uniform(1, 30, n).round(1)
house['Price'] = (50 * house['Area_sqft'] + 300000 * house['Bathrooms']
                  - 15000 * house['Age_years'] - 20000 * house['Distance_km'] + rng.normal(0, 60000, n)).round(0)

features = house.drop(columns='Price')
corr = features.corr()

plt.figure(figsize=(7, 5.5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1,
            mask=np.triu(np.ones_like(corr, dtype=bool), k=1))
plt.title('Feature correlation matrix (house data)')
plt.show()
````

````python
def high_corr_pairs(df, threshold=0.8):
    '''Return feature pairs whose absolute correlation is above the threshold.'''
    corr = df.corr(numeric_only=True).abs()
    upper = corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1))   # keep each pair once
    pairs = upper.stack().reset_index()
    pairs.columns = ['Feature_1', 'Feature_2', 'Abs_Correlation']
    return pairs[pairs['Abs_Correlation'] > threshold].sort_values('Abs_Correlation', ascending=False)

high_corr_pairs(features, threshold=0.8).round(3)
````

### Line-by-line (`high_corr_pairs`)
- `.abs()` → treat −0.9 as strongly as +0.9.
- `np.triu(..., k=1)` → True above the diagonal only, so each pair appears **once** and self-correlations (1.0) are skipped.
- `.stack()` → turns the matrix into a long list of (row, column, value).
- Filter by the threshold and sort.

### 8. Expected output
- `Area_sqft` ↔ `Area_sqm` = **1.00** (perfect redundancy).
- `Area_sqft` ↔ `Rooms` high (≈ 0.9).
- `Rooms` ↔ `Bathrooms` moderately high.

### The limitation: a hidden combination

````python
n = 80
fin = pd.DataFrame({
    'Income': rng.normal(100000, 20000, n),
    'Expenses': rng.normal(60000, 15000, n),
})
fin['Savings'] = fin['Income'] - fin['Expenses'] + rng.normal(0, 500, n)   # almost exact combination

print('Pairwise correlations:')
print(fin.corr().round(2))
print('\nHigh pairs (> 0.8):')
print(high_corr_pairs(fin, 0.8))
````

No pair is above 0.8 — correlation screening says "all fine". But `Savings` is almost perfectly predictable from `Income − Expenses`. We will catch this with VIF in the next section.

### 9. Common mistakes
- Using only the correlation matrix and declaring "no multicollinearity".
- Including the **target** in the multicollinearity check (the target *should* correlate with the features).

# 13. Variance Inflation Factor (VIF)

### 1. What it means

> **Analogy:** For each witness, ask: *"Could the other witnesses have told me everything this witness said?"* If yes, this witness is redundant. VIF gives each feature a **redundancy score**.

**VIF** measures how much a feature can be **predicted from all the other features together** — and therefore how much the **variance (uncertainty) of its coefficient is inflated** by multicollinearity.

### 3. How it works (mathematical intuition)
For each feature *j*:
1. Fit a linear regression that predicts feature *j* **using all other features**.
2. Get the R² of that regression (how well the others explain feature *j*, from 0 to 1).
3. Compute:

$$\text{VIF}_j = \frac{1}{1 - R_j^2}$$

| R² of predicting the feature from others | VIF | Meaning |
|---|---|---|
| 0.00 | 1.0 | Completely independent — no multicollinearity |
| 0.50 | 2.0 | Some overlap |
| 0.80 | 5.0 | High overlap |
| 0.90 | 10.0 | Very high overlap |
| 0.99 | 100 | Almost totally redundant |
| 1.00 | ∞ | Perfect multicollinearity |

**Why "inflation factor"?** VIF = 5 means the variance of that coefficient is **5× larger** than it would be if the feature were unrelated to the others (its standard error is √5 ≈ 2.2× larger).

### Common interpretation thresholds (rules of thumb)

| VIF | Interpretation |
|---|---|
| **1** | No multicollinearity |
| **1 – 5** | Moderate — usually OK |
| **5 – 10** | High — investigate |
| **> 10** | Serious multicollinearity — usually act |
| **∞ / huge** | Perfect multicollinearity (duplicate or exact combination) |

### 6. Numerical example
If `Experience` can be predicted from `Age` and `Years_in_company` with R² = 0.95:
VIF = 1 / (1 − 0.95) = 1 / 0.05 = **20** → serious.

### ⚠️ Important detail: the intercept (constant)
`variance_inflation_factor` from statsmodels expects the feature matrix to **include a constant column**. Without it, VIFs can be badly wrong (inflated for features with large means). We add it with `sm.add_constant` and then skip the constant's own VIF.

````python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

def calculate_vif(df):
    '''
    Calculate the Variance Inflation Factor for every column of a numeric DataFrame.
    Returns a DataFrame sorted from highest VIF to lowest.
    '''
    X = df.select_dtypes(include='number').dropna()      # VIF needs numeric data without missing values
    X = sm.add_constant(X, has_constant='add')            # add intercept column named 'const'

    vif = pd.DataFrame({
        'Feature': X.columns,
        'VIF': [variance_inflation_factor(X.values, i)    # VIF of column i vs all other columns
                for i in range(X.shape[1])]
    })

    vif = vif[vif['Feature'] != 'const']                  # the constant's VIF is not meaningful
    return vif.sort_values('VIF', ascending=False).reset_index(drop=True)

calculate_vif(fin)
````

### Line-by-line explanation of `calculate_vif`
1. `df.select_dtypes(include='number')` → VIF only works with numbers; text columns are removed.
2. `.dropna()` → the regression inside VIF cannot handle missing values.
3. `sm.add_constant(X, has_constant='add')` → adds a column of 1s called `const` (the intercept), which statsmodels' VIF formula assumes is present.
4. `variance_inflation_factor(X.values, i)` → for column `i`, regress it on all other columns and return `1 / (1 − R²)`.
5. The list comprehension does this for every column.
6. We drop the `const` row and sort so the worst features appear first.

### 8. Expected output
`Income`, `Expenses`, and `Savings` all have **huge VIFs** (hundreds or more) — VIF found the hidden combination that the pairwise correlation check missed!

### Checking our manual formula

````python
# Manual VIF for 'Savings': regress Savings on Income and Expenses
X_others = fin[['Income', 'Expenses']]
r2 = LinearRegression().fit(X_others, fin['Savings']).score(X_others, fin['Savings'])
print('R² of Savings ~ Income + Expenses:', round(r2, 5))
print('Manual VIF = 1 / (1 - R²)          :', round(1 / (1 - r2), 1))
````

### VIF on the house dataset

````python
calculate_vif(features).round(2)
````

- `Area_sqft` and `Area_sqm` → **huge** VIFs (they are the same thing).
- `Rooms` → elevated (strongly related to area).
- `Age_years`, `Distance_km` → ≈ 1 (independent).

### Iteratively removing the worst feature
A common approach: remove the feature with the highest VIF, recalculate, repeat until all VIFs are below a threshold. **But always check with domain knowledge** which feature you'd prefer to keep.

````python
def reduce_vif(df, threshold=5.0, verbose=True):
    '''Repeatedly drop the feature with the highest VIF until all VIFs are <= threshold.'''
    current = df.copy()
    while True:
        vif = calculate_vif(current)
        worst = vif.iloc[0]
        if worst['VIF'] <= threshold:
            break
        if verbose:
            print(f"Dropping {worst['Feature']:<12} (VIF = {worst['VIF']:.1f})")
        current = current.drop(columns=worst['Feature'])
    return current, vif

reduced, final_vif = reduce_vif(features, threshold=5)
print('\nRemaining features:', list(reduced.columns))
final_vif.round(2)
````

**Notice:** `Area_sqft` and `Area_sqm` had (almost) the same VIF, so the algorithm dropped `Area_sqft` and kept `Area_sqm` — a basically **arbitrary** choice. If your business works in square feet, you would keep `Area_sqft` instead. The algorithm finds *what* is redundant; **you** decide *which* one to keep.

### 4. When to use VIF
- Before fitting linear/logistic regression when you will interpret coefficients.
- To find redundancy that involves **more than two** features.

### 5. When NOT to rely on VIF
- For categorical variables encoded as many dummies (VIF of individual dummies can mislead; consider the group as a whole — advanced "GVIF").
- For tree models focused only on prediction (less important).
- Blindly: VIF doesn't know which feature is more meaningful or cheaper to collect.

### 9. Common mistakes
- Forgetting to add a constant → misleading VIFs.
- Including the **target** in the VIF calculation.
- Dropping all high-VIF features at once — drop one, recalculate (dropping one often fixes the others).
- Computing VIF on dummy columns including **all** categories (perfect multicollinearity → infinite VIF; see Section 15).

# 14. Fixing Multicollinearity

| # | Approach | Idea | Good when |
|---|---|---|---|
| 1 | **Remove a feature** | Keep one of the redundant features | Clear duplicates (cm & m) |
| 2 | **Combine features** | Average or sum related features into one | Several measures of the same concept |
| 3 | **Feature engineering** | Replace with a ratio/difference that carries new information | `Age` & `Experience` → `Age_at_Start = Age − Experience` |
| 4 | **PCA** | Transform correlated features into uncorrelated components | Many correlated numeric features; interpretation less important |
| 5 | **Regularization** | Ridge/Lasso penalize large coefficients → stabilize them | You want to keep all features for prediction |
| 6 | **Domain knowledge** | Decide which feature is most meaningful, reliable, or available | Always! |

### 1. Remove one highly correlated feature
```text
Height_cm
Height_m    ← exactly Height_cm / 100
```
Keeping both adds **zero** new information and creates perfect multicollinearity.

````python
n = 60
body = pd.DataFrame({'Height_cm': rng.normal(170, 9, n)})
body['Height_m'] = body['Height_cm'] / 100
body['Weight_kg'] = 0.9 * body['Height_cm'] - 85 + rng.normal(0, 7, n)
body['Waist_cm'] = 0.5 * body['Weight_kg'] + 45 + rng.normal(0, 4, n)
body['Hip_cm'] = 0.9 * body['Waist_cm'] + 20 + rng.normal(0, 2, n)

print('Before:')
print(calculate_vif(body).round(1))

body_fixed = body.drop(columns='Height_m')
print('\nAfter removing Height_m:')
print(calculate_vif(body_fixed).round(1))
````

(The VIF of perfectly collinear columns shows as a gigantic number or `inf`.)

### 2. Combine features
`Waist_cm` and `Hip_cm` are strongly related. Instead of both, we can create **one** combined feature:

````python
combo = body_fixed.copy()
combo['Waist_Hip_Avg'] = (combo['Waist_cm'] + combo['Hip_cm']) / 2
combo = combo.drop(columns=['Waist_cm', 'Hip_cm'])
calculate_vif(combo).round(2)
````

### 3. Feature engineering — replace with a more informative feature
`Waist_to_Hip_Ratio` is a well-known health indicator (domain knowledge!) and is much less correlated with the other features than waist or hip alone.

````python
eng = body_fixed.copy()
eng['Waist_to_Hip'] = eng['Waist_cm'] / eng['Hip_cm']
eng = eng.drop(columns=['Hip_cm'])
print(calculate_vif(eng).round(2))

# Age & Experience example
ae = make_sample(1)
ae['Age_at_Career_Start'] = ae['Age'] - ae['Experience']
print('\nVIF with Age + Experience:')
print(calculate_vif(ae[['Age', 'Experience']]).round(1))
print('\nVIF with Experience + Age_at_Career_Start:')
print(calculate_vif(ae[['Experience', 'Age_at_Career_Start']]).round(2))
````

### 4. PCA (Principal Component Analysis) — short introduction

> **Analogy:** Instead of three witnesses repeating the same story, you write one **summary** of the story (component 1), then note any **extra detail** (component 2).

PCA creates new features (**principal components**) that are combinations of the originals and are **completely uncorrelated** with each other. Downside: components are hard to interpret ("PC1 = 0.6 × Height + 0.5 × Weight + …"). Always **scale** before PCA. (Full PCA is a future topic.)

````python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X_body = body_fixed[['Height_cm', 'Weight_kg', 'Waist_cm', 'Hip_cm']]
X_scaled = StandardScaler().fit_transform(X_body)       # PCA needs scaled data

pca = PCA(n_components=2)
pcs = pd.DataFrame(pca.fit_transform(X_scaled), columns=['PC1', 'PC2'])

print('Share of information (variance) kept by each component:', pca.explained_variance_ratio_.round(3))
print('Total kept:', pca.explained_variance_ratio_.sum().round(3))
print('\nCorrelation between components:')
print(pcs.corr().round(3))
print('\nVIF of components:')
print(calculate_vif(pcs).round(2))
````

Two uncorrelated components keep most of the information of 4 correlated features, and their VIFs are exactly 1.

### 5. Regularization (Ridge)

**Ridge regression** adds a penalty for large coefficients. When two features are redundant, Ridge **shares** the effect between them in a stable way instead of producing wild, opposite coefficients. (**Lasso** can push one of them to exactly 0 — automatic feature selection.) Scale features first.

````python
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline

rows = []
for seed in range(6):
    d = make_sample(seed)
    Xs = d[['Age', 'Experience']]
    ols = make_pipeline(StandardScaler(), LinearRegression()).fit(Xs, d['Salary'])
    ridge = make_pipeline(StandardScaler(), Ridge(alpha=10)).fit(Xs, d['Salary'])
    rows.append({'sample': seed,
                 'OLS_Age': ols[-1].coef_[0], 'OLS_Exp': ols[-1].coef_[1],
                 'Ridge_Age': ridge[-1].coef_[0], 'Ridge_Exp': ridge[-1].coef_[1]})

reg_table = pd.DataFrame(rows).round(0)
print(reg_table)
print('\nStd of coefficients across samples:')
print(reg_table.drop(columns='sample').std().round(0))
````

(Coefficients here are "per 1 standard deviation" because we scaled.) Ordinary least squares (OLS) coefficients swing wildly; Ridge coefficients are much more stable.

### 6. Domain knowledge
Ask:
- Which feature is **more meaningful** for the business question?
- Which is **more reliable / less noisy**?
- Which will be **available at prediction time**?
- Which is **cheaper** to collect?

Example: between `Age` and `Experience` for salary prediction, `Experience` is usually the more direct cause of salary → keep `Experience`.

### 9. Common mistakes
- Removing a feature just because it correlates with the **target** (that is the useful kind!).
- Using PCA when stakeholders need to understand the features.
- Applying Ridge/Lasso without scaling.

# 15. The Dummy Variable Trap

### 1. What it means
When we one-hot encode a categorical column (Topic 6), each category becomes a 0/1 column:

```text
Gender   →  Gender_Male  Gender_Female
Male              1            0
Female            0            1
Male              1            0
```

Notice: **`Gender_Male + Gender_Female = 1` for every row.** One column is perfectly predictable from the other:

```text
Gender_Female = 1 − Gender_Male
```

A linear model with an **intercept** already has a hidden column of all 1s. So:

```text
Intercept column (1)  =  Gender_Male + Gender_Female
```

That is **perfect multicollinearity** — the "dummy variable trap". The model cannot find a unique set of coefficients.

### 3. How `drop_first=True` fixes it
Drop one dummy column (the **reference / baseline** category). The dropped category is represented when all remaining dummies are 0.

```text
Colour: Red, Green, Blue  →  keep Colour_Green, Colour_Red (drop Blue)
Blue  = (0, 0)   ← baseline
Green = (1, 0)
Red   = (0, 1)
```

The coefficients are then interpreted **relative to the baseline**: "Green customers spend X more than Blue customers."

`pd.get_dummies` drops the **first category in alphabetical order** (for normal text columns).

````python
dummy_df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore', 'Karachi', 'Lahore'],
    'Spend': [500, 700, 450, 800, 600, 750]
})

all_dummies = pd.get_dummies(dummy_df, columns=['Gender', 'City'], dtype=int)
print('ALL dummies (trap!):')
print(all_dummies)
print('\nGender_Male + Gender_Female =', (all_dummies['Gender_Male'] + all_dummies['Gender_Female']).tolist())
print('Sum of City dummies        =', all_dummies.filter(like='City_').sum(axis=1).tolist())
````

````python
print('VIF with ALL dummies (expect inf / huge values):')
print(calculate_vif(all_dummies.drop(columns='Spend')).round(2))
````

````python
safe_dummies = pd.get_dummies(dummy_df, columns=['Gender', 'City'], drop_first=True, dtype=int)
print('drop_first=True:')
print(safe_dummies)
print('\nVIF after drop_first=True:')
print(calculate_vif(safe_dummies.drop(columns='Spend')).round(2))
````

### 8. Expected output
- With all dummies: VIFs are `inf` (or astronomically large).
- With `drop_first=True`: `Gender_Female` and `City_Islamabad` were dropped (first alphabetically); VIFs are finite.
- Reading the result: `Gender_Male = 0` means Female (the baseline); `City_Karachi = 0` and `City_Lahore = 0` means Islamabad (the baseline).

### 4. When to use `drop_first=True`
- Linear regression / logistic regression **with an intercept**, especially when interpreting coefficients.

### 5. When it's not necessary
- **Tree-based models** — they don't have this problem; keeping all dummies can even be clearer.
- **Regularized models** (Ridge/Lasso) — the penalty makes a unique solution possible; many practitioners keep all dummies.
- scikit-learn's `LinearRegression` won't crash with all dummies (it uses a least-squares solver that picks one solution), but the coefficients are then **not uniquely meaningful**.

### scikit-learn equivalent
```python
OneHotEncoder(drop='first')        # like drop_first=True
OneHotEncoder(drop='if_binary')    # drop one column only for 2-category features
```

### 9. Common mistakes
- Using `drop_first=True` separately on train and test data that have different categories → different columns! Use `OneHotEncoder` fitted on train (Topic 6).
- Forgetting which category is the baseline when explaining coefficients.
- Thinking `drop_first=True` loses information — it doesn't; the dropped category is "all zeros".

# 16. Multicollinearity with a Real ML Model

We build an employee dataset with **intentionally correlated** features:
- `Age`, `Experience`, `Years_in_company` — all strongly related
- `Experience_months` — an exact copy of Experience × 12 (perfect redundancy)
- `Education`, `Department` — categorical

Then:
1. Train Linear Regression with **all** features → look at coefficients, performance, VIF.
2. Remove redundant features → train again → compare.

````python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error

n = 200
r16 = np.random.default_rng(16)
experience = r16.uniform(0, 30, n)
emp = pd.DataFrame({
    'Age': (22 + experience + r16.normal(0, 1.5, n)).round(0),
    'Experience': experience.round(1),
    'Education': r16.choice(['Bachelor', 'Master', 'PhD'], n, p=[0.5, 0.35, 0.15]),
    'Department': r16.choice(['IT', 'HR', 'Finance', 'Sales'], n),
})
emp['Years_in_company'] = (emp['Experience'] * r16.uniform(0.2, 1.0, n)).round(1)
emp['Experience_months'] = emp['Experience'] * 12

edu_bonus = emp['Education'].map({'Bachelor': 0, 'Master': 15000, 'PhD': 35000})
dept_bonus = emp['Department'].map({'IT': 20000, 'Finance': 15000, 'Sales': 5000, 'HR': 0})
emp['Salary'] = (40000 + 3000 * emp['Experience'] + 500 * emp['Years_in_company']
                 + edu_bonus + dept_bonus + r16.normal(0, 8000, n)).round(0)

emp.head()
````

````python
def fit_and_report(df, feature_cols, label):
    '''One-hot encode, split, fit LinearRegression, return metrics and coefficients.'''
    X = pd.get_dummies(df[feature_cols], drop_first=True, dtype=int)   # avoid the dummy trap
    y = df['Salary']
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

    model = LinearRegression().fit(X_train, y_train)
    pred = model.predict(X_test)

    metrics = {'Model': label,
               'Train_R2': r2_score(y_train, model.predict(X_train)),
               'Test_R2': r2_score(y_test, pred),
               'Test_MAE': mean_absolute_error(y_test, pred),
               'n_features': X.shape[1]}
    coefs = pd.Series(model.coef_, index=X.columns, name=label)
    return metrics, coefs, X

cat_cols = ['Education', 'Department']
all_features = numeric_cols + cat_cols
clean_features = ['Experience', 'Years_in_company'] + cat_cols      # remove Age and Experience_months

m_all, c_all, X_all = fit_and_report(emp, all_features, 'All features')
m_clean, c_clean, X_clean = fit_and_report(emp, clean_features, 'Redundant removed')

print(pd.DataFrame([m_all, m_clean]).round(3).to_string(index=False))
````

### Coefficient stability check (bootstrap)
We refit both models on 30 random resamples of the data and look at how much the `Experience`-related coefficients move.

````python
def bootstrap_coefs(df, feature_cols, n_boot=30):
    X = pd.get_dummies(df[feature_cols], drop_first=True, dtype=int)
    y = df['Salary']
    rows = []
    rb = np.random.default_rng(0)
    for _ in range(n_boot):
        idx = rb.integers(0, len(X), len(X))              # sample rows WITH replacement
        rows.append(LinearRegression().fit(X.iloc[idx], y.iloc[idx]).coef_)
    return pd.DataFrame(rows, columns=X.columns)

# NOTE: we use Age (strongly but not perfectly correlated) for the 'redundant' model here.
# With an EXACT copy (Experience_months) the solver always picks the same 'minimum-size' split,
# which looks stable but is still meaningless — so Age shows the instability more honestly.
redundant_features = ['Age', 'Experience', 'Years_in_company'] + cat_cols
boot_all = bootstrap_coefs(emp, redundant_features)
boot_clean = bootstrap_coefs(emp, clean_features)

stab = pd.DataFrame({
    'Std (Age + Experience + Years)': boot_all[['Age', 'Experience', 'Years_in_company']].std(),
    'Std (cleaned)': boot_clean[['Experience', 'Years_in_company']].std()
}).round(1)
stab
````

### What changed? (Explanation)

1. **Performance (R², MAE):** almost the **same** for both models (the cleaned model is even slightly better on the test set). Removing redundant features did **not** hurt predictions — the removed columns carried no new information.
2. **Coefficients:**
   - The true effect used to generate the data is **+3000 per year of experience**.
   - With all features, that effect is **split** in an arbitrary way: `Experience` ≈ 14, `Experience_months` ≈ 162 per *month* (≈ 1,940 per year), and `Age` ≈ 1,219. None of these numbers means "the effect of experience". Because `Experience_months` is an *exact* copy, the solver simply picks one of infinitely many equally good splits.
   - After cleaning, `Experience` ≈ +3,160 per year — close to the true 3,000. **Interpretable.** (`Years_in_company` ≈ +300 vs a true +500 — it is a smaller effect estimated with more uncertainty.)
   - Education and department coefficients barely change, because they were never involved in the multicollinearity.
3. **VIF:** from `inf` / 36 → about 3.
4. **Stability:** in the bootstrap, the `Experience` coefficient's standard deviation drops by about **4×** once `Age` is removed.
5. **Simpler model:** fewer features, same accuracy.

> **Key lesson:** multicollinearity usually hurts **interpretation and stability** more than **prediction accuracy**.

# 17. Correlation vs Multicollinearity — Comparison

| Concept | Meaning | Range | Used For |
|---|---|---|---|
| **Covariance** | Whether two variables vary together (direction), in their original units | −∞ to +∞ | Building block for correlation, PCA, finance |
| **Pearson correlation** | Strength & direction of a **linear** relationship between two numeric variables | −1 to +1 | EDA, feature–target relationships, detecting redundant pairs |
| **Spearman correlation** | Strength & direction of a **monotonic** relationship (uses ranks) | −1 to +1 | Curved-but-monotonic data, ordinal data, outliers |
| **Kendall's tau** | Agreement of pair orderings | −1 to +1 | Small samples, many ties, ordinal data |
| **Point-biserial** | Relationship between a binary and a numeric variable | −1 to +1 | Yes/No vs numeric |
| **Cramér's V** | Strength of association between two categorical variables | 0 to 1 | Category vs category |
| **Multicollinearity** | **Features** (inputs) that are highly related to each other → redundant information | — (a situation, not a number) | Diagnosing unstable, uninterpretable linear models |
| **VIF** | How well a feature is predicted by **all other features**; how much its coefficient variance is inflated | 1 to ∞ | Measuring multicollinearity, including multi-feature combinations |

### The key difference in one sentence
- **Correlation** is a **measurement** between two variables (any two, including feature ↔ target).
- **Multicollinearity** is a **problem** in a model when **features** overlap; correlation is one way to **detect** it, VIF is a better way.

```text
Feature ↔ Target high correlation     → GOOD (useful feature)
Feature ↔ Feature high correlation    → possible MULTICOLLINEARITY (redundant features)
```

# 18. Common Mistakes

| # | Mistake | Why it's wrong | Do this instead |
|---|---|---|---|
| 1 | **Assuming correlation means causation** | Confounders, coincidence, reverse causation | Use experiments, control confounders, domain logic (Section 8) |
| 2 | **Only looking at the correlation number** | Curves, clusters and outliers hide behind one number (Anscombe) | Always make scatter plots / pair plots |
| 3 | **Ignoring outliers** | One point can create or destroy a correlation | Plot, investigate, compare Pearson vs Spearman (Section 9) |
| 4 | **Removing every correlated feature** | Feature–target correlation is useful; moderate feature–feature correlation is normal | Remove only truly redundant features; check VIF + domain knowledge |
| 5 | **Using Pearson on categorical variables** | Label codes (1, 2, 3) are arbitrary | Point-biserial, chi-square, Cramér's V, group means (Section 7) |
| 6 | **Thinking correlation always detects multicollinearity** | It only checks pairs; combinations can be hidden (Savings = Income − Expenses) | Use VIF (Section 13) |
| 7 | **Using VIF without understanding the features** | VIF says *what* is redundant, not *which* to keep | Combine VIF with domain knowledge |
| 8 | **Blindly dropping features** | You may drop the more meaningful one or remove useful signal | Drop one at a time, recalc VIF, compare model performance |
| 9 | Including the **target** in the VIF/multicollinearity check | Target *should* relate to features | Only check features |
| 10 | Forgetting the constant in statsmodels VIF | Gives misleading VIFs | `sm.add_constant` (as in `calculate_vif`) |
| 11 | Worrying about multicollinearity for pure-prediction tree models | Trees don't use coefficients | Focus on it when interpreting linear models |
| 12 | Keeping all dummy columns in linear regression with an intercept | Dummy variable trap | `drop_first=True` or `OneHotEncoder(drop='first')` |
| 13 | Treating r = 0 as "no relationship" | Could be a U-shape | Look at the scatter plot |
| 14 | Treating a small p-value as a strong relationship | Large samples make tiny r significant | Look at the size of r too |

````python
p1 = pd.DataFrame({'Ad_Spend': [2, 4, 6, 8, 10], 'Sales': [20, 26, 35, 41, 48]})

x_dev = p1['Ad_Spend'] - p1['Ad_Spend'].mean()       # mean = 6
y_dev = p1['Sales'] - p1['Sales'].mean()             # mean = 34
products = x_dev * y_dev
cov_manual = products.sum() / (len(p1) - 1)

std_x = np.sqrt((x_dev ** 2).sum() / (len(p1) - 1))
std_y = np.sqrt((y_dev ** 2).sum() / (len(p1) - 1))
r_manual = cov_manual / (std_x * std_y)

print(pd.DataFrame({'x_dev': x_dev, 'y_dev': y_dev, 'product': products}))
print('\nManual covariance:', cov_manual, '| Pandas:', p1['Ad_Spend'].cov(p1['Sales']))
print('Manual Pearson r :', round(r_manual, 4), '| Pandas:', round(p1['Ad_Spend'].corr(p1['Sales']), 4))
````

````python
r5 = np.random.default_rng(5)
n = 150
p5 = pd.DataFrame({'Income': r5.normal(80000, 15000, n)})
p5['Monthly_Income'] = p5['Income'] / 12
p5['Rent'] = 0.3 * p5['Income'] + r5.normal(0, 3000, n)
p5['Car_Value'] = 0.8 * p5['Income'] + r5.normal(0, 20000, n)
p5['Age'] = r5.integers(20, 65, n)
p5['Credit_Score'] = r5.normal(650, 60, n)

def find_high_corr(df, threshold=0.8):
    corr = df.corr(numeric_only=True)
    pairs = []
    cols = corr.columns
    for i in range(len(cols)):
        for j in range(i + 1, len(cols)):          # j > i → each pair once, no self-pairs
            r = corr.iloc[i, j]
            if abs(r) > threshold:
                pairs.append((cols[i], cols[j], round(r, 3)))
    out = pd.DataFrame(pairs, columns=['Feature_1', 'Feature_2', 'r'])
    return out.reindex(out['r'].abs().sort_values(ascending=False).index).reset_index(drop=True)

find_high_corr(p5, 0.7)
````

````python
r6 = np.random.default_rng(6)
p6 = pd.DataFrame({'Temperature': r6.uniform(20, 35, 25)})
p6['Electricity_Units'] = 10 * p6['Temperature'] + r6.normal(0, 15, 25)

def both_corr(d):
    return (round(float(d['Temperature'].corr(d['Electricity_Units'])), 3),
            round(float(d['Temperature'].corr(d['Electricity_Units'], method='spearman')), 3))

print('Clean      (Pearson, Spearman):', both_corr(p6))
p6_bad = pd.concat([p6, pd.DataFrame({'Temperature': [30], 'Electricity_Units': [5000]})], ignore_index=True)
print('With error (Pearson, Spearman):', both_corr(p6_bad))

# IQR rule on Electricity_Units
q1, q3 = p6_bad['Electricity_Units'].quantile([0.25, 0.75])
iqr = q3 - q1
keep = p6_bad['Electricity_Units'].between(q1 - 1.5 * iqr, q3 + 1.5 * iqr)
p6_fixed = p6_bad[keep]
print('After IQR  (Pearson, Spearman):', both_corr(p6_fixed), '| rows removed:', (~keep).sum())

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].scatter(p6['Temperature'], p6['Electricity_Units']); ax[0].set_title(f'Clean: r = {both_corr(p6)[0]}')
ax[1].scatter(p6_bad['Temperature'], p6_bad['Electricity_Units'])
ax[1].scatter([30], [5000], color='red', s=100); ax[1].set_title(f'With outlier: r = {both_corr(p6_bad)[0]}')
plt.tight_layout(); plt.show()
````

- After removing one income column, all VIFs are ≤ 5, so `Rent` survives here. With a stricter threshold (e.g. 3) it would be dropped, even though it may be a genuinely useful feature. Converting it to `Rent_to_Income` (what share of income goes to rent) keeps its information **and** removes the overlap with income — every VIF drops to ≈ 1–1.5.

### Solution 10

````python
r10 = np.random.default_rng(10)
n = 250
area = r10.normal(1500, 350, n)
p10 = pd.DataFrame({
    'Area_sqft': area.round(0),
    'Area_sqm': (area * 0.0929).round(1),
    'Rooms': (area / 300 + r10.normal(0, 0.6, n)).round(0),
    'Age_years': r10.integers(0, 40, n),
    'Floor': r10.integers(0, 15, n),
    'City': r10.choice(['Lahore', 'Karachi', 'Islamabad'], n),
    'Has_Parking': r10.choice(['Yes', 'No'], n),
})
city_effect = p10['City'].map({'Lahore': 0, 'Karachi': 1_000_000, 'Islamabad': 1_500_000})
p10['Price'] = (4000 * p10['Area_sqft'] - 40000 * p10['Age_years'] + 30000 * p10['Floor']
                + city_effect + np.where(p10['Has_Parking'] == 'Yes', 500000, 0)
                + r10.normal(0, 400000, n)).round(-3)

# 1. Explore
num10 = ['Area_sqft', 'Area_sqm', 'Rooms', 'Age_years', 'Floor']
plt.figure(figsize=(6.5, 5))
sns.heatmap(p10[num10 + ['Price']].corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('House data correlations'); plt.show()

print("Cramér's V (City vs Has_Parking):", round(cramers_v(p10['City'], p10['Has_Parking']), 3))
r_pb10, _ = stats.pointbiserialr((p10['Has_Parking'] == 'Yes').astype(int), p10['Price'])
print('Point-biserial (Has_Parking vs Price):', round(r_pb10, 3))

# 2. VIF
print('\nVIF (numeric features):')
print(calculate_vif(p10[num10]).round(2))
````

````python
# 3-4. Encode + train before/after
def run(cols, label):
    X = pd.get_dummies(p10[cols], drop_first=True, dtype=int)        # drop_first avoids the dummy trap
    y = p10['Price']
    X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=42)
    m = LinearRegression().fit(X_tr, y_tr)
    pred = m.predict(X_te)
    return ({'Model': label, 'Test_R2': round(r2_score(y_te, pred), 4),
             'Test_MAE': round(mean_absolute_error(y_te, pred)), 'n_features': X.shape[1]},
            pd.Series(m.coef_, index=X.columns, name=label).round(0))

before_cols = num10 + ['City', 'Has_Parking']
after_cols = ['Area_sqft', 'Age_years', 'Floor', 'City', 'Has_Parking']   # dropped Area_sqm and Rooms

m_b, c_b = run(before_cols, 'Before')
m_a, c_a = run(after_cols, 'After')
print(pd.DataFrame([m_b, m_a]).to_string(index=False), '\n')
print(pd.concat([c_b, c_a], axis=1))
print('\nVIF after:')
print(calculate_vif(p10[['Area_sqft', 'Age_years', 'Floor']]).round(2))
````

**Explanation (5):**
- `Area_sqm` is `Area_sqft` in a different unit (VIF in the hundreds of thousands), and `Rooms` is derived from area (VIF ≈ 4, moderate). `City` and `Has_Parking` are unrelated (Cramér's V ≈ 0). Parking has a small positive point-biserial correlation with price (≈ 0.14) — small because area and city dominate price, but the regression still finds its effect.
- **Test R² and MAE are almost identical** before and after (slightly better after) → the removed columns added no information.
- **Before:** the area effect is split between `Area_sqft` (+126,000!), `Area_sqm` (−1,300,000!) and `Rooms` (negative) — absurd, uninterpretable coefficients.
- **After:** `Area_sqft` ≈ 4,000 per sq ft (true: 4,000), `Age_years` ≈ −38,000 (true: −40,000), `Floor` ≈ +21,500 (true: +30,000 — a small effect estimated with more noise), `City_Karachi` ≈ −500,000 and `City_Lahore` ≈ −1,500,000 relative to the baseline **Islamabad** (dropped because it is first alphabetically), and `Has_Parking_Yes` close to +500,000 — close to the true rules used to generate the data. VIFs are all ≈ 1.

## Step 1 — Load the dataset (50 employees)
In a real project you'd use `pd.read_csv('employees.csv')`. Here we generate a realistic dataset so the notebook runs anywhere.

````python
rp = np.random.default_rng(2025)
n = 50

experience = rp.uniform(0, 28, n).round(1)
age = (23 + experience + rp.normal(0, 2, n)).round(0)
years_in_company = (experience * rp.uniform(0.3, 1.0, n)).round(1)
education = rp.choice(['Bachelor', 'Master', 'PhD'], n, p=[0.5, 0.35, 0.15])
department = rp.choice(['IT', 'Finance', 'HR', 'Sales'], n)
performance = rp.integers(1, 6, n)                                          # rating 1–5
training_hours = rp.integers(0, 120, n)

edu_add = pd.Series(education).map({'Bachelor': 0, 'Master': 12000, 'PhD': 30000}).to_numpy()
dept_add = pd.Series(department).map({'IT': 18000, 'Finance': 14000, 'Sales': 6000, 'HR': 0}).to_numpy()

base_salary = (45000 + 2800 * experience + 600 * years_in_company + 9000 * performance
               + edu_add + dept_add + rp.normal(0, 7000, n)).round(-2)

employees = pd.DataFrame({
    'Employee_ID': [f'E{100 + i}' for i in range(n)],
    'Age': age,
    'Experience': experience,
    'Experience_Months': (experience * 12).round(0),          # redundant copy of Experience
    'Years_in_Company': years_in_company,
    'Education': education,
    'Department': department,
    'Performance_Rating': performance,
    'Training_Hours': training_hours,
    'Monthly_Salary': (base_salary / 12).round(0),            # target in monthly units
})
employees['Annual_Bonus'] = (0.1 * base_salary * performance / 5 + rp.normal(0, 1000, n)).round(0)
# ⚠️ Annual_Bonus is paid AFTER salary is set and is computed from salary → leakage risk (discussed later)

employees.head()
````

## Step 8 — Identify multicollinearity

| Feature | Problem | Evidence |
|---|---|---|
| `Experience_Months` | Exact copy of `Experience` × 12 | r = 1.00, VIF = ∞/huge |
| `Experience` | Duplicated by Experience_Months, and ≈ Age − 23 | VIF = ∞/huge |
| `Age` | Almost fully explained by Experience | r ≈ 0.96, VIF ≈ 13 |
| `Performance_Rating` ↔ `Annual_Bonus` | The bonus is calculated from performance | r ≈ 0.89, VIF ≈ 11 each |
| `Years_in_Company` | Related to Experience but with its own variation | VIF ≈ 4 (acceptable) |
| `Training_Hours` | Independent | VIF ≈ 1 |

### ⚠️ A leakage issue (not multicollinearity, but just as important)
`Annual_Bonus` is **calculated from the salary** and paid afterwards. It correlates with salary, but at prediction time (e.g. making an offer to a new employee) we would **not know it**. It must be removed regardless of VIF (Topic 11 — data leakage).

## Step 9 — Remove / rework redundant features

Decisions (VIF + domain knowledge):
1. Drop `Experience_Months` — pure duplicate.
2. Drop `Age` — salary is driven by experience; age adds little and is also a sensitive attribute in HR decisions.
3. Keep `Experience` and `Years_in_Company` — both meaningful (total vs company-specific experience).
4. Drop `Annual_Bonus` — leakage.
5. Drop `Employee_ID` — identifier.

````python
features_before = numerical_features + categorical_features          # everything (incl. leakage & redundancy)
features_after = ['Experience', 'Years_in_Company', 'Performance_Rating', 'Training_Hours'] + categorical_features

vif_after = calculate_vif(employees[[c for c in features_after if c not in categorical_features]])
print('VIF AFTER cleaning:')
vif_after.round(2)
````

## Step 10 — Train Linear Regression

````python
from sklearn.model_selection import cross_val_score, KFold

def train_eval(cols, label):
    X = pd.get_dummies(employees[cols], columns=categorical_features, drop_first=True, dtype=int)
    y = employees[target]
    X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=7)
    model = LinearRegression().fit(X_tr, y_tr)
    pred = model.predict(X_te)
    cv = cross_val_score(LinearRegression(), X, y, cv=KFold(5, shuffle=True, random_state=0), scoring='r2')
    return ({'Model': label, 'Features': X.shape[1],
             'Train_R2': round(model.score(X_tr, y_tr), 3),
             'Test_R2': round(r2_score(y_te, pred), 3),
             'Test_MAE': round(mean_absolute_error(y_te, pred)),
             'CV_R2_mean': round(cv.mean(), 3)},
            pd.Series(model.coef_, index=X.columns, name=label))

res_before, coef_before = train_eval(features_before, 'Before (all features)')

# Also: 'before' WITHOUT the leakage column, to isolate the multicollinearity effect
features_before_no_leak = [c for c in features_before if c != 'Annual_Bonus']
res_mid, coef_mid = train_eval(features_before_no_leak, 'Redundant kept, no leakage')

res_after, coef_after = train_eval(features_after, 'After (cleaned)')
````

````python
# Coefficient stability: bootstrap the 'redundant kept' and 'cleaned' models
def boot(cols, n_boot=40):
    X = pd.get_dummies(employees[cols], columns=categorical_features, drop_first=True, dtype=int)
    y = employees[target]
    rb = np.random.default_rng(1)
    out = []
    for _ in range(n_boot):
        idx = rb.integers(0, len(X), len(X))
        out.append(LinearRegression().fit(X.iloc[idx], y.iloc[idx]).coef_)
    return pd.DataFrame(out, columns=X.columns)

b_mid = boot(features_before_no_leak)
b_after = boot(features_after)

stability = pd.DataFrame({
    'Std — redundant kept': b_mid[['Age', 'Experience', 'Experience_Months', 'Years_in_Company']].std(),
    'Std — cleaned': b_after[['Experience', 'Years_in_Company']].std(),
}).round(1)
stability
````

## Step 12 — Explain the results

**1. Correlation analysis**
- `Experience`, `Experience_Months`, `Age` and `Years_in_Company` are strongly correlated with salary (r ≈ 0.75–0.80) — but also with **each other** (r ≈ 0.82–1.00).
- `Performance_Rating` and `Training_Hours` have weak positive correlations with salary (≈ 0.19). A weak correlation does **not** mean useless: once experience is accounted for in the regression, performance still has a clear effect.
- PhD > Master > Bachelor, and HR is paid less than the other departments (group means).

**2. Multicollinearity**
- `Experience_Months` is a perfect copy → VIF in the hundreds of thousands.
- `Age` is almost fully explained by experience → VIF ≈ 13.
- `Annual_Bonus` and `Performance_Rating` inflate each other (VIF ≈ 11) because the bonus is computed from performance.
- After cleaning, every VIF is below 4.

**3. Leakage**
- The "all features" model includes `Annual_Bonus`, which is computed from salary. Its score may look slightly better, but that improvement is **fake** — the bonus is not known when predicting salary.

**4. Performance**
- The "redundant kept, no leakage" and "cleaned" models have **very similar** scores; the cleaned model is even slightly better on the test set and in cross-validation. Removing redundant features did not cost accuracy.
- With only 50 rows the single test set has just 13 employees, so its R² jumps around a lot. The **5-fold cross-validated R²** (`CV_R2_mean`) is the more reliable comparison.

**5. Coefficients & stability**
- With redundant features, the experience effect is split in arbitrary ways between `Age`, `Experience`, and `Experience_Months`; bootstrap standard deviations are large and signs can flip.
- After cleaning, the `Experience` coefficient (≈ 200) is close to the true monthly effect (2800/12 ≈ 233 per year of experience), and `Performance_Rating` (≈ 626) is close to its true effect (9000/12 = 750 per rating point). The bootstrap standard deviation of the `Experience` coefficient falls from thousands to about 20 — **far more stable**. `Years_in_Company`, performance, education and department coefficients are interpretable relative to their baselines (Bachelor, Finance — first alphabetically).

**6. Takeaways**
```text
✔ Correlation told us WHAT relates to salary and WHICH features overlap.
✔ VIF confirmed and quantified the redundancy.
✔ Domain knowledge decided WHICH redundant feature to keep.
✔ Removing redundancy kept accuracy and made coefficients trustworthy.
✔ Leakage must be removed even if it "improves" the score.
```

# 21. Cheat Sheet — Correlation & Multicollinearity

| Topic | Key idea | Code |
|---|---|---|
| **Covariance** | Direction of joint variation; depends on units; −∞..+∞ | `df.cov()`, `df['a'].cov(df['b'])` |
| **Pearson r** | Linear relationship; −1..+1; sensitive to outliers | `df.corr(numeric_only=True)`, `df['a'].corr(df['b'])`, `stats.pearsonr` |
| **Spearman ρ** | Monotonic relationship via ranks; robust to outliers | `df.corr(method='spearman')`, `stats.spearmanr` |
| **Kendall τ** | Concordant vs discordant pairs; small data, ties | `df.corr(method='kendall')`, `stats.kendalltau` |
| **Point-biserial** | Binary vs numeric | `stats.pointbiserialr(binary, numeric)` |
| **Chi-square** | Are two categorical variables independent? | `stats.chi2_contingency(pd.crosstab(a, b))` |
| **Cramér's V** | Strength of categorical association; 0..1 | `sqrt(chi2 / (n * (min(r, c) - 1)))` |
| **Correlation matrix** | All pairwise correlations | `df.corr(numeric_only=True)` |
| **Heatmap** | Visualize the matrix | `sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)` |
| **Scatter / pair plot** | See shape, curves, outliers | `sns.scatterplot`, `sns.regplot`, `sns.pairplot` |
| **Outliers** | One point can create/destroy r | Plot; compare Pearson vs Spearman; IQR; log |
| **Causation** | Correlation ≠ causation | Confounders, spurious trends, experiments |
| **Multicollinearity** | Features overlapping each other | Correlation screen + VIF |
| **High-corr pairs** | Quick screen, pairs only | `corr.abs().where(np.triu(ones, k=1)).stack()` |
| **VIF** | `1 / (1 − R²)` of feature vs all others | `variance_inflation_factor(sm.add_constant(X).values, i)` |
| **Dummy variable trap** | All one-hot columns + intercept = perfect collinearity | `pd.get_dummies(df, drop_first=True)`, `OneHotEncoder(drop='first')` |

### Interpretation guides

```text
|r|:   0–0.19 very weak | 0.2–0.39 weak | 0.4–0.59 moderate | 0.6–0.79 strong | 0.8–1 very strong
VIF:   1 none | 1–5 moderate (OK) | 5–10 high (investigate) | >10 serious | inf perfect
Cramér's V: <0.1 negligible | 0.1–0.3 weak | 0.3–0.5 moderate | >0.5 strong
```

### Ways to fix multicollinearity

```text
1. Remove one of the redundant features   (Height_cm vs Height_m)
2. Combine features                       (average / sum of related measures)
3. Feature engineering                    (ratios, differences: Age − Experience)
4. PCA                                    (uncorrelated components; less interpretable)
5. Regularization                         (Ridge stabilizes; Lasso can zero-out)
6. Domain knowledge                       (keep the meaningful, reliable, available feature)
```

### Decision flow

```text
Numeric ↔ numeric?        → scatter plot → Pearson (linear) / Spearman (monotonic, outliers) / Kendall (small, ties)
Binary ↔ numeric?         → point-biserial + box plot
Category ↔ numeric?       → group means / box plots
Category ↔ category?      → crosstab → chi-square → Cramér's V
Interpreting a linear model? → check feature–feature correlations → VIF → fix → re-check
```

### Golden rules
1. **Plot before you trust a number.**
2. **Correlation ≠ causation.**
3. **Feature ↔ target correlation is good; feature ↔ feature correlation can be a problem.**
4. **Use VIF, not only pairwise correlation, to detect multicollinearity.**
5. **Multicollinearity hurts interpretation and stability more than prediction.**
6. **Fix it with domain knowledge — never blindly.**
