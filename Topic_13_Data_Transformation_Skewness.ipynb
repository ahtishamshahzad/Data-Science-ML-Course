# 1. What is Data Transformation?

### 1. What it means
**Data transformation** means applying the **same mathematical function to every value** in a column to change the **shape** of its distribution.

```text
Raw salaries:      20k   25k   30k   40k   1,000k
log10(salary):     4.30  4.40  4.48  4.60  6.00
```

The order stays the same (the richest person is still the richest), but the huge gap between 40k and 1,000k has become much smaller.

### 2. Why we need it
Transformations can help to:
- **reduce skewness** (make a lopsided distribution more balanced),
- **reduce the effect of extreme values** (a billionaire no longer dominates),
- **make relationships easier to learn** for some models (e.g. a curve becomes a straight line),
- **stabilize variance** (spread that grows with the value becomes more constant),
- make data fit **statistical assumptions** better (e.g. approximately normal residuals in linear regression).

### ⚠️ Transformation does NOT automatically improve every ML model
Tree-based models (Random Forest, XGBoost) often gain nothing. Even for linear models it can help, do nothing, or hurt — you must **test** it.

### Raw vs transformed data

| | Raw | Transformed (log) |
|---|---|---|
| Values | Original units (rupees) | New units (log-rupees) |
| Order of rows | — | **Same order** (for monotonic transforms) |
| Gaps between big values | Huge | Compressed |
| Interpretation | Easy | Harder — must convert back |

### How is transformation different from other techniques you learned?

| Technique | What it changes | Shape changes? | Example |
|---|---|---|---|
| **Transformation** | The **shape** of the distribution | ✅ Yes | `log(x)`, `sqrt(x)`, Box-Cox |
| **Normalization** (Min-Max) | The **range** → usually [0, 1] | ❌ No | `(x − min) / (max − min)` |
| **Standardization** | The **center & spread** → mean 0, std 1 | ❌ No | `(x − mean) / std` |
| **Feature engineering** | **Creates new information** from existing columns | New column | `Price × Quantity`, `Debt / Income` |

Simple way to remember:

```text
Scaling (normalization / standardization) = zoom and move the ruler  → same shape, different numbers
Transformation                            = bend the ruler           → different shape
Feature engineering                       = build a new ruler        → new information
```

Transformations like `log(x)` are often *used inside* feature engineering (Topic 11), but the idea here is specifically **reshaping a distribution**.

````python
demo = pd.DataFrame({'Salary': [20000, 25000, 30000, 40000, 1_000_000]})

demo['Log10_Salary'] = np.log10(demo['Salary'])                                             # transformation
demo['MinMax_Salary'] = (demo['Salary'] - demo['Salary'].min()) / (demo['Salary'].max() - demo['Salary'].min())  # normalization
demo['Z_Salary'] = (demo['Salary'] - demo['Salary'].mean()) / demo['Salary'].std()          # standardization

demo.round(3)
````

### Expected output — what to notice
- **MinMax** and **Z-score**: the first four people are still squashed together near 0 and the millionaire is far away → the **shape did not change**.
- **Log10**: values 4.30, 4.40, 4.48, 4.60, 6.00 → the gap between normal salaries is now visible and the millionaire is much closer → the **shape changed**.

### Common mistakes
- Thinking `StandardScaler` removes skewness — it does **not**.
- Transforming without keeping track of how to reverse it (needed to explain predictions).
- Transforming everything "just because".

# 2. Understanding Data Distribution

### 1. What it means
A **distribution** describes **which values appear and how often**. A histogram is a picture of a distribution.

| Word | Simple meaning |
|---|---|
| **Shape** | What the histogram looks like (bell, lopsided, flat, two humps) |
| **Center** | A "typical" value — mean or median |
| **Spread** | How far values are from each other — standard deviation, IQR, range |
| **Tails** | The thin ends of the histogram where rare, extreme values live |
| **Symmetric** | Left side is a mirror image of the right side |
| **Asymmetric (skewed)** | One tail is much longer than the other |

### 2. Why we need it
You cannot choose the right transformation (or even the right summary statistic) without knowing the shape.

### Visualization tools
- **Histogram:** bars counting values in ranges (bins).
- **KDE (Kernel Density Estimate):** a smooth curve version of the histogram.

````python
n = 2000
symmetric = rng.normal(loc=50, scale=10, size=n)                   # bell-shaped, e.g. exam scores / heights
right_skewed = rng.lognormal(mean=10.5, sigma=0.7, size=n)         # long right tail, e.g. salary
left_skewed = 100 - rng.gamma(shape=2, scale=6, size=n)            # long left tail, e.g. easy-exam scores

dist = pd.DataFrame({'Symmetric': symmetric, 'Right_Skewed': right_skewed, 'Left_Skewed': left_skewed})

fig, axes = plt.subplots(1, 3, figsize=(16, 4))
titles = ['Approximately symmetric', 'Right-skewed (long RIGHT tail)', 'Left-skewed (long LEFT tail)']
for ax, col, t in zip(axes, dist.columns, titles):
    sns.histplot(dist[col], bins=40, kde=True, ax=ax, stat='density')   # histogram + KDE curve
    ax.set_title(f'{t}\nskew = {dist[col].skew():.2f}')
plt.tight_layout(); plt.show()

dist.describe().round(1)
````

### Expected output
- **Symmetric:** bell shape, mean ≈ median ≈ 50, skew ≈ 0.
- **Right-skewed:** most values on the left, a long thin tail to the right, positive skew.
- **Left-skewed:** most values on the right (near 100), a long tail to the left, negative skew.

### Common mistakes
- Using too few bins (hides the shape) or too many (looks noisy).
- Judging the shape from `describe()` alone without a plot.

# 3. What is Skewness?

### 1. What it means — in extremely simple language

> **Analogy:** Picture a playground seesaw with kids sitting on it. If most kids sit on the left but **one heavy adult** sits far out on the right, the seesaw tips. **Skewness measures how lopsided the data is and which side has the long "tail".**

```text
Right-skewed (positive skew):
Most values are smaller.
A few very large values create a long RIGHT tail.

   ██
   ███
   █████
   ███████▄▄▄▂▂▂▁▁▁▁___________→  (long tail to the right)

Left-skewed (negative skew):
Most values are larger.
A few very small values create a long LEFT tail.

                          ██
                         ███
                       █████
   ←___________▁▁▁▂▂▂▄▄▄███████   (long tail to the left)

Approximately symmetric (skew ≈ 0):
Both sides look like mirror images.
```

**Tip to remember:** the skew is named after the side with the **long tail**, not the side with most of the data.

### Real-world examples

| Variable | Usual skew | Why |
|---|---|---|
| **Salary / income** | Right | Most people earn normal salaries; a few earn enormous amounts |
| **House prices** | Right | Most homes are average; a few mansions cost many times more |
| **Number of purchases** | Right | Most customers buy a few times; a few loyal customers buy hundreds of times |
| **Customer spending** | Right | Many small spenders, a few "whales" |
| **Website visits per user** | Right | Most visit rarely; power users visit constantly |
| **Exam score on an easy test** | Left | Most students score high; a few score very low |
| **Age at retirement** | Left | Most retire around 60–65; a few retire very early |
| **Height of adults** | ≈ Symmetric | Most near average, few very short or very tall |

In business data, **right skew is much more common** than left skew, because many quantities start at 0 and have no upper limit.

````python
rw = np.random.default_rng(1)
n = 1000
real_world = pd.DataFrame({
    'Salary': rw.lognormal(11, 0.6, n).round(-2),
    'House_Price': rw.lognormal(16, 0.5, n).round(-3),
    'Purchases': rw.negative_binomial(2, 0.25, n),            # counts with a long tail
    'Customer_Spending': rw.gamma(1.5, 4000, n).round(0),
})

fig, axes = plt.subplots(1, 4, figsize=(18, 3.8))
for ax, col in zip(axes, real_world.columns):
    sns.histplot(real_world[col], bins=40, kde=True, ax=ax)
    ax.set_title(f'{col}\nskew = {real_world[col].skew():.2f}')
plt.tight_layout(); plt.show()
````

All four have long right tails → positive skewness.

### Common mistakes
- Calling data "right-skewed" because most values are on the **right** — it's the opposite; the **tail** is on the right.
- Thinking skew is always bad. It is simply a description of the shape.

# 4. Mean, Median and Skewness

### 1. What it means

| Shape | Relationship |
|---|---|
| Right skew | **Mean > Median** |
| Left skew | **Mean < Median** |
| Approximately symmetric | **Mean ≈ Median** |

### 3. How it works — why the mean is pulled toward the tail
- The **median** is the **middle** value after sorting. It only cares about *order*: making the largest value even larger does not move the middle.
- The **mean** adds up **all** values. One enormous value adds a lot to the sum, so the mean **moves toward the long tail**.

> **Analogy:** 9 friends each earn 50k and one friend earns 5 million. The "average income" of the group is ~545k — nobody in the group actually earns close to that! The median (50k) describes the typical friend much better.

### 4. Simple numerical examples

**Right skew** — salaries (thousands): `30, 32, 35, 38, 40, 300`
- Mean = 475 / 6 = **79.2**
- Median = (35 + 38) / 2 = **36.5**
- Mean > Median ✅ (the 300 pulled the mean up)

**Left skew** — exam scores: `20, 85, 88, 90, 92, 95`
- Mean = 470 / 6 = **78.3**
- Median = (88 + 90) / 2 = **89**
- Mean < Median ✅ (the 20 pulled the mean down)

**Symmetric** — `10, 20, 30, 40, 50`
- Mean = **30**, Median = **30** → equal ✅

````python
examples = {
    'Right skew (salaries)': [30, 32, 35, 38, 40, 300],
    'Left skew (scores)': [20, 85, 88, 90, 92, 95],
    'Symmetric': [10, 20, 30, 40, 50],
}
for name, values in examples.items():
    s = pd.Series(values)
    print(f'{name:<22} mean = {s.mean():6.1f}   median = {s.median():6.1f}   skew = {s.skew():5.2f}')
````

````python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for ax, col in zip(axes, dist.columns):
    sns.histplot(dist[col], bins=40, ax=ax, color='lightsteelblue')
    ax.axvline(dist[col].mean(), color='red', lw=2, label=f'Mean = {dist[col].mean():.0f}')
    ax.axvline(dist[col].median(), color='green', lw=2, ls='--', label=f'Median = {dist[col].median():.0f}')
    ax.set_title(col); ax.legend()
plt.tight_layout(); plt.show()
````

### Expected output
- Right-skewed: the **red** mean line is to the **right** of the green median line.
- Left-skewed: the mean is to the **left** of the median.
- Symmetric: the two lines almost overlap.

### When to use this check
As a quick, number-only hint of skew direction (e.g. inside `describe()` output).

### When NOT to rely on it alone
With multi-modal (two-hump) data or small samples, mean vs median can mislead. Always also plot.

### Common mistakes
- Reporting only the **mean** of salary or house price in a business report — the median is usually the better "typical" value for right-skewed data.

# 5. Measuring Skewness

### 1. What it means
The **skewness coefficient** is a single number describing the lopsidedness.

### 3. How it works (intuition, not heavy math)
1. For each value, compute its distance from the mean in units of standard deviation: `z = (x − mean) / std`.
2. **Cube** each z: `z³`. Cubing **keeps the sign** (negative stays negative) and makes **large distances much larger** (2³ = 8, 5³ = 125).
3. Average them.

So values far out on the **right** add big positive numbers; values far out on the **left** add big negative numbers. If both sides balance, the result is ≈ 0.

$$\text{skew} \approx \text{average of } \left(\frac{x - \bar{x}}{s}\right)^3$$

**Pandas** `.skew()` uses a slightly adjusted version (the "adjusted Fisher–Pearson coefficient", also called G1) which corrects for small samples. **SciPy** `stats.skew()` by default uses the **unadjusted** version (g1). With `bias=False`, SciPy gives the same number as Pandas. For large datasets the difference is tiny.

### General interpretation

| Skewness value | Interpretation |
|---|---|
| ≈ 0 | Approximately symmetric |
| Positive | Right-skewed (long right tail) |
| Negative | Left-skewed (long left tail) |
| Larger absolute value | Stronger skew |

A **common rule of thumb**:

| \|skew\| | Description |
|---|---|
| < 0.5 | Approximately symmetric |
| 0.5 – 1 | Moderately skewed |
| > 1 | Highly skewed |

### ⚠️ IMPORTANT: there is no universal magic cutoff
- These thresholds are **guidelines**, not laws.
- Skewness is sensitive to **sample size** and to a **few outliers**.
- Whether skew is a **problem** depends on the **model** (Section 7) and the **goal** (prediction vs interpretation).
- A skew of 1.2 might be fine for a Random Forest and worth fixing for a KNN model.

````python
sal = pd.DataFrame({'Salary': [30, 32, 35, 38, 40, 300]})

# Manual calculation (unadjusted g1)
x = sal['Salary']
z = (x - x.mean()) / x.std(ddof=0)          # ddof=0 → population std, used in the g1 formula
g1_manual = (z ** 3).mean()

print('Manual g1 (unadjusted)       :', round(g1_manual, 4))
print('SciPy skew (default, bias=True):', round(stats.skew(x), 4))
print('SciPy skew (bias=False)        :', round(stats.skew(x, bias=False), 4))
print('Pandas .skew()                 :', round(x.skew(), 4))
````

````python
# On larger data the two versions are almost identical
print('n = 2000 right-skewed:')
print('  Pandas:', round(dist['Right_Skewed'].skew(), 4))
print('  SciPy :', round(stats.skew(dist['Right_Skewed']), 4))

print('\nSkewness of every column (Pandas):')
print(dist.skew().round(3))
````

### Expected output
- For the tiny 6-value example, Pandas ≈ 2.44 and SciPy (default) ≈ 1.78 — noticeably different because the sample is tiny.
- With `bias=False` SciPy matches Pandas exactly.
- For 2,000 values, both are nearly identical.

### Common mistakes
- Comparing a Pandas skew value from one report with a SciPy value from another without knowing which formula was used.
- Trusting skewness from 10 rows.
- Treating 0.99 and 1.01 as fundamentally different.

# 6. Detecting Skewness Visually

### Tools

| Tool | What to look for |
|---|---|
| **Histogram** | Where is the long tail? |
| **KDE plot** | Smooth curve; peak shifted to one side |
| **Boxplot** | Median line off-center in the box; one whisker much longer; many outlier dots on one side |
| **Mean vs median** | Mean pulled toward the tail |

### A reusable function

````python
def analyze_skewness(df, column, bins=40):
    '''Show histogram (+KDE), boxplot, and key statistics for one numeric column.'''
    data = df[column].dropna()                       # 1. remove missing values so stats/plots work

    mean = data.mean()                               # 2. average (pulled toward the tail)
    median = data.median()                           # 3. middle value (robust)
    skew = data.skew()                               # 4. Pandas adjusted skewness

    # 5. Describe the skew in words using a common rule of thumb (a guideline, not a law)
    if abs(skew) < 0.5:
        label = 'approximately symmetric'
    elif abs(skew) < 1:
        label = 'moderately ' + ('right' if skew > 0 else 'left') + '-skewed'
    else:
        label = 'highly ' + ('right' if skew > 0 else 'left') + '-skewed'

    # 6. Two plots side by side: histogram (wider) and boxplot
    fig, axes = plt.subplots(1, 2, figsize=(13, 3.8), gridspec_kw={'width_ratios': [2, 1]})

    sns.histplot(data, bins=bins, kde=True, ax=axes[0], color='lightsteelblue')
    axes[0].axvline(mean, color='red', lw=2, label=f'Mean = {mean:,.2f}')
    axes[0].axvline(median, color='green', lw=2, ls='--', label=f'Median = {median:,.2f}')
    axes[0].set_title(f'{column}: skew = {skew:.2f} ({label})')
    axes[0].legend()

    sns.boxplot(x=data, ax=axes[1], color='lightsteelblue')
    axes[1].set_title(f'{column}: boxplot')
    plt.tight_layout(); plt.show()

    # 7. Return the numbers as a small table so they can be used later
    return pd.Series({'mean': mean, 'median': median, 'skewness': skew, 'interpretation': label}, name=column)

analyze_skewness(real_world, 'Customer_Spending')
````

### Line-by-line explanation
1. `.dropna()` — skewness, mean and plots need non-missing values.
2–4. Calculate mean, median and skewness.
5. Convert the number into words using the rule-of-thumb table from Section 5.
6. `plt.subplots(1, 2, ..., gridspec_kw={'width_ratios': [2, 1]})` — two plots in one row, the histogram twice as wide as the boxplot. `axvline` draws the mean (red) and median (green, dashed) as vertical lines.
7. Return a `pd.Series` so you can collect results for many columns into a table.

### Expected output
For `Customer_Spending`: a long right tail, mean to the right of median, a boxplot with a long right whisker and many outlier dots on the right, and a positive skew labelled "highly right-skewed".

````python
analyze_skewness(dist, 'Left_Skewed')
````

### Common mistakes
- Forgetting `dropna()` — some functions then return `NaN` or fail.
- Reading boxplot "outliers" in skewed data as errors — in right-skewed data many points beyond the whisker are **normal** members of the long tail.

# 7. Why Skewness Can Matter in Machine Learning

### ⚠️ First: skewness is **not automatically a problem**
Many excellent models are trained on skewed features. The question is always: **does this model, for this goal, care about the shape?**

### Models that can be sensitive to feature distributions

| Model | Why skew can matter |
|---|---|
| **Linear Regression** | Assumes a **linear** relationship. Skewed features often have curved relationships with the target, and extreme values act as **high-leverage points** that pull the line. Residual assumptions (normality, constant variance) matter for confidence intervals/p-values. |
| **Logistic Regression** | Also linear (in log-odds). A few huge values can dominate the coefficient; a log may make the relationship more linear. |
| **KNN** | Uses **distances**. In a skewed feature, most points are crushed together and a few are far away → distances are dominated by the tail. |
| **SVM** | Distance/kernel-based → the same issue as KNN; also sensitive to scale. |
| **Neural Networks** | Train with gradient descent. Very large or long-tailed inputs make training slower/less stable; they usually prefer inputs on a similar, well-behaved scale. |

### Models that are generally less sensitive

| Model | Why |
|---|---|
| **Decision Tree** | Splits on thresholds like `Income < 50,000`. A **monotonic** transformation (log, sqrt, Box-Cox with fixed λ) keeps the **order** of values, so the same split exists in the transformed data (`log(Income) < 10.82`). The tree finds equivalent splits. |
| **Random Forest** | An ensemble of trees → the same reasoning. |
| **Gradient Boosting** (XGBoost, LightGBM, CatBoost) | Also tree-based → the same reasoning. |

### Careful: "less sensitive" ≠ "completely unaffected"
- Some tree implementations **bin** feature values into histograms (LightGBM, `HistGradientBoosting`). Changing the shape changes which values share a bin, so results can differ slightly.
- Transforming the **target** (not a feature) changes what the tree optimizes — e.g. predicting `log(Price)` minimizes **relative** errors instead of absolute ones. That can matter a lot.
- **Non-monotonic** transformations (e.g. `x²` on data with negative values) change the order and can change tree results.
- Random seeds and tie-breaking can make tiny differences.

### A quick demonstration

````python
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

rd = np.random.default_rng(7)
n = 600
income = rd.lognormal(11, 0.9, n)                          # strongly right-skewed feature
age = rd.uniform(20, 65, n)
spend = 3000 * np.log(income) + 50 * age + rd.normal(0, 1500, n)   # target depends on log(income)

X_raw = pd.DataFrame({'Income': income, 'Age': age})
X_log = X_raw.assign(Income=np.log(X_raw['Income']))         # same data, income log-transformed

idx_train, idx_test = train_test_split(np.arange(n), test_size=0.3, random_state=0)

def score(model, X):
    model.fit(X.iloc[idx_train], spend[idx_train])
    return r2_score(spend[idx_test], model.predict(X.iloc[idx_test]))

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

results = pd.DataFrame({
    'Raw Income': [score(LinearRegression(), X_raw),
                   score(make_pipeline(StandardScaler(), KNeighborsRegressor(10)), X_raw),
                   score(DecisionTreeRegressor(max_depth=5, random_state=0), X_raw)],
    'log(Income)': [score(LinearRegression(), X_log),
                    score(make_pipeline(StandardScaler(), KNeighborsRegressor(10)), X_log),
                    score(DecisionTreeRegressor(max_depth=5, random_state=0), X_log)],
}, index=['Linear Regression', 'KNN (scaled)', 'Decision Tree']).round(3)
print('Test R² (higher is better):')
results
````

### Expected output
- **Linear Regression:** clearly better with `log(Income)` because the true relationship is linear in log-income.
- **KNN:** better with log, because distances are no longer dominated by a few huge incomes.
- **Decision Tree:** **identical** R² — the log kept the order of incomes, so the tree found equivalent splits.

### When to transform for ML
- Linear/logistic regression, KNN, SVM, neural networks, especially with **strong skew** or **curved** relationships.
- When model diagnostics (residual plots) look bad.

### When NOT to bother
- Pure tree-based models focused on prediction (usually little gain).
- When interpretability in original units is essential and the gain is small.

### Common mistakes
- Transforming all features for a Random Forest and expecting improvement.
- Assuming "skewed = must fix" without testing with cross-validation.

# 8. Transformation Categories — Roadmap

```text
Data
 │
 ├── Log transformation          log(x), log1p(x)
 ├── Square-root transformation  sqrt(x)
 ├── Cube-root transformation    cbrt(x)
 ├── Reciprocal transformation   1 / x
 ├── Power transformation        (learns the best power automatically)
 │      ├── Box-Cox              (strictly positive data only)
 │      └── Yeo-Johnson          (works with zero and negative data)
 └── Quantile transformation     (maps ranks to a uniform or normal shape)
```

### Strength of the "squeeze" on large values (weakest → strongest)

```text
cube root  <  square root  <  log  <  reciprocal
 (gentle)                               (very strong)
```

| Transformation | Formula | Allowed values | Useful for | Section |
|---|---|---|---|---|
| **Log** | `log(x)` | x > 0 | Strong right skew, multiplicative data (income, prices) | 9–10 |
| **log1p** | `log(1 + x)` | x > −1 (typically x ≥ 0) | Right skew **with zeros** (counts, spending) | 9–10 |
| **Square root** | `√x` | x ≥ 0 | Moderate right skew, **counts** | 11 |
| **Cube root** | `∛x` | **any** real number | Mild skew, data with **negatives** | 12 |
| **Reciprocal** | `1/x` | x ≠ 0 | Very strong skew; rates like "time per task" ↔ "tasks per time" | 13 |
| **Box-Cox** | `(x^λ − 1)/λ` | x > 0 | Automatically find the best power | 14 |
| **Yeo-Johnson** | Box-Cox-like, extended | **any** real number | Automatic power with zero/negative values | 15 |
| **Quantile** | Based on ranks | any | Force a uniform/normal shape; very robust to outliers | 16 |

# 9. Log Transformation

### 1. What it means
The **logarithm** answers: *"what power do I raise the base to, to get this number?"*

```text
log10(10)      = 1      because 10¹ = 10
log10(100)     = 2      because 10² = 100
log10(1,000)   = 3
log10(10,000)  = 4
```

`np.log(x)` uses base *e* ≈ 2.718 (the **natural log**). The base only changes the numbers by a constant factor — the **shape** after the transformation is the same for any base.

### 2. Why we need it
Right-skewed data like salary, spending, prices and populations is often **multiplicative**: a raise is "+10%", not "+5,000". Log turns multiplication into addition and pulls in the long right tail.

### 3. How it works — why large values are compressed more
Every time the value is **multiplied by 10**, the log only goes up by **1** (base 10):

| x | log10(x) | Change in x | Change in log |
|---|---|---|---|
| 10 | 1 | — | — |
| 100 | 2 | +90 | +1 |
| 1,000 | 3 | +900 | +1 |
| 10,000 | 4 | +9,000 | +1 |

- Going from 10 → 100 (small values, +90) gets the **same** +1 as 1,000 → 10,000 (+9,000).
- So **large values are compressed much more** than small values; small values change relatively little.
- Result: the long right tail is pulled in → skewness drops.

### `np.log(x)` vs `np.log1p(x)`

| | `np.log(x)` | `np.log1p(x)` = `log(1 + x)` |
|---|---|---|
| x = 0 | `-inf` ❌ | 0 ✅ |
| Works for | x > 0 | x > −1 (in practice x ≥ 0) |
| Reverse with | `np.exp(y)` | `np.expm1(y)` |
| Typical use | Salaries, prices (never 0) | Counts, spending, visits (can be 0) |

For large values, `log1p(x) ≈ log(x)`. For small values `log1p` is also more numerically accurate.

### 4. Numerical example
Spending: `[0, 10, 100, 1000, 10000]`
- `log1p` → `[0, 2.40, 4.62, 6.91, 9.21]` — the range shrinks from 0–10,000 to 0–9.2 and the values are nicely spread out.

````python
vals = np.array([0, 10, 100, 1000, 10000])
print('log1p:', np.log1p(vals).round(2))
print('log   :', np.log(np.where(vals > 0, vals, np.nan)).round(2), '← 0 is impossible (we replaced it with NaN)')
print('back  :', np.expm1(np.log1p(vals)).round(2), '← expm1 reverses log1p')
````

````python
rs = np.random.default_rng(9)
n = 1000
money = pd.DataFrame({
    'Salary': rs.lognormal(mean=11, sigma=0.7, size=n).round(-2),     # always > 0
    'Spending': rs.gamma(shape=0.8, scale=5000, size=n).round(0),    # includes zeros after rounding
})
money.loc[rs.choice(n, 60, replace=False), 'Spending'] = 0           # some customers spent nothing

money['Log_Salary'] = np.log(money['Salary'])                        # safe: all salaries > 0
money['Log1p_Spending'] = np.log1p(money['Spending'])                # safe: handles zeros

summary = pd.DataFrame({
    'Original skew': [money['Salary'].skew(), money['Spending'].skew()],
    'After log skew': [money['Log_Salary'].skew(), money['Log1p_Spending'].skew()],
}, index=['Salary (log)', 'Spending (log1p)']).round(3)
print('Zeros in Spending:', (money['Spending'] == 0).sum())
summary
````

````python
fig, axes = plt.subplots(2, 2, figsize=(13, 7))
pairs = [('Salary', 'Log_Salary'), ('Spending', 'Log1p_Spending')]
for row, (orig, trans) in enumerate(pairs):
    sns.histplot(money[orig], bins=40, kde=True, ax=axes[row, 0], color='tab:orange')
    axes[row, 0].set_title(f'Original {orig} (skew = {money[orig].skew():.2f})')
    sns.histplot(money[trans], bins=40, kde=True, ax=axes[row, 1], color='tab:blue')
    axes[row, 1].set_title(f'{trans} (skew = {money[trans].skew():.2f})')
plt.tight_layout(); plt.show()
````

### Expected output & interpretation
- **Salary:** skew drops from about 2 to about 0 → the log version looks bell-shaped.
- **Spending:** the skew changes from about **+2.2 to about −1.9** — the transformation **over-corrected** and made the data *left*-skewed! The 60 customers who spent nothing all land at exactly 0, far to the left of everyone else (log1p of a typical spend is ≈ 7–9). Log cannot "spread out" identical values.
  - Lesson: **always check the result** — a transformation can overshoot.
  - Tip: add a flag column `Spent_Nothing = (Spending == 0)` so the model knows about this group (Topic 11).

### 8. When to use
- Strong right skew, **positive** values, multiplicative behaviour (income, prices, population, spending).
- When a scatter plot of the feature vs target curves like a "diminishing returns" shape.

### 9. When NOT to use
- Negative values (Section 10).
- Already symmetric or left-skewed data (log makes left skew **worse**).
- When the model is tree-based and only prediction matters (little effect).

### 10. Common mistakes
- `np.log` on data with zeros → `-inf`, which silently breaks scaling and models.
- Forgetting to **reverse** the transform (`np.exp` / `np.expm1`) when reporting predictions of a log-transformed **target**.
- Thinking log "removes" outliers — they're still the largest values, just closer to the rest.

# 10. Important Log Transformation Issue — Zero and Negative Values

### 1. What it means
The logarithm is only defined (for ordinary real numbers) for **positive** values:

```text
log(5)   = 1.609   ✅
log(1)   = 0       ✅
log(0)   = −∞      ❌ (minus infinity)
log(−5)  = NaN     ❌ (no real number y gives e^y = −5, because e^y is always positive)
```

Real-world columns with zero or negative values: account balance, profit/loss, temperature, change in sales, spending (zero), number of purchases (zero).

````python
with np.errstate(divide='ignore', invalid='ignore'):     # silence NumPy's warnings for the demo
    print('np.log(5)  =', np.log(5))
    print('np.log(0)  =', np.log(0))
    print('np.log(-5) =', np.log(-5))
````

### How to handle zero and negative values

| Situation | Option | Notes |
|---|---|---|
| Zeros, no negatives | **`np.log1p(x)`** | The standard choice for counts/spending |
| Negatives present | **Shift:** `np.log(x − min(x) + 1)` | Simple, but has problems (below) |
| Negatives present | **Yeo-Johnson** (Section 15) | Learns the best transformation and handles any real value |
| Negatives present, want something simple | **Cube root** (Section 12) | Keeps the sign |
| Negatives present, symmetric-ish tails | **Signed log:** `sign(x) · log1p(|x|)` | Compresses both tails, keeps the sign |

### ⚠️ Why blindly adding a constant is not always the best solution
1. **The constant is arbitrary.** `log(x + 1)`, `log(x + 10)`, `log(x + 1000)` give **very different shapes**. There's no single "correct" constant.
2. **Leakage / instability.** If you shift by `min(x)` of the **training** data, a new customer with an even lower value in production produces `log(negative)` → `NaN`.
3. **Meaning changes.** Adding a large constant makes the log almost linear over the data range, so it barely reduces skewness.
4. **Zero may be special.** "Balance = 0" might be a separate group; forcing it into a continuous curve hides that.

````python
rb = np.random.default_rng(3)
balance = pd.Series(np.concatenate([rb.lognormal(9, 1.0, 900), -rb.lognormal(7, 1.0, 100)]), name='Balance')
print('Min balance:', round(balance.min(), 1), '| Negative values:', (balance < 0).sum())
print('Original skew:', round(balance.skew(), 2))

for c in [1, 100, 10_000]:
    shifted = np.log(balance - balance.min() + c)
    print(f'log(x - min + {c:>6}) → skew = {shifted.skew():6.2f}')

signed_log = np.sign(balance) * np.log1p(balance.abs())
print('signed log          → skew =', round(signed_log.skew(), 2))
````

### Expected output
The skewness after shifting depends heavily on the constant you choose — proof that the choice is arbitrary. The signed log and Yeo-Johnson (Section 15) avoid picking a constant.

### Common mistakes
- Using `np.log(x + 1)` on data that contains `−5` → still `log(−4)` = NaN.
- Computing the shift constant on the **full** dataset (leakage) — compute it on the training data, and decide what happens to new values below it.

# 11. Square Root Transformation

### 1. What it means
Replace each value with its square root: `√x`.

```text
√0 = 0   √1 = 1   √4 = 2   √9 = 3   √100 = 10   √10,000 = 100
```

### 2. Why we need it
It compresses large values, but **more gently than log**. It is a classic choice for **count data** (purchases, visits, support tickets, accidents), where the spread tends to grow with the average (for counts, variance ≈ mean). The square root **stabilizes this variance**.

### 3. How it works
- To double √x you must multiply x by 4. To increase log x by the same amount you'd multiply x by a fixed factor — log compresses harder.
- `√0 = 0`, so **zeros are fine**. Negative values are not allowed.

### 4. Numerical example
Support tickets: `[0, 1, 4, 9, 16, 100]` → `√` → `[0, 1, 2, 3, 4, 10]`. The outlier 100 is now only 2.5× the value 16 instead of 6.25×.

````python
rc = np.random.default_rng(11)
n = 1000
counts = pd.DataFrame({
    'Purchases': rc.poisson(lam=rc.gamma(1.2, 4, n)),         # counts with overdispersion (long tail)
    'Website_Visits': rc.poisson(lam=rc.gamma(0.9, 20, n)),
    'Support_Tickets': rc.poisson(lam=rc.gamma(0.6, 2, n)),
})

rows = []
for col in counts.columns:
    rows.append({'Column': col,
                 'Zeros': int((counts[col] == 0).sum()),
                 'Original skew': counts[col].skew(),
                 'sqrt skew': np.sqrt(counts[col]).skew(),
                 'log1p skew': np.log1p(counts[col]).skew()})
pd.DataFrame(rows).round(2)
````

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 3.8))
col = 'Website_Visits'
for ax, (name, values) in zip(axes, [('Original', counts[col]),
                                     ('sqrt', np.sqrt(counts[col])),
                                     ('log1p', np.log1p(counts[col]))]):
    sns.histplot(values, bins=30, ax=ax)
    ax.set_title(f'{col} — {name} (skew = {values.skew():.2f})')
plt.tight_layout(); plt.show()
````

### Expected output
- **Purchases / Website_Visits:** sqrt reduces the skew clearly; log1p reduces it even more and can **overshoot** into a negative (left) skew.
- **Support_Tickets:** more than half the values are 0. **Neither** transformation can fix a spike of identical values — here log1p is not even better than sqrt. For such columns, a flag like `Has_Ticket` is often more useful than any transformation.

This is why sqrt is often preferred for moderately skewed counts: it is gentler and rarely overshoots.

### When to use
Counts, moderately right-skewed data, data with zeros but no negatives.

### When NOT to use
Negative values; very strong skew (log may be needed); left-skewed data.

### Common mistakes
Using sqrt on negative values → `NaN`. Assuming the "most symmetric" result is automatically the best for the model.

# 12. Cube Root Transformation

### 1. What it means
`∛x` — the number which multiplied by itself three times gives x.

```text
∛8 = 2    ∛27 = 3    ∛1000 = 10    ∛0 = 0    ∛(−8) = −2    ∛(−1000) = −10
```

### 2. Why we need it
Unlike `log` and `sqrt`, the cube root is defined for **negative numbers** and **keeps the sign**. It compresses large values in **both** directions.

### 3. How it works
It is weaker than sqrt and log for large positive values, but it works on the whole real line. `np.cbrt` handles negatives correctly (note: `x ** (1/3)` in NumPy returns NaN for negatives — a common trap!).

### 4. Numerical example
Profit/loss: `[−1000, −8, 0, 8, 27, 1000]` → `[−10, −2, 0, 2, 3, 10]`

````python
pl = np.array([-1000, -8, 0, 8, 27, 1000], dtype=float)
print('np.cbrt :', np.cbrt(pl))
with np.errstate(invalid='ignore'):
    print('x**(1/3):', pl ** (1/3), '← NaN for negatives! Use np.cbrt instead.')
````

````python
rp = np.random.default_rng(5)
profit = pd.Series(np.concatenate([rp.gamma(1.2, 20000, 800), -rp.gamma(1.0, 8000, 200)]), name='Profit')

cbrt_profit = np.cbrt(profit)
fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
sns.histplot(profit, bins=40, ax=axes[0]); axes[0].set_title(f'Profit (skew = {profit.skew():.2f})')
sns.histplot(cbrt_profit, bins=40, ax=axes[1]); axes[1].set_title(f'cbrt(Profit) (skew = {cbrt_profit.skew():.2f})')
plt.tight_layout(); plt.show()
````

### Expected output
Skewness drops strongly. Note that cube root often creates a **"gap" or dip around 0** because it stretches values near zero (the slope of ∛x is very steep close to 0). Always look at the plot.

### When to use
Mild-to-moderate skew with **negative** values (profit/loss, changes, temperatures, balances).

### When NOT to use
When values near zero are extremely common and their small differences are just noise (cube root exaggerates them).

### Common mistakes
Using `x ** (1/3)` instead of `np.cbrt(x)`.

# 13. Reciprocal Transformation

### 1. What it means
`1 / x` — flip the value.

```text
x:    1     2     4     10     100
1/x:  1   0.5   0.25   0.1   0.01
```

### 2. Why we need it
- It is the **strongest** of the common transformations for right skew: huge values become tiny numbers near 0.
- Sometimes the reciprocal **has a natural meaning**:
  - *minutes per task* ↔ *tasks per minute*
  - *km per litre* ↔ *litres per km*
  - *time to complete a download* ↔ *download speed*

### 3. How it works — it REVERSES the order
If `a < b` (both positive), then `1/a > 1/b`. So:
- the **largest** original value becomes the **smallest** transformed value,
- **relationships flip direction**: if price increases with size, price will **decrease** with `1/size`.

Use `−1/x` if you want to keep the original order.

### ⚠️ Dangerous with zero
`1 / 0` = `inf`. And values **close to zero** explode: `1 / 0.001 = 1000`. A tiny measurement error near zero becomes a huge outlier.

### 4. Numerical example
Minutes to complete a task: `[1, 2, 4, 10, 60]` → tasks per minute: `[1.0, 0.5, 0.25, 0.1, 0.017]`

````python
minutes = pd.Series([1, 2, 4, 10, 60], name='Minutes_per_Task')
recip = pd.DataFrame({'Minutes_per_Task': minutes,
                      'Tasks_per_Minute (1/x)': 1 / minutes,
                      'Order-keeping (-1/x)': -1 / minutes})
print(recip.round(3))

with np.errstate(divide='ignore'):
    print('\n1 / 0     =', np.float64(1) / 0)
print('1 / 0.001 =', 1 / 0.001, '← tiny values explode')
````

````python
rt = np.random.default_rng(13)
task_time = pd.Series(rt.lognormal(1.5, 0.8, 1000) + 0.5, name='Minutes')   # strictly positive
fig, axes = plt.subplots(1, 3, figsize=(15, 3.8))
for ax, (name, v) in zip(axes, [('Original', task_time), ('log', np.log(task_time)), ('1/x', 1 / task_time)]):
    sns.histplot(v, bins=40, ax=ax); ax.set_title(f'{name} (skew = {v.skew():.2f})')
plt.tight_layout(); plt.show()
````

### Expected output
Reciprocal changes the shape very strongly — sometimes it **over-corrects** (right skew becomes left skew or stays skewed in the other direction). Log is often more moderate.

### When to use
Very strong right skew with values far from zero, or when the reciprocal has a real meaning (rates).

### When NOT to use
Data with zeros or values near zero; negative values mixed with positive ones (the order reverses differently on each side); when you need an easy interpretation.

### Common mistakes
Forgetting that coefficients/correlations **change sign** after `1/x`.

# 14. Box-Cox Transformation

### 1. What it means
Instead of **you** choosing log or sqrt, **Box-Cox searches for the best power** automatically.

$$y = \begin{cases} \dfrac{x^{\lambda} - 1}{\lambda} & \text{if } \lambda \neq 0 \\[2mm] \log(x) & \text{if } \lambda = 0 \end{cases}$$

### Why "power transformation"?
Because it raises x to a **power λ** (lambda). Different λ values reproduce familiar transformations (apart from shifting/rescaling, which doesn't change the shape):

| λ | Equivalent shape | Meaning |
|---|---|---|
| 2 | x² | Stretches large values (fixes **left** skew) |
| 1 | x | **No change** in shape |
| 0.5 | √x | Square root |
| 0.33 | ∛x | Cube root |
| 0 | log(x) | Log |
| −0.5 | 1/√x | Reciprocal square root |
| −1 | 1/x | Reciprocal |

### 3. How it finds λ
SciPy tries many λ values and picks the one that makes the result look **most like a normal distribution** (technically: it maximizes the *log-likelihood* assuming the result is normal). You don't need the math — just understand: **"λ is chosen to make the data as bell-shaped as possible."**

### ⚠️ Requirement: strictly positive data
Every value must be **> 0**. Zero or negative values → error. (Use Yeo-Johnson instead.)

### 4. Numerical intuition
For salary data, Box-Cox usually finds λ close to **0** → it's essentially telling you "use a log".

````python
from scipy.stats import boxcox

salary = money['Salary']                               # strictly positive
salary_bc, best_lambda = boxcox(salary)                 # returns (transformed values, chosen lambda)

print('Chosen lambda:', round(best_lambda, 3))
print('Skew before  :', round(salary.skew(), 3))
print('Skew after   :', round(pd.Series(salary_bc).skew(), 3))
print('Skew of log  :', round(np.log(salary).skew(), 3))
````

````python
fig, axes = plt.subplots(1, 3, figsize=(16, 3.8))
sns.histplot(salary, bins=40, kde=True, ax=axes[0]); axes[0].set_title(f'Original (skew {salary.skew():.2f})')
sns.histplot(salary_bc, bins=40, kde=True, ax=axes[1]); axes[1].set_title(f'Box-Cox λ = {best_lambda:.2f}')
stats.probplot(salary_bc, dist='norm', plot=axes[2]); axes[2].set_title('Q-Q plot after Box-Cox')
plt.tight_layout(); plt.show()
````

### Expected output — explaining it
- `boxcox` returns **two things**: the transformed array and the chosen λ.
- λ is close to 0 → Box-Cox essentially chose a log, which makes sense for lognormal salaries.
- Skew after ≈ 0.
- **Q-Q plot** (quantile–quantile plot): compares the data to a perfect normal distribution. If the points lie on the red straight line, the data is approximately normal. After Box-Cox they lie nicely on the line.

### Applying the SAME λ to new data
In ML you must use the λ learned from the **training** data on the test data:

````python
from scipy.special import boxcox as boxcox_with_lambda     # applies a GIVEN lambda
from scipy.special import inv_boxcox                          # reverses the transformation

new_salaries = np.array([25000, 60000, 250000])
transformed = boxcox_with_lambda(new_salaries, best_lambda)
print('New data transformed with training λ:', transformed.round(3))
print('Reversed back:', inv_boxcox(transformed, best_lambda).round(0))

try:
    boxcox(np.array([100, 0, 50]))
except ValueError as e:
    print('\nBox-Cox with a zero → ValueError:', e)
````

### Box-Cox with other λ values (to see the family)

````python
x = np.linspace(0.1, 10, 200)
plt.figure(figsize=(7, 4.5))
for lam in [2, 1, 0.5, 0, -0.5, -1]:
    y = np.log(x) if lam == 0 else (x ** lam - 1) / lam
    plt.plot(x, y, label=f'λ = {lam}')
plt.ylim(-5, 10); plt.xlabel('x'); plt.ylabel('Box-Cox(x)')
plt.title('The Box-Cox family: smaller λ → stronger compression of large values')
plt.legend(); plt.show()
````

### When to use
Strictly positive, skewed data where you want an **automatic, data-driven** choice of power.

### When NOT to use
Zeros/negatives; tree models; when you need a simple explanation ("log salary" is easier to explain than "salary to the power 0.07").

### Common mistakes
- Calling `boxcox()` separately on train and test → **different λ** → inconsistent features and leakage. Learn λ on train, apply to test (or use `PowerTransformer`, Section 17).
- Adding +1 to make data positive without thinking (Section 10).

# 15. Yeo-Johnson Transformation

### 1. What it means
**Yeo-Johnson** is an extension of Box-Cox that works for **any real number** — zero, negative, and positive.

### 2. Why we need it
Real data often includes zeros and negatives (account balance, profit, temperature change). Box-Cox refuses them; Yeo-Johnson handles them without an arbitrary shift.

### 3. How it works (simple version)
- For **x ≥ 0**: it applies a Box-Cox-like power to **(x + 1)**.
- For **x < 0**: it applies a mirrored version to **(−x + 1)** with power **(2 − λ)**, then puts the minus sign back.
- One **λ** is learned for the whole column (again, chosen to make the result as normal as possible).

$$y = \begin{cases}
\dfrac{(x+1)^{\lambda} - 1}{\lambda} & x \ge 0,\ \lambda \ne 0 \\[2mm]
\log(x+1) & x \ge 0,\ \lambda = 0 \\[2mm]
-\dfrac{(-x+1)^{2-\lambda} - 1}{2-\lambda} & x < 0,\ \lambda \ne 2 \\[2mm]
-\log(-x+1) & x < 0,\ \lambda = 2
\end{cases}$$

You never need to calculate this by hand — scikit-learn does it.

`PowerTransformer` by default also **standardizes** the output (`standardize=True`) → mean 0, std 1.

````python
from sklearn.preprocessing import PowerTransformer

acct = pd.DataFrame({'Balance': balance})                   # has negative values (from Section 10)

pt = PowerTransformer(method='yeo-johnson')                 # standardize=True by default
acct['Balance_YJ'] = pt.fit_transform(acct[['Balance']])    # scikit-learn expects a 2D input → [['col']]

print('Learned lambda:', pt.lambdas_.round(3))
print('Skew before:', round(acct['Balance'].skew(), 3))
print('Skew after :', round(acct['Balance_YJ'].skew(), 3))
print('Mean / std after:', round(acct['Balance_YJ'].mean(), 3), '/', round(acct['Balance_YJ'].std(ddof=0), 3))

fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
sns.histplot(acct['Balance'], bins=50, ax=axes[0]); axes[0].set_title('Balance (with negatives)')
sns.histplot(acct['Balance_YJ'], bins=50, ax=axes[1]); axes[1].set_title('Yeo-Johnson')
plt.tight_layout(); plt.show()
````

### ⚠️ Result above: the skew only dropped from ≈ 4.5 to ≈ 2.1 — why?
**Yeo-Johnson depends on the units of the data.** Its formula uses `x + 1`, and "+1" means something very different for values around 1 than for values around 10,000. With balances in rupees (tens of thousands) and a small group of negative values, a single λ cannot bend both sides well.

**Fix: bring the data to a small, unit-free scale first** (e.g. `StandardScaler`, or express the money in thousands), *then* apply Yeo-Johnson.

````python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

for label, values in [('Rupees (raw)', acct[['Balance']]),
                      ('Thousands of rupees', acct[['Balance']] / 1000)]:
    out = PowerTransformer().fit_transform(values).ravel()
    print(f'{label:<28} → Yeo-Johnson skew = {pd.Series(out).skew():6.2f}')

scaled_yj = make_pipeline(StandardScaler(), PowerTransformer(method='yeo-johnson'))
out = scaled_yj.fit_transform(acct[['Balance']]).ravel()
print(f"{'StandardScaler → Yeo-Johnson':<28} → Yeo-Johnson skew = {pd.Series(out).skew():6.2f}")
````

Rescaling first fixes most of the problem (thousands: skew ≈ 0.1; standardized: ≈ −0.4).

**Practical rule:**
- **Only positive values** (even with zeros) → apply Yeo-Johnson **directly**. It already works well, and centering the data first would *create* artificial negatives and make the result slightly worse.
- **Negative and positive values with large units** (balances, profit in rupees) → **rescale first** (`StandardScaler`, or convert to thousands), then Yeo-Johnson.

Box-Cox does not have this issue (for positive data, multiplying x by a constant doesn't change its best shape). We use this rule in Sections 24 and 26.

### Box-Cox vs Yeo-Johnson

````python
pos = money[['Salary']]                                    # strictly positive → both methods allowed

bc = PowerTransformer(method='box-cox').fit(pos)
yj = PowerTransformer(method='yeo-johnson').fit(pos)

print('Box-Cox λ    :', bc.lambdas_.round(3), '| skew after:', round(pd.Series(bc.transform(pos).ravel()).skew(), 3))
print('Yeo-Johnson λ:', yj.lambdas_.round(3), '| skew after:', round(pd.Series(yj.transform(pos).ravel()).skew(), 3))

try:
    PowerTransformer(method='box-cox').fit(acct[['Balance']])
except ValueError as e:
    print('\nBox-Cox on negative data →', str(e)[:80], '...')
````

| | Box-Cox | Yeo-Johnson |
|---|---|---|
| Allowed values | **x > 0 only** | **Any** real number |
| Zero | ❌ | ✅ |
| Negative | ❌ | ✅ |
| On positive data | Classic, slightly simpler | Very similar results |
| Depends on units? | No (for positive data) | **Yes** → standardize first for mixed-sign / large values |
| SciPy | `stats.boxcox` | `stats.yeojohnson` |
| scikit-learn | `PowerTransformer(method='box-cox')` | `PowerTransformer(method='yeo-johnson')` (**default**) |

### When to use Yeo-Johnson
Skewed data with zeros/negatives, or as a safe default in pipelines when you don't know the value range of future data.

### When NOT to use
Tree-only models; when a simple, explainable log is enough; bimodal data (a single power can't fix two humps).

### Common mistakes
- Passing a 1D Series (`df['col']`) to scikit-learn → error. Use `df[['col']]`.
- Forgetting that the output is **standardized** by default (values are no longer in rupees).

# 16. Quantile Transformation

### 1. What it means

> **Analogy:** Instead of using students' raw marks, you give each student their **percentile rank** ("better than 80% of the class"). Then you can convert that percentile into any shape you like — a flat distribution, or a bell curve.

`QuantileTransformer`:
1. Converts each value into its **quantile / percentile rank** (between 0 and 1) based on the training data.
2. Maps that rank to a target distribution:
   - `output_distribution='uniform'` → values spread **evenly** between 0 and 1 (flat histogram).
   - `output_distribution='normal'` → values follow a **standard normal** (bell curve).

### 2. Why we need it
It can turn **almost any** shape (even very weird ones) into uniform/normal, and it is extremely **robust to outliers** — an outlier is just "rank 100%".

### 3. How it works — numerical example
Values: `[1, 2, 3, 1000]`
- Ranks → quantiles: `[0, 0.33, 0.67, 1.0]`
- The outlier 1000 is simply "the top value" — its huge size no longer matters.

### Advantages and disadvantages

| ✅ Advantages | ❌ Disadvantages |
|---|---|
| Works for any shape, including multi-modal | **Non-linear distortion**: distances between values are changed (equal gaps may become unequal) |
| Very robust to outliers | Needs enough data to estimate quantiles well |
| Guarantees an (approximately) uniform/normal output | Values beyond the training range are **clipped** to the min/max |
| Handles negatives and zeros | Less interpretable; can hurt linear relationships |

````python
from sklearn.preprocessing import QuantileTransformer

spend = money[['Spending']]

qt_uniform = QuantileTransformer(output_distribution='uniform', n_quantiles=500, random_state=0)
qt_normal = QuantileTransformer(output_distribution='normal', n_quantiles=500, random_state=0)
# n_quantiles must not exceed the number of rows (default 1000)

spend_u = qt_uniform.fit_transform(spend).ravel()
spend_n = qt_normal.fit_transform(spend).ravel()

fig, axes = plt.subplots(1, 3, figsize=(16, 3.8))
sns.histplot(spend['Spending'], bins=40, ax=axes[0]); axes[0].set_title(f"Original (skew {spend['Spending'].skew():.2f})")
sns.histplot(spend_u, bins=40, ax=axes[1]); axes[1].set_title('Quantile → uniform')
sns.histplot(spend_n, bins=40, ax=axes[2]); axes[2].set_title(f'Quantile → normal (skew {pd.Series(spend_n).skew():.2f})')
plt.tight_layout(); plt.show()
````

````python
tiny = pd.DataFrame({'x': [1, 2, 3, 1000]})
qt = QuantileTransformer(n_quantiles=4, output_distribution='uniform')
print(pd.DataFrame({'x': tiny['x'], 'quantile': qt.fit_transform(tiny).ravel().round(3)}))

print('\nNew value beyond the training max (5000) →', qt.transform(pd.DataFrame({'x': [5000]})).ravel(),
      '(clipped to 1)')
````

### Expected output
- Uniform: a flat histogram between 0 and 1.
- Normal: a bell shape… except the **60 zeros** all share the same rank, so they form a **spike** on the left. Tied values can't be spread apart.
- The tiny example shows 1000 becomes just `1.0`.

### When to use
Extremely skewed or odd-shaped features, heavy outliers, distance-based models (KNN), neural networks.

### When NOT to use
Small datasets; when relationships should stay linear (linear regression interpretation); when extrapolating beyond the training range matters.

### Common mistakes
- Leaving `n_quantiles` at the default 1000 when you have fewer rows (scikit-learn warns and lowers it).
- Believing the output is "the same data, just normal" — it is a **rank-based** representation.

# 17. PowerTransformer — `fit`, `transform`, `fit_transform`

```python
PowerTransformer(method='yeo-johnson')   # default; any values
PowerTransformer(method='box-cox')       # strictly positive only
```

Useful parameters/attributes:
- `standardize=True` (default) → output mean 0, std 1.
- `.lambdas_` → the learned λ for each column.
- `.inverse_transform()` → go back to original units.

### `fit` vs `transform` vs `fit_transform`

| Method | What it does | Use on |
|---|---|---|
| `fit(X_train)` | **Learns** λ (and mean/std for standardizing) | **Train only** |
| `transform(X)` | **Applies** the learned λ | Train **and** test |
| `fit_transform(X_train)` | Learn + apply in one step | **Train only** |

```text
X_train ──fit──▶ [learns λ, mean, std]
                         │
              transform  │  transform
             ┌───────────┴───────────┐
             ▼                       ▼
     X_train_transformed     X_test_transformed
```

````python
cust = pd.DataFrame({
    'Income': rs.lognormal(11, 0.6, 400),
    'Visits': rs.poisson(rs.gamma(0.8, 10, 400)),
    'Balance_k': np.concatenate([rs.lognormal(9, 1, 340), -rs.lognormal(7, 1, 60)]) / 1000,   # in THOUSANDS (Section 15 tip)
})

X_train, X_test = train_test_split(cust, test_size=0.25, random_state=42)

pt = PowerTransformer(method='yeo-johnson')
pt.fit(X_train)                                  # 1) learn λ from TRAIN only
X_train_t = pt.transform(X_train)                # 2) apply to train
X_test_t = pt.transform(X_test)                  # 3) apply the SAME λ to test

print('Learned λ per column:', dict(zip(cust.columns, pt.lambdas_.round(3))))
print('\nSkew train before:', X_train.skew().round(2).to_dict())
print('Skew train after :', pd.DataFrame(X_train_t, columns=cust.columns).skew().round(2).to_dict())
print('Skew test  after :', pd.DataFrame(X_test_t, columns=cust.columns).skew().round(2).to_dict())

# Reverse
back = pt.inverse_transform(X_test_t)
print('\nInverse transform matches original test data:', np.allclose(back, X_test.values))
````

### Avoiding data leakage
- ✅ `fit` only on training data.
- ✅ Put the transformer **inside a Pipeline** so cross-validation re-fits it on each training fold automatically (Section 21).
- ❌ Never `fit` or `fit_transform` on the test set or on train + test together (Section 22).

### Common mistakes
- `pt.fit_transform(X_test)` — learns a new λ from the test data. Wrong!
- Using `method='box-cox'` on a column that could contain zeros in future data.

# 18. Transformation vs Scaling (VERY IMPORTANT)

### Transformation — changes the **shape**
`log`, `sqrt`, Box-Cox, Yeo-Johnson, Quantile → the histogram looks different afterwards.

### Standardization — changes the **scale** (not the shape)
```text
z = (x − mean) / std   →  mean = 0, standard deviation = 1
```

### Normalization — usually rescales to a **range** (or a vector norm)
```text
Min-Max: (x − min) / (max − min)   →  values in [0, 1]
(Normalizer in scikit-learn rescales each ROW to length 1 — a different idea.)
```

### Robust scaling — scale using **median and IQR**
```text
(x − median) / IQR   →  less affected by outliers, shape unchanged
```

A **linear** operation (subtract, divide by a constant) can **never** change the skewness. Only a **non-linear** function (log, sqrt, power, quantile) can.

````python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

x = money[['Salary']]
methods = {
    'Original': x.values.ravel(),
    'Log': np.log(x.values.ravel()),
    'MinMaxScaler': MinMaxScaler().fit_transform(x).ravel(),
    'StandardScaler': StandardScaler().fit_transform(x).ravel(),
    'RobustScaler': RobustScaler().fit_transform(x).ravel(),
    'PowerTransformer (YJ)': PowerTransformer().fit_transform(x).ravel(),
    'QuantileTransformer (normal)': QuantileTransformer(output_distribution='normal', n_quantiles=500).fit_transform(x).ravel(),
}

compare = pd.DataFrame({name: {'min': v.min(), 'median': np.median(v), 'mean': v.mean(),
                               'max': v.max(), 'std': v.std(), 'skew': stats.skew(v)}
                        for name, v in methods.items()}).T.round(3)
compare
````

````python
fig, axes = plt.subplots(2, 4, figsize=(18, 7))
for ax, (name, v) in zip(axes.ravel(), methods.items()):
    sns.histplot(v, bins=40, ax=ax)
    ax.set_title(f'{name}\nskew = {stats.skew(v):.2f}', fontsize=10)
axes.ravel()[-1].axis('off')
plt.tight_layout(); plt.show()
````

### Expected output
- **MinMax, Standard, Robust**: the histograms look **identical** to the original (only the x-axis numbers change) and the skew is **exactly the same**.
- **Log, PowerTransformer, QuantileTransformer**: the shape changes and the skew moves close to 0.

### Detailed comparison table

| Method | Changes shape? | Output range / center | Handles negatives/zeros | Outlier-robust? | Typical use |
|---|---|---|---|---|---|
| **Log / log1p** | ✅ Yes | Unbounded | ❌ / log1p: zeros ✅ | Reduces influence | Right-skewed positive features |
| **MinMaxScaler** | ❌ No | [0, 1] | ✅ | ❌ (outliers squash the rest) | Neural nets, bounded inputs, image pixels |
| **StandardScaler** | ❌ No | mean 0, std 1 | ✅ | ❌ (mean/std affected) | Linear models, SVM, KNN, PCA |
| **RobustScaler** | ❌ No | median 0, IQR 1 | ✅ | ✅ (scale only) | Features with outliers |
| **PowerTransformer** | ✅ Yes | ≈ mean 0, std 1 (standardize=True) | YJ ✅ / BC ❌ | Partly | Skewed features for linear/distance models |
| **QuantileTransformer** | ✅ Yes (strongly) | [0,1] or ≈ N(0,1) | ✅ | ✅✅ Very | Heavy outliers, weird shapes |

### Transformation + scaling together
They solve **different** problems, so it's common to do **both**:
```text
log / Yeo-Johnson  →  fixes the SHAPE
StandardScaler     →  fixes the SCALE
```
(`PowerTransformer` with `standardize=True` already does both.)

### Common mistakes
- "I used StandardScaler, so my data is normal now." — **No**, the shape is unchanged.
- Using MinMaxScaler on data with extreme outliers → almost all values end up squeezed near 0.

# 19. Transformation and Outliers

### 1. What it means
Transformations like log **compress** large values, so extreme values become **less extreme relative to the rest**. They do **not delete** outliers — the largest value is still the largest value.

### 2. Why we need it
Extreme values can dominate means, standard deviations, regression lines, and distances. Reducing their **influence** (without throwing away real data) is often better than deleting them (Topic 5).

### 4. Numerical example
Values: `[10, 12, 15, 20, 25, 5000]`
- Original: 5000 is **200×** larger than 25.
- `log`: log(5000) = 8.52 vs log(25) = 3.22 → only **2.6×** larger.

````python
ro = np.random.default_rng(21)
income_o = pd.Series(np.concatenate([ro.normal(60000, 12000, 95), [400000, 650000, 900000, 1_500_000, 2_000_000]]),
                     name='Income')
df_o = pd.DataFrame({'Original': income_o})
df_o['Log'] = np.log(income_o)
df_o['RobustScaled'] = RobustScaler().fit_transform(df_o[['Original']]).ravel()

def iqr_outliers(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return int(((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum())

def max_z(s):
    return ((s - s.mean()) / s.std()).abs().max()

summary_o = pd.DataFrame({
    'skew': df_o.skew(),
    'IQR outliers': df_o.apply(iqr_outliers),
    'max |z-score|': df_o.apply(max_z),
    'max / median': df_o['Original'].max() / df_o['Original'].median(),
}).round(2)
summary_o.loc[['Log', 'RobustScaled'], 'max / median'] = [
    df_o['Log'].max() / df_o['Log'].median(), np.nan]
summary_o.round(2)
````

````python
fig, axes = plt.subplots(1, 3, figsize=(16, 3.8))
for ax, col in zip(axes, df_o.columns):
    sns.boxplot(x=df_o[col], ax=ax)
    ax.set_title(f'{col} (skew = {df_o[col].skew():.2f})')
plt.tight_layout(); plt.show()
````

### Expected output — interpretation
- **RobustScaled:** the boxplot looks **exactly like the original** (same outliers, same skew). Robust scaling uses the median and IQR so the *scaling* isn't distorted by outliers — but it **doesn't change the outliers themselves**.
- **Log:** the extreme incomes are pulled much closer; skew and the max z-score drop a lot. They are still the largest values, and some may still be flagged by the IQR rule — the transformation **reduced their influence but did not remove them**.

### When to use
When extreme values are **real** (not errors) and you want to keep them but limit their influence.

### When NOT to use
When the "outliers" are **data errors** (e.g. age = 999) — fix or remove those instead. Transforming an error just hides it.

### Common mistakes
- Saying "log removed the outliers".
- Using RobustScaler expecting the skew to disappear.

# 20. Transformation and Linear Regression

### Scenario
A company's **sales** grow with **advertising budget**, but with **diminishing returns**: the first 10,000 rupees of advertising help a lot; going from 1,000,000 to 1,010,000 barely matters. The ad budget itself is right-skewed (a few huge campaigns).

A straight line on the raw budget cannot capture a curve. A straight line on **log(budget)** can.

- **Model 1:** `Sales ~ Ad_Budget`
- **Model 2:** `Sales ~ log(Ad_Budget)`

Metrics:
- **R²** — share of the variation in sales explained (1 = perfect).
- **MAE** — average absolute error (in sales units).
- **RMSE** — root mean squared error; punishes large errors more.
- **Residuals** — actual − predicted. A good model's residuals look like **random noise around 0**, with **no pattern**.

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error

rr = np.random.default_rng(20)
n = 400
ad_budget = rr.lognormal(mean=10, sigma=1.0, size=n)                       # right-skewed budgets
sales = 5000 + 8000 * np.log(ad_budget) + rr.normal(0, 4000, n)           # diminishing returns

ads = pd.DataFrame({'Ad_Budget': ad_budget, 'Sales': sales})
ads['Log_Ad_Budget'] = np.log(ads['Ad_Budget'])

train_ads, test_ads = train_test_split(ads, test_size=0.3, random_state=0)

def evaluate(feature):
    model = LinearRegression().fit(train_ads[[feature]], train_ads['Sales'])
    pred = model.predict(test_ads[[feature]])
    return model, pred, {
        'Feature': feature,
        'R2': r2_score(test_ads['Sales'], pred),
        'MAE': mean_absolute_error(test_ads['Sales'], pred),
        'RMSE': np.sqrt(mean_squared_error(test_ads['Sales'], pred)),
    }

m1, pred1, res1 = evaluate('Ad_Budget')
m2, pred2, res2 = evaluate('Log_Ad_Budget')
pd.DataFrame([res1, res2]).set_index('Feature').round(3)
````

````python
fig, axes = plt.subplots(2, 2, figsize=(13, 8))

# Fitted lines
order = np.argsort(test_ads['Ad_Budget'].values)
axes[0, 0].scatter(test_ads['Ad_Budget'], test_ads['Sales'], alpha=0.5)
axes[0, 0].plot(test_ads['Ad_Budget'].values[order], pred1[order], color='red')
axes[0, 0].set_title('Model 1: straight line on raw budget'); axes[0, 0].set_xlabel('Ad_Budget')

axes[0, 1].scatter(test_ads['Log_Ad_Budget'], test_ads['Sales'], alpha=0.5)
axes[0, 1].plot(test_ads['Log_Ad_Budget'].values[order], pred2[order], color='red')
axes[0, 1].set_title('Model 2: straight line on log(budget)'); axes[0, 1].set_xlabel('log(Ad_Budget)')

# Residuals vs predicted
for ax, pred, title in [(axes[1, 0], pred1, 'Model 1 residuals'), (axes[1, 1], pred2, 'Model 2 residuals')]:
    resid = test_ads['Sales'] - pred
    ax.scatter(pred, resid, alpha=0.5)
    ax.axhline(0, color='red', ls='--')
    ax.set_title(title); ax.set_xlabel('Predicted sales'); ax.set_ylabel('Residual (actual − predicted)')
plt.tight_layout(); plt.show()
````

### Expected output & explanation
- **Model 2** has a clearly **higher R²** and **lower MAE/RMSE**.
- **Model 1 residuals** show a **pattern**: a curve, and most points squeezed on the left — the straight line is systematically wrong for small and large budgets.
- **Model 2 residuals** look like a random horizontal band around 0 → the model captured the relationship.

### Why transformation can help linear models
1. It can make a **curved** relationship **linear** (the model can only draw straight lines/planes).
2. It reduces the **leverage** of a few huge values that would otherwise pull the line.
3. It can make residuals more **constant in spread** (homoscedastic) and closer to normal — needed for reliable confidence intervals and p-values.

### Interpretation after log
`Sales = a + b · log(Ad_Budget)` → **increasing the budget by 1%** increases sales by about **b / 100** units. A 10% budget increase adds about `b · log(1.1) ≈ 0.095 · b`.

### When NOT to expect improvement
If the true relationship is already linear in the raw feature, the log version will be **worse**. Always compare on a **test set** or with cross-validation.

# 21. Transformation and ML Pipelines

### Correct workflow

```text
Train data
   ↓
Fit transformation   (learn λ / quantiles / mean & std)
   ↓
Transform train
   ↓
Transform test using the SAME fitted transformer
```

### ❌ Incorrect vs ✅ correct

````python
rpl = np.random.default_rng(22)
n = 500
data21 = pd.DataFrame({
    'Income': rpl.lognormal(11, 0.7, n),
    'Visits': rpl.poisson(rpl.gamma(0.8, 10, n)),
    'Age': rpl.uniform(18, 70, n),
    'City': rpl.choice(['Lahore', 'Karachi', 'Islamabad'], n),
})
data21['Spend'] = (2000 * np.log(data21['Income']) + 300 * np.sqrt(data21['Visits'])
                   + 20 * data21['Age'] + data21['City'].map({'Lahore': 0, 'Karachi': 1500, 'Islamabad': 3000})
                   + rpl.normal(0, 1200, n))

X = data21.drop(columns='Spend')
y = data21['Spend']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

# ❌ INCORRECT: separate transformers fitted on train and test
wrong_train = PowerTransformer().fit_transform(X_train[['Income']])
wrong_test = PowerTransformer().fit_transform(X_test[['Income']])      # learns a DIFFERENT λ from test data
print('λ learned on train:', PowerTransformer().fit(X_train[['Income']]).lambdas_.round(4))
print('λ learned on test :', PowerTransformer().fit(X_test[['Income']]).lambdas_.round(4), '← different! inconsistent features')

# ✅ CORRECT: one transformer, fitted on train, applied to both
pt = PowerTransformer().fit(X_train[['Income']])
right_train = pt.transform(X_train[['Income']])
right_test = pt.transform(X_test[['Income']])
````

### A full pipeline with `ColumnTransformer`

Different columns need different treatment:
- skewed numeric → `PowerTransformer` (Yeo-Johnson; standardizes too),
- roughly symmetric numeric → `StandardScaler`,
- categorical → `OneHotEncoder`.

````python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import cross_val_score, KFold

skewed_cols = ['Income', 'Visits']
normal_cols = ['Age']
cat_cols = ['City']

preprocess = ColumnTransformer([
    ('skewed', PowerTransformer(method='yeo-johnson'), skewed_cols),
    ('numeric', StandardScaler(), normal_cols),
    ('categorical', OneHotEncoder(handle_unknown='ignore'), cat_cols),
])

pipe_transformed = Pipeline([('prep', preprocess), ('model', LinearRegression())])

# Baseline: only scaling, no shape transformation
preprocess_scale_only = ColumnTransformer([
    ('numeric', StandardScaler(), skewed_cols + normal_cols),
    ('categorical', OneHotEncoder(handle_unknown='ignore'), cat_cols),
])
pipe_scaled = Pipeline([('prep', preprocess_scale_only), ('model', LinearRegression())])

cv = KFold(5, shuffle=True, random_state=0)
for name, p in [('Scaling only', pipe_scaled), ('PowerTransformer + scaling', pipe_transformed)]:
    cv_r2 = cross_val_score(p, X_train, y_train, cv=cv, scoring='r2')     # transformer re-fitted in each fold
    p.fit(X_train, y_train)
    print(f'{name:<28} CV R² = {cv_r2.mean():.3f}   Test R² = {p.score(X_test, y_test):.3f}')

print('\nLearned λ inside the pipeline:',
      dict(zip(skewed_cols, pipe_transformed.named_steps['prep'].named_transformers_['skewed'].lambdas_.round(3))))
````

### Why pipelines matter
- `pipe.fit(X_train, y_train)` fits **every** step on training data only.
- `pipe.predict(X_test)` only **transforms** the test data.
- `cross_val_score(pipe, ...)` re-fits the transformer inside **each fold** → no leakage during model selection.
- One object to save and deploy → production uses exactly the same λ values.

# 22. Data Leakage in Transformations

### What it means
**Data leakage** = information from the test set (or the future) sneaks into training. Transformers **learn** things from data:

| Transformer | What it learns |
|---|---|
| `PowerTransformer` | λ, mean, std |
| `QuantileTransformer` | The quantiles (ranks) of the data |
| `StandardScaler` | mean, std |
| `MinMaxScaler` | min, max |
| `RobustScaler` | median, IQR |
| Shift constant for log | min of the data |

If they learn from the **test set**, your evaluation is no longer an honest simulation of "new, unseen data".

### ❌ Bad
```python
transformer.fit(all_data)          # sees the test rows
X_all = transformer.transform(all_data)
X_train, X_test = split(X_all)
```

### ✅ Correct
```text
Split
 ↓
Fit transformer on train
 ↓
Transform train
 ↓
Transform test
```

Note: **stateless** transforms such as `np.log1p`, `np.sqrt`, `np.cbrt` learn **nothing** from the data, so applying them before the split doesn't leak. The danger is with transformers that **fit parameters**.

### Complete example
The quantile transformer below is where leakage shows most clearly: when fitted on all data, the test rows' **own ranks** help define the mapping.

````python
from sklearn.neighbors import KNeighborsRegressor

rl = np.random.default_rng(99)
n = 300
leak_df = pd.DataFrame({'Income': rl.lognormal(11, 1.0, n), 'Visits': rl.poisson(rl.gamma(0.7, 12, n))})
leak_y = 1500 * np.log(leak_df['Income']) + 400 * np.sqrt(leak_df['Visits']) + rl.normal(0, 1500, n)

X_tr, X_te, y_tr, y_te = train_test_split(leak_df, leak_y, test_size=0.3, random_state=1)

# ❌ BAD: fit on ALL data, then split
qt_bad = QuantileTransformer(n_quantiles=200, output_distribution='normal').fit(leak_df)
knn_bad = KNeighborsRegressor(10).fit(qt_bad.transform(X_tr), y_tr)
r2_bad = r2_score(y_te, knn_bad.predict(qt_bad.transform(X_te)))

# ✅ GOOD: fit on TRAIN only
qt_good = QuantileTransformer(n_quantiles=200, output_distribution='normal').fit(X_tr)
knn_good = KNeighborsRegressor(10).fit(qt_good.transform(X_tr), y_tr)
r2_good = r2_score(y_te, knn_good.predict(qt_good.transform(X_te)))

print(f'R² with leakage (fit on all data): {r2_bad:.4f}')
print(f'R² correct (fit on train only)   : {r2_good:.4f}')

# The learned parameters themselves differ:
print('\nMedian income learned — all data :', round(float(np.median(qt_bad.quantiles_[:, 0])), 0))
print('Median income learned — train only:', round(float(np.median(qt_good.quantiles_[:, 0])), 0))
````

### Interpretation
- The two scores are often **close** on a dataset like this — leakage from a transformer is usually **small**, but it is still **wrong**, and on small datasets, time-series data, or with target-based transformations it can be large.
- The parameters differ because the leaky version **used information from the test rows**.
- The correct habit is simple: **split first, and put every learned transformation inside a Pipeline.**

### Common mistakes
- `df['x'] = PowerTransformer().fit_transform(df[['x']])` in the EDA notebook, then splitting later.
- Choosing the transformation by looking at the skewness of the **whole** dataset — decide using the training set.

# 23. How to Choose a Transformation — Decision Guide

```text
Start: is the feature numeric and skewed?  (check histogram + skewness on TRAINING data)
   │
   ├── No  → usually no transformation needed (maybe just scaling)
   │
   └── Yes
        │
        ├── Is my model tree-based and I only care about prediction?
        │        └── Yes → usually skip it (test if unsure)
        │
        ├── Right-skewed?
        │        ├── All values > 0         → log  /  Box-Cox  /  sqrt (milder)
        │        ├── Zeros, no negatives    → log1p  /  sqrt  /  Yeo-Johnson
        │        └── Negatives present      → Yeo-Johnson  /  cube root  /  signed log
        │
        ├── Left-skewed?
        │        └── Yeo-Johnson / Box-Cox (λ > 1)  or  reflect: log(max + 1 − x)
        │
        ├── Is it a count (purchases, visits, tickets)?
        │        └── sqrt or log1p
        │
        ├── Is a rate/ratio the natural meaning?
        │        └── consider reciprocal (1/x), careful with zeros
        │
        ├── Need approximately normal output / heavy outliers / weird shape?
        │        └── QuantileTransformer / PowerTransformer
        │
        └── Always: compare with cross-validation and keep it only if it helps
```

This is a **guideline, not a strict rule**. The final judge is:
1. model performance on validation data,
2. residual/diagnostic plots,
3. interpretability needs,
4. the behaviour of future data (zeros? negatives? new extremes?).

# 24. Before vs After Comparison — Reusable Function

````python
def compare_transformations(df, column, plot=True):
    '''
    Compare Original, Log (log1p), Square root, Cube root and Yeo-Johnson for one column.
    Returns a table of mean, median, std and skewness.
    Transformations that are impossible for the data (e.g. sqrt of negatives) are skipped.
    '''
    data = df[column].dropna()
    has_negative = (data < 0).any()

    versions = {'Original': data}
    if not has_negative:
        versions['Log (log1p)'] = np.log1p(data)           # safe with zeros
        versions['Square root'] = np.sqrt(data)
    versions['Cube root'] = np.cbrt(data)                   # works with negatives
    # Yeo-Johnson: rescale first ONLY when there are negative values (Section 15 rule)
    if has_negative:
        yj = make_pipeline(StandardScaler(), PowerTransformer(method='yeo-johnson', standardize=False))
    else:
        yj = PowerTransformer(method='yeo-johnson', standardize=False)
    versions['Yeo-Johnson'] = pd.Series(yj.fit_transform(data.to_frame()).ravel(), index=data.index)

    table = pd.DataFrame({
        name: {'Mean': v.mean(), 'Median': v.median(), 'Std': v.std(), 'Skewness': v.skew()}
        for name, v in versions.items()
    }).T
    table['|Skewness|'] = table['Skewness'].abs()

    if plot:
        fig, axes = plt.subplots(1, len(versions), figsize=(4 * len(versions), 3.5))
        for ax, (name, v) in zip(axes, versions.items()):
            sns.histplot(v, bins=35, kde=True, ax=ax)
            ax.set_title(f'{name}\nskew = {v.skew():.2f}', fontsize=10)
            ax.set_xlabel('')
        fig.suptitle(f'Transformations of {column}', y=1.03)
        plt.tight_layout(); plt.show()

    return table.sort_values('|Skewness|').round(3)

compare_transformations(money, 'Salary')
````

````python
compare_transformations(acct, 'Balance')        # contains negatives → log/sqrt skipped automatically
````

### Line-by-line (key parts)
- `has_negative` decides which transformations are **mathematically allowed**.
- Yeo-Johnson is applied directly to non-negative data, and **after `StandardScaler`** when the column has negative values (the Section 15 rule — Yeo-Johnson depends on units). `standardize=False` keeps the result un-standardized; the skewness is what matters here.
- The table is sorted by **absolute** skewness → the most symmetric result is on top.

### ⚠️ Remember
The **smallest skewness is not automatically the best choice**. Prefer the transformation that:
1. reduces skew *enough*,
2. is simple and explainable,
3. works for future data (zeros/negatives),
4. improves validation performance.

# 26. Customer Spending Transformation Project

**Goal:** predict each customer's **MonthlySpending** from their profile, and decide which transformations actually help.

Columns: `CustomerID, Age, Income, MonthlySpending, Purchases, WebsiteVisits, SupportTickets, TotalOrders, AccountBalance`

We intentionally create realistic skewness:
- `Income`, `MonthlySpending` → lognormal (right-skewed, positive)
- `Purchases`, `WebsiteVisits`, `SupportTickets`, `TotalOrders` → counts with long tails and zeros
- `AccountBalance` → mostly positive but some customers are in **debt** (negative values)
- `Age` → roughly symmetric

## Step 0 — Create the dataset (100 customers)

````python
rproj = np.random.default_rng(2026)
n = 100

age = rproj.normal(38, 11, n).clip(18, 75).round(0)
income = rproj.lognormal(11.2, 0.6, n).round(-2)
purchases = rproj.poisson(rproj.gamma(1.0, 5, n))
website_visits = rproj.poisson(rproj.gamma(0.9, 25, n)) + purchases
support_tickets = rproj.poisson(rproj.gamma(0.5, 2, n))
total_orders = purchases * 6 + rproj.poisson(rproj.gamma(1.0, 8, n))          # lifetime orders
account_balance = np.where(rproj.random(n) < 0.15,
                           -rproj.lognormal(8, 0.8, n),                          # customers in debt
                           rproj.lognormal(9.5, 1.1, n)).round(0)

# Spending depends on log(income) and sqrt(purchases) → multiplicative / diminishing effects
log_spend = (-6.5 + 1.0 * np.log(income) + 0.25 * np.sqrt(purchases)
             + 0.03 * np.sqrt(website_visits) - 0.05 * support_tickets + rproj.normal(0, 0.25, n))
monthly_spending = np.exp(log_spend).round(0)

customers = pd.DataFrame({
    'CustomerID': [f'C{1000 + i}' for i in range(n)],
    'Age': age,
    'Income': income,
    'MonthlySpending': monthly_spending,
    'Purchases': purchases,
    'WebsiteVisits': website_visits,
    'SupportTickets': support_tickets,
    'TotalOrders': total_orders,
    'AccountBalance': account_balance,
})
customers.head(10)
````

## Step 2 & 3 — Detect skewed columns and calculate skewness

````python
numeric = customers.drop(columns='CustomerID')
skew_report = pd.DataFrame({
    'skew': numeric.skew(),
    'mean': numeric.mean(),
    'median': numeric.median(),
    'has_zero': (numeric == 0).any(),
    'has_negative': (numeric < 0).any(),
})
skew_report['skew_level'] = pd.cut(skew_report['skew'].abs(), bins=[-0.01, 0.5, 1, np.inf],
                                   labels=['≈ symmetric', 'moderate', 'high'])
skew_report.sort_values('skew', key=abs, ascending=False).round(2)
````

## Step 7 — Choose transformations

| Column | Observation | Choice | Reason |
|---|---|---|---|
| `Age` | ≈ symmetric | **StandardScaler only** | No skew problem |
| `Income` | Right-skewed, strictly positive | **Yeo-Johnson** (≈ log) | Classic multiplicative variable; Yeo-Johnson is safe in a pipeline |
| `Purchases`, `WebsiteVisits`, `TotalOrders` | Counts with long tails, zeros possible | **Yeo-Johnson** | Handles zeros; learns sqrt-/log-like strength |
| `SupportTickets` | Many zeros, few large counts | **Yeo-Johnson** (+ optionally a `Has_Ticket` flag) | A spike at 0 cannot be fully removed by any transformation |
| `AccountBalance` | Negative values (debt) | **StandardScaler → Yeo-Johnson** | log/sqrt/Box-Cox impossible; rescale first because of the negatives (Section 15) |
| `MonthlySpending` (**target**) | Right-skewed, positive | Try **log target** via `TransformedTargetRegressor` | Spending is multiplicative; errors are more naturally relative |

`CustomerID` is an identifier → excluded.

## Step 8 — Build preprocessing pipelines

````python
from sklearn.compose import TransformedTargetRegressor
from sklearn.ensemble import RandomForestRegressor

target = 'MonthlySpending'
feature_cols = ['Age', 'Income', 'Purchases', 'WebsiteVisits', 'SupportTickets', 'TotalOrders', 'AccountBalance']
skewed_features = ['Income', 'Purchases', 'WebsiteVisits', 'SupportTickets', 'TotalOrders', 'AccountBalance']
positive_skewed = ['Income', 'Purchases', 'WebsiteVisits', 'SupportTickets', 'TotalOrders']   # no negatives
signed_skewed = ['AccountBalance']                                                             # has negatives

X = customers[feature_cols]
y = customers[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

scale_only = ColumnTransformer([('scale', StandardScaler(), feature_cols)])

transform_and_scale = ColumnTransformer([
    ('power', PowerTransformer(method='yeo-johnson'), positive_skewed),                        # shape + scale
    ('power_signed', make_pipeline(StandardScaler(), PowerTransformer(method='yeo-johnson')),
     signed_skewed),                                                                           # rescale first (negatives)
    ('scale', StandardScaler(), ['Age']),                                                      # scale only
])

models = {
    'A. Linear — scaling only':
        Pipeline([('prep', scale_only), ('model', LinearRegression())]),
    'B. Linear — Yeo-Johnson features':
        Pipeline([('prep', transform_and_scale), ('model', LinearRegression())]),
    'C. Linear — Yeo-Johnson features + log target':
        TransformedTargetRegressor(
            regressor=Pipeline([('prep', transform_and_scale), ('model', LinearRegression())]),
            func=np.log1p, inverse_func=np.expm1),                         # predict log1p(spend), return rupees
    'D. Random Forest — raw features':
        RandomForestRegressor(n_estimators=300, random_state=0),
    'E. Random Forest — Yeo-Johnson features':
        Pipeline([('prep', transform_and_scale), ('model', RandomForestRegressor(n_estimators=300, random_state=0))]),
}
print('Pipelines ready:', list(models))
````

## Step 9 & 10 — Train models and compare performance

````python
cv = KFold(5, shuffle=True, random_state=0)
rows = []
for name, model in models.items():
    cv_r2 = cross_val_score(model, X_train, y_train, cv=cv, scoring='r2')
    cv_mae = -cross_val_score(model, X_train, y_train, cv=cv, scoring='neg_mean_absolute_error')
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    rows.append({'Model': name,
                 'CV R² (mean)': cv_r2.mean(), 'CV MAE': cv_mae.mean(),
                 'Test R²': r2_score(y_test, pred),
                 'Test MAE': mean_absolute_error(y_test, pred),
                 'Test RMSE': np.sqrt(mean_squared_error(y_test, pred))})

results = pd.DataFrame(rows).set_index('Model').round(3)
results
````

````python
fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))
for ax, name in zip(axes, ['A. Linear — scaling only', 'C. Linear — Yeo-Johnson features + log target',
                           'D. Random Forest — raw features']):
    pred = models[name].predict(X_test)
    ax.scatter(y_test, pred, alpha=0.7)
    lim = [0, max(y_test.max(), pred.max()) * 1.05]
    ax.plot(lim, lim, 'r--', label='perfect prediction')
    ax.set_xlim(lim); ax.set_ylim(lim)
    ax.set_title(f'{name}\nTest R² = {r2_score(y_test, pred):.2f}', fontsize=10)
    ax.set_xlabel('Actual spending'); ax.set_ylabel('Predicted')
    ax.legend()
plt.tight_layout(); plt.show()
````

````python
# Skewness of the training features after the chosen transformation (fitted on train only)
fitted_prep = models['B. Linear — Yeo-Johnson features'].named_steps['prep']
Xt = pd.DataFrame(fitted_prep.transform(X_train), columns=positive_skewed + signed_skewed + ['Age'])
lambdas = np.concatenate([fitted_prep.named_transformers_['power'].lambdas_,
                          fitted_prep.named_transformers_['power_signed'][-1].lambdas_])

pd.DataFrame({'Skew before (train)': X_train[positive_skewed + signed_skewed + ['Age']].skew(),
              'Skew after (train)': Xt.skew(),
              'Learned λ': list(lambdas.round(3)) + [np.nan]}).round(3)
````

## Step 11 — Explain the final decisions

**1. Skewness detection**
- `Income`, `MonthlySpending`, the count columns and `AccountBalance` are clearly skewed; `Age` is roughly symmetric.
- `AccountBalance` has negative values and several columns contain zeros → only **Yeo-Johnson** (or cube root) is safe for all of them in one consistent step.

**2. Transformations**
- Yeo-Johnson brought the feature skewness close to 0 for most columns. `SupportTickets` improves (2.3 → ≈ 0.4) but keeps a spike at 0 — no transformation can separate identical values.
- The learned λ values tell a story: `Income` and `Purchases` ≈ 0 → **log-like**; `WebsiteVisits` and `TotalOrders` ≈ 0.15–0.25 → between log and square root; `SupportTickets` ≈ −1 → a very strong, reciprocal-like squeeze. Yeo-Johnson rediscovered the kinds of transformations we would have picked by hand.

**3. Model comparison** (use the **CV** columns as the main evidence — the test set has only 25 customers)
- **A → B: transforming only the features made the linear model WORSE.** Why? In this data, spending is roughly **proportional** to income (`spending ∝ income × other factors`). In raw units that is close to a straight line, so model A already did reasonably well. After log-transforming income but *not* spending, the relationship between `log(income)` and raw spending becomes a **curve** (exponential) — harder for a straight line.
- **B → C: transforming the target too gives the best model.** Taking `log(spending)` turns the multiplicative relationship into an additive one: `log(spending) ≈ log(income) + effect of purchases + …` — a straight line in the transformed space. Model C has the highest CV R² and the lowest MAE.
- **D vs E (Random Forest):** practically identical — trees are largely insensitive to monotonic feature transformations (Section 7).
- The best linear model (C) beats the Random Forest here, and it is much easier to explain.

> **Key lesson:** transformation is not "always good". Transforming the wrong side of a relationship can hurt. Think about **how the variables relate** (additive or multiplicative?) and let cross-validation decide.

**4. Final choice**
```text
Features:  Yeo-Johnson on skewed columns, StandardScaler on Age   (inside a Pipeline)
Target:    log1p(MonthlySpending), converted back with expm1      (TransformedTargetRegressor)
Model:     LinearRegression
Why:       best / near-best cross-validated performance, simple, interpretable, leakage-safe
```

**5. Leakage safety**
Every transformer was inside a `Pipeline` / `TransformedTargetRegressor`, split first, and cross-validation re-fitted them on each fold.

# 27. Cheat Sheet

| Technique | Handles Negative? | Handles Zero? | Main Purpose | Common Use | Code |
|---|---|---|---|---|---|
| **Log** | ❌ | ❌ | Strongly reduce right skew | Income, prices (always > 0) | `np.log(x)` |
| **log1p** | ❌ (x > −1 only) | ✅ | Reduce right skew with zeros | Counts, spending, visits | `np.log1p(x)`, reverse `np.expm1` |
| **sqrt** | ❌ | ✅ | Mildly reduce right skew; stabilize count variance | Counts | `np.sqrt(x)` |
| **cbrt** | ✅ | ✅ | Mild compression keeping the sign | Profit/loss, changes | `np.cbrt(x)` |
| **Reciprocal** | ⚠️ reverses order differently on each side | ❌ | Very strong compression; rate conversion | Time ↔ speed | `1 / x` |
| **Box-Cox** | ❌ | ❌ | Automatically find the best power (λ) | Positive skewed data | `stats.boxcox(x)`, `PowerTransformer('box-cox')` |
| **Yeo-Johnson** | ✅ | ✅ | Automatic power for any real values | Safe default in pipelines | `PowerTransformer('yeo-johnson')` |
| **QuantileTransformer** | ✅ | ✅ | Force uniform/normal shape; robust to outliers | Heavy outliers, odd shapes, KNN/NN | `QuantileTransformer(output_distribution='normal')` |
| **StandardScaler** | ✅ | ✅ | Mean 0, std 1 (**shape unchanged**) | Linear models, SVM, KNN, PCA | `StandardScaler()` |
| **MinMaxScaler** | ✅ | ✅ | Rescale to [0, 1] (**shape unchanged**) | Neural nets, bounded inputs | `MinMaxScaler()` |
| **RobustScaler** | ✅ | ✅ | Scale by median & IQR (**shape unchanged**) | Features with outliers | `RobustScaler()` |

### Key facts to remember

```text
Right skew → long RIGHT tail → mean > median → skew > 0
Left skew  → long LEFT tail  → mean < median → skew < 0
Pandas .skew() = SciPy skew(bias=False)
|skew| < 0.5 ≈ symmetric, 0.5–1 moderate, > 1 high      (guideline, NOT a law)

Compression strength:  cbrt < sqrt < log < 1/x
Box-Cox λ:  1 = none | 0.5 = sqrt | 0 = log | −1 = reciprocal
Scaling changes SCALE; transformation changes SHAPE.
Trees: mostly insensitive to monotonic feature transforms. Linear/KNN/SVM/NN: can benefit.
Fit on TRAIN only. Put it in a Pipeline.
```

## How to choose the right transformation (in very simple English)

1. **Look first.** Draw a histogram and check the skewness of your **training** data.
2. **Ask if it matters.** Using trees only for prediction? You can usually skip it. Using linear models, KNN, SVM or neural networks? Transformation may help.
3. **Check the values.**
   - All values bigger than 0 → **log** (or Box-Cox).
   - Some zeros → **log1p** or **sqrt**.
   - Some negatives → **Yeo-Johnson** or **cube root**.
4. **Counts** (visits, orders, tickets) → start with **sqrt** or **log1p**.
5. **Not sure?** → use **Yeo-Johnson** inside a Pipeline — it chooses the strength for you and works with any values.
6. **Huge outliers or a very strange shape?** → try **QuantileTransformer**.
7. **Don't confuse** transformation with scaling. Scaling does not remove skew. Often you need **both**.
8. **Test it.** Keep the transformation only if **cross-validation** says it helps (or if you need it for statistical assumptions).
9. **Avoid leakage.** Split first, fit on train, transform test, use a Pipeline.
10. **Remember to reverse** a transformed **target** (e.g. `np.expm1`) before reporting predictions.
