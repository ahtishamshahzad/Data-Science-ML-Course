# 1. What Is Encoding?

Suppose we have:

| Name | Gender | City |
|---|---|---|
| Ali | Male | Lahore |
| Sara | Female | Karachi |
| Ahmed | Male | Islamabad |

The values `Male`, `Female`, `Lahore`, `Karachi`, and `Islamabad` are categorical values.

A machine-learning model usually needs numerical input.

Encoding converts categorical information into numerical representation.

Conceptually:

```text
Categorical data
       ↓
    Encoding
       ↓
Numerical representation
       ↓
Machine-learning model
```

Encoding does NOT mean that the original category itself becomes mathematically ordered. The correct encoding technique depends on the type of categorical variable.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'John'],
    'Gender': ['Male', 'Female', 'Male', 'Male'],
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

df
````

<a id="sec-2"></a>
# 2. What Is Categorical Data?

A categorical column contains values that represent groups or categories.

Examples:

```text
Gender: Male, Female
City: Lahore, Karachi, Islamabad
Color: Red, Blue, Green
Browser: Chrome, Safari, Firefox
```

There are two important types:

## Nominal categorical data

There is no meaningful order.

Example:

```text
Lahore
Karachi
Islamabad
```

There is no mathematical meaning to saying one city is greater than another.

## Ordinal categorical data

There is a meaningful order.

Example:

```text
School < Bachelor < Master < PhD
```

Another example:

```text
Low < Medium < High
```

This distinction is one of the most important ideas when choosing an encoding method.

<a id="sec-3"></a>
# 3. Binary Categories

A binary categorical variable has only two possible categories.

Examples:

```text
Male / Female
Yes / No
True / False
Pass / Fail
```

A simple representation is:

```text
No  → 0
Yes → 1
```

For binary variables, mapping to 0 and 1 is often straightforward.

````python
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male']
})

df['Gender_encoded'] = df['Gender'].map({
    'Female': 0,
    'Male': 1
})

df
````

<a id="sec-4"></a>
# 4. Encoding with `map()`

`map()` is very useful when you explicitly know the mapping you want.

Example:

```python
df['Gender'].map({'Female': 0, 'Male': 1})
```

Advantages:

- Easy to understand
- Explicit mapping
- Good for binary categories
- Good for ordinal categories when you define the correct order

Be careful: if a value does not exist in the mapping dictionary, `map()` can produce `NaN`.

````python
df = pd.DataFrame({
    'Status': ['Active', 'Inactive', 'Active', 'Active']
})

df['Status_encoded'] = df['Status'].map({
    'Inactive': 0,
    'Active': 1
})

df
````

# 5. Encoding with `replace()`

`replace()` can also replace category names with numerical values.

Example:

```python
df['Gender'] = df['Gender'].replace({
    'Male': 1,
    'Female': 0
})
```

Difference in idea:

- `map()` is commonly used to create a transformed Series from a mapping.
- `replace()` is a more general replacement operation and can work across DataFrame values.

````python
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female']
})

df['Gender_encoded'] = df['Gender'].replace({
    'Male': 1,
    'Female': 0
})

df
````

# 6. Pandas `category` Data Type

Pandas provides a special categorical data type.

You can convert a column using:

```python
df['City'] = df['City'].astype('category')
```

This is useful when a column contains repeated categorical values.

Categorical dtype can also provide a structured list of categories and can save memory in appropriate datasets.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Lahore', 'Islamabad', 'Karachi']
})

df['City'] = df['City'].astype('category')

print(df)
print(df['City'].cat.categories)
````

<a id="sec-7"></a>
# 7. `cat.codes`

After converting a Series to categorical dtype, Pandas provides category codes.

Example:

```python
df['City'].astype('category').cat.codes
```

Important warning:

`cat.codes` assigns integer codes to categories, but those codes should not automatically be interpreted as meaningful numerical order.

For example:

```text
Lahore     → 1
Karachi    → 0
Islamabad  → 2
```

does NOT mean:

```text
Islamabad > Lahore > Karachi
```

For nominal categories, one-hot encoding is often safer.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

df['City_category'] = df['City'].astype('category')
df['City_code'] = df['City_category'].cat.codes

df
````

# 8. Ordinal Encoding

Ordinal encoding is appropriate when categories have a meaningful order.

Example:

```text
School
Bachelor
Master
PhD
```

We can encode:

```text
School   → 0
Bachelor → 1
Master   → 2
PhD      → 3
```

The numerical order represents the real-world order.

````python
df = pd.DataFrame({
    'Education': [
        'Bachelor',
        'Master',
        'School',
        'PhD',
        'Master'
    ]
})

education_map = {
    'School': 0,
    'Bachelor': 1,
    'Master': 2,
    'PhD': 3
}

df['Education_encoded'] = df['Education'].map(education_map)

df
````

# 9. One-Hot Encoding

One-Hot Encoding creates a separate binary column for each category.

Suppose:

```text
City
Lahore
Karachi
Islamabad
```

One-hot representation:

```text
City        Lahore  Karachi  Islamabad
Lahore         1       0         0
Karachi        0       1         0
Islamabad      0       0         1
```

This avoids creating an artificial numerical order between nominal categories.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

encoded = pd.get_dummies(
    df,
    columns=['City'],
    dtype=int
)

encoded
````

# 10. `drop_first=True`

Pandas allows you to remove one dummy column:

```python
pd.get_dummies(
    df,
    columns=['City'],
    drop_first=True,
    dtype=int
)
```

If there are three categories, two dummy columns can be enough for many linear-model setups because the omitted category can be represented by all zeros.

However, `drop_first=True` should not be treated as a universal rule. The appropriate representation depends on the model and preprocessing design.

````python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

encoded = pd.get_dummies(
    df,
    columns=['City'],
    drop_first=True,
    dtype=int
)

encoded
````

# 13. Frequency Encoding

Frequency encoding replaces each category with how frequently it appears.

Example:

```text
Lahore      appears 3 times
Karachi     appears 2 times
Islamabad   appears 1 time
```

Frequency encoding:

```text
Lahore     → 3
Karachi    → 2
Islamabad  → 1
```

This can be useful for high-cardinality categorical variables.

````python
df = pd.DataFrame({
    'City': [
        'Lahore',
        'Lahore',
        'Lahore',
        'Karachi',
        'Karachi',
        'Islamabad'
    ]
})

frequency = df['City'].value_counts()

df['City_frequency'] = df['City'].map(frequency)

df
````

# 15. Count Encoding

Count encoding is closely related to frequency encoding.

Instead of representing the category itself, we represent how many times that category appears.

Example:

```text
Chrome → 500
Safari → 300
Firefox → 200
```

The key idea is:

```python
counts = df['Browser'].value_counts()
df['Browser_count'] = df['Browser'].map(counts)
```

````python
df = pd.DataFrame({
    'Browser': [
        'Chrome', 'Chrome', 'Safari',
        'Firefox', 'Chrome', 'Safari'
    ]
})

counts = df['Browser'].value_counts()

df['Browser_count'] = df['Browser'].map(counts)

df
````

# 16. Target Encoding

Target encoding replaces each category with a statistic calculated from the target variable, commonly the mean target value for that category.

Example:

| City | Purchased |
|---|---:|
| Lahore | 1 |
| Lahore | 1 |
| Lahore | 0 |
| Karachi | 0 |
| Karachi | 0 |
| Islamabad | 1 |

Mean purchase rate:

```text
Lahore     → (1 + 1 + 0) / 3 = 0.67
Karachi    → (0 + 0) / 2 = 0.00
Islamabad  → 1 / 1 = 1.00
```

The category is therefore represented using information about the target.

````python
df = pd.DataFrame({
    'City': [
        'Lahore', 'Lahore', 'Lahore',
        'Karachi', 'Karachi', 'Islamabad'
    ],
    'Purchased': [1, 1, 0, 0, 0, 1]
})

target_mean = df.groupby('City')['Purchased'].mean()

df['City_target_encoded'] = df['City'].map(target_mean)

df
````

# 17. Data Leakage in Target Encoding

Target encoding has an important risk: **data leakage**.

If you calculate target statistics using the entire dataset before splitting into training and test data, information from the test set can influence the features.

This is especially dangerous when the category has few observations.

Correct high-level workflow:

```text
Raw data
   ↓
Train/Test split
   ↓
Learn encoding from training data
   ↓
Transform training data
   ↓
Transform test data using training mapping
```

Never allow test-target information to influence training preprocessing.

# 18. Multi-Hot Encoding

Sometimes one row can contain multiple categories.

Example:

| User | Skills |
|---|---|
| Ali | Python,SQL |
| Sara | Python,Java |
| John | Java,React |

A user can have multiple skills, so each row can contain multiple `1` values.

Result:

```text
User   Python  SQL  Java  React
Ali       1      1     0      0
Sara      1      0     1      0
John      0      0     1      1
```

This is commonly called multi-hot encoding.

````python
df = pd.DataFrame({
    'User': ['Ali', 'Sara', 'John'],
    'Skills': [
        'Python,SQL',
        'Python,Java',
        'Java,React'
    ]
})

skills_encoded = df['Skills'].str.get_dummies(sep=',')

skills_encoded
````

<a id="sec-24"></a>
# 24. `fit()` vs `transform()`

This distinction is critical in machine learning.

### `fit()`

The encoder learns information from data.

For one-hot encoding, it learns the available categories.

### `transform()`

The encoder applies what it already learned to new data.

Correct workflow:

```python
encoder.fit(X_train)
X_train_encoded = encoder.transform(X_train)
X_test_encoded = encoder.transform(X_test)
```

Shortcut for training data:

```python
X_train_encoded = encoder.fit_transform(X_train)
```

Do NOT fit a separate encoder on the test data.

````python
train = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

test = pd.DataFrame({
    'City': ['Lahore', 'Multan']
})

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

X_train = encoder.fit_transform(train[['City']])
X_test = encoder.transform(test[['City']])

print('Train:')
print(X_train)

print('\nTest:')
print(X_test)
````

# 25. Encoding With `ColumnTransformer`

Real datasets contain both numerical and categorical columns.

Example:

```text
Age       → numerical
Salary    → numerical
Gender    → categorical
City      → categorical
```

We can apply different preprocessing to different columns using `ColumnTransformer`.

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

X = pd.DataFrame({
    'Age': [25, 30, 28, 35],
    'Salary': [50000, 70000, 60000, 90000],
    'Gender': ['Male', 'Female', 'Male', 'Female'],
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

numeric_features = ['Age', 'Salary']
categorical_features = ['Gender', 'City']

preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numeric_features
        ),
        (
            'cat',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        )
    ]
)

X_transformed = preprocessor.fit_transform(X)

print(X_transformed)
````

```text
Yes → 1

```

Use for two-category variables when a binary representation makes sense.

## Ordinal

Use commonly for nominal categories with manageable cardinality.

Represent categories using their occurrence frequency.

Powerful but requires careful leakage prevention.

Useful techniques to investigate when cardinality becomes very large.

## Solution 16 — Leakage detection

1. **No, it is not safe.**
2. The encoding for each category was computed using the targets of the *test* rows too. The training features
   now contain information about the test answers, so the test score is optimistic — the model has "seen" part of the test.
   The effect is largest for rare categories (a category with 1 row gets exactly its own target!).
3. **Split first.** Compute the category means on the training set only, map them onto both train and test,
   fill unseen categories with the global training mean, and ideally use out-of-fold (cross-validated) means for the
   training rows themselves (this is what `sklearn.preprocessing.TargetEncoder` does).

## Solution 18 — Group rare categories into `Other`

````python
cities = pd.DataFrame({'City': ['Lahore'] * 4 + ['Karachi'] * 3 + ['Islamabad'] * 3 +
                                 ['Multan', 'Quetta', 'Peshawar', 'Multan']})
counts = cities['City'].value_counts()
rare = counts[counts < 3].index
cities['City_grouped'] = cities['City'].where(~cities['City'].isin(rare), 'Other')
print(counts, '\n')
print(cities['City_grouped'].value_counts())
````

## Solution 19 — Unknown categories

The encoder only knows the categories it saw during `fit` (Lahore, Karachi, Islamabad). When `Multan` appears at
prediction time there is no column for it. With the default `handle_unknown='error'`, `transform` raises an error and the
production system crashes.

`handle_unknown='ignore'` makes the encoder output **all zeros** for that row's City columns, so the row is still
processed — the model simply gets "none of the known cities".

## Solution 20 — Binary encoding arithmetic

1. One-hot: **16 columns** (15 with `drop_first`).
2. Binary: 2⁴ = 16, so **4 bits** (in practice libraries often need 5 because code 0 is reserved — still tiny).
3. Binary encoding needs about log₂(k) columns instead of k, so dimensionality grows very slowly as categories increase.

## Solution 21 — Hash encoding

1. A hash function turns each category string into a number, and that number (modulo *n*) chooses one of *n* fixed output columns.
2. The number of columns is fixed in advance, no matter how many categories exist (even ones never seen before), and no
   category dictionary has to be stored.
3. A **collision** is when two different categories land in the same column, so the model cannot tell them apart.
   More columns → fewer collisions.

## Solution 22 — `OneHotEncoder(handle_unknown="ignore")`

## Solution 25 — Numbers that are really categories

| Column | Type | Why |
|---|---|---|
| `Age` | numeric measurement | differences are meaningful (30 is 10 more than 20) |
| `Salary` | numeric measurement | amounts can be compared and averaged |
| `NumberOfChildren` | numeric (count) | 2 children really is more than 1 |
| `PostalCode` | **category** | 75000 is not "bigger" than 54000; it's a location label |
| `ProductID` | **identifier** | just a name for a product; usually dropped, or frequency/target encoded |

Rule: *if averaging the column makes no sense, it is not a quantity.*

## Solution 26 — Choose the encoding

| Feature | Encoding | Why |
|---|---|---|
| Gender (2 values) | binary 0/1 | only two categories |
| City (3 values) | one-hot | nominal, few categories |
| Education | ordinal | clear order School < Bachelor < Master < PhD |
| Satisfaction | ordinal | clear order Bad < Okay < Good < Excellent |
| Browser | one-hot | nominal, few categories |
| Skills (many per row) | multi-hot (`str.get_dummies`) | a row can have several skills |
| Country (150+) | group rare → Other + one-hot, or frequency / target encoding | too many columns for plain one-hot |
| ProductID (100k) | frequency, target (leak-safe), or hash encoding — or drop | one-hot would create 100k columns |

## Solution 30 — Train/test encoding with a full pipeline

````python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

rng = np.random.default_rng(0)
n = 60
d30 = pd.DataFrame({
    'Age': rng.integers(20, 60, n),
    'Income': rng.normal(60, 20, n).round(1),
    'Gender': rng.choice(['Male', 'Female'], n),
    'City': rng.choice(['Lahore', 'Karachi', 'Islamabad', 'Multan'], n),
})
# target depends on Income and City so the model has something to learn
score = 0.05 * (d30['Income'] - 60) + (d30['City'] == 'Lahore') * 1.0
d30['Purchased'] = (score + rng.normal(0, 0.5, n) > 0).astype(int)

X = d30.drop(columns='Purchased')
y = d30['Purchased']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

num_cols = X.select_dtypes(include='number').columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'string', 'category']).columns.tolist()
print('numeric:', num_cols, '| categorical:', cat_cols)

pipe30 = Pipeline([
    ('pre', ColumnTransformer([
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ])),
    ('model', LogisticRegression())
])
pipe30.fit(X_train, y_train)                 # everything is learned from TRAIN only
pred = pipe30.predict(X_test)
print('Test accuracy:', round(accuracy_score(y_test, pred), 3))
print(classification_report(y_test, pred, zero_division=0))
````

## Solution 31 — Leak-free target encoding

````python
rng = np.random.default_rng(1)
d31 = pd.DataFrame({
    'City': rng.choice(['Lahore', 'Karachi', 'Islamabad'], 40),
    'Age': rng.integers(20, 60, 40),
    'Income': rng.normal(60, 15, 40).round(1),
    'Purchased': rng.integers(0, 2, 40)
})
train31, test31 = train_test_split(d31, test_size=0.25, random_state=42)
test31 = test31.copy()
test31.iloc[0, test31.columns.get_loc('City')] = 'Quetta'      # a city never seen in training

city_rate = train31.groupby('City')['Purchased'].mean()        # 2. training data only
global_rate = train31['Purchased'].mean()

train31 = train31.assign(City_te=train31['City'].map(city_rate))                    # 3
test31['City_te'] = test31['City'].map(city_rate).fillna(global_rate)               # 4 + 5
print(city_rate, '\nglobal:', round(global_rate, 3))
test31.head()
````

**6. What goes wrong if you compute the mean before splitting?** The test rows' targets leak into the feature.
The model looks better on the test set than it will be on truly new data. Unseen categories like `Quetta` are
filled with the global training rate instead of crashing or getting `NaN`.

## Solution 33 — Final Project: Customer Purchase Prediction

````python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.metrics import confusion_matrix

rng = np.random.default_rng(7)
n = 200
proj = pd.DataFrame({
    'Age': rng.integers(18, 65, n),
    'Income': rng.normal(70, 25, n).clip(15).round(1),
    'Gender': rng.choice(['Male', 'Female'], n),
    'Education': rng.choice(['School', 'Bachelor', 'Master', 'PhD'], n, p=[0.2, 0.4, 0.3, 0.1]),
    'City': rng.choice(['Lahore', 'Karachi', 'Islamabad', 'Multan'], n),
    'Browser': rng.choice(['Chrome', 'Safari', 'Firefox'], n, p=[0.6, 0.25, 0.15]),
    'Device': rng.choice(['Mobile', 'Desktop', 'Tablet'], n, p=[0.55, 0.35, 0.10]),
})
edu_rank = proj['Education'].map({'School': 0, 'Bachelor': 1, 'Master': 2, 'PhD': 3})
logit = 0.04 * (proj['Income'] - 70) + 0.5 * edu_rank + 0.8 * (proj['Device'] == 'Desktop') - 0.8
proj['Purchased'] = (rng.random(n) < 1 / (1 + np.exp(-logit))).astype(int)

# Part 1 — inspection
print(proj.head(), '\n')
print('shape:', proj.shape)
print(proj.dtypes, '\n')
X = proj.drop(columns='Purchased')
y = proj['Purchased']
cat_cols = X.select_dtypes(include=['object', 'string', 'category']).columns.tolist()
num_cols = X.select_dtypes(include='number').columns.tolist()
print('categorical:', cat_cols)
print('numerical  :', num_cols)
````

````python
# Part 5 — split BEFORE fitting anything
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# Parts 2-4 + 6 — one ColumnTransformer does binary, ordinal and one-hot encoding
pre33 = ColumnTransformer([
    ('num', StandardScaler(), ['Age', 'Income']),
    ('gender', OneHotEncoder(drop='if_binary'), ['Gender']),                                       # binary 0/1
    ('edu', OrdinalEncoder(categories=[['School', 'Bachelor', 'Master', 'PhD']]), ['Education']),  # ordered
    ('nominal', OneHotEncoder(handle_unknown='ignore'), ['City', 'Browser', 'Device'])              # one-hot
])

# Parts 7-8 — pipeline + model
model33 = Pipeline([('pre', pre33), ('model', LogisticRegression(max_iter=1000))])
model33.fit(X_train, y_train)

# Part 9 — evaluation
pred = model33.predict(X_test)
print('Accuracy:', round(accuracy_score(y_test, pred), 3))
print('Confusion matrix:\n', confusion_matrix(y_test, pred))
print(classification_report(y_test, pred, zero_division=0))
print('Features after encoding:', len(model33.named_steps['pre'].get_feature_names_out()))
````

**Part 10 — explanation**

- **Gender** has two values, so one 0/1 column is enough (`drop='if_binary'`).
- **Education** has a real order, so `OrdinalEncoder` with an explicit category order keeps School < Bachelor < Master < PhD.
- **City** has no order — label encoding (0,1,2,3) would invent one, so it is one-hot encoded.
- **Leakage:** the scaler and encoders are fitted inside the pipeline on the training split only; the test set is only transformed.
- **Unknown categories:** `handle_unknown='ignore'` means a new city/browser/device in production produces zeros instead of an error.
