Find P(app **or** website), P(neither), and P(app **only**). Draw the situation as a small table.

````python
# ✍️ Problem 3 — your attempt
````

### Solution 3

(a) P(3 coin flips are all heads). (b) Draw 2 cards: P(both red) **without** replacement vs **with** replacement. (c) In the data below, check whether `Newsletter` and `Purchased` are independent:
```python
rng4 = np.random.default_rng(4)
d4 = pd.DataFrame({'Newsletter': rng4.random(5000) < 0.4})
d4['Purchased'] = np.where(d4['Newsletter'], rng4.random(5000) < 0.3, rng4.random(5000) < 0.1)
```

````python
# ✍️ Problem 4 — your attempt
````

### Solution 4
(a) (1/2)³ = **1/8 = 0.125** (b) without replacement 26/52 × 25/51 ≈ **0.245**; with replacement (1/2)² = **0.25**

````python
print('All heads:', Fraction(1, 2) ** 3)
print('Both red, no replacement  :', Fraction(26, 52) * Fraction(25, 51), '≈', round(26 / 52 * 25 / 51, 4))
print('Both red, with replacement:', Fraction(1, 4))

rng4 = np.random.default_rng(4)
d4 = pd.DataFrame({'Newsletter': rng4.random(5000) < 0.4})
d4['Purchased'] = np.where(d4['Newsletter'], rng4.random(5000) < 0.3, rng4.random(5000) < 0.1)
pa, pb, pab = d4['Newsletter'].mean(), d4['Purchased'].mean(), (d4['Newsletter'] & d4['Purchased']).mean()
print(f'\nP(N ∩ P) = {pab:.4f} vs P(N)·P(P) = {pa * pb:.4f} → dependent')
print(d4.groupby('Newsletter')['Purchased'].mean().rename('P(Purchased | Newsletter)'))
````

**Explanation:** the joint probability is clearly larger than the product, and P(purchase | newsletter) ≈ 0.30 vs ≈ 0.10 without → **dependent**.

### Problem 5 — Conditional probability from a table

500 users: 300 on mobile (45 converted), 200 on desktop (50 converted).
Find P(convert), P(convert | mobile), P(convert | desktop), and P(desktop | convert). Build the table with Pandas and explain why the last two answers differ.

````python
# ✍️ Problem 5 — your attempt
````

### Solution 5
P(convert) = 95/500 = **0.19**; P(convert | mobile) = 45/300 = **0.15**; P(convert | desktop) = 50/200 = **0.25**; P(desktop | convert) = 50/95 ≈ **0.526**

````python
users = pd.DataFrame({
    'Device': ['Mobile'] * 300 + ['Desktop'] * 200,
    'Converted': [1] * 45 + [0] * 255 + [1] * 50 + [0] * 150,
})
print(pd.crosstab(users['Device'], users['Converted'], margins=True), '\n')
print('P(convert)           =', users['Converted'].mean())
print(users.groupby('Device')['Converted'].mean().rename('P(convert | device)'))
print('P(desktop | convert) =', round((users.loc[users['Converted'] == 1, 'Device'] == 'Desktop').mean(), 3))
````

**Explanation:** P(convert | desktop) = 0.25 asks *"of desktop users, how many convert?"*. P(desktop | convert) ≈ 0.53 asks *"of the converters, how many used desktop?"* — a different group in the denominator, so a different answer.

### Problem 6 — Bayes' theorem (fraud)

0.5% of transactions are fraudulent. The detector flags 95% of fraud and wrongly flags 2% of legitimate transactions.
(a) P(flagged) (b) P(fraud | flagged) (c) What if the false-alarm rate drops to 0.5%? (d) Verify (b) with a simulation of 1,000,000 transactions and explain the result using counts per 100,000 transactions.

````python
# ✍️ Problem 6 — your attempt
````

### Solution 6
(a) P(flag) = 0.95 × 0.005 + 0.02 × 0.995 = 0.00475 + 0.0199 = **0.02465**
(b) P(fraud | flag) = 0.00475 / 0.02465 ≈ **0.193** (c) with 0.5% false alarms: 0.00475 / (0.00475 + 0.004975) ≈ **0.488**

````python
post_b, p_flag = bayes(prior=0.005, likelihood=0.95, false_positive_rate=0.02)
post_c, _ = bayes(prior=0.005, likelihood=0.95, false_positive_rate=0.005)
print(f'(a) P(flag) = {p_flag:.5f}\n(b) P(fraud | flag) = {post_b:.4f}\n(c) with 0.5% false alarms = {post_c:.4f}')

n6 = 1_000_000
fraud = rng.random(n6) < 0.005
flag = np.where(fraud, rng.random(n6) < 0.95, rng.random(n6) < 0.02)
print(f'(d) simulated P(fraud | flag) = {fraud[flag].mean():.4f}')
````

**Explanation (per 100,000 transactions):** 500 are fraud → 475 flagged. 99,500 are legitimate → 1,990 falsely flagged. Of 2,465 flags only 475 (≈ 19%) are fraud. The legitimate group is so large that even a 2% false-alarm rate swamps the true cases — the **base-rate effect**. Reducing false alarms matters more than improving detection here.

### Problem 7 — Naive Bayes by hand

Training reviews: 5 positive ("great" appears in 4, "slow" in 1), 5 negative ("great" in 1, "slow" in 4).
Using Laplace smoothing (+1 / +2), compute P(positive | review contains "great" and does NOT contain "slow").

````python
# ✍️ Problem 7 — your attempt
````

### Solution 7
- P(great | pos) = (4+1)/(5+2) = 5/7; P(no slow | pos) = 1 − (1+1)/(5+2) = 5/7 → score = 0.5 × 5/7 × 5/7 = 0.5 × 25/49
- P(great | neg) = (1+1)/7 = 2/7; P(no slow | neg) = 1 − (4+1)/7 = 2/7 → score = 0.5 × 4/49
- P(pos | review) = 25 / (25 + 4) = **25/29 ≈ 0.862**

````python
pos = 0.5 * Fraction(5, 7) * Fraction(5, 7)
neg = 0.5 * Fraction(2, 7) * Fraction(2, 7)
print('P(positive | "great", no "slow") =', Fraction(pos / (pos + neg)).limit_denominator(), '≈', round(float(pos / (pos + neg)), 3))
````

### Problem 9 — Binomial distribution

20 visitors arrive; each converts independently with p = 0.1.
Find P(exactly 2 conversions), P(at least 1), the expected number and the standard deviation. Plot the PMF.

````python
# ✍️ Problem 9 — your attempt
````

### Solution 9
P(X = 2) = C(20,2) × 0.1² × 0.9¹⁸ ≈ **0.285**; P(X ≥ 1) = 1 − 0.9²⁰ ≈ **0.878**; mean = np = **2**; sd = √(np(1−p)) = √1.8 ≈ **1.342**

### Problem 10 — Poisson distribution

A website crashes on average 3 times per week.
Find P(no crash in a week), P(more than 5 crashes in a week), P(no crash in 2 weeks). Plot the PMF for one week.

````python
# ✍️ Problem 10 — your attempt
````

### Solution 10
P(0) = e⁻³ ≈ **0.050**; P(X > 5) ≈ **0.084**; two weeks → λ = 6 → P(0) = e⁻⁶ ≈ **0.0025**

````python
p10 = stats.poisson(3)
print(f'P(0 in a week) = {p10.pmf(0):.4f} | P(>5 in a week) = {p10.sf(5):.4f} | P(0 in 2 weeks) = {stats.poisson(6).pmf(0):.4f}')
k = np.arange(0, 12)
plt.figure(figsize=(7, 3.2)); plt.bar(k, p10.pmf(k), color='tab:green'); plt.title('Poisson(3): crashes per week'); plt.show()
````

**Explanation:** over 2 weeks the average count doubles (λ = 6); Poisson rates scale with the length of the interval.

### Problem 12 — Exponential distribution

Calls arrive at a call centre with an average of 4 minutes between calls.
Find P(next call within 2 minutes), P(waiting more than 10 minutes), and show the memoryless property numerically: P(wait > 10 | already waited 6) vs P(wait > 4).

````python
# ✍️ Problem 12 — your attempt
````

### Solution 12
Mean 4 min → P(wait ≤ 2) = 1 − e^(−2/4) ≈ **0.393**; P(wait > 10) = e^(−2.5) ≈ **0.082**; memoryless: P(> 10 | > 6) = e^(−10/4) / e^(−6/4) = e^(−1) = P(> 4) ≈ **0.368**

````python
e12 = stats.expon(scale=4)
print(f'P(≤2) = {e12.cdf(2):.4f} | P(>10) = {e12.sf(10):.4f}')
print(f'P(>10 | >6) = {e12.sf(10) / e12.sf(6):.4f} | P(>4) = {e12.sf(4):.4f}  → identical (memoryless)')
````

### Problem 13 — Monte Carlo simulation (Chevalier de Méré's problem)

Which is more likely: (A) at least one 6 in 4 rolls of a die, or (B) at least one double-six in 24 rolls of two dice?
Compute both exactly using complements, then estimate both with 200,000 simulations each.

````python
# ✍️ Problem 13 — your attempt
````

### Solution 13
(A) 1 − (5/6)⁴ ≈ **0.518** (B) 1 − (35/36)²⁴ ≈ **0.491** → (A) is slightly more likely. (This puzzle, raised by a French gambler in 1654, helped start probability theory!)

````python
print('Exact A =', round(1 - (5 / 6) ** 4, 4), '| Exact B =', round(1 - (35 / 36) ** 24, 4))
n13 = 200_000
sim_A = (rng.integers(1, 7, (n13, 4)) == 6).any(axis=1).mean()
double_six = (rng.integers(1, 7, (n13, 24)) == 6) & (rng.integers(1, 7, (n13, 24)) == 6)
sim_B = double_six.any(axis=1).mean()
print('Simulated A =', round(sim_A, 4), '| Simulated B =', round(sim_B, 4))
````

### Problem 14 — Sigmoid, softmax and log loss

(a) Convert logits `[-2, 0, 1.5]` to probabilities with the sigmoid. (b) Apply softmax to `[3, 1, 0]` and verify it sums to 1. (c) Three examples all have true label 1 with predicted probabilities `[0.9, 0.6, 0.2]` — compute the log loss by hand and with `sklearn.metrics.log_loss`.

````python
# ✍️ Problem 14 — your attempt
````

### Solution 14
(a) σ(−2) ≈ 0.119, σ(0) = 0.5, σ(1.5) ≈ 0.818 (b) softmax([3, 1, 0]) ≈ [0.844, 0.114, 0.042] (c) log loss = (−ln 0.9 − ln 0.6 − ln 0.2) / 3 = (0.105 + 0.511 + 1.609) / 3 ≈ **0.742**

````python
print('(a) sigmoid:', sigmoid(np.array([-2, 0, 1.5])).round(4))
sm = softmax([3, 1, 0])
print('(b) softmax:', sm.round(4), '| sum =', sm.sum())
p14 = np.array([0.9, 0.6, 0.2])
print('(c) by hand :', round(np.mean(-np.log(p14)), 4))
print('    sklearn :', round(log_loss([1, 1, 1], p14, labels=[0, 1]), 4))
````

**Explanation:** the third example (p = 0.2 for a true positive) contributes most of the loss — confident mistakes dominate log loss.

# 23. Final Project — Customer Churn Probability Analysis

**Business goal:** a subscription company wants to (1) understand **churn probabilities** for different customers, (2) estimate **revenue at risk**, and (3) decide **which customers should receive a retention offer**.

We'll use almost every idea from this notebook: empirical, conditional and joint probability, independence, Bayes' theorem, Poisson distribution, model probabilities, calibration, log loss, expected value and Monte Carlo simulation.

## Step 0 — Create the dataset (2,000 customers)

````python
rp = np.random.default_rng(2026)
n = 2000

plan = rp.choice(['Basic', 'Standard', 'Premium'], n, p=[0.5, 0.3, 0.2])
contract = rp.choice(['Monthly', 'Yearly'], n, p=[0.65, 0.35])
tenure = rp.exponential(scale=18, size=n).clip(1, 72).round()                      # months with the company
ticket_rate = pd.Series(plan).map({'Basic': 1.2, 'Standard': 0.9, 'Premium': 0.6}).to_numpy()
tickets = rp.poisson(ticket_rate * 3)                                              # support tickets last 3 months
monthly_charge = pd.Series(plan).map({'Basic': 1500, 'Standard': 2500, 'Premium': 4000}).to_numpy() \
                 * rp.uniform(0.9, 1.1, n)

# True (hidden) churn mechanism on the log-odds scale
logit = (-1.6
         + np.where(contract == 'Monthly', 1.1, -0.6)
         + 0.35 * tickets
         - 0.035 * tenure
         + pd.Series(plan).map({'Basic': 0.3, 'Standard': 0.0, 'Premium': -0.4}).to_numpy())
churn = (rp.random(n) < 1 / (1 + np.exp(-logit))).astype(int)

cust = pd.DataFrame({
    'CustomerID': [f'C{10000 + i}' for i in range(n)],
    'Plan': plan, 'Contract': contract, 'Tenure_Months': tenure,
    'Support_Tickets': tickets, 'Monthly_Charge': monthly_charge.round(0), 'Churned': churn,
})
cust.head()
````

## Step 1 — Empirical (marginal) probabilities

````python
p_churn = cust['Churned'].mean()
print(f'P(churn)     = {p_churn:.3f}')
print(f'P(no churn)  = {1 - p_churn:.3f}   (complement)')
print('\nP(Plan):');     print(cust['Plan'].value_counts(normalize=True).round(3))
print('\nP(Contract):'); print(cust['Contract'].value_counts(normalize=True).round(3))
````

## Step 2 — Conditional probabilities: who churns?

````python
cust['Ticket_Group'] = pd.cut(cust['Support_Tickets'], bins=[-1, 0, 2, 4, 100], labels=['0', '1-2', '3-4', '5+'])
cust['Tenure_Group'] = pd.cut(cust['Tenure_Months'], bins=[0, 6, 12, 24, 72], labels=['≤6', '7-12', '13-24', '25+'])

fig, axes = plt.subplots(1, 4, figsize=(20, 3.8), sharey=True)
for ax, col in zip(axes, ['Contract', 'Plan', 'Ticket_Group', 'Tenure_Group']):
    cond = cust.groupby(col, observed=True)['Churned'].mean()
    ax.bar(cond.index.astype(str), cond.values, color='tab:red')
    ax.axhline(p_churn, color='black', ls='--', label='overall P(churn)')
    ax.set_title(f'P(churn | {col})')
    for i, v in enumerate(cond.values):
        ax.text(i, v + 0.01, f'{v:.2f}', ha='center')
axes[0].legend(); plt.tight_layout(); plt.show()

print(pd.crosstab([cust['Contract'], cust['Plan']], cust['Churned'], normalize='index')[1]
      .rename('P(churn | contract, plan)').round(3))
````

## Step 3 — Joint probability and an independence check

````python
monthly = cust['Contract'] == 'Monthly'
churned = cust['Churned'] == 1
p_m, p_c, p_mc = monthly.mean(), churned.mean(), (monthly & churned).mean()
print(f'P(Monthly ∩ Churn) = {p_mc:.3f} vs P(Monthly)·P(Churn) = {p_m * p_c:.3f} → '
      f'{"DEPENDENT" if abs(p_mc - p_m * p_c) > 0.01 else "≈ independent"}')

print('\nJoint distribution P(Contract, Churned):')
print(pd.crosstab(cust['Contract'], cust['Churned'], normalize='all', margins=True).round(3))
````

## Step 4 — Bayes' theorem: from "churners have many tickets" to "many tickets → churn?"

The support team knows **P(≥ 3 tickets | churn)** from churned customers' histories. Management wants **P(churn | ≥ 3 tickets)**. Bayes flips it.

````python
many = cust['Support_Tickets'] >= 3
p_many_given_churn = many[churned].mean()           # likelihood
p_many_given_stay = many[~churned].mean()           # "false positive" rate
posterior, p_many = bayes(prior=p_churn, likelihood=p_many_given_churn, false_positive_rate=p_many_given_stay)

print(f'Prior P(churn)                = {p_churn:.3f}')
print(f'P(≥3 tickets | churn)         = {p_many_given_churn:.3f}')
print(f'P(≥3 tickets | no churn)      = {p_many_given_stay:.3f}')
print(f'Bayes → P(churn | ≥3 tickets) = {posterior:.3f}')
print(f'Direct from data              = {churned[many].mean():.3f}  ✔ same answer')
````

## Step 5 — Distributions: are support tickets Poisson?

````python
fig, axes = plt.subplots(1, 3, figsize=(17, 3.8))
for ax, p_name in zip(axes, ['Basic', 'Standard', 'Premium']):
    t = cust.loc[cust['Plan'] == p_name, 'Support_Tickets']
    k = np.arange(0, t.max() + 1)
    ax.bar(k, [np.mean(t == i) for i in k], alpha=0.6, label='observed')
    ax.plot(k, stats.poisson(t.mean()).pmf(k), 'ro-', label=f'Poisson(λ = {t.mean():.2f})')
    ax.set_title(f'{p_name}: mean {t.mean():.2f}, variance {t.var():.2f}'); ax.legend()
plt.tight_layout(); plt.show()

lam_basic = cust.loc[cust['Plan'] == 'Basic', 'Support_Tickets'].mean()
print(f'Basic plan: P(a customer files 8+ tickets in 3 months) = {stats.poisson(lam_basic).sf(7):.4f} → rare → worth flagging')
````

For each plan the mean ≈ variance and the Poisson PMF matches the bars → a Poisson model is reasonable. It lets us flag **unusually high** ticket counts as early churn warnings.

## Step 6 — Predict churn probabilities with a model

````python
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import roc_auc_score, brier_score_loss

features = ['Plan', 'Contract', 'Tenure_Months', 'Support_Tickets', 'Monthly_Charge']
X = cust[features]; y = cust['Churned']
X_train, X_test, y_train, y_test, idx_train, idx_test = train_test_split(
    X, y, cust.index, test_size=0.3, random_state=0, stratify=y)

model = make_pipeline(
    make_column_transformer((OneHotEncoder(drop='first'), ['Plan', 'Contract']),
                            (StandardScaler(), ['Tenure_Months', 'Support_Tickets', 'Monthly_Charge'])),
    LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
p_test = model.predict_proba(X_test)[:, 1]

baseline = np.full_like(p_test, y_train.mean())          # always predict the training churn rate
print(f'Log loss — model    : {log_loss(y_test, p_test):.4f}')
print(f'Log loss — baseline : {log_loss(y_test, baseline):.4f}  (always predict P(churn) = {y_train.mean():.3f})')
print(f'ROC AUC  — model    : {roc_auc_score(y_test, p_test):.3f}  (ranking quality: 0.5 = random, 1 = perfect)')
print(f'Brier score         : {brier_score_loss(y_test, p_test):.4f}  (mean squared error of probabilities)')
````

## Step 7 — Expected value: who should get a retention offer?

Assumptions:
- Customer value if retained = 12 months × monthly charge.
- The offer costs 2,000 rupees and **reduces churn probability by 40%** (relative), i.e. new P = 0.6 × P.

Expected gain of making the offer = (P − 0.6·P) × value − cost = 0.4 · P · value − cost. Give the offer only when this is **positive**.

````python
test = cust.loc[idx_test].copy()
test['P_churn'] = p_test
test['Value'] = 12 * test['Monthly_Charge']
offer_cost, reduction = 2000, 0.40
test['Expected_Gain'] = reduction * test['P_churn'] * test['Value'] - offer_cost
test['Give_Offer'] = test['Expected_Gain'] > 0

strategies = {
    'Nobody': np.zeros(len(test), bool),
    'Everybody': np.ones(len(test), bool),
    'Top 20% risk': test['P_churn'] >= test['P_churn'].quantile(0.8),
    'Expected-value rule': test['Give_Offer'].to_numpy(),
}
rows = []
for name, target in strategies.items():
    gain = test.loc[target, 'Expected_Gain'].sum() if target.any() else 0.0
    rows.append({'Strategy': name, 'Customers targeted': int(target.sum()), 'Expected net gain': gain})
pd.DataFrame(rows).set_index('Strategy').round(0)
````

## Step 8 — Monte Carlo: how much revenue is at risk next year?

Each customer churns (Bernoulli) with their predicted probability. Simulate the whole test portfolio many times to get the **distribution** of lost revenue, not just a single average.

````python
n_sim = 10_000
churn_draws = rp.random((n_sim, len(test))) < test['P_churn'].to_numpy()        # each row = one possible future
lost_revenue = churn_draws @ test['Value'].to_numpy()                            # revenue lost in each simulated future

expected_loss = (test['P_churn'] * test['Value']).sum()                          # E[sum] = sum of E (Section 14)
lo, hi = np.percentile(lost_revenue, [5, 95])
print(f'Expected revenue lost (formula)   = {expected_loss:,.0f}')
print(f'Expected revenue lost (simulated) = {lost_revenue.mean():,.0f}')
print(f'90% of simulated futures lose between {lo:,.0f} and {hi:,.0f}')
print(f'P(loss > {expected_loss * 1.1:,.0f}, i.e. 10% worse than expected) = {np.mean(lost_revenue > expected_loss * 1.1):.3f}')

plt.figure(figsize=(8, 3.8))
sns.histplot(lost_revenue, bins=60, stat='density')
plt.axvline(expected_loss, color='red', label='expected'); plt.axvline(lo, color='gray', ls='--'); plt.axvline(hi, color='gray', ls='--', label='5th–95th pct')
plt.xlabel('Revenue lost over 12 months (test customers)'); plt.title('Monte Carlo: revenue at risk'); plt.legend(); plt.show()
````

## Step 9 — Conclusions (read alongside the outputs above)

1. **Baseline risk.** The overall churn probability (Step 1) is the **prior**. Every segment is compared with it.
2. **Conditional probabilities** (Step 2) show the strongest signals. **Monthly** contracts churn far more than yearly ones. Churn rises steadily with the number of **support tickets** and is highest for **new** customers (short tenure). Premium customers churn least.
3. **Dependence** (Step 3): contract type and churn are clearly **not independent**: P(Monthly ∩ Churn) is well above P(Monthly)·P(Churn).
4. **Bayes** (Step 4) turned P(many tickets | churn), which the support team can measure, into P(churn | many tickets), which management needs. It gives exactly the same answer as counting directly.
5. **Poisson** (Step 5) describes ticket counts per plan well (mean ≈ variance). That gives a principled way to flag customers with unusually many tickets.
6. **Model probabilities** (Step 6): logistic regression beats the "always predict the base rate" baseline on log loss, ranks customers well (AUC), and its calibration curve lies close to the diagonal. Its probabilities can therefore be used as real probabilities.
7. **Decisions** (Step 7): multiplying probabilities by money at stake turns predictions into actions. The **expected-value rule** gives the highest expected net gain. Offering everyone wastes money on customers who would stay anyway; offering nobody loses preventable churn.
8. **Risk** (Step 8): Monte Carlo gives a **range** of revenue loss, not just one number, so finance can plan for a bad (95th-percentile) year.

**Main lesson:** probability connects data → model → decision. Conditional probabilities find the signals, calibrated model probabilities quantify risk per customer, and expected value plus simulation turn risk into money and actions.
