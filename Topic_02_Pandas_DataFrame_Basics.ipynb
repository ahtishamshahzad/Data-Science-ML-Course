# 3. Creating DataFrames in Different Ways

Common approaches:

- Dictionary of lists
- List of lists
- List of dictionaries
- Empty DataFrame

````python
import pandas as pd

# ============================================
# 1. Dictionary of Lists
# ============================================

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Score": [85, 92, 78]
}

df1 = pd.DataFrame(data)

print("1. Dictionary of Lists:")
print(df1)


# ============================================
# 2. List of Lists
# ============================================

data = [
    ["Ali", 20, 85],
    ["Ahmed", 21, 92],
    ["Sara", 19, 78]
]

df2 = pd.DataFrame(
    data,
    columns=["Name", "Age", "Score"]
)

print("\n2. List of Lists:")
print(df2)


# ============================================
# 3. List of Dictionaries
# ============================================

data = [
    {"Name": "Ali", "Age": 20, "Score": 85},
    {"Name": "Ahmed", "Age": 21, "Score": 92},
    {"Name": "Sara", "Age": 19, "Score": 78}
]

df3 = pd.DataFrame(data)

print("\n3. List of Dictionaries:")
print(df3)


# ============================================
# 4. Empty DataFrame
# ============================================

df4 = pd.DataFrame()

print("\n4. Empty DataFrame:")
print(df4)


# ============================================
# Empty DataFrame with Columns
# ============================================

df5 = pd.DataFrame(columns=["Name", "Age", "Score"])

print("\n5. Empty DataFrame with Columns:")
print(df5)
````

# 9. Filtering

Filtering creates a Boolean condition and uses it to select rows.

```python
df[df["Age"] > 25]
```

For multiple conditions:

- `&` = AND
- `|` = OR
- `~` = NOT

Always put individual conditions in parentheses.

````python
print(df[df["Age"] > 25])
print()
print(df[(df["Age"] > 25) & (df["Salary"] > 80000)])
print()
print(df[(df["City"] == "Lahore") | (df["City"] == "Karachi")])
````

# 12. Conditional Columns with NumPy

`np.where()` is useful for two outcomes.

```python
np.where(condition, value_if_true, value_if_false)
```

`np.select()` is useful for multiple conditions.

````python
employees = df.copy()

work_df["SalaryLevel"] = np.where(
    work_df["Salary"] >= 100000,
    "High",
    "Normal"
)

print(np.where(
    work_df["Salary"] >= 100000,
    "High",
    "Normal"
))

conditions = [
    work_df["Age"] < 30,
    work_df["Age"].between(30, 39),
    work_df["Age"] >= 40
]

choices = ["Young", "Middle", "Senior"]

work_df["AgeGroup"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

work_df
````

# 17. Missing Values

Important tools:

- `isna()`

- `dropna()`
- `fillna()`
- `ffill()`
- `bfill()`

````python
missing_df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Hamza", "Zara"],
    "Age": [25, np.nan, 28, 35, np.nan],
    "Salary": [80000, 90000, np.nan, 120000, 95000],
    "City": ["Lahore", "Karachi", None, "Islamabad", "Lahore"]
})

print(missing_df)
print()
print(missing_df.isna())
print()
print(missing_df.isna().sum())
````

````python
filled = missing_df.copy()

filled["Age"] = filled["Age"].fillna(filled["Age"].median())
filled["Salary"] = filled["Salary"].fillna(filled["Salary"].mean())
filled["City"] = filled["City"].fillna("Unknown")

filled
````

````python
print("Drop rows:")
print(missing_df.dropna())

print()
print("Forward fill:")
print(missing_df.ffill())

print()
print("Backward fill:")
print(missing_df.bfill())
````

# 52. A Professional Data-Cleaning Workflow

A common real-world process is:

1. Load data
2. Inspect shape/types
3. Check missing values
4. Check duplicates

````python
import pandas as pd

# ==========================================
# 1. Load data
# ==========================================

data = pd.read_csv("data/customer_sales_dirty.csv")

print(data.head())


# ==========================================
# 2. Inspect shape and data types
# ==========================================

print("Shape:", data.shape)

print("\nData types:")
print(data.dtypes)

print("\nBasic information:")
print(data.info())


# ==========================================
# 3. Check missing values
# ==========================================

print("\nMissing values:")
print(data.isna().sum())


# ==========================================
# 4. Check duplicates
# ==========================================

print("\nNumber of duplicate rows:")
print(data.duplicated().sum())

print("\nDuplicate rows:")
print(data[data.duplicated()])


# Remove duplicates
data = data.drop_duplicates()


# ==========================================
# 5. Fix data types
# ==========================================

# Age contains "abc", so convert invalid values to NaN
data["Age"] = pd.to_numeric(data["Age"], errors="coerce")

# Convert numeric columns
data["Quantity"] = pd.to_numeric(data["Quantity"], errors="coerce")
data["Unit_Price"] = pd.to_numeric(data["Unit_Price"], errors="coerce")
data["Rating"] = pd.to_numeric(data["Rating"], errors="coerce")

# Convert date
data["Order_Date"] = pd.to_datetime(
    data["Order_Date"],
    errors="coerce"
)

print("\nData types after conversion:")
print(data.dtypes)


# ==========================================
# 6. Clean strings
# ==========================================

# Remove extra spaces and make names consistent
data["Name"] = data["Name"].str.strip()

data["City"] = data["City"].str.strip().str.title()

data["Product"] = data["Product"].str.strip().str.title()

data["Email"] = data["Email"].str.strip().str.lower()


# ==========================================
# 7. Handle missing values
# ==========================================

# Age → median
data["Age"] = data["Age"].fillna(data["Age"].median())


# Unit_Price → median of the SAME product
data["Unit_Price"] = (
    data.groupby("Product")["Unit_Price"]
    .transform(lambda x: x.fillna(x.median()))
)


# Rating → median
data["Rating"] = data["Rating"].fillna(data["Rating"].median())


# Email → unknown
data["Email"] = data["Email"].fillna("unknown")


# ==========================================
# 8. Validate ranges
# ==========================================

# Find invalid ages
invalid_age = data[
    (data["Age"] < 18) |
    (data["Age"] > 100)
]

print("\nInvalid ages:")
print(invalid_age)


# Find invalid quantities
invalid_quantity = data[
    data["Quantity"] < 0
]

print("\nInvalid quantities:")
print(invalid_quantity)


# Find invalid ratings
invalid_rating = data[
    (data["Rating"] < 1) |
    (data["Rating"] > 5)
]

print("\nInvalid ratings:")
print(invalid_rating)


# Find invalid prices
invalid_price = data[
    data["Unit_Price"] < 0
]

print("\nInvalid prices:")
print(invalid_price)


# Fix invalid values
data.loc[
    (data["Age"] < 18) | (data["Age"] > 100),
    "Age"
] = data["Age"].median()


data.loc[
    data["Quantity"] < 0,
    "Quantity"
] = data.loc[
    data["Quantity"] >= 0,
    "Quantity"
].median()


data.loc[
    (data["Rating"] < 1) | (data["Rating"] > 5),
    "Rating"
] = data["Rating"].median()


# ==========================================
# 9. Create useful columns
# ==========================================

# Total sales
data["Total_Sales"] = (
    data["Quantity"] * data["Unit_Price"]
)


# Age group
conditions = [
    data["Age"] < 25,
    (data["Age"] >= 25) & (data["Age"] < 40),
    data["Age"] >= 40
]

choices = [
    "Young",
    "Adult",
    "Senior"
]

data["Age_Group"] = np.select(
    conditions,
    choices,
    default="Unknown"
)


# Order month
data["Order_Month"] = data["Order_Date"].dt.month_name()


print("\nNew columns:")
print(
    data[
        [
            "Product",
            "Quantity",
            "Unit_Price",
            "Total_Sales",
            "Age_Group",
            "Order_Month"
        ]
    ].head()
)


# ==========================================
# 10. Analyze / Group
# ==========================================

# Total sales by product
sales_by_product = (
    data.groupby("Product")["Total_Sales"]
    .sum()
)

print("\nTotal sales by product:")
print(sales_by_product)


# Average price by product
average_price = (
    data.groupby("Product")["Unit_Price"]
    .mean()
)

print("\nAverage price by product:")
print(average_price)


# Number of orders by city
orders_by_city = (
    data.groupby("City")["Customer_ID"]
    .count()
)

print("\nOrders by city:")
print(orders_by_city)


# ==========================================
# 11. Export clean data
# ==========================================

data.to_csv(
    "data/customer_sales_clean.csv",
    index=False
)

print("\nClean data exported successfully!")
````

````python
sales = pd.DataFrame({"Date": pd.date_range("2026-01-01", periods=7, freq="D"),
                      "Sales": [100, 150, 80, 200, 170, 90, 220]})
sales["CumulativeSales"] = sales["Sales"].cumsum()
sales["RunningMax"] = sales["Sales"].cummax()
sales["PrevDay"] = sales["Sales"].shift(1)
sales["Change"] = sales["Sales"] - sales["PrevDay"]            # same as .diff()
sales["PctChange"] = (sales["Sales"].pct_change() * 100).round(1)
sales["MA3"] = sales["Sales"].rolling(3).mean().round(1)       # first 2 rows are NaN (not enough history)
sales
````

## Capstone — full solution

````python
cap = pd.DataFrame({
    "EmployeeID": range(1, 11),
    "Name": [" Ali ", "Ahmed", "Sara", "Hamza", "Ayesha", "Bilal", "Zara", "Usman", "Hina", "Omar"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "IT", "HR", "Finance"],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Lahore", "Islamabad", "Lahore", "Karachi", "Islamabad"],
    "Age": [25, 32, 28, 41, 24, 35, 29, 38, 27, 34],
    "Salary": [80000, 65000, 95000, 120000, 70000, 110000, 90000, 105000, 72000, 115000],
    "Experience": [2, 6, 4, 12, 2, 8, 5, 10, 3, 9],
    "JoinDate": ["2024-01-10", "2020-05-12", "2022-03-15", "2014-07-01", "2024-02-20",
                 "2018-09-10", "2021-01-05", "2016-11-20", "2023-06-10", "2017-04-18"],
})

# Part A — inspect
print(cap.shape, cap.columns.tolist(), sep="\n")
print(cap.dtypes)
cap.info()
print(cap.describe())

# Part B — clean
cap["Name"] = cap["Name"].str.strip()
cap["JoinDate"] = pd.to_datetime(cap["JoinDate"])
print("missing:", cap.isna().sum().sum(), "| duplicates:", cap.duplicated().sum())

# Part C — new columns
cap["AnnualSalary"] = cap["Salary"] * 12
cap["Bonus"] = cap["Salary"] * 0.10
cap["ExperienceLevel"] = np.select([cap["Experience"] < 3, cap["Experience"] <= 7],
                                   ["Junior", "Mid"], default="Senior")
cap["JoinYear"] = cap["JoinDate"].dt.year
cap.head()
````

````python
# Part D — filter
print(cap[cap["Department"] == "IT"]["Name"].tolist())
print(cap[cap["Salary"] > 100000]["Name"].tolist())
print(cap[(cap["City"] == "Lahore") & (cap["Experience"] > 5)]["Name"].tolist())
print(cap[(cap["ExperienceLevel"] == "Senior") & (cap["Salary"] > 100000)]["Name"].tolist(), "\n")

# Part E — analysis
print(cap.groupby("Department")["Salary"].agg(["mean", "sum"]), "\n")
print(cap["City"].value_counts(), "\n")
print("Highest paid:", cap.loc[cap["Salary"].idxmax(), "Name"])
cap["SalaryRank"] = cap["Salary"].rank(ascending=False).astype(int)

# Part F — reshaping
print(cap.pivot_table(values="Salary", index="Department", columns="City", aggfunc="mean"), "\n")
print(cap.pivot_table(values="EmployeeID", index="Department", columns="City", aggfunc="count", fill_value=0), "\n")

# Part G — final report
report = cap.sort_values("Salary", ascending=False)
print(report[["Name", "Department", "Salary", "SalaryRank"]].head(3))
report.to_csv(os.path.join(OUT, "employees_clean.csv"), index=False)
print("saved to", os.path.join(OUT, "employees_clean.csv"))
````
