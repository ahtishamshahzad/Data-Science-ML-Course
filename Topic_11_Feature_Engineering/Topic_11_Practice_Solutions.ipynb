### Practice Problem 5.1

Given exam scores `[35, 42, 55, 61, 68, 72, 77, 81, 88, 95, 99, 50]`:
1. Use `pd.cut` to create grades: 0–49 = F, 50–59 = D, 60–69 = C, 70–79 = B, 80–100 = A.
2. Use `pd.qcut` to split students into 3 equal-sized groups: Low, Medium, High.
3. Count students in each grade.

````python
# ✍️ Your attempt for Practice Problem 5.1
````

### Solution 5.1

````python
exam = pd.DataFrame({'Score': [35, 42, 55, 61, 68, 72, 77, 81, 88, 95, 99, 50]})

# right=False → intervals are [0,50), [50,60), ... so 50 is a D and 49.9 is an F
exam['Grade'] = pd.cut(
    exam['Score'],
    bins=[0, 50, 60, 70, 80, 101],
    labels=['F', 'D', 'C', 'B', 'A'],
    right=False
)

exam['Level'] = pd.qcut(exam['Score'], q=3, labels=['Low', 'Medium', 'High'])

print(exam['Grade'].value_counts().sort_index())
exam
````

Create a DataFrame of 6 app logins with columns `User` and `Login_Time` (include times on weekends and late at night).

### Solution 6.1

````python
logins = pd.DataFrame({
    'User': ['u1', 'u1', 'u2', 'u1', 'u2', 'u3'],
    'Login_Time': pd.to_datetime([
        '2025-03-01 02:30', '2025-03-03 10:00', '2025-03-02 23:15',
        '2025-03-15 04:45', '2025-03-20 18:00', '2025-03-22 09:30'
    ])
})

lt = logins['Login_Time']
logins['Month'] = lt.dt.month
logins['Day_Name'] = lt.dt.day_name()
logins['Hour'] = lt.dt.hour
logins['Is_Weekend'] = (lt.dt.dayofweek >= 5).astype(int)
logins['Is_Night'] = (logins['Hour'] < 6).astype(int)

# First login of each user, repeated on each row (transform from Section 4!)
first_login = logins.groupby('User')['Login_Time'].transform('min')
logins['Days_Since_First_Login'] = (lt - first_login).dt.days

logins
````

### Practice Problem 8.1

1. Create a column `Followers` with 300 values using `rng.exponential(scale=2000, size=300).round()` — make sure some values are 0 by setting the first 20 values to 0.
2. Calculate skewness.
3. Apply `np.log1p` and calculate skewness again.
4. Plot both histograms side by side.
5. Explain why `np.log` would be a problem here.

````python
# ✍️ Your attempt for Practice Problem 8.1
````

### Solution 8.1

````python
rng = np.random.default_rng(7)
fol = pd.DataFrame({'Followers': rng.exponential(scale=2000, size=300).round()})
fol.loc[:19, 'Followers'] = 0              # first 20 users have zero followers

fol['Log_Followers'] = np.log1p(fol['Followers'])

print('Skew before:', round(fol['Followers'].skew(), 2))
print('Skew after :', round(fol['Log_Followers'].skew(), 2))

fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
ax[0].hist(fol['Followers'], bins=30, edgecolor='black'); ax[0].set_title('Followers')
ax[1].hist(fol['Log_Followers'], bins=30, edgecolor='black'); ax[1].set_title('log1p(Followers)')
plt.tight_layout(); plt.show()

# np.log(0) = -inf, so the 20 users with 0 followers would get -inf values.
# np.log1p(0) = log(1) = 0, which is safe.
````

**Important observation:** here the skew goes from about **+1.9** to about **−1.8**! The 20 zeros become a separate lump far to the left of all the other log values, creating a new *left* tail. Log transformation is not magic — **always check the result**. With many zeros, a better option can be a zero-indicator column (`Followers == 0`) plus the log of the non-zero values.

3. How many output columns are there? List them.
4. Repeat with `interaction_only=True`. What changed?

````python
# ✍️ Your attempt for Practice Problem 10.1
````

### Solution 10.1

````python
abc = pd.DataFrame({'A': [1, 2, 3], 'B': [2, 4, 6], 'C': [0, 1, 0]})

p = PolynomialFeatures(degree=2, include_bias=False)
out = pd.DataFrame(p.fit_transform(abc), columns=p.get_feature_names_out(abc.columns))
print('Columns (degree 2):', out.shape[1], list(out.columns))

p2 = PolynomialFeatures(degree=2, include_bias=False, interaction_only=True)
out2 = pd.DataFrame(p2.fit_transform(abc), columns=p2.get_feature_names_out(abc.columns))
print('Columns (interaction_only):', out2.shape[1], list(out2.columns))
# interaction_only removes the squared terms A^2, B^2, C^2 → 9 columns become 6
out
````

Daily website visits for 10 days: `[500, 520, 480, 610, 700, 650, 720, 800, 760, 900]` starting 2025-06-01.
1. Create `Visits_Lag1` and `Visits_Lag2`.
2. Create a **leakage-safe** 3-day rolling mean, max and std.
3. Create `Pct_Change` using `pct_change()`.
4. Drop rows with missing values and report how many rows remain.

````python
# ✍️ Your attempt for Practice Problem 17.1
````

### Solution 17.1

````python
web = pd.DataFrame({
    'Date': pd.date_range('2025-06-01', periods=10),
    'Visits': [500, 520, 480, 610, 700, 650, 720, 800, 760, 900]
})
web['Visits_Lag1'] = web['Visits'].shift(1)
web['Visits_Lag2'] = web['Visits'].shift(2)

past = web['Visits'].shift(1)
web['Past3_Mean'] = past.rolling(3).mean().round(1)
web['Past3_Max'] = past.rolling(3).max()
web['Past3_Std'] = past.rolling(3).std().round(1)
web['Pct_Change'] = web['Visits'].pct_change().round(3)

web_clean = web.dropna()
print('Rows remaining:', len(web_clean))   # first 3 rows lost because of the 3-day past window
web
````

# 23. Complete End-to-End Project — Customer Feature Engineering

**Goal:** prepare a customer dataset for a future churn/value model. We will **not** build a complex model — the focus is feature engineering.

Steps:
1. Inspect the dataset
2. Create features
3. Check missing values
4. Check distributions
5. Check correlations
6. Identify potentially useless features
7. Discuss leakage risks
8. Prepare data for ML

## Step 0 — Create the dataset (40 customers)

````python
rng = np.random.default_rng(2025)
n = 40
snapshot_date = pd.Timestamp('2025-06-30')     # the day the data was extracted

cities = ['Lahore', 'Karachi', 'Islamabad', 'Multan', 'Faisalabad', 'Quetta', 'Sialkot']
city_probs = [0.30, 0.25, 0.15, 0.10, 0.10, 0.05, 0.05]

signup = snapshot_date - pd.to_timedelta(rng.integers(30, 1200, n), unit='D')
last_login = signup + pd.to_timedelta(
    (rng.random(n) * (snapshot_date - signup).days).astype(int), unit='D')

income = rng.lognormal(mean=11.2, sigma=0.6, size=n).round(-2)       # skewed incomes
orders_ = rng.poisson(lam=8, size=n)
spending = (orders_ * rng.normal(2500, 800, n).clip(300)).round(0)

customers = pd.DataFrame({
    'Customer_ID': [f'CUST{1000 + i}' for i in range(n)],
    'Age': rng.integers(18, 70, n).astype(float),
    'Gender': rng.choice(['Male', 'Female'], n),
    'City': rng.choice(cities, n, p=city_probs),
    'Income': income,
    'Signup_Date': signup.normalize(),
    'Last_Login': last_login.normalize(),
    'Total_Orders': orders_,
    'Total_Spending': spending,
    'Discount': (spending * rng.uniform(0, 0.3, n)).round(0),   # total discount amount received
})

# Add realistic messiness: missing values
customers.loc[rng.choice(n, 4, replace=False), 'Age'] = np.nan
customers.loc[rng.choice(n, 5, replace=False), 'Income'] = np.nan
customers.loc[rng.choice(n, 2, replace=False), 'Last_Login'] = pd.NaT

customers.head(10)
````

## Step 1 — Inspect the dataset

````python
print('Shape:', customers.shape)
print('\nData types:')
print(customers.dtypes)
print('\nUnique values per column:')
print(customers.nunique())
````

````python
customers.describe().round(1)
````

## Step 2 — Create features

````python
fe = customers.copy()     # keep the original untouched

# --- Missing indicators (BEFORE any filling) ---
fe['Age_Missing'] = fe['Age'].isna().astype(int)
fe['Income_Missing'] = fe['Income'].isna().astype(int)
fe['Last_Login_Missing'] = fe['Last_Login'].isna().astype(int)

# --- Binning ---
fe['Age_Group'] = pd.cut(fe['Age'], bins=[0, 18, 30, 50, 120],
                         labels=['Teen', 'Young Adult', 'Adult', 'Senior'])
fe['Income_Group'] = pd.qcut(fe['Income'], q=4, labels=['Low', 'Lower-Mid', 'Upper-Mid', 'High'])

# --- Date features ---
fe['Days_Since_Signup'] = (snapshot_date - fe['Signup_Date']).dt.days
fe['Days_Since_Last_Login'] = (snapshot_date - fe['Last_Login']).dt.days
fe['Signup_Month'] = fe['Signup_Date'].dt.month
fe['Signup_Month_Sin'] = np.sin(2 * np.pi * (fe['Signup_Month'] - 1) / 12)
fe['Signup_Month_Cos'] = np.cos(2 * np.pi * (fe['Signup_Month'] - 1) / 12)

# --- Ratio features (safe division) ---
orders_safe = fe['Total_Orders'].replace(0, np.nan)
fe['Average_Order_Value'] = (fe['Total_Spending'] / orders_safe).round(1)
fe['Spending_Per_Order'] = fe['Average_Order_Value']        # same formula → redundant on purpose (we'll detect it)
fe['Discount_Percentage'] = (fe['Discount'] / (fe['Total_Spending'] + fe['Discount']).replace(0, np.nan) * 100).round(1)
fe['Spending_to_Income'] = (fe['Total_Spending'] / fe['Income']).round(3)

# --- Categorical features ---
fe['Is_Major_City'] = fe['City'].isin(['Lahore', 'Karachi', 'Islamabad']).astype(int)
fe['City_Frequency'] = fe['City'].map(fe['City'].value_counts())   # (see Step 7: should be fit on train)
fe['Gender_City'] = fe['Gender'] + '_' + fe['City']

# --- Transformations ---
fe['Income_Log'] = np.log1p(fe['Income'])
fe['Spending_Log'] = np.log1p(fe['Total_Spending'])

# --- Customer activity features ---
tenure_months = (fe['Days_Since_Signup'] / 30).clip(lower=1)
fe['Orders_per_Month'] = (fe['Total_Orders'] / tenure_months).round(2)
fe['Is_Active_30d'] = (fe['Days_Since_Last_Login'] <= 30).astype(int)
fe['Is_New_Customer'] = (fe['Days_Since_Signup'] <= 90).astype(int)
fe['Never_Ordered'] = (fe['Total_Orders'] == 0).astype(int)

# A column we add just to show detection of useless features
fe['Data_Source'] = 'CRM'

print('Columns before:', customers.shape[1], ' after:', fe.shape[1])
fe.head()
````

## Step 3 — Check missing values

````python
missing = pd.DataFrame({
    'Missing_Count': fe.isna().sum(),
    'Missing_%': (fe.isna().mean() * 100).round(1)
})
missing[missing['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False)
````

**Observation:** Missing values in `Age`, `Income`, and `Last_Login` also propagate into engineered features (`Age_Group`, `Income_Group`, `Income_Log`, `Days_Since_Last_Login`, `Spending_to_Income`, …). That's expected — we'll impute inside the ML pipeline (fitted on train). The `_Missing` indicator columns preserve the information.

## Step 4 — Check distributions

````python
cols_to_plot = ['Income', 'Income_Log', 'Total_Spending', 'Average_Order_Value',
                'Days_Since_Signup', 'Days_Since_Last_Login']

fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.ravel(), cols_to_plot):
    ax.hist(fe[col].dropna(), bins=12, edgecolor='black')
    ax.set_title(f'{col}\nskew = {fe[col].skew():.2f}')
plt.tight_layout(); plt.show()
````

## Step 7 — Discuss leakage risks

| Feature | Leakage risk | Fix |
|---|---|---|
| `City_Frequency` | Counts computed on the **full** dataset (train + test) | Fit value counts on train only |
| `Income_Group` (`qcut`) | Quartile edges learned from the full dataset | Learn edges on train (or use fixed domain edges) |
| Imputation (median Age/Income) | Median learned from full data | Use `SimpleImputer` inside a Pipeline |
| Target encoding (not used) | Would use target | Train-only, out-of-fold |
| `Days_Since_Last_Login` | If predicting **churn**, and churn is defined as "no login in 60 days", this feature almost *is* the target definition | Compute login features from a window **before** the churn-observation period |
| `Total_Spending`, `Total_Orders` | If they include purchases **after** the prediction date | Compute only from history before the snapshot |
| Stateless features (ratios, logs, date parts, flags) | No learned statistics | Safe to create before split |

## Step 8 — Prepare data for ML

````python
# 8a. Drop useless / redundant / raw columns
drop_cols = ['Customer_ID', 'Data_Source', 'Spending_Per_Order', 'Signup_Date', 'Last_Login',
             'Gender_City', 'Income', 'Total_Spending', 'Signup_Month', 'Never_Ordered',
             # learned-on-full-data features: we re-create them safely below
             'City_Frequency', 'Income_Group']
model_df = fe.drop(columns=drop_cols)

# A demo target for this project (in real life this comes from future behaviour):
# customers inactive for > 180 days are labelled as churned
model_df['Churned'] = (fe['Days_Since_Last_Login'] > 180).astype(int)

# ⚠️ Days_Since_Last_Login DEFINES this demo target → it must be removed (target leakage!)
# Is_Active_30d is also derived from last login → remove as well.
model_df = model_df.drop(columns=['Days_Since_Last_Login', 'Is_Active_30d', 'Last_Login_Missing'])

X = model_df.drop(columns='Churned')
y = model_df['Churned']

# 8b. Split FIRST
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# 8c. Learned features, fitted on train only
X_train, X_test = X_train.copy(), X_test.copy()
city_freq = X_train['City'].value_counts()
X_train['City_Frequency'] = X_train['City'].map(city_freq)
X_test['City_Frequency'] = X_test['City'].map(city_freq).fillna(0)

print('Train:', X_train.shape, ' Test:', X_test.shape)
print('Churn rate train/test:', round(y_train.mean(), 2), '/', round(y_test.mean(), 2))
X_train.head()
````

````python
# 8d. Leakage-safe preprocessing pipeline (imputation → encoding → scaling)
numeric_cols = X_train.select_dtypes(include='number').columns.tolist()
categorical_cols = ['Gender', 'City', 'Age_Group']

prep = ColumnTransformer([
    ('num', Pipeline([
        ('impute', SimpleImputer(strategy='median')),
        ('scale', StandardScaler())
    ]), numeric_cols),
    ('cat', Pipeline([
        ('to_str', FunctionTransformer(lambda d: d.astype(object),     # category dtype → plain objects
                                        feature_names_out='one-to-one')),
        ('impute', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    ]), categorical_cols)
])

X_train_ready = prep.fit_transform(X_train)      # FIT on train
X_test_ready = prep.transform(X_test)            # only TRANSFORM test

ready_cols = prep.get_feature_names_out()
print('Final feature count:', len(ready_cols))
print('Any missing left in train?', np.isnan(X_train_ready).any())
pd.DataFrame(X_train_ready, columns=ready_cols).round(2).head()
````

### Problem 11 — Leakage-safe time-series features

For **each store separately**:

2. Past 3-day mean and past 7-day max (no leakage — current day excluded)

Explain why a random split would be wrong here.

````python
# ✍️ Problem 11 — your attempt
````

### Solution 11

````python
rng = np.random.default_rng(9)
p11 = pd.DataFrame({
    'Date': list(pd.date_range('2025-01-01', periods=15)) * 2,
    'Store': ['S1'] * 15 + ['S2'] * 15,
    'Sales': np.concatenate([rng.integers(80, 120, 15), rng.integers(300, 400, 15)])
})
p11 = p11.sort_values(['Store', 'Date']).reset_index(drop=True)

g = p11.groupby('Store')['Sales']
p11['Lag_1'] = g.shift(1)
p11['Lag_7'] = g.shift(7)
p11['Past3_Mean'] = g.transform(lambda s: s.shift(1).rolling(3).mean()).round(1)
p11['Past7_Max'] = g.transform(lambda s: s.shift(1).rolling(7).max())

cutoff = pd.Timestamp('2025-01-11')
train11 = p11[p11['Date'] <= cutoff]
test11 = p11[p11['Date'] > cutoff]
print('Train:', train11.shape, ' Test:', test11.shape)
p11.head(10)
````

**Explanation:**
- `groupby('Store')` ensures S2's first lag is `NaN`, not S1's last sale.
- `shift(1)` before `rolling` excludes the current day.
- A **random** split would put, e.g., 14 January in train and 5 January in test — the model would learn from the future to predict the past, which is impossible in reality and inflates scores. Time data must be split by date.

### Solution 12

````python
rng = np.random.default_rng(12)
p12 = pd.DataFrame({
    'Product_Category': rng.choice(['Electronics', 'Clothing', 'Books', 'Toys', 'Beauty', 'Garden'],
                                   size=60, p=[.3, .25, .2, .15, .07, .03]),
    'Returned': rng.integers(0, 2, 60)
})
tr12, te12 = train_test_split(p12, test_size=0.3, random_state=0)
tr12, te12 = tr12.copy(), te12.copy()

m = 5
gmean = tr12['Returned'].mean()
stats = tr12.groupby('Product_Category')['Returned'].agg(['mean', 'count'])
smoothed = (stats['count'] * stats['mean'] + m * gmean) / (stats['count'] + m)
print('Global train mean:', round(gmean, 3))
print(pd.DataFrame({'count': stats['count'], 'raw_mean': stats['mean'].round(3),
                    'smoothed': smoothed.round(3)}), '\n')

tr12['Cat_TE'] = tr12['Product_Category'].map(smoothed)
te12['Cat_TE'] = te12['Product_Category'].map(smoothed).fillna(gmean)

# WRONG: computed on the full dataset, including the test rows' own answers
wrong = p12.groupby('Product_Category')['Returned'].mean()
te12['Cat_TE_WRONG'] = te12['Product_Category'].map(wrong)

te12.round(3).head(10)
````

**Explanation:**
- Smoothing pulls small categories toward the global mean: a category with 2 rows barely moves from `gmean`, while a category with 15 rows stays close to its own mean.
- The **WRONG** column used the test rows' own `Returned` values when calculating the means, so it is secretly correlated with the answer → optimistic, unrealistic evaluation.
- Unseen categories in test (if any) get the global training mean instead of `NaN`.

### Solution 13

````python
p13 = pd.DataFrame({
    'Customer': ['A','B','A','C','D','B','A','E','C','D','F','B','E','A','F'],
    'Date': pd.to_datetime(['2025-01-03','2025-01-10','2025-02-14','2025-02-20','2025-03-01','2025-03-15',
                            '2025-03-20','2025-04-01','2025-04-11','2025-05-05','2025-05-06','2025-05-20',
                            '2025-06-01','2025-06-10','2025-06-25']),
    'Amount': [500, 1200, 700, 3000, 250, 900, 650, 4000, 2500, 300, 150, 1100, 3800, 800, 200]
})
snap13 = pd.Timestamp('2025-07-01')

rfm13 = p13.groupby('Customer').agg(
    Recency=('Date', lambda d: (snap13 - d.max()).days),
    Frequency=('Date', 'count'),
    Monetary=('Amount', 'sum')
)
rfm13['R'] = pd.qcut(rfm13['Recency'].rank(method='first'), 3, labels=[3, 2, 1]).astype(int)
rfm13['F'] = pd.qcut(rfm13['Frequency'].rank(method='first'), 3, labels=[1, 2, 3]).astype(int)
rfm13['M'] = pd.qcut(rfm13['Monetary'].rank(method='first'), 3, labels=[1, 2, 3]).astype(int)
rfm13['RFM_Score'] = rfm13[['R', 'F', 'M']].sum(axis=1)
rfm13['Label'] = pd.cut(rfm13['RFM_Score'], bins=[0, 4, 7, 9], labels=['At Risk', 'Regular', 'VIP'])
rfm13.sort_values('RFM_Score', ascending=False)
````

**Explanation:** a `lambda` inside `agg` computes days since each customer's last purchase. `rank(method='first')` breaks ties so `qcut` never fails. For recency, labels are reversed (`[3, 2, 1]`) because more recent (smaller) is better. `pd.cut` on the score converts it into segments.

### Solution 14

````python
rng = np.random.default_rng(14)
n = 200
p14 = pd.DataFrame({
    'User_ID': np.arange(n),
    'Platform': ['Android'] * n,
    'Premium': rng.choice([0, 1], n, p=[0.98, 0.02]),
    'Age': rng.integers(18, 60, n),
    'Sessions': rng.poisson(10, n),
})
p14['Sessions_x2'] = p14['Sessions'] * 2
p14['Minutes'] = p14['Sessions'] * 12 + rng.normal(0, 5, n)
p14['Noise'] = rng.normal(size=n)

nunique = p14.nunique()
top_share = p14.apply(lambda c: c.value_counts(normalize=True).iloc[0])

constant = nunique[nunique <= 1].index.tolist()
near_constant = top_share[(top_share >= 0.97) & (nunique > 1)].index.tolist()
# Identifier-like: 100% unique AND integer type.
# (Continuous columns like Minutes and Noise are also 100% unique, so unique ratio alone
#  cannot tell IDs from real measurements — we add the integer check + domain knowledge.)
is_int = p14.dtypes.apply(pd.api.types.is_integer_dtype)
id_like = nunique[(nunique == len(p14)) & is_int].index.tolist()

corr14 = p14.select_dtypes('number').corr().abs()
upper = corr14.where(np.triu(np.ones(corr14.shape, dtype=bool), k=1)).stack()
high_pairs = upper[upper > 0.95]

print('Constant      :', constant)
print('Near-constant :', near_constant)
print('ID-like       :', id_like)
print('Highly correlated pairs:\n', high_pairs.round(3))

to_drop = constant + id_like + ['Sessions_x2']     # keep one of each correlated pair
p14_clean = p14.drop(columns=to_drop)
print('\nKept columns:', list(p14_clean.columns))
````

**Explanation:** We drop the constant `Platform`, the identifier `User_ID`, and `Sessions_x2` (exact copy × 2). `Minutes` is also highly correlated with `Sessions` — whether to keep both is a judgement call (both are kept here). `Premium` is near-constant: we keep it for investigation because a rare premium flag could be very predictive.

### Solution 15

````python
rng = np.random.default_rng(15)
n = 80
p15 = pd.DataFrame({
    'Age': rng.integers(18, 70, n).astype(float),
    'Income': rng.lognormal(11, 0.5, n).round(-2),
    'City': rng.choice(['Lahore', 'Karachi', 'Islamabad', 'Multan'], n),
    'Signup_Date': pd.Timestamp('2025-01-01') - pd.to_timedelta(rng.integers(10, 900, n), unit='D'),
    'Orders': rng.poisson(5, n),
    'Spending': rng.integers(0, 50000, n),
})
p15.loc[rng.choice(n, 8, replace=False), 'Income'] = np.nan
p15['Bought_Again'] = ((p15['Orders'] > 5) | (p15['Spending'] > 30000)).astype(int)

# 1. Stateless features
snap15 = pd.Timestamp('2025-01-01')
p15['Days_Since_Signup'] = (snap15 - p15['Signup_Date']).dt.days
p15['Avg_Order_Value'] = p15['Spending'] / p15['Orders'].replace(0, np.nan)
p15['Income_Log'] = np.log1p(p15['Income'])
p15['Income_Missing'] = p15['Income'].isna().astype(int)

# 2. Drop raw date, split
X15 = p15.drop(columns=['Signup_Date', 'Bought_Again'])
y15 = p15['Bought_Again']
X_tr, X_te, y_tr, y_te = train_test_split(X15, y15, test_size=0.25, random_state=0, stratify=y15)

# 3. Preprocessing + model
num15 = ['Age', 'Income', 'Orders', 'Spending', 'Days_Since_Signup', 'Avg_Order_Value',
         'Income_Log', 'Income_Missing']
cat15 = ['City']
pre15 = ColumnTransformer([
    ('num', Pipeline([('imp', SimpleImputer(strategy='median')), ('sc', StandardScaler())]), num15),
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat15)
])
pipe15 = Pipeline([('prep', pre15), ('model', LogisticRegression(max_iter=1000))])

# 4. Fit + evaluate
pipe15.fit(X_tr, y_tr)
print('Test accuracy:', round(pipe15.score(X_te, y_te), 3))
print('Final features:', list(pipe15.named_steps['prep'].get_feature_names_out()))

# 5. Leakage discussion:
# 'Bought_Again' is DEFINED from Orders and Spending (Orders > 5 or Spending > 30000).
# So 'Orders', 'Spending' and 'Avg_Order_Value' (built from them) contain the answer → target leakage.
# The very high accuracy is a warning sign. In a real project, these features must be computed from a
# period BEFORE the 'buy again' window, or removed.
````

**Explanation:** the pipeline makes preprocessing leakage-safe (imputer and scaler fitted on train only). But a pipeline **cannot** protect you from target leakage created by *feature definitions* — that requires thinking about what is known at prediction time.
