### Topic 01 — Python Basics

**Q1.** What does `type(3 / 2)` return?

- **A)** `float`  ✅
- **B)** `int`
- **C)** `str`
- **D)** `Fraction`

> ✅ **Answer: A** — `/` always returns a float (1.5). Use `//` for floor division (`3 // 2` → 1).

**Q2.** Which of these data types is **immutable** (cannot be changed after creation)?

- **A)** list
- **B)** tuple  ✅
- **C)** dict
- **D)** set

> ✅ **Answer: B** — Tuples cannot be modified. Lists, dicts and sets can be changed in place.

**Q3.** What is the result of `[x**2 for x in range(5) if x % 2 == 0]`?

- **A)** `[1, 9]`
- **B)** `[0, 4, 16]`  ✅
- **C)** `[0, 1, 4, 9, 16]`
- **D)** `[4, 16]`

> ✅ **Answer: B** — `range(5)` is 0–4; the even ones are 0, 2, 4 and their squares are 0, 4, 16.

**Q4.** You write `sum = 0` and later call `sum([1, 2, 3])`. What happens?

- **A)** It returns 6
- **B)** It returns 0
- **C)** `SyntaxError`
- **D)** `TypeError: 'int' object is not callable`  ✅

> ✅ **Answer: D** — `sum = 0` replaces Python's built-in `sum()` function with the number 0. Never use built-in names (`sum`, `list`, `max`, `str` …) as variable names — use `total` instead.

### Topic 02 — Pandas DataFrame Basics

**Q5.** A DataFrame has the default index 0, 1, 2, … How many rows do `df.iloc[0:3]` and `df.loc[0:3]` return?

- **A)** 3 and 4  ✅
- **B)** 3 and 3
- **C)** 4 and 4
- **D)** 4 and 3

> ✅ **Answer: A** — `iloc` uses positions and excludes the end (0, 1, 2). `loc` uses labels and **includes** the end (0, 1, 2, 3).

**Q6.** Which expression returns a **DataFrame** (not a Series)?

- **A)** `df[['Age']]`  ✅
- **B)** `df['Age']`
- **C)** `df.Age`
- **D)** `df['Age'].values`

> ✅ **Answer: A** — Double brackets pass a *list* of columns, so the result is a DataFrame. Single brackets give a Series.

**Q7.** How do you select rows where Age > 30 **and** City is 'Lahore'?

- **A)** `df[df.Age > 30 and df.City == 'Lahore']`
- **B)** `df[df.Age > 30 & df.City == 'Lahore']`
- **C)** `df.where(Age > 30, City='Lahore')`
- **D)** `df[(df.Age > 30) & (df.City == 'Lahore')]`  ✅

> ✅ **Answer: D** — Use `&` (not `and`) and wrap each condition in parentheses, because `&` binds more tightly than `>` and `==`.

**Q8.** What does `df.groupby('City')['Salary'].mean()` return?

- **A)** A Series indexed by City  ✅
- **B)** A DataFrame with the same number of rows as `df`
- **C)** A single number
- **D)** A Python list

> ✅ **Answer: A** — One mean per city, returned as a Series whose index is the city names.

### Topic 03 — Handling Missing Values

**Q9.** What does `df.isna().mean()` give you?

- **A)** The number of missing values per column
- **B)** The mean of the non-missing values
- **C)** The rows that contain missing values
- **D)** The fraction (0–1) of missing values per column  ✅

> ✅ **Answer: D** — `isna()` is True/False; the mean of True/False is the share of True values. Multiply by 100 for a percentage.

**Q10.** A salary column is skewed and has a few huge outliers. Which simple value is usually the best fill for its missing values?

- **A)** Mean
- **B)** 0
- **C)** Median  ✅
- **D)** Maximum

> ✅ **Answer: C** — The median is not pulled by outliers, so it represents a typical salary better than the mean.

**Q11.** What does `df.dropna(thresh=3)` do?

- **A)** Drops rows with 3 or more missing values
- **B)** Drops the first 3 rows
- **C)** Keeps only rows that have **at least 3 non-missing** values  ✅
- **D)** Keeps only 3 columns

> ✅ **Answer: C** — `thresh` is the minimum number of non-missing values a row needs to be kept.

**Q12.** Why should an imputer (e.g. `SimpleImputer`) be **fitted on the training data only**?

- **A)** It is faster
- **B)** To avoid data leakage from the test set  ✅
- **C)** pandas requires it
- **D)** It removes more NaNs

> ✅ **Answer: B** — If the fill value (mean/median) is computed using test rows, information from the test set leaks into training.

### Topic 04 — Normalization & Standardization

**Q13.** Min-Max scale the value 20 from the data `[10, 20, 30]`. What is the result?

- **A)** 0.33
- **B)** 20
- **C)** 1
- **D)** 0.5  ✅

> ✅ **Answer: D** — (20 − min) / (max − min) = (20 − 10) / (30 − 10) = 0.5.

**Q14.** After `StandardScaler`, each column has…

- **A)** values between 0 and 1
- **B)** median 0 and IQR 1
- **C)** only whole numbers
- **D)** mean 0 and standard deviation 1  ✅

> ✅ **Answer: D** — Standardization computes z = (x − mean) / std.

**Q15.** Your data has strong outliers. Which scaler is least affected by them?

- **A)** `MinMaxScaler`
- **B)** `StandardScaler`
- **C)** `RobustScaler`  ✅
- **D)** None of them

> ✅ **Answer: C** — `RobustScaler` uses the median and IQR, which outliers barely change.

**Q16.** What is the correct order?

- **A)** Scale the whole dataset, then split
- **B)** Split, fit the scaler on train, transform train and test  ✅
- **C)** Fit the scaler on the test set
- **D)** Fit one scaler on train and another on test

> ✅ **Answer: B** — Learn the scaling from training data only, then apply that same scaling to the test data.

### Topic 05 — Outliers

**Q17.** Q1 = 20 and Q3 = 40. What is the upper fence of the 1.5 × IQR rule?

- **A)** 70  ✅
- **B)** 60
- **C)** 50
- **D)** 80

> ✅ **Answer: A** — IQR = 40 − 20 = 20. Upper fence = Q3 + 1.5 × IQR = 40 + 30 = 70.

**Q18.** Which Z-score rule is most commonly used to flag outliers?

- **A)** |z| > 1
- **B)** z > 0
- **C)** |z| > 3  ✅
- **D)** |z| > 10

> ✅ **Answer: C** — Values more than 3 standard deviations from the mean are rare in roughly normal data.

**Q19.** What does **capping (winsorization)** do?

- **A)** Deletes the outlier rows
- **B)** Sets outliers to NaN
- **C)** Replaces values beyond the bounds with the bound values  ✅
- **D)** Takes the log of every value

> ✅ **Answer: C** — Extreme values are pulled back to the lower/upper bound, so no rows are lost.

**Q20.** Which statement is true?

- **A)** Every outlier is an error and must be removed
- **B)** Detection is not treatment — investigate before deciding  ✅
- **C)** The Z-score method works best on very skewed data
- **D)** Boxplots cannot show outliers

> ✅ **Answer: B** — An outlier can be a real, important value (e.g. a large genuine order). Always check the cause first.

### Task 3 — Topic 03 — Fill the missing values

Create `clean`: a copy of `raw` where missing **Age** is filled with the **median** age and missing **City** with the **most frequent** city (mode).

```python
raw = pd.DataFrame({
    'Age': [25, np.nan, 31, 40, np.nan, 29],
    'City': ['Lahore', 'Karachi', None, 'Lahore', 'Lahore', None],
})

clean = raw.copy()
clean['Age'] = clean['Age'].fillna(clean['Age'].median())
clean['City'] = clean['City'].fillna(clean['City'].mode()[0])
clean

check(3)
```

### Task 4 — Topic 04 — Min-Max scaling by hand

Min-Max scale `marks` **without scikit-learn** and store the result in `scaled` (values between 0 and 1).

```python
marks = np.array([50, 60, 70, 80, 100])

scaled = (marks - marks.min()) / (marks.max() - marks.min())
scaled

check(4)
```

### Task 5 — Topic 05 — Find outliers with the IQR rule

Using the 1.5 × IQR rule, put the outlier values of `values` into a list `outliers`.

```python
values = pd.Series([12, 14, 15, 15, 16, 18, 19, 95, 2])

q1, q3 = values.quantile(0.25), values.quantile(0.75)
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = values[(values < lower) | (values > upper)].tolist()
print(lower, upper, outliers)

check(5)
```
