# 1. Selecting Data

### What & why
Picking the **columns** and/or **rows** you need — the most common operation in any analysis.

### Syntax
| Syntax | Returns | Meaning |
|---|---|---|
| `df['col']` | **Series** | One column |
| `df[['col1', 'col2']]` | **DataFrame** | Several columns (note the **double brackets** = a list) |
| `df.loc[rows, cols]` | Series/DataFrame/value | Select by **labels** (index labels & column names) |
| `df.iloc[rows, cols]` | Series/DataFrame/value | Select by **integer positions** (0, 1, 2 …) |

### `loc` vs `iloc` — the key difference
| | `loc` | `iloc` |
|---|---|---|
| Uses | **Labels** (names) | **Positions** (numbers) |
| Slice end | **Included**: `loc[0:3]` → labels 0, 1, 2, **3** | **Excluded**: `iloc[0:3]` → positions 0, 1, 2 |
| Accepts boolean masks | ✅ `loc[df['Age'] > 30]` | Only arrays, not Series |
| Column selection | By name: `loc[:, 'Salary']` | By number: `iloc[:, 5]` |

**Memory trick:** **l**oc = **l**abel, **i**loc = **i**nteger.

With the default index (0, 1, 2, …) labels and positions look the same — the difference becomes obvious once the index is something else (e.g. after sorting or `set_index`).

````python
print(type(emp['Name']), '← single brackets → Series')
print(type(emp[['Name', 'Salary']]), '← double brackets → DataFrame\n')

print('loc[0:3, ["Name","Age"]] → labels 0..3 INCLUDED')
print(emp.loc[0:3, ['Name', 'Age']], '\n')
print('iloc[0:3, [1, 2]] → positions 0..2 (3 EXCLUDED)')
print(emp.iloc[0:3, [1, 2]])
````

````python
# The difference becomes clear with a non-default index
by_name = emp.set_index('Name')                      # index labels are now names
print('loc["Sara", "Salary"]  →', by_name.loc['Sara', 'Salary'])      # by label
print('iloc[1, 4]             →', by_name.iloc[1, 4])                 # 2nd row, 5th column (Salary)
print(by_name.loc['Sara':'Omar', ['City', 'Salary']])                # label slice, end INCLUDED

sorted_emp = emp.sort_values('Salary', ascending=False)               # index is now shuffled
print('\nAfter sorting by salary: index =', sorted_emp.index.tolist())
print('sorted_emp.loc[0, "Name"]  →', sorted_emp.loc[0, 'Name'], '  (label 0 = Ali, wherever he is)')
print('sorted_emp.iloc[0]["Name"] →', sorted_emp.iloc[0]['Name'], ' (first ROW = highest salary)')
````

### Explanation
After sorting, label `0` still belongs to Ali, but position `0` is now the highest earner (Bilal). Mixing these up is a classic bug.

### When not
Don't use chained selection to **assign** values (`df['col'][0] = x`) — see Section 3.

### Common mistakes
- `df['col1', 'col2']` (missing inner brackets) → `KeyError`.
- Expecting `iloc[0:3]` to include position 3 (it doesn't) or `loc[0:3]` to exclude label 3 (it doesn't).
- Using `loc` with positions after the index changed.

# 2. Filtering Rows

### What & why
Keeping only the rows that meet a **condition** ("employees in Lahore older than 30").

### How it works
A condition like `df['Age'] > 30` produces a **boolean Series** (True/False per row). Passing it inside `df[...]` keeps the True rows.

| Operator | Meaning | Example |
|---|---|---|
| `&` | AND | `(df['Age'] > 30) & (df['City'] == 'Lahore')` |
| `\|` | OR | `(df['City'] == 'Lahore') \| (df['City'] == 'Karachi')` |
| `~` | NOT | `~(df['Department'] == 'IT')` |
| `.isin([...])` | value in a list | `df['City'].isin(['Lahore', 'Karachi'])` |
| `.between(a, b)` | a ≤ value ≤ b (inclusive by default) | `df['Age'].between(25, 35)` |
| `.query('...')` | condition as a string | `df.query('Age > 30 and City == "Lahore"')` |

### ⚠️ Why parentheses are required
`&` and `|` have **higher priority** than `>` and `==` in Python. So
`df['Age'] > 30 & df['City'] == 'Lahore'` is read as `df['Age'] > (30 & df['City']) == 'Lahore'` → error or nonsense. Always wrap each condition in `( )`.

Also: use `&`, `|`, `~` — **not** Python's `and`, `or`, `not` (those don't work element-wise on Series → `ValueError: The truth value of a Series is ambiguous`).

````python
print(emp[emp['Age'] > 30][['Name', 'Age']], '\n')
print(emp[emp['City'] == 'Lahore'][['Name', 'City']], '\n')

lahore_senior = emp[(emp['City'] == 'Lahore') & (emp['Age'] > 30)]           # AND
lahore_or_karachi = emp[(emp['City'] == 'Lahore') | (emp['City'] == 'Karachi')]  # OR
not_it = emp[~(emp['Department'] == 'IT')]                                     # NOT
print('AND :', lahore_senior['Name'].tolist())
print('OR  :', lahore_or_karachi['Name'].tolist())
print('NOT :', not_it['Name'].tolist())
````

````python
print('isin    :', emp[emp['Department'].isin(['IT', 'Finance'])]['Name'].tolist())
print('between :', emp[emp['Age'].between(25, 35)]['Name'].tolist())
min_salary = 90000
print('query   :', emp.query('Salary >= @min_salary and City != "Karachi"')['Name'].tolist())   # @ uses a Python variable

try:
    emp[emp['Age'] > 30 & emp['City'] == 'Lahore']          # ❌ missing parentheses
except Exception as e:
    print('\nWithout parentheses →', type(e).__name__)
````

### When not
- `query()` is great for readability, but column names with spaces need backticks (`` `Monthly Salary` > 5 ``), and it's harder to build dynamically.
- For many OR conditions on one column, use `isin()` instead of chaining `|`.

### Common mistakes
Missing parentheses · using `and`/`or` · forgetting that `between` includes both ends · filtering creates a **new** DataFrame (assign it to a variable).

# 3. Adding and Modifying Columns

### Syntax
```python
df['new'] = expression                               # add or overwrite a column
df.loc[condition, 'col'] = value                     # modify only some rows (safe way)
df['flag'] = np.where(condition, 'yes', 'no')        # if/else
df['level'] = np.select([cond1, cond2], ['A', 'B'], default='C')   # if / elif / else
df = df.assign(new=lambda d: d['a'] * 2)             # returns a NEW DataFrame (chainable)
```

| Tool | Use when |
|---|---|
| Arithmetic (`df['a'] * df['b']`) | Formula columns |
| `np.where(cond, x, y)` | Two outcomes |
| `np.select(conds, choices, default)` | Several ordered conditions (first True wins) |
| `assign()` | Method chaining; creating several columns in one expression |

### ⚠️ Copy-on-Write (pandas 3): no chained assignment
`df['Salary'][0] = 1` or `df[df['Age'] > 30]['Salary'] = 1` **never** modify `df` in pandas 3 (you get a `ChainedAssignmentError` warning). Use **one** `.loc[...] = ...` instead.

````python
e = emp.copy()                                                     # work on a copy so `emp` stays unchanged
e['Annual_Salary'] = e['Salary'] * 12                                 # arithmetic
e['Salary_per_Exp_Year'] = (e['Salary'] / e['Experience']).round(0)
e['Senior'] = np.where(e['Experience'] >= 10, 'Yes', 'No')           # two outcomes
e['Band'] = np.select(
    [e['Salary'] >= 110000, e['Salary'] >= 80000],                    # checked in order
    ['High', 'Medium'],
    default='Low')
e['Salary'] = e['Salary'].astype(float)                              # int → float first (a 10% raise creates decimals)
e.loc[e['Department'] == 'Sales', 'Salary'] = e['Salary'] * 1.10     # modify some rows safely with ONE .loc
e[['Name', 'Department', 'Salary', 'Annual_Salary', 'Senior', 'Band']]
````

````python
# assign(): build new columns in a chain without touching the original
summary = (emp
           .assign(Bonus=lambda d: d['Salary'] * 0.1,
                   Total=lambda d: d['Salary'] + d['Bonus'])        # can use Bonus created just before
           .loc[:, ['Name', 'Salary', 'Bonus', 'Total']])
print(summary.head(3))
print('\nIs there a Bonus column in the original emp?', 'Bonus' in emp.columns, '← assign() did not modify it')
````

### Common mistakes
- `np.select` conditions in the wrong order (the first True wins — put the strictest first).
- Chained assignment (silently does nothing in pandas 3).
- Writing floats into an **integer** column with `.loc` → pandas 3 raises `TypeError: Invalid value ... for dtype 'int64'`. Convert first (`astype(float)`) or round the new values to integers. (Writing `NaN` into an int column is allowed; the column becomes float.)
- Integer division surprises / dividing by zero → `inf` (check with `replace([np.inf, -np.inf], np.nan)`).

# 4. Sorting

### Syntax
```python
df.sort_values('col', ascending=False)                         # by values
df.sort_values(['col1', 'col2'], ascending=[True, False])      # by several columns
df.sort_values('col', na_position='first')                     # where NaNs go ('last' by default)
df.sort_index()                                                # by the index labels
```

### Why / when
Rankings, top-N lists, time order (before rolling/shift/diff!), readable reports.

### When not
Sorting huge DataFrames just to get the top 5 → `nlargest(5, 'col')` / `nsmallest` is faster.

````python
print(emp.sort_values('Salary', ascending=False)[['Name', 'Salary']].head(3), '\n')
print(emp.sort_values(['City', 'Salary'], ascending=[True, False])[['City', 'Name', 'Salary']], '\n')

with_nan = emp[['Name', 'Salary']].copy()
with_nan.loc[[2, 5], 'Salary'] = np.nan
print(with_nan.sort_values('Salary', na_position='first').head(4), '\n')
print('sort_index after sorting by salary:', emp.sort_values('Salary').sort_index().index.tolist())
print('nlargest(3):', emp.nlargest(3, 'Salary')['Name'].tolist())
````

### Explanation
Multiple columns sort by the first column, then break ties by the second. `ascending` can be a list (one per column). Missing values go to the end unless `na_position='first'`.

### Common mistakes
Forgetting to assign the result (`df.sort_values(...)` returns a new DataFrame) · forgetting to `reset_index(drop=True)` if you later use positions.

# 5. Renaming

```python
df.rename(columns={'old': 'new'})          # columns
df.rename(index={0: 'first'})              # index labels
df.columns = [...]                         # replace ALL names at once (must match length)
df.rename(columns=str.lower)               # apply a function to every name
```
**Why:** clean, consistent, meaningful names (no spaces, consistent case) make code simpler and prevent bugs.

````python
renamed = emp.rename(columns={'Emp_ID': 'Employee_ID', 'Salary': 'Monthly_Salary'})
print(renamed.columns.tolist())
print(emp.rename(columns=str.lower).columns.tolist())
print(emp.set_index('Name').rename(index={'Ali': 'Ali Khan'}).index[:3].tolist())

messy = pd.DataFrame(columns=[' Order ID', 'Customer Name ', 'UNIT PRICE'])
messy.columns = messy.columns.str.strip().str.lower().str.replace(' ', '_')   # clean all names at once
print(messy.columns.tolist())
````

### Common mistakes
Misspelling an old name in `rename` (no error — the column is just not renamed) · assigning `df.columns = [...]` with the wrong number of names.

# 6. Dropping Data

```python
df.drop(columns=['col1', 'col2'])          # drop columns (clearest form)
df.drop(index=[0, 3])                      # drop rows by label
df.drop('col', axis=1)                     # axis=1 → columns, axis=0 → rows
df.dropna()                                # drop rows with ANY missing value
df.dropna(subset=['Salary'])               # only consider some columns
df.dropna(thresh=5)                        # keep rows with at least 5 non-missing values
df.dropna(axis=1)                          # drop COLUMNS containing missing values
```

### `axis` in one sentence
`axis=0` = **rows** (move down), `axis=1` = **columns** (move across). Using `columns=` / `index=` avoids the confusion.

### Why `inplace=True` is often avoided
- It **doesn't** make things faster or save memory in practice; pandas usually creates a copy internally anyway.
- It returns `None`, so it breaks **method chaining** and causes bugs like `df = df.drop(..., inplace=True)` → `df` becomes `None`!
- The pandas developers discourage it. Prefer `df = df.drop(...)`.

````python
print(emp.drop(columns=['Emp_ID', 'Experience']).columns.tolist())
print(emp.drop(index=[0, 1]).index.tolist())

m = emp[['Name', 'Age', 'Salary']].copy()
m.loc[[1, 4], 'Salary'] = np.nan; m.loc[4, 'Age'] = np.nan
print('\ndropna()            →', len(m.dropna()), 'rows')
print('dropna(subset=Age)  →', len(m.dropna(subset=['Age'])), 'rows')
print('dropna(axis=1)      →', m.dropna(axis=1).columns.tolist())

result = m.drop(columns='Age', inplace=True)
print('\nWhat inplace=True returns:', result, '← so never write m = m.drop(..., inplace=True)')
````

### Common mistakes
`dropna()` without `subset` dropping far more rows than intended · the `inplace=True` + assignment trap · dropping rows before understanding **why** values are missing (Topic 3).

# 7. Index Manipulation

```python
df.set_index('col')                 # use a column as the index (row labels)
df.reset_index()                    # move the index back into a column
df.reset_index(drop=True)           # discard the old index, get 0..n-1
```

### Why indexes matter
- **Fast, readable lookups** by label: `df.loc['Sara']`.
- **Automatic alignment**: arithmetic between two Series lines up by index labels.
- Many operations produce a meaningful index: `groupby`, `pivot_table`, `resample` (time index).

### When not
Don't hide important columns in the index when you'll need them as normal columns (merging, plotting, exporting) — `reset_index()` brings them back.

````python
by_id = emp.set_index('Emp_ID')
print(by_id.loc[105, ['Name', 'Salary']], '\n')

bonus = pd.Series({103: 5000, 101: 2000, 999: 1000})             # different order, one unknown ID
print('Automatic alignment by index (NaN where no match):')
print((by_id['Salary'] + bonus).head(4), '\n')

city_avg = emp.groupby('City')['Salary'].mean()                   # City becomes the index
print(city_avg, '\n')
print(city_avg.reset_index())                                     # back to a normal column

filtered = emp[emp['Age'] > 35]
print('\nIndex after filtering:', filtered.index.tolist(), '→ reset_index(drop=True):', filtered.reset_index(drop=True).index.tolist())
````

### Common mistakes
Forgetting `drop=True` (adds an unwanted `index` column) · duplicate index labels making `loc` return several rows · unexpected NaNs caused by index alignment.

# 8. Applying Functions — `apply`, `map`, `DataFrame.map` (formerly `applymap`)

| Method | Works on | Function receives | Typical use |
|---|---|---|---|
| `Series.map(func or dict)` | One column | **Each value** | Transform/look up values: `s.map({'M': 'Male'})` |
| `Series.apply(func)` | One column | Each value | Like `map` for functions |
| `DataFrame.apply(func, axis=0)` | DataFrame | **Each column** (a Series) | Column-wise summaries |
| `DataFrame.apply(func, axis=1)` | DataFrame | **Each row** (a Series) | Logic that needs several columns (slow!) |
| `DataFrame.map(func)` | DataFrame | **Each cell** | Element-wise formatting (was `applymap`) |

⚠️ **`applymap()` was deprecated in pandas 2.1 and removed in pandas 3.0** → use `DataFrame.map()`.

### Modern alternative: vectorized operations
`apply` with a Python function runs a **Python loop** internally. Built-in vectorized operations (`df['a'] * df['b']`, `np.where`, `.str`, `.dt` methods) run in optimized C code and are often **10–100× faster**. Use `apply` only when no vectorized option exists.

````python
print(emp['Department'].map({'IT': 'Tech', 'HR': 'People', 'Finance': 'Money', 'Sales': 'Revenue'}).head(4).tolist())
print(emp['Name'].map(len).head(4).tolist(), '← map with a function')
print(emp['Salary'].apply(lambda s: 'high' if s > 90000 else 'normal').head(4).tolist())

print('\napply on columns (axis=0):')
print(emp[['Age', 'Salary', 'Experience']].apply(lambda col: col.max() - col.min()))

print('\napply on rows (axis=1):')
emp_rows = emp.apply(lambda r: f"{r['Name']} ({r['City']})", axis=1)
print(emp_rows.head(3).tolist())

print('\nDataFrame.map (element-wise, replaces applymap):')
print(emp[['Salary', 'Age']].head(3).map(lambda v: f'{v:,}'))
````

````python
import time
big = pd.DataFrame({'price': np.random.default_rng(0).uniform(10, 1000, 300_000),
                    'qty': np.random.default_rng(1).integers(1, 10, 300_000)})

t0 = time.perf_counter(); r1 = big.apply(lambda r: r['price'] * r['qty'], axis=1); t_apply = time.perf_counter() - t0
t0 = time.perf_counter(); r2 = big['price'] * big['qty'];                             t_vec = time.perf_counter() - t0
print(f'apply(axis=1): {t_apply:.3f} s | vectorized: {t_vec:.4f} s | same result: {np.allclose(r1, r2)}')
print(f'→ vectorized is about {t_apply / t_vec:,.0f}× faster')
````

### Common mistakes
Using `apply(axis=1)` for simple arithmetic · `map` with a dict that misses some keys → those values become **NaN** · calling the removed `applymap` in pandas 3.

# 9. GroupBy (deep dive)

### What
**Split → Apply → Combine**:
```text
Split the rows into groups (e.g. by City) → apply a function to each group (e.g. mean salary) → combine the results into one table
```

### Syntax
```python
df.groupby('City')['Salary'].mean()                        # one column, one statistic
df.groupby('City')[['Salary', 'Age']].agg(['mean', 'max'])  # several statistics
df.groupby(['City', 'Department'])['Salary'].sum()          # several grouping keys → MultiIndex
df.groupby('City').agg(                                     # NAMED aggregation (recommended)
    avg_salary=('Salary', 'mean'),
    max_salary=('Salary', 'max'),
    employees=('Salary', 'count'),
)
```

Common functions: `sum, mean, median, count, size, min, max, std, nunique, first, last`.
- `count` = non-missing values; `size` = all rows (including NaN).

### Why / when
Any question with "**per** …" or "**by** …": average salary **per** city, total sales **by** month, number of orders **per** customer.

### When not
When you need the group statistic **on every row** → `transform` (Section 10). For a 2-D summary table → `pivot_table` (Section 12).

````python
print('Mean salary by city:'); print(emp.groupby('City')['Salary'].mean().round(0), '\n')
for f in ['sum', 'median', 'count', 'min', 'max', 'std']:
    print(f'{f:<7}', emp.groupby('City')['Salary'].agg(f).round(0).to_dict())
````

````python
print('Multiple aggregations:')
print(emp.groupby('Department')[['Salary', 'Experience']].agg(['mean', 'max']).round(1), '\n')

print('Named aggregation (clean column names):')
city_summary = emp.groupby('City').agg(
    avg_salary=('Salary', 'mean'),
    max_salary=('Salary', 'max'),
    employees=('Salary', 'count'),
    avg_age=('Age', 'mean'),
).round(1)
print(city_summary, '\n')

print('Two grouping keys:')
print(emp.groupby(['City', 'Department'])['Salary'].sum().reset_index())
````

### Expected output
Named aggregation gives a tidy table with **one row per city** and readable column names (`avg_salary`, `max_salary`, …).

### Common mistakes
- Forgetting `reset_index()` when you need the group keys as columns (or use `as_index=False`).
- `count` vs `size` confusion.
- Grouping by a column with NaN: by default NaN groups are **dropped** (`dropna=False` keeps them).
- Multi-level column names from `.agg(['mean', 'max'])` that are awkward to use → prefer named aggregation.

# 10. `groupby().transform()`

### What
Computes a group statistic but returns a result **the same length as the original DataFrame** — each row gets the value of **its own group**.

| | `agg()` | `transform()` |
|---|---|---|
| Output rows | **One per group** | **Same as original** (one per row) |
| Use for | Summary tables | New columns: group mean on each row, % of group, rank within group, standardise within group, fill NaN with group mean |

### Syntax
```python
df['City_Avg'] = df.groupby('City')['Salary'].transform('mean')
```

````python
t = emp[['Name', 'City', 'Salary']].copy()
print('agg → one row per city:', emp.groupby('City')['Salary'].agg('mean').shape)
t['City_Avg'] = t.groupby('City')['Salary'].transform('mean').round(0)               # repeated for each row
t['Diff_vs_City'] = t['Salary'] - t['City_Avg']
t['Share_of_City_Total_%'] = (t['Salary'] / t.groupby('City')['Salary'].transform('sum') * 100).round(1)
t['Rank_in_City'] = t.groupby('City')['Salary'].rank(ascending=False).astype(int)
t.sort_values(['City', 'Rank_in_City'])
````

````python
# Practical: fill missing salaries with the average of the same department
s_missing = emp[['Name', 'Department', 'Salary']].copy()
s_missing.loc[[0, 6], 'Salary'] = np.nan
s_missing['Salary_Filled'] = s_missing['Salary'].fillna(s_missing.groupby('Department')['Salary'].transform('mean'))
s_missing
````

### Common mistakes
Using `agg` and then trying to assign it as a column (lengths don't match / misaligned) · forgetting `transform` keeps the **original index** (that's why assignment works).

# 11. `groupby().filter()`

### What
Keeps or drops **whole groups** based on a condition computed on the group. The result has the **original rows** (not a summary).

```python
df.groupby('City').filter(lambda g: len(g) >= 3)           # keep cities with at least 3 employees
df.groupby('Department').filter(lambda g: g['Salary'].mean() > 80000)
```

### When
"Only analyse customers with ≥ 5 orders", "only products sold in ≥ 3 cities", "remove tiny groups before computing group statistics".

### When not
Filtering individual rows → use a normal boolean filter. On large data with many groups, `filter` with a Python function can be slow → a `transform`-based mask is faster:
`df[df.groupby('City')['City'].transform('size') >= 3]`.

````python
big_cities = emp.groupby('City').filter(lambda g: len(g) >= 4)
print('Cities with ≥ 4 employees:', big_cities['City'].unique().tolist(), '| rows kept:', len(big_cities))

rich_depts = emp.groupby('Department').filter(lambda g: g['Salary'].mean() > 80000)
print('Departments with mean salary > 80k:', rich_depts['Department'].unique().tolist())

fast = emp[emp.groupby('City')['City'].transform('size') >= 4]
print('Same result with transform mask:', fast.equals(big_cities))
````

# 12. Pivot Tables — `pd.pivot_table()`

### What
A 2-D summary table (like Excel's PivotTable): **rows** from one column, **columns** from another, **cells** = an aggregate of a third column.

```python
pd.pivot_table(df,
               index='City',            # rows
               columns='Department',    # columns
               values='Salary',         # what to summarise
               aggfunc='mean',          # how: 'mean', 'sum', 'count', ['mean', 'max'], {'Salary': 'mean', 'Age': 'max'}
               fill_value=0,            # replace empty cells
               margins=True,            # add an 'All' row/column with totals
               margins_name='Total')
```

### When / when not
✅ Two-dimensional summaries for reports and heatmaps. ❌ Just counts of two categories → `crosstab` is shorter. ❌ One grouping key → `groupby` is simpler.

````python
pt = pd.pivot_table(emp, index='City', columns='Department', values='Salary', aggfunc='mean', fill_value=0,
                    margins=True, margins_name='Total').round(0)
print(pt, '\n')
print(pd.pivot_table(emp, index='City', values=['Salary', 'Age'], aggfunc={'Salary': 'sum', 'Age': 'mean'}).round(1))
````

### Explanation
Each cell is the average salary of that city–department combination; `0` means no employees there (from `fill_value=0`). The `Total` row/column is computed from the **raw data**, not by averaging the averages.

### Common mistakes
`fill_value=0` for a **mean** can be misleading (0 ≠ "no data" — consider leaving NaN) · forgetting `values=` (pandas aggregates every numeric column) · confusing `pivot` (no aggregation, fails on duplicates — Section 21) with `pivot_table`.

# 14. String Manipulation — the `.str` accessor

| Method | What it does | Example result |
|---|---|---|
| `.str.lower()` / `.str.upper()` / `.str.title()` | Change case | `'lahore'`, `'LAHORE'`, `'Lahore'` |
| `.str.strip()` | Remove spaces at both ends | `'  Ali '` → `'Ali'` |
| `.str.contains('x', case=False)` | True if the text contains a pattern | Filter emails from gmail |
| `.str.startswith('+92')` / `.str.endswith('.pk')` | Check beginning / end | Phone/domain checks |
| `.str.replace('-', '', regex=False)` | Replace text | Clean phone numbers |
| `.str.split('@')` / `.str.split('@', expand=True)` | Split into a list / into columns | User & domain |
| `.str.len()` | Length of each text | Detect short/invalid codes |
| `.str.extract(r'(\d+)')` | Pull out a **regex** group | Numbers inside text |

### Why
Real text data is messy: extra spaces, mixed case, inconsistent formats. Vectorized `.str` methods clean an entire column in one line (and handle missing values gracefully).

````python
customers = pd.DataFrame({
    'Name':  ['  ali khan', 'SARA AHMED ', 'Omar farooq', 'hina  Malik', None],
    'Email': ['ali.khan@gmail.com', 'sara@company.pk', 'OMAR@Gmail.com', 'hina@yahoo.com', 'test@company.pk'],
    'Phone': ['+92-300-1234567', '0321 7654321', '+92 333 1112223', '0300-9998887', '+92-345-0001112'],
    'Product_Code': ['ELEC-1001', 'FASH-2040', 'ELEC-1500', 'BOOK-3002', 'FASH-2999'],
    'Address': ['House 12, Street 5, Lahore', 'Flat 3B, Clifton, Karachi', 'Plot 44, F-7, Islamabad', 'House 9, Gulberg, Lahore', 'Street 2, Saddar, Karachi'],
})

c = customers.copy()
c['Name'] = c['Name'].str.strip().str.replace(r'\s+', ' ', regex=True).str.title()   # trim, collapse spaces, Title Case
c['Email'] = c['Email'].str.lower()
c['Is_Gmail'] = c['Email'].str.contains('gmail', case=False)
c['Is_PK_Domain'] = c['Email'].str.endswith('.pk')
c['Phone_Clean'] = c['Phone'].str.replace(r'[^\d]', '', regex=True).str.replace(r'^92', '0', regex=True)  # keep digits, +92 → 0
c['Phone_OK'] = c['Phone_Clean'].str.len() == 11
c['Intl_Format'] = c['Phone'].str.startswith('+92')
c[['User', 'Domain']] = c['Email'].str.split('@', expand=True)                       # split into two columns
c['Category_Code'] = c['Product_Code'].str.extract(r'^([A-Z]+)-')                    # letters before the dash
c['Code_Number'] = c['Product_Code'].str.extract(r'-(\d+)$').astype(int)             # digits after the dash
c['City'] = c['Address'].str.split(',').str[-1].str.strip()                         # last part of the address
c.drop(columns=['Phone', 'Address'])
````

### Explanation
- Missing names stay missing (`None`/NaN) — `.str` methods don't crash on them.
- `regex=True` lets you use patterns: `\s+` = one or more spaces, `[^\d]` = anything that is not a digit, `^92` = "92" at the start.
- `str.extract` returns the part matched inside `( )`.

### Common mistakes
Forgetting `regex=False` when replacing special characters like `.` or `+` (they have special regex meanings) · `str.contains` on columns with NaN inside a filter → use `na=False` · calling `.str` on a numeric column (convert with `astype(str)` first).

# 15. Date/Time Manipulation

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')   # text → datetime (format avoids ambiguity)
df['Date'].dt.year / .dt.month / .dt.day / .dt.day_name() / .dt.hour / .dt.quarter / .dt.dayofweek
df[(df['Date'] >= '2025-03-01') & (df['Date'] < '2025-04-01')]    # date filtering
df.set_index('Date').loc['2025-03']                                # partial string selection (sorted index)
df.groupby(df['Date'].dt.to_period('M'))['Sales'].sum()            # monthly grouping
df.set_index('Date').resample('W')['Sales'].sum()                  # weekly totals
```

### Why
Dates stored as text can't be compared, sorted correctly (`'10/01'` < `'2/01'` as text!) or grouped by month. `datetime` dtype unlocks all time-based analysis.

### Common mistakes
- Ambiguous formats: `01/02/2025` is 1 February in Pakistan/UK but January 2 in the US → always pass `format=` (or `dayfirst=True`).
- Bad values → use `errors='coerce'` (they become `NaT`) and then inspect them.
- `resample` needs a datetime **index** (or `on='Date'`).

````python
orders = pd.DataFrame({
    'Order_ID': range(1, 11),
    'Order_Time': ['05/01/2025 09:15', '17/01/2025 14:30', '02/02/2025 20:05', '14/02/2025 11:45', '28/02/2025 16:20',
                   '03/03/2025 08:10', '15/03/2025 19:55', '22/03/2025 13:00', 'not a date', '30/03/2025 22:40'],
    'Amount': [1200, 3400, 800, 5600, 2300, 4100, 900, 3000, 1500, 2700],
})
orders['Order_Time'] = pd.to_datetime(orders['Order_Time'], format='%d/%m/%Y %H:%M', errors='coerce')
print('Unparseable rows:', orders['Order_Time'].isna().sum())
ot = orders['Order_Time'].dt
orders = orders.assign(Year=ot.year.astype('Int64'), Month=ot.month.astype('Int64'), Day=ot.day.astype('Int64'),
                       Weekday=ot.day_name(), Hour=ot.hour.astype('Int64'),   # 'Int64' = integers that allow missing (<NA>)
                       Is_Weekend=ot.dayofweek >= 5, Evening=ot.hour >= 18)
orders
````

````python
valid = orders.dropna(subset=['Order_Time'])
print('March orders (date filter):', valid[(valid['Order_Time'] >= '2025-03-01') & (valid['Order_Time'] < '2025-04-01')]['Order_ID'].tolist())
print('\nMonthly revenue (to_period):'); print(valid.groupby(valid['Order_Time'].dt.to_period('M'))['Amount'].sum())
print('\nWeekly revenue (resample):'); print(valid.set_index('Order_Time').resample('W')['Amount'].sum().head(4))
print('\nDays between first and last order:', (valid['Order_Time'].max() - valid['Order_Time'].min()).days)
````

# 16. Duplicate Handling

```python
df.duplicated()                                   # True for rows that repeat an EARLIER row
df.duplicated(subset=['Email'], keep=False)       # mark ALL copies
df.drop_duplicates()                              # remove exact duplicate rows (keeps the first)
df.drop_duplicates(subset=['Email'], keep='last') # one row per email, keep the latest
```

| `keep=` | Marks as duplicate |
|---|---|
| `'first'` (default) | All copies except the first |
| `'last'` | All copies except the last |
| `False` | **All** copies (useful to inspect them) |

### When not
Don't blindly drop duplicates: two identical orders might be **real** (a customer bought the same item twice). Define what makes a row unique (e.g. `Order_ID`) and use `subset=`.

````python
dups = pd.DataFrame({
    'Customer': ['Ali', 'Sara', 'Ali', 'Omar', 'Sara', 'Hina'],
    'Email': ['ali@x.com', 'sara@x.com', 'ali@x.com', 'omar@x.com', 'SARA@x.com', 'hina@x.com'],
    'Signup': pd.to_datetime(['2025-01-01', '2025-01-05', '2025-01-01', '2025-02-01', '2025-03-10', '2025-02-15']),
})
print('Exact duplicates:', dups.duplicated().sum())
print(dups[dups.duplicated(keep=False)], '\n')

dups['Email_Clean'] = dups['Email'].str.lower()                               # 'SARA@x.com' is the same person!
print('Duplicates by cleaned email:', dups.duplicated(subset=['Email_Clean']).sum())
latest = dups.sort_values('Signup').drop_duplicates(subset=['Email_Clean'], keep='last')
latest
````

### Common mistakes
Checking duplicates **before** cleaning text (case/spaces hide duplicates) · forgetting to sort before `keep='last'` · removing legitimate repeated transactions.

# 17. Missing Value Manipulation

| Method | What it does | Use when |
|---|---|---|
| `fillna(value)` | Replace NaN with a constant | Categorical "Unknown", 0 when missing truly means zero |
| `fillna(df['col'].median())` | Fill with a statistic | Numeric, roughly random missingness (fit on training data in ML!) |
| `ffill()` | Copy the **previous** value forward | Time series / sensor readings where the last known value still holds (stock price, status) |
| `bfill()` | Copy the **next** value backward | Gaps at the start of a series; when the next value is the best guess |
| `interpolate()` | Estimate between neighbours (linear by default; `method='time'` for dates) | Smooth numeric series (temperature, gradual trends) |

### When not
- `ffill`/`bfill`/`interpolate` on **unordered** data (sort first!) or across **different groups** (use `groupby().ffill()`).
- Interpolating categories or long gaps.
- Filling a target variable, or filling with statistics from the test set (leakage).

````python
temps = pd.DataFrame({'Date': pd.date_range('2025-06-01', periods=8),
                      'Temp': [30.0, np.nan, 34.0, np.nan, np.nan, 40.0, np.nan, 38.0],
                      'Status': ['OK', None, None, 'Maintenance', None, 'OK', None, None]})
temps['fillna_mean'] = temps['Temp'].fillna(temps['Temp'].mean()).round(1)
temps['ffill'] = temps['Temp'].ffill()
temps['bfill'] = temps['Temp'].bfill()
temps['interpolate'] = temps['Temp'].interpolate()
temps['Status_ffill'] = temps['Status'].ffill()
temps
````

````python
# Group-aware forward fill: never carry a value from one store into another
stock = pd.DataFrame({'Store': ['A', 'A', 'A', 'B', 'B', 'B'], 'Day': [1, 2, 3, 1, 2, 3], 'Stock': [50, np.nan, 45, np.nan, 30, np.nan]})
stock['Wrong_ffill'] = stock['Stock'].ffill()
stock['Group_ffill'] = stock.groupby('Store')['Stock'].ffill()
stock
````

### Explanation
- Linear interpolation fills 34 → 40 with 36 and 38 (a straight line); `ffill` repeats 34; `bfill` uses 40.
- In the store example, the wrong `ffill` copies Store A's value (45) into Store B's first day; the group-aware version leaves it as NaN.

# 18. Replace and Map

| | `replace()` | `map()` |
|---|---|---|
| Values **not** in the dictionary | **Kept unchanged** | Become **NaN** |
| Works on | Series or whole DataFrame | Series |
| Also supports | Regex, lists, nested dicts | Functions, Series lookups |
| Best for | Fixing a few specific values (typos, codes) | Complete recodings / lookups (every value has a mapping) |

```python
df['Gender'].replace({'M': 'Male', 'F': 'Female'})
df['Size'].map({'S': 1, 'M': 2, 'L': 3})          # ordinal encoding (Topic 6)
df.replace({'N/A': np.nan, '-': np.nan})          # standardise missing-value markers everywhere
```

````python
survey = pd.DataFrame({'Gender': ['M', 'F', 'Female', 'm', 'Other'], 'Size': ['S', 'L', 'M', 'XL', 'M'],
                       'Rating': ['5', 'N/A', '4', '-', '3']})
s = survey.copy()
s['Gender_replace'] = s['Gender'].str.upper().replace({'M': 'Male', 'F': 'Female', 'FEMALE': 'Female'})   # unmatched kept
s['Gender_map'] = s['Gender'].str.upper().map({'M': 'Male', 'F': 'Female'})                             # unmatched → NaN
s['Size_Code'] = s['Size'].map({'S': 1, 'M': 2, 'L': 3, 'XL': 4})
s['Rating'] = s['Rating'].replace({'N/A': np.nan, '-': np.nan}).astype(float)
s
````

### Common mistakes
Using `map` with an incomplete dictionary (silent NaNs) · forgetting to clean case/spaces first · `replace` on a whole DataFrame changing values in columns you didn't intend.

## 21.1 Reshaping: MultiIndex, `stack`, `unstack`, `melt`, `pivot`, `pivot_table`, `wide_to_long`

| Function | From → to | Simple meaning |
|---|---|---|

## 21.3 Ranking and ordered calculations: `rank`, `shift`, `diff`, `pct_change`, cumulative functions

| Method | Meaning |
|---|---|
| `rank(ascending=False, method='dense')` | Position in sorted order (ties handled by `method`: `'average'`, `'min'`, `'dense'`, `'first'`) |
| `shift(1)` | Value from the **previous** row (lag) |
| `diff()` | Current − previous |
| `pct_change()` | (Current − previous) / previous |
| `cumsum()` / `cumprod()` | Running total / running product |
| `cummax()` / `cummin()` | Highest / lowest value so far |

⚠️ These depend on **row order** — sort first (usually by date), and use `groupby(...)` when rows belong to different entities.

````python
monthly = pd.DataFrame({'Month': pd.period_range('2025-01', periods=8, freq='M'),
                        'Revenue': [100, 120, 115, 140, 160, 150, 175, 190],
                        'Growth_Factor': [1.00, 1.05, 0.98, 1.10, 1.02, 0.95, 1.08, 1.04]})
m = monthly.copy()
m['Rank'] = m['Revenue'].rank(ascending=False, method='dense').astype(int)
m['Prev_Month'] = m['Revenue'].shift(1)
m['Change'] = m['Revenue'].diff()
m['Pct_Change_%'] = (m['Revenue'].pct_change() * 100).round(1)
m['Cumulative'] = m['Revenue'].cumsum()
m['Best_So_Far'] = m['Revenue'].cummax()
m['Worst_So_Far'] = m['Revenue'].cummin()
m['Compounded_Index'] = m['Growth_Factor'].cumprod().round(3)
m
````

<a id="sec-25"></a>
## 21.4 Window functions: `rolling` and `expanding`

| | `rolling(3)` | `expanding()` |
|---|---|---|
| Window | The **last 3 rows** (moves along) | **All rows so far** (grows) |
| Example | 3-month moving average (smooths noise) | Average of all months to date |

````python
w = monthly[['Month', 'Revenue']].copy()
w['Rolling_3_Mean'] = w['Revenue'].rolling(3).mean().round(1)
w['Rolling_3_Sum'] = w['Revenue'].rolling(3).sum()
w['Rolling_3_Mean_min1'] = w['Revenue'].rolling(3, min_periods=1).mean().round(1)   # allow partial windows at the start
w['Expanding_Mean'] = w['Revenue'].expanding().mean().round(1)
w['Expanding_Max'] = w['Revenue'].expanding().max()
w
````

### Common mistakes (Section 21)
- `pivot` on data with duplicate index/column pairs (use `pivot_table`).
- `melt` without `id_vars` (all columns get melted).
- `shift`/`diff`/`rolling` on unsorted data or across groups.
- Forgetting that `rolling(3)` gives NaN for the first 2 rows (`min_periods`).

# 22. Performance

| Topic | Advice |
|---|---|
| **Vectorization** | Operate on whole columns (`df['a'] * df['b']`, `np.where`, `.str`, `.dt`) — runs in fast compiled code |
| **Loops** | `for` loops / `iterrows()` run Python code per row → slow; avoid for data transformations |
| **`apply()`** | Convenient but still a Python loop; prefer vectorized alternatives, especially `apply(axis=1)` |
| **Memory usage** | Check with `df.memory_usage(deep=True)` / `df.info(memory_usage='deep')` |
| **Efficient dtypes** | Downcast numbers (`int64` → `int32`/`int16`, `float64` → `float32`) when the range allows |
| **`category` dtype** | For text columns with **few distinct values** (city, gender, status): stores each label once + small integer codes → much less memory, faster groupby |
| **Avoid unnecessary copies** | Don't `.copy()` at every step; select only needed columns early; with Copy-on-Write (pandas 3), copies are made lazily only when you modify data |
| **Read less** | `pd.read_csv(..., usecols=[...], dtype={...})` |

````python
n = 500_000
rng = np.random.default_rng(22)
big = pd.DataFrame({'city': rng.choice(['Lahore', 'Karachi', 'Islamabad', 'Multan'], n),
                    'qty': rng.integers(1, 50, n),
                    'price': rng.uniform(10, 5000, n)})

# 1) Loop vs vectorized
t0 = time.perf_counter()
loop_total = [q * p for q, p in zip(big['qty'].to_list()[:100_000], big['price'].to_list()[:100_000])]
t_loop = (time.perf_counter() - t0) * 5                                         # scaled to 500k rows
t0 = time.perf_counter(); vec_total = big['qty'] * big['price']; t_vec = time.perf_counter() - t0
print(f'Python loop (estimated for 500k rows): {t_loop:.3f} s | vectorized: {t_vec:.4f} s')

# 2) Memory: object/str vs category, int64 vs int8, float64 vs float32
before = big.memory_usage(deep=True).sum() / 1e6
opt = big.assign(city=big['city'].astype('category'), qty=big['qty'].astype('int8'), price=big['price'].astype('float32'))
after = opt.memory_usage(deep=True).sum() / 1e6
print(f'\nMemory before: {before:.1f} MB | after dtype optimisation: {after:.1f} MB')
print(opt.dtypes)

# 3) groupby speed on category vs text
t0 = time.perf_counter(); big.groupby('city')['price'].mean(); t_obj = time.perf_counter() - t0
t0 = time.perf_counter(); opt.groupby('city', observed=True)['price'].mean(); t_cat = time.perf_counter() - t0
print(f'\ngroupby on text: {t_obj:.4f} s | on category: {t_cat:.4f} s')
````

### When not
- `category` for columns with **many unique values** (IDs, free text) — no savings and slower.
- Downcasting beyond the value range (`int8` holds only −128…127 → overflow!).
- Premature optimisation on tiny data — readability first.

### Common mistakes
`iterrows()` for calculations · converting to `category` and then comparing with strings that aren't categories (works, but watch for unused categories in groupby — use `observed=True`) · chained operations that repeatedly copy big DataFrames.

**Scenario:** you receive a raw export of 260 orders from an online store. It has messy column names, inconsistent text, dates stored as text, missing values and duplicate rows. Your job: clean it and produce a business summary using the techniques from this notebook.

## Step 0 — Create the raw dataset

````python
rr = np.random.default_rng(2025)
n = 250
products = {'Laptop': ('Electronics', 145000), 'Smartphone': ('Electronics', 85000), 'Earbuds': ('Electronics', 6500),
            'T-Shirt': ('Fashion', 1800), 'Jeans': ('Fashion', 4200), 'Sneakers': ('Fashion', 9500),
            'Novel': ('Books', 1200), 'Cookbook': ('Books', 2500), 'Face Wash': ('Beauty', 950), 'Perfume': ('Beauty', 7500)}
first = ['Ali', 'Sara', 'Omar', 'Hina', 'Bilal', 'Zara', 'Usman', 'Ayesha', 'Hamza', 'Fatima', 'Imran', 'Mariam']
last = ['Khan', 'Ahmed', 'Farooq', 'Malik', 'Raza', 'Iqbal', 'Tariq', 'Noor', 'Butt', 'Sheikh']
all_names = [f'{f} {l}' for f in first for l in last]
names = list(rr.choice(all_names, 40, replace=False))                     # 40 distinct customers
cust_ids = [f'C{200 + i}' for i in range(len(names))]
prod_names = rr.choice(list(products), n, p=[0.06, 0.09, 0.12, 0.14, 0.12, 0.08, 0.11, 0.07, 0.12, 0.09])
cust_idx = rr.integers(0, len(names), n)
raw = pd.DataFrame({
    'Order ID': [f'ORD-{5000 + i}' for i in range(n)],
    ' customer_id': np.array(cust_ids)[cust_idx],
    'Customer Name': [rr.choice([nm, nm.upper(), '  ' + nm.lower() + ' ']) for nm in np.array(names)[cust_idx]],
    'AGE': rr.integers(18, 60, len(names))[cust_idx].astype(float),
    'gender': rr.choice(['M', 'F'], len(names))[cust_idx],
    'City ': rr.choice(['lahore', 'Lahore ', 'KARACHI', 'Karachi', 'Islamabad', 'islamabad', 'Multan'], n),
    'Product': prod_names,
    'Category': [products[p][0] for p in prod_names],
    'Price': [round(products[p][1] * rr.uniform(0.95, 1.05), -1) for p in prod_names],
    'Quantity': np.where(np.isin(prod_names, ['Laptop', 'Smartphone']), 1, rr.integers(1, 5, n)),
    'Discount': rr.choice([0, 0.05, 0.10, 0.15, 0.20], n, p=[0.35, 0.2, 0.25, 0.12, 0.08]),
    'Order Date': (pd.to_datetime('2024-07-01') + pd.to_timedelta(rr.integers(0, 365, n), unit='D')).strftime('%d-%m-%Y'),
    'Payment Method': rr.choice(['Card', 'cash on delivery', 'COD', 'Wallet', 'card'], n),
    'Rating': rr.choice([1, 2, 3, 4, 5], n, p=[0.04, 0.06, 0.2, 0.4, 0.3]).astype(float),
})
# Inject realistic problems
raw.loc[rr.choice(n, 12, replace=False), 'AGE'] = np.nan
raw.loc[rr.choice(n, 20, replace=False), 'Rating'] = np.nan
raw.loc[rr.choice(n, 6, replace=False), 'Payment Method'] = np.nan
raw.loc[rr.choice(n, 3, replace=False), 'Order Date'] = 'unknown'
raw = pd.concat([raw, raw.sample(10, random_state=1)], ignore_index=True)       # duplicate rows
print(raw.shape)
raw.head()
````

## Step 1 — Clean column names

````python
df = raw.copy()
df.columns = df.columns.str.strip().str.replace(' ', '_').str.title().str.replace('Id', 'ID')
print(df.columns.tolist())
````

**Why:** consistent `Title_Case` names without spaces are easy to type and avoid `KeyError`s caused by hidden spaces (`' customer_id'`, `'City '`).

## Step 2 — Convert dates

````python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d-%m-%Y', errors='coerce')
print('Unparseable dates:', df['Order_Date'].isna().sum(), '| range:', df['Order_Date'].min().date(), '→', df['Order_Date'].max().date())
````

`errors='coerce'` turns the three 'unknown' values into `NaT` instead of crashing; `format=` avoids day/month confusion.

## Step 3 — Handle missing values

````python
print('Missing before:\n', df.isna().sum()[df.isna().sum() > 0], '\n')
df['Age'] = df['Age'].fillna(df.groupby('Customer_ID')['Age'].transform('first'))       # same customer → same age
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Rating_Missing'] = df['Rating'].isna().astype(int)                                   # keep the information
df['Rating'] = df['Rating'].fillna(df.groupby('Product')['Rating'].transform('median'))  # typical rating of that product
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')
df = df.dropna(subset=['Order_Date'])                                                    # can't analyse orders without a date
print('Missing after:\n', df.isna().sum()[df.isna().sum() > 0].to_dict() or 'none')
````

**Decisions:** age is a customer attribute → borrow it from the same customer's other orders; ratings → product median + a flag; payment → an explicit 'Unknown' category; the 3 undated orders are dropped (tiny share).

## Step 4 — Remove duplicates

````python
print('Duplicate rows:', df.duplicated().sum(), '| duplicate Order_IDs:', df['Order_ID'].duplicated().sum())
df = df.drop_duplicates(subset='Order_ID', keep='first').reset_index(drop=True)
print('Rows now:', len(df))
````

`Order_ID` defines a unique order, so we deduplicate on it.

## Step 5–7 — Revenue, Discount_Amount, Net_Revenue

````python
df['Revenue'] = df['Price'] * df['Quantity']
df['Discount_Amount'] = (df['Revenue'] * df['Discount']).round(0)
df['Net_Revenue'] = df['Revenue'] - df['Discount_Amount']
df[['Order_ID', 'Product', 'Price', 'Quantity', 'Discount', 'Revenue', 'Discount_Amount', 'Net_Revenue']].head()
````

## Step 8–9 — Age_Group and Price_Category

````python
df['Age_Group'] = pd.cut(df['Age'], bins=[17, 25, 35, 45, 60], labels=['18–25', '26–35', '36–45', '46–60'])
df['Price_Category'] = pd.cut(df['Price'], bins=[0, 3000, 20000, np.inf], labels=['Budget', 'Mid-range', 'Premium'])
print(df['Age_Group'].value_counts().sort_index().to_dict())
print(df['Price_Category'].value_counts().to_dict())
````

## Step 15 (early) — String operations: clean names, cities, payment methods

````python
df['Customer_Name'] = df['Customer_Name'].str.strip().str.replace(r'\s+', ' ', regex=True).str.title()
df['City'] = df['City'].str.strip().str.title()
df['Payment_Method'] = (df['Payment_Method'].str.strip().str.lower()
                        .replace({'cod': 'Cash on Delivery', 'cash on delivery': 'Cash on Delivery', 'card': 'Card',
                                  'wallet': 'Wallet', 'unknown': 'Unknown'}))
df['Gender'] = df['Gender'].map({'M': 'Male', 'F': 'Female'})
df['First_Name'] = df['Customer_Name'].str.split().str[0]
df['Order_Number'] = df['Order_ID'].str.extract(r'(\d+)').astype(int)
print(df['City'].value_counts().to_dict())
print(df['Payment_Method'].value_counts().to_dict())
print('Distinct customer name spellings now:', df['Customer_Name'].nunique(), '(was', raw['Customer Name'].nunique(), ')')
````

Text cleaning is done **before** grouping — otherwise 'lahore', 'Lahore ' and 'Lahore' would be three different cities. (It's step 15 in the task list but must logically come before steps 10–14.)

## Step 10 — Group by City

````python
city_summary = df.groupby('City').agg(
    Orders=('Order_ID', 'count'), Customers=('Customer_ID', 'nunique'),
    Net_Revenue=('Net_Revenue', 'sum'), Avg_Order_Value=('Net_Revenue', 'mean'), Avg_Rating=('Rating', 'mean'),
).sort_values('Net_Revenue', ascending=False).round(1)
city_summary
````

## Step 11 — Group by Product

````python
product_summary = df.groupby('Product').agg(
    Category=('Category', 'first'), Units=('Quantity', 'sum'), Net_Revenue=('Net_Revenue', 'sum'),
    Avg_Discount_pct=('Discount', lambda s: s.mean() * 100), Avg_Rating=('Rating', 'mean'),
).sort_values('Net_Revenue', ascending=False).round(1)
product_summary
````

## Step 12 — Group by Category

````python
cat_summary = df.groupby('Category').agg(Orders=('Order_ID', 'count'), Units=('Quantity', 'sum'),
                                         Net_Revenue=('Net_Revenue', 'sum'), Total_Discount=('Discount_Amount', 'sum'))
cat_summary['Revenue_Share_%'] = (cat_summary['Net_Revenue'] / cat_summary['Net_Revenue'].sum() * 100).round(1)
cat_summary.sort_values('Net_Revenue', ascending=False)
````

## Step 13 — Pivot tables

````python
print('Net revenue (thousand PKR) by city × category:')
print((pd.pivot_table(df, index='City', columns='Category', values='Net_Revenue', aggfunc='sum', fill_value=0,
                      margins=True, margins_name='Total') / 1000).round(0), '\n')
print('Average rating by age group × category:')
print(pd.pivot_table(df, index='Age_Group', columns='Category', values='Rating', aggfunc='mean', observed=True).round(2))
````

## Step 14 — Crosstabs

````python
print('Payment method share by city (%):')
print((pd.crosstab(df['City'], df['Payment_Method'], normalize='index') * 100).round(0), '\n')
print('Orders by gender × price category:')
print(pd.crosstab(df['Gender'], df['Price_Category'], margins=True))
````

## Step 16 — Date analysis

````python
df = df.sort_values('Order_Date').reset_index(drop=True)
od = df['Order_Date'].dt
df['Month'] = od.to_period('M')
df['Weekday'] = od.day_name()
df['Is_Weekend'] = od.dayofweek >= 5

monthly = df.groupby('Month').agg(Orders=('Order_ID', 'count'), Net_Revenue=('Net_Revenue', 'sum'))
print(monthly.assign(Net_Revenue_k=(monthly['Net_Revenue'] / 1000).round(0)).drop(columns='Net_Revenue'), '\n')
print('Orders by weekday:'); print(df['Weekday'].value_counts().reindex(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']))
print('\nWeekend share of net revenue: {:.1%}'.format(df.loc[df['Is_Weekend'], 'Net_Revenue'].sum() / df['Net_Revenue'].sum()))
````

## Step 17–18 — Cumulative and rolling revenue

````python
daily_rev = df.groupby('Order_Date')['Net_Revenue'].sum().asfreq('D', fill_value=0)      # include days without orders
trend = pd.DataFrame({'Daily': daily_rev,
                      'Cumulative': daily_rev.cumsum(),
                      'Rolling_7d': daily_rev.rolling(7).sum(),
                      'Rolling_30d_Avg': daily_rev.rolling(30).mean().round(0)})
monthly['Cumulative_Net_Revenue'] = monthly['Net_Revenue'].cumsum()
monthly['MoM_Change_%'] = (monthly['Net_Revenue'].pct_change() * 100).round(1)
print(trend.tail(5), '\n')
monthly
````

## Step 19 — Rank products

````python
ranked = product_summary.copy()
ranked['Revenue_Rank'] = ranked['Net_Revenue'].rank(ascending=False, method='min').astype(int)
ranked['Units_Rank'] = ranked['Units'].rank(ascending=False, method='min').astype(int)
ranked['Rank_in_Category'] = ranked.groupby('Category')['Net_Revenue'].rank(ascending=False, method='min').astype(int)
ranked.sort_values('Revenue_Rank')[['Category', 'Net_Revenue', 'Revenue_Rank', 'Units', 'Units_Rank', 'Rank_in_Category']]
````

## Step 20 — Final business summary

````python
customer_summary = df.groupby(['Customer_ID', 'Customer_Name']).agg(
    Orders=('Order_ID', 'count'), Net_Revenue=('Net_Revenue', 'sum'), Avg_Rating=('Rating', 'mean')
).sort_values('Net_Revenue', ascending=False)

best_month = monthly['Net_Revenue'].idxmax()
summary = pd.Series({
    'Clean orders analysed': len(df),
    'Unique customers': df['Customer_ID'].nunique(),
    'Gross revenue (PKR)': f"{df['Revenue'].sum():,.0f}",
    'Discounts given (PKR)': f"{df['Discount_Amount'].sum():,.0f}",
    'Net revenue (PKR)': f"{df['Net_Revenue'].sum():,.0f}",
    'Average order value (PKR)': f"{df['Net_Revenue'].mean():,.0f}",
    'Top city': city_summary.index[0],
    'Top category (share)': f"{cat_summary['Net_Revenue'].idxmax()} ({cat_summary['Revenue_Share_%'].max()}%)",
    'Top product': product_summary.index[0],
    'Most units sold': product_summary['Units'].idxmax(),
    'Best month': str(best_month),
    'Top customer': customer_summary.index[0][1],
    'Average rating': round(df['Rating'].mean(), 2),
    'Most common payment': df['Payment_Method'].value_counts().idxmax(),
})
print(summary.to_string())
print('\nTop 5 customers:'); print(customer_summary.head().round(1))
````

### Project conclusions
- **Cleaning mattered:** hidden spaces, mixed case and duplicate spellings would otherwise have split cities, customers and payment methods into several fake groups; 10 duplicate orders and 3 undated orders were removed.
- **Revenue is concentrated** in the Electronics category (laptops and smartphones), while cheaper items (T-shirts, novels, face wash) drive the **number of units**.
- The **city**, **monthly trend**, **top customers** and **payment mix** tables above are the inputs for the marketing and inventory teams; read the exact numbers from your output.
- Every step used a vectorized pandas operation (no Python loops), so the same code scales to millions of rows.

# Pandas Data Manipulation Cheat Sheet

| Task | Code |
|---|---|
| Select column / columns | `df['c']` · `df[['a', 'b']]` |
| Select by label / position | `df.loc[rows, cols]` (end **included**) · `df.iloc[rows, cols]` (end **excluded**) |
| Filter | `df[(df.a > 1) & (df.b == 'x')]` · `\|` or · `~` not · `.isin([...])` · `.between(a, b)` · `df.query('a > 1 and b == "x"')` |
| New column | `df['c'] = df.a * df.b` · `np.where(cond, x, y)` · `np.select(conds, choices, default)` · `df.assign(c=lambda d: ...)` |
| Modify some rows | `df.loc[cond, 'c'] = value` (never chained assignment) |
| Sort | `df.sort_values(['a', 'b'], ascending=[True, False], na_position='first')` · `df.sort_index()` · `nlargest(n, 'c')` |
| Rename | `df.rename(columns={'old': 'new'})` · `df.columns.str.strip().str.lower()` |
| Drop | `df.drop(columns=[...])` · `df.drop(index=[...])` · `df.dropna(subset=[...], how='any'/'all', thresh=)` |
| Index | `df.set_index('c')` · `df.reset_index(drop=True)` |
| Functions | `s.map(dict/func)` · `s.apply(func)` · `df.apply(func, axis=0/1)` · `df.map(func)` (was `applymap`) — prefer vectorized ops |
| Group & summarise | `df.groupby('k').agg(new=('col', 'mean'), n=('col', 'count'))` |
| Group value on each row | `df.groupby('k')['c'].transform('mean')` |
| Keep whole groups | `df.groupby('k').filter(lambda g: len(g) >= 5)` |
| Pivot table | `pd.pivot_table(df, index=, columns=, values=, aggfunc=, fill_value=, margins=True)` |
| Frequency table | `pd.crosstab(df.a, df.b, normalize='index', margins=True)` |
| Strings | `.str.lower/upper/title/strip/contains/startswith/endswith/replace/split/len/extract` |
| Dates | `pd.to_datetime(s, format=, errors='coerce')` · `.dt.year/month/day/day_name()/hour/dayofweek` · `.dt.to_period('M')` · `resample('W')` |
| Duplicates | `df.duplicated(subset=, keep=)` · `df.drop_duplicates(subset=, keep='last')` |
| Missing values | `fillna(v)` · `ffill()` · `bfill()` · `interpolate()` · `groupby().ffill()` · `fillna(groupby().transform('median'))` |
| Recode | `replace({...})` (keeps unmatched) · `map({...})` (unmatched → NaN) |
| Bin | `pd.cut(s, bins=[...], labels=[...])` · `pd.qcut(s, q=4)` |
| Lists to rows | `df.assign(c=df.c.str.split(', ')).explode('c')` |
| Reshape | `melt(id_vars=)` · `pivot(index=, columns=, values=)` · `pivot_table` · `stack()` · `unstack()` · `pd.wide_to_long` |
| Conditional values | `s.where(cond, other)` · `s.mask(cond, other)` · `s.clip(lo, hi)` |
| Ordered calcs | `rank()` · `shift()` · `diff()` · `pct_change()` · `cumsum/cumprod/cummax/cummin()` |
| Windows | `rolling(7).mean()` · `rolling(7, min_periods=1)` · `expanding().mean()` |
| Performance | vectorize · avoid `iterrows`/`apply(axis=1)` · `astype('category')` · downcast · `memory_usage(deep=True)` |

### Golden rules
1. **Clean text before grouping** (strip, lower/title case).
2. **Convert dates** with an explicit `format=` and `errors='coerce'`.
3. Use **one `.loc[...] = ...`** for assignments (Copy-on-Write).
4. Prefer **`df = df.method(...)`** over `inplace=True`.
5. **agg** for summaries, **transform** for per-row group values, **filter** for whole groups.
6. **Sort** before `shift`/`diff`/`rolling`/`cumsum`, and use `groupby` for multiple entities.
7. Prefer **vectorized** operations over `apply` and loops.
