### Problem 2 — Linear regression by hand

x = `[1, 2, 3, 4, 5]`, y = `[2.1, 3.9, 6.2, 7.8, 10.1]`.
Compute the slope b₁ = Σ(x−x̄)(y−ȳ) / Σ(x−x̄)² and intercept b₀ = ȳ − b₁x̄ by hand (with NumPy), then verify with `LinearRegression`. Predict y for x = 6.

````python
# ✍️ Problem 2 — your attempt
````

### Solution 2

````python
x2 = np.array([1, 2, 3, 4, 5]); y2 = np.array([2.1, 3.9, 6.2, 7.8, 10.1])
b1 = np.sum((x2 - x2.mean()) * (y2 - y2.mean())) / np.sum((x2 - x2.mean()) ** 2)
b0 = y2.mean() - b1 * x2.mean()
lr2 = LinearRegression().fit(x2.reshape(-1, 1), y2)
print(f'By hand : b0 = {b0:.3f}, b1 = {b1:.3f}')
print(f'sklearn : b0 = {lr2.intercept_:.3f}, b1 = {lr2.coef_[0]:.3f}')
print(f'Prediction at x = 6: {b0 + b1 * 6:.3f}')
````

**Explanation:** the slope is the covariance of x and y divided by the variance of x (Topic 12/14); the line passes through (x̄, ȳ). Each +1 in x adds ≈ 2 to y.

### Problem 3 — Regression metrics by hand

Actual `[50, 60, 70, 80, 90]`, predicted `[55, 58, 75, 70, 95]`. Compute MAE, MSE, RMSE and R² by hand and with scikit-learn. Which metric is most affected by the single 10-point error?

````python
# ✍️ Problem 3 — your attempt
````

### Solution 3

````python
a3 = np.array([50, 60, 70, 80, 90]); p3 = np.array([55, 58, 75, 70, 95])
e3 = a3 - p3
print('errors:', e3)
print(f'MAE  by hand = {np.mean(np.abs(e3)):.2f} | sklearn = {mean_absolute_error(a3, p3):.2f}')
print(f'MSE  by hand = {np.mean(e3 ** 2):.2f} | sklearn = {mean_squared_error(a3, p3):.2f}')
print(f'RMSE by hand = {np.sqrt(np.mean(e3 ** 2)):.3f} | sklearn = {rmse(a3, p3):.3f}')
r2_hand = 1 - np.sum(e3 ** 2) / np.sum((a3 - a3.mean()) ** 2)
print(f'R²   by hand = {r2_hand:.4f} | sklearn = {r2_score(a3, p3):.4f}')
````

**Explanation:** errors are 5, 2, 5, 10, 5 in size. The 10-point error contributes 100 of the 179 total squared error → **MSE/RMSE** are most affected; MAE treats it as just 10/27 of the total.

### Problem 5 — Polynomial degree selection

Data: `rng5 = np.random.default_rng(5); x = rng5.uniform(-2, 2, 120); y = np.sin(2 * x) + rng5.normal(0, 0.25, 120)`.
Use 5-fold CV RMSE to choose the best polynomial degree from 1 to 12. Plot CV RMSE vs degree.

````python
# ✍️ Problem 5 — your attempt
````

### Solution 5

````python
rng5 = np.random.default_rng(5)
x5 = rng5.uniform(-2, 2, 120); y5 = np.sin(2 * x5) + rng5.normal(0, 0.25, 120)
degrees = range(1, 13)
cv_rmse5 = [-cross_val_score(make_pipeline(PolynomialFeatures(d, include_bias=False), StandardScaler(), LinearRegression()),
                             x5.reshape(-1, 1), y5, cv=KFold(5, shuffle=True, random_state=0),
                             scoring='neg_root_mean_squared_error').mean() for d in degrees]
best5 = list(degrees)[int(np.argmin(cv_rmse5))]
print('Best degree by CV:', best5, '| CV RMSE:', round(min(cv_rmse5), 4))
plt.figure(figsize=(7, 3.5)); plt.plot(list(degrees), cv_rmse5, 'o-'); plt.axvline(best5, color='red', ls='--')
plt.xlabel('degree'); plt.ylabel('CV RMSE'); plt.title('Polynomial degree selection'); plt.show()
````

**Explanation:** low degrees underfit the sine wave (high CV RMSE). The error drops sharply up to degree ~5 and then flattens; CV picks degree 7 here. When several degrees are within the CV noise of each other, prefer the **smallest** one (simpler is safer), and very high degrees start to get worse again.

### Problem 8 — Classification metrics by hand

A fraud model's confusion matrix: TP = 40, FP = 10, FN = 20, TN = 930.
Compute accuracy, precision, recall, specificity and F1. What would the accuracy of a model that predicts "never fraud" be?

````python
# ✍️ Problem 8 — your attempt
````

### Solution 8

````python
TP, FP, FN, TN = 40, 10, 20, 930
acc = (TP + TN) / (TP + FP + FN + TN); prec = TP / (TP + FP); rec = TP / (TP + FN); spec = TN / (TN + FP)
f1 = 2 * prec * rec / (prec + rec)
print(f'Accuracy = {acc:.3f} | Precision = {prec:.3f} | Recall = {rec:.3f} | Specificity = {spec:.3f} | F1 = {f1:.3f}')
print(f'"Never fraud" accuracy = {(FP + TN) / (TP + FP + FN + TN):.3f}  (catches 0 of {TP + FN} frauds)')
````

**Explanation:** the model catches 2/3 of frauds (recall 0.667) and 80% of its alerts are real (precision 0.8). A useless always-negative model has 94% accuracy — accuracy hides everything that matters here.

### Problem 12 — Decision tree impurity

(a) A node has 30 samples: 20 class A, 10 class B. A split creates left = (18 A, 2 B) and right = (2 A, 8 B). Compute the Gini of parent and children and the information gain.
(b) Fit a depth-2 tree on iris and print its rules with `export_text`.

````python
# ✍️ Problem 12 — your attempt
````

### Solution 12

````python
g_parent, g_left, g_right = gini([20, 10]), gini([18, 2]), gini([2, 8])
gain = g_parent - (20 / 30 * g_left + 10 / 30 * g_right)
print(f'Gini parent = {g_parent:.4f} | left = {g_left:.4f} | right = {g_right:.4f} | information gain = {gain:.4f}')
tree12 = DecisionTreeClassifier(max_depth=2, random_state=0).fit(iris.data, iris.target)
print(export_text(tree12, feature_names=list(iris.data.columns)))
````

**Explanation:** parent Gini = 1 − (2/3)² − (1/3)² = 0.444; left = 0.18, right = 0.32; weighted = 0.227 → gain ≈ 0.218. The tree splits on **petal width** first (petal width ≤ 0.8 isolates setosa perfectly — petal length would work equally well, and the tree picks one of the tied options), then on petal width again to separate versicolor from virginica.

### Problem 13 — Random forest importance

On wine (70/30, stratified, `random_state=0`), fit a random forest. Compare the top-5 features by impurity importance and by permutation importance (on the test set). Are they the same?

````python
# ✍️ Problem 13 — your attempt
````

### Solution 13

### Solution 21

````python
r21 = np.random.default_rng(21)
n = 600
d21 = pd.DataFrame({'Age': r21.integers(18, 70, n).astype(float), 'Balance': r21.lognormal(9, 1, n),
                    'Products': r21.integers(1, 5, n), 'Region': r21.choice(['North', 'South', 'East', 'West'], n),
                    'Card': r21.choice(['Gold', 'Silver', 'Basic'], n)})
d21['Exited'] = (r21.random(n) < 1 / (1 + np.exp(-(-3 + 0.05 * d21['Age'] - 0.4 * d21['Products'] + (d21['Region'] == 'East'))))).astype(int)
d21.loc[r21.choice(n, 50, replace=False), 'Balance'] = np.nan
d21.loc[r21.choice(n, 30, replace=False), 'Card'] = np.nan

num21, cat21 = ['Age', 'Balance', 'Products'], ['Region', 'Card']
pipe21 = Pipeline([
    ('prep', ColumnTransformer([
        ('num', Pipeline([('imp', SimpleImputer(strategy='median')), ('sc', StandardScaler())]), num21),
        ('cat', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('oh', OneHotEncoder(handle_unknown='ignore'))]), cat21)])),
    ('model', RandomForestClassifier(n_estimators=300, min_samples_leaf=5, random_state=0, n_jobs=-1))])
X21, y21 = d21[num21 + cat21], d21['Exited']
print('5-fold CV ROC-AUC:', round(cross_val_score(pipe21, X21, y21, cv=StratifiedKFold(5, shuffle=True, random_state=0), scoring='roc_auc').mean(), 3))
pipe21.fit(X21, y21)
new21 = pd.DataFrame([{'Age': 58, 'Balance': np.nan, 'Products': 1, 'Region': 'Central', 'Card': 'Platinum'}])
print('P(exit) for a new customer (missing balance, unseen region & card):', round(pipe21.predict_proba(new21)[0, 1], 3))
````

**Explanation:** imputation, scaling and encoding are learned inside each CV fold; unseen categories become all-zero one-hot columns thanks to `handle_unknown='ignore'`, so the pipeline never crashes on new data.

# 38. PROJECT 1 — House Price Prediction (Regression)

**Business problem:** a property website wants to suggest a fair asking price for new listings. Errors in rupees matter to users → we'll report **MAE** (typical error), **RMSE** (punishes big misses) and **R²**.

## Step 0 — Create a realistic dataset (400 houses)

````python
rh = np.random.default_rng(38)
n = 400
location = rh.choice(['Downtown', 'Suburb', 'Outskirts'], n, p=[0.25, 0.5, 0.25])
distance = np.where(location == 'Downtown', rh.uniform(0.5, 5, n),
                    np.where(location == 'Suburb', rh.uniform(4, 15, n), rh.uniform(12, 35, n))).round(1)
area = rh.lognormal(7.3, 0.35, n).round(0)                                    # right-skewed sq ft
bedrooms = np.clip(np.round(area / 450 + rh.normal(0, 0.7, n)), 1, 7).astype(int)
bathrooms = np.clip(np.round(bedrooms * 0.7 + rh.normal(0, 0.5, n)), 1, 5).astype(int)
age = rh.integers(0, 50, n)
parking = rh.choice([0, 1, 2], n, p=[0.25, 0.5, 0.25])
condition = rh.choice(['Poor', 'Fair', 'Good', 'Excellent'], n, p=[0.1, 0.3, 0.45, 0.15])

loc_mult = pd.Series(location).map({'Downtown': 1.6, 'Suburb': 1.0, 'Outskirts': 0.7}).to_numpy()
cond_mult = pd.Series(condition).map({'Poor': 0.8, 'Fair': 0.92, 'Good': 1.0, 'Excellent': 1.15}).to_numpy()
price = (9000 * area ** 0.95 * loc_mult * cond_mult * (1 - 0.004 * age) * (1 + 0.04 * parking)
         * np.exp(-0.01 * distance) * rh.lognormal(0, 0.12, n)).round(-3)          # multiplicative → skewed prices

houses = pd.DataFrame({'Area': area, 'Bedrooms': bedrooms, 'Bathrooms': bathrooms, 'Location': location, 'Age': age,
                       'Parking': parking, 'DistanceToCenter': distance, 'Condition': condition, 'Price': price})
# Realistic data problems
houses.loc[rh.choice(n, 15, replace=False), 'Age'] = np.nan
houses.loc[rh.choice(n, 10, replace=False), 'Condition'] = np.nan
houses.loc[[5, 17], 'Area'] = [0, 25]                               # data-entry errors (impossible areas)
houses = pd.concat([houses, houses.iloc[[3, 8, 12]]], ignore_index=True)   # duplicate listings
houses.head()
````

(Missing values are **not** filled here — imputation is a learned step and belongs **inside the pipeline**, fitted on training data only.)

## Step 3 — Feature engineering

````python
h['Total_Rooms'] = h['Bedrooms'] + h['Bathrooms']
h['Area_per_Room'] = h['Area'] / h['Total_Rooms']
h['Is_New'] = (h['Age'] <= 5).astype(float).where(h['Age'].notna())          # NaN stays NaN → imputed later
condition_order = [['Poor', 'Fair', 'Good', 'Excellent']]                     # ordinal category (Topic 6)
h[['Area', 'Total_Rooms', 'Area_per_Room', 'Is_New']].describe().round(2)
````

## Step 4 — Skewness analysis & Step 5 — Transformation

````python
num_feats = ['Area', 'Bedrooms', 'Bathrooms', 'Age', 'Parking', 'DistanceToCenter', 'Total_Rooms', 'Area_per_Room']
skew_tbl = pd.DataFrame({'skew': h[num_feats + ['Price']].skew(),
                         'skew after log1p': np.log1p(h[num_feats + ['Price']]).skew()}).round(2)
print(skew_tbl)

fig, axes = plt.subplots(1, 2, figsize=(12, 3.5))
sns.histplot(h['Price'], bins=40, ax=axes[0]); axes[0].set_title('Price')
sns.histplot(np.log(h['Price']), bins=40, ax=axes[1]); axes[1].set_title('log(Price)')
plt.tight_layout(); plt.show()
````

**Decisions:**
- `Price` is right-skewed and generated **multiplicatively** (location, condition… multiply the price) → model **log(Price)** with `TransformedTargetRegressor` (predictions are converted back to rupees automatically).
- `Area` and `DistanceToCenter` are skewed → log-transform them for the **linear** models (trees don't need it).

## Step 6 — Train/test split & Step 7 — Preprocessing pipelines

````python
from sklearn.compose import TransformedTargetRegressor
from sklearn.preprocessing import OrdinalEncoder, FunctionTransformer

features_h = num_feats + ['Is_New', 'Location', 'Condition']
X_h, y_h = h[features_h], h['Price']
Xh_train, Xh_test, yh_train, yh_test = train_test_split(X_h, y_h, test_size=0.2, random_state=RANDOM_STATE)
print('Train:', Xh_train.shape, '| Test:', Xh_test.shape)

log_cols = ['Area', 'DistanceToCenter', 'Area_per_Room']
other_num = [c for c in num_feats + ['Is_New'] if c not in log_cols]

linear_prep = ColumnTransformer([
    ('log_num', Pipeline([('imp', SimpleImputer(strategy='median')), ('log', FunctionTransformer(np.log1p)), ('sc', StandardScaler())]), log_cols),
    ('num', Pipeline([('imp', SimpleImputer(strategy='median')), ('sc', StandardScaler())]), other_num),
    ('loc', OneHotEncoder(drop='first', handle_unknown='ignore'), ['Location']),
    ('cond', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('ord', OrdinalEncoder(categories=condition_order))]), ['Condition']),
])
tree_prep = ColumnTransformer([
    ('num', SimpleImputer(strategy='median'), num_feats + ['Is_New']),
    ('loc', OneHotEncoder(handle_unknown='ignore'), ['Location']),
    ('cond', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('ord', OrdinalEncoder(categories=condition_order))]), ['Condition']),
])

def log_target(regressor):
    '''Train on log(price); predictions are automatically converted back with exp.'''
    return TransformedTargetRegressor(regressor=regressor, func=np.log, inverse_func=np.exp)

house_models = {
    'Linear Regression': log_target(Pipeline([('prep', linear_prep), ('model', LinearRegression())])),
    'Ridge': log_target(Pipeline([('prep', linear_prep), ('model', Ridge(alpha=1.0))])),
    'Random Forest': log_target(Pipeline([('prep', tree_prep), ('model', RandomForestRegressor(n_estimators=300, min_samples_leaf=2, random_state=RANDOM_STATE, n_jobs=-1))])),
    'Gradient Boosting': log_target(Pipeline([('prep', tree_prep), ('model', GradientBoostingRegressor(random_state=RANDOM_STATE))])),
}
````

## Steps 8–12 — Train Linear Regression, Ridge, Random Forest, Gradient Boosting with cross-validation

````python
cv_h = KFold(5, shuffle=True, random_state=RANDOM_STATE)
baseline_mae = mean_absolute_error(yh_train, np.full(len(yh_train), yh_train.median()))
print(f'Baseline (predict the median price) training MAE: {baseline_mae:,.0f}\n')

rows = []
for name, m in house_models.items():
    r = cross_validate(m, Xh_train, yh_train, cv=cv_h, scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error', 'r2'])
    rows.append({'Model': name, 'CV MAE': -r['test_neg_mean_absolute_error'].mean(),
                 'CV RMSE': -r['test_neg_root_mean_squared_error'].mean(), 'CV R²': r['test_r2'].mean(), 'R² std': r['test_r2'].std()})
cv_house = pd.DataFrame(rows).set_index('Model').sort_values('CV MAE').round(3)
cv_house
````

## Step 13 — Hyperparameter tuning (Ridge and Gradient Boosting)

````python
ridge_search = GridSearchCV(house_models['Ridge'], {'regressor__model__alpha': np.logspace(-2, 2, 9)},
                            cv=cv_h, scoring='neg_mean_absolute_error', n_jobs=-1).fit(Xh_train, yh_train)
gb_search = RandomizedSearchCV(house_models['Gradient Boosting'],
                               {'regressor__model__n_estimators': randint(100, 500), 'regressor__model__learning_rate': uniform(0.02, 0.15),
                                'regressor__model__max_depth': [2, 3, 4], 'regressor__model__subsample': [0.7, 0.85, 1.0]},
                               n_iter=12, cv=cv_h, scoring='neg_mean_absolute_error', random_state=RANDOM_STATE, n_jobs=-1).fit(Xh_train, yh_train)
print(f'Ridge best alpha: {ridge_search.best_params_} | CV MAE {-ridge_search.best_score_:,.0f}')
print(f'GB best params  : { {k.split("__")[-1]: (round(v, 3) if isinstance(v, float) else v) for k, v in gb_search.best_params_.items()} } | CV MAE {-gb_search.best_score_:,.0f}')
````

## Step 14 — Compare MAE / RMSE / R² on the test set (once)

````python
final_house = {
    'Linear Regression': house_models['Linear Regression'].fit(Xh_train, yh_train),
    'Ridge (tuned)': ridge_search.best_estimator_,
    'Random Forest': house_models['Random Forest'].fit(Xh_train, yh_train),
    'Gradient Boosting (tuned)': gb_search.best_estimator_,
}
rows = []
for name, m in final_house.items():
    p = m.predict(Xh_test)
    rows.append({'Model': name, 'Test MAE': mean_absolute_error(yh_test, p), 'Test RMSE': rmse(yh_test, p),
                 'Test R²': r2_score(yh_test, p), 'MAPE %': np.mean(np.abs((yh_test - p) / yh_test)) * 100})
test_house = pd.DataFrame(rows).set_index('Model').sort_values('Test MAE').round(3)
test_house
````

## Step 15 — Error analysis

````python
best_house_name = test_house.index[0]
best_house = final_house[best_house_name]
eh = Xh_test.copy(); eh['Actual'] = yh_test; eh['Pred'] = best_house.predict(Xh_test)
eh['Error'] = eh['Actual'] - eh['Pred']; eh['AbsPctError'] = (eh['Error'].abs() / eh['Actual'] * 100)
eh['Price_Band'] = pd.qcut(eh['Actual'], 4, labels=['cheapest 25%', 'Q2', 'Q3', 'most expensive 25%'])
print(f'Best model: {best_house_name}\n')
print('Mean absolute % error by location:'); print(eh.groupby('Location')['AbsPctError'].mean().round(1))
print('\nMean absolute % error by price band:'); print(eh.groupby('Price_Band', observed=True)['AbsPctError'].mean().round(1))

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].scatter(eh['Actual'], eh['Pred'], s=15); lim = [eh['Actual'].min(), eh['Actual'].max()]
axes[0].plot(lim, lim, 'r--'); axes[0].set_title(f'{best_house_name}: actual vs predicted'); axes[0].set_xlabel('actual'); axes[0].set_ylabel('predicted')
axes[1].scatter(eh['Pred'], eh['AbsPctError'], s=15); axes[1].set_title('Absolute % error vs predicted price'); axes[1].set_xlabel('predicted')
plt.tight_layout(); plt.show()
print('\nLargest misses:'); print(eh.sort_values('AbsPctError', ascending=False).head(3)[['Area', 'Location', 'Condition', 'Age', 'Actual', 'Pred']].round(0))
````

## Step 16 — Save the final model

````python
os.makedirs('ml_models', exist_ok=True)
joblib.dump(best_house, 'ml_models/house_price_model.joblib')
json.dump({'model': best_house_name, 'features': features_h, 'target': 'Price (model trained on log price)',
           'test_MAE': float(test_house.loc[best_house_name, 'Test MAE']), 'sklearn_version': sklearn.__version__},
          open('ml_models/house_price_metadata.json', 'w'), indent=2)

reloaded_house = joblib.load('ml_models/house_price_model.joblib')
new_listing = pd.DataFrame([{'Area': 1800, 'Bedrooms': 3, 'Bathrooms': 2, 'Age': 8, 'Parking': 1, 'DistanceToCenter': 6.5,
                             'Total_Rooms': 5, 'Area_per_Room': 360, 'Is_New': 0, 'Location': 'Suburb', 'Condition': 'Good'}])
print(f'Suggested price for the new listing: {reloaded_house.predict(new_listing[features_h])[0]:,.0f}')
````

### Project 1 — conclusions
- All four models beat the median baseline by a wide margin.
- Modelling **log(price)** suits the multiplicative way prices are formed and keeps percentage errors similar across cheap and expensive houses.
- With log-transformed skewed features and a log target, the **linear models are very competitive** with the tree ensembles here, because the (log) relationships are close to linear. The best model is chosen by test MAE, but differences smaller than the CV spread should not be over-interpreted; the simpler linear/Ridge model is easier to explain to users.
- Error analysis shows which locations/price bands have the largest percentage errors → the place to collect more data or engineer features.

# 39. PROJECT 2 — Customer Churn (Classification)

**Business problem:** a subscription company wants to identify customers likely to cancel, so the retention team can contact them. Missing a churner (FN) costs more than an unnecessary call (FP) → **recall** and **PR-AUC** matter more than accuracy.

## Dataset (1,000 customers)

````python
rc = np.random.default_rng(39)
n = 1000
churn_df = pd.DataFrame({
    'Age': rc.integers(18, 75, n),
    'Tenure': rc.integers(1, 72, n),
    'Contract': rc.choice(['Month-to-month', 'One year', 'Two year'], n, p=[0.55, 0.25, 0.2]),
    'MonthlyCharges': rc.uniform(20, 120, n).round(2),
    'SupportCalls': rc.poisson(1.5, n),
    'LoginFrequency': rc.gamma(2, 6, n).round(1),                        # logins per month
    'PaymentMethod': rc.choice(['Card', 'Bank transfer', 'E-wallet', 'Cash'], n, p=[0.4, 0.25, 0.2, 0.15]),
    'SatisfactionScore': rc.integers(1, 6, n),
})
logit_c = (-2.4 + 1.3 * (churn_df['Contract'] == 'Month-to-month') - 0.8 * (churn_df['Contract'] == 'Two year')
           - 0.03 * churn_df['Tenure'] + 0.012 * churn_df['MonthlyCharges'] + 0.35 * churn_df['SupportCalls']
           - 0.05 * churn_df['LoginFrequency'] - 0.45 * (churn_df['SatisfactionScore'] - 3)
           + 0.5 * (churn_df['PaymentMethod'] == 'Cash'))
churn_df['Churn'] = (rc.random(n) < 1 / (1 + np.exp(-logit_c))).astype(int)
churn_df.loc[rc.choice(n, 40, replace=False), 'LoginFrequency'] = np.nan
print('Churn rate:', churn_df['Churn'].mean().round(3), '→ imbalanced')
churn_df.head()
````

````python
fig, axes = plt.subplots(1, 4, figsize=(21, 3.8))
churn_df.groupby('Contract')['Churn'].mean().plot(kind='bar', ax=axes[0], title='Churn rate by contract', rot=0)
churn_df.groupby('SatisfactionScore')['Churn'].mean().plot(kind='bar', ax=axes[1], title='Churn rate by satisfaction', rot=0)
churn_df.groupby(churn_df['SupportCalls'].clip(upper=5))['Churn'].mean().plot(kind='bar', ax=axes[2], title='Churn rate by support calls (5 = 5+)', rot=0)
sns.boxplot(data=churn_df, x='Churn', y='Tenure', ax=axes[3]); axes[3].set_title('Tenure by churn')
plt.tight_layout(); plt.show()
````

````python
num_c = ['Age', 'Tenure', 'MonthlyCharges', 'SupportCalls', 'LoginFrequency', 'SatisfactionScore']
cat_c = ['Contract', 'PaymentMethod']
Xch, ych = churn_df[num_c + cat_c], churn_df['Churn']
Xch_tr, Xch_te, ych_tr, ych_te = train_test_split(Xch, ych, test_size=0.25, stratify=ych, random_state=RANDOM_STATE)

prep_c = ColumnTransformer([
    ('num', Pipeline([('imp', SimpleImputer(strategy='median')), ('sc', StandardScaler())]), num_c),
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_c)])

churn_models = {
    'Logistic Regression': LogisticRegression(max_iter=2000),
    'KNN (k=25)': KNeighborsClassifier(25),
    'Decision Tree (depth 4)': DecisionTreeClassifier(max_depth=4, random_state=RANDOM_STATE),
    'Random Forest': RandomForestClassifier(n_estimators=300, min_samples_leaf=3, random_state=RANDOM_STATE, n_jobs=-1),
    'Gradient Boosting': GradientBoostingClassifier(random_state=RANDOM_STATE),
}
cv_c = StratifiedKFold(5, shuffle=True, random_state=RANDOM_STATE)
rows, fitted_churn = [], {}
for name, model in churn_models.items():
    pipe = Pipeline([('prep', prep_c), ('model', model)])
    cv_pr = cross_val_score(pipe, Xch_tr, ych_tr, cv=cv_c, scoring='average_precision').mean()
    pipe.fit(Xch_tr, ych_tr); fitted_churn[name] = pipe
    pred, score = pipe.predict(Xch_te), pipe.predict_proba(Xch_te)[:, 1]
    rows.append({'Model': name, 'CV PR-AUC': cv_pr, 'Accuracy': accuracy_score(ych_te, pred), 'Precision': precision_score(ych_te, pred, zero_division=0),
                 'Recall': recall_score(ych_te, pred), 'F1': f1_score(ych_te, pred), 'ROC-AUC': roc_auc_score(ych_te, score),
                 'PR-AUC': average_precision_score(ych_te, score)})
churn_results = pd.DataFrame(rows).set_index('Model').sort_values('PR-AUC', ascending=False).round(3)
churn_results
````

````python
fig, axes = plt.subplots(1, 5, figsize=(24, 4))
for ax, (name, pipe) in zip(axes, fitted_churn.items()):
    ConfusionMatrixDisplay.from_estimator(pipe, Xch_te, ych_te, ax=ax, colorbar=False); ax.set_title(name, fontsize=10)
plt.tight_layout(); plt.show()

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
for name, pipe in fitted_churn.items():
    RocCurveDisplay.from_estimator(pipe, Xch_te, ych_te, ax=axes[0], name=name)
    PrecisionRecallDisplay.from_estimator(pipe, Xch_te, ych_te, ax=axes[1], name=name)
axes[0].plot([0, 1], [0, 1], 'k--'); axes[1].axhline(ych_te.mean(), color='k', ls='--')
axes[0].set_title('ROC curves'); axes[1].set_title('Precision–Recall curves'); plt.tight_layout(); plt.show()
````

### Handling the class imbalance
At threshold 0.5 most models miss many churners (low recall). Two practical fixes: **class weights** and a **lower threshold chosen on training CV predictions**.

````python
from sklearn.model_selection import cross_val_predict
best_churn_name = churn_results.index[0]
lr_bal = Pipeline([('prep', prep_c), ('model', LogisticRegression(max_iter=2000, class_weight='balanced'))]).fit(Xch_tr, ych_tr)

oof_c = cross_val_predict(fitted_churn[best_churn_name], Xch_tr, ych_tr, cv=cv_c, method='predict_proba')[:, 1]
ths = np.linspace(0.1, 0.6, 51)
f1s = [f1_score(ych_tr, (oof_c >= t).astype(int)) for t in ths]
t_best = ths[int(np.argmax(f1s))]

rows = []
for name, pred in [(f'{best_churn_name} @ 0.50', fitted_churn[best_churn_name].predict(Xch_te)),
                   (f'{best_churn_name} @ {t_best:.2f} (tuned on CV)', (fitted_churn[best_churn_name].predict_proba(Xch_te)[:, 1] >= t_best).astype(int)),
                   ('Logistic Regression, class_weight=balanced', lr_bal.predict(Xch_te))]:
    rows.append({'Approach': name, 'Precision': precision_score(ych_te, pred), 'Recall': recall_score(ych_te, pred), 'F1': f1_score(ych_te, pred)})
pd.DataFrame(rows).set_index('Approach').round(3)
````

### Project 2 — conclusions
- Accuracy looks decent for every model mainly because most customers stay; **PR-AUC and recall** separate the models properly.
- Month-to-month contracts, many support calls, low satisfaction and short tenure are the main churn signals (see the EDA and the logistic coefficients).
- **Logistic regression is competitive** with the ensembles here because the true relationship is close to linear in log-odds; it's also the easiest to explain to the retention team.
- Lowering the threshold (chosen on CV predictions, not the test set) or using class weights catches far more churners at an acceptable precision — pick the operating point from the cost of a retention call vs. the value of a saved customer (expected value, Topic 15).

# 40. PROJECT 3 — Customer Segmentation (Unsupervised)

**Business problem:** marketing wants a few customer **segments** to tailor campaigns. No labels exist → clustering.

## Dataset (600 customers)

````python
rs3 = np.random.default_rng(40)
segments_true = rs3.choice(4, 600, p=[0.3, 0.25, 0.25, 0.2])        # hidden groups (unknown to us in reality)
profiles = {0: (25, 35_000, 800, 2, 25),     # young, low income, frequent browsers
            1: (45, 120_000, 6000, 6, 12),  # affluent, big spenders
            2: (38, 70_000, 2500, 10, 30),  # loyal frequent buyers
            3: (60, 55_000, 600, 1, 4)}     # older, occasional
cust_seg = pd.DataFrame([{'Age': rs3.normal(p[0], 5), 'Income': rs3.normal(p[1], p[1] * 0.15), 'Spending': rs3.normal(p[2], p[2] * 0.25),
                          'PurchaseFrequency': max(0, rs3.normal(p[3], 1.2)), 'WebsiteVisits': max(0, rs3.normal(p[4], 5))}
                         for p in (profiles[s] for s in segments_true)]).round(1)
cust_seg.describe().round(1)
````

````python
sns.pairplot(cust_seg.sample(300, random_state=0), corner=True, plot_kws={'s': 10}, height=1.8); plt.show()
````

## Scaling → K-Means → elbow & silhouette

````python
scaler_seg = StandardScaler()
Xs_seg = scaler_seg.fit_transform(cust_seg)          # ⚠️ essential: Income (tens of thousands) would dominate otherwise

rows = []
for k_ in range(2, 10):
    km_ = KMeans(n_clusters=k_, n_init=10, random_state=RANDOM_STATE).fit(Xs_seg)
    rows.append({'K': k_, 'inertia': km_.inertia_, 'silhouette': silhouette_score(Xs_seg, km_.labels_)})
k_tbl = pd.DataFrame(rows)
fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
axes[0].plot(k_tbl['K'], k_tbl['inertia'], 'o-'); axes[0].set_title('Elbow method'); axes[0].set_xlabel('K')
axes[1].plot(k_tbl['K'], k_tbl['silhouette'], 's-', color='tab:green'); axes[1].set_title('Silhouette score'); axes[1].set_xlabel('K')
plt.tight_layout(); plt.show()
print(k_tbl.round(3).to_string(index=False))
K_SEG = int(k_tbl.loc[k_tbl['silhouette'].idxmax(), 'K'])
print('Chosen K:', K_SEG)
````

````python
km_seg = KMeans(n_clusters=K_SEG, n_init=10, random_state=RANDOM_STATE).fit(Xs_seg)
cust_seg['Segment'] = km_seg.labels_

# PCA for visualisation only
pca_seg = PCA(n_components=2).fit(Xs_seg)
Z = pca_seg.transform(Xs_seg)
centers_2d = pca_seg.transform(km_seg.cluster_centers_)
plt.figure(figsize=(8, 5.5))
plt.scatter(Z[:, 0], Z[:, 1], c=cust_seg['Segment'], cmap='tab10', s=12)
plt.scatter(centers_2d[:, 0], centers_2d[:, 1], c='black', marker='X', s=200)
for i, (x_, y_) in enumerate(centers_2d):
    plt.annotate(f'Segment {i}', (x_, y_), xytext=(8, 8), textcoords='offset points', fontweight='bold')
plt.xlabel(f'PC1 ({pca_seg.explained_variance_ratio_[0]:.0%} of variance)'); plt.ylabel(f'PC2 ({pca_seg.explained_variance_ratio_[1]:.0%})')
plt.title('Customer segments (K-Means) shown in 2 principal components'); plt.show()
````

## Interpret the clusters

````python
profile = cust_seg.groupby('Segment').agg(Customers=('Age', 'size'), Age=('Age', 'mean'), Income=('Income', 'mean'),
                                           Spending=('Spending', 'mean'), PurchaseFrequency=('PurchaseFrequency', 'mean'),
                                           WebsiteVisits=('WebsiteVisits', 'mean')).round(0)

def name_segment(r):
    if r['Spending'] > 4000: return 'Affluent big spenders'
    if r['PurchaseFrequency'] >= 7: return 'Loyal frequent buyers'
    if r['Age'] >= 52: return 'Older occasional shoppers'
    return 'Young browsers'
profile['Name'] = profile.apply(name_segment, axis=1)
profile
````

````python
z_profile = pd.DataFrame(km_seg.cluster_centers_, columns=cust_seg.columns[:5]).round(2)
plt.figure(figsize=(8, 3.2)); sns.heatmap(z_profile, annot=True, cmap='coolwarm', center=0)
plt.title('Cluster centres in standardized units (red = above average, blue = below)'); plt.ylabel('Segment'); plt.show()
from sklearn.metrics import adjusted_rand_score      # 1 = identical grouping, 0 = random
print('Agreement with the hidden true groups (adjusted Rand index):', round(adjusted_rand_score(segments_true, km_seg.labels_), 3))
````

|---|---|---|
| **Affluent big spenders** | Middle-aged, high income, very high spending | Premium products, VIP programme |
| **Loyal frequent buyers** | Many purchases, many visits | Loyalty points, subscriptions, referrals |
| **Young browsers** | Young, lower income, many visits but low spend | Discounts, first-purchase offers, mobile campaigns |

# 41. FINAL ML CAPSTONE — Predicting Next-Month Purchases

This capstone walks through **all 24 steps** of a professional project. Use it as a template for your own projects.

## Step 1 — Define the business problem

**Context:** an online store sends a monthly promotion. Sending it to everyone is expensive and annoys customers who wouldn't buy anyway.

**Goal:** predict which customers are likely to **make a purchase in the next 30 days**, and target the campaign at them.

| Item | Decision |
|---|---|
| ML task | **Binary classification** (`Purchased_Next_30d` = 1/0) |
| Unit of prediction | One customer, at the start of the month |
| Available information | Only data known **before** the month starts (no leakage) |
| Primary metric | **PR-AUC** (the positive class is the minority; we care about ranking buyers first) |
| Secondary metrics | ROC-AUC, recall & precision at the chosen threshold, **expected campaign profit** |
| Business rule for the threshold | Contact a customer if expected profit > 0: P(buy) × 150 rupees margin > 45 rupees cost (discount voucher + message) → contact if **P(buy) > 0.30** |
| Success criterion | Beat the "contact everyone" and "contact nobody" strategies on expected profit, with PR-AUC clearly above the base rate |

## Step 2 — Understand the dataset

````python
rcap = np.random.default_rng(41)
n = 1500
cap = pd.DataFrame({
    'CustomerID': [f'U{100000 + i}' for i in range(n)],
    'Age': rcap.integers(18, 72, n).astype(float),
    'Income': rcap.lognormal(10.9, 0.55, n).round(-2),
    'Region': rcap.choice(['North', 'South', 'East', 'West', 'Central'], n, p=[0.25, 0.2, 0.2, 0.2, 0.15]),
    'Device': rcap.choice(['Mobile', 'Desktop', 'Tablet'], n, p=[0.62, 0.3, 0.08]),
    'Membership': rcap.choice(['None', 'Silver', 'Gold'], n, p=[0.6, 0.28, 0.12]),
    'Sessions_30d': rcap.poisson(rcap.gamma(1.5, 4, n)),
    'Past_Purchases': rcap.poisson(rcap.gamma(1.2, 3, n)),
    'Days_Since_Last_Purchase': rcap.exponential(60, n).round().clip(0, 400),
    'Email_Opens_30d': rcap.poisson(3, n),
    'Discount_Usage_Rate': rcap.beta(2, 5, n).round(3),
})
cap['Time_on_Site_min'] = (cap['Sessions_30d'] * rcap.gamma(2, 3, n)).round(1)
cap['Cart_Adds_30d'] = rcap.binomial(cap['Sessions_30d'].clip(upper=60), 0.25)
cap['Time_on_Site_sec'] = cap['Time_on_Site_min'] * 60 + rcap.normal(0, 5, n)            # redundant duplicate in other units
logit_cap = -4.3 + 1.8 * (0.18 * np.sqrt(cap['Sessions_30d']) + 0.5 * np.log1p(cap['Cart_Adds_30d'])
                          + 0.35 * np.log1p(cap['Past_Purchases']) - 0.012 * cap['Days_Since_Last_Purchase']
                          + 0.08 * cap['Email_Opens_30d'] + cap['Membership'].map({'None': 0, 'Silver': 0.4, 'Gold': 0.9})
                          + 1.2 * cap['Discount_Usage_Rate'] + 0.25 * (cap['Device'] == 'Desktop'))
cap['Purchased_Next_30d'] = (rcap.random(n) < 1 / (1 + np.exp(-logit_cap))).astype(int)

# Realistic data problems
cap.loc[rcap.choice(n, 90, replace=False), 'Income'] = np.nan
cap.loc[rcap.choice(n, 60, replace=False), 'Time_on_Site_min'] = np.nan
cap.loc[rcap.choice(n, 40, replace=False), 'Device'] = np.nan
cap.loc[rcap.choice(n, 4, replace=False), 'Income'] = 99_999_999                         # typos
cap.loc[rcap.choice(n, 3, replace=False), 'Age'] = -1                                     # impossible values
cap = pd.concat([cap, cap.sample(12, random_state=1)], ignore_index=True)                 # duplicate rows

print('Shape:', cap.shape)
print('Target rate (buyers):', cap['Purchased_Next_30d'].mean().round(3))
print(cap.dtypes.value_counts())
cap.head()
````

## Step 4 — Clean the data

````python
clean = cap.drop_duplicates().copy()
print('Duplicates removed:', len(cap) - len(clean))

print('Impossible ages:', (clean['Age'] < 0).sum(), '→ NaN')
clean.loc[clean['Age'] < 0, 'Age'] = np.nan
income_cap = clean['Income'].quantile(0.99) * 5
print('Income typos (> 5 × 99th percentile):', (clean['Income'] > income_cap).sum(), '→ NaN')
clean.loc[clean['Income'] > income_cap, 'Income'] = np.nan
clean = clean.drop(columns='CustomerID')
````

## Step 5 — Missing values

````python
missing = clean.isna().mean().mul(100).round(1)
print('Missing %:\n', missing[missing > 0])
print('\nPurchase rate when Income is missing vs present:')
print(clean.groupby(clean['Income'].isna())['Purchased_Next_30d'].mean().round(3))
````

**Decision:** missingness is modest and not strongly related to the target → **median / most-frequent imputation inside the pipeline** (learned from training data), plus a missing-indicator for numeric columns (`add_indicator=True`) so the model can use missingness if it matters.

## Step 6 — Detect outliers

````python
num_cap = ['Age', 'Income', 'Sessions_30d', 'Past_Purchases', 'Days_Since_Last_Purchase', 'Email_Opens_30d',
           'Discount_Usage_Rate', 'Time_on_Site_min', 'Cart_Adds_30d', 'Time_on_Site_sec']
def iqr_outlier_share(s):
    q1, q3 = s.quantile([0.25, 0.75]); iqr = q3 - q1
    return ((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).mean() * 100
print('IQR-outlier share (%):'); print(clean[num_cap].apply(iqr_outlier_share).round(1))
````

**Decision:** after fixing the typos, the remaining "outliers" are real heavy users / long-inactive customers (right-skewed counts), not errors → **keep them**; reduce their influence with transformations (next step) and use robust tree models.

## Step 7 — Analyze skewness & Step 8 — Transform features where appropriate

````python
skewed = clean[num_cap].skew().sort_values(ascending=False)
transformed = pd.DataFrame({'skew': skewed, 'skew after log1p': np.log1p(clean[skewed.index].clip(lower=0)).skew()}).round(2)
print(transformed)
log_candidates = [c for c in transformed.index if transformed.loc[c, 'skew'] > 1]
print('\nFeatures to log1p-transform for the linear model:', log_candidates)
````

## Step 9 — Engineer features

````python
def engineer(df):
    '''Stateless, row-wise features (safe to apply before splitting — nothing is learned from other rows).'''
    df = df.copy()
    no_sessions = df['Sessions_30d'] == 0
    df['Cart_per_Session'] = np.where(no_sessions, 0, df['Cart_Adds_30d'] / df['Sessions_30d'].replace(0, np.nan))
    df['Minutes_per_Session'] = np.where(no_sessions, 0, df['Time_on_Site_min'] / df['Sessions_30d'].replace(0, np.nan))
    # 0 sessions → ratio defined as 0; a missing Time_on_Site stays NaN and is imputed later
    df['Recent_Buyer'] = (df['Days_Since_Last_Purchase'] <= 30).astype(int)
    df['Inactive_30d'] = (df['Sessions_30d'] == 0).astype(int)
    return df

fe = engineer(clean)
print(fe[['Cart_per_Session', 'Minutes_per_Session', 'Recent_Buyer', 'Inactive_30d']].describe().round(2))
print('\nPurchase rate for recent vs not-recent buyers:'); print(fe.groupby('Recent_Buyer')['Purchased_Next_30d'].mean().round(3))
````

## Step 10 — Analyze correlation & Step 11 — Check multicollinearity

````python
numeric_all = num_cap + ['Cart_per_Session', 'Minutes_per_Session', 'Recent_Buyer', 'Inactive_30d']
corr_cap = fe[numeric_all + ['Purchased_Next_30d']].corr()
plt.figure(figsize=(11, 8)); sns.heatmap(corr_cap, cmap='coolwarm', center=0, annot=True, fmt='.2f', annot_kws={'size': 7})
plt.title('Correlation matrix'); plt.show()
print('Correlation with the target:'); print(corr_cap['Purchased_Next_30d'].drop('Purchased_Next_30d').sort_values(ascending=False).round(3))

def vif_table(df):
    '''VIF_j = 1 / (1 − R²) of regressing feature j on all other features (Topic 12).'''
    d = df.dropna()
    out = {}
    for col in d.columns:
        others = d.drop(columns=col)
        r2_ = LinearRegression().fit(others, d[col]).score(others, d[col])
        out[col] = np.inf if r2_ >= 0.999999 else 1 / (1 - r2_)
    return pd.Series(out).sort_values(ascending=False)

print('\nVIF (numeric features):'); print(vif_table(fe[numeric_all]).round(1).head(8))
````

**Findings:** `Time_on_Site_sec` is the same information as `Time_on_Site_min` (r ≈ 1, huge VIF) → **drop it**. `Sessions_30d`, `Time_on_Site_min` and `Cart_Adds_30d` are also strongly related (heavy users do everything more) — acceptable for tree models; regularization handles it for the linear model.

## Step 12 — Select features

````python
features_cap = ['Age', 'Income', 'Sessions_30d', 'Past_Purchases', 'Days_Since_Last_Purchase', 'Email_Opens_30d',
                'Discount_Usage_Rate', 'Time_on_Site_min', 'Cart_Adds_30d', 'Cart_per_Session', 'Minutes_per_Session',
                'Recent_Buyer', 'Inactive_30d', 'Region', 'Device', 'Membership']
dropped = ['Time_on_Site_sec (redundant, VIF ≈ ∞)', 'CustomerID (identifier)']
print('Kept', len(features_cap), 'features. Dropped:', dropped)
````

(A data-driven check of which features the final model really uses follows in Step 19 with permutation importance — done on held-out data, so it does not leak.)

## Step 13 — Split train/test

````python
X_cap, y_cap = fe[features_cap], fe['Purchased_Next_30d']
Xc_train, Xc_test, yc_train, yc_test = train_test_split(X_cap, y_cap, test_size=0.2, stratify=y_cap, random_state=RANDOM_STATE)
print('Train:', Xc_train.shape, 'positive rate', yc_train.mean().round(3), '| Test:', Xc_test.shape, 'positive rate', yc_test.mean().round(3))
````

## Step 14 — Build the preprocessing pipeline

````python
num_features = [c for c in features_cap if c not in ['Region', 'Device', 'Membership']]
log_features = [c for c in num_features if c in log_candidates or c in ['Cart_per_Session', 'Minutes_per_Session']]
plain_features = [c for c in num_features if c not in log_features]
cat_features = ['Region', 'Device', 'Membership']

def log1p_clip(a):
    '''log(1 + x) after clipping negatives to 0. A NAMED function (not a lambda) so the pipeline can be saved with joblib.'''
    return np.log1p(np.clip(a, 0, None))

prep_linear = ColumnTransformer([
    ('log', Pipeline([('imp', SimpleImputer(strategy='median', add_indicator=True)),
                      ('log', FunctionTransformer(log1p_clip, feature_names_out='one-to-one')),
                      ('sc', StandardScaler())]), log_features),
    ('num', Pipeline([('imp', SimpleImputer(strategy='median', add_indicator=True)), ('sc', StandardScaler())]), plain_features),
    ('cat', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('oh', OneHotEncoder(handle_unknown='ignore'))]), cat_features),
])
prep_tree = ColumnTransformer([
    ('num', SimpleImputer(strategy='median', add_indicator=True), num_features),
    ('cat', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('oh', OneHotEncoder(handle_unknown='ignore'))]), cat_features),
])
print('Log-transformed:', log_features)
print('Scaled only    :', plain_features)
print('One-hot        :', cat_features)
````

## Step 15 — Train multiple models & Step 16 — Cross-validate

````python
cap_models = {
    'Baseline (most frequent)': Pipeline([('prep', prep_tree), ('model', DummyClassifier(strategy='most_frequent'))]),
    'Logistic Regression': Pipeline([('prep', prep_linear), ('model', LogisticRegression(max_iter=3000))]),
    'KNN (k=31)': Pipeline([('prep', prep_linear), ('model', KNeighborsClassifier(31))]),
    'Random Forest': Pipeline([('prep', prep_tree), ('model', RandomForestClassifier(n_estimators=400, min_samples_leaf=5, random_state=RANDOM_STATE, n_jobs=-1))]),
    'Hist Gradient Boosting': Pipeline([('prep', prep_tree), ('model', HistGradientBoostingClassifier(learning_rate=0.05, max_iter=300, random_state=RANDOM_STATE))]),
}
cv_cap = StratifiedKFold(5, shuffle=True, random_state=RANDOM_STATE)
rows = []
for name, m in cap_models.items():
    r = cross_validate(m, Xc_train, yc_train, cv=cv_cap, scoring=['average_precision', 'roc_auc'], return_train_score=True)
    rows.append({'Model': name, 'CV PR-AUC': r['test_average_precision'].mean(), 'PR-AUC std': r['test_average_precision'].std(),
                 'CV ROC-AUC': r['test_roc_auc'].mean(), 'Train ROC-AUC': r['train_roc_auc'].mean(), 'Fit time (s)': r['fit_time'].mean()})
cv_cap_tbl = pd.DataFrame(rows).set_index('Model').sort_values('CV PR-AUC', ascending=False).round(3)
cv_cap_tbl
````

## Step 17 — Tune hyperparameters (top two model families)

````python
lr_tune = GridSearchCV(cap_models['Logistic Regression'], {'model__C': [0.01, 0.03, 0.1, 0.3, 1, 3]},
                       cv=cv_cap, scoring='average_precision', n_jobs=-1).fit(Xc_train, yc_train)
hgb_tune = RandomizedSearchCV(cap_models['Hist Gradient Boosting'],
                              {'model__learning_rate': uniform(0.02, 0.12), 'model__max_iter': randint(100, 500),
                               'model__max_leaf_nodes': [8, 15, 31], 'model__min_samples_leaf': [10, 20, 40],
                               'model__l2_regularization': [0.0, 0.1, 1.0]},
                              n_iter=15, cv=cv_cap, scoring='average_precision', random_state=RANDOM_STATE, n_jobs=-1).fit(Xc_train, yc_train)
print(f'Logistic Regression best C = {lr_tune.best_params_["model__C"]} | CV PR-AUC = {lr_tune.best_score_:.3f}')
print(f'HistGB best params = { {k.split("__")[1]: (round(v, 3) if isinstance(v, float) else v) for k, v in hgb_tune.best_params_.items()} } | CV PR-AUC = {hgb_tune.best_score_:.3f}')
````

## Step 18 — Evaluate models on the test set

````python
MARGIN, CONTACT_COST = 150, 45                       # business numbers from Step 1
break_even = CONTACT_COST / MARGIN                   # contact if P(buy) > 0.30

def campaign_profit(y_true, proba, threshold):
    contact = proba >= threshold
    return int((y_true[contact] * MARGIN).sum() - contact.sum() * CONTACT_COST)

candidates = {'Logistic Regression (tuned)': lr_tune.best_estimator_,
              'Hist Gradient Boosting (tuned)': hgb_tune.best_estimator_,
              'Random Forest': cap_models['Random Forest'].fit(Xc_train, yc_train)}
rows = []
y_arr = yc_test.to_numpy()
for name, m in candidates.items():
    p = m.predict_proba(Xc_test)[:, 1]
    pred = (p >= break_even).astype(int)
    rows.append({'Model': name, 'PR-AUC': average_precision_score(yc_test, p), 'ROC-AUC': roc_auc_score(yc_test, p),
                 f'Recall @ {break_even:.3f}': recall_score(yc_test, pred), f'Precision @ {break_even:.3f}': precision_score(yc_test, pred),
                 'Customers contacted': int(pred.sum()), 'Campaign profit': campaign_profit(y_arr, p, break_even)})
rows.append({'Model': 'Contact everyone', 'Customers contacted': len(y_arr), 'Campaign profit': campaign_profit(y_arr, np.ones(len(y_arr)), 0.5)})
rows.append({'Model': 'Contact nobody', 'Customers contacted': 0, 'Campaign profit': 0})
test_cap = pd.DataFrame(rows).set_index('Model').round(3)
test_cap
````

## Step 19 — Error analysis

````python
chosen_name = 'Logistic Regression (tuned)'          # provisional; confirmed in Step 20
chosen = candidates[chosen_name]
p_test = chosen.predict_proba(Xc_test)[:, 1]
ea_cap = Xc_test.copy(); ea_cap['y'] = y_arr; ea_cap['p'] = p_test; ea_cap['pred'] = (p_test >= break_even).astype(int)
ea_cap['type'] = np.select([(ea_cap.y == 1) & (ea_cap.pred == 0), (ea_cap.y == 0) & (ea_cap.pred == 1)], ['FN', 'FP'], 'OK')

print('Error types:', ea_cap['type'].value_counts().to_dict())
print('\nSegment analysis (PR-AUC by membership):')
for seg, g in ea_cap.groupby('Membership', dropna=False):
    if g['y'].nunique() == 2:
        print(f'  {str(seg):<6} n = {len(g):>3} | buy rate {g.y.mean():.2f} | PR-AUC {average_precision_score(g.y, g.p):.3f}')
print('\nMissed buyers (FN) vs caught buyers (TP) — averages:')
print(ea_cap[ea_cap.y == 1].groupby('pred')[['Sessions_30d', 'Cart_Adds_30d', 'Days_Since_Last_Purchase', 'Past_Purchases']].mean().round(1)
      .rename(index={0: 'missed (FN)', 1: 'caught (TP)'}))

perm_cap = permutation_importance(chosen, Xc_test, yc_test, scoring='average_precision', n_repeats=10, random_state=RANDOM_STATE, n_jobs=-1)
pd.Series(perm_cap.importances_mean, index=features_cap).sort_values().plot(kind='barh', figsize=(7, 5),
                                                                           title='Permutation importance (drop in test PR-AUC)')
plt.show()
````

**Reading the error analysis:**
- **Missed buyers (FN)** are much **less active** than the buyers we caught: about half the sessions and cart adds, and a longer time since their last purchase. They buy with little warning in our data. The available features don't signal their intent, so better data (e.g. wishlist activity, seasonal events) would be needed to catch them.
- The model is weakest for customers **without a membership** (lowest PR-AUC). That is the largest group and also the one with the lowest buying rate, which makes it a natural place for new features.
- **Permutation importance** shows that recency, cart activity, membership and past purchases drive the predictions. All are plausible and known before the month starts, so there is no sign of leakage.

## Step 20 — Select a model based on documented evaluation criteria

````python
criteria = pd.DataFrame({
    'Test PR-AUC': test_cap.loc[list(candidates), 'PR-AUC'],
    'CV PR-AUC': np.round([lr_tune.best_score_, hgb_tune.best_score_, cv_cap_tbl.loc['Random Forest', 'CV PR-AUC']], 3),
    'Campaign profit (test)': test_cap.loc[list(candidates), 'Campaign profit'],
    'Interpretability': ['High (coefficients)', 'Medium (needs PDP/SHAP)', 'Medium'],
    'Training/serving cost': ['Very low', 'Low', 'Medium (400 trees)'],
})
criteria
````

**Decision rule (documented before looking at the test set):** choose the model with the best **cross-validated PR-AUC**; if the difference between models is smaller than about one standard deviation of the CV scores, prefer the **simpler, more interpretable** model. Use test profit only as a final sanity check, not to pick.

→ **Result:** the tuned logistic regression has the best **CV PR-AUC** (≈ 0.54). The random forest (≈ 0.54) and gradient boosting (≈ 0.53) are within about one standard deviation of it (CV std ≈ 0.02–0.05). On the test set all three give similar PR-AUC. The random forest earns a little more campaign profit on these 300 test customers, but that gap is within noise too. The rule says: **select the tuned logistic regression**. It has equal-best performance, the highest interpretability (marketing can see *why* someone was targeted) and a trivial serving cost. The random forest is kept as a **challenger** to re-evaluate at the next retraining.

All three models beat both simple strategies: contacting **everyone** loses money (only ~22% buy, and each contact costs 45), and contacting **nobody** earns nothing.

## Step 21 — Save the model using joblib

````python
final_cap = chosen                                                  # already refitted on the full training set by GridSearchCV
# ⚠️ The pipeline references engineer() and log1p_clip(): any script that loads this file must define/import them too
#    (in real projects, put such functions in a small Python module that both training and serving code import).

artifact_path = 'ml_models/purchase_model.joblib'
joblib.dump(final_cap, artifact_path)
json.dump({'model': chosen_name, 'features': features_cap, 'target': 'Purchased_Next_30d', 'threshold': break_even,
           'business_rule': f'contact if P(buy) >= {break_even:.3f} (margin {MARGIN}, cost {CONTACT_COST})',
           'test_pr_auc': round(float(average_precision_score(yc_test, final_cap.predict_proba(Xc_test)[:, 1])), 4),
           'sklearn_version': sklearn.__version__, 'trained_rows': int(len(Xc_train))},
          open('ml_models/purchase_model_metadata.json', 'w'), indent=2)
print('Saved:', artifact_path)
````

## Step 24 — How this model could be deployed

**Batch (recommended here):** the campaign is monthly, so real-time scoring isn't needed.
```text
1st of each month (scheduled job, e.g. cron / Airflow):
  1. Extract last-30-day activity for all customers from the database
  2. engineer() → loaded pipeline.predict_proba()
  3. Save CustomerID, P_Buy, Send_Campaign, model_version, date → table used by the email tool
  4. Log input statistics and the share of customers selected
```
**Monitoring:** compare each month's feature distributions with training (PSI/KS, Section 35); when purchases are known 30 days later, compute PR-AUC and campaign profit for that month; alert if they drop.
**Retraining:** quarterly, or when drift/performance alerts fire; the challenger model (gradient boosting) is re-evaluated at every retraining.
**Real-time option:** if the website later wants to show an offer while the customer browses, wrap the same pipeline in a small API (FastAPI) that loads the model once and scores one customer per request (`predict_one`, Section 34).
