# 1. What is Machine Learning?

### AI, ML, Deep Learning, Generative AI

```text
┌──────────────────────── Artificial Intelligence ────────────────────────┐
│  Any technique that makes computers do "intelligent" tasks              │
│  (rules, search, planning, chess engines, expert systems, ML ...)       │
│   ┌────────────────────── Machine Learning ──────────────────────────┐  │
│   │  Systems that LEARN patterns from data instead of hand-written   │  │
│   │  rules (linear regression, trees, random forests, SVM, K-Means)  │  │
│   │   ┌──────────────────── Deep Learning ─────────────────────────┐ │  │
│   │   │  ML with many-layered neural networks (images, audio, text)│ │  │
│   │   │   ┌──────────────── Generative AI ──────────────────────┐  │ │  │
│   │   │   │  Models that CREATE new content: text (ChatGPT,     │  │ │  │
│   │   │   │  Claude), images, code, music                       │  │ │  │
│   │   │   └─────────────────────────────────────────────────────┘  │ │  │
│   │   └────────────────────────────────────────────────────────────┘ │  │
│   └──────────────────────────────────────────────────────────────────┘  │
└─────────────────────────────────────────────────────────────────────────┘
```

| Term | Simple meaning | Example |
|---|---|---|
| **Artificial Intelligence** | Making machines do tasks that need "intelligence" | A chess program, a route planner, a chatbot |
| **Machine Learning** | Learning patterns from examples | Predicting house prices from past sales |
| **Deep Learning** | ML with large neural networks, great for unstructured data | Recognising faces, speech-to-text |
| **Generative AI** | Deep learning models that generate new content | Writing an email, creating an image |

For **tabular business data** (spreadsheets, databases), classic ML (this notebook) is often the best tool — it's fast, cheap, and strong.

### Traditional programming vs machine learning

```text
Traditional Programming:   Rules + Data            → Output
Machine Learning:          Data + Expected Output  → Model (the "rules")
                           then:  Model + New Data → Predictions
```

**Example — spam filtering**
- Traditional: a programmer writes `if 'free money' in email: spam`. Spammers change words → rules break → endless maintenance.
- ML: give the computer 100,000 emails labelled spam/not spam. It **learns** which patterns matter and adapts when retrained on new data.

### What does a model actually learn?
A model learns **parameters** — numbers that define a function mapping inputs to outputs.
- Linear regression learns **weights**: `price = 50,000 + 3,000 × area − 1,000 × age`.
- A decision tree learns **if/else split rules**: `if area > 1500 and location == 'City'`.
- A neural network learns **millions of weights**.

The goal is not to memorise the training data but to **generalise** — to make good predictions on **new, unseen** data.

````python
# Traditional programming: WE write the rule
def price_rule(area):
    return 20_000 + 150 * area                # a human guessed these numbers

# Machine learning: the MODEL learns the rule from examples
area = rng.uniform(500, 3000, 1000)
price = 35_000 + 120 * area + rng.normal(0, 15_000, 1000)     # the true (unknown) relationship + noise

model = LinearRegression().fit(area.reshape(-1, 1), price)       # learn from data + expected output
print(f'Learned rule: price = {model.intercept_:,.0f} + {model.coef_[0]:.1f} × area')
print(f'Prediction for a 2000 sq ft house (ML)  : {model.predict([[2000]])[0]:,.0f}')
print(f'Prediction for a 2000 sq ft house (rule): {price_rule(2000):,.0f}')
````

**Interpretation:** the model rediscovered numbers close to the true ones (35,000 and 120) purely from examples. The hand-written rule was just a guess.

# 2. Types of Machine Learning

| Type | Data | Goal | Examples | Common algorithms |
|---|---|---|---|---|
| **Supervised** | Features **+ labels** (the right answers) | Predict the label for new data | House price, churn, spam, fraud, demand forecasting | Linear/logistic regression, trees, random forest, gradient boosting, SVM, KNN |
| **Unsupervised** | Features only, **no labels** | Find hidden structure | Customer segments, anomaly detection, compressing features, topic discovery | K-Means, hierarchical clustering, DBSCAN, PCA |
| **Semi-supervised** | A **few** labels + **many** unlabelled rows | Use unlabelled data to improve learning | Medical images where only 5% are labelled by doctors | Self-training, label propagation |
| **Reinforcement learning** | An **agent** acts in an **environment** and receives **rewards** | Learn a strategy that maximises long-term reward | Game-playing AI, robotics, ad bidding, recommendation, training LLMs from human feedback (RLHF) | Q-learning, policy gradients |

### When is each used?
- **Supervised:** you have historical examples **with the correct answer** — the most common type in business.
- **Unsupervised:** you have **no labels** and want to explore, group, compress or detect unusual items.
- **Semi-supervised:** labels are **expensive** (need experts) but raw data is plentiful.
- **Reinforcement:** decisions happen **in sequence** and feedback is a **reward** (often delayed).

### Supervised learning has two main tasks
- **Regression** → predict a **number** (Section 6).
- **Classification** → predict a **category** (Section 6).

### Unsupervised learning has two main tasks
- **Clustering** → group similar rows (Sections 20–21).
- **Dimensionality reduction** → compress many features into fewer (Section 22).

````python
from sklearn.datasets import make_blobs

X_demo, y_demo = make_blobs(n_samples=300, centers=3, cluster_std=1.2, random_state=RANDOM_STATE)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
axes[0].scatter(X_demo[:, 0], X_demo[:, 1], c=y_demo, cmap='viridis', s=20)
axes[0].set_title('Supervised: we KNOW the label (colour) of each point\n→ learn to predict it for new points')
axes[1].scatter(X_demo[:, 0], X_demo[:, 1], color='gray', s=20)
axes[1].set_title('Unsupervised: NO labels\n→ discover the groups ourselves')
plt.tight_layout(); plt.show()
````

# 3. ML Terminology

| Term | Meaning | In the table below |
|---|---|---|
| **Dataset** | The whole table of examples | All rows |
| **Sample** (row, observation, instance) | One example | One house |
| **Feature** (input, predictor, X) | A column used to make predictions | Area, Bedrooms, Age |
| **Target** (output, y) | The column we want to predict | Price |
| **Label** | The known value of the target for a sample (the "answer") | 250,000 for house 1 |
| **Model** | The learned function f(X) → y | `LinearRegression` after `fit` |
| **Training** (fitting) | Learning parameters from data | `model.fit(X_train, y_train)` |
| **Prediction** | Output of the model for new inputs | `model.predict(X_new)` |
| **Parameter** | Learned **by the model** from data | Coefficients, tree split thresholds |
| **Hyperparameter** | Chosen **by you** before training; controls how learning happens | `max_depth` of a tree, `alpha` in Ridge, `k` in KNN |
| **Inference** | Using a trained model to make predictions (often in production) | The website calling the model to price a new listing |

````python
houses = pd.DataFrame({
    'Area': [1200, 1500, 900, 2000, 1750, 1100],
    'Bedrooms': [2, 3, 2, 4, 3, 2],
    'Age': [10, 5, 30, 2, 8, 20],
    'Price': [250_000, 320_000, 160_000, 450_000, 380_000, 210_000],
})
X = houses[['Area', 'Bedrooms', 'Age']]       # FEATURES (capital X by convention: a 2-D table)
y = houses['Price']                           # TARGET  (lowercase y: a 1-D column of labels)

model = LinearRegression()                    # MODEL (not trained yet); no hyperparameters needed here
model.fit(X, y)                               # TRAINING: learn the PARAMETERS

print('Learned parameters:')
print('  intercept   :', round(model.intercept_, 1))
print('  coefficients:', dict(zip(X.columns, model.coef_.round(1))))

new_house = pd.DataFrame({'Area': [1600], 'Bedrooms': [3], 'Age': [12]})
print('\nINFERENCE — predicted price for a new house:', round(model.predict(new_house)[0]))

tree = DecisionTreeRegressor(max_depth=2)     # max_depth is a HYPERPARAMETER (we choose it)
tree.fit(X, y)
print('Tree hyperparameter max_depth =', tree.get_params()['max_depth'], '| learned split parameters:', tree.tree_.threshold[:3].round(1))
````

### Line-by-line
- `X = houses[[...]]` → double brackets → a DataFrame (2-D), which scikit-learn expects for features.
- `y = houses['Price']` → single brackets → a Series (1-D).
- `.fit()` learns parameters; `.predict()` uses them. Every scikit-learn model follows this same **`fit` / `predict`** pattern — once you learn it, you know how to use them all.
- Learned attributes end with an underscore: `coef_`, `intercept_`, `feature_importances_`.

# 4. The Complete Machine Learning Workflow

```text
Problem Definition → Data Collection → Data Understanding → EDA → Data Cleaning → Feature Engineering
   → Train/Test Split → Preprocessing → Model Training → Validation → Hyperparameter Tuning
   → Evaluation → Model Selection → Deployment → Monitoring   (→ back to the start when things change)
```

| Stage | What you do | Key questions | Topic where you learned it |
|---|---|---|---|
| **1. Problem definition** | Translate the business need into an ML task | What decision will the prediction support? Regression or classification? What metric defines success? What does an error cost? | — |
| **2. Data collection** | Gather data from databases, APIs, logs, surveys | Is the data relevant, legal, representative? Is the target available? | Topic 2, 10 |
| **3. Data understanding** | Types, meaning, units, granularity | What does each column mean? What is one row? | Topic 2 |
| **4. EDA** | Distributions, relationships, anomalies | Skewed? Outliers? Correlations? Class balance? | Topics 7, 8 |
| **5. Data cleaning** | Missing values, duplicates, errors, outliers | Why is data missing? Are outliers errors or real? | Topics 3, 5 |
| **6. Feature engineering** | Create useful inputs | What domain knowledge can we add? | Topic 11 |
| **7. Train/test split** | Hold back data the model never sees during development | Random, stratified, or time-based split? | Section 5 |
| **8. Preprocessing** | Imputation, encoding, scaling, transformation — **fitted on training data only** | Which steps does each model need? | Topics 4, 6, 13 |
| **9. Model training** | Fit one or more algorithms | Start with a simple baseline! | Sections 7–18 |
| **10. Validation** | Estimate performance on unseen data (validation set / cross-validation) | Is it overfitting? | Section 25 |
| **11. Hyperparameter tuning** | Search for the best settings using validation data | Grid or random search? | Section 26 |
| **12. Evaluation** | Final, one-time check on the **test set** | Does it meet the success criteria? | Sections 8, 12 |
| **13. Model selection** | Choose a model based on performance **and** simplicity, speed, interpretability, cost | Is the extra complexity worth it? | Section 32 |
| **14. Deployment** | Save the model; serve predictions (batch or real-time) | How will it be called? | Section 34 |
| **15. Monitoring** | Track data drift, performance, errors; retrain when needed | Is the world changing? | Section 35 |

**Real projects are loops, not straight lines.** Error analysis often sends you back to feature engineering or data collection.

### The most important habit: **start with a baseline**
Before any clever model, measure a **dumb baseline**: predict the mean (regression) or the most common class (classification). If your model can't beat it, something is wrong.

````python
from sklearn.dummy import DummyRegressor, DummyClassifier

# Regression baseline: always predict the mean price
y_price = pd.Series(35_000 + 120 * area + rng.normal(0, 25_000, len(area)))
X_area = pd.DataFrame({'Area': area})
Xtr, Xte, ytr, yte = train_test_split(X_area, y_price, test_size=0.25, random_state=RANDOM_STATE)

baseline = DummyRegressor(strategy='mean').fit(Xtr, ytr)
linear = LinearRegression().fit(Xtr, ytr)
print(f'Baseline MAE (predict the mean): {mean_absolute_error(yte, baseline.predict(Xte)):,.0f}')
print(f'Linear regression MAE          : {mean_absolute_error(yte, linear.predict(Xte)):,.0f}')
````

# 5. Train / Validation / Test

> **Analogy:** A student has **practice questions** (training set), **mock exams** to check progress and adjust study strategy (validation set), and **one final exam** they have never seen (test set). If the student sees the final exam in advance, their score says nothing about what they really learned.

| Set | Used for | Typical share |
|---|---|---|
| **Training** | Fitting the model's parameters | 60–80% |
| **Validation** | Choosing models, hyperparameters, features, thresholds | 10–20% (or use cross-validation on the training set) |
| **Test** | **One final**, unbiased estimate of performance on new data | 10–20% |

### Why must the test set stay unseen?
Every decision you make while looking at a dataset (a feature, a hyperparameter, a threshold) slightly **fits your model to that dataset**. If you tune on the test set, the test score becomes **optimistic** — your model will perform worse in the real world. Use the test set **once**, at the very end.

### `train_test_split` — key arguments
- `test_size=0.2` → 20% goes to the test set.
- `random_state=42` → reproducible split.
- `stratify=y` → keeps the **class proportions** the same in train and test (important for imbalanced classification).
- `shuffle=False` → for **time series**, don't shuffle; train on the past, test on the future.

````python
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer(as_frame=True)
Xc, yc = cancer.data, cancer.target                     # 569 tumours, 30 features, target 1 = benign, 0 = malignant

# Step 1: split off the TEST set (20%) — lock it away
X_temp, X_test, y_temp, y_test = train_test_split(Xc, yc, test_size=0.20, stratify=yc, random_state=RANDOM_STATE)
# Step 2: split the rest into TRAIN (75% of the rest = 60% overall) and VALIDATION (25% = 20% overall)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.25, stratify=y_temp, random_state=RANDOM_STATE)

for name, yy in [('Full', yc), ('Train', y_train), ('Validation', y_val), ('Test', y_test)]:
    print(f'{name:<11} rows = {len(yy):>3} | share benign = {yy.mean():.3f}')
````

````python
# Use the validation set to choose a hyperparameter, then evaluate ONCE on the test set
val_scores = {}
for depth in [1, 2, 3, 4, 6, 8, None]:
    m = DecisionTreeClassifier(max_depth=depth, random_state=RANDOM_STATE).fit(X_train, y_train)
    val_scores[str(depth)] = accuracy_score(y_val, m.predict(X_val))
print('Validation accuracy by max_depth:', {k: round(v, 3) for k, v in val_scores.items()})

best_depth = max(val_scores, key=val_scores.get)
best_depth = None if best_depth == 'None' else int(best_depth)
final = DecisionTreeClassifier(max_depth=best_depth, random_state=RANDOM_STATE).fit(X_temp, y_temp)  # refit on train+val
print(f'Chosen max_depth = {best_depth} → FINAL test accuracy = {accuracy_score(y_test, final.predict(X_test)):.3f}')
````

### Data leakage (preview — full details in Section 33)
**Data leakage** = information from outside the training data (the test set, the future, or the target itself) sneaks into training. The model looks great in testing and fails in production.

Classic example: `StandardScaler().fit(all_data)` **before** splitting → the scaler learned the test set's mean and std. Fix: **split first**, fit preprocessing on training data only — best done with a **Pipeline** (Section 27).

### When NOT to use a random split
- **Time series** → split by time.
- **Grouped data** (several rows per customer/patient) → split by group (`GroupShuffleSplit`) so the same person isn't in train and test.

# 7. Linear Regression

### What it is
Linear regression draws the **best straight line** (or plane) through the data.

**Simple** (one feature):
$$\hat{y} = b_0 + b_1 x$$

**Multiple** (many features):
$$\hat{y} = b_0 + b_1 x_1 + b_2 x_2 + \dots + b_p x_p$$

| Term | Meaning | Example (salary model) |
|---|---|---|
| **Intercept b₀** | Prediction when all features are 0 | Starting salary at 0 years experience |
| **Coefficient bᵢ** | Change in ŷ when xᵢ increases by 1, **holding other features fixed** | +3,000 per year of experience |
| **Prediction ŷ** | The model's output | 95,000 |
| **Residual** | Actual − predicted = the error | 100,000 − 95,000 = +5,000 |

### How does it choose the line? — **Ordinary Least Squares (OLS)**
It picks the coefficients that **minimise the sum of squared residuals**. Squaring makes big errors count much more and makes the maths solvable exactly.

### WHY / WHEN / WHEN NOT
- ✅ Fast, simple, **interpretable** coefficients, a great **baseline**, works well when relationships are roughly linear.
- ❌ Can't capture curves/interactions unless you engineer them; sensitive to outliers; coefficients unstable with multicollinearity (Topic 12); extrapolation beyond the data range can be nonsense.
- Scaling is **not needed** for predictions, but helps when comparing coefficient sizes or using regularization.

````python
rs = np.random.default_rng(7)
n = 500
salary_df = pd.DataFrame({'Experience': rs.uniform(0, 20, n).round(1),
                          'Education_Years': rs.integers(12, 20, n),
                          'Remote': rs.integers(0, 2, n)})
salary_df['Salary'] = (30_000 + 3_000 * salary_df['Experience'] + 2_500 * salary_df['Education_Years']
                       - 4_000 * salary_df['Remote'] + rs.normal(0, 8_000, n)).round(-2)

# ---- Simple linear regression: Salary ~ Experience
simple = LinearRegression().fit(salary_df[['Experience']], salary_df['Salary'])
print(f'SIMPLE:   Salary = {simple.intercept_:,.0f} + {simple.coef_[0]:,.0f} × Experience')

# ---- Multiple linear regression: Salary ~ all features
features = ['Experience', 'Education_Years', 'Remote']
multi = LinearRegression().fit(salary_df[features], salary_df['Salary'])
print(f'MULTIPLE: Salary = {multi.intercept_:,.0f}', ' '.join(f'+ ({c:,.0f} × {f})' for f, c in zip(features, multi.coef_)))

salary_df['Predicted'] = multi.predict(salary_df[features])
salary_df['Residual'] = salary_df['Salary'] - salary_df['Predicted']
salary_df.head()
````

````python
fig, axes = plt.subplots(1, 3, figsize=(18, 4.2))
xs = np.linspace(0, 20, 100).reshape(-1, 1)
axes[0].scatter(salary_df['Experience'], salary_df['Salary'], alpha=0.5)
axes[0].plot(xs, simple.predict(pd.DataFrame(xs, columns=['Experience'])), color='red', lw=2)
axes[0].set_title('Simple linear regression: best-fit line'); axes[0].set_xlabel('Experience')

axes[1].scatter(salary_df['Predicted'], salary_df['Salary'], alpha=0.5)
lims = [salary_df['Salary'].min(), salary_df['Salary'].max()]
axes[1].plot(lims, lims, 'r--'); axes[1].set_title('Multiple regression: actual vs predicted')
axes[1].set_xlabel('Predicted'); axes[1].set_ylabel('Actual')

axes[2].scatter(salary_df['Predicted'], salary_df['Residual'], alpha=0.5); axes[2].axhline(0, color='red', ls='--')
axes[2].set_title('Residuals vs predicted (should look like random noise)'); axes[2].set_xlabel('Predicted')
plt.tight_layout(); plt.show()
````

### Interpretation
- The learned coefficients are close to the true values used to generate the data (+3,000 per year of experience, +2,500 per year of education, −4,000 for remote). They are not exact because of random noise — more data brings them closer.
- **Residuals** form a random cloud around 0 with no pattern → a linear model fits this data well. A **curved** or **funnel** pattern would mean non-linearity or unequal variance (Topic 14, assumptions).
- The simple model's experience coefficient differs a little from the multiple model's because the simple model can't separate the effects of the other features.

### Code explanation
- `LinearRegression().fit(X, y)` solves OLS.
- `.intercept_` and `.coef_` hold b₀ and b₁…bₚ (in the order of `X`'s columns).
- `' '.join(f'...' for f, c in zip(...))` just builds a readable equation string.

# 8. Regression Metrics

| Metric | Formula | Units | Meaning | Use when |
|---|---|---|---|---|
| **MAE** | mean(\|y − ŷ\|) | Same as target | Average size of the error | You want an easy-to-explain error; outliers shouldn't dominate |
| **MSE** | mean((y − ŷ)²) | Target² | Average squared error | Mainly for optimisation; punishes big errors strongly |
| **RMSE** | √MSE | Same as target | "Typical" error, but big errors weigh more | Big errors are especially costly |
| **R²** | 1 − SS_res / SS_tot | None (≤ 1) | Share of the target's variation explained (1 = perfect, 0 = as good as predicting the mean, < 0 = worse than the mean) | Comparing models on the same data; communicating fit |
| **Adjusted R²** | 1 − (1 − R²)(n − 1)/(n − p − 1) | None | R² penalised for the number of features p | Comparing linear models with different numbers of features |

### Numerical example
Actual `[100, 200, 300, 400]`, predicted `[110, 190, 330, 380]` → errors `[−10, 10, −30, 20]`
- MAE = (10 + 10 + 30 + 20) / 4 = **17.5**
- MSE = (100 + 100 + 900 + 400) / 4 = **375**
- RMSE = √375 ≈ **19.36**
- R²: mean = 250, SS_tot = 150² + 50² + 50² + 150² = 50,000, SS_res = 1,500 → R² = 1 − 1,500/50,000 = **0.97**

### Why Adjusted R²?
Adding **any** feature (even random noise) never lowers training R². Adjusted R² subtracts a penalty for each extra feature, so useless features make it go **down**.

````python
y_true = np.array([100, 200, 300, 400])
y_pred = np.array([110, 190, 330, 380])
print('MAE  =', mean_absolute_error(y_true, y_pred))
print('MSE  =', mean_squared_error(y_true, y_pred))
print('RMSE =', round(rmse(y_true, y_pred), 2))
print('R²   =', r2_score(y_true, y_pred))

def adjusted_r2(r2, n, p):
    return 1 - (1 - r2) * (n - 1) / (n - p - 1)

# Adding random noise features: R² creeps up, adjusted R² goes down
X_base = salary_df[features].copy()
for n_noise in [0, 5, 20, 50]:
    Xn = X_base.copy()
    for i in range(n_noise):
        Xn[f'noise_{i}'] = rs.normal(size=len(Xn))
    r2 = LinearRegression().fit(Xn, salary_df['Salary']).score(Xn, salary_df['Salary'])
    print(f'{n_noise:>2} noise features → training R² = {r2:.4f} | adjusted R² = {adjusted_r2(r2, len(Xn), Xn.shape[1]):.4f}')
````

````python
# Outliers affect RMSE much more than MAE
y_true = np.array([100, 200, 300, 400, 500])
good = np.array([105, 195, 310, 390, 505])
one_big_miss = np.array([105, 195, 310, 390, 700])
for name, p in [('small errors', good), ('one big miss', one_big_miss)]:
    print(f'{name:<13} MAE = {mean_absolute_error(y_true, p):6.1f} | RMSE = {rmse(y_true, p):6.1f}')
````

### Interpretation
One large miss raises MAE modestly but RMSE a lot. Pick the metric that matches the **business cost**:
- A delivery that is 60 minutes late is much worse than six 10-minute delays → **RMSE**.
- Every rupee of error costs the same → **MAE**.
- Relative errors matter (a 10k error on a 50k house is worse than on a 5-crore house) → **MAPE** or a log-target model.

### Common mistakes
- Comparing R² across **different datasets** (it depends on how variable the target is).
- Reporting only R² — always add an error in **real units** (MAE/RMSE).
- Using **training** metrics to judge a model — use validation/test data.

# 9. Polynomial Regression

### What it is
Linear regression can only draw straight lines. **Polynomial regression** adds powers of the feature (x², x³, …) so a *linear* model can fit a **curve**:

$$\hat{y} = b_0 + b_1 x + b_2 x^2 + b_3 x^3 + \dots$$

It's still "linear regression" — linear in the **coefficients** — on engineered features (Topic 11).

### Degree = model complexity
| Degree | Shape | Risk |
|---|---|---|
| 1 | Straight line | **Underfitting** a curved pattern (too simple) |
| 2–3 | Gentle curves | Often just right |
| 10+ | Wiggly curve that passes near every point | **Overfitting** the noise (too complex) |

### WHY / WHEN / WHEN NOT
- ✅ A scatter plot shows a clear curve; you want to stay with an interpretable linear model.
- ❌ Many features (feature count explodes), high degrees (wild behaviour at the edges), extrapolation.
- Always **scale** after creating polynomial features if you also regularize.

````python
rp = np.random.default_rng(9)
x_poly = np.sort(rp.uniform(-3, 3, 40))
y_poly = 0.5 * x_poly ** 3 - x_poly ** 2 + x_poly + rp.normal(0, 2.5, 40)   # the true pattern is a cubic curve
Xp = x_poly.reshape(-1, 1)
Xp_tr, Xp_te, yp_tr, yp_te = train_test_split(Xp, y_poly, test_size=0.3, random_state=RANDOM_STATE)

x_plot = np.linspace(-3, 3, 300).reshape(-1, 1)
fig, axes = plt.subplots(1, 5, figsize=(22, 4), sharey=True)
rows = []
for ax, degree in zip(axes, [1, 2, 3, 5, 10]):
    model = make_pipeline(PolynomialFeatures(degree, include_bias=False), StandardScaler(), LinearRegression())
    model.fit(Xp_tr, yp_tr)
    tr_err, te_err = rmse(yp_tr, model.predict(Xp_tr)), rmse(yp_te, model.predict(Xp_te))
    rows.append({'degree': degree, 'train RMSE': tr_err, 'test RMSE': te_err})
    ax.scatter(Xp_tr, yp_tr, s=15, label='train'); ax.scatter(Xp_te, yp_te, s=15, color='orange', label='test')
    ax.plot(x_plot, model.predict(x_plot), color='red'); ax.set_ylim(-25, 20)
    ax.set_title(f'degree {degree}\ntrain {tr_err:.2f} | test {te_err:.2f}')
axes[0].legend(); plt.tight_layout(); plt.show()
print('One random split:'); print(pd.DataFrame(rows).round(3).to_string(index=False))
````

````python
# One split of 40 points is noisy — repeat over 50 different random splits for a reliable picture
rows = []
for degree in [1, 2, 3, 5, 10]:
    tr_errs, te_errs = [], []
    for seed in range(50):
        a_tr, a_te, b_tr, b_te = train_test_split(Xp, y_poly, test_size=0.3, random_state=seed)
        m = make_pipeline(PolynomialFeatures(degree, include_bias=False), StandardScaler(), LinearRegression()).fit(a_tr, b_tr)
        tr_errs.append(rmse(b_tr, m.predict(a_tr))); te_errs.append(rmse(b_te, m.predict(a_te)))
    rows.append({'degree': degree, 'mean train RMSE': np.mean(tr_errs), 'median test RMSE': np.median(te_errs)})
avg = pd.DataFrame(rows)
print(avg.round(3).to_string(index=False))

plt.figure(figsize=(7, 3.8))
plt.plot(avg['degree'], avg['mean train RMSE'], 'o-', label='train'); plt.plot(avg['degree'], avg['median test RMSE'], 's-', label='test')
plt.xlabel('polynomial degree (model complexity)'); plt.ylabel('RMSE'); plt.xticks(avg['degree'])
plt.title('Training error keeps falling; test error is lowest at degree 3'); plt.legend(); plt.show()
````

### Interpretation — what happens as complexity increases?
- **Degree 1:** a straight line misses the curve → high error on **both** train and test (**underfitting**).
- **Degree 3:** matches the true cubic shape → low train **and** low test error (**good fit**).
- **Degree 10:** training error keeps dropping, but the curve wiggles to chase noise, and test error rises again (**overfitting**).
- A single split of only 40 points can mislead (by luck, a complex model may look fine on one split). The **average over 50 splits** gives the reliable picture — this is exactly the idea behind **cross-validation** (Section 25).

**Rule:** training error always falls as complexity rises; **test/validation error** tells you when to stop.

### Code explanation
`make_pipeline(PolynomialFeatures(d), StandardScaler(), LinearRegression())` creates x, x², …, xᵈ, scales them (high powers are huge numbers), and fits a linear model — all as one object.

# 10. Regularization — Ridge, Lasso, ElasticNet

### Why is it needed?
With many features, correlated features, or complex features (polynomials), linear models can **overfit**: coefficients become huge and unstable, fitting noise.

**Regularization** adds a **penalty on coefficient size** to the loss:
$$\text{Loss} = \underbrace{\sum (y - \hat{y})^2}_{\text{fit the data}} + \underbrace{\alpha \times \text{penalty}}_{\text{keep coefficients small}}$$

| Model | Penalty | Effect on coefficients | Use when |
|---|---|---|---|
| **Ridge** (L2) | α × Σ bᵢ² | Shrinks all coefficients **toward** 0 smoothly; never exactly 0 | Many useful, correlated features (multicollinearity) |
| **Lasso** (L1) | α × Σ \|bᵢ\| | Pushes some coefficients **exactly to 0** → automatic **feature selection** | You suspect only a few features matter |
| **ElasticNet** | mix: `l1_ratio` × L1 + (1 − `l1_ratio`) × L2 | Some zeros + stable with correlated groups | Many correlated features and you still want selection |

- **α (alpha)** = penalty strength. α = 0 → ordinary linear regression. Larger α → simpler model (more bias, less variance).
- ⚠️ **Always scale features** first: the penalty treats all coefficients equally, so features in large units would be unfairly penalised.

````python
rr = np.random.default_rng(10)
n, p = 80, 20
X_reg = pd.DataFrame(rr.normal(size=(n, p)), columns=[f'x{i}' for i in range(p)])
X_reg['x1_copy'] = X_reg['x1'] + rr.normal(0, 0.05, n)                 # nearly duplicated feature (multicollinearity)
true_coefs = np.zeros(X_reg.shape[1]); true_coefs[[0, 1, 2]] = [5, 3, -4]   # only x0, x1, x2 truly matter
y_reg = X_reg.to_numpy() @ true_coefs + rr.normal(0, 2, n)

Xr_tr, Xr_te, yr_tr, yr_te = train_test_split(X_reg, y_reg, test_size=0.3, random_state=RANDOM_STATE)

models = {
    'Linear': LinearRegression(),
    'Ridge (α=1)': Ridge(alpha=1.0),
    'Lasso (α=0.2)': Lasso(alpha=0.2),
    'ElasticNet (α=0.2, l1=0.5)': ElasticNet(alpha=0.2, l1_ratio=0.5),
}
coef_table, results = {}, []
for name, m in models.items():
    pipe = make_pipeline(StandardScaler(), m).fit(Xr_tr, yr_tr)
    coef_table[name] = pipe[-1].coef_
    results.append({'Model': name, 'Train RMSE': rmse(yr_tr, pipe.predict(Xr_tr)),
                    'Test RMSE': rmse(yr_te, pipe.predict(Xr_te)),
                    'Zero coefficients': int(np.sum(np.abs(pipe[-1].coef_) < 1e-6))})
print(pd.DataFrame(results).round(3).to_string(index=False))
coefs = pd.DataFrame(coef_table, index=X_reg.columns).round(2)
coefs.head(6)
````

````python
alphas = np.logspace(-3, 1.5, 60)
lasso_path = [make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=10_000)).fit(Xr_tr, yr_tr)[-1].coef_ for a in alphas]
ridge_path = [make_pipeline(StandardScaler(), Ridge(alpha=a * 50)).fit(Xr_tr, yr_tr)[-1].coef_ for a in alphas]

fig, axes = plt.subplots(1, 2, figsize=(14, 4.2))
axes[0].plot(alphas * 50, ridge_path); axes[0].set_xscale('log'); axes[0].set_title('Ridge: coefficients shrink smoothly, never exactly 0')
axes[1].plot(alphas, lasso_path); axes[1].set_xscale('log'); axes[1].set_title('Lasso: coefficients hit exactly 0 one by one')
for ax in axes: ax.set_xlabel('alpha (penalty strength)'); ax.set_ylabel('coefficient')
plt.tight_layout(); plt.show()
````

### Interpretation
- Plain **linear regression** gives non-zero coefficients to all 21 features and splits x1's effect between x1 and its near-copy in a wild way (one large negative, one large positive coefficient).
- **Ridge** keeps all features but shrinks them; x1 and its copy **share** the effect in a stable, sensible way. With this small α its test error is similar to plain regression — α should be **tuned** (see below and Section 26).
- **Lasso** sets most of the useless coefficients to exactly **0** — it found the few real features — and it has a **higher train error but lower test error** → less overfitting. ElasticNet behaves similarly.

### WHEN NOT to use
- Small numbers of clearly useful, uncorrelated features → plain linear regression is fine.
- Lasso with groups of highly correlated features → it picks one somewhat arbitrarily (ElasticNet is more stable).
- Choose α with **cross-validation** (`RidgeCV`, `LassoCV`, or `GridSearchCV`), not by guessing.

# 11. Logistic Regression

### What it is
Despite its name, logistic regression is a **classification** algorithm. It predicts the **probability** that a row belongs to class 1.

1. Compute a linear score (the **log-odds**): z = b₀ + b₁x₁ + … + bₚxₚ
2. Squash it into a probability with the **sigmoid**: P(y = 1) = 1 / (1 + e^(−z))  (Topic 15)
3. Predict class 1 if P ≥ threshold (default 0.5).

### Decision boundary
The set of points where P = 0.5 (i.e. z = 0). For logistic regression it is a **straight line** (plane) in feature space.

### Log-odds interpretation
Each coefficient is the change in **log-odds** for +1 unit of the feature. `exp(coef)` is the **odds ratio**: e.g. exp(0.7) ≈ 2 → the odds of class 1 **double** per unit increase.

### WHY / WHEN / WHEN NOT
- ✅ Strong **baseline** for classification, fast, gives probabilities (usually well calibrated), interpretable.
- ✅ Scale features (it uses regularization by default: `C` = inverse of regularization strength).
- ❌ Complex non-linear boundaries (use trees/boosting/SVM-RBF or engineer features).

````python
rl = np.random.default_rng(11)
n = 400
study = pd.DataFrame({'Hours_Studied': rl.uniform(0, 10, n), 'Previous_Score': rl.uniform(30, 100, n)})
logit = -9 + 0.9 * study['Hours_Studied'] + 0.08 * study['Previous_Score']
study['Passed'] = (rl.random(n) < 1 / (1 + np.exp(-logit))).astype(int)

Xl = study[['Hours_Studied', 'Previous_Score']]; yl = study['Passed']
Xl_tr, Xl_te, yl_tr, yl_te = train_test_split(Xl, yl, test_size=0.25, stratify=yl, random_state=RANDOM_STATE)

log_model = LogisticRegression()          # scaling isn't essential for 2 features on similar scales, but see pipelines later
log_model.fit(Xl_tr, yl_tr)

proba = log_model.predict_proba(Xl_te)     # shape (n_rows, 2): column 0 = P(fail), column 1 = P(pass)
print('predict_proba for 5 students:\n', np.round(proba[:5], 3))
print('predict (threshold 0.5):', log_model.predict(Xl_te)[:5])
print('\nIntercept:', round(log_model.intercept_[0], 3))
for f, c in zip(Xl.columns, log_model.coef_[0]):
    print(f'{f:<15} coef = {c:.3f} | odds ratio = {np.exp(c):.3f}')
print('\nTest accuracy:', round(accuracy_score(yl_te, log_model.predict(Xl_te)), 3))
````

````python
# Decision boundary & probability surface
xx, yy = np.meshgrid(np.linspace(0, 10, 200), np.linspace(30, 100, 200))
grid = pd.DataFrame({'Hours_Studied': xx.ravel(), 'Previous_Score': yy.ravel()})
zz = log_model.predict_proba(grid)[:, 1].reshape(xx.shape)

plt.figure(figsize=(8, 5.5))
cs = plt.contourf(xx, yy, zz, levels=20, cmap='RdYlGn', alpha=0.6); plt.colorbar(cs, label='P(pass)')
plt.contour(xx, yy, zz, levels=[0.5], colors='black', linewidths=2)
plt.scatter(Xl_te['Hours_Studied'], Xl_te['Previous_Score'], c=yl_te, cmap='RdYlGn', edgecolor='k', s=30)
plt.xlabel('Hours studied'); plt.ylabel('Previous score')
plt.title('Logistic regression: probabilities and the P = 0.5 decision boundary (black line)'); plt.show()
````

### Interpreting `predict_proba`
- Each row sums to 1: `[P(class 0), P(class 1)]`.
- `predict()` = "class 1 if column 1 ≥ 0.5".
- Odds ratio for Hours_Studied ≈ e^0.9 ≈ 2.5 → each extra study hour multiplies the **odds** of passing by about 2.5 (holding previous score fixed).

### Common mistakes
- Treating coefficients as changes in **probability** (they are changes in **log-odds**).
- Forgetting to scale when features have very different units (regularization then treats them unequally).
- Always using threshold 0.5 (Section 12).

# 12. Classification Metrics

### Confusion matrix (positive class = "fraud", "churn", "disease" — the thing we want to catch)

|  | Predicted **negative** | Predicted **positive** |
|---|---|---|
| **Actual negative** | TN (true negative) | FP (false positive — false alarm, Type I error) |
| **Actual positive** | FN (false negative — missed case, Type II error) | TP (true positive) |

| Metric | Formula | Question it answers | Care about it when |
|---|---|---|---|
| **Accuracy** | (TP + TN) / all | How often is the model right overall? | Classes are balanced and errors cost the same |
| **Precision** | TP / (TP + FP) | When it says "positive", how often is it right? | False alarms are expensive (spam filter, costly interventions) |
| **Recall** (sensitivity, TPR) | TP / (TP + FN) | Of all real positives, how many did we catch? | Missing a case is expensive (fraud, disease, churn) |
| **Specificity** (TNR) | TN / (TN + FP) | Of all real negatives, how many did we correctly clear? | Medical tests; false alarms on healthy people |
| **F1-score** | 2·P·R / (P + R) | Balance of precision and recall (harmonic mean) | You need one number and both errors matter |
| **ROC-AUC** | Area under TPR vs FPR curve | How well does the model **rank** positives above negatives, across all thresholds? (0.5 = random, 1 = perfect) | Comparing models' ranking ability, fairly balanced data |
| **PR-AUC** (average precision) | Area under precision vs recall curve | Ranking quality focused on the **positive** class | **Imbalanced** data — more honest than ROC-AUC there |

### ⚠️ When accuracy misleads
With 2% fraud, a model that says **"never fraud"** has **98% accuracy** and catches **zero** fraud.

````python
from sklearn.datasets import make_classification

X_imb, y_imb = make_classification(n_samples=5000, n_features=10, n_informative=5, weights=[0.95, 0.05],
                                   flip_y=0.01, random_state=RANDOM_STATE)
Xi_tr, Xi_te, yi_tr, yi_te = train_test_split(X_imb, y_imb, test_size=0.3, stratify=y_imb, random_state=RANDOM_STATE)
print('Share of positives:', y_imb.mean().round(3))

dummy = DummyClassifier(strategy='most_frequent').fit(Xi_tr, yi_tr)
clf = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(Xi_tr, yi_tr)

def classification_summary(y_true, y_pred, y_score=None):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    out = {'Accuracy': accuracy_score(y_true, y_pred),
           'Precision': precision_score(y_true, y_pred, zero_division=0),
           'Recall': recall_score(y_true, y_pred),
           'Specificity': tn / (tn + fp),
           'F1': f1_score(y_true, y_pred, zero_division=0)}
    if y_score is not None:
        out['ROC-AUC'] = roc_auc_score(y_true, y_score)
        out['PR-AUC'] = average_precision_score(y_true, y_score)
    return out

summary = pd.DataFrame({
    'Always "negative" (dummy)': classification_summary(yi_te, dummy.predict(Xi_te), dummy.predict_proba(Xi_te)[:, 1]),
    'Logistic regression': classification_summary(yi_te, clf.predict(Xi_te), clf.predict_proba(Xi_te)[:, 1]),
}).round(3)
summary
````

````python
fig, axes = plt.subplots(1, 3, figsize=(18, 4.6))
ConfusionMatrixDisplay.from_predictions(yi_te, clf.predict(Xi_te), ax=axes[0], colorbar=False)
axes[0].set_title('Confusion matrix (threshold 0.5)')
RocCurveDisplay.from_estimator(clf, Xi_te, yi_te, ax=axes[1]); axes[1].plot([0, 1], [0, 1], 'k--')
axes[1].set_title('ROC curve')
PrecisionRecallDisplay.from_estimator(clf, Xi_te, yi_te, ax=axes[2])
axes[2].axhline(yi_te.mean(), color='k', ls='--', label='random = positive rate'); axes[2].legend()
axes[2].set_title('Precision–Recall curve')
plt.tight_layout(); plt.show()

print(classification_report(yi_te, clf.predict(Xi_te), digits=3))
````

````python
# Changing the threshold trades precision for recall
scores = clf.predict_proba(Xi_te)[:, 1]
rows = []
for t in [0.1, 0.2, 0.3, 0.5, 0.7]:
    rows.append({'threshold': t, **classification_summary(yi_te, (scores >= t).astype(int))})
pd.DataFrame(rows).round(3)
````

### Interpretation
- The dummy model has high **accuracy** but **zero recall** — it's useless for catching positives.
- The logistic model's **ROC-AUC** looks high even on imbalanced data; **PR-AUC** is lower and more honest about how hard it is to find the rare class (its random baseline is the positive rate ≈ 0.05, not 0.5).
- Lowering the **threshold** increases recall but lowers precision. Pick the threshold from **business costs**, using validation data.

### Reading `classification_report`
It shows precision, recall and F1 **per class**, plus averages: *macro* (unweighted mean of classes) and *weighted* (weighted by class size).

### Common mistakes
- Reporting only accuracy on imbalanced data.
- Choosing the threshold on the **test** set (use validation data).
- Comparing ROC-AUC of models evaluated on different test sets.

# 13. K-Nearest Neighbors (KNN)

### Intuition
> "Tell me who your neighbours are, and I'll tell you who you are."

To predict a new point, KNN finds the **K most similar** training points (nearest by **distance**) and:
- **Classification:** takes a **majority vote** of their classes.
- **Regression:** takes the **average** of their target values.

There is no real "training" — KNN just **stores** the data (a *lazy learner*). All work happens at prediction time.

### Distance
**Euclidean** (straight-line): d = √((x₁ − x₂)² + (y₁ − y₂)²). Also Manhattan, Minkowski.

### Why scaling matters ⚠️
If one feature is Income (0–1,000,000) and another is Age (18–70), the distance is dominated **entirely** by income; age is ignored. Always **standardize** before KNN.

### Choosing K
| K | Behaviour |
|---|---|
| Small (1–3) | Very flexible, sensitive to noise → **overfitting** (high variance) |
| Large (50+) | Very smooth, ignores local patterns → **underfitting** (high bias) |
Choose K with cross-validation; for binary classification odd K avoids ties.

### WHY / WHEN / WHEN NOT
- ✅ Simple, no assumptions about the shape of the boundary, good for small/medium datasets with meaningful distances.
- ❌ Slow predictions on big data, poor with many features (**curse of dimensionality**: in high dimensions, everything is "far"), needs scaling, sensitive to irrelevant features.

````python
rk = np.random.default_rng(13)
n = 400
people = pd.DataFrame({'Age': rk.uniform(18, 70, n), 'Income': rk.uniform(20_000, 200_000, n)})
people['Buys'] = ((people['Age'] > 40) ^ (rk.random(n) < 0.1)).astype(int)   # depends on AGE only (+10% noise)
Xk = people[['Age', 'Income']]; yk = people['Buys']
Xk_tr, Xk_te, yk_tr, yk_te = train_test_split(Xk, yk, test_size=0.3, stratify=yk, random_state=RANDOM_STATE)

unscaled = KNeighborsClassifier(n_neighbors=7).fit(Xk_tr, yk_tr)
scaled = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=7)).fit(Xk_tr, yk_tr)
print(f'KNN WITHOUT scaling: accuracy = {accuracy_score(yk_te, unscaled.predict(Xk_te)):.3f}  (distance dominated by Income)')
print(f'KNN WITH scaling   : accuracy = {accuracy_score(yk_te, scaled.predict(Xk_te)):.3f}')
````

````python
# Choosing K with cross-validation
ks = list(range(1, 60, 2))
cv_acc = [cross_val_score(make_pipeline(StandardScaler(), KNeighborsClassifier(k)), Xk_tr, yk_tr, cv=5).mean() for k in ks]
train_acc = [make_pipeline(StandardScaler(), KNeighborsClassifier(k)).fit(Xk_tr, yk_tr).score(Xk_tr, yk_tr) for k in ks]

plt.figure(figsize=(8, 4))
plt.plot(ks, train_acc, 'o-', label='training accuracy'); plt.plot(ks, cv_acc, 's-', label='5-fold CV accuracy')
best_k = ks[int(np.argmax(cv_acc))]; plt.axvline(best_k, color='red', ls='--', label=f'best K = {best_k}')
plt.xlabel('K'); plt.ylabel('Accuracy'); plt.title('Small K overfits (train ≫ CV); large K smooths'); plt.legend(); plt.show()

# KNN regression
knn_reg = make_pipeline(StandardScaler(), KNeighborsRegressor(n_neighbors=10)).fit(Xp_tr, yp_tr)
print(f'KNeighborsRegressor on the cubic data: test RMSE = {rmse(yp_te, knn_reg.predict(Xp_te)):.3f}')
````

### Interpretation
- Without scaling, KNN measures distance almost only by income (which is irrelevant here) → poor accuracy. Scaling fixes it.
- K = 1 memorises the training set (100% training accuracy) but generalises worse; CV picks a K in between.

# 14. Naive Bayes

### Connection to Bayes' theorem (Topic 15)
$$P(\text{class} \mid \text{features}) \propto P(\text{class}) \times P(x_1 \mid \text{class}) \times P(x_2 \mid \text{class}) \times \dots$$

- **Prior** P(class): how common each class is.
- **Likelihood** P(xᵢ | class): how typical each feature value is **within** each class (conditional probability).
- **Naive assumption:** features are **independent given the class**, so we can simply **multiply** their likelihoods. Rarely exactly true, but it often works well anyway.

| Variant | Assumes each feature is… | Good for |
|---|---|---|
| **GaussianNB** | Normally distributed within each class | Continuous numerical features |
| **MultinomialNB** | Counts (e.g. word counts) | **Text** classification (spam, sentiment, topics) |
| BernoulliNB | Binary (present/absent) | Binary features, short texts |

### WHY / WHEN / WHEN NOT
- ✅ Extremely fast, works with little data and many features (text), strong baseline for text.
- ❌ Strongly dependent features, when you need well-calibrated probabilities (NB probabilities are often too extreme), complex interactions.

````python
# GaussianNB on the breast-cancer data (continuous features)
gnb = GaussianNB().fit(X_train, y_train)
print(f'GaussianNB  — breast cancer validation accuracy: {accuracy_score(y_val, gnb.predict(X_val)):.3f}')
print('Class priors learned:', gnb.class_prior_.round(3))

# MultinomialNB on text
from sklearn.feature_extraction.text import CountVectorizer
texts = ['win a free prize now', 'free cash offer click now', 'limited offer win money', 'claim your free gift',
         'cheap loans no credit check', 'you won the lottery claim now',
         'meeting at 10 tomorrow', 'please review the attached report', 'lunch with the team today',
         'project deadline moved to friday', 'can you send the invoice', 'notes from the client call']
labels = [1] * 6 + [0] * 6
text_model = make_pipeline(CountVectorizer(), MultinomialNB(alpha=1.0)).fit(texts, labels)

for msg in ['free money for the team', 'send the report before the meeting', 'claim your prize now']:
    print(f'P(spam) = {text_model.predict_proba([msg])[0, 1]:.3f}  ←  "{msg}"')
````

### Interpretation
- GaussianNB reaches good accuracy on the cancer data with **zero** tuning, in milliseconds.
- MultinomialNB turns word counts into spam probabilities; mixed messages land in between.

# 15. Decision Trees

### Intuition
A decision tree is a flowchart of **yes/no questions** learned from data — like the game "20 questions".

```text
                    [Root] Contract == Monthly?
                    /                         \
                  yes                          no
                  /                              \
      [Node] SupportCalls > 3?              [Leaf] predict: STAY
          /             \
        yes              no
        /                  \
[Leaf] predict: CHURN   [Leaf] predict: STAY
```

| Term | Meaning |
|---|---|
| **Root** | The first (top) question |
| **Node** (internal node) | Any question further down |
| **Split** | Dividing the data into two groups by a question |
| **Decision rule** | The question, e.g. `SupportCalls <= 3.5` |
| **Leaf** | A final box with a prediction (majority class or average value) |
| **Depth** | Number of questions from root to the deepest leaf |

### How does the tree choose a split? — **impurity**
A node is **pure** if all its samples have the same class. The tree tries every feature and threshold and picks the split that makes the children **as pure as possible**.

- **Gini impurity** = 1 − Σ pₖ² (0 = pure; 0.5 = 50/50 for two classes)
- **Entropy** = −Σ pₖ log₂ pₖ (0 = pure; 1 = 50/50 for two classes)
- **Information gain** = impurity(parent) − weighted average impurity(children). The best split has the **largest gain**.
- Regression trees use **variance/MSE** instead of impurity.

### Numerical example
Parent: 10 customers, 5 churn / 5 stay → Gini = 1 − (0.5² + 0.5²) = **0.5**
Split by "Monthly contract?":
- Left (6 rows): 5 churn, 1 stay → Gini = 1 − ((5/6)² + (1/6)²) = **0.278**
- Right (4 rows): 0 churn, 4 stay → Gini = **0** (pure)
- Weighted child Gini = 0.6 × 0.278 + 0.4 × 0 = 0.167 → **gain = 0.5 − 0.167 = 0.333**

### WHY / WHEN / WHEN NOT
- ✅ Very **interpretable** (rules), handles non-linear patterns & interactions, **no scaling needed**, handles mixed feature types.
- ❌ A single deep tree **overfits badly** (high variance) and is unstable (small data changes → different tree). Control with `max_depth`, `min_samples_leaf`, `min_samples_split`, or use ensembles (Sections 16–17).

````python
def gini(counts):
    p = np.array(counts) / np.sum(counts)
    return 1 - np.sum(p ** 2)

def entropy(counts):
    p = np.array(counts) / np.sum(counts)
    p = p[p > 0]
    return float(-np.sum(p * np.log2(p))) + 0.0      # + 0.0 turns -0.0 into 0.0 for tidy printing

parent, left, right = [5, 5], [5, 1], [0, 4]
w_left, w_right = 6 / 10, 4 / 10
print(f'Gini   parent = {gini(parent):.3f} | left = {gini(left):.3f} | right = {gini(right):.3f} '
      f'| information gain = {gini(parent) - (w_left * gini(left) + w_right * gini(right)):.3f}')
print(f'Entropy parent = {entropy(parent):.3f} | left = {entropy(left):.3f} | right = {entropy(right):.3f} '
      f'| information gain = {entropy(parent) - (w_left * entropy(left) + w_right * entropy(right)):.3f}')
````

````python
iris_X, iris_y = iris.data, iris.target
Xir_tr, Xir_te, yir_tr, yir_te = train_test_split(iris_X, iris_y, test_size=0.3, stratify=iris_y, random_state=RANDOM_STATE)

tree_clf = DecisionTreeClassifier(max_depth=3, criterion='gini', random_state=RANDOM_STATE).fit(Xir_tr, yir_tr)
print('Iris test accuracy (depth 3):', round(tree_clf.score(Xir_te, yir_te), 3))
print(export_text(tree_clf, feature_names=list(iris_X.columns)))

plt.figure(figsize=(16, 7))
plot_tree(tree_clf, feature_names=iris_X.columns, class_names=iris.target_names, filled=True, rounded=True, fontsize=9)
plt.title('Decision tree for iris species (each box: rule, gini, samples, class counts, prediction)'); plt.show()
````

````python
# Depth controls over/underfitting
rows = []
for depth in [1, 2, 3, 5, 8, None]:
    t = DecisionTreeClassifier(max_depth=depth, random_state=RANDOM_STATE).fit(X_train, y_train)
    rows.append({'max_depth': str(depth), 'train acc': t.score(X_train, y_train), 'validation acc': t.score(X_val, y_val),
                 'leaves': t.get_n_leaves()})
print('Breast cancer data:'); print(pd.DataFrame(rows).round(3).to_string(index=False))

# Regression tree: a step-function fit
tree_reg = DecisionTreeRegressor(max_depth=3, random_state=RANDOM_STATE).fit(Xp_tr, yp_tr)
plt.figure(figsize=(7, 3.5)); plt.scatter(Xp_tr, yp_tr, s=15); plt.plot(x_plot, tree_reg.predict(x_plot), 'r')
plt.title(f'DecisionTreeRegressor (depth 3): predictions are steps | test RMSE {rmse(yp_te, tree_reg.predict(Xp_te)):.2f}'); plt.show()
````

### Interpretation
- The tree's top rule uses petal measurements — the most informative split for iris.
- An unlimited-depth tree reaches **100% training accuracy** (memorisation) but not better validation accuracy → overfitting. `max_depth` is the main control.
- Regression trees predict the **average** of each leaf, so the fit looks like **steps**.

# 16. Random Forest

### Intuition — the wisdom of crowds
One decision tree is like asking **one** expert with strong opinions: often wrong in odd ways. A **random forest** asks **hundreds of different trees** and averages their answers. Their individual mistakes **cancel out**.

### How it builds diverse trees — **bagging** + random features
1. **Bootstrap sample:** each tree trains on a random sample of rows **drawn with replacement** (same size as the data; about 63% unique rows, the rest are duplicates).
2. **Random feature selection:** at each split, a tree may only choose among a **random subset of features** (`max_features`, e.g. √p for classification). This stops all trees from using the same strong feature at the top.
3. **Aggregate:** classification → majority vote (or average probability); regression → average prediction.

**Bagging** = **B**ootstrap **agg**regat**ing**. It mainly reduces **variance** (overfitting) of deep trees.

**Out-of-bag (OOB) score:** each tree didn't see ~37% of rows; predicting those gives a free validation estimate (`oob_score=True`).

### Feature importance
- **Impurity-based** (`feature_importances_`): total impurity decrease from splits on each feature. Fast, but biased toward high-cardinality/continuous features and computed on **training** data.
- **Permutation importance** (Section 30): drop in validation score when a feature is shuffled — more reliable.

### WHY / WHEN / WHEN NOT
- ✅ Strong default for tabular data, little tuning, **no scaling**, robust to outliers, captures non-linearity/interactions.
- ❌ Less interpretable than one tree, bigger/slower models, **can't extrapolate** beyond the training target range (regression), usually beaten slightly by well-tuned gradient boosting.
- Key hyperparameters: `n_estimators`, `max_depth`, `max_features`, `min_samples_leaf`.

````python
rf = RandomForestClassifier(n_estimators=300, max_features='sqrt', oob_score=True, n_jobs=-1, random_state=RANDOM_STATE)
rf.fit(X_train, y_train)
print(f'Random forest OOB accuracy: {rf.oob_score_:.3f}  (free estimate from out-of-bag rows)')

# A single validation split is noisy (114 rows); 10-fold cross-validation gives a fairer comparison
cv10 = StratifiedKFold(10, shuffle=True, random_state=RANDOM_STATE)
for name, m in [('Single decision tree', DecisionTreeClassifier(random_state=RANDOM_STATE)),
                ('Random forest (300 trees)', RandomForestClassifier(n_estimators=300, n_jobs=-1, random_state=RANDOM_STATE))]:
    s = cross_val_score(m, Xc, yc, cv=cv10)
    print(f'{name:<26} 10-fold CV accuracy = {s.mean():.3f} ± {s.std():.3f}')

importances = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
importances.head(10).sort_values().plot(kind='barh', figsize=(7, 4), title='Random forest: top-10 impurity importances')
plt.show()
````

````python
# Bootstrap sample illustration
idx = rng.integers(0, 10, 10)
print('Rows drawn for one tree (with replacement):', sorted(idx.tolist()), '| unique rows:', len(set(idx.tolist())), 'of 10')

# RandomForestRegressor vs a single tree on the cubic data
rf_reg = RandomForestRegressor(n_estimators=300, min_samples_leaf=3, random_state=RANDOM_STATE).fit(Xp_tr, yp_tr)
plt.figure(figsize=(7, 3.5)); plt.scatter(Xp_tr, yp_tr, s=15)
plt.plot(x_plot, DecisionTreeRegressor(random_state=0).fit(Xp_tr, yp_tr).predict(x_plot), 'gray', alpha=0.6, label='one deep tree')
plt.plot(x_plot, rf_reg.predict(x_plot), 'r', label='random forest'); plt.legend()
plt.title(f'Averaging many trees smooths the fit | RF test RMSE {rmse(yp_te, rf_reg.predict(Xp_te)):.2f}'); plt.show()
````

# 17. Gradient Boosting

### Intuition — learning from mistakes
Boosting builds trees **one after another**. Each new small tree focuses on the **errors** the previous trees made.

```text
Step 1: tree 1 predicts           → residual errors remain
Step 2: tree 2 predicts the ERRORS of tree 1 → add a small part of it (learning rate)
Step 3: tree 3 predicts the remaining errors …
Final prediction = start + lr·tree1 + lr·tree2 + lr·tree3 + …
```

"**Gradient**" because each tree follows the gradient (direction of steepest decrease) of the loss function — for squared error that's simply the residuals.

### Bagging vs Boosting

| | **Bagging** (Random Forest) | **Boosting** (Gradient Boosting) |
|---|---|---|
| Trees built | **In parallel**, independently | **Sequentially**, each fixes the previous |
| Tree size | Deep trees (low bias, high variance) | Shallow trees (high bias, low variance) |
| Mainly reduces | **Variance** | **Bias** |
| Overfitting risk | Low; more trees rarely hurt | Higher; too many trees / high learning rate overfit |
| Tuning needed | Little | More (learning rate, n_estimators, depth) |
| Typical accuracy on tabular data | Very good | Often the best |

### scikit-learn versions
- `GradientBoostingClassifier / Regressor`: classic, accurate, slower on large data.
- `HistGradientBoostingClassifier / Regressor`: bins features into histograms → **much faster** on large data, handles **missing values natively**, supports categorical features. (Similar in spirit to LightGBM/XGBoost.)

### Key hyperparameters
`n_estimators` (number of trees), `learning_rate` (step size; smaller = needs more trees but generalises better), `max_depth` (tree size, often 2–5), `subsample` (<1 adds randomness), and for Hist-GB `max_iter`, `early_stopping`.

### WHY / WHEN / WHEN NOT
- ✅ Often the most accurate model for **tabular** data; handles non-linearity, interactions; no scaling needed.
- ❌ Needs tuning; can overfit noisy small datasets; less interpretable; sequential training is slower (classic version).

````python
gb = GradientBoostingClassifier(n_estimators=200, learning_rate=0.1, max_depth=3, random_state=RANDOM_STATE).fit(X_train, y_train)
hgb = HistGradientBoostingClassifier(max_iter=200, learning_rate=0.1, random_state=RANDOM_STATE).fit(X_train, y_train)
print(f'GradientBoostingClassifier     validation accuracy: {gb.score(X_val, y_val):.3f}')
print(f'HistGradientBoostingClassifier validation accuracy: {hgb.score(X_val, y_val):.3f}')

# Watching boosting learn step by step (regression on the cubic data)
gbr = GradientBoostingRegressor(n_estimators=300, learning_rate=0.05, max_depth=2, random_state=RANDOM_STATE).fit(Xp_tr, yp_tr)
train_curve = [rmse(yp_tr, p) for p in gbr.staged_predict(Xp_tr)]      # predictions after 1, 2, 3, … trees
test_curve = [rmse(yp_te, p) for p in gbr.staged_predict(Xp_te)]

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(train_curve, label='train RMSE'); axes[0].plot(test_curve, label='test RMSE')
axes[0].set_xlabel('number of trees'); axes[0].set_title('Boosting: each tree reduces the remaining error'); axes[0].legend()
axes[1].scatter(Xp_tr, yp_tr, s=15)
for n_trees, c in [(5, 'lightgray'), (30, 'orange'), (300, 'red')]:
    stage_pred = list(gbr.staged_predict(x_plot))[n_trees - 1]
    axes[1].plot(x_plot, stage_pred, color=c, label=f'after {n_trees} trees')
axes[1].legend(); axes[1].set_title('The fit improves as trees are added')
plt.tight_layout(); plt.show()
````

# 18. Support Vector Machines (SVM)

### Intuition — the widest street
For two classes, many lines can separate them. SVM picks the line (in higher dimensions: a **hyperplane**) with the **widest margin** — the widest "street" between the classes. A wide street is more likely to classify new points correctly.

| Term | Meaning |
|---|---|
| **Hyperplane** | The decision boundary (a line in 2-D, a plane in 3-D, …) |
| **Margin** | Distance between the boundary and the closest points of each class |
| **Support vectors** | The training points **on or inside** the margin — they alone define the boundary |
| **C** | Penalty for points inside the margin/misclassified. Large C → narrow margin, fits training data tightly (overfit risk); small C → wider, softer margin |

### Kernels — curved boundaries
When classes aren't separable by a straight line, a **kernel** implicitly maps the data into a higher-dimensional space where a straight separator exists — without computing that space explicitly (the "kernel trick").
- `kernel='linear'` → straight boundary.
- `kernel='rbf'` (Radial Basis Function, default) → flexible curved boundaries; `gamma` controls how far each point's influence reaches (large gamma → wiggly, overfit risk).
- `kernel='poly'` → polynomial boundaries.

### Why scaling is important ⚠️
SVM uses **distances** (margins, RBF similarity). Unscaled features with large ranges dominate. Always standardize.

### SVR (regression)
Fits a "tube" of width ε around the data; only points outside the tube are penalised.

### WHY / WHEN / WHEN NOT
- ✅ Strong on small-to-medium datasets with clear margins, high-dimensional data (text), flexible via kernels.
- ❌ Slow on large datasets (training time grows roughly with n² or worse), needs scaling and tuning (C, gamma), no native probabilities (`probability=True` adds cost), less interpretable.

````python
from sklearn.datasets import make_moons

Xm, ym = make_moons(n_samples=400, noise=0.2, random_state=RANDOM_STATE)
Xm_tr, Xm_te, ym_tr, ym_te = train_test_split(Xm, ym, test_size=0.3, random_state=RANDOM_STATE)

svms = {
    'Linear SVM (C=1)': make_pipeline(StandardScaler(), SVC(kernel='linear', C=1)),
    'RBF SVM (C=1, gamma=scale)': make_pipeline(StandardScaler(), SVC(kernel='rbf', C=1)),
    'RBF SVM (C=100, gamma=10)': make_pipeline(StandardScaler(), SVC(kernel='rbf', C=100, gamma=10)),
}
xx, yy = np.meshgrid(np.linspace(-2, 3, 300), np.linspace(-1.5, 2, 300))
fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))
for ax, (name, m) in zip(axes, svms.items()):
    m.fit(Xm_tr, ym_tr)
    Z = m.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z > 0, alpha=0.25, cmap='coolwarm')
    ax.contour(xx, yy, Z, levels=[-1, 0, 1], colors='k', linestyles=['--', '-', '--'])   # margin lines at ±1
    ax.scatter(Xm_tr[:, 0], Xm_tr[:, 1], c=ym_tr, cmap='coolwarm', s=12, edgecolor='k', linewidth=0.3)
    sv = m[-1].support_vectors_
    ax.set_title(f'{name}\ntrain {m.score(Xm_tr, ym_tr):.2f} | test {m.score(Xm_te, ym_te):.2f} | {len(sv)} support vectors')
plt.tight_layout(); plt.show()
````

````python
# Scaling matters: breast cancer features have very different ranges
raw_svc = SVC().fit(X_train, y_train)
scaled_svc = make_pipeline(StandardScaler(), SVC()).fit(X_train, y_train)
print(f'SVC without scaling: validation accuracy = {raw_svc.score(X_val, y_val):.3f}')
print(f'SVC with scaling   : validation accuracy = {scaled_svc.score(X_val, y_val):.3f}')

svr = make_pipeline(StandardScaler(), SVR(kernel='rbf', C=20, epsilon=0.5)).fit(Xp_tr, yp_tr)
print(f'SVR on the cubic data: test RMSE = {rmse(yp_te, svr.predict(Xp_te)):.3f}')
````

### Interpretation
- The **linear** SVM can't follow the moon shapes. **RBF** with moderate settings fits the curve well.
- RBF with large C and gamma draws tiny islands around training points → high training accuracy, lower test accuracy → **overfitting**.
- Scaling makes a large difference on real data with mixed feature ranges.

# 19. Unsupervised Learning

### What is different?
There is **no target variable** — no "right answer" to learn from. The goal is to **find hidden structure** in the features themselves.

| Task | Goal | Examples | Algorithms |
|---|---|---|---|
| **Clustering** | Group similar rows together | Customer segments, grouping documents, grouping similar products | K-Means, hierarchical, DBSCAN, Gaussian mixtures |
| **Dimensionality reduction** | Compress many features into a few while keeping most information | Visualising high-dimensional data, speeding up models, removing noise/multicollinearity | PCA, t-SNE, UMAP |
| **Anomaly detection** | Find unusual rows | Fraud, faulty sensors | Isolation Forest, LOF |
| **Association rules** | Items that occur together | "Customers who buy bread also buy butter" | Apriori |

### Challenges
- **No ground truth** to evaluate against → use internal metrics (inertia, silhouette) **and business sense**.
- Results depend heavily on **scaling**, feature choice and hyperparameters (e.g. number of clusters).
- Clusters must be **interpreted** by humans to become useful.

# 20. K-Means Clustering

### Algorithm (step by step)
1. Choose **K** (number of clusters).
2. **Initialise** K centroids (cluster centres) — scikit-learn uses the smart `k-means++` method.
3. **Assignment:** give every point to its **nearest** centroid.
4. **Update:** move each centroid to the **mean** of the points assigned to it.
5. **Iterate** steps 3–4 until assignments stop changing.

It minimises **inertia** = the sum of squared distances from each point to its centroid (how tight the clusters are).

### Choosing K
- **Elbow method:** plot inertia vs K. Inertia always decreases; look for the "elbow" where adding clusters stops helping much.
- **Silhouette score** (−1 to 1): for each point, compares its distance to its own cluster vs the nearest other cluster. Higher = better-separated clusters. Pick the K with the highest average silhouette (and one that makes business sense).

### WHY / WHEN / WHEN NOT
- ✅ Fast, simple, scales to large data, works well for roughly **round, similar-sized** clusters.
- ❌ You must choose K; sensitive to **scaling** and outliers; assumes spherical clusters (fails on moons/rings/very different densities); results depend on initialisation (`n_init`).

````python
X_blobs, y_blobs_true = make_blobs(n_samples=500, centers=4, cluster_std=[1.0, 1.5, 0.8, 1.2], random_state=7)
X_blobs = StandardScaler().fit_transform(X_blobs)

# Watch the algorithm: centroids after 1, 2 and final iterations
fig, axes = plt.subplots(1, 3, figsize=(17, 4.2))
init = X_blobs[[0, 1, 2, 3]]                                   # deliberately poor starting centroids
for ax, iters in zip(axes, [1, 2, 50]):
    km = KMeans(n_clusters=4, init=init, n_init=1, max_iter=iters, random_state=0).fit(X_blobs)
    ax.scatter(X_blobs[:, 0], X_blobs[:, 1], c=km.labels_, cmap='tab10', s=10)
    ax.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1], c='black', marker='X', s=200)
    ax.set_title(f'After {iters} iteration(s): inertia = {km.inertia_:.1f}')
plt.tight_layout(); plt.show()
````

````python
ks = range(2, 10)
inertias, silhouettes = [], []
for k in ks:
    km = KMeans(n_clusters=k, n_init=10, random_state=RANDOM_STATE).fit(X_blobs)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_blobs, km.labels_))

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].plot(list(ks), inertias, 'o-'); axes[0].set_title('Elbow method'); axes[0].set_xlabel('K'); axes[0].set_ylabel('inertia')
axes[1].plot(list(ks), silhouettes, 's-', color='tab:green'); axes[1].set_title('Silhouette score'); axes[1].set_xlabel('K')
plt.tight_layout(); plt.show()
print('Best K by silhouette:', list(ks)[int(np.argmax(silhouettes))])
````

````python
# K-Means fails on non-round shapes
Xmoon, _ = make_moons(n_samples=400, noise=0.06, random_state=0)
km_moon = KMeans(n_clusters=2, n_init=10, random_state=0).fit(Xmoon)
plt.figure(figsize=(6, 3.5)); plt.scatter(Xmoon[:, 0], Xmoon[:, 1], c=km_moon.labels_, cmap='coolwarm', s=12)
plt.title('K-Means cuts the moons with a straight boundary (wrong grouping)'); plt.show()
````

# 21. Hierarchical Clustering

### Agglomerative (bottom-up) clustering
1. Start with **every point as its own cluster**.
2. **Merge** the two closest clusters.
3. Repeat until everything is one cluster.
4. The history of merges forms a tree called a **dendrogram**. **Cut** it at a chosen height to get the number of clusters.

### Linkage — how is the distance between two **clusters** measured?
| Linkage | Distance between clusters | Behaviour |
|---|---|---|
| **Ward** (default) | Increase in within-cluster variance after merging | Compact, similar-sized clusters (like K-Means) |
| **Complete** | Farthest pair of points | Compact clusters |
| **Average** | Average of all pairs | In-between |
| **Single** | Closest pair | Can follow chains / odd shapes, but can create "chaining" |

### Choosing the number of clusters
Look for a **big vertical gap** in the dendrogram (merges that join very different groups) and cut there; confirm with silhouette and business sense.

### WHY / WHEN / WHEN NOT
- ✅ No need to fix K in advance; the dendrogram shows structure at every level; can capture non-spherical shapes (single linkage).
- ❌ Slow on large datasets (memory grows with n²); merges can't be undone; sensitive to scaling.

````python
from scipy.cluster.hierarchy import linkage, dendrogram

sample = X_blobs[rng.choice(len(X_blobs), 60, replace=False)]
Z = linkage(sample, method='ward')

plt.figure(figsize=(14, 4.5))
dendrogram(Z, no_labels=True, color_threshold=6)
plt.axhline(6, color='red', ls='--', label='cut here → 4 clusters')
plt.title('Dendrogram (Ward linkage)'); plt.ylabel('merge distance'); plt.legend(); plt.show()

rows = []
for linkage_type in ['ward', 'complete', 'average', 'single']:
    agg = AgglomerativeClustering(n_clusters=4, linkage=linkage_type).fit(X_blobs)
    rows.append({'linkage': linkage_type, 'silhouette': silhouette_score(X_blobs, agg.labels_),
                 'cluster sizes': sorted(np.bincount(agg.labels_).tolist(), reverse=True)})
pd.DataFrame(rows).round(3)
````

````python
agg_moon = AgglomerativeClustering(n_clusters=2, linkage='single').fit(Xmoon)
plt.figure(figsize=(6, 3.5)); plt.scatter(Xmoon[:, 0], Xmoon[:, 1], c=agg_moon.labels_, cmap='coolwarm', s=12)
plt.title('Single-linkage hierarchical clustering follows the moon shapes'); plt.show()
````

# 22. PCA — Principal Component Analysis

### Intuition
Imagine photographing a 3-D object. Some camera angles show its shape clearly; others hide it. PCA finds the **best "camera angles"** for your data: new axes (**principal components**) along which the data **varies the most**.

- **PC1** = direction of **maximum variance**.
- **PC2** = the next largest variance, **perpendicular** (uncorrelated) to PC1, and so on.
- Each PC is a **weighted combination** of the original features; the weights are called **loadings**.
- **Explained variance ratio** = share of the total information (variance) each PC keeps.

### Uses
Visualising high-dimensional data in 2-D; compressing features; removing multicollinearity (PCs are uncorrelated — Topic 12); denoising.

### ⚠️ Scale first — PCA chases variance, so large-unit features would dominate.

### Feature selection vs PCA

| | **Feature selection** (Section 29) | **PCA** (feature extraction) |
|---|---|---|
| Output | A **subset** of the original features | **New** features (combinations of all originals) |
| Interpretability | High — "we kept Age and Income" | Low — "PC1 = 0.4·Age + 0.6·Income − …" |
| Uses the target? | Often yes | **No** (unsupervised) — may drop information useful for prediction |
| Removes multicollinearity | Partly | Completely (components are uncorrelated) |

````python
Xcan_scaled = StandardScaler().fit_transform(Xc)          # breast cancer: 30 features
pca_full = PCA().fit(Xcan_scaled)
cum = np.cumsum(pca_full.explained_variance_ratio_)
print('Explained variance ratio of first 5 PCs:', pca_full.explained_variance_ratio_[:5].round(3))
print('Components needed for 90% of the variance:', int(np.argmax(cum >= 0.90)) + 1, 'out of 30')

pca2 = PCA(n_components=2).fit(Xcan_scaled)
X2 = pca2.transform(Xcan_scaled)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].bar(range(1, 31), pca_full.explained_variance_ratio_, alpha=0.6, label='per component')
axes[0].plot(range(1, 31), cum, 'o-', color='red', label='cumulative'); axes[0].axhline(0.9, ls='--', color='gray')
axes[0].set_xlabel('principal component'); axes[0].set_title('Explained variance'); axes[0].legend()
sc = axes[1].scatter(X2[:, 0], X2[:, 1], c=yc, cmap='coolwarm', s=12)
axes[1].set_xlabel(f'PC1 ({pca2.explained_variance_ratio_[0]:.0%})'); axes[1].set_ylabel(f'PC2 ({pca2.explained_variance_ratio_[1]:.0%})')
axes[1].set_title('30 features compressed to 2 (colour = true diagnosis, not used by PCA)')
plt.tight_layout(); plt.show()

loadings = pd.DataFrame(pca2.components_.T, index=Xc.columns, columns=['PC1', 'PC2'])
print('Top loadings on PC1:'); print(loadings['PC1'].abs().sort_values(ascending=False).head(5).round(3))
````

````python
# PCA inside a pipeline: fewer features, similar accuracy
for n_comp in [2, 5, 10, None]:
    steps = [StandardScaler()] + ([PCA(n_components=n_comp)] if n_comp else []) + [LogisticRegression(max_iter=2000)]
    score = cross_val_score(make_pipeline(*steps), Xc, yc, cv=5).mean()
    print(f'{"all 30 features" if n_comp is None else f"{n_comp:>2} PCs":<16} → CV accuracy = {score:.3f}')
````

### Interpretation
- About 7 components keep 90% of the information from 30 features.
- In just 2 dimensions, benign and malignant tumours are already mostly separated — even though PCA never saw the labels.
- PC1 has similar-sized loadings on **many** features at once — concavity, concave points, compactness and perimeter — so it acts like an overall **"how abnormal / large is the tumour"** axis. This is typical: a PC is a blend, which is why PCA is less interpretable than feature selection.
- A handful of PCs give almost the same accuracy as all 30 features.

# 23. Overfitting and Underfitting

| | **Underfitting** | **Good fit** | **Overfitting** |
|---|---|---|---|
| Analogy | A student who didn't study — fails practice **and** exam | Understood the ideas — does well on both | Memorised the practice answers — perfect on practice, fails the exam |
| Training error | High | Low | **Very low** |
| Validation/test error | High | Low | **High** (big gap vs training) |
| Cause | Model too simple, too few features, too much regularization | Right complexity | Model too complex, too little data, noisy features, too little regularization |
| Fixes | More complex model, better features, less regularization, train longer | — | More data, simpler model, regularization, pruning/max_depth, fewer features, cross-validation, early stopping, ensembles |

### How to identify them
1. Compare **training vs validation** scores (the gap is the key signal).
2. Plot a **validation curve** (score vs one hyperparameter).
3. Plot a **learning curve** (score vs training-set size):
   - Both curves low and close → **underfitting** (more data won't help; need a better model/features).
   - Big gap that shrinks with more data → **overfitting** (more data helps).

````python
from sklearn.model_selection import validation_curve, learning_curve

depths = np.arange(1, 16)
train_sc, val_sc = validation_curve(DecisionTreeClassifier(random_state=RANDOM_STATE), Xc, yc,
                                    param_name='max_depth', param_range=depths, cv=5, scoring='accuracy')

fig, axes = plt.subplots(1, 3, figsize=(20, 4.3))
axes[0].plot(depths, train_sc.mean(axis=1), 'o-', label='training'); axes[0].plot(depths, val_sc.mean(axis=1), 's-', label='validation (CV)')
axes[0].axvspan(0.5, 2.5, color='orange', alpha=0.15, label='underfit'); axes[0].axvspan(6.5, 15.5, color='red', alpha=0.1, label='overfit')
axes[0].set_xlabel('max_depth'); axes[0].set_ylabel('accuracy'); axes[0].set_title('Validation curve: decision tree depth'); axes[0].legend(fontsize=8)

for ax, (name, est) in zip(axes[1:], [('Logistic regression (simple, well-regularized)', make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))),
                                      ('Unpruned decision tree (complex)', DecisionTreeClassifier(random_state=RANDOM_STATE))]):
    sizes, tr, va = learning_curve(est, Xc, yc, cv=5, train_sizes=np.linspace(0.1, 1.0, 8), scoring='accuracy')
    ax.plot(sizes, tr.mean(axis=1), 'o-', label='training'); ax.plot(sizes, va.mean(axis=1), 's-', label='validation')
    ax.set_ylim(0.85, 1.01); ax.set_xlabel('training set size'); ax.set_title(f'Learning curve: {name}', fontsize=10); ax.legend()
plt.tight_layout(); plt.show()
````

### Interpretation
- **Validation curve:** depth 1–2 → both scores lower (underfitting); after depth ~5 training accuracy hits 100% while validation stops improving (overfitting).
- **Learning curves:** the unpruned tree keeps a **gap** (training 100%, validation lower) — classic overfitting; the logistic regression's curves are close together and high — a good fit.

# 24. Bias–Variance Tradeoff

> **Analogy — archery:**
> **Bias** = your arrows consistently land **off-centre** (a systematic error — the model is too simple to hit the target).
> **Variance** = your arrows are **scattered** (the model changes a lot depending on which training data it saw).

$$\text{Expected test error} = \text{Bias}^2 + \text{Variance} + \text{Irreducible noise}$$

| | Low complexity | High complexity |
|---|---|---|
| Bias | **High** (underfits) | Low |
| Variance | Low | **High** (overfits) |

Increasing complexity lowers bias but raises variance; the best model balances them.

### Connecting to algorithms

| Algorithm | Knob | Low setting → | High setting → |
|---|---|---|---|
| **Linear models** | Features / polynomial degree; regularization α | α large: high bias, low variance | α ≈ 0 with many features: low bias, higher variance |
| **Decision tree** | `max_depth` | Shallow: high bias | Deep: high variance |
| **Random forest** | Number of trees, depth | — | Averaging many deep trees **reduces variance** while keeping low bias |
| **Gradient boosting** | Number of trees, learning rate | Few trees: high bias | Many trees, high lr: high variance — boosting mainly **reduces bias** |
| **KNN** | K | Large K: high bias (smooth) | K = 1: high variance |

### Demonstration — measuring bias² and variance by simulation
We draw **many different training sets** from the same process, fit a model to each, and look at its predictions at fixed test points.
- **Bias²** = (average prediction − truth)²
- **Variance** = how much predictions vary between training sets

````python
def true_f(x):
    return np.sin(1.5 * x) * 3 + 0.3 * x ** 2

x_test_bv = np.linspace(-3, 3, 50).reshape(-1, 1)
models_bv = {
    'Linear regression': lambda: LinearRegression(),
    'Polynomial (deg 4)': lambda: make_pipeline(PolynomialFeatures(4), StandardScaler(), LinearRegression()),
    'KNN (K=30)': lambda: KNeighborsRegressor(30),
    'KNN (K=1)': lambda: KNeighborsRegressor(1),
    'Deep decision tree': lambda: DecisionTreeRegressor(random_state=0),
    'Random forest': lambda: RandomForestRegressor(n_estimators=60, random_state=0, n_jobs=-1),
}
rbv = np.random.default_rng(24)
rows, preds_store = [], {}
for name, make_model in models_bv.items():
    preds = []
    for _ in range(60):                                   # 60 different training sets
        x_tr = rbv.uniform(-3, 3, 80).reshape(-1, 1)
        y_tr = true_f(x_tr.ravel()) + rbv.normal(0, 1.0, 80)
        preds.append(make_model().fit(x_tr, y_tr).predict(x_test_bv))
    preds = np.array(preds)
    preds_store[name] = preds
    bias2 = np.mean((preds.mean(axis=0) - true_f(x_test_bv.ravel())) ** 2)
    variance = np.mean(preds.var(axis=0))
    rows.append({'Model': name, 'Bias²': bias2, 'Variance': variance, 'Bias² + Variance': bias2 + variance})
pd.DataFrame(rows).set_index('Model').round(3)
````

````python
fig, axes = plt.subplots(1, 3, figsize=(18, 4))
for ax, name in zip(axes, ['Linear regression', 'KNN (K=1)', 'Random forest']):
    for p in preds_store[name][:25]:
        ax.plot(x_test_bv, p, color='steelblue', alpha=0.2)
    ax.plot(x_test_bv, preds_store[name].mean(axis=0), color='red', lw=2, label='average prediction')
    ax.plot(x_test_bv, true_f(x_test_bv), 'k--', lw=2, label='truth')
    ax.set_title(name); ax.legend(fontsize=8)
plt.suptitle('Blue = models trained on different datasets', y=1.02); plt.tight_layout(); plt.show()
````

### Interpretation
- **Linear regression:** all the blue lines agree (low variance) but they are all wrong in the same way (high bias).
- **KNN K=1 / deep tree:** on average close to the truth (low bias) but every training set gives a very different, jagged line (high variance).
- **Random forest:** averages many high-variance trees → **variance drops** while bias stays low → lowest total error here.

# 25. Cross-Validation

### Why is it needed?
A single train/validation split is **noisy**: with a different random split you might get a noticeably different score (we saw this in Sections 9 and 16). Cross-validation uses the data more efficiently and gives a **more reliable estimate of generalisation** — plus a measure of its **uncertainty**.

### K-Fold
```text
Data split into K = 5 folds:
Round 1: [VAL] [train] [train] [train] [train]
Round 2: [train] [VAL] [train] [train] [train]
Round 3: [train] [train] [VAL] [train] [train]
Round 4: [train] [train] [train] [VAL] [train]
Round 5: [train] [train] [train] [train] [VAL]
→ 5 scores → report mean ± std
```

- **Stratified K-Fold:** every fold keeps the **same class proportions** — the default for classifiers in `cross_val_score`; essential for imbalanced data.
- **Other variants:** `GroupKFold` (same person never in train and validation), `TimeSeriesSplit` (always train on the past, validate on the future), `RepeatedStratifiedKFold` (repeat with different shuffles).

### `cross_val_score` vs `cross_validate`
- `cross_val_score(model, X, y, cv=5, scoring='roc_auc')` → one metric, array of fold scores.
- `cross_validate(model, X, y, cv=5, scoring=[...], return_train_score=True)` → several metrics, fit/score times, and training scores (to spot overfitting).

### Important rules
- Do CV on the **training set**; keep the test set for the final check.
- Put preprocessing inside a **Pipeline** so it is re-fitted within each fold (otherwise: leakage).

````python
model_cv = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))

kf = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

for name, cv in [('KFold', kf), ('StratifiedKFold', skf)]:
    fold_pos_rates = [yc.iloc[val_idx].mean() for _, val_idx in cv.split(Xc, yc)]
    scores = cross_val_score(model_cv, Xc, yc, cv=cv, scoring='accuracy')
    print(f'{name:<16} fold accuracies = {scores.round(3)} → mean {scores.mean():.3f} ± {scores.std():.3f}')
    print(f'{"":<16} share benign per fold = {np.round(fold_pos_rates, 3)}')

res = cross_validate(model_cv, Xc, yc, cv=skf, scoring=['accuracy', 'precision', 'recall', 'roc_auc'], return_train_score=True)
cv_table = pd.DataFrame(res).drop(columns=['fit_time', 'score_time'])
print('\ncross_validate — mean of each column:')
print(cv_table.mean().round(4))
````

````python
# Why a single split can mislead: the score depends on the random split
single_scores = []
for seed in range(100):
    a, b, c, d = train_test_split(Xc, yc, test_size=0.2, stratify=yc, random_state=seed)
    single_scores.append(model_cv.fit(a, c).score(b, d))
plt.figure(figsize=(8, 3.5))
sns.histplot(single_scores, bins=15)
plt.axvline(cross_val_score(model_cv, Xc, yc, cv=skf).mean(), color='red', lw=2, label='5-fold CV mean')
plt.title(f'Accuracy from 100 different single splits ranges from {min(single_scores):.3f} to {max(single_scores):.3f}')
plt.xlabel('test accuracy'); plt.legend(); plt.show()
````

# 26. Hyperparameter Tuning

### Parameters vs hyperparameters
| | Parameters | Hyperparameters |
|---|---|---|
| Set by | **The algorithm**, learned during `fit` | **You**, before `fit` |
| Examples | Linear coefficients, tree split thresholds, K-Means centroids | `C`, `alpha`, `max_depth`, `n_estimators`, `n_neighbors`, `learning_rate` |
| Chosen using | Training data | **Validation data / cross-validation** |

### Grid Search vs Random Search
| | `GridSearchCV` | `RandomizedSearchCV` |
|---|---|---|
| How | Tries **every** combination in a grid | Tries **n_iter** random combinations from distributions |
| Cost | Grows multiplicatively (4 × 4 × 3 = 48 combos × 5 folds = 240 fits) | You control the budget |
| Best for | Few hyperparameters, small grids | Many hyperparameters, continuous ranges; often finds good settings faster |

### Good practice
- Tune on the **training set** with CV; evaluate the final model **once** on the test set.
- Tune the **whole pipeline** (e.g. `logisticregression__C`) so preprocessing is inside the CV.
- Choose a **scoring** metric that matches the business goal.
- Don't over-tune tiny differences — if two settings are within the CV noise, prefer the simpler one.

````python
Xtr_c, Xte_c, ytr_c, yte_c = train_test_split(Xc, yc, test_size=0.2, stratify=yc, random_state=RANDOM_STATE)
cv5 = StratifiedKFold(5, shuffle=True, random_state=RANDOM_STATE)

# ---- 1. Logistic regression: grid search over C and penalty
lr_pipe = Pipeline([('scaler', StandardScaler()), ('model', LogisticRegression(max_iter=5000, solver='liblinear'))])
lr_grid = GridSearchCV(lr_pipe, param_grid={'model__C': [0.01, 0.1, 1, 10, 100], 'model__penalty': ['l1', 'l2']},
                       cv=cv5, scoring='roc_auc', n_jobs=-1)
lr_grid.fit(Xtr_c, ytr_c)
print('Logistic regression best params:', lr_grid.best_params_, '| best CV ROC-AUC:', round(lr_grid.best_score_, 4))

# ---- 2. KNN: grid search over K and weighting
knn_pipe = Pipeline([('scaler', StandardScaler()), ('model', KNeighborsClassifier())])
knn_grid = GridSearchCV(knn_pipe, param_grid={'model__n_neighbors': [3, 5, 7, 11, 15, 21, 31], 'model__weights': ['uniform', 'distance']},
                        cv=cv5, scoring='roc_auc', n_jobs=-1)
knn_grid.fit(Xtr_c, ytr_c)
print('KNN best params:', knn_grid.best_params_, '| best CV ROC-AUC:', round(knn_grid.best_score_, 4))

# ---- 3. Random forest: randomized search over several hyperparameters
from scipy.stats import randint
rf_search = RandomizedSearchCV(
    RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
    param_distributions={'n_estimators': randint(100, 400), 'max_depth': [None, 4, 6, 8, 12],
                         'min_samples_leaf': randint(1, 8), 'max_features': ['sqrt', 'log2', 0.5]},
    n_iter=15, cv=cv5, scoring='roc_auc', random_state=RANDOM_STATE, n_jobs=-1)
rf_search.fit(Xtr_c, ytr_c)
print('Random forest best params:', rf_search.best_params_, '| best CV ROC-AUC:', round(rf_search.best_score_, 4))
````

````python
# Inspect the grid results (top rows) and evaluate the tuned models ONCE on the test set
print(pd.DataFrame(lr_grid.cv_results_)[['param_model__C', 'param_model__penalty', 'mean_test_score', 'std_test_score', 'rank_test_score']]
      .sort_values('rank_test_score').head(5).round(4).to_string(index=False))

final_rows = []
for name, search in [('Logistic regression', lr_grid), ('KNN', knn_grid), ('Random forest', rf_search)]:
    best = search.best_estimator_                      # already refitted on the full training set (refit=True)
    proba = best.predict_proba(Xte_c)[:, 1]
    final_rows.append({'Model': name, 'CV ROC-AUC (train)': search.best_score_, 'Test ROC-AUC': roc_auc_score(yte_c, proba),
                       'Test accuracy': accuracy_score(yte_c, best.predict(Xte_c))})
print(); print(pd.DataFrame(final_rows).round(4).to_string(index=False))
````

### Code explanation
- `param_grid={'model__C': [...]}` → the **double underscore** means "the `C` parameter of the pipeline step named `model`".
- `scoring='roc_auc'` → the metric the search optimises.
- `best_params_`, `best_score_` (mean CV score), `best_estimator_` (the pipeline refitted on all training data with the best settings), `cv_results_` (every combination's scores).
- `randint(100, 400)` → a distribution to sample integers from in random search.

### Interpretation
All three tuned models reach high CV ROC-AUC. The **test** scores are the honest final numbers; small differences between models may be within noise.

# 27. Pipelines

### Why pipelines?
Without pipelines you must remember to apply the same preprocessing — with the **same fitted parameters** — to train, validation, test, and production data. It's easy to:
- fit the scaler/imputer/encoder on **all** data (leakage),
- forget a step at prediction time,
- apply steps in a different order,
- leak inside cross-validation.

A **Pipeline** bundles preprocessing + model into **one object**: `fit` learns everything from training data only; `predict` applies everything consistently. **ColumnTransformer** applies different steps to different columns.

```text
Numerical  → Imputation (median) → Scaling
Categorical → Imputation (most frequent) → One-Hot Encoding
Both        → (combined) → Model
```

````python
rpl = np.random.default_rng(27)
n = 800
loan = pd.DataFrame({
    'Age': rpl.integers(21, 70, n).astype(float),
    'Income': rpl.lognormal(11, 0.5, n).round(-2),
    'Loan_Amount': rpl.lognormal(12, 0.6, n).round(-3),
    'Credit_History_Years': rpl.integers(0, 30, n).astype(float),
    'Employment': rpl.choice(['Salaried', 'Self-employed', 'Unemployed'], n, p=[0.65, 0.28, 0.07]),
    'City': rpl.choice(['Lahore', 'Karachi', 'Islamabad', 'Multan'], n),
})
logit = (-5.0 + 1.3 * (loan['Loan_Amount'] / loan['Income']) - 0.05 * loan['Credit_History_Years']
         + loan['Employment'].map({'Salaried': -0.5, 'Self-employed': 0.2, 'Unemployed': 1.5}))
loan['Default'] = (rpl.random(n) < 1 / (1 + np.exp(-logit))).astype(int)
# Add realistic missing values
for col, frac in [('Income', 0.08), ('Credit_History_Years', 0.10), ('Employment', 0.05)]:
    loan.loc[rpl.choice(n, int(frac * n), replace=False), col] = np.nan

print('Missing values:\n', loan.isna().sum()[loan.isna().sum() > 0], '\n')
print('Default rate:', loan['Default'].mean().round(3))
loan.head()
````

````python
num_cols = ['Age', 'Income', 'Loan_Amount', 'Credit_History_Years']
cat_cols = ['Employment', 'City']

numeric_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),       # fill missing numbers with the TRAINING median
    ('scaler', StandardScaler()),                        # scale with TRAINING mean/std
])
categorical_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),        # fill missing categories
    ('onehot', OneHotEncoder(handle_unknown='ignore')),          # unseen categories at prediction time → all zeros
])
preprocessor = ColumnTransformer([
    ('num', numeric_pipe, num_cols),
    ('cat', categorical_pipe, cat_cols),
])
loan_model = Pipeline([
    ('preprocess', preprocessor),
    ('model', LogisticRegression(max_iter=2000, class_weight='balanced')),
])

X_loan, y_loan = loan[num_cols + cat_cols], loan['Default']
Xl_train, Xl_test, yl_train, yl_test = train_test_split(X_loan, y_loan, test_size=0.25, stratify=y_loan, random_state=RANDOM_STATE)

cv_auc = cross_val_score(loan_model, Xl_train, yl_train, cv=cv5, scoring='roc_auc')   # preprocessing re-fitted in every fold
loan_model.fit(Xl_train, yl_train)
print(f'CV ROC-AUC: {cv_auc.mean():.3f} ± {cv_auc.std():.3f} | Test ROC-AUC: {roc_auc_score(yl_test, loan_model.predict_proba(Xl_test)[:, 1]):.3f}')

print('\nFeatures the model actually sees:')
print(loan_model.named_steps['preprocess'].get_feature_names_out())

# The pipeline handles a brand-new, messy row automatically (missing value + unseen city)
new_applicant = pd.DataFrame([{'Age': 35, 'Income': np.nan, 'Loan_Amount': 400_000, 'Credit_History_Years': 3,
                               'Employment': 'Self-employed', 'City': 'Quetta'}])
print('\nNew applicant P(default) =', loan_model.predict_proba(new_applicant)[0, 1].round(3))
loan_model
````

### Code explanation
- Each `Pipeline` is a list of `(name, step)` pairs; the last step can be a model.
- `ColumnTransformer([('num', numeric_pipe, num_cols), ...])` sends each group of columns to its own pipeline and concatenates the results.
- `cross_val_score(loan_model, ...)` re-fits **imputer, scaler, encoder and model** inside every fold → no leakage.
- `get_feature_names_out()` shows the final columns (e.g. `cat__City_Lahore`).
- The new applicant had a missing income and a city never seen in training — the pipeline imputed and encoded it without crashing.

### WHEN to use
**Always**, for any model you evaluate or deploy.

# 28. Imbalanced Classification

### The problem
When one class is rare (fraud 1%, churn 5%, disease 0.5%):
- **Accuracy is misleading** (predicting "never" is 99% accurate — Section 12).
- Models tend to **ignore the rare class**, because doing so barely hurts their loss.

### Approaches

| Approach | How | Use when | Watch out |
|---|---|---|---|
| **Better metrics** | Recall, precision, F1, **PR-AUC**, confusion matrix | **Always** | — |
| **Threshold tuning** | Lower the decision threshold (Section 12) | Model ranks well but misses positives at 0.5 | Tune on validation data |
| **Class weights** | Penalise mistakes on the rare class more (`class_weight='balanced'`) | First thing to try; no data changes | Probabilities become less calibrated |
| **Random oversampling** | Duplicate rare-class rows (training data only!) | Small datasets | Overfitting to duplicated rows |
| **Random undersampling** | Drop majority-class rows | Very large datasets | Throws away information |
| **SMOTE** (concept) | Create **synthetic** rare-class rows by interpolating between a rare point and one of its rare neighbours | Moderate imbalance, numeric features | Can create unrealistic points; needs `imbalanced-learn` in practice |
| **Collect more rare examples / better features** | — | Always helpful | — |

⚠️ **Resample only the training data** (inside each CV fold). Never resample the test set — it must reflect reality.

````python
from sklearn.utils import resample
from sklearn.neighbors import NearestNeighbors

X_im, y_im = make_classification(n_samples=6000, n_features=12, n_informative=6, weights=[0.97, 0.03],
                                 class_sep=0.8, flip_y=0.005, random_state=RANDOM_STATE)
Xim_tr, Xim_te, yim_tr, yim_te = train_test_split(X_im, y_im, test_size=0.3, stratify=y_im, random_state=RANDOM_STATE)
print(f'Training class counts: {np.bincount(yim_tr)} (positives = {yim_tr.mean():.1%})')

def random_oversample(X, y, seed=0):
    minority = np.where(y == 1)[0]
    extra = resample(minority, replace=True, n_samples=(y == 0).sum() - len(minority), random_state=seed)
    idx = np.concatenate([np.arange(len(y)), extra])
    return X[idx], y[idx]

def random_undersample(X, y, seed=0):
    majority = np.where(y == 0)[0]
    keep = resample(majority, replace=False, n_samples=(y == 1).sum(), random_state=seed)
    idx = np.concatenate([keep, np.where(y == 1)[0]])
    return X[idx], y[idx]

def simple_smote(X, y, k=5, seed=0):
    '''Minimal SMOTE idea: new point = rare point + random fraction × (rare neighbour − rare point).'''
    r = np.random.default_rng(seed)
    X_min = X[y == 1]
    n_new = (y == 0).sum() - len(X_min)
    nn = NearestNeighbors(n_neighbors=k + 1).fit(X_min)
    neighbours = nn.kneighbors(X_min, return_distance=False)[:, 1:]          # skip the point itself
    base = r.integers(0, len(X_min), n_new)
    partner = neighbours[base, r.integers(0, k, n_new)]
    synthetic = X_min[base] + r.random((n_new, 1)) * (X_min[partner] - X_min[base])
    return np.vstack([X, synthetic]), np.concatenate([y, np.ones(n_new, dtype=int)])

strategies = {
    'Plain logistic regression': (Xim_tr, yim_tr, None),
    'class_weight="balanced"': (Xim_tr, yim_tr, 'balanced'),
    'Random oversampling': (*random_oversample(Xim_tr, yim_tr), None),
    'Random undersampling': (*random_undersample(Xim_tr, yim_tr), None),
    'SMOTE (simple version)': (*simple_smote(Xim_tr, yim_tr), None),
}
rows = []
for name, (Xs, ys, cw) in strategies.items():
    m = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, class_weight=cw)).fit(Xs, ys)
    pred, score = m.predict(Xim_te), m.predict_proba(Xim_te)[:, 1]
    rows.append({'Strategy': name, 'Train rows': len(ys), 'Accuracy': accuracy_score(yim_te, pred),
                 'Precision': precision_score(yim_te, pred, zero_division=0), 'Recall': recall_score(yim_te, pred),
                 'F1': f1_score(yim_te, pred), 'PR-AUC': average_precision_score(yim_te, score)})
pd.DataFrame(rows).set_index('Strategy').round(3)
````

### Interpretation
- The plain model has the **highest accuracy** but catches only a small share of positives (low **recall**).
- Class weights, oversampling, SMOTE and undersampling all **raise recall a lot** at the cost of precision (more false alarms). Undersampling trains on very few rows.
- **PR-AUC** (ranking quality across all thresholds) does **not** improve — here it is even slightly lower. These methods mainly **shift the decision threshold** rather than teaching the model new patterns. That's why simple **class weights or threshold tuning** are often as good as resampling, and why you should compare approaches with threshold-independent metrics too.
- In practice use `imbalanced-learn` (`from imblearn.over_sampling import SMOTE` with `imblearn.pipeline.Pipeline`) so resampling happens **only inside training folds**.

# 29. Feature Selection

(Connects to Topics 11–12; covered in depth in **Topic 16 — Feature Selection**.) Fewer, better features → simpler, faster, more interpretable, often more robust models.

| Method | Family | Idea | scikit-learn |
|---|---|---|---|
| **Correlation filtering** | Filter | Drop one of each highly correlated feature pair (redundancy); keep features correlated with the target | pandas `.corr()` |
| **Variance threshold** | Filter | Drop (near-)constant features | `VarianceThreshold` |
| **SelectKBest** | Filter | Keep the K features with the best univariate statistic (ANOVA F-test, chi²) | `SelectKBest(f_classif, k=...)` |
| **Mutual information** | Filter | Captures **non-linear** dependence between each feature and the target | `mutual_info_classif` |
| **RFE** | Wrapper | Repeatedly fit a model and remove the weakest feature | `RFE`, `RFECV` |
| **Lasso / L1** | Embedded | L1 penalty sets weak coefficients to 0 | `Lasso`, `LogisticRegression(penalty='l1')`, `SelectFromModel` |
| **Tree importance** | Embedded | Impurity-based importance from trees/forests | `.feature_importances_` |
| **Permutation importance** | Model-agnostic | Score drop when a feature is shuffled on **validation** data | `permutation_importance` |

⚠️ Feature selection is a **learned** step → do it **inside a pipeline / within CV** on training data only. Selecting features on the full data before CV is **leakage** (Section 33).

````python
from sklearn.feature_selection import VarianceThreshold, SelectKBest, f_classif, mutual_info_classif, RFE, SelectFromModel
from sklearn.inspection import permutation_importance

X_fs, y_fs = make_classification(n_samples=1500, n_features=20, n_informative=5, n_redundant=3, n_repeated=0,
                                 shuffle=False, random_state=RANDOM_STATE)
feat_names = [f'inf_{i}' for i in range(5)] + [f'red_{i}' for i in range(3)] + [f'noise_{i}' for i in range(12)]
X_fs = pd.DataFrame(X_fs, columns=feat_names)
X_fs['constant'] = 1.0                                        # a useless constant column
Xfs_tr, Xfs_te, yfs_tr, yfs_te = train_test_split(X_fs, y_fs, test_size=0.3, stratify=y_fs, random_state=RANDOM_STATE)

# 1. Variance threshold
vt = VarianceThreshold(threshold=0.0).fit(Xfs_tr)
print('VarianceThreshold removes:', list(Xfs_tr.columns[~vt.get_support()]))
Xv_tr = Xfs_tr.loc[:, vt.get_support()]

# 2. Correlation filtering (between features)
corr = Xv_tr.corr().abs()
upper = corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1))
print('Highly correlated pairs (|r| > 0.8):', [(a, b, round(v, 2)) for (a, b), v in upper.stack().items() if v > 0.8])

# 3–7. Several selectors, each keeping 6 features
k = 6
selected = {}
selected['SelectKBest (F-test)'] = Xv_tr.columns[SelectKBest(f_classif, k=k).fit(Xv_tr, yfs_tr).get_support()]
mi = pd.Series(mutual_info_classif(Xv_tr, yfs_tr, random_state=RANDOM_STATE), index=Xv_tr.columns)
selected['Mutual information'] = mi.nlargest(k).index
rfe = RFE(LogisticRegression(max_iter=2000), n_features_to_select=k).fit(StandardScaler().fit_transform(Xv_tr), yfs_tr)
selected['RFE (logistic)'] = Xv_tr.columns[rfe.support_]
l1 = LogisticRegression(penalty='l1', C=0.02, solver='liblinear').fit(StandardScaler().fit_transform(Xv_tr), yfs_tr)
selected['L1 / Lasso logistic'] = Xv_tr.columns[np.abs(l1.coef_[0]) > 1e-6]
rf_fs = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1).fit(Xv_tr, yfs_tr)
selected['Random forest importance'] = pd.Series(rf_fs.feature_importances_, index=Xv_tr.columns).nlargest(k).index
perm = permutation_importance(rf_fs, Xfs_te.loc[:, vt.get_support()], yfs_te, n_repeats=10, random_state=RANDOM_STATE, n_jobs=-1)
selected['Permutation importance'] = pd.Series(perm.importances_mean, index=Xv_tr.columns).nlargest(k).index

for name, cols in selected.items():
    n_noise = sum(c.startswith('noise') for c in cols)
    print(f'{name:<26} → {sorted(cols)}  (noise features picked: {n_noise})')
````

````python
# Does selection help? Compare CV scores — selection done INSIDE the pipeline (no leakage)
for name, pipe in [
    ('All 21 features', make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))),
    ('SelectKBest k=6 in pipeline', make_pipeline(VarianceThreshold(), StandardScaler(), SelectKBest(f_classif, k=6), LogisticRegression(max_iter=2000))),
    ('L1 SelectFromModel in pipeline', make_pipeline(VarianceThreshold(), StandardScaler(),
                                                     SelectFromModel(LogisticRegression(penalty='l1', C=0.05, solver='liblinear')),
                                                     LogisticRegression(max_iter=2000))),
]:
    s = cross_val_score(pipe, Xfs_tr, yfs_tr, cv=cv5, scoring='roc_auc')
    print(f'{name:<32} CV ROC-AUC = {s.mean():.4f} ± {s.std():.4f}')
````

### Interpretation
- Most methods recover the informative (`inf_`) and redundant (`red_`) features and ignore the noise.
- Redundant features are combinations of informative ones → some methods keep them instead of the originals (they carry the same information).
- The model with **6 selected features** performs about as well as with all 21 → simpler model, same quality.
- Filters (F-test) only see **one feature at a time**; wrappers/embedded methods see features **together**; permutation importance uses held-out data and is the most trustworthy view of what the model relies on.

# 30. Model Interpretation

### Why interpretability matters
- **Trust:** doctors, managers and customers need to understand **why** a prediction was made.
- **Debugging:** a model relying on a strange feature (an ID, a leakage column) is a red flag.
- **Fairness & regulation:** credit and hiring decisions may legally require explanations.
- **Business insight:** "support calls drive churn" is actionable.

| Technique | Scope | Works with | Tells you |
|---|---|---|---|
| **Linear coefficients** | Global | Linear/Ridge/Lasso | Change in prediction per unit (or per std if scaled) |
| **Logistic coefficients / odds ratios** | Global | Logistic regression | Multiplicative change in odds |
| **Impurity feature importance** | Global | Trees, forests, boosting | How much each feature was used for splits (training data; biased toward high-cardinality features) |
| **Permutation importance** | Global | **Any** model | How much the **validation** score drops when a feature is shuffled |
| **Partial dependence (PDP)** | Global | Any model | The **average** effect of changing one feature on the prediction |
| **SHAP** (concept) | **Local** (per prediction) + global | Any model (fast for trees) | How much each feature pushed **this particular** prediction up or down from the average |

⚠️ All of these describe **what the model learned** — not necessarily **causation** in the real world. Correlated features share/steal importance from each other.

````python
# 1. Linear coefficients (salary model from Section 7) — scaled so sizes are comparable
lin_pipe = make_pipeline(StandardScaler(), LinearRegression()).fit(salary_df[features], salary_df['Salary'])
print('Linear regression coefficients (per 1 standard deviation of each feature):')
print(pd.Series(lin_pipe[-1].coef_, index=features).round(0), '\n')

# 2. Logistic coefficients & odds ratios (loan pipeline from Section 27)
names = loan_model.named_steps['preprocess'].get_feature_names_out()
coefs = loan_model.named_steps['model'].coef_[0]
odds = pd.DataFrame({'coef (log-odds)': coefs, 'odds ratio': np.exp(coefs)}, index=names).sort_values('coef (log-odds)')
print('Logistic regression — loan default:'); print(odds.round(3))
````

**Reading the logistic table:** numeric features were standardized, so an odds ratio of e.g. 2.0 for `Loan_Amount` means +1 standard deviation of loan amount **doubles the odds** of default. For one-hot columns, compare categories with each other (e.g. `Unemployed` vs `Salaried`).

````python
# 3. Impurity importance vs 4. permutation importance (random forest on breast cancer, held-out data)
rf_int = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1).fit(Xtr_c, ytr_c)
imp = pd.Series(rf_int.feature_importances_, index=Xc.columns)
perm = permutation_importance(rf_int, Xte_c, yte_c, n_repeats=15, random_state=RANDOM_STATE, n_jobs=-1)
perm_s = pd.Series(perm.importances_mean, index=Xc.columns)

top = perm_s.nlargest(8).index.union(imp.nlargest(8).index)
comp = pd.DataFrame({'impurity importance': imp[top], 'permutation importance': perm_s[top]}).sort_values('permutation importance')
comp.plot(kind='barh', figsize=(9, 5.5), title='Impurity vs permutation importance (random forest)'); plt.show()
````

Correlated features (radius, perimeter, area…) **share** importance: shuffling one barely hurts because the others carry the same information. Permutation importance on held-out data is the more honest measure of what the model **needs**.

````python
# 5. Partial dependence: average effect of a feature on the prediction
from sklearn.inspection import PartialDependenceDisplay

gbr_pd = GradientBoostingRegressor(random_state=RANDOM_STATE).fit(diabetes.data, diabetes.target)
fig, ax = plt.subplots(1, 3, figsize=(16, 3.8))
PartialDependenceDisplay.from_estimator(gbr_pd, diabetes.data, ['bmi', 'bp', 's5'], ax=ax)
fig.suptitle('Partial dependence (diabetes progression): higher BMI, blood pressure and s5 → higher predicted progression', y=1.05)
plt.show()
````

### SHAP (conceptually)
SHAP (SHapley Additive exPlanations) comes from game theory: treat features as "players" who share the "payout" (the prediction). For **one** prediction:

$$\text{prediction} = \text{average prediction} + \sum_{\text{features}} \text{contribution}_i$$

Each contribution says how much that feature **pushed this prediction** above or below the average. For a **linear** model it's exactly `coef × (x − mean(x))`, as the cell below shows. For trees, the `shap` library computes it efficiently (`pip install shap`; not needed for this notebook).

````python
# SHAP-style explanation for ONE prediction of a linear model: contribution = coef × (x − mean)
lin_model = LinearRegression().fit(salary_df[features], salary_df['Salary'])
person = salary_df[features].iloc[[0]]
base = lin_model.predict(salary_df[features]).mean()              # average prediction
contrib = lin_model.coef_ * (person.to_numpy()[0] - salary_df[features].mean().to_numpy())

explain = pd.Series(contrib, index=features).round(0)
print(f'Average prediction           : {base:,.0f}')
for f, c in explain.items():
    print(f'  {f:<16} value = {person[f].iloc[0]:>5}  → contribution {c:+,.0f}')
print(f'Base + contributions         : {base + contrib.sum():,.0f}')
print(f'Model prediction for person  : {lin_model.predict(person)[0]:,.0f}  ✔ identical')
````

# 31. Error Analysis

Metrics tell you **how much** a model is wrong; error analysis tells you **where and why** — and therefore **what to fix**.

### Checklist
1. **Residuals** (regression): plot residuals vs predictions and vs each feature. Patterns = missing information.
2. **Error distribution:** are errors symmetric? A few huge errors (outliers, data errors)?
3. **False positives / false negatives** (classification): look at real examples of each.
4. **Segment-based errors:** compute the metric per group (city, device, customer type, time period). A model can be great on average and terrible for one segment.
5. **Hypothesise → fix → re-measure:** new feature, more data for a segment, a different model, a data fix.

### Demonstration
House prices where the **effect of area depends on location** (an interaction). A plain linear model can't express this.

````python
rea = np.random.default_rng(31)
n = 1200
homes = pd.DataFrame({
    'Area': rea.uniform(600, 3500, n),
    'Location': rea.choice(['City', 'Suburb', 'Rural'], n, p=[0.3, 0.45, 0.25]),
    'Age': rea.uniform(0, 40, n),
})
price_per_sqft = homes['Location'].map({'City': 400, 'Suburb': 220, 'Rural': 110})     # interaction!
homes['Price'] = 50_000 + price_per_sqft * homes['Area'] - 1_500 * homes['Age'] + rea.normal(0, 60_000, n)

Xh = homes[['Area', 'Location', 'Age']]; yh = homes['Price']
Xh_tr, Xh_te, yh_tr, yh_te = train_test_split(Xh, yh, test_size=0.3, random_state=RANDOM_STATE)

prep_h = ColumnTransformer([('num', StandardScaler(), ['Area', 'Age']), ('cat', OneHotEncoder(drop='first'), ['Location'])])
v1 = Pipeline([('prep', prep_h), ('model', LinearRegression())]).fit(Xh_tr, yh_tr)
pred_v1 = v1.predict(Xh_te)
print(f'Version 1 (plain linear): MAE = {mean_absolute_error(yh_te, pred_v1):,.0f} | R² = {r2_score(yh_te, pred_v1):.3f}')

err = Xh_te.copy(); err['Actual'] = yh_te; err['Pred'] = pred_v1; err['Residual'] = err['Actual'] - err['Pred']
err['AbsError'] = err['Residual'].abs()
print('\nMAE by segment:'); print(err.groupby('Location')['AbsError'].mean().round(0))

fig, axes = plt.subplots(1, 3, figsize=(18, 4))
axes[0].scatter(err['Pred'], err['Residual'], alpha=0.4, s=10); axes[0].axhline(0, color='red')
axes[0].set_title('Residuals vs predicted'); axes[0].set_xlabel('predicted')
for loc, g in err.groupby('Location'):
    axes[1].scatter(g['Area'], g['Residual'], s=10, alpha=0.5, label=loc)
axes[1].axhline(0, color='k'); axes[1].legend(); axes[1].set_title('Residuals vs Area, by location → clear patterns!')
sns.histplot(err['Residual'], bins=40, ax=axes[2]); axes[2].set_title('Error distribution')
plt.tight_layout(); plt.show()
````

**Diagnosis:** residuals vs area show **opposite slopes** for different locations: in the city the model under-predicts large homes; in rural areas it over-predicts them. → The price per square foot **depends on location**. Fix: add an **Area × Location interaction** feature (or use a model that learns interactions, like gradient boosting).

````python
def add_interactions(df):
    df = df.copy()
    for loc in ['City', 'Suburb', 'Rural']:
        df[f'Area_x_{loc}'] = df['Area'] * (df['Location'] == loc)          # area only counts in its own location
    return df

from sklearn.preprocessing import FunctionTransformer
num_int = ['Area', 'Age', 'Area_x_City', 'Area_x_Suburb', 'Area_x_Rural']
v2 = Pipeline([
    ('features', FunctionTransformer(add_interactions)),
    ('prep', ColumnTransformer([('num', StandardScaler(), num_int), ('cat', OneHotEncoder(drop='first'), ['Location'])])),
    ('model', LinearRegression())]).fit(Xh_tr, yh_tr)
v3 = Pipeline([('prep', ColumnTransformer([('cat', OneHotEncoder(), ['Location'])], remainder='passthrough')),
               ('model', GradientBoostingRegressor(random_state=RANDOM_STATE))]).fit(Xh_tr, yh_tr)

rows = []
for name, m in [('v1 plain linear', v1), ('v2 linear + interactions', v2), ('v3 gradient boosting', v3)]:
    p = m.predict(Xh_te)
    seg = pd.Series(np.abs(yh_te - p)).groupby(Xh_te['Location'].values).mean()
    rows.append({'Model': name, 'MAE': mean_absolute_error(yh_te, p), 'R²': r2_score(yh_te, p), **{f'MAE {k}': v for k, v in seg.items()}})
pd.DataFrame(rows).set_index('Model').round(3)
````

````python
# Classification error analysis: false positives / false negatives of the loan model, by segment
pred_loan = loan_model.predict(Xl_test)
ea = Xl_test.copy(); ea['Actual'] = yl_test.values; ea['Pred'] = pred_loan
ea['Error_Type'] = np.select([(ea.Actual == 1) & (ea.Pred == 0), (ea.Actual == 0) & (ea.Pred == 1)],
                             ['False negative (missed default)', 'False positive (false alarm)'], 'Correct')
print(ea['Error_Type'].value_counts(), '\n')
print('Error rate by employment type:')
print(ea.assign(Wrong=ea['Error_Type'] != 'Correct').groupby('Employment', dropna=False)['Wrong'].mean().round(3))
print('\nTypical false negatives (averages):')
print(ea[ea['Error_Type'].str.startswith('False neg')][['Income', 'Loan_Amount', 'Credit_History_Years']].mean().round(0))
````

### How error analysis leads to better models
- The segment table showed the plain model failing **differently per location** → an interaction feature cut the error dramatically; gradient boosting found the interaction automatically.
- For classification, look at **which customers** are false negatives/positives and **which segments** have high error rates — then add features, collect data, or set segment-specific thresholds.

# 33. Data Leakage

**Leakage** = the model sees information during training/evaluation that it **won't have** at real prediction time. Scores look amazing; production performance is disappointing.

| Leakage type | ❌ Bad workflow | ✅ Correct workflow |
|---|---|---|
| **Scaling before splitting** | `scaler.fit(X_all)` → split | Split → fit scaler on train → transform test (Pipeline) |
| **Encoding / imputing with all data** | Target encoding or median imputation on the full dataset | Fit encoders/imputers inside the pipeline on training folds |
| **Feature selection using test data** | `SelectKBest.fit(X_all, y_all)` → then CV | Put selection **inside** the pipeline |
| **Target-derived features** | A feature computed from the target or only known after it ("refund issued" to predict "will complain") | Only use information available **at prediction time** |
| **Future information (time series)** | Random split; rolling means including the current day | Time-based split; `shift(1)` before rolling (Topic 11) |
| **Duplicates / same entity in train & test** | Same customer's rows in both sets | Group-based split (`GroupKFold`) |
| **Tuning on the test set** | Picking the model with the best test score | Tune with CV; test once |

### Demonstration 1 — feature selection leakage (the most dramatic one)
Pure **random noise** features and a **random** target: the true achievable accuracy is 50%.

````python
rlk = np.random.default_rng(33)
X_noise = rlk.normal(size=(100, 5000))            # 100 rows, 5,000 random features
y_noise = rlk.integers(0, 2, 100)                 # random labels — nothing to learn!

# ❌ BAD: select the 20 "best" features using ALL rows, then cross-validate
X_selected_bad = SelectKBest(f_classif, k=20).fit_transform(X_noise, y_noise)
bad = cross_val_score(LogisticRegression(max_iter=2000), X_selected_bad, y_noise, cv=5).mean()

# ✅ CORRECT: selection inside the pipeline → done separately in each training fold
good = cross_val_score(make_pipeline(SelectKBest(f_classif, k=20), LogisticRegression(max_iter=2000)),
                       X_noise, y_noise, cv=5).mean()
print(f'❌ Selection before CV: accuracy = {bad:.3f}   ← looks like a great model... on pure noise!')
print(f'✅ Selection inside CV: accuracy = {good:.3f}   ← the truth: ≈ 50%, nothing to learn')
````

````python
# Demonstration 2 — a target-derived feature
rt = np.random.default_rng(34)
n = 2000
cust_lk = pd.DataFrame({'Tenure': rt.integers(1, 60, n), 'Monthly_Charge': rt.uniform(500, 5000, n)})
cust_lk['Churn'] = (rt.random(n) < 1 / (1 + np.exp(-(0.5 - 0.05 * cust_lk['Tenure'])))).astype(int)
# 'Cancellation_Survey_Done' only exists AFTER a customer has decided to leave
cust_lk['Cancellation_Survey_Done'] = ((cust_lk['Churn'] == 1) & (rt.random(n) < 0.9)).astype(int)

for cols, label in [(['Tenure', 'Monthly_Charge'], 'Honest features'), (['Tenure', 'Monthly_Charge', 'Cancellation_Survey_Done'], 'With leaky feature')]:
    s = cross_val_score(make_pipeline(StandardScaler(), LogisticRegression()), cust_lk[cols], cust_lk['Churn'], cv=5, scoring='roc_auc').mean()
    print(f'{label:<20} CV ROC-AUC = {s:.3f}')
````

````python
# Demonstration 3 — time series: random split vs time-based split
from sklearn.model_selection import TimeSeriesSplit

days = pd.date_range('2022-01-01', periods=900, freq='D')
trend = np.linspace(100, 300, 900)                                         # demand grows over time
ts = pd.DataFrame({'Date': days, 'Demand': trend + 20 * np.sin(np.arange(900) * 2 * np.pi / 7) + rng.normal(0, 10, 900)})
ts['DayOfWeek'] = ts['Date'].dt.dayofweek
ts['Lag_1'] = ts['Demand'].shift(1)                                        # yesterday's demand (known today) ✔
ts = ts.dropna()
X_ts, y_ts = ts[['DayOfWeek', 'Lag_1']], ts['Demand']

rf_ts = RandomForestRegressor(n_estimators=200, random_state=0, n_jobs=-1)
random_cv = -cross_val_score(rf_ts, X_ts, y_ts, cv=KFold(5, shuffle=True, random_state=0), scoring='neg_mean_absolute_error').mean()
time_cv = -cross_val_score(rf_ts, X_ts, y_ts, cv=TimeSeriesSplit(5), scoring='neg_mean_absolute_error').mean()
print(f'❌ Random K-Fold (trains on the future): MAE = {random_cv:.1f}')
print(f'✅ TimeSeriesSplit (past → future)     : MAE = {time_cv:.1f}  ← the realistic error when forecasting')
````

### Interpretation
- **Demo 1:** selecting features on all rows made random noise look predictive. Inside the pipeline, the honest ≈ 50% appears.
- **Demo 2:** the survey column almost reveals the answer → a near-perfect score that is useless in production (you'd only know it after the customer churned).
- **Demo 3:** the random split lets the random forest learn from future demand levels; the time-based split shows the true (larger) forecasting error — random forests can't extrapolate a rising trend.

# 34. Model Deployment Basics

### Saving and loading models
A trained model (ideally the **whole pipeline**, so preprocessing comes with it) is saved to a file and loaded wherever predictions are needed.

`joblib` is the standard tool for scikit-learn objects (efficient for large NumPy arrays).

⚠️ Security: only load model files from **trusted** sources — loading a pickle/joblib file can execute code.
⚠️ Load with the **same library versions** that saved the model (record them!).

### Two ways to serve predictions
| | **Batch inference** | **Real-time (online) inference** |
|---|---|---|
| When | On a schedule (e.g. every night) | On demand, per request |
| Input | A whole table of rows | One row (or a few) |
| Latency | Minutes/hours are fine | Milliseconds |
| Example | Score all customers' churn risk nightly → CRM | Price a house as the user types; approve a card transaction |
| Typical tech | Scheduled Python job, Airflow, Spark | A web API (FastAPI/Flask) inside Docker, cloud endpoints |

### Prediction API concept
```text
Client (website / app)  ──HTTP POST {"Age": 35, "Income": 90000, ...}──▶  API server
                                                                           │ loads model once at startup
                                                                           │ validates input
                                                                           │ model.predict_proba(row)
Client ◀──────────────────── {"default_probability": 0.23} ────────────────┘
```
(We'll build real APIs with FastAPI in a later topic.)

````python
import joblib, os, json, sklearn

os.makedirs('ml_models', exist_ok=True)
model_path = 'ml_models/loan_default_pipeline.joblib'

joblib.dump(loan_model, model_path)                               # save the ENTIRE pipeline
metadata = {'model': 'LogisticRegression pipeline', 'features': num_cols + cat_cols, 'target': 'Default',
            'sklearn_version': sklearn.__version__, 'test_roc_auc': round(roc_auc_score(yl_test, loan_model.predict_proba(Xl_test)[:, 1]), 4)}
with open('ml_models/loan_default_metadata.json', 'w') as f:
    json.dump(metadata, f, indent=2)                              # always save metadata next to the model
print('Saved:', os.listdir('ml_models'), '| size:', round(os.path.getsize(model_path) / 1024, 1), 'KB')

loaded = joblib.load(model_path)                                  # e.g. in another script / server
print('Loaded model gives identical predictions:', np.allclose(loaded.predict_proba(Xl_test), loan_model.predict_proba(Xl_test)))
````

````python
# BATCH inference: score a whole table at once
new_batch = Xl_test.head(5).copy()
new_batch['Default_Probability'] = loaded.predict_proba(new_batch[num_cols + cat_cols])[:, 1].round(3)
new_batch['Decision'] = np.where(new_batch['Default_Probability'] >= 0.5, 'Review manually', 'Approve')
print(new_batch[['Income', 'Loan_Amount', 'Employment', 'Default_Probability', 'Decision']])

# REAL-TIME inference: one request (a dict, like JSON from an API)
def predict_one(request: dict, model=loaded) -> dict:
    row = pd.DataFrame([{col: request.get(col, np.nan) for col in num_cols + cat_cols}])   # missing fields → NaN → imputed
    start = time.perf_counter()
    p = float(model.predict_proba(row)[0, 1])
    return {'default_probability': round(p, 4), 'latency_ms': round((time.perf_counter() - start) * 1000, 2)}

print('\nReal-time request →', predict_one({'Age': 29, 'Income': 55000, 'Loan_Amount': 300000, 'Employment': 'Salaried', 'City': 'Lahore'}))
````

# 35. Production ML Considerations & MLOps

A model in a notebook is ~10% of the work. In production it must stay **correct, reproducible and monitored** as the world changes.

| Topic | What it means | Practices |
|---|---|---|
| **Reproducibility** | Anyone can re-create the same model | Fixed random seeds, pinned library versions (`requirements.txt`), code in Git, documented data snapshot |
| **Versioning** | Track versions of **code, data and models** | Git, DVC / data snapshots, a model registry (e.g. MLflow) with metrics per version |
| **Model monitoring** | Watch the model after deployment | Prediction distributions, latency, errors, business KPIs |
| **Data drift** | The **input** distribution changes (e.g. customers get younger, a new city appears) | Compare live features to training data (KS test, PSI); alert on large shifts |
| **Concept drift** | The **relationship** between inputs and target changes (e.g. after a price change, the same customer behaves differently) | Track performance when true labels arrive; retrain |
| **Retraining** | Update the model with new data | Scheduled (monthly) or triggered by drift/performance drop; always re-validate before replacing |
| **Logging** | Record inputs, predictions, model version, timestamps | Enables debugging, audits and later training data |
| **Performance monitoring** | Compare predictions with actual outcomes as they arrive | Dashboards of MAE / recall over time and per segment |

**MLOps** = DevOps practices applied to ML: automation (CI/CD for training and deployment), testing (data validation, model checks), monitoring and governance — so ML systems are reliable, not one-off notebooks.

### A tiny data-drift check

````python
train_income = Xl_train['Income'].dropna()
live_income_ok = pd.Series(rng.lognormal(11, 0.5, 500))          # production data similar to training
live_income_drift = pd.Series(rng.lognormal(11.4, 0.5, 500))     # incomes shifted up ~50% (e.g. inflation / new market)

def psi(expected, actual, bins=10):
    '''Population Stability Index: < 0.1 stable, 0.1–0.25 moderate shift, > 0.25 major shift (common rule of thumb).'''
    edges = np.quantile(expected, np.linspace(0, 1, bins + 1)); edges[0], edges[-1] = -np.inf, np.inf
    e = np.histogram(expected, edges)[0] / len(expected)
    a = np.histogram(actual, edges)[0] / len(actual)
    e, a = np.clip(e, 1e-6, None), np.clip(a, 1e-6, None)
    return float(np.sum((a - e) * np.log(a / e)))

for name, live in [('No drift', live_income_ok), ('Drifted', live_income_drift)]:
    ks = stats.ks_2samp(train_income, live)
    print(f'{name:<9} KS p-value = {ks.pvalue:.4f} | PSI = {psi(train_income, live):.3f}')
````

# 36. Common ML Mistakes

| # | Mistake | Why it hurts | Fix |
|---|---|---|---|
| 1 | **Data leakage** | Unrealistically good scores; failure in production | Split first; pipelines; only features available at prediction time (Section 33) |
| 2 | **Overfitting** | Great training score, poor on new data | CV, regularization, simpler models, more data (Section 23) |
| 3 | **Wrong metric** | Optimising accuracy when missed fraud is what matters | Choose metrics from business costs (Sections 8, 12) |
| 4 | **Training on test data** | Test score no longer measures generalisation | Lock the test set away; use it once |
| 5 | **Not using validation / CV** | Decisions based on one noisy split | Cross-validation (Section 25) |
| 6 | **Ignoring class imbalance** | Model never predicts the rare class | PR-AUC, class weights, thresholds (Section 28) |
| 7 | **Blind hyperparameter tuning** | Huge grids, overfit to CV noise, wasted compute | Understand each knob, random search, sensible ranges, prefer simpler ties |
| 8 | **Too many features** | Noise, overfitting, slow, hard to explain | Feature selection, regularization (Section 29) |
| 9 | **Too little data** | High variance, unreliable estimates | Simpler models, CV, collect more data, learning curves |
| 10 | **Ignoring outliers** | Distorted linear models and scalers | Investigate; robust models/scalers; transformations (Topics 5, 13) |
| 11 | **Incorrect scaling** | KNN/SVM/regularized models behave badly; or scaler fit on all data | Scale inside pipelines for models that need it |
| 12 | **Incorrect encoding** | Label-encoding nominal categories creates fake order; unseen categories crash | One-hot (`handle_unknown='ignore'`), ordinal only for ordered categories (Topic 6) |
| 13 | **Ignoring the business objective** | A "better" model that doesn't change decisions or costs too much | Define success with stakeholders; expected value of decisions |
| 14 | **Comparing models unfairly** | Different splits, metrics or tuning effort | Same CV folds, same metric, similar tuning budget (Section 32) |
| 15 | **Not analysing errors** | Missing easy, high-impact fixes | Residuals, FP/FN review, segment errors (Section 31) |
| 16 | **No baseline** | Can't tell if the model adds value | Always compare with a dummy / simple model |
| 17 | **Random split on time series** | Leakage from the future | Time-based split |
| 18 | **Trusting uncalibrated probabilities** | Wrong expected-value decisions | Check calibration (Topic 15) |
| 19 | **Not saving preprocessing with the model** | Different transformations in production | Save the whole pipeline (Section 34) |
| 20 | **Deploy and forget** | Silent degradation due to drift | Monitoring and retraining (Section 35) |

# 43. Machine Learning Cheat Sheet

| Algorithm | Type | Task | Scaling? | Strengths | Weaknesses | Key Hyperparameters |
|---|---|---|---|---|---|---|
| **Linear Regression** | Supervised | Regression | Not required (helps interpretation) | Fast, interpretable, great baseline | Only linear patterns; outliers; multicollinearity | — (`fit_intercept`) |
| **Ridge** | Supervised | Regression | **Yes** | Handles correlated features, reduces overfitting | Keeps all features | `alpha` |
| **Lasso** | Supervised | Regression | **Yes** | Sparse models = feature selection | Unstable with correlated groups | `alpha` |
| **ElasticNet** | Supervised | Regression | **Yes** | Selection + stability with correlated features | Two knobs to tune | `alpha`, `l1_ratio` |
| **Logistic Regression** | Supervised | Classification | **Yes** | Probabilities, interpretable, strong baseline | Linear decision boundary | `C`, `penalty`, `class_weight` |
| **KNN** | Supervised | Both | **Yes** | Simple, flexible, no training | Slow prediction, high-dimensional data, irrelevant features | `n_neighbors`, `weights`, `metric` |
| **Naive Bayes** | Supervised | Classification | No | Extremely fast, great for text, small data | Independence assumption; poorly calibrated | `alpha` (Multinomial), `var_smoothing` (Gaussian) |
| **Decision Tree** | Supervised | Both | No | Readable rules, non-linear, mixed data | Overfits, unstable | `max_depth`, `min_samples_leaf`, `criterion` |
| **Random Forest** | Supervised | Both | No | Robust, accurate, little tuning, OOB score | Large, less interpretable, no extrapolation | `n_estimators`, `max_depth`, `max_features`, `min_samples_leaf` |
| **Gradient Boosting** | Supervised | Both | No | Often best on tabular data, handles NaN (Hist) | Needs tuning, can overfit | `learning_rate`, `n_estimators`/`max_iter`, `max_depth`/`max_leaf_nodes` |

| **PCA** | Unsupervised | Dimensionality reduction | **Yes** | Compress, visualise, remove correlation | Components hard to interpret; ignores the target | `n_components` |

# "Which algorithm should I try?"

⚠️ **No algorithm is universally best** ("no free lunch"). Use this guide to pick sensible **candidates**, then **compare them with cross-validation** on your data.

```text
1. What is the TARGET?
   ├── No target ─────────────▶ Unsupervised:
   │                             groups → K-Means (round clusters) / hierarchical (small data, dendrogram) / DBSCAN (odd shapes, noise)
   │                             too many features / visualise → PCA
   ├── A number ──────────────▶ Regression
   └── A category ────────────▶ Classification

2. ALWAYS start with a BASELINE
   Regression → DummyRegressor(mean) + Linear/Ridge
   Classification → DummyClassifier + Logistic Regression

3. How big is the dataset?
   ├── Small (< ~1k rows) ─────▶ Linear/logistic (regularized), Naive Bayes, KNN, SVM, small Random Forest
   ├── Medium (1k – 1M) ───────▶ Random Forest, Gradient Boosting (HistGradientBoosting), linear models
   └── Very large / text / images ▶ Linear models (SGD), HistGradientBoosting; unstructured data → deep learning

4. Linear or non-linear relationship?
   ├── Roughly linear (check scatter/residual plots) ▶ Linear / Ridge / Lasso / Logistic
   └── Curves, thresholds, interactions ▶ Tree ensembles (RF, Gradient Boosting), SVM-RBF, KNN
       (or engineer features/polynomials for linear models)

5. How important is interpretability?
   ├── Must explain every decision ▶ Linear/logistic (coefficients), shallow decision tree
   └── Accuracy first ▶ Gradient Boosting / Random Forest + permutation importance / PDP / SHAP

6. Scaling?
   ├── Distance / gradient / penalty-based (KNN, SVM, linear with regularization, logistic, K-Means, PCA) ▶ SCALE (in a pipeline)
   └── Tree-based (DT, RF, GB) ▶ scaling not needed

7. Class imbalance?
   ▶ Use PR-AUC/recall/F1, stratified CV, class_weight='balanced', threshold tuning;
     resampling (SMOTE) only inside training folds; prefer models that output probabilities

8. Many features vs rows?
   ▶ Regularization (Lasso/Ridge/ElasticNet), feature selection, PCA; avoid unregularized models and KNN

9. Missing values & categories?
   ▶ Pipelines with SimpleImputer + OneHotEncoder (any model), or HistGradientBoosting (native NaN support)

10. Finally
   ▶ Compare 3–5 candidates with the SAME CV folds and metric → tune the best 1–2 → evaluate ONCE on the test set
   ▶ Choose using performance AND simplicity, speed, interpretability, maintenance cost
```

### Golden rules of ML
1. Understand the **business problem** and pick the metric first.
2. Always build a **baseline**.
3. **Split first**; keep the test set untouched until the end.
4. Put all preprocessing in a **Pipeline**.
5. Use **cross-validation** for every decision.
6. Beware of **leakage** — only use information available at prediction time.
7. Watch for **overfitting**: compare training vs validation.
8. For imbalanced data, don't trust **accuracy**.
9. **Analyse errors** — they tell you what to fix.
10. Prefer the **simplest model** that meets the goal.
11. Save the **whole pipeline** + metadata; monitor it in production.
