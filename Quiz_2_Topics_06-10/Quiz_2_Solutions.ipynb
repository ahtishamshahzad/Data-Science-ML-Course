### Topic 06 — Encoding Techniques

**Q1.** A `Size` column has the ordered values Small < Medium < Large. Which encoding fits best?

- **A)** Ordinal encoding that keeps the order  ✅
- **B)** One-hot encoding
- **C)** Hash encoding
- **D)** Drop the column

> ✅ **Answer: A** — Ordinal data has a natural order, so map it to 0, 1, 2 in that order.

**Q2.** Why use `drop_first=True` in `pd.get_dummies()`?

- **A)** To make it faster
- **B)** To fill missing values
- **C)** To sort the categories
- **D)** To avoid the dummy variable trap (perfect multicollinearity)  ✅

> ✅ **Answer: D** — With k dummy columns, one is always predictable from the others. Dropping one removes that redundancy.

**Q3.** What is the main risk of **target encoding**?

- **A)** It uses too much memory
- **B)** Data leakage when the category means are computed from the same rows (or the test set)  ✅
- **C)** It cannot handle text
- **D)** It is always slow

> ✅ **Answer: B** — Compute target means on training data only (ideally out-of-fold), otherwise the model sees the answer.

**Q4.** An `OneHotEncoder(handle_unknown='ignore')` meets a category at transform time that it never saw while fitting. What happens?

- **A)** It raises an error
- **B)** A new column is created
- **C)** That feature's one-hot columns are all zeros for that row  ✅
- **D)** The row becomes NaN

> ✅ **Answer: C** — 'ignore' encodes unknown categories as all zeros instead of failing.

**Q5.** What should you do **first** in EDA?

- **A)** Look at shape, dtypes, `head()` and missing values  ✅
- **B)** Train a model

> ✅ **Answer: A** — Get to know the data before changing anything.

- **A)** Histogram or boxplot  ✅
- **B)** Crosstab
- **C)** Scatter plot

> ✅ **Answer: A** — Univariate = one variable: look at its distribution and spread.

- **A)** A great feature
- **B)** A missing value
- **C)** A leakage red flag  ✅
- **D)** An outlier

> ✅ **Answer: C** — It contains the answer itself, so it would not be available at prediction time.

**Q8.** How do you study the relationship between **two categorical** variables?

- **A)** Scatter plot
- **B)** Histogram
- **C)** Crosstab / grouped bar chart  ✅
- **D)** Line chart

> ✅ **Answer: C** — `pd.crosstab()` counts every combination of the two categories.

### Topic 09 — Pandas Data Manipulation

**Q13.** What does `df.groupby('City')['Sales'].transform('mean')` return?

- **A)** One row per city
- **B)** A single number
- **C)** A Series with the same length as `df`, each row holding its city's mean  ✅
- **D)** The rows sorted by city

> ✅ **Answer: C** — `transform` broadcasts the group result back to every original row — useful for new columns.

**Q14.** What is the difference between `pd.cut` and `pd.qcut`?

- **A)** `cut` makes equal-frequency bins; `qcut` makes equal-width bins
- **B)** `cut` makes equal-width bins (or your edges); `qcut` makes equal-frequency bins (quantiles)  ✅
- **C)** There is no difference
- **D)** `qcut` is for text

> ✅ **Answer: B** — `qcut` puts about the same number of rows in each bin.

**Q15.** What does `df.pivot_table(index='City', columns='Month', values='Sales', aggfunc='sum')` produce?

- **A)** A long table
- **B)** A Series
- **C)** A list
- **D)** A wide table: cities as rows, months as columns, summed sales in the cells  ✅

> ✅ **Answer: D** — A pivot table reshapes long data into a summary grid.

**Q16.** `df['Name'].str.lower()` works on…

- **A)** the whole DataFrame
- **B)** column names only
- **C)** numbers only
- **D)** each string in that Series  ✅

> ✅ **Answer: D** — The `.str` accessor applies a string method to every value in a Series.

### Topic 10 — Combining DataFrames

**Q17.** By default, what does `pd.concat([jan, feb])` do?

- **A)** Stacks the rows of feb under jan  ✅
- **B)** Joins them on a key column
- **C)** Puts them side by side
- **D)** Keeps only common rows

> ✅ **Answer: A** — `axis=0` (the default) adds rows. Use `axis=1` to put DataFrames side by side.

**Q18.** You want **every customer**, even those with no orders: `customers.merge(orders, on='id', how=?)`

- **A)** `'inner'`
- **B)** `'right'`
- **C)** `'cross'`
- **D)** `'left'`  ✅

> ✅ **Answer: D** — A left join keeps all rows of the left table (customers); missing order data becomes NaN.

**Q19.** What does `validate='one_to_one'` do in `merge()`?

- **A)** Makes the merge faster
- **B)** Raises an error if a key is duplicated on either side  ✅
- **C)** Drops duplicate rows
- **D)** Sorts the result

> ✅ **Answer: B** — It protects you from silent row duplication caused by unexpected duplicate keys.

**Q20.** `indicator=True` adds a `_merge` column with which values?

- **A)** `both`, `left_only`, `right_only`  ✅
- **B)** `True` / `False`
- **C)** Row counts
- **D)** The index

> ✅ **Answer: A** — It tells you where each row came from — great for checking a join.

Create `size_code`: the `Size` column encoded as **Small = 0, Medium = 1, Large = 2** (as a Series or list).

### Task 2 — Topic 07 — Quick data-quality audit

For `students`, create a dict `audit` with three keys: `'rows'` (number of rows), `'duplicates'` (number of fully duplicated rows) and `'missing_city'` (number of missing values in City).

```python
students = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ali', 'Hina', 'Omar', 'Sara'],
    'City': ['Lahore', None, 'Lahore', 'Karachi', None, None],
    'Score': [70, 85, 70, 90, 65, 85],
})

audit = {
    'rows': len(students),
    'duplicates': int(students.duplicated().sum()),
    'missing_city': int(students['City'].isna().sum()),
}
audit

check(2)
```

Draw a histogram of `ages` with **5 bins** on an Axes called `ax`, and give it a title. (Hint: `fig, ax = plt.subplots()` then `ax.hist(...)`.)

### Task 4 — Topic 09 — Share of the city total

Add a column `pct_of_city` to `sales`: each row's `Amount` divided by the **total Amount of its City** (use `groupby(...).transform('sum')`).

```python
sales = pd.DataFrame({
    'City': ['Lahore', 'Lahore', 'Karachi', 'Karachi', 'Karachi'],
    'Amount': [300, 100, 200, 200, 600],
})

sales['pct_of_city'] = sales['Amount'] / sales.groupby('City')['Amount'].transform('sum')
sales

check(4)
```

<a id="task-5"></a>
### Task 5 — Topic 10 — Orders per customer (including zero)

Create `orders_per_customer`: a Series indexed by customer **Name** with the number of orders of **every** customer — customers without orders must show **0**.

```python
customers = pd.DataFrame({'CustomerID': [1, 2, 3, 4], 'Name': ['Ali', 'Sara', 'Hina', 'Omar']})
orders = pd.DataFrame({'OrderID': [10, 11, 12, 13], 'CustomerID': [1, 1, 3, 1]})

merged = customers.merge(orders, on='CustomerID', how='left')
orders_per_customer = merged.groupby('Name')['OrderID'].count()   # count() ignores NaN -> 0
orders_per_customer

check(5)
```
