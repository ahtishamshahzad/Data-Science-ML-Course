p_spam = 0.01
p_flag_given_spam = 0.95
p_flag_given_ham = 0.02

p_flag = p_flag_given_spam * p_spam + p_flag_given_ham * (1 - p_spam)   # total probability

<details><summary>💡 <b>Show solution</b> (try it yourself first!)</summary>

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X_blobs, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=7)
