### Solution — Practice 3 — Manual Standardization

````python
vals = pd.Series([10, 20, 30, 40, 50])
mean, std = vals.mean(), vals.std()            # pandas std() uses the sample formula (n − 1)
z = (vals - mean) / std
print('Mean:', mean, '| Std:', round(std, 4))
print(pd.DataFrame({'Value': vals, 'Z_Score': z.round(4)}))
print('Mean of z-scores ≈', round(z.mean(), 10))
````

**Explanation:** Mean = 30, sample std ≈ 15.81 → z-scores ≈ −1.26, −0.63, 0, 0.63, 1.26. Their mean is 0 (up to rounding).

# Practice 4 — IQR Outlier Detection

Create a Series:

`[10, 12, 11, 13, 12, 14, 15, 100]`

Use the IQR method to:

1. Calculate Q1.
2. Calculate Q3.
3. Calculate IQR.
4. Calculate lower and upper bounds.
5. Find the outlier.

````python
data = pd.Series([10, 12, 11, 13, 12, 14, 15, 100])
multiplier=1.5
q1 = data.quantile(0.25)
q3 = data.quantile(0.75)
iqr = q3 - q1
lower = q1 - multiplier * iqr
upper = q3 + multiplier * iqr

print(data[(data < lower ) | (data > upper)  ])
````

### Solution — Practice 4 — IQR detection

````python
s4 = pd.Series([10, 12, 11, 13, 12, 14, 15, 100])
q1, q3 = s4.quantile(0.25), s4.quantile(0.75)
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print(f'Q1={q1}, Q3={q3}, IQR={iqr}, bounds=({lower}, {upper})')
s4[(s4 < lower) | (s4 > upper)]
````

**Explanation:** Only 100 lies above the upper bound → it is the outlier.

# Practice 5 — Capping

Create a Series:

`[10, 12, 15, 14, 13, 100]`

Use the IQR method to calculate the upper bound and cap values above the upper bound.

### Solution — Practice 5 — Capping

````python
s5 = pd.Series([10, 12, 15, 14, 13, 100])
q1, q3 = s5.quantile(0.25), s5.quantile(0.75)
upper = q3 + 1.5 * (q3 - q1)
print('Upper bound:', upper)
pd.DataFrame({'Original': s5, 'Capped': s5.clip(upper=upper)})
````

**Explanation:** 100 is replaced by the upper bound; the other values are unchanged. Capping keeps the row but limits the extreme value.

Use only Pandas and NumPy.

Given:

`data = [5, 6, 7, 5, 8, 6, 7, 100, 5, 6]`

Find the outlier using:

1. IQR.
2. Z-score.
3. Modified Z-score.

Compare the results.

### Solution — Practice 8 — Outliers with IQR, z-score and modified z-score

````python
d8 = pd.Series([5, 6, 7, 5, 8, 6, 7, 100, 5, 6])
q1, q3 = d8.quantile(0.25), d8.quantile(0.75); iqr = q3 - q1
iqr_out = d8[(d8 < q1 - 1.5 * iqr) | (d8 > q3 + 1.5 * iqr)]
z = (d8 - d8.mean()) / d8.std()
med = d8.median(); mad = np.median(np.abs(d8 - med))
mod_z = 0.6745 * (d8 - med) / mad
print('IQR outliers     :', iqr_out.tolist())
print('|z| > 3          :', d8[z.abs() > 3].tolist(), '(max |z| =', round(z.abs().max(), 2), ')')
print('|modified z|>3.5 :', d8[mod_z.abs() > 3.5].tolist())
````

**Explanation:** IQR and modified z-score both flag 100. The ordinary z-score only just reaches about 2.84, because 100 inflates the mean and std it is measured against. That's why robust methods (IQR, MAD) are preferred for small or skewed samples.
