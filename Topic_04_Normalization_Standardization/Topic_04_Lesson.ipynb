# 3. Why Do We Need Scaling?

Suppose we have these features:

- Age: 18 - 60
- Experience: 0 - 20
- Salary: 50,000 - 500,000
- Score: 0 - 100

The features have very different numerical ranges.

Some machine-learning algorithms are sensitive to these differences in scale, especially:

- K-Nearest Neighbors
- K-Means
- Support Vector Machines
- Neural Networks
- PCA
- Gradient-based models

Scaling puts features onto comparable numerical ranges.

<a id="sec-4"></a>
# 4. Min-Max Normalization

Min-Max normalization transforms values to a fixed range, normally **0 to 1**.

Formula:

$$x' = \frac{x - min(x)}{max(x) - min(x)}$$

After transformation:

- Minimum value becomes 0
- Maximum value becomes 1
- Other values fall between 0 and 1

### Example

For `[10, 20, 30]`:

`10 -> 0`

`20 -> 0.5`

`30 -> 1`

````python
values = pd.Series([10, 20, 30, 40, 50])

normalized = (values - values.min()) / (values.max() - values.min())

result = pd.DataFrame({
    'Original': values,
    'Normalized': normalized
})

result
````

## 5. Manual Min-Max Normalization with Pandas

<a id="sec-7"></a>
# 7. Standardization / Z-Score Scaling

Standardization transforms data so that the feature has approximately:

- Mean = 0
- Standard deviation = 1

Formula:

$$z = \frac{x - \mu}{\sigma}$$

Where:

- `x` = original value
- `μ` = mean
- `σ` = standard deviation

A positive z-score means the value is above the mean.

A negative z-score means the value is below the mean.

````python
values = pd.Series([10, 20, 30, 40, 50])

mean = values.mean()
std = values.std()

z_scores = (values - mean) / std

pd.DataFrame({
    'Original': values,
    'Z_Score': z_scores
})
````

The important idea is that standardization does **not** force values between 0 and 1.

You can have values such as:

`-2.1, -1.0, 0, 0.8, 2.5`

````python
df_standardized = df.copy()

df_standardized["Age"] = (
    df_standardized["Age"] - df_standardized["Age"].mean()
) / df_standardized["Age"].std()

df_standardized
````

<a id="sec-9"></a>
# 9. Normalization vs Standardization

| Method | Main Idea | Typical Range | Sensitive to Outliers? |
|---|---|---|---|
| Min-Max Normalization | Rescale using min/max | 0 to 1 | Yes |
| Standardization | Mean 0, std 1 | No fixed range | Yes |
| Robust Scaling | Median and IQR | No fixed range | Less sensitive |

### Use Min-Max when

- You want a fixed range.
- The data does not contain serious outliers.
- Algorithms benefit from bounded values.

### Use Standardization when

- You want mean 0 and standard deviation 1.
- You are using models such as SVM, PCA, or many optimization-based algorithms.

### Use Robust Scaling when

- Your data contains significant outliers.
- Median and IQR provide a better representation of the central distribution.

<a id="sec-10"></a>
# 10. Robust Scaling

Robust scaling uses the median and interquartile range instead of mean and standard deviation.

Formula:

$$x' = \frac{x - median(x)}{IQR}$$

Where:

`IQR = Q3 - Q1`

This makes Robust Scaling less sensitive to extreme values.

````python
robust_scaler = RobustScaler()

robust_values = robust_scaler.fit_transform(df[["Age"]])

df_robust = df.copy()
df_robust["Age"] = robust_values

df_robust
````

<a id="sec-11"></a>
# 11. What Is an Outlier?

An **outlier** is a value that is unusually far away from the majority of observations.

Example:

`[10, 11, 12, 10, 13, 11, 100]`

`100` may be an outlier.

However, an outlier is not automatically an error.

For example:

- A salary of $500,000 may be valid.
- A person's age of 250 is probably invalid.
- A transaction of $1,000,000 may be legitimate or fraudulent.

Always investigate the business meaning before deleting an outlier.

<a id="sec-12"></a>
# 12. IQR Method

The Interquartile Range method is one of the most common approaches to detecting outliers.

First calculate:

- `Q1` = 25th percentile
- `Q3` = 75th percentile
- `IQR = Q3 - Q1`

Then calculate the boundaries:

$$Lower = Q1 - 1.5 \times IQR$$

$$Upper = Q3 + 1.5 \times IQR$$

Values below the lower boundary or above the upper boundary are potential outliers.

````python
salary = df["Salary"]
print(df["Salary"])
Q1 = salary.quantile(0.25)
Q3 = salary.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', IQR)
print('Lower Bound:', lower_bound)
print('Upper Bound:', upper_bound)
````

<a id="sec-13"></a>
# 13. Detect Outlier Values Using IQR

````python
outlier_values = salary[(salary < lower_bound) | (salary > upper_bound)]

outlier_values
````

<a id="sec-14"></a>
# 14. Detect Complete Outlier Rows

Usually, we want the entire DataFrame rows containing outliers rather than only the values.

````python
salary_outlier_mask = (
    (df['Salary'] < lower_bound) |
    (df['Salary'] > upper_bound)
)

outlier_rows = df[salary_outlier_mask]

outlier_rows
````

<a id="sec-15"></a>
# 15. Create a Reusable IQR Function

A function makes it easier to detect outliers across many columns.

````python
def detect_iqr_outliers(series, multiplier=1.5):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - multiplier * iqr
    upper = q3 + multiplier * iqr
    mask = (series < lower) | (series > upper)
    return mask, lower, upper


mask, lower, upper = detect_iqr_outliers(df['Salary'])

print('Lower:', lower)
print('Upper:', upper)
print('Outliers:')
print(df.loc[mask, ['Employee', 'Salary']])
````

<a id="sec-16"></a>
# 16. Detect Outliers Across Multiple Columns

We can create a separate boolean column for each feature.

````python
numeric_cols = ['Age', 'Experience', 'Salary', 'Score']

outlier_flags = pd.DataFrame(index=df.index)

for column in numeric_cols:
    mask, lower, upper = detect_iqr_outliers(df[column])
    outlier_flags[column + '_outlier'] = mask

outlier_flags
````

### Any Outlier in the Row

````python
df_with_flags = df.copy()
df_with_flags['has_outlier'] = outlier_flags.any(axis=1)

df_with_flags
````

<a id="sec-17"></a>
# 17. Z-Score Outlier Detection

A z-score tells us how many standard deviations a value is away from the mean.

A common rule is:

`abs(z) > 3`

may indicate an outlier.

The threshold is not universal. Depending on the problem, thresholds such as 2, 2.5, 3, or another domain-specific value may be appropriate.

````python
salary_z = (df['Salary'] - df['Salary'].mean()) / df['Salary'].std()

df_zscore = df.copy()
df_zscore['Salary_Z'] = salary_z
df_zscore['Salary_Outlier'] = salary_z.abs() > 3

df_zscore
````

## Visualize the Z-Scores

````python
plt.figure(figsize=(10, 5))
plt.bar(df['Employee'], salary_z)
plt.axhline(3, linestyle='--')
plt.axhline(-3, linestyle='--')
plt.title('Salary Z-Scores')
plt.xlabel('Employee')
plt.ylabel('Z-Score')
plt.xticks(rotation=45)
plt.show()
````

For the mathematical definition of a z-score, remember:

`z = (x - mean) / standard_deviation`.

<a id="sec-18"></a>
# 18. Modified Z-Score and MAD

The traditional z-score uses mean and standard deviation. Both can be strongly affected by extreme outliers.

A more robust alternative uses the **median** and **Median Absolute Deviation (MAD)**.

One common modified z-score formula is:

$$ModifiedZ = 0.6745 \times \frac{x - median(x)}{MAD}$$

A common threshold is approximately `3.5`, although the appropriate threshold depends on the application.

````python
values = df['Salary']

median = values.median()
mad = np.median(np.abs(values - median))

if mad != 0:
    modified_z = 0.6745 * (values - median) / mad
else:
    modified_z = pd.Series(0, index=values.index)

df_modified_z = df.copy()
df_modified_z['Modified_Z'] = modified_z
df_modified_z['Outlier'] = modified_z.abs() > 3.5

df_modified_z
````

<a id="sec-20"></a>
# 20. Removing Outliers

If you determine that an outlier is invalid or should be excluded, you can remove the corresponding row.

Do not remove observations simply because they are statistically unusual. First determine whether the observation is a data error, measurement problem, fraud case, rare but valid event, or an important business observation.

````python
salary_mask, lower, upper = detect_iqr_outliers(df['Salary'])

df_without_salary_outliers = df[~salary_mask].copy()

df_without_salary_outliers
````

<a id="sec-21"></a>
# 21. Capping Outliers

Instead of deleting outliers, we can cap extreme values at the calculated boundaries.

This is sometimes called **Winsorization**.

````python
df_capped = df.copy()

df_capped['Salary'] = df_capped['Salary'].clip(
    lower=lower_bound,
    upper=upper_bound
)

df_capped[['Employee', 'Salary']]
````

# 22. Log Transformation

Highly right-skewed numerical data can sometimes be transformed using a logarithm.

This can reduce the influence of very large values and make a distribution more symmetric.

For positive values, a common transformation is:

`log1p(x) = log(1 + x)`

`np.log1p()` is useful because it also works when the original value is zero.

````python
df_log = df.copy()
df_log['Salary_Log'] = np.log1p(df_log['Salary'])

df_log[['Salary', 'Salary_Log']]
````

<a id="sec-23"></a>
# 23. Effect of an Outlier on Min-Max Scaling

An important property of Min-Max scaling is that extreme values affect the minimum and maximum.

Consider:

`[10, 20, 30, 40, 1000]`

Because `1000` becomes the maximum, most normal values become compressed near 0.

````python
example = pd.Series([10, 20, 30, 40, 1000])

example_normalized = (example - example.min()) / (example.max() - example.min())

pd.DataFrame({
    'Original': example,
    'Normalized': example_normalized
})
````

<a id="sec-24"></a>
# 24. Standardization Can Also Be Affected by Outliers

Standardization uses the mean and standard deviation.

Both statistics can be influenced by extreme observations.

Therefore, if your dataset has strong outliers, consider:

- Investigating the outliers first
- Removing invalid observations
- Capping extreme values
- Transforming the feature
- Using RobustScaler

````python
comparison = pd.DataFrame({
    'Original': df['Salary'],
    'MinMax': MinMaxScaler().fit_transform(df[['Salary']]).flatten(),
    'Standardized': StandardScaler().fit_transform(df[['Salary']]).flatten(),
    'Robust': RobustScaler().fit_transform(df[['Salary']]).flatten()
})

comparison
````

<a id="sec-25"></a>
# 25. Train/Test Data Leakage

This is extremely important when using scaling for machine learning.

### Incorrect approach

Fit the scaler using the complete dataset and then split into training and testing data.

This allows information from the test set to influence the scaling parameters.

### Correct approach

1. Split the data.
2. Fit the scaler only on training data.
3. Transform training data.
4. Transform test data using the same fitted scaler.

````python
from sklearn.model_selection import train_test_split

X = df[['Age', 'Experience', 'Salary', 'Score']]

X_train, X_test = train_test_split(
    X,
    test_size=0.2,
    random_state=42
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('Training data:')
print(X_train_scaled)

print('\nTest data:')
print(X_test_scaled)
````

# 28. Complete IQR Outlier Pipeline

The following function detects outliers for every numeric column and returns a report.

````python
def iqr_outlier_report(dataframe):
    numeric_cols = dataframe.select_dtypes(include=np.number).columns
    rows = []

    for column in numeric_cols:
        series = dataframe[column].dropna()
        q1 = series.quantile(0.25)
        q3 = series.quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        mask = (series < lower) | (series > upper)

        rows.append({
            'Column': column,
            'Q1': q1,
            'Q3': q3,
            'IQR': iqr,
            'Lower_Bound': lower,
            'Upper_Bound': upper,
            'Outlier_Count': mask.sum()
        })

    return pd.DataFrame(rows)


iqr_outlier_report(df)
````

# 29. Complete Data Preparation Example

A realistic workflow can look like this:

```text
Raw Data
   ↓
Inspect Data
   ↓
Handle Missing Values
   ↓
Check Data Types
   ↓
Detect Outliers
   ↓
Investigate Outliers
   ↓
Remove / Cap / Transform if justified
   ↓
Scale Numeric Features
   ↓
Machine Learning Model
```

The exact order can change depending on the problem. In machine-learning workflows, make sure preprocessing is fitted only on training data.

````python
pipeline_df = df.copy()

# Step 1: Identify numeric columns
numeric_cols = pipeline_df.select_dtypes(include=np.number).columns.tolist()

# Step 2: Fill missing numeric values if any
pipeline_df[numeric_cols] = pipeline_df[numeric_cols].fillna(
    pipeline_df[numeric_cols].median()
)

# Step 3: Cap IQR outliers
for column in numeric_cols:
    q1 = pipeline_df[column].quantile(0.25)
    q3 = pipeline_df[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    pipeline_df[column] = pipeline_df[column].clip(lower, upper)

# Step 4: Standardize numeric features
scaler = StandardScaler()
pipeline_df[numeric_cols] = scaler.fit_transform(pipeline_df[numeric_cols])

pipeline_df
````
