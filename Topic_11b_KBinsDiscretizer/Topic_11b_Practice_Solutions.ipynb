### ✍️ Practice Problem 7

Create a `Spending` column with 3 visible groups, e.g. small spenders (~100–200), medium (~1,000–1,300) and large (~5,000–6,000), 3–4 values each. Apply `strategy='kmeans'` with 3 bins, print `bin_edges_`, and check whether each natural group ended up in its own bin.

*(Solution in the **Practice Solutions** section at the end.)*

````python
# ✍️ Practice Problem 7 — write your solution here
````

### Solution 7

````python
spend7 = pd.DataFrame({'Spending': [110, 140, 170, 200, 1000, 1100, 1250, 1300, 5000, 5400, 5900]})
kb7 = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='kmeans')
spend7['Bin'] = kb7.fit_transform(spend7[['Spending']]).astype(int).ravel()
print('KBinsDiscretizer (kmeans) edges:', kb7.bin_edges_[0].round(0))
print(spend7)

# Check against a full K-Means with 10 different starts (as in Section 7)
from sklearn.cluster import KMeans
centres7 = np.sort(KMeans(n_clusters=3, n_init=10, random_state=0).fit(spend7[['Spending']]).cluster_centers_.ravel())
edges7 = (centres7[:-1] + centres7[1:]) / 2
print('\nFull K-Means centres:', centres7.round(0), '→ edges', edges7.round(0))
print('Bins with full K-Means:', np.digitize(spend7['Spending'], edges7).tolist())
````

**Explanation:** You might expect one bin per natural group, but here `KBinsDiscretizer` merges the small and medium spenders into bin 0 and splits the large spenders into two bins. It is the same **local-optimum** pitfall as in Section 7: `KBinsDiscretizer` runs K-Means only once, starting from evenly spaced centres, and the big gap up to ~5,000 pulls the result the wrong way. A full `KMeans(n_init=10)` finds the three natural groups (≈ 155, ≈ 1,160, ≈ 5,430). Lesson: always **inspect** `bin_edges_` and the bins, and don't assume `kmeans` found the natural groups.

Given `skewed = [1, 1, 2, 2, 3, 3, 4, 5, 6, 8, 12, 20, 45, 120, 400]`, which strategy would you **test first**, and why? Apply all three strategies with 4 bins, print the rows per bin for each, and write your conclusion in a comment.

*(Solution in the **Practice Solutions** section at the end.)*

````python
# ✍️ Practice Problem 8 — write your solution here
````

### Solution 8

````python
skewed8 = pd.DataFrame({'x': [1, 1, 2, 2, 3, 3, 4, 5, 6, 8, 12, 20, 45, 120, 400]})
for strat in ['uniform', 'quantile', 'kmeans']:
    kb8 = KBinsDiscretizer(n_bins=4, encode='ordinal', strategy=strat)
    counts8 = np.bincount(kb8.fit_transform(skewed8).astype(int).ravel(), minlength=4)
    print(f'{strat:<9} rows per bin = {counts8.tolist()} | edges = {kb8.bin_edges_[0].round(1)}')
# Test QUANTILE first: the data is strongly right-skewed, so uniform crowds almost everything into bin 0.
````

**Explanation:** For skewed data, quantile is the natural first choice. Uniform gives counts like [13, 1, 0, 1] (almost everything in bin 0, one empty bin). Kmeans gives [12, 1, 1, 1]: it spends three bins on the few extreme values. Quantile gives balanced groups [2, 5, 4, 4] (repeated values prevent a perfect split).

### ✍️ Practice Problem 15

Find the data-leakage problem in this code and rewrite it correctly:
```python
kb = KBinsDiscretizer(n_bins=5, encode='ordinal', strategy='quantile')
X_all_binned = kb.fit_transform(customers[['Age', 'Income']])
X_train, X_test, y_train, y_test = train_test_split(X_all_binned, customers['Churn'], test_size=0.2, random_state=1)
model = LogisticRegression().fit(X_train, y_train)
print(model.score(X_test, y_test))
```
Create a small `customers` DataFrame (at least 40 rows, random data is fine) so your corrected code runs.

*(Solution in the **Practice Solutions** section at the end.)*

````python
# ✍️ Practice Problem 15 — write your solution here
````

### Solution 15

````python
from sklearn.linear_model import LogisticRegression
rng15s = np.random.default_rng(15)
customers15 = pd.DataFrame({'Age': rng15s.integers(18, 70, 120), 'Income': rng15s.integers(20000, 150000, 120)})
customers15['Churn'] = (rng15s.random(120) < 0.3).astype(int)

# PROBLEM: fit_transform on ALL rows before splitting → the test rows helped choose the bin edges (leakage).
# FIX: split first, fit on train, transform both.
X15 = customers15[['Age', 'Income']]; y15 = customers15['Churn']
X15_train, X15_test, y15_train, y15_test = train_test_split(X15, y15, test_size=0.2, random_state=1)
kb15 = KBinsDiscretizer(n_bins=5, encode='ordinal', strategy='quantile').fit(X15_train)
model15 = LogisticRegression().fit(kb15.transform(X15_train), y15_train)
print('Test accuracy:', round(model15.score(kb15.transform(X15_test), y15_test), 3))

# Even better — a Pipeline does it automatically:
pipe15 = Pipeline([('bins', KBinsDiscretizer(n_bins=5, encode='onehot', strategy='quantile')), ('model', LogisticRegression())])
print('Pipeline test accuracy:', round(pipe15.fit(X15_train, y15_train).score(X15_test, y15_test), 3))
````

**Explanation:** The fix is the order of operations: **split → fit on train → transform train and test**. (The target is random here, so accuracy itself is not meaningful — the workflow is the point.)

# Interview Answers

| # | Question | Answer |
|---|---|---|
| 1 | What is discretization? | Converting a continuous numerical variable into a finite number of intervals (bins/groups). |
| 2 | What is binning? | The practical process of choosing interval boundaries and assigning each value to the interval it falls in; used as a synonym for discretization. |
| 3 | What is `KBinsDiscretizer`? | A scikit-learn transformer that learns bin edges for each numeric feature (uniform, quantile or kmeans) and outputs bin IDs or one-hot encodings; works with `fit`/`transform` and pipelines. |
| 4 | What does `n_bins` mean? | The number of bins per feature (an int, or a list with one value per feature). Must be ≥ 2; fewer bins may be used if edges are duplicates (`n_bins_`). |
| 5 | What is `uniform`? | Equal-**width** bins: (max − min) / n_bins. Simple; sensitive to skew and outliers. |
| 6 | What is `quantile`? | Equal-**frequency** bins: edges at percentiles so each bin has about the same number of observations. Good for skewed data. |
| 7 | What is `kmeans`? | Runs 1-D K-Means; edges are midpoints between neighbouring cluster centres, so values close to each other share a bin. Good for natural groups; results depend on data (single run can be a local optimum). |
| 8 | `pd.cut()` vs `pd.qcut()`? | `cut` = equal-width bins or custom edges; `qcut` = equal-frequency (quantile) bins. |
| 9 | `pd.cut()` vs `KBinsDiscretizer`? | `pd.cut` is a Pandas one-off function (supports custom edges and labels); `KBinsDiscretizer` learns edges on training data, reuses them on new data, offers kmeans and one-hot output, and fits into ML pipelines. |
| 10 | `fit()` vs `transform()`? | `fit` learns the bin edges from data; `transform` applies stored edges to (any) data; `fit_transform` does both on the same data (training only). |
| 11 | What is `bin_edges_`? | The learned boundaries: an array per feature with `n_bins + 1` values; bin *i* is `[edge_i, edge_{i+1})`, the last bin includes its right edge. |
| 12 | What is data leakage? | When information unavailable at prediction time — e.g. the test set — influences training or preprocessing, producing over-optimistic evaluation. |
| 13 | Why fit on training data only? | So the edges reflect only "seen" data; the test set then honestly simulates new data. Fitting on all data lets test values shape the boundaries. |
| 14 | Ordinal encoding & its risk? | Each value becomes a single bin ID (0, 1, 2…). Risk: models like linear regression treat the IDs as numbers with equal spacing, which may not be true. |
| 15 | One-hot encoding; `onehot` vs `onehot-dense`? | One 0/1 column per bin. `onehot` returns a memory-efficient sparse matrix; `onehot-dense` returns a regular NumPy array. |
| 16 | Disadvantages of discretization? | Information loss, boundary effects, arbitrary edges, more columns (one-hot), possible overfitting with many bins, often worse for tree models. |
| 17 | When to avoid discretization? | When exact values matter, with tree-based models, with small data, when boundaries would be arbitrary, or when scaling/transformation solves the real problem. |
| 18 | Values outside the training range? | They are clipped: below the first edge → bin 0; above the last edge → last bin. |
| 19 | Why fewer bins than requested? | Duplicate edges (many repeated values, or constant data) are removed with a warning; check `n_bins_`. |
| 20 | Supervised discretization? | Using the target to choose edges, e.g. fitting a small decision tree on one feature and taking its split thresholds (chosen by information gain/Gini). Must be fitted on training data only and validated to avoid overfitting. |

- `Purchased` is the **target**.

### 4. Split X and y · 5. Train/test split

````python
features_f = ['Age', 'Income', 'SpendingScore', 'YearsAsCustomer']
X_f = final_df[features_f]
y_f = final_df['Purchased']
Xf_train, Xf_test, yf_train, yf_test = train_test_split(X_f, y_f, test_size=0.25, stratify=y_f, random_state=42)
print('Train:', Xf_train.shape, '| Test:', Xf_test.shape)
````

### 6–7. Choose a discretization strategy (and why)
- `Income` (skew ≈ 3.0) and `YearsAsCustomer` (skew ≈ 2.5) are clearly **right-skewed** → uniform bins would crowd most customers into bin 0.
- `Age` and `SpendingScore` are roughly evenly spread.
- **Choice: `quantile`** for all four features. It gives balanced bins for the skewed features and is still sensible for the even ones. It is also a good default when we want every bin to have enough customers.

### 8–10. Create 4 bins, ordinal encoding, inspect the boundaries

````python
kb_final = KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='quantile')
kb_final.fit(Xf_train)                                   # learn edges on TRAINING data only
for col, e in zip(features_f, kb_final.bin_edges_):
    print(f'{col:<16} edges: {e.round(1)}')

Xf_train_b = kb_final.transform(Xf_train)
Xf_test_b = kb_final.transform(Xf_test)
print('\nCustomers per bin (train):')
print(pd.DataFrame(Xf_train_b.astype(int), columns=features_f).apply(lambda c: c.value_counts().sort_index()))
````

### 11. Train a simple ML model · 12. Evaluate

````python
tree_f = DecisionTreeClassifier(max_depth=4, random_state=0).fit(Xf_train_b, yf_train)
pred_f = tree_f.predict(Xf_test_b)
proba_f = tree_f.predict_proba(Xf_test_b)[:, 1]
print('Quantile bins + decision tree')
print('  accuracy:', round(accuracy_score(yf_test, pred_f), 3), '| ROC-AUC:', round(roc_auc_score(yf_test, proba_f), 3))
print('  confusion matrix:\n', confusion_matrix(yf_test, pred_f))
````

### 13–14. Try another strategy and compare (with pipelines + cross-validation for a fair comparison)

````python
cv_f = StratifiedKFold(5, shuffle=True, random_state=0)
candidates_f = {
    'Continuous (no bins) + tree': DecisionTreeClassifier(max_depth=4, random_state=0),
    'Quantile bins + tree': Pipeline([('bins', KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='quantile')),
                                      ('tree', DecisionTreeClassifier(max_depth=4, random_state=0))]),
    'Uniform bins + tree': Pipeline([('bins', KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='uniform')),
                                     ('tree', DecisionTreeClassifier(max_depth=4, random_state=0))]),
    'Continuous + LogReg (scaled)': Pipeline([('scale', StandardScaler()), ('lr', LogisticRegression(max_iter=1000))]),
    'Quantile one-hot + LogReg': Pipeline([('bins', KBinsDiscretizer(n_bins=4, encode='onehot', strategy='quantile')),
                                           ('lr', LogisticRegression(max_iter=1000))]),
}
rows_f = []
for name, m in candidates_f.items():
    cv_auc = cross_val_score(m, Xf_train, yf_train, cv=cv_f, scoring='roc_auc')
    m.fit(Xf_train, yf_train)
    rows_f.append({'Model': name, 'CV ROC-AUC': cv_auc.mean(), 'CV std': cv_auc.std(),
                   'Test ROC-AUC': roc_auc_score(yf_test, m.predict_proba(Xf_test)[:, 1]),
                   'Test accuracy': accuracy_score(yf_test, m.predict(Xf_test))})
pd.DataFrame(rows_f).set_index('Model').sort_values('CV ROC-AUC', ascending=False).round(3)
````

### 15. Did discretization help? — interpreting the results

(The numbers are reproducible because of the fixed random seeds.)

| Model | CV ROC-AUC (± std) | What it tells us |
|---|---|---|
| Continuous + LogReg (scaled) | ≈ 0.72 ± 0.06 | **Best CV score** |
| Quantile one-hot + LogReg | ≈ 0.69 ± 0.07 | Slightly lower, within noise |
| Quantile bins + tree | ≈ 0.67 ± 0.06 | Clearly better than the tree on raw values |
| Uniform bins + tree | ≈ 0.65 ± 0.06 | A bit behind quantile (skewed features) |
| Continuous + tree | ≈ 0.62 ± 0.09 | Worst and most unstable |

**Conclusions**
- **For the decision tree, discretization helped.** With only 225 training rows, a tree on raw values picks very specific, noisy thresholds that don't generalise. Four bins per feature act as **regularization**: the tree can only split at a few sensible places. Quantile bins did a little better than uniform bins, which matches the skewed `Income` and `YearsAsCustomer` (within noise, though).
- **For logistic regression, discretization did not help.** The scaled continuous features scored slightly higher. The one-hot bins can express the step effects, but they lose detail inside each bin, and here that roughly cancels the benefit.
- **Best overall:** logistic regression on continuous features, with the binned models close behind. The differences between the top models are about one CV standard deviation, so they are not decisive.
- **Test-set numbers** (75 rows) bounce around a lot; use the **CV** columns for decisions.

**Final answer:** *"Quantile discretization was chosen because two features are strongly right-skewed. It clearly improved the decision tree, mainly by regularizing it, but it did not beat logistic regression on scaled continuous features. The bins remain useful for explaining the model, e.g. 'customers in the top income quartile buy more'."*
