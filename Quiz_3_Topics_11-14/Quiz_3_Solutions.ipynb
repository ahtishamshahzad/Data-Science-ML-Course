### Topic 11 — Feature Engineering

**Q1.** Hour 23 and hour 0 are only 1 hour apart. Which encoding captures that?

- **A)** Keep the raw hour number
- **B)** Cyclical sin/cos encoding  ✅
- **C)** Always one-hot
- **D)** Log transform

> ✅ **Answer: B** — sin(2π·h/24) and cos(2π·h/24) place 23 and 0 next to each other on a circle.

**Q2.** You create `CTR = clicks / impressions`, but some rows have 0 impressions. What happens?

- **A)** Nothing, it's fine
- **B)** The result is negative
- **C)** Division by zero gives inf/NaN — handle it safely  ✅
- **D)** pandas sets it to 0 automatically

> ✅ **Answer: C** — Replace 0 with NaN first (or use `np.where`) and decide how to fill those rows.

**Q3.** What is a **missing-indicator** feature?

- **A)** It replaces NaN with 0
- **B)** A 0/1 column that records whether the value was missing  ✅
- **C)** It drops rows with NaN
- **D)** It is the target

> ✅ **Answer: B** — Sometimes *being missing* is informative, so the model gets that as its own feature.

**Q4.** Engineering features with statistics of the **whole** dataset before the train/test split risks…

- **A)** Underfitting
- **B)** Slower code only
- **C)** Data leakage  ✅
- **D)** Nothing

> ✅ **Answer: C** — Statistics must come from training data only, otherwise test information leaks into the features.

### Topic 11b — KBinsDiscretizer

**Q5.** What does `strategy='quantile'` do?

- **A)** Equal-width bins
- **B)** Bins from k-means clusters
- **C)** Bins with about the same number of samples  ✅
- **D)** Random bins

> ✅ **Answer: C** — Quantile bins are equal-frequency.

**Q6.** What does `encode='ordinal'` output?

- **A)** One column per bin
- **B)** Strings
- **C)** Floats between 0 and 1
- **D)** One integer bin number per feature  ✅

> ✅ **Answer: D** — Each value is replaced by the index of its bin (0, 1, 2 …).

**Q7.** On which data should you `fit` a `KBinsDiscretizer`?

- **A)** Training data only, then `transform` the test data  ✅
- **B)** All data
- **C)** Test data only
- **D)** Train and test separately

> ✅ **Answer: A** — Bin edges are learned parameters — learn them from training data only.

**Q8.** With `strategy='uniform'` and one extreme outlier, what usually happens?

- **A)** Most values end up crowded in one or two bins  ✅
- **B)** Every bin has the same count
- **C)** An error is raised
- **D)** Nothing changes

> ✅ **Answer: A** — Equal-width bins stretch to cover the outlier, so the normal values share very few bins.

### Topic 12 — Correlation & Multicollinearity

**Q9.** A Pearson correlation of **−0.9** means…

- **A)** a weak relationship
- **B)** a strong negative linear relationship  ✅
- **C)** no relationship
- **D)** X causes Y to fall

> ✅ **Answer: B** — Close to −1: as one variable goes up the other goes down, almost on a straight line. Correlation ≠ causation.

**Q10.** The relationship is monotonic but **not linear** (e.g. it curves). Which measure fits better?

- **A)** Pearson
- **B)** Spearman  ✅
- **C)** Covariance
- **D)** None

> ✅ **Answer: B** — Spearman uses ranks, so it measures any consistently increasing/decreasing relationship.

**Q11.** A feature has a **VIF above 10**. That suggests…

- **A)** strong multicollinearity  ✅
- **B)** no correlation
- **C)** missing values
- **D)** a perfect model

> ✅ **Answer: A** — VIF measures how well a feature is explained by the other features. High VIF = redundant information.

**Q12.** Ice-cream sales and drownings are correlated because both rise in summer. This is an example of…

- **A)** causation
- **B)** multicollinearity
- **C)** a confounding variable  ✅
- **D)** sampling error

> ✅ **Answer: C** — Temperature (summer) drives both — a hidden third variable.

### Topic 13 — Data Transformation & Skewness

**Q13.** In a **right-skewed** distribution, usually…

- **A)** mean < median
- **B)** mean = median
- **C)** all values are negative
- **D)** mean > median  ✅

> ✅ **Answer: D** — The long right tail pulls the mean up above the median.

**Q14.** The **Box-Cox** transformation requires…

- **A)** strictly positive values  ✅
- **B)** negative values
- **C)** categorical data
- **D)** already-normal data

> ✅ **Answer: A** — Box-Cox uses logs/powers that are undefined for 0 and negative numbers.

**Q15.** What is the advantage of **Yeo-Johnson** over Box-Cox?

- **A)** It also works with zero and negative values  ✅
- **B)** It is always faster
- **C)** It has no parameters
- **D)** It only works on categories

> ✅ **Answer: A** — That is why `PowerTransformer` uses Yeo-Johnson by default.

**Q16.** `np.log(0)` gives `-inf`. What is the usual fix for count data with zeros?

- **A)** Use `np.log(x) * 0`
- **B)** Drop the column
- **C)** Use `np.log1p(x)` (log of 1 + x)  ✅
- **D)** Use `np.exp(x)`

> ✅ **Answer: C** — `log1p(0) = 0`, so zeros stay valid.

<a id="a-5"></a>
### Topic 14 — Statistics for Data Science

**Q17.** A test gives **p = 0.03** and you use α = 0.05. What do you conclude?

- **A)** Fail to reject H₀
- **B)** H₁ is proven true
- **C)** There is a 3% chance that H₀ is true
- **D)** Reject H₀  ✅

> ✅ **Answer: D** — p < α, so the result is statistically significant. A p-value is not the probability that H₀ is true.

**Q18.** A **Type I error** is…

- **A)** failing to reject a false H₀
- **B)** a biased sample
- **C)** using a large sample
- **D)** rejecting a true H₀ (a false positive)  ✅

> ✅ **Answer: D** — Type II is the opposite: missing a real effect (false negative).

**Q19.** What does the **Central Limit Theorem** say?

- **A)** All data is normally distributed
- **B)** For large n, the sampling distribution of the mean is approximately normal  ✅
- **C)** The median equals the mean
- **D)** The variance becomes 0

> ✅ **Answer: B** — Even for skewed populations, averages of large samples are approximately normal.

**Q20.** You want to compare the means of **three or more** groups. Which test?

- **A)** t-test
- **B)** Chi-square test
- **C)** Pearson correlation
- **D)** ANOVA  ✅

> ✅ **Answer: D** — A t-test compares two groups; ANOVA compares several at once.

### Task 1 — Topic 11 — Date features

Convert `OrderDate` to datetime and add two columns to `orders`: `month` (1–12) and `is_weekend` (1 if Saturday/Sunday, else 0).

```python
orders = pd.DataFrame({'OrderDate': ['2025-01-03', '2025-01-04', '2025-02-09', '2025-03-12']})

orders['OrderDate'] = pd.to_datetime(orders['OrderDate'])
orders['month'] = orders['OrderDate'].dt.month
orders['is_weekend'] = (orders['OrderDate'].dt.dayofweek >= 5).astype(int)   # Mon=0 ... Sun=6
orders

check(1)
```

### Task 2 — Topic 11b — Discretize ages

Use `KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform')` on `ages` and store the bin numbers as a 1-D integer array `age_bins`.

```python
from sklearn.preprocessing import KBinsDiscretizer

ages = np.array([18, 22, 25, 31, 38, 45, 52, 60, 66, 72]).reshape(-1, 1)

kb = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform')
age_bins = kb.fit_transform(ages).astype(int).ravel()
print(kb.bin_edges_[0], age_bins)

check(2)
```

<a id="task-3"></a>
### Task 3 — Topic 12 — Pearson vs Spearman

Compute the **Pearson** and **Spearman** correlation between `x` and `y` and store them in `pearson` and `spearman` (floats).

```python
x = pd.Series([1, 2, 3, 4, 5, 6, 7, 8])
y = x ** 3

pearson = x.corr(y)                      # method='pearson' is the default
spearman = x.corr(y, method='spearman')
print(round(pearson, 3), spearman)       # Spearman = 1.0: perfectly monotonic

check(3)
```

<a id="task-4"></a>
### Task 4 — Topic 13 — Reduce skewness

Create `log_income = np.log1p(income)`, then store `skew_before` (skew of `income`) and `skew_after` (skew of `log_income`).

```python
income = pd.Series([18, 20, 22, 25, 27, 30, 33, 38, 45, 60, 90, 250, 900])

log_income = np.log1p(income)
skew_before = income.skew()
skew_after = log_income.skew()
print(f'skew before {skew_before:.2f} -> after {skew_after:.2f}')

check(4)
```

<a id="task-5"></a>
### Task 5 — Topic 14 — Two-sample t-test

Run an independent two-sample t-test (`scipy.stats.ttest_ind`) between `group_a` and `group_b`. Store the p-value in `p_value` and set `decision` to `'reject'` if p < 0.05, else `'fail to reject'`.

```python
from scipy import stats

group_a = [72, 75, 78, 71, 74, 77, 73, 76]
group_b = [80, 83, 79, 85, 82, 84, 81, 86]

t_stat, p_value = stats.ttest_ind(group_a, group_b)
decision = 'reject' if p_value < 0.05 else 'fail to reject'
print(f't = {t_stat:.2f}, p = {p_value:.6f} -> {decision} H0')

check(5)
```
