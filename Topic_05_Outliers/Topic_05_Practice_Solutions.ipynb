### Solution — Practice Problem 1 — Basic IQR Detection

````python
p1 = pd.DataFrame({'values': [10, 12, 13, 15, 16, 17, 18, 20, 100]})
q1, q3 = p1['values'].quantile(0.25), p1['values'].quantile(0.75)
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers_p1 = p1[(p1['values'] < lower) | (p1['values'] > upper)]
print(f'Q1={q1}, Q3={q3}, IQR={iqr}, lower={lower}, upper={upper}')
print('Outliers:', outliers_p1['values'].tolist(), '| count:', len(outliers_p1))
````

**Explanation:** Q1 = 13, Q3 = 18, IQR = 5 → bounds 5.5 and 25.5. Only 100 lies outside.

# 38. Practice Problem 3 — Outlier Percentage

Use this dataset:

```python
values = [10, 11, 12, 13, 14, 15, 16, 17, 100, 200]
```

Tasks:

1. Detect outliers using IQR.
2. Count the outliers.
3. Calculate the percentage of outliers.

### Solution — Practice Problem 3 — Outlier Percentage

````python
p3 = pd.Series([10, 11, 12, 13, 14, 15, 16, 17, 100, 200])
q1, q3 = p3.quantile([0.25, 0.75]); iqr = q3 - q1
out3 = p3[(p3 < q1 - 1.5 * iqr) | (p3 > q3 + 1.5 * iqr)]
print('Outliers:', out3.tolist(), '| count:', len(out3), f'| percentage: {len(out3) / len(p3) * 100:.1f}%')
````

**Explanation:** Both 100 and 200 are outliers → 2 of 10 values = 20%. A high percentage is a warning sign: maybe the data contains two groups rather than errors.

# 39. Practice Problem 4 — Remove Outliers

Use:

```python
scores = [55, 60, 62, 65, 67, 70, 72, 75, 78, 200]
```

Tasks:

1. Detect the outlier.
2. Create a new DataFrame without the outlier.
3. Compare the number of rows before and after cleaning.

### Solution — Practice Problem 4 — Remove Outliers

````python
scores = pd.DataFrame({'score': [55, 60, 62, 65, 67, 70, 72, 75, 78, 200]})
q1, q3 = scores['score'].quantile([0.25, 0.75]); iqr = q3 - q1
mask = (scores['score'] < q1 - 1.5 * iqr) | (scores['score'] > q3 + 1.5 * iqr)
scores_clean = scores[~mask]
print('Outlier:', scores.loc[mask, 'score'].tolist(), '| rows before:', len(scores), '| after:', len(scores_clean))
````

**Explanation:** 200 is removed. A score of 200 is impossible if scores go up to 100, so it's a data error and removing it is reasonable.

# 40. Practice Problem 5 — Median Replacement

Dataset:

```python
income = [30000, 32000, 34000, 35000, 37000, 40000, 450000]
```

Tasks:

1. Detect the outlier using IQR.
2. Replace the outlier with `NaN`.
3. Calculate the median of the remaining values.
4. Replace the `NaN` with the median.
5. Compare the original and treated DataFrames.

### Solution — Practice Problem 5 — Median Replacement

````python
inc = pd.DataFrame({'income': [30000, 32000, 34000, 35000, 37000, 40000, 450000]})
q1, q3 = inc['income'].quantile([0.25, 0.75]); iqr = q3 - q1
treated = inc.copy()
out_mask = (treated['income'] < q1 - 1.5 * iqr) | (treated['income'] > q3 + 1.5 * iqr)
treated.loc[out_mask, 'income'] = np.nan
median_rest = treated['income'].median()
treated['income'] = treated['income'].fillna(median_rest)
print('Median of the remaining values:', median_rest)
pd.DataFrame({'original': inc['income'], 'treated': treated['income']})
````

**Explanation:** 450,000 → NaN → replaced by the median of the other six values (34,500).

### Solution — Practice Problem 6 — Capping Outliers

````python
pr = pd.DataFrame({'price': [100, 110, 120, 130, 140, 150, 160, 1000]})
q1, q3 = pr['price'].quantile([0.25, 0.75]); iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print(f'Q1={q1}, Q3={q3}, IQR={iqr}, bounds=({lower}, {upper})')
pr['price_capped'] = pr['price'].clip(lower=lower, upper=upper)
pr
````

**Explanation:** Q1 = 117.5, Q3 = 152.5, IQR = 35 → upper bound 205. 1000 is capped at 205; all other prices are unchanged.

Tasks:

3. Add the Z-score as a new DataFrame column.
4. Use `abs(z_score) > 3` to identify potential outliers.
5. Compare your result with the IQR method.

### Solution — Practice Problem 7 — Z-Score

````python
from scipy.stats import zscore
p7 = pd.DataFrame({'value': [10, 11, 12, 13, 14, 15, 16, 100]})
p7['z_score'] = zscore(p7['value'])                      # scipy uses the population std (ddof=0)
print(p7.round(3))
print('|z| > 3:', p7.loc[p7['z_score'].abs() > 3, 'value'].tolist())
q1, q3 = p7['value'].quantile([0.25, 0.75]); iqr = q3 - q1
print('IQR    :', p7.loc[(p7['value'] < q1 - 1.5 * iqr) | (p7['value'] > q3 + 1.5 * iqr), 'value'].tolist())
````

**Explanation:** The z-score of 100 is only about 2.6, so the |z| > 3 rule misses it. IQR catches it. With few data points, one extreme value inflates the standard deviation and hides itself. IQR is more reliable here.

Create this DataFrame:

```python
df = pd.DataFrame({
    "age": [21, 22, 23, 24, 25, 26, 27, 100],

    "experience": [1, 2, 2, 3, 4, 5, 6, 30]
```

2. Loop through every numerical column.
3. Calculate Q1, Q3, and IQR.

### Solution — Practice Problem 8 — Multiple Columns

````python
p8 = pd.DataFrame({
    "age": [21, 22, 23, 24, 25, 26, 27, 100],
    "salary": [30000, 32000, 34000, 35000, 37000, 39000, 42000, 500000],
    "experience": [1, 2, 2, 3, 4, 5, 6, 30]
})
for column in p8.select_dtypes(include='number').columns:
    q1, q3 = p8[column].quantile([0.25, 0.75]); iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    vals = p8.loc[(p8[column] < lower) | (p8[column] > upper), column].tolist()
    print(f'{column:<11} Q1={q1:<9} Q3={q3:<9} IQR={iqr:<9} outliers={vals}')
````

**Explanation:** Each column gets its own bounds. The last row is an outlier in all three columns.

# 44. Practice Problem 9 — Write Your Own Function

Create a function named:

```python
find_outliers(df, column)
```

The function should:

1. Calculate Q1.
2. Calculate Q3.
3. Calculate IQR.
4. Calculate lower and upper bounds.
5. Return the outlier rows.

Test your function on at least three different columns.

### Solution — Practice Problem 9 — Write Your Own Function

````python
def find_outliers(df, column):
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return df[(df[column] < lower) | (df[column] > upper)]

for col in ['age', 'salary', 'experience']:
    print(col, '→ outlier rows:', find_outliers(p8, col).index.tolist())
````

**Explanation:** The function returns whole rows, so you can see the other values of the unusual record.

Create the following dataset:

```python
employees = pd.DataFrame({

    "age": [22, 25, 28, 30, 32, 35, 38, 120],

    "experience": [1, 2, 3, 5, 7, 8, 10, 50]
})
```

Complete the following workflow:

1. Inspect the DataFrame.
2. Generate descriptive statistics.
3. Identify numerical columns.
4. Detect outliers using IQR.
5. Count outliers for every column.

### Solution — Practice Problem 10 — Real-World Employee Dataset

````python
employees_p10 = pd.DataFrame({
    "name": ["Ali", "Sara", "John", "Maria", "David", "Emma", "Tom", "Ayesha"],
    "age": [22, 25, 28, 30, 32, 35, 38, 120],
    "salary": [40000, 45000, 50000, 55000, 60000, 65000, 70000, 600000],
    "experience": [1, 2, 3, 5, 7, 8, 10, 50]
})
employees_p10.info()
print(employees_p10.describe().round(1))
num_cols = employees_p10.select_dtypes(include='number').columns

summary = []
for col in num_cols:
    out = find_outliers(employees_p10, col)
    summary.append({'column': col, 'outliers': len(out), 'percent': round(len(out) / len(employees_p10) * 100, 1)})
print(pd.DataFrame(summary))

fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))
for ax, col in zip(axes, num_cols):
    ax.boxplot(employees_p10[col]); ax.set_title(col)
plt.tight_layout(); plt.show()

# Treatment: age 120 and 50 years of experience are impossible for this person → data errors → set to NaN and impute with the median.
# The 600,000 salary could be real (a senior manager) → keep it, but cap it so it doesn't dominate averages.
cleaned_p10 = employees_p10.copy()
cleaned_p10.loc[cleaned_p10['age'] > 100, 'age'] = np.nan
cleaned_p10.loc[cleaned_p10['experience'] > cleaned_p10['age'].fillna(0) - 16, 'experience'] = np.nan
cleaned_p10[['age', 'experience']] = cleaned_p10[['age', 'experience']].fillna(cleaned_p10[['age', 'experience']].median())
q1, q3 = cleaned_p10['salary'].quantile([0.25, 0.75])
cleaned_p10['salary'] = cleaned_p10['salary'].clip(upper=q3 + 1.5 * (q3 - q1))
pd.DataFrame({'original_mean': employees_p10[num_cols].mean(), 'cleaned_mean': cleaned_p10[num_cols].mean(),
              'original_median': employees_p10[num_cols].median(), 'cleaned_median': cleaned_p10[num_cols].median()}).round(1)
````

**Explanation:** Different treatments for different causes: **impossible** values (age 120, 50 years of experience) are errors, so replace them. A **plausible but extreme** salary is kept but capped. Means move a lot after treatment, medians barely do.

# 46. Challenge — Build an Outlier Treatment Pipeline

Build a reusable workflow that accepts:

```python
df
column
```

and performs:

1. Q1 calculation
2. Q3 calculation
3. IQR calculation
4. Lower-bound calculation
5. Upper-bound calculation
6. Outlier detection
7. Outlier count
8. Outlier percentage
9. Optional removal
10. Optional capping
11. Return of the cleaned DataFrame

Try to design the function so that the treatment can be selected using a parameter such as:

```python
method="remove"
```

or:

```python
method="cap"
```

Do this yourself before looking for a more advanced implementation.

### Solution — Challenge — Outlier Treatment Pipeline

````python
def treat_outliers(df, column, method='cap', k=1.5):
    '''Detect IQR outliers in one column and optionally remove or cap them.

    method: 'report' (no change), 'remove', or 'cap'
    Returns (cleaned DataFrame, summary dict).
    '''
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - k * iqr, q3 + k * iqr
    mask = (df[column] < lower) | (df[column] > upper)
    summary = {'column': column, 'Q1': q1, 'Q3': q3, 'IQR': iqr, 'lower': lower, 'upper': upper,
               'outliers': int(mask.sum()), 'percent': round(float(mask.mean()) * 100, 1), 'method': method}
    if method == 'remove':
        cleaned = df[~mask].copy()
    elif method == 'cap':
        cleaned = df.copy()
        cleaned[column] = cleaned[column].clip(lower, upper)
    elif method == 'report':
        cleaned = df.copy()
    else:
        raise ValueError("method must be 'report', 'remove' or 'cap'")
    return cleaned, summary

for m in ['report', 'remove', 'cap']:
    out_df, info = treat_outliers(employees_p10, 'salary', method=m)
    print(m, '→ rows:', len(out_df), '| max salary:', out_df['salary'].max(), '|', {k: info[k] for k in ['outliers', 'percent']})
````

**Explanation:** One function, three behaviours. The original DataFrame is never modified (Best Practice 3), and the summary makes the decision traceable.

### Solution — Final Question — Should outliers always be removed?

````python
# Answer (in words):
# No. An outlier is only a value that is statistically unusual.
# - If it is an ERROR (age 150, negative quantity, typo) → correct it, set it to NaN and impute, or remove the row.
# - If it is a VALID rare value (a CEO salary, a big order, a real fraud case) → keep it; it may be the most
#   important information. Reduce its influence instead (capping, log transform, robust scaling, robust models).
# - If it belongs to a DIFFERENT GROUP → analyse that group separately.
# Always investigate the cause, keep the original data, and compare results before and after treatment.
````

**Explanation:** Detection and treatment are separate decisions. Domain knowledge decides the treatment.
