## 21. Forward Fill — `ffill()`

Forward fill uses the previous available value. It is useful for ordered or time-series data when carrying the latest known value forward makes sense.

````python
temperature = pd.DataFrame({
    "Temperature": [30, 31, np.nan, np.nan, 35]
})
print("Before:")
print(temperature)

temperature_ffill = temperature.copy()
temperature_ffill["Temperature"] = temperature_ffill["Temperature"].ffill()
print("\nAfter forward fill:")
print(temperature_ffill)
````

## 28. Group-Based Imputation

Sometimes a global statistic is inappropriate. You can calculate the median within each group.

````python
group_data = pd.DataFrame({
    "City": ["Lahore", "Lahore", "Karachi", "Karachi", "Lahore", "Karachi"],
    "Salary": [50000, np.nan, 70000, 75000, 60000, np.nan]
})

print("Before:")
print(group_data)

group_data["Salary"] = group_data.groupby("City")["Salary"].transform(
    lambda x: x.fillna(x.median())
)

print("\nAfter group-based median:")
print(group_data)
````

## 31. Complete Missing-Value Workflow

A practical workflow is:

1. Inspect
2. Detect
3. Quantify
4. Understand why data is missing
5. Choose a strategy
6. Clean
7. Verify

````python
workflow_df = df.copy()

print("Initial info:")
workflow_df.info()

print("\nMissing counts:")
print(workflow_df.isna().sum())

print("\nMissing percentages:")
print(workflow_df.isna().mean() * 100)

workflow_df["Age"] = workflow_df["Age"].fillna(workflow_df["Age"].median())
workflow_df["Salary"] = workflow_df["Salary"].fillna(workflow_df["Salary"].median())
workflow_df["Department"] = workflow_df["Department"].fillna(workflow_df["Department"].mode()[0])

print("\nCleaned data:")
print(workflow_df)

print("\nRemaining missing values:")
print(workflow_df.isna().sum())
````

Before choosing *how* to fill, ask *why* the value is missing. Statisticians use three categories:

| Type | Meaning | Example | What it means for you |
|---|---|---|---|
| **MCAR** — Missing Completely At Random | missingness has nothing to do with any value | a lab sample tube broke | dropping or simple filling does not bias results (you only lose data) |
| **MAR** — Missing At Random | missingness depends on *other observed* columns | older customers skip the "Instagram handle" field | fill using the other columns (group-based, KNN, iterative) |
| **MNAR** — Missing Not At Random | missingness depends on the *missing value itself* | people with very high income refuse to report income | any fill is biased; keep a **"was missing" flag** so the model can learn from it |

You usually cannot *prove* which type you have, but you can look for clues: compare other columns for rows with and
without the missing value.

````python
rng = np.random.default_rng(0)
people = pd.DataFrame({"Age": rng.integers(20, 70, 200),
                       "Income": rng.normal(80, 20, 200).round(1)})
# MAR example: older people are more likely to skip the Income question
skip = rng.random(200) < (people["Age"] - 20) / 60
people.loc[skip, "Income"] = np.nan

people["Income_missing"] = people["Income"].isna()
print(people.groupby("Income_missing")["Age"].mean().round(1))
print("\nMissing rows are clearly older -> the missingness depends on Age (MAR), not pure chance.")
````

## 32c. Scikit-learn Imputers and the Train/Test Rule

Pandas `fillna()` is perfect for exploring. For **machine learning**, use scikit-learn imputers because they
*remember* what they learned on the training data and apply the same values to new data:

| Imputer | What it does | When to use |
|---|---|---|
| `SimpleImputer(strategy='mean' / 'median' / 'most_frequent' / 'constant')` | one value per column | default choice |
| `SimpleImputer(..., add_indicator=True)` / `MissingIndicator` | also adds a 0/1 "was missing" column | when missingness itself may carry information (MNAR) |
| `KNNImputer(n_neighbors=k)` | average of the k most similar rows | MAR, numeric data, small/medium datasets |
| `IterativeImputer` (experimental) | predicts each column from the others, repeatedly | MAR with strong relationships between columns |

⚠️ **Leakage rule:** learn the fill values from the **training** data only (`fit` on train), then `transform` both
train and test. Filling with the mean of the *whole* dataset lets test information leak into training.
Inside a `Pipeline` (Topic 17) this happens automatically.

````python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer, KNNImputer
rng = np.random.default_rng(0)
people = pd.DataFrame({"Age": rng.integers(20, 70, 200),
                       "Income": rng.normal(80, 20, 200).round(1)})
# print(people)
X = people[["Age", "Income"]]
X_train, X_test = train_test_split(X, test_size=0.25, random_state=42)

imp = SimpleImputer(strategy="median", add_indicator=True)
imp.fit(X_train)                                     # learns medians from TRAIN only
print("learned medians:", imp.statistics_)
print(imp.get_feature_names_out())

X_test_filled = pd.DataFrame(imp.transform(X_test), columns=imp.get_feature_names_out(), index=X_test.index)
X_test_filled.head()
````

````python
# KNNImputer: each missing Income is filled with the average Income of the 5 rows with the most similar Age
knn = KNNImputer(n_neighbors=5)
knn.fit(X_train)
knn_filled = pd.DataFrame(knn.transform(X_test), columns=X.columns, index=X_test.index)

compare = pd.DataFrame({"Age": X_test["Age"],
                        "median_fill": X_test_filled["Income"],
                        "knn_fill": knn_filled["Income"].round(1)})[X_test["Income"].isna()]
print(compare.head(8))
print("\nMedian fill gives every row the same value; KNN adapts the value to each row's Age.")
````

# 36. Key Concepts to Remember

1. **Detect before cleaning.**
2. Use `isna()` / `isnull()` to find missing data.
3. Use `isna().sum()` to count it.
4. Use percentages to understand the scale of missingness.
5. Use `dropna()` only when removing data is justified.
6. Use `fillna()` when retaining rows is appropriate.
7. Mean is sensitive to outliers.
8. Median is more robust to outliers.
9. Mode is commonly useful for categorical data.
10. `ffill()` and `bfill()` are appropriate when order matters.
11. `interpolate()` estimates numerical values between observations.
12. Different columns can require different strategies.
13. Group-based imputation can preserve group-level differences.
14. Do not blindly replace missing values with `0`.
15. Always verify the cleaned dataset afterward.
