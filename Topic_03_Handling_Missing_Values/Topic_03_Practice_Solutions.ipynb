```python
import pandas as pd
import numpy as np

def fresh_df():
    return pd.DataFrame({
        "Name": ["Ali", "Ahmed", "Sara", "Usman", "Ayesha"],
        "Age": [25, np.nan, 30, 28, np.nan],
        "Salary": [50000, 60000, np.nan, 55000, 70000],
        "City": ["Lahore", "Karachi", None, "Islamabad", "Lahore"]
    })

df = fresh_df()
# Practice 1-7: detect, count, percentage, report
print(df.head(3), df.shape, df.columns.tolist(), df.dtypes, sep="\n\n")
print(df.isna())
print(df.isna().sum())
print("total missing:", df.isna().sum().sum())
report = pd.DataFrame({"Missing Count": df.isna().sum(),
                       "Missing Percentage": (df.isna().mean() * 100).round(1)})
report
```

## ✅ Solution — Practice 8–14 — Selecting and dropping

```python
df = fresh_df()
print(df[df.isna().any(axis=1)], "\n")        # 8 incomplete rows
print(df[df.notna().all(axis=1)], "\n")       # 9 complete rows
complete = df.dropna()                          # 10
print(df.dropna(subset=["Age"]), "\n")        # 11
print(df.dropna(axis=1), "\n")                # 12 only complete columns (Name)

p13 = pd.DataFrame({"A": [1, np.nan, np.nan], "B": [2, np.nan, 5], "C": [3, np.nan, np.nan]})
print(p13.dropna(how="any"), "\n")   # drops rows 1 and 2
print(p13.dropna(how="all"), "\n")   # drops only the fully empty row 1

print(df.dropna(thresh=3))             # 14 keep rows with >= 3 non-missing values
```

## ✅ Solution — Practice 26–30 — Column strategies, groups, filtering, assignment

```python
df = fresh_df()
p26 = df.copy()
p26["Age"] = p26["Age"].fillna(p26["Age"].median())
p26["Salary"] = p26["Salary"].fillna(p26["Salary"].mean())
p26["City"] = p26["City"].fillna(p26["City"].mode()[0])
print(p26, "\n")

num_cols = df.select_dtypes(include="number").columns                 # 27 no hard-coded names
p27 = df.copy()
p27[num_cols] = p27[num_cols].fillna(p27[num_cols].mean())
print(p27, "\n")

dept = pd.DataFrame({"Department": ["IT", "IT", "HR", "HR", "IT", "HR"],
                     "Salary": [90000, np.nan, 60000, np.nan, 110000, 64000]})
dept["Salary"] = dept.groupby("Department")["Salary"].transform(lambda s: s.fillna(s.median()))   # 28
print(dept, "\n")

print(df[df["Salary"].isna()], "\n")                                 # 29.1
print(df[df["Salary"].notna()], "\n")                                # 29.2

p30 = df.copy()                                                        # 30
p30["Age"] = p30["Age"].fillna(p30["Age"].median())
print("original still has", df["Age"].isna().sum(), "missing ages; copy has", p30["Age"].isna().sum())
```

For each situation, decide the technique you would use and explain why in comments.

1. `Salary` has 2% missing values and several very high salaries.
2. `City` has 1% missing values.
3. Hourly sensor temperatures have gaps between known readings.

## 33 — Advanced decision-making (suggested answers)

1. **Salary, 2% missing, some very high values** → median (robust to the outliers), or group median by Department.
2. **City, 1% missing** → mode, or `"Unknown"` if guessing would be misleading.
3. **Hourly sensor temperatures with gaps** → `interpolate()` (time-ordered, smooth signal); `ffill` for short gaps.
4. **Column 95% missing** → usually drop it; if the fact of being missing matters, keep only a 0/1 "has value" flag.
5. **Row with 1 of 10 values** → drop the row (`dropna(thresh=...)`) — filling 9 values would mostly be invention.
6. **Department missing and must not be guessed** → fill with `"Unknown"` (a separate category), never the mode.

```python
rng = np.random.default_rng(3)
n = 16
exam = pd.DataFrame({
    "Name": [f"Emp{i:02d}" for i in range(1, n + 1)],
    "Age": rng.integers(22, 58, n).astype(float),
    "Salary": rng.normal(90000, 25000, n).round(-3),
    "Department": rng.choice(["IT", "HR", "Finance"], n),
    "City": rng.choice(["Lahore", "Karachi", "Islamabad"], n),
    "Experience": rng.integers(0, 25, n).astype(float),
    "Performance": rng.normal(3.5, 0.6, n).round(1),
}).astype({"Department": object, "City": object})
exam.loc[15, "Salary"] = 450000                             # one extreme salary
for col, rows in {"Age": [1, 7], "Salary": [3, 9, 12], "Department": [4, 10],
                  "City": [2, 13], "Experience": [5], "Performance": [6, 11]}.items():
    exam.loc[rows, col] = np.nan

# Part A — detection
before = pd.DataFrame({"count": exam.isna().sum(), "pct": (exam.isna().mean() * 100).round(1)})
print(before)
print("incomplete rows:", exam.index[exam.isna().any(axis=1)].tolist())

# Part B — cleaning
print("dropna experiment keeps", len(exam.dropna()), "of", len(exam), "rows")
clean = exam.copy()
print("Salary mean vs median:", round(exam["Salary"].mean()), exam["Salary"].median())   # Part C comparison
clean["Salary"] = clean.groupby("Department")["Salary"].transform(lambda s: s.fillna(s.median()))  # group-based
clean["Salary"] = clean["Salary"].fillna(clean["Salary"].median())       # rows whose Department was missing
clean["Age"] = clean["Age"].fillna(clean["Age"].median())
clean["Experience"] = clean["Experience"].fillna(clean["Experience"].median())
clean["Department"] = clean["Department"].fillna("Unknown")
clean["City"] = clean["City"].fillna(clean["City"].mode()[0])
clean["Performance"] = clean["Performance"].interpolate()                  # interpolation demo

demo = pd.Series([1.0, np.nan, np.nan, 4.0])
print("ffill:", demo.ffill().tolist(), "bfill:", demo.bfill().tolist())

# Part C/D — before/after report + verification
print(pd.DataFrame({"before": exam.isna().sum(), "after": clean.isna().sum()}))
clean
```
