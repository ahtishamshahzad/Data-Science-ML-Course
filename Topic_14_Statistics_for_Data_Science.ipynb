# Part 1 — Statistics Fundamentals

### 1. Simple explanation
**Statistics** is the science of **collecting, describing, and drawing conclusions from data**, while being honest about **uncertainty**.

Two big branches:
- **Descriptive statistics** — summarize the data you have (mean, median, charts).
- **Inferential statistics** — use a **sample** to make conclusions about a bigger **population** (confidence intervals, hypothesis tests).

### Key vocabulary

| Term | Simple meaning | Example (salaries in Pakistan's IT sector) |
|---|---|---|
| **Population** | **Everyone/everything** you want to know about | All IT employees in Pakistan |
| **Sample** | The **part** of the population you actually measure | 500 IT employees you surveyed |
| **Parameter** | A number describing the **population** (usually unknown) | The true average salary of *all* IT employees (μ) |
| **Statistic** | A number calculated from the **sample** (used to estimate a parameter) | The average salary of your 500 employees (x̄) |
| **Observation** | One row / one measured unit | One employee |
| **Variable** | A characteristic that can vary between observations | Salary, city, experience |
| **Feature** | (ML word) an input variable used for prediction | Experience, education, city |
| **Target** | (ML word) the variable we want to predict | Salary |

### Symbols you'll see

| | Population (parameter) | Sample (statistic) |
|---|---|---|
| Mean | μ ("mu") | x̄ ("x-bar") |
| Standard deviation | σ ("sigma") | s |
| Size | N | n |
| Proportion | p | p̂ ("p-hat") |

### 2. Real-world examples — population vs sample

| Question | Population | Sample |
|---|---|---|
| Average height of adult Pakistani men? | All adult men in Pakistan | 2,000 randomly selected men |
| Will voters choose party X? | All voters | 1,500 people in an opinion poll |
| Does the new website design increase sales? | All future visitors | Visitors during a 2-week A/B test |
| Is this batch of medicine safe? | All tablets in the batch | 50 tablets tested in a lab |

We use samples because measuring the whole population is usually **too expensive, too slow, or impossible** (you can't test every tablet — testing destroys it!).

### 4. Python — a population and a sample

````python
# Pretend we know the WHOLE population (in reality we almost never do)
population = pd.DataFrame({'Salary': rng.lognormal(mean=11.4, sigma=0.45, size=100_000).round(-2)})

mu = population['Salary'].mean()                          # PARAMETER (population mean)

sample = population.sample(n=500, random_state=1)         # a random sample of 500 people
x_bar = sample['Salary'].mean()                           # STATISTIC (sample mean)

print(f'Population size N = {len(population):,}')
print(f'Sample size     n = {len(sample)}')
print(f'Parameter  μ  (population mean) = {mu:,.0f}')
print(f'Statistic  x̄ (sample mean)      = {x_bar:,.0f}')
print(f'Difference = {x_bar - mu:,.0f}  ← sampling error: the sample is close, but not exact')
````

### 6. Interpretation
The sample mean is **close** to the population mean but not identical. That gap is called **sampling error**. A large part of statistics (Parts 11–14) is about measuring **how big this error is likely to be**.

### 7. When to use these ideas
Every time you analyse data: always ask "is this my whole population, or a sample of something bigger?"

### 8. Common mistakes
- Treating a sample statistic as if it were the exact truth.
- Using a **biased** sample (e.g. surveying only LinkedIn users about national salaries) — Part 10.
- Confusing "feature" and "target" in ML problems.

# Part 2 — Types of Data

### 1. Simple explanation

```text
Data
 ├── Numerical (quantitative) — numbers where arithmetic makes sense
 │     ├── Discrete    — countable, whole numbers           (children, orders, goals)
 │     └── Continuous  — any value in a range, measured     (height, salary, time)
 │
 └── Categorical (qualitative) — labels / groups
       ├── Nominal     — no natural order                    (city, colour, blood group)
       ├── Ordinal     — natural order, unequal/unknown gaps (low/medium/high, education level)
       └── Binary      — only two categories                 (yes/no, churned/not)
```

### 2. Real-world examples

| Variable | Type | Why |
|---|---|---|
| Number of children | Numerical — discrete | You can't have 2.5 children |
| Height (cm) | Numerical — continuous | 172.36 cm is possible |
| Salary | Numerical — continuous (in practice) | Many possible values |
| City | Categorical — nominal | Lahore isn't "greater than" Karachi |
| T-shirt size (S, M, L, XL) | Categorical — ordinal | S < M < L < XL, but the gaps aren't measurable |
| Customer rating 1–5 stars | Ordinal (often treated as numeric) | Ordered, but is 4→5 the same "distance" as 1→2? |
| Churned (Yes/No) | Categorical — binary | Two possible values |
| Zip code | Categorical — nominal (even though it's a number!) | Averaging zip codes is meaningless |

### Why it matters
The data type decides **which statistics, charts, tests and encodings** are valid:
- mean → numerical; mode → any type; median → numerical or ordinal
- histogram → numerical; bar chart → categorical
- t-test → numerical outcome; chi-square → categorical vs categorical
- one-hot encoding → nominal; ordinal encoding → ordinal (Topic 6)

````python
types_df = pd.DataFrame({
    'Children':     [0, 2, 1, 3, 2],                                    # numerical – discrete
    'Height_cm':    [165.2, 178.9, 170.0, 158.4, 182.1],                # numerical – continuous
    'City':         ['Lahore', 'Karachi', 'Lahore', 'Multan', 'Quetta'],   # categorical – nominal
    'Education':    ['Bachelor', 'Master', 'PhD', 'Bachelor', 'Master'],  # categorical – ordinal
    'Churned':      ['No', 'Yes', 'No', 'No', 'Yes'],                   # categorical – binary
    'Zip_Code':     ['54000', '74200', '54000', '60000', '87300'],     # nominal (stored as text!)
})

# Tell pandas that Education has an ORDER
types_df['Education'] = pd.Categorical(types_df['Education'],
                                       categories=['Bachelor', 'Master', 'PhD'], ordered=True)
types_df['City'] = types_df['City'].astype('category')

print(types_df.dtypes, '\n')
print('Ordinal comparison works:', (types_df['Education'] > 'Bachelor').tolist())
print('Numerical columns :', types_df.select_dtypes('number').columns.tolist())
print('Category columns  :', types_df.select_dtypes('category').columns.tolist())
````

### 6. Interpretation
`pd.Categorical(..., ordered=True)` stores the order, so comparisons like `Education > 'Bachelor'` work. Zip codes are stored as **text** so nobody accidentally averages them.

### 8. Common mistakes
- Computing the **mean of zip codes / IDs** because they look like numbers.
- Treating ordinal ratings as perfectly numerical without thinking.
- One-hot encoding an ordinal variable and losing its order (or label-encoding a nominal one and inventing an order).

# Part 3 — Measures of Central Tendency

### 1. Simple explanation
"Central tendency" = **one number that represents a typical value**.

| Measure | Meaning | How |
|---|---|---|
| **Mean** | The average | Add everything, divide by the count |
| **Median** | The middle value | Sort, pick the middle (average of the two middle values if n is even) |
| **Mode** | The most frequent value | Count how often each value appears |

### 3. Numerical example
Data: `2, 3, 3, 5, 7, 10`
- **Mean** = (2 + 3 + 3 + 5 + 7 + 10) / 6 = 30 / 6 = **5**
- **Median**: sorted already; middle two are 3 and 5 → (3 + 5) / 2 = **4**
- **Mode** = **3** (appears twice)

Now add one billionaire-sized value: `2, 3, 3, 5, 7, 10, 1000`
- Mean = 1030 / 7 = **147.1** (jumped massively!)
- Median = **5** (barely moved)
- Mode = **3** (unchanged)

````python
data = pd.Series([2, 3, 3, 5, 7, 10])
print('Mean  :', data.mean())
print('Median:', data.median())
print('Mode  :', data.mode().tolist())       # mode() returns a Series because there can be several modes

with_outlier = pd.Series([2, 3, 3, 5, 7, 10, 1000])
print('\nWith an outlier → mean:', round(with_outlier.mean(), 1), '| median:', with_outlier.median(),
      '| mode:', with_outlier.mode().tolist())
````

````python
# On a DataFrame
emp = pd.DataFrame({
    'Salary':     [45000, 52000, 48000, 61000, 52000, 250000],
    'Experience': [1, 3, 2, 6, 3, 20],
    'City':       ['Lahore', 'Karachi', 'Lahore', 'Lahore', 'Islamabad', 'Karachi'],
})
print('df.mean(numeric_only=True):\n', emp.mean(numeric_only=True), '\n')
print('df.median(numeric_only=True):\n', emp.median(numeric_only=True), '\n')
print('df.mode():')
print(emp.mode())
````

**Note:** `df.mode()` returns a **DataFrame**, because a column can have several modes (ties). Row 0 holds the first mode of each column; `NaN` appears when a column has fewer modes than others.

````python
skewed = pd.Series(rng.lognormal(11, 0.6, 5000))
plt.figure(figsize=(8, 3.8))
sns.histplot(skewed, bins=60, color='lightsteelblue')
plt.axvline(skewed.mean(), color='red', lw=2, label=f'Mean = {skewed.mean():,.0f}')
plt.axvline(skewed.median(), color='green', lw=2, ls='--', label=f'Median = {skewed.median():,.0f}')
plt.title('Salary-like data: the mean is pulled toward the long right tail')
plt.legend(); plt.show()
````

### When is the mean better? When is the median better?

| Use the **mean** when… | Use the **median** when… |
|---|---|
| Data is roughly **symmetric** | Data is **skewed** (salary, house prices) |
| No extreme outliers | There are **outliers** |
| You need the **total** (mean × n = total sales) | You want the "**typical**" person |
| Further maths needs it (variance, t-tests) | Data is **ordinal** |

**Mode** is best for **categorical** data ("most common city") and for finding peaks.

### 8. Common mistakes
- Reporting the **mean salary** of a company where the CEO earns 100× more than everyone.
- Forgetting that `mode()` can return **several** values.
- Computing the mean of an ordinal scale like "Low/Medium/High" coded 1/2/3 without saying so.

# Part 4 — Measures of Spread (Dispersion)

### 1. Simple explanation
Two classes can have the **same average** score (70) but be completely different:
- Class A: everyone scored 68–72 → **low spread** (consistent).
- Class B: scores from 20 to 100 → **high spread** (unpredictable).

Spread tells us **how far values are from each other / from the center**.

| Measure | Meaning | Outlier-sensitive? |
|---|---|---|
| **Range** | max − min | Very |
| **Variance** | Average **squared** distance from the mean | Yes |
| **Standard deviation (std)** | √variance — typical distance from the mean, **in original units** | Yes |
| **IQR** | Q3 − Q1 — spread of the **middle 50%** | No (robust) |
| **MAD** | **Median absolute deviation** — median of \|x − median\| | No (robust) |

### Variance and standard deviation — very simply
1. Find the mean.
2. For each value, find its distance from the mean.
3. **Square** each distance (so negatives don't cancel positives, and big distances count more).
4. Average the squares → **variance**.
5. Take the square root → **standard deviation** (back in the original units, e.g. rupees instead of rupees²).

### 3. Numerical example
Data: `2, 4, 4, 4, 5, 5, 7, 9` (n = 8), mean = 40 / 8 = **5**

| x | x − mean | (x − mean)² |
|---|---|---|
| 2 | −3 | 9 |
| 4 | −1 | 1 |
| 4 | −1 | 1 |
| 4 | −1 | 1 |
| 5 | 0 | 0 |
| 5 | 0 | 0 |
| 7 | 2 | 4 |
| 9 | 4 | 16 |
| | | **sum = 32** |

- **Population variance** = 32 / **n** = 32 / 8 = **4** → std = **2**
- **Sample variance** = 32 / **(n − 1)** = 32 / 7 = **4.571** → std = **2.138**
- Range = 9 − 2 = **7**

### Sample vs population variance — why `n − 1`?
A sample's values are, on average, a little **closer to their own sample mean** than to the true population mean. Dividing by `n` would therefore **underestimate** the true spread. Dividing by `n − 1` (called **Bessel's correction**) fixes this bias.

| | Formula | NumPy | Pandas |
|---|---|---|---|
| Population | ÷ n | `np.var(x)` (default `ddof=0`) | `s.var(ddof=0)` |
| Sample | ÷ (n − 1) | `np.var(x, ddof=1)` | `s.var()` (default `ddof=1`) |

⚠️ **NumPy and Pandas have different defaults!** This is a classic interview question.

````python
x = np.array([2, 4, 4, 4, 5, 5, 7, 9])
s = pd.Series(x)

print('Range                     :', x.max() - x.min())
print('np.var(x)  (population)   :', np.var(x))
print('np.var(x, ddof=1) (sample):', round(np.var(x, ddof=1), 3))
print('s.var()    (pandas sample):', round(s.var(), 3))
print('np.std(x)  (population)   :', np.std(x))
print('s.std()    (pandas sample):', round(s.std(), 3))

q1, q3 = np.percentile(x, [25, 75])
print('\nQ1, Q3 =', q1, q3, '→ IQR =', q3 - q1)
print('MAD (median absolute deviation):', stats.median_abs_deviation(x))
print('Mean absolute deviation       :', np.mean(np.abs(x - x.mean())))
````

### MAD — two different things with the same abbreviation
- **Median absolute deviation** = median(|x − median|) → very robust; used in robust outlier detection. `scipy.stats.median_abs_deviation`.
- **Mean absolute deviation** = mean(|x − mean|) → easier to explain, less robust.
Always say which one you mean.

````python
class_a = rng.normal(70, 3, 200)
class_b = rng.normal(70, 15, 200)

fig, ax = plt.subplots(figsize=(8, 3.8))
sns.kdeplot(class_a, fill=True, label=f'Class A: mean {class_a.mean():.0f}, std {class_a.std(ddof=1):.1f}', ax=ax)
sns.kdeplot(class_b, fill=True, label=f'Class B: mean {class_b.mean():.0f}, std {class_b.std(ddof=1):.1f}', ax=ax)
ax.set_title('Same mean, very different spread'); ax.legend(); plt.show()
````

### 6. Interpretation
**Rule of thumb for bell-shaped data (68–95–99.7 rule):** about 68% of values lie within 1 std of the mean, 95% within 2 std, 99.7% within 3 std.

### 7. When to use
- **Std** for roughly symmetric data (and most statistical formulas).
- **IQR / MAD** for skewed data or data with outliers.
- **Range** only for a quick look.

### 8. Common mistakes
- Mixing `np.std` (population) and `pandas .std()` (sample) and getting "different answers".
- Interpreting variance in original units (it's in **squared** units).
- Using std to describe heavily skewed data.

# Part 5 — Percentiles and Quartiles

### 1. Simple explanation
The **p-th percentile** is the value **below which p% of the data falls**.
"You scored in the 90th percentile" → you did better than 90% of students.

**Quartiles** split sorted data into **four equal parts**:

| Name | Percentile | Meaning |
|---|---|---|
| **Q1** | 25th | 25% of data is below it |
| **Q2** | 50th | The **median** |
| **Q3** | 75th | 75% of data is below it |
| **IQR** | Q3 − Q1 | Range of the **middle 50%** |

### 3. Numerical example
Data (n = 8): `2, 4, 4, 4, 5, 5, 7, 9`
NumPy's default (linear interpolation) uses position `p × (n − 1)`:
- Q1: 0.25 × 7 = 1.75 → between the 2nd value (4) and the 3rd (4) → **4**
- Q2: 0.50 × 7 = 3.5 → between 4 and 5 → **4.5**
- Q3: 0.75 × 7 = 5.25 → between 5 and 7 → 5 + 0.25 × 2 = **5.5**
- IQR = 5.5 − 4 = **1.5**

(Different software can use slightly different interpolation methods — small differences are normal.)

### Connection to boxplots and outliers (Topic 5)

```text
      lower fence            Q1   median   Q3            upper fence
          |-------------------[=====|=====]-------------------|      •  ← outlier
     Q1 − 1.5×IQR                                    Q3 + 1.5×IQR
```
- The **box** goes from Q1 to Q3; the line inside is the median.
- Whiskers extend to the most extreme points **inside** the fences.
- Points outside the fences are drawn as **outliers**.

````python
x = np.array([2, 4, 4, 4, 5, 5, 7, 9])
print('Q1, Q2, Q3:', np.percentile(x, [25, 50, 75]))
print('90th percentile:', np.percentile(x, 90))

scores = pd.Series(np.concatenate([rng.normal(65, 10, 300).round(), [5, 12, 110]]))
q1, q3 = np.percentile(scores, [25, 75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = scores[(scores < lower) | (scores > upper)]

print(f'\nQ1 = {q1}, Q3 = {q3}, IQR = {iqr}')
print(f'Fences: [{lower}, {upper}]')
print('Outliers:', sorted(outliers.tolist()))

fig, ax = plt.subplots(figsize=(9, 2.8))
sns.boxplot(x=scores, ax=ax, color='lightsteelblue')
for v, lab in [(q1, 'Q1'), (np.median(scores), 'Median'), (q3, 'Q3')]:
    ax.axvline(v, color='gray', ls=':', lw=1)
    ax.text(v, -0.45, lab, ha='center')
ax.axvline(lower, color='red', ls='--'); ax.axvline(upper, color='red', ls='--')
ax.set_title('Boxplot with quartiles and 1.5×IQR fences (red)'); plt.show()
````

### 7. When to use
- Describe skewed data (median + IQR).
- Outlier detection (IQR rule).
- Business thresholds ("top 10% customers" = above the 90th percentile).
- Feature engineering: quantile bins (`pd.qcut`, Topic 11).

### 8. Common mistakes
- Confusing **percentage** (a proportion) with **percentile** (a position in the data).
- Deleting every IQR "outlier" in skewed data — in right-skewed data many are real, normal values.

# Part 6 — Distributions

### 1. Simple explanation
A **distribution** shows **which values occur and how often**.

- **Frequency distribution:** a table of values (or ranges) and their counts.
- **Histogram:** a bar chart of a frequency distribution for numerical data.
- **Density:** the histogram scaled so the total **area** is 1 — lets you compare datasets of different sizes. A smooth version is the **KDE** curve.
- **Shape:** symmetric, skewed, flat, one hump or two humps…

### Common shapes

| Distribution | Shape | Real-world example |
|---|---|---|
| **Normal** (Gaussian) | Symmetric bell | Heights, measurement errors, exam scores |
| **Uniform** | Flat — every value equally likely | A fair die; random number generators |
| **Right-skewed** | Long right tail | Salary, house prices, time on site |
| **Left-skewed** | Long left tail | Scores on an easy exam, age at retirement |
| **Bimodal** | Two humps | Restaurant visits (lunch & dinner peaks); heights of mixed men & women |

````python
scores = rng.integers(40, 100, 30)
freq_table = pd.cut(pd.Series(scores), bins=[40, 50, 60, 70, 80, 90, 100], right=False).value_counts().sort_index()
print('Frequency distribution of 30 exam scores:')
print(freq_table)
````

````python
n = 5000
shapes = {
    'Normal': rng.normal(50, 10, n),
    'Uniform': rng.uniform(0, 100, n),
    'Right-skewed': rng.gamma(2, 10, n),
    'Left-skewed': 100 - rng.gamma(2, 10, n),
    'Bimodal': np.concatenate([rng.normal(12, 1.5, n // 2), rng.normal(19.5, 1.5, n // 2)]),  # lunch & dinner
}
fig, axes = plt.subplots(1, 5, figsize=(21, 3.8))
for ax, (name, v) in zip(axes, shapes.items()):
    sns.histplot(v, bins=40, stat='density', kde=True, ax=ax)
    ax.set_title(f'{name}\nmean {np.mean(v):.1f} | median {np.median(v):.1f}', fontsize=10)
plt.tight_layout(); plt.show()
````

### 6. Interpretation
- Normal: mean ≈ median, bell shape.
- Uniform: flat; no value is more likely.
- Right-skewed: mean > median. Left-skewed: mean < median.
- **Bimodal**: the mean (~15.7, between the humps) describes **nobody** — a sign the data contains **two groups** that should maybe be analysed separately.

### 7. When to use
Always plot the distribution of important variables before choosing statistics, tests or models.

### 8. Common mistakes
- Summarizing bimodal data with a single mean.
- Assuming "normal" without checking.

# Part 7 — Skewness and Kurtosis

### Skewness — lopsidedness
| Value | Meaning | Mean vs median |
|---|---|---|
| ≈ 0 | Symmetric | mean ≈ median |
| > 0 (**positive**) | Long **right** tail | mean > median |
| < 0 (**negative**) | Long **left** tail | mean < median |

Rule of thumb: |skew| < 0.5 ≈ symmetric, 0.5–1 moderate, > 1 high (a guideline, not a law).

### Kurtosis — "tailedness"
Kurtosis measures how **heavy the tails** are — how often **extreme values** occur compared with a normal distribution.

- Pandas `.kurt()` and SciPy `stats.kurtosis()` (default) report **excess kurtosis** = kurtosis − 3, so a **normal distribution = 0**.

| Excess kurtosis | Name | Tails | Example |
|---|---|---|---|
| ≈ 0 | Mesokurtic | Like a normal distribution | Heights |
| **> 0** | Leptokurtic | **Heavy tails** — extreme values more common | Stock returns (crashes!), insurance claims |
| **< 0** | Platykurtic | **Light tails** — extremes rarer | Uniform distribution |

### 2. Real-world example
Daily stock returns look roughly bell-shaped, but crashes (−10% days) happen **far more often** than a normal distribution predicts → high kurtosis. Risk models that assume normality underestimate crash risk.

### Why skewness & kurtosis matter for ML
- Skewed features can hurt **linear models, KNN, SVM, neural nets** → consider transformations (Topic 13).
- Heavy tails mean **more extreme values** → outlier handling, robust losses (MAE/Huber), robust scalers.
- Skewed **targets** (e.g. house price) often benefit from predicting `log(price)`.
- Many classical tests assume approximate normality (Part 22).

````python
n = 20000
samples = {
    'Normal': rng.normal(0, 1, n),
    'Heavy tails (t, df=3)': rng.standard_t(3, n),
    'Light tails (uniform)': rng.uniform(-1.7, 1.7, n),
    'Right-skewed (exponential)': rng.exponential(1, n),
}
summary = pd.DataFrame({name: {'skew (pandas)': pd.Series(v).skew(),
                               'excess kurtosis (pandas)': pd.Series(v).kurt(),
                               'skew (scipy)': stats.skew(v),
                               'excess kurtosis (scipy)': stats.kurtosis(v)}
                        for name, v in samples.items()}).T.round(2)
summary
````

````python
fig, axes = plt.subplots(1, 3, figsize=(16, 3.8))
for ax, name in zip(axes, ['Normal', 'Heavy tails (t, df=3)', 'Light tails (uniform)']):
    v = samples[name]
    sns.histplot(v[(v > -6) & (v < 6)], bins=80, stat='density', ax=ax)
    ax.set_title(f'{name}\nexcess kurtosis = {pd.Series(v).kurt():.1f}'); ax.set_xlim(-6, 6)
plt.tight_layout(); plt.show()
````

### 6. Interpretation
- The t-distribution has a **large positive** kurtosis: it looks similar to the normal in the middle, but values beyond ±4 appear far more often.
- The uniform distribution has **negative** excess kurtosis (≈ −1.2): no tails at all.
- The exponential is both skewed (≈ 2) and heavy-tailed (≈ 6).

### 8. Common mistakes
- Thinking kurtosis measures "peakedness" — it's mainly about the **tails**.
- Forgetting whether a tool reports **excess** kurtosis (normal = 0) or **raw** kurtosis (normal = 3). `stats.kurtosis(x, fisher=False)` gives raw kurtosis.

# Part 8 — Z-Score

### 1. Simple explanation
$$z = \frac{x - \text{mean}}{\text{standard deviation}}$$

A **z-score** tells you **how many standard deviations a value is above or below the mean**.

| z | Meaning |
|---|---|
| **0** | Exactly average |
| **+1** | 1 std **above** average |
| **−1** | 1 std **below** average |
| **+2** | 2 std above — unusually high (top ~2.5% if normal) |
| **−2** | 2 std below — unusually low (bottom ~2.5% if normal) |
| \|z\| > 3 | Very rare if the data is normal → possible outlier |

### 2. Real-world example
Ali scored **80** in Maths (class mean 70, std 5) and **85** in English (class mean 80, std 10). In which subject did he do **relatively** better?
- Maths: z = (80 − 70) / 5 = **2.0**
- English: z = (85 − 80) / 10 = **0.5**

The raw English score is higher, but relative to classmates, **Maths is much more impressive**. Z-scores make different scales comparable.

### Z-score and standardization
Converting **every** value of a column into its z-score **is** standardization (`StandardScaler`, Topic 4): the result has mean 0 and std 1. The **shape** of the distribution stays the same.

````python
def z_score(x, mean, std):
    return (x - mean) / std

print('Maths z  :', z_score(80, 70, 5))
print('English z:', z_score(85, 80, 10))

marks = pd.Series([55, 60, 65, 70, 75, 80, 85, 120])
z = (marks - marks.mean()) / marks.std(ddof=0)          # population std, like StandardScaler
print('\nZ-scores:')
print(pd.DataFrame({'Mark': marks, 'z': z.round(2)}))
print('\nMean of z:', round(z.mean(), 10), '| std of z:', round(z.std(ddof=0), 10))
print('scipy.stats.zscore gives the same:', np.allclose(stats.zscore(marks), z))
````

````python
xs = np.linspace(-4, 4, 400)
plt.figure(figsize=(9, 3.8))
plt.plot(xs, stats.norm.pdf(xs), color='black')
for k, c in [(1, '#9ecae1'), (2, '#4292c6'), (3, '#08519c')][::-1]:
    mask = (xs >= -k) & (xs <= k)
    plt.fill_between(xs[mask], stats.norm.pdf(xs[mask]), color=c, alpha=0.6,
                     label=f'|z| ≤ {k}: {100 * (stats.norm.cdf(k) - stats.norm.cdf(-k)):.1f}%')
plt.xlabel('z-score'); plt.title('Standard normal distribution: the 68–95–99.7 rule'); plt.legend(); plt.show()
````

### 6. Interpretation
For normally distributed data: 68.3% of z-scores are between −1 and 1, 95.4% between −2 and 2, and 99.7% between −3 and 3.

### 7. When to use
Comparing values on different scales, detecting outliers in roughly normal data (|z| > 3), standardizing features.

### 8. Common mistakes
- Using |z| > 3 outlier detection on **skewed** data (the mean/std are themselves distorted — use IQR/MAD).
- Calculating z-scores for the test set using the **test** mean/std (leakage — use the training mean/std).

# Part 9 — Covariance and Correlation

(Topic 12 covered this in depth — here is the statistics summary.)

| Measure | Meaning | Range |
|---|---|---|
| **Covariance** | Do two variables move together? Direction only; depends on units | −∞ to +∞ |
| **Pearson r** | Strength & direction of a **linear** relationship | −1 to +1 |
| **Spearman ρ** | Strength & direction of a **monotonic** relationship (uses ranks; robust) | −1 to +1 |

$$\text{Cov}(X,Y) = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{n-1} \qquad r = \frac{\text{Cov}(X,Y)}{s_X \, s_Y}$$

### 3. Numerical example
Hours `[1, 2, 3, 4, 5]`, Score `[50, 55, 65, 70, 80]` → Cov = 75 / 4 = **18.75**, r ≈ **0.993**.

### ⚠️ Correlation ≠ causation
Ice-cream sales and drownings are correlated because **hot weather** causes both (a **confounder**).

````python
hours = pd.DataFrame({'Hours': [1, 2, 3, 4, 5], 'Score': [50, 55, 65, 70, 80]})
print('Covariance:', hours['Hours'].cov(hours['Score']))
print('Pearson   :', round(hours['Hours'].corr(hours['Score']), 3))
print('Spearman  :', round(hours['Hours'].corr(hours['Score'], method='spearman'), 3))

n = 300
study = pd.DataFrame({'Study_Hours': rng.uniform(0, 10, n), 'Sleep_Hours': rng.uniform(4, 9, n),
                      'Phone_Hours': rng.uniform(0, 8, n)})
study['Exam_Score'] = 40 + 4 * study['Study_Hours'] + 2 * study['Sleep_Hours'] - 2.5 * study['Phone_Hours'] + rng.normal(0, 5, n)

corr = study.corr()
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, ax=axes[0])
axes[0].set_title('Correlation matrix')
sns.regplot(data=study, x='Study_Hours', y='Exam_Score', ax=axes[1], scatter_kws={'alpha': 0.4})
axes[1].set_title(f"Study hours vs score (r = {corr.loc['Study_Hours', 'Exam_Score']:.2f})")
plt.tight_layout(); plt.show()
````

### 6. Interpretation
Study hours correlate strongly and positively with the score; phone hours negatively; sleep weakly positively. The predictors are nearly uncorrelated with each other (no multicollinearity).

### 7. When to use
EDA, feature selection, detecting redundant features, checking assumptions.

### 8. Common mistakes
Correlation ≠ causation · Pearson only sees **linear** patterns · one outlier can create or destroy r · always look at scatter plots.

# Part 10 — Sampling

### 1. Simple explanation
**Sampling** = choosing which part of the population to measure. A good sample **represents** the population; a bad sample gives confidently wrong answers.

| Method | How it works | Real-world example |
|---|---|---|
| **Simple random sampling** | Every member has an **equal chance** of being picked | Draw 500 employee IDs at random |
| **Stratified sampling** | Split the population into groups (**strata**), then sample randomly **within each group** in the right proportions | Sample 60% Lahore, 30% Karachi, 10% Quetta employees to match the company |
| **Systematic sampling** | Pick every **k-th** member after a random start | Check every 10th product on an assembly line |
| **Cluster sampling** | Split into **clusters** (e.g. schools, branches), randomly pick **whole clusters**, measure everyone in them | Randomly pick 5 bank branches and survey all their staff |

**Random sampling** is the umbrella idea: chance (not the researcher's choice) decides who is included.

### Sampling bias — when the sample does not represent the population

| Bias | Example |
|---|---|
| **Selection bias** | Surveying salaries only through LinkedIn → misses people without LinkedIn |
| **Voluntary response bias** | Online reviews — mostly very happy or very angry customers reply |
| **Survivorship bias** | Studying only successful start-ups to learn "what makes start-ups succeed" |
| **Non-response bias** | Busy, high-earning people ignore the survey |
| **Undercoverage** | A phone survey misses people without phones |
| **Convenience sampling** | Asking only your friends or classmates |

In ML, a biased **training set** gives a biased **model** (e.g. a face-recognition model trained mostly on one skin tone).

````python
# A company population of 10,000 employees
pop = pd.DataFrame({
    'Employee_ID': np.arange(10_000),
    'City': rng.choice(['Lahore', 'Karachi', 'Quetta'], 10_000, p=[0.6, 0.3, 0.1]),
    'Branch': rng.integers(1, 101, 10_000),                       # 100 branches (clusters)
})
city_salary = pop['City'].map({'Lahore': 90000, 'Karachi': 100000, 'Quetta': 60000})
pop['Salary'] = (city_salary + rng.normal(0, 15000, len(pop))).round(-2)

n = 500

# 1. Simple random sampling
srs = pop.sample(n=n, random_state=0)

# 2. Stratified sampling: same fraction from each city
stratified = pop.groupby('City', group_keys=False).sample(frac=n / len(pop), random_state=0)

# 3. Systematic sampling: every k-th row after a random start
k = len(pop) // n
start = rng.integers(0, k)
systematic = pop.iloc[start::k]

# 4. Cluster sampling: pick 5 random branches, take everyone in them
chosen_branches = rng.choice(pop['Branch'].unique(), 5, replace=False)
cluster = pop[pop['Branch'].isin(chosen_branches)]

# 5. BIASED sample: only Quetta employees answered
biased = pop[pop['City'] == 'Quetta'].sample(n=n, random_state=0)

comparison = pd.DataFrame({
    name: {'n': len(s), 'Mean salary': s['Salary'].mean(),
           '% Lahore': (s['City'] == 'Lahore').mean() * 100,
           '% Karachi': (s['City'] == 'Karachi').mean() * 100,
           '% Quetta': (s['City'] == 'Quetta').mean() * 100}
    for name, s in [('POPULATION', pop), ('Simple random', srs), ('Stratified', stratified),
                    ('Systematic', systematic), ('Cluster', cluster), ('Biased (Quetta only)', biased)]
}).T.round(1)
comparison
````

### 6. Interpretation
- Simple random, stratified, systematic and cluster samples all give a mean salary **close to the population**.
- **Stratified** matches the city percentages almost exactly — its main advantage.
- The **biased** sample is badly wrong (it thinks the average salary is ~60,000). A **bigger** biased sample would still be wrong — **size does not fix bias**.

### 7. When to use which
| Method | Use when |
|---|---|
| Simple random | You have a complete list of the population |
| Stratified | Important subgroups must be represented (also: `train_test_split(..., stratify=y)` for imbalanced classes!) |
| Systematic | Items arrive in a stream/list (quality control) — beware of periodic patterns |
| Cluster | The population is naturally grouped and visiting everyone is expensive |

### 8. Common mistakes
- Believing a huge sample is automatically representative (a million biased responses are still biased).
- Systematic sampling when the list has a **repeating pattern** that matches k (e.g. every 7th day is always a Sunday).
- Not stratifying the train/test split for imbalanced classification.

# Part 11 — Sampling Distribution

### 1. Simple explanation
If 1,000 different researchers each took their **own** random sample of 50 employees and computed the average salary, they would get **1,000 slightly different averages**. The distribution of those averages is the **sampling distribution of the mean**.

> **Analogy:** Every spoonful from the soup pot tastes slightly different. The sampling distribution describes how much spoonfuls vary.

### Why do sample means vary?
Because **each sample contains different people by chance**. Sometimes you happen to pick a few high earners, sometimes a few low earners.

### The **standard error (SE)**
The **standard deviation of the sampling distribution** is called the **standard error**. It tells you how much a sample mean typically differs from the true population mean.

$$SE = \frac{\sigma}{\sqrt{n}} \approx \frac{s}{\sqrt{n}}$$

### 3. Numerical example
Population std σ = 20,000, sample size n = 100 → SE = 20,000 / √100 = **2,000**.
So a sample mean is typically about ±2,000 away from the true mean.

````python
population_salary = population['Salary'].to_numpy()          # the population from Part 1
mu, sigma = population_salary.mean(), population_salary.std()

n = 50
n_samples = 2000
sample_means = np.array([rng.choice(population_salary, n, replace=False).mean() for _ in range(n_samples)])

print(f'Population mean μ         = {mu:,.0f}')
print(f'Mean of the sample means  = {sample_means.mean():,.0f}  ← centred on μ')
print(f'Std of the sample means   = {sample_means.std():,.0f}  ← the standard error (simulated)')
print(f'Formula σ / √n            = {sigma / np.sqrt(n):,.0f}')
````

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(population_salary, bins=80, ax=axes[0], color='gray')
axes[0].axvline(mu, color='red', lw=2, label='μ'); axes[0].legend()
axes[0].set_title('Population: individual salaries (skewed, very spread out)')

sns.histplot(sample_means, bins=50, ax=axes[1], color='tab:blue')
axes[1].axvline(mu, color='red', lw=2, label='μ'); axes[1].legend()
axes[1].set_title(f'Sampling distribution: {n_samples} sample means (n = {n})')
axes[1].set_xlim(axes[0].get_xlim())
plt.tight_layout(); plt.show()
````

### 6. Interpretation
- Individual salaries vary **a lot**; averages of 50 salaries vary **much less**.
- The sample means are centred on μ → the sample mean is an **unbiased** estimate.
- Their spread (the SE) matches **σ / √n**.
- Even though the population is skewed, the sample means look **bell-shaped** — that's the Central Limit Theorem (next).

### 8. Common mistakes
Confusing the spread of **individual values** (std) with the spread of **sample means** (SE).

# Part 12 — Central Limit Theorem (CLT)

### 1. Simple explanation
> **No matter what shape the population has** (skewed, flat, weird), if you take **many random samples of a reasonably large size** and compute their **means**, those means will form an **approximately normal (bell-shaped) distribution**, centred on the population mean, with spread σ/√n.

Conditions (simply):
- samples are random and independent,
- the population has a finite variance,
- n is "large enough" — often **n ≥ 30** is a rule of thumb; very skewed populations need more.

### 2. Real-world example
Individual customer purchases are very skewed (many small, a few huge). But the **average purchase per day** (averaging hundreds of customers) behaves like a bell curve — which is why we can put confidence intervals around daily averages and run A/B tests on them.

### Demonstration: start from a very NON-normal population

````python
population_exp = rng.exponential(scale=1000, size=200_000)      # strongly right-skewed (e.g. purchase amounts)

fig, axes = plt.subplots(1, 5, figsize=(22, 3.8))
sns.histplot(population_exp, bins=60, ax=axes[0], color='gray')
axes[0].set_title(f'Population (skew = {stats.skew(population_exp):.2f})')

for ax, n in zip(axes[1:], [2, 5, 30, 100]):
    means = rng.choice(population_exp, size=(5000, n)).mean(axis=1)   # 5000 samples of size n at once
    sns.histplot(means, bins=50, stat='density', ax=ax)
    xs = np.linspace(means.min(), means.max(), 200)
    ax.plot(xs, stats.norm.pdf(xs, population_exp.mean(), population_exp.std() / np.sqrt(n)),
            color='red', lw=2)                                        # the normal curve CLT predicts
    ax.set_title(f'Means of samples, n = {n}\n(skew = {stats.skew(means):.2f})')
plt.tight_layout(); plt.show()
````

### 6. Interpretation
- n = 2: sample means are still skewed.
- n = 5: less skewed.
- n = 30 and n = 100: almost perfectly bell-shaped and matching the red normal curve; the distribution also gets **narrower** as n grows (SE = σ/√n).

### Why CLT is important in Data Science
- It justifies **confidence intervals** and **t-tests/z-tests** for means even when the data isn't normal.
- It's the foundation of **A/B testing** (comparing average conversion or revenue).
- It explains why **averages are more stable** than individual values.
- Model evaluation: average error across many test examples has an approximately normal sampling distribution → confidence intervals for model accuracy.

### 8. Common mistakes
- Thinking CLT makes the **data itself** normal — it's about the distribution of **sample means**.
- Applying it with tiny samples from extremely skewed/heavy-tailed data.
- Applying it to **medians or maxima** in the same simple way (CLT is about sums/means).

# Part 13 — Standard Error

### Standard deviation vs standard error

| | Standard deviation (s, σ) | Standard error (SE) |
|---|---|---|
| Measures | Spread of **individual values** | Spread of a **statistic** (e.g. sample mean) across samples |
| Question | "How different are employees from each other?" | "How precise is my estimate of the average?" |
| Changes with n? | **No** (it's a property of the population) | **Yes** — shrinks as n grows |
| Formula | — | SE = s / √n |

### Why SE decreases with larger samples
Averaging more values lets random highs and lows **cancel out**. But it decreases with **√n**, not n:
- 4× more data → SE **halves**.
- 100× more data → SE becomes **10×** smaller.

### 3. Numerical example
s = 20,000
- n = 25 → SE = 20,000 / 5 = 4,000
- n = 100 → SE = 20,000 / 10 = 2,000
- n = 400 → SE = 20,000 / 20 = 1,000

````python
sizes = [5, 10, 25, 50, 100, 250, 500, 1000]
rows = []
for n in sizes:
    means = rng.choice(population_salary, size=(3000, n)).mean(axis=1)
    one_sample = rng.choice(population_salary, n)
    rows.append({'n': n,
                 'Sample std (one sample)': one_sample.std(ddof=1),
                 'Simulated SE': means.std(),
                 'Formula σ/√n': sigma / np.sqrt(n),
                 'SE from ONE sample (s/√n)': stats.sem(one_sample)})
se_table = pd.DataFrame(rows).round(0)
print(se_table.to_string(index=False))

plt.figure(figsize=(8, 4))
plt.plot(se_table['n'], se_table['Simulated SE'], 'o-', label='Simulated SE')
plt.plot(se_table['n'], se_table['Formula σ/√n'], '--', label='σ/√n')
plt.plot(se_table['n'], se_table['Sample std (one sample)'], 's:', label='Std of individual values')
plt.xscale('log'); plt.xlabel('Sample size n (log scale)'); plt.ylabel('Rupees')
plt.title('Std stays the same; standard error shrinks with √n'); plt.legend(); plt.show()
````

### 6. Interpretation
- The **std** of individual salaries stays around the population σ regardless of n.
- The **SE** shrinks steadily; the formula σ/√n matches the simulation.
- `scipy.stats.sem(sample)` estimates SE from **one** sample (s/√n) — that's what you use in real life.

### 8. Common mistakes
- Reporting "mean ± SE" and letting readers think it's "mean ± std" (SE bars look much smaller!).
- Expecting 2× data to halve the error (it only reduces it by √2 ≈ 1.41×).

# Part 14 — Confidence Intervals

### 1. Simple explanation
- **Point estimate:** a single best guess, e.g. "the average salary is 97,300".
- **Confidence interval (CI):** a **range** of plausible values, e.g. "95% CI: 94,100 to 100,500".

$$\text{CI} = \text{estimate} \pm \text{critical value} \times SE$$

For a mean with an unknown population std we use the **t-distribution** (a slightly wider bell curve that accounts for the extra uncertainty of small samples):

$$\bar{x} \pm t^{*} \cdot \frac{s}{\sqrt{n}}$$

| Confidence level | Critical value (large n, ≈ z) | Interval width |
|---|---|---|
| 90% | 1.645 | Narrowest |
| 95% | 1.960 | Medium |
| 99% | 2.576 | Widest |

Higher confidence → wider interval. (To be more sure of catching the truth you need a bigger net.)

### ✅ What a 95% confidence interval actually means (frequentist interpretation)

> The **method** we used to build the interval **captures the true population mean in about 95% of repeated samples.**

Think of a fisherman throwing a net: his throwing **technique** catches the fish 95% of the time. After one particular throw, the fish is either in the net or not — we just don't know which.

- The **true mean μ is fixed** (not random). The **interval is random** — it changes from sample to sample.
- "95%" describes the **long-run success rate of the procedure**, not a probability about this one specific interval.

### ❌ Incorrect interpretations
- ❌ "There is a 95% probability that μ is inside **this** interval." (After computing it, μ is either inside or not.)
- ❌ "95% of the data lies inside the CI." (That's a completely different idea — a CI is about the **mean**, not individual values.)
- ❌ "If I repeat the study, there's a 95% chance the new mean falls in this interval."

A good wording for reports: *"We are 95% confident that the true average salary lies between 94,100 and 100,500"*, where "95% confident" means *"we used a method that works 95% of the time"*.

### 3. Numerical example
n = 36, x̄ = 50,000, s = 12,000 → SE = 12,000 / 6 = 2,000. t* for 95%, df = 35 ≈ 2.03.
CI = 50,000 ± 2.03 × 2,000 = **50,000 ± 4,060 → (45,940, 54,060)**.

````python
# Numerical example with SciPy
n, x_bar, s = 36, 50000, 12000
se = s / np.sqrt(n)
t_crit = stats.t.ppf(0.975, df=n - 1)          # 97.5th percentile → leaves 2.5% in each tail
print(f't* = {t_crit:.3f}, SE = {se:.0f}')
print('95% CI (manual):', (round(x_bar - t_crit * se), round(x_bar + t_crit * se)))
print('95% CI (SciPy) :', tuple(np.round(stats.t.interval(0.95, df=n - 1, loc=x_bar, scale=se))))
````

````python
# Real sample: 90%, 95%, 99% CIs for the average salary
sample = rng.choice(population_salary, 60, replace=False)
x_bar, se, df = sample.mean(), stats.sem(sample), len(sample) - 1

rows = []
for level in [0.90, 0.95, 0.99]:
    low, high = stats.t.interval(level, df=df, loc=x_bar, scale=se)
    rows.append({'Confidence': f'{int(level * 100)}%', 'Lower': low, 'Upper': high, 'Width': high - low,
                 'Contains true μ?': low <= mu <= high})
print(f'Point estimate x̄ = {x_bar:,.0f}   (true μ = {mu:,.0f}, normally unknown!)')
pd.DataFrame(rows).round(0)
````

````python
# Simulation: what "95%" really means — build 100 intervals from 100 different samples
n_intervals, n = 100, 40
fig, ax = plt.subplots(figsize=(10, 7))
misses = 0
for i in range(n_intervals):
    smp = rng.choice(population_salary, n, replace=False)
    low, high = stats.t.interval(0.95, df=n - 1, loc=smp.mean(), scale=stats.sem(smp))
    hit = low <= mu <= high
    misses += not hit
    ax.plot([low, high], [i, i], color='tab:blue' if hit else 'red', lw=1.5)
    ax.plot(smp.mean(), i, 'o', color='tab:blue' if hit else 'red', ms=3)
ax.axvline(mu, color='black', ls='--', label='true μ (fixed)')
ax.set_title(f'100 different 95% CIs: {n_intervals - misses} contain μ, {misses} miss (red)')
ax.set_xlabel('Salary'); ax.set_ylabel('Sample number'); ax.legend(); plt.show()
````

### 6. Interpretation
Each horizontal line is a CI from a **different** sample. The true μ never moves; the intervals do. About 95 of 100 catch it — exactly what "95% confidence" promises about the **method**.

### 7. When to use
Report CIs whenever you report an estimate (mean, proportion, difference in A/B tests, model accuracy). They show **precision**, which a single number hides.

### 8. Common mistakes
- The "95% probability μ is in this interval" wording.
- Using z = 1.96 with very small samples instead of the t critical value.
- Forgetting that a CI only accounts for **random sampling error**, not bias (a biased sample gives a confidently wrong CI).

# Part 15 — Hypothesis Testing

### 1. Simple explanation

> **Analogy — a court trial.** The defendant is **presumed innocent** (H0). The prosecutor brings **evidence**. If the evidence would be **very unlikely if the person were innocent**, the jury **rejects innocence** ("guilty"). Otherwise the verdict is "**not guilty**" — which is **not** the same as "proven innocent".

| Term | Meaning | Court analogy |
|---|---|---|
| **Null hypothesis (H0)** | "Nothing special is happening": no effect, no difference | Innocent |
| **Alternative hypothesis (H1 / Ha)** | What we suspect: there *is* an effect/difference | Guilty |
| **Test statistic** | A number measuring how far the data is from what H0 predicts (t, χ², F…) | Strength of the evidence |
| **p-value** | Probability of seeing data **at least this extreme if H0 were true** | "How surprising is this evidence if innocent?" |
| **Significance level α** | The threshold we set **before** the test (commonly 0.05) | "Beyond reasonable doubt" |
| **Reject H0** | p ≤ α → evidence is strong enough | Guilty |
| **Fail to reject H0** | p > α → not enough evidence | Not guilty (not "innocent"!) |

### Steps
```text
1. State H0 and H1
2. Choose α (e.g. 0.05) BEFORE looking at the data
3. Check assumptions and pick a test
4. Compute the test statistic and p-value
5. Decide: p ≤ α → reject H0; p > α → fail to reject H0
6. Interpret in plain language + report the effect size / CI
```

### One-sided vs two-sided
- **Two-sided** H1: "the mean is **different** from X" (either direction) — the default.
- **One-sided** H1: "the mean is **greater** than X" — only when you decided the direction in advance.

### 2. Real-world example
A food-delivery company claims the **average delivery time is 30 minutes**. You record 40 deliveries.
- H0: μ = 30 minutes
- H1: μ ≠ 30 minutes

````python
delivery = rng.normal(loc=32.5, scale=6, size=40).round(1)       # (unknown to us) true mean is 32.5

t_stat, p_value = stats.ttest_1samp(delivery, popmean=30)
alpha = 0.05
print(f'Sample mean = {delivery.mean():.2f} min, n = {len(delivery)}')
print(f't statistic = {t_stat:.3f}')
print(f'p-value     = {p_value:.4f}')
print('Decision    :', 'Reject H0' if p_value <= alpha else 'Fail to reject H0')
````

### 6. Interpretation
The **t statistic** says how many standard errors the sample mean is away from 30. If the p-value is below 0.05, we conclude: *"The data provides evidence that the average delivery time is different from 30 minutes"* (here: longer). If it's above 0.05 we'd say *"we did not find enough evidence that it differs from 30"* — **not** "it is 30".

### 8. Common mistakes
- "Accepting" H0. We only **fail to reject** it.
- Choosing α or one-sided/two-sided **after** seeing the result.
- Reporting only "significant / not significant" without the effect size and CI.

# Part 16 — The p-value (explained carefully)

### 1. Definition in simple English
> **The p-value is the probability of getting a result at least as extreme as the one you observed, _assuming the null hypothesis is true_.**

It answers: *"If nothing special were going on, how surprising would my data be?"*

- **Small p-value** (e.g. 0.01): data like this would be **rare** if H0 were true → evidence **against** H0.
- **Large p-value** (e.g. 0.40): data like this is **common** under H0 → **no evidence** against H0 (but also not proof that H0 is true).

### 3. Numerical example — a coin
You flip a coin 100 times and get **62 heads**. H0: the coin is fair (p = 0.5).
If the coin were fair, getting 62 or more heads (or 38 or fewer, for two-sided) happens with probability ≈ **0.02**. That's rare → evidence the coin is biased.

````python
result = stats.binomtest(62, n=100, p=0.5, alternative='two-sided')
print('Exact p-value for 62/100 heads:', round(result.pvalue, 4))

# Simulate: flip a FAIR coin 100 times, repeated 100,000 times
heads = rng.binomial(100, 0.5, 100_000)
p_sim = np.mean(np.abs(heads - 50) >= 12)                      # as or more extreme than 62 (|62 − 50| = 12)
print('Simulated p-value              :', round(p_sim, 4))

plt.figure(figsize=(9, 3.8))
bins = np.arange(heads.min(), heads.max() + 2) - 0.5
plt.hist(heads, bins=bins, color='lightgray')
plt.hist(heads[np.abs(heads - 50) >= 12], bins=bins, color='red', label='as or more extreme than 62')
plt.axvline(62, color='black', ls='--')
plt.title('What a FAIR coin produces in 100 flips (the null distribution)')
plt.xlabel('Number of heads'); plt.legend(); plt.show()
````

The red area **is** the p-value: the share of fair-coin experiments that look at least as unusual as ours.

### ❌ What the p-value does NOT mean

| Misconception | Why it's wrong |
|---|---|
| "p = 0.03 means there's a 3% chance H0 is true" | The p-value **assumes** H0 is true; it can't be the probability that H0 is true |
| "p = 0.03 means a 97% chance my hypothesis is correct" | Same error |
| "p > 0.05 proves there is no effect" | Absence of evidence ≠ evidence of absence (maybe the sample was too small) |
| "A smaller p-value means a bigger / more important effect" | p depends heavily on **sample size**; tiny effects become 'significant' with huge n |
| "p = 0.049 is real, p = 0.051 is nothing" | 0.05 is a convention, not a law of nature |
| "p is the probability the result happened by chance" | It's the probability of the data (or more extreme) **given** chance alone |

### Demonstration 1 — with no real effect, p-values are spread evenly
If H0 is true, p-values are **uniformly distributed** between 0 and 1: about 5% fall below 0.05 **purely by chance**.

````python
p_null = [stats.ttest_ind(rng.normal(100, 15, 30), rng.normal(100, 15, 30)).pvalue for _ in range(5000)]
p_null = np.array(p_null)
print(f'Share of p < 0.05 when there is NO difference: {np.mean(p_null < 0.05):.3f}')

plt.figure(figsize=(8, 3.5))
sns.histplot(p_null, bins=20, color='lightsteelblue')
plt.axvline(0.05, color='red', ls='--', label='α = 0.05')
plt.title('5,000 A/B tests with NO real difference: p-values are uniform'); plt.legend(); plt.show()
````

→ If you run **20 tests** on random noise, you should expect about **one** "significant" result. This is the **multiple testing** problem (fix: Bonferroni correction — divide α by the number of tests).

### Demonstration 2 — statistical vs practical significance

````python
rows = []
for n in [50, 500, 5000, 50000]:
    a = rng.normal(100.0, 15, n)
    b = rng.normal(100.5, 15, n)                   # a tiny true difference: 0.5 points
    rows.append({'n per group': n, 'Observed difference': b.mean() - a.mean(),
                 'p-value': stats.ttest_ind(a, b).pvalue})
pd.DataFrame(rows).round(4)
````

The true difference (0.5 points on a scale where people vary by ±15) is practically meaningless, yet with 50,000 per group it becomes "highly significant". **Always report the effect size and a confidence interval, not just p.**

# Part 17 — Type I and Type II Errors

### 1. Simple explanation — a medical test for a disease
H0: **the patient is healthy.** H1: the patient has the disease.

| | **Truth: healthy (H0 true)** | **Truth: sick (H0 false)** |
|---|---|---|
| **Test says: sick** (reject H0) | ❌ **Type I error — False Positive** (healthy person told they're sick) | ✅ Correct (True Positive) |
| **Test says: healthy** (fail to reject H0) | ✅ Correct (True Negative) | ❌ **Type II error — False Negative** (sick person sent home) |

| Symbol | Meaning |
|---|---|
| **α (alpha)** | Probability of a **Type I** error = the significance level (e.g. 0.05) |
| **β (beta)** | Probability of a **Type II** error |
| **Power = 1 − β** | Probability of **detecting a real effect** when it exists (commonly aim for ≥ 0.80) |

### Which error is worse? It depends
- **Cancer screening:** a false **negative** (missing cancer) is worse → accept more false positives, follow up with more tests.
- **Spam filter:** a false **positive** (an important email goes to spam) is worse.
- **Court:** convicting an innocent person (Type I) is considered worse → strict α.

### The trade-off
Making α smaller (stricter) → fewer false positives but **more** false negatives (lower power). The only way to reduce **both** is more data (or a better test).

### What increases power?
1. Larger **sample size** n
2. Larger true **effect size**
3. Larger α (at the cost of more Type I errors)
4. Less noise (lower variability)

````python
def simulated_power(n, effect, sd=15, alpha=0.05, n_sim=2000):
    '''Share of experiments that detect a real difference of `effect` points.'''
    hits = 0
    for _ in range(n_sim):
        a = rng.normal(100, sd, n)
        b = rng.normal(100 + effect, sd, n)
        hits += stats.ttest_ind(a, b).pvalue < alpha
    return hits / n_sim

sizes = [10, 20, 40, 80, 160]
effects = [3, 6, 10]
power_table = pd.DataFrame({f'effect = {e}': [simulated_power(n, e) for n in sizes] for e in effects},
                           index=pd.Index(sizes, name='n per group'))
print(power_table.round(2))

power_table.plot(marker='o', figsize=(8, 4))
plt.axhline(0.8, color='red', ls='--', label='80% power target')
plt.ylabel('Power (1 − β)'); plt.title('Power grows with sample size and effect size'); plt.legend(); plt.show()
````

### 6. Interpretation
- To detect a **small** effect (3 points) reliably you need far more data than for a **large** effect (10 points).
- A study with low power often "finds nothing" even when the effect is real → **plan the sample size before the experiment** (power analysis).

### Connection to ML (classification)
- False positive rate ↔ Type I error; false negative rate ↔ Type II error.
- **Precision/recall** and the **decision threshold** control this trade-off (future topic).

### 8. Common mistakes
- Thinking α = 0.05 means "5% of my significant results are false" (that's the **false discovery rate**, a different quantity).
- Running an under-powered A/B test and concluding "the new design doesn't work".

# Part 18 — Statistical Tests

### Which test should I use?

```text
What is the OUTCOME variable?
 │
 ├── Numerical
 │     ├── 1 group vs a known value ............ One-sample t-test
 │     ├── 2 independent groups ................ Independent t-test  (non-normal/ordinal → Mann-Whitney U)
 │     ├── 2 related measurements (before/after) Paired t-test       (non-normal → Wilcoxon signed-rank)
 │     └── 3+ independent groups ............... ANOVA               (non-normal → Kruskal-Wallis)
 │
 └── Categorical
       └── relationship between 2 categorical variables ... Chi-square test of independence
```

- **Parametric** tests (t-tests, ANOVA) compare **means** and assume roughly normal data (or large samples via CLT).
- **Non-parametric** tests (Mann-Whitney, Wilcoxon, Kruskal-Wallis) use **ranks**, make fewer assumptions, and are robust to outliers/skew. They compare distributions / typical values rather than means.

---
<a id="sec-19"></a>
## 18.1 One-sample t-test

| | |
|---|---|
| **Purpose** | Is a group's mean different from a specific known/claimed value? |
| **Data** | One numerical variable |
| **Assumptions** | Independent observations; roughly normal data or n ≳ 30 |
| **H0** | μ = μ₀ |
| **H1** | μ ≠ μ₀ (or > / <) |
| **Example** | A bottling machine should fill 500 ml. Is the average fill different? |

````python
fills = rng.normal(497.8, 4, 35).round(1)
res = stats.ttest_1samp(fills, popmean=500)
ci = res.confidence_interval(0.95)
print(f'Mean fill = {fills.mean():.2f} ml | t = {res.statistic:.3f} | p = {res.pvalue:.4f}')
print(f'95% CI for the true mean: ({ci.low:.2f}, {ci.high:.2f})')
````

**Interpretation:** if p < 0.05, the machine's average fill differs from 500 ml; the CI shows by how much (and whether 500 lies inside it — if it doesn't, that matches p < 0.05).

---
<a id="sec-20"></a>
## 18.2 Independent two-sample t-test

| | |
|---|---|
| **Purpose** | Do **two separate groups** have different means? |
| **Data** | Numerical outcome + a 2-level group variable; different people in each group |
| **Assumptions** | Independence; roughly normal in each group (or large n); equal variances for the classic version — use **Welch's t-test** (`equal_var=False`) which doesn't need them |
| **H0** | μ₁ = μ₂ |
| **H1** | μ₁ ≠ μ₂ |
| **Example** | A/B test: do users of website design B spend more than users of design A? |

````python
design_a = rng.normal(1200, 300, 150)
design_b = rng.normal(1320, 310, 150)

res = stats.ttest_ind(design_a, design_b, equal_var=False)       # Welch's t-test (safe default)
diff = design_b.mean() - design_a.mean()
print(f'Mean A = {design_a.mean():.0f}, Mean B = {design_b.mean():.0f}, difference = {diff:.0f}')
print(f't = {res.statistic:.3f}, p = {res.pvalue:.4f}')
print('95% CI for mean(A) − mean(B):', np.round(res.confidence_interval(0.95), 0))

# Effect size: Cohen's d = difference / pooled std  (≈0.2 small, 0.5 medium, 0.8 large)
pooled_sd = np.sqrt((design_a.var(ddof=1) + design_b.var(ddof=1)) / 2)
print(f"Cohen's d = {diff / pooled_sd:.2f}")
````

**Interpretation:** p < 0.05 → evidence that the two designs lead to different average spending. The CI gives a plausible range for the difference, and **Cohen's d** tells us how large it is in standard-deviation units.

---
<a id="sec-21"></a>
## 18.3 Paired t-test

| | |
|---|---|
| **Purpose** | Compare **two measurements on the same subjects** (before/after, left/right) |
| **Data** | Two numerical columns, row-matched |
| **Assumptions** | Pairs are independent of each other; the **differences** are roughly normal |
| **H0** | Mean difference = 0 |
| **H1** | Mean difference ≠ 0 |
| **Example** | Employees' productivity scores before and after a training course |

Why not an independent test? People differ a lot from each other; pairing removes that person-to-person variation and makes the test **much more powerful**.

````python
before = rng.normal(60, 12, 30).round(1)
after = (before + rng.normal(3, 3, 30)).round(1)          # same people, improved by ~3 points

paired = stats.ttest_rel(after, before)
wrong = stats.ttest_ind(after, before)                    # ❌ ignores the pairing
print(f'Mean improvement = {(after - before).mean():.2f}')
print(f'Paired t-test      : t = {paired.statistic:.3f}, p = {paired.pvalue:.5f}')
print(f'(Wrong) independent: t = {wrong.statistic:.3f}, p = {wrong.pvalue:.5f}  ← misses the effect')
````

<a id="sec-24"></a>
## 18.6 Mann-Whitney U test (non-parametric alternative to the independent t-test)

| | |
|---|---|
| **Purpose** | Do two independent groups differ? — tends to be larger in one group? |
| **Data** | Numerical or **ordinal** (ratings), skewed data, outliers |
| **Assumptions** | Independent observations; similar shapes if you want to interpret it as a difference in medians |
| **H0** | The two distributions are the same (a random value from A is equally likely to be above or below one from B) |
| **H1** | One group tends to have larger values |
| **Example** | Customer satisfaction ratings (1–5) for two support teams; time-on-site (very skewed) for two app versions |

````python
time_a = rng.lognormal(3.0, 0.9, 60)          # very skewed session times (minutes)
time_b = rng.lognormal(3.4, 0.9, 60)
mw = stats.mannwhitneyu(time_a, time_b, alternative='two-sided')
tt = stats.ttest_ind(time_a, time_b, equal_var=False)
print(f'Median A = {np.median(time_a):.1f}, Median B = {np.median(time_b):.1f}')
print(f'Mann-Whitney U = {mw.statistic:.0f}, p = {mw.pvalue:.4f}')
print(f"(Welch t-test p = {tt.pvalue:.4f} — less reliable here because of the heavy skew)")
````

<a id="sec-25"></a>
## 18.7 Wilcoxon signed-rank test (non-parametric alternative to the paired t-test)

| | |
|---|---|
| **Purpose** | Compare two **related** measurements when the differences are not normal |
| **Data** | Paired numerical/ordinal data |
| **Assumptions** | Pairs independent; differences roughly symmetric around their median |
| **H0** | The median of the differences is 0 |
| **H1** | The median difference ≠ 0 |
| **Example** | Pain score (0–10) before and after a treatment, for the same patients |

````python
pain_before = rng.integers(4, 10, 20)
pain_after = np.clip(pain_before - rng.choice([0, 1, 1, 2, 3], 20), 0, 10)
w = stats.wilcoxon(pain_before, pain_after)
print(f'Median before = {np.median(pain_before)}, after = {np.median(pain_after)}')
print(f'Wilcoxon statistic = {w.statistic}, p = {w.pvalue:.5f}')
````

| Test | Groups | Paired? | Outcome | Compares | SciPy |
|---|---|---|---|---|---|
| One-sample t | 1 vs value | — | Numerical | Mean | `stats.ttest_1samp` |
| Independent t (Welch) | 2 | No | Numerical | Means | `stats.ttest_ind(a, b, equal_var=False)` |
| Paired t | 2 | **Yes** | Numerical | Mean difference | `stats.ttest_rel` |
| Mann-Whitney U | 2 | No | Numerical/ordinal | Ranks | `stats.mannwhitneyu` |
| Wilcoxon signed-rank | 2 | **Yes** | Numerical/ordinal | Ranks of differences | `stats.wilcoxon` |
| One-way ANOVA | 3+ | No | Numerical | Means | `stats.f_oneway` |
| Kruskal-Wallis | 3+ | No | Numerical/ordinal | Ranks | `stats.kruskal` |
| Chi-square | 2 categorical vars | No | Counts | Independence | `stats.chi2_contingency` |

### Common mistakes
- Using an independent test on paired data (loses power) or a paired test on independent groups (wrong).
- Running many t-tests between 3+ groups instead of ANOVA (inflates Type I error).
- Using a t-test on tiny, heavily skewed samples — use the non-parametric version.

# Part 19 — ANOVA (Analysis of Variance)

### 1. Why ANOVA exists
With 3 departments you could run 3 t-tests (IT–HR, IT–Finance, HR–Finance). But each test has a 5% false-positive risk, so the chance of **at least one** false positive becomes ≈ 1 − 0.95³ ≈ **14%**. With 10 groups (45 pairs) it's ≈ 90%! ANOVA tests **all groups at once** with a single α.

### 3. How it works — the F-statistic
ANOVA compares two kinds of variation:

```text
F = (variation BETWEEN group means) / (variation WITHIN groups)
```

- If the group means are far apart **compared with** the natural spread inside each group → **large F** → small p-value → at least one mean differs.
- If the group means are close compared with the within-group spread → F ≈ 1 → no evidence of a difference.

> **Analogy:** three classes' average scores differ by 2 points, but students within each class vary by ±20 points → that 2-point gap is probably noise (small F). If the averages differed by 20 points and students varied by only ±3, the gap is clearly real (large F).

### 2. Example: salary across 3 departments
- H0: μ_IT = μ_HR = μ_Finance
- H1: at least one department's mean salary is different

````python
dept = pd.DataFrame({
    'Department': np.repeat(['IT', 'HR', 'Finance'], 30),
    'Salary': np.concatenate([rng.normal(120000, 15000, 30),
                              rng.normal(95000, 15000, 30),
                              rng.normal(110000, 15000, 30)]).round(-2),
})
print(dept.groupby('Department')['Salary'].agg(['mean', 'std', 'count']).round(0))

groups = [g['Salary'].to_numpy() for _, g in dept.groupby('Department')]
f_stat, p_value = stats.f_oneway(*groups)
print(f'\nF = {f_stat:.2f}, p = {p_value:.6f}')

plt.figure(figsize=(7, 4))
sns.boxplot(data=dept, x='Department', y='Salary', order=['IT', 'Finance', 'HR'])
sns.stripplot(data=dept, x='Department', y='Salary', order=['IT', 'Finance', 'HR'], color='black', alpha=0.4, size=3)
plt.title(f'Salary by department (ANOVA F = {f_stat:.1f}, p = {p_value:.2g})'); plt.show()
````

````python
# Manual F calculation (to see where the number comes from)
grand_mean = dept['Salary'].mean()
k, N = dept['Department'].nunique(), len(dept)
g = dept.groupby('Department')['Salary']

ss_between = (g.count() * (g.mean() - grand_mean) ** 2).sum()           # how far group means are from the overall mean
ss_within = ((dept['Salary'] - g.transform('mean')) ** 2).sum()          # spread inside each group

ms_between = ss_between / (k - 1)
ms_within = ss_within / (N - k)
print(f'MS between = {ms_between:,.0f}')
print(f'MS within  = {ms_within:,.0f}')
print(f'F = {ms_between / ms_within:.2f}  (matches SciPy)')
````

### Post-hoc testing — WHICH groups differ?
ANOVA only says "**at least one** mean is different". To find **which pairs** differ we run a **post-hoc test** that controls the overall error rate, e.g. **Tukey's HSD** (Honestly Significant Difference). Conceptually it's "all pairwise comparisons, with an adjustment so the total false-positive risk stays at α".

````python
tukey = stats.tukey_hsd(*groups)
names = [name for name, _ in dept.groupby('Department')]
rows = []
for i in range(len(names)):
    for j in range(i + 1, len(names)):
        rows.append({'Group 1': names[i], 'Group 2': names[j],
                     'Mean diff': groups[i].mean() - groups[j].mean(),
                     'p-value (adjusted)': tukey.pvalue[i, j]})
pd.DataFrame(rows).round(4)
````

### 6. Interpretation
A small ANOVA p-value says the department means are not all equal; Tukey's table shows **which** pairs differ after adjusting for multiple comparisons.

### Non-parametric alternative
`stats.kruskal(*groups)` — Kruskal-Wallis, for skewed data or ordinal outcomes.

### 8. Common mistakes
- Stopping after a significant ANOVA and claiming "all groups differ".
- Running many unadjusted t-tests instead of ANOVA + post-hoc.
- Ignoring very unequal variances (use Welch's ANOVA or Kruskal-Wallis).

# Part 20 — Chi-Square Test of Independence

### 1. Simple explanation
For two **categorical** variables (e.g. `Gender` and `Preferred_Product`), we ask: are they **independent** (knowing one tells you nothing about the other), or **associated**?

| Term | Meaning |
|---|---|
| **Observed frequency (O)** | The actual counts in each cell of the table |
| **Expected frequency (E)** | The counts we'd expect **if the variables were independent**: `E = (row total × column total) / grand total` |
| **Independence** | The proportions in each row are the same |

$$\chi^2 = \sum \frac{(O - E)^2}{E}$$

Big differences between observed and expected → large χ² → small p-value → evidence of association.

### 3. Numerical example
200 customers, Gender × Product:

| | Laptop | Phone | Row total |
|---|---|---|---|
| Male | 70 | 30 | 100 |
| Female | 40 | 60 | 100 |
| Col total | 110 | 90 | 200 |

Expected (Male, Laptop) = 100 × 110 / 200 = **55**, but observed = 70 → a big gap.
Expected: Male (55, 45), Female (55, 45).
χ² = (70−55)²/55 + (30−45)²/45 + (40−55)²/55 + (60−45)²/45 = 4.09 + 5.00 + 4.09 + 5.00 = **18.18** (without continuity correction).

````python
table = pd.DataFrame({'Laptop': [70, 40], 'Phone': [30, 60]}, index=['Male', 'Female'])
chi2, p, dof, expected = stats.chi2_contingency(table, correction=False)   # correction=False to match the hand calculation
print('Observed:\n', table, '\n')
print('Expected:\n', pd.DataFrame(expected, index=table.index, columns=table.columns), '\n')
print(f'χ² = {chi2:.2f}, df = {dof}, p = {p:.6f}')
````

(By default SciPy applies **Yates' continuity correction** for 2×2 tables, which gives a slightly smaller χ² — safer for small counts.)

### Building a contingency table from raw data

````python
n = 400
cust = pd.DataFrame({'City': rng.choice(['Lahore', 'Karachi', 'Islamabad'], n, p=[0.45, 0.35, 0.2])})
payment_probs = {'Lahore': [0.5, 0.3, 0.2], 'Karachi': [0.3, 0.5, 0.2], 'Islamabad': [0.6, 0.2, 0.2]}
cust['Payment'] = [rng.choice(['Card', 'Cash', 'Wallet'], p=payment_probs[c]) for c in cust['City']]

ct = pd.crosstab(cust['City'], cust['Payment'])
chi2, p, dof, expected = stats.chi2_contingency(ct)
print(ct, '\n')
print(f'χ² = {chi2:.2f}, df = {dof}, p = {p:.6f}')
print('Smallest expected count:', expected.min().round(1), '(should be ≥ 5)')

# Cramér's V = strength of association (0 = none, 1 = perfect)
cramers_v = np.sqrt(chi2 / (ct.to_numpy().sum() * (min(ct.shape) - 1)))
print(f"Cramér's V = {cramers_v:.3f}")

pd.crosstab(cust['City'], cust['Payment'], normalize='index').plot(kind='bar', stacked=True, figsize=(7, 4))
plt.ylabel('Share within city'); plt.title('Payment method share by city'); plt.xticks(rotation=0); plt.show()
````

### 6. Interpretation
A small p-value → payment method and city are **associated** (e.g. Karachi prefers cash, Islamabad prefers cards). Cramér's V tells **how strong** the association is. The stacked bar chart shows the pattern.

### 7. When to use
Categorical vs categorical: gender vs product, city vs churn, ad version vs clicked (yes/no).

### 8. Common mistakes
- Using percentages instead of **counts** in the table.
- Expected counts below 5 in many cells (combine categories or use Fisher's exact test: `stats.fisher_exact` for 2×2).
- Treating p as the strength of association (use Cramér's V).

# Part 21 — Probability Fundamentals (preparing for the next topic)

| Term | Simple meaning | Example (rolling a die) |
|---|---|---|
| **Experiment** | A process with an uncertain result | Rolling a die |
| **Outcome** | One possible result | Rolling a 4 |
| **Sample space (S)** | The set of **all** possible outcomes | {1, 2, 3, 4, 5, 6} |
| **Event** | A set of outcomes we care about | "Even number" = {2, 4, 6} |
| **Probability P(A)** | How likely an event is, from 0 (impossible) to 1 (certain) | P(even) = 3/6 = 0.5 |

For equally likely outcomes:
$$P(A) = \frac{\text{number of outcomes in } A}{\text{number of outcomes in } S}$$

### Rules
- 0 ≤ P(A) ≤ 1, and the probabilities of all outcomes add to 1.
- **Complement:** P(not A) = 1 − P(A).
- **Addition (OR):** P(A or B) = P(A) + P(B) − P(A and B).

### Independent vs dependent events

| | Meaning | Rule | Example |
|---|---|---|---|
| **Independent** | One event doesn't change the probability of the other | P(A and B) = P(A) × P(B) | Two coin flips: P(H, H) = 0.5 × 0.5 = 0.25 |
| **Dependent** | One event **changes** the other's probability | P(A and B) = P(A) × P(B \| A) | Drawing 2 aces **without** replacement: 4/52 × 3/51 ≈ 0.0045 |

`P(B | A)` = "probability of B **given** A has happened" — **conditional probability** (the basis of Bayes' theorem, next topic).

### Simulation — the Law of Large Numbers
As you repeat an experiment more times, the observed frequency gets closer to the true probability.

````python
rolls = rng.integers(1, 7, 10_000)
running_share_even = np.cumsum(rolls % 2 == 0) / np.arange(1, len(rolls) + 1)

print('P(even) theory  : 0.5')
print('P(even) observed:', round(np.mean(rolls % 2 == 0), 4))

# Independent: two coins
coins = rng.integers(0, 2, size=(100_000, 2))
print('\nP(two heads) theory 0.25, simulated:', round(np.mean(coins.sum(axis=1) == 2), 4))

# Dependent: two aces without replacement
deck = np.array([1] * 4 + [0] * 48)                    # 1 = ace
draws = np.array([rng.choice(deck, 2, replace=False).sum() == 2 for _ in range(100_000)])
print('P(two aces) theory', round(4 / 52 * 3 / 51, 5), ', simulated:', round(draws.mean(), 5))

plt.figure(figsize=(8, 3.5))
plt.plot(running_share_even); plt.axhline(0.5, color='red', ls='--')
plt.xscale('log'); plt.xlabel('Number of rolls (log scale)'); plt.ylabel('Share of even numbers')
plt.title('Law of Large Numbers: observed frequency → true probability'); plt.show()
````

### 8. Common mistakes
- **Gambler's fallacy:** "5 heads in a row, so tails is due" — independent flips have no memory.
- Multiplying probabilities of **dependent** events as if they were independent.

# Part 22 — Statistical Assumptions

Most tests and models give **trustworthy** p-values and CIs only if certain assumptions (approximately) hold.

| Assumption | Meaning | Used by | How to check |
|---|---|---|---|
| **Random sampling** | The sample represents the population | Everything | Study design (not a test!) |
| **Independence** | One observation doesn't influence another | t-tests, ANOVA, chi-square, regression | Design; for time data, residual autocorrelation |
| **Normality** | Data (or residuals / differences) roughly bell-shaped | t-tests, ANOVA, linear-regression inference | Histogram, **Q-Q plot**, Shapiro-Wilk test |
| **Homoscedasticity** | Groups (or residuals) have **similar variance** | Classic t-test, ANOVA, linear regression | Boxplots, residual plot, **Levene's test** |
| **Linearity** | The relationship is a straight line | Pearson correlation, linear regression | Scatter plot, residual plot |

### Why they matter
If assumptions are badly violated, p-values and confidence intervals can be **wrong** (too optimistic or too pessimistic), even though the software still prints numbers.

### What to do if an assumption fails
- Non-normal → larger samples (CLT), **transformation** (Topic 13), or a **non-parametric** test.
- Unequal variances → **Welch's t-test** / Welch's ANOVA.
- Non-linear → transform variables or use a non-linear model.
- Non-independent (repeated measures, time series) → paired tests, mixed models, time-series methods.

````python
normal_data = rng.normal(50, 10, 80)
skewed_data = rng.exponential(10, 80)

print('Shapiro-Wilk normality test (H0: data is normal)')
for name, d in [('normal_data', normal_data), ('skewed_data', skewed_data)]:
    stat, p = stats.shapiro(d)
    print(f'  {name:<12} W = {stat:.3f}, p = {p:.4f} →', 'looks normal' if p > 0.05 else 'NOT normal')

print("\nLevene's test (H0: equal variances)")
g1, g2 = rng.normal(0, 1, 50), rng.normal(0, 3, 50)
print(f'  p = {stats.levene(g1, g2).pvalue:.6f} → variances differ → use Welch (equal_var=False)')

fig, axes = plt.subplots(1, 3, figsize=(16, 4))
stats.probplot(normal_data, dist='norm', plot=axes[0]); axes[0].set_title('Q-Q plot: normal data (points on the line)')
stats.probplot(skewed_data, dist='norm', plot=axes[1]); axes[1].set_title('Q-Q plot: skewed data (curves away)')

# Linearity & homoscedasticity via a residual plot
x = rng.uniform(1, 10, 150)
y = 3 * x + rng.normal(0, x, 150)                     # noise grows with x → heteroscedastic
slope, intercept = np.polyfit(x, y, 1)
resid = y - (slope * x + intercept)
axes[2].scatter(slope * x + intercept, resid, alpha=0.6); axes[2].axhline(0, color='red', ls='--')
axes[2].set_title('Residuals fan out → heteroscedasticity'); axes[2].set_xlabel('Fitted value')
plt.tight_layout(); plt.show()
````

### 8. Common mistakes
- Relying only on Shapiro-Wilk: with **huge** samples it rejects normality for tiny, harmless deviations; with **tiny** samples it can't detect anything. Look at the **Q-Q plot** too.
- Checking normality of the raw data when the assumption is about **residuals** (regression) or **differences** (paired test).
- Ignoring independence — the most important and least testable assumption.

# Part 23 — Statistics in Machine Learning

| ML stage | Where statistics appears | Examples |
|---|---|---|
| **Data preprocessing** | Summaries, missing values, scaling | Mean/median imputation; standardization uses mean & std (z-scores) |
| **Outlier detection** | Spread & distribution | IQR rule, z-score > 3, MAD |
| **Feature engineering** | Distributions & transformations | Log for skewed features, binning by quantiles, frequency/target encoding (means) |
| **Feature selection** | Relationships & tests | Correlation, chi-square test for categorical features, ANOVA F-test (`SelectKBest(f_classif)`), VIF |
| **A/B testing** | Hypothesis tests, power, CIs | Does the new recommendation model increase conversion? |
| **Model evaluation** | Sampling, CIs, CLT | Cross-validation mean ± std; CI for accuracy; comparing two models with a paired test on fold scores |
| **Confidence intervals** | Precision of estimates | "Accuracy = 87% ± 2%" |
| **Hypothesis testing** | Deciding if differences are real | Is model B really better than model A, or is it noise? |
| **Experiment analysis** | Design, bias, randomization | Randomly assign users, avoid peeking, correct for multiple tests |
| **Model assumptions** | Linearity, normal residuals, homoscedasticity | Linear regression diagnostics |
| **Probabilistic models** | Probability & distributions | Naive Bayes, logistic regression, uncertainty estimates |

### Example: confidence interval for a model's accuracy

````python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold

X, y = make_classification(n_samples=600, n_features=10, n_informative=5, random_state=0)
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=0)

scores_lr = cross_val_score(LogisticRegression(max_iter=1000), X, y, cv=cv)
scores_rf = cross_val_score(RandomForestClassifier(n_estimators=200, random_state=0), X, y, cv=cv)

for name, sc in [('Logistic Regression', scores_lr), ('Random Forest', scores_rf)]:
    low, high = stats.t.interval(0.95, df=len(sc) - 1, loc=sc.mean(), scale=stats.sem(sc))
    print(f'{name:<20} accuracy = {sc.mean():.3f}  (95% CI {low:.3f} – {high:.3f})')

# Same folds for both models → the scores are PAIRED
res = stats.ttest_rel(scores_rf, scores_lr)
print(f'\nPaired t-test on fold scores: p = {res.pvalue:.4f}')
````

### Interpretation
- Each model's accuracy comes with a CI that shows how much it could vary.
- Because both models were evaluated on the **same folds**, a **paired** test is used.
- ⚠️ Caveat: cross-validation folds share training data, so the fold scores are not fully independent — this simple test is a useful rough check, not a perfect one (more advanced corrected tests exist).

### Solution 12

````python
rng_p = np.random.default_rng(12)
def power(n, effect=5, sd=15, sims=3000):
    return np.mean([stats.ttest_ind(rng_p.normal(0, sd, n), rng_p.normal(effect, sd, n)).pvalue < 0.05
                    for _ in range(sims)])
print('Power with n = 30 per group :', power(30))
print('Power with n = 150 per group:', power(150))
````

**Explanation:**
- **Type I (false positive):** a legitimate transaction is flagged as fraud → annoyed customer, blocked card. Costly for the **customer** experience.
- **Type II (false negative):** a fraudulent transaction is **not** flagged → direct money loss. Usually more costly for the **bank**.
- Power: with 30 per group a 5-point difference is detected only about a quarter of the time; with 150 per group, most of the time. Small studies miss real effects.

### Solution 13

````python
rng_p = np.random.default_rng(13)
team_a = rng_p.normal(72, 10, 40)
team_b = rng_p.normal(77, 12, 45)
w13 = stats.ttest_ind(team_a, team_b, equal_var=False)
d13 = (team_b.mean() - team_a.mean()) / np.sqrt((team_a.var(ddof=1) + team_b.var(ddof=1)) / 2)
print(f"(a) Welch t = {w13.statistic:.2f}, p = {w13.pvalue:.4f}, Cohen's d = {d13:.2f}")

before = rng_p.normal(80, 8, 20)
after = before - rng_p.normal(2, 3, 20)
print(f'(b) Paired t-test      p = {stats.ttest_rel(before, after).pvalue:.5f}')
print(f'    Independent (wrong) p = {stats.ttest_ind(before, after).pvalue:.5f}')
````

**Explanation:**
- (a) Welch's test handles the different group sizes and variances. Here the observed effect is **small** (Cohen's d ≈ 0.2) and p > 0.05, so we **fail to reject H0**. The data was generated with a true 5-point difference, but ~40 people per group gives **low power** for an effect this size. "Not significant" ≠ "no difference" (Part 16/17).
- (b) The paired test detects the ~2-point drop clearly because it compares each person with **themselves**; the independent test is swamped by the large differences **between** people and misses it.

### Solution 14

````python
survey = pd.DataFrame({'Tea': [40, 75], 'Coffee': [70, 50], 'Juice': [40, 25]}, index=['Age < 30', 'Age ≥ 30'])
chi2_14, p14, dof14, exp14 = stats.chi2_contingency(survey)
print('Expected counts:\n', pd.DataFrame(exp14, index=survey.index, columns=survey.columns).round(1))
print(f'\nχ² = {chi2_14:.2f}, df = {dof14}, p = {p14:.6f}')
v14 = np.sqrt(chi2_14 / (survey.to_numpy().sum() * (min(survey.shape) - 1)))
print(f"Cramér's V = {v14:.3f}")
````

**Explanation:** the very small p-value means drink preference is **not independent** of age: younger people choose coffee and juice more, older people choose tea more. Cramér's V ≈ 0.25 → a weak-to-moderate association.

### Solution 15

````python
rng_p = np.random.default_rng(15)
A = rng_p.normal(30, 5, 25); B = rng_p.normal(33, 5, 25); C = rng_p.normal(30.5, 5, 25)
print('H0: μA = μB = μC | H1: at least one mean differs')
f15, p15 = stats.f_oneway(A, B, C)
print(f'F = {f15:.2f}, p = {p15:.4f}')
print(f"Levene p = {stats.levene(A, B, C).pvalue:.3f} (> 0.05 → equal variances OK)")

t15 = stats.tukey_hsd(A, B, C)
labels = ['A', 'B', 'C']
for i in range(3):
    for j in range(i + 1, 3):
        print(f'{labels[i]} vs {labels[j]}: diff = {[A, B, C][i].mean() - [A, B, C][j].mean():6.2f}, adjusted p = {t15.pvalue[i, j]:.4f}')
````

**Explanation:**
- H0: all three companies have the same mean delivery time; H1: at least one differs.
- Here the ANOVA p-value is **well above 0.05** → we **fail to reject H0**. Normally you would **stop here** and not run a post-hoc test; the Tukey table is printed only to show that no pair differs either.
- Yet company B's true mean was 3 minutes slower! With 25 deliveries per company and a spread of ±5 minutes, this experiment has **low power** for a 3-minute difference. A power analysis before collecting data would have suggested a larger sample.
- Levene's p ≈ 0.99 → the equal-variance assumption is fine.

# Part 25 — Final Project: Employee Salary Statistical Analysis

**Questions we want to answer:**
1. What is a typical salary, and how spread out are salaries?
2. What does the salary distribution look like? Is it skewed? Are there outliers?
3. How are salary, experience and age related?
4. What is a plausible range for the company's true average salary?
5. Do men and women earn different average salaries?
6. Do departments differ in salary?
7. Is department related to education level?

## Step 0 — The dataset (80 employees)

````python
rproj = np.random.default_rng(2025)
n = 80
department = rproj.choice(['IT', 'Finance', 'HR', 'Sales'], n, p=[0.35, 0.25, 0.15, 0.25])
edu_probs = {'IT': [0.3, 0.55, 0.15], 'Finance': [0.35, 0.55, 0.10], 'HR': [0.7, 0.3, 0.0], 'Sales': [0.75, 0.25, 0.0]}
education = np.array([rproj.choice(['Bachelor', 'Master', 'PhD'], p=edu_probs[d]) for d in department])
experience = rproj.gamma(2.2, 3.5, n).clip(0, 35).round(1)
age = (23 + experience + rproj.normal(2, 2.5, n)).round(0)
gender = rproj.choice(['Male', 'Female'], n, p=[0.6, 0.4])

dept_base = pd.Series(department).map({'IT': 110000, 'Finance': 100000, 'HR': 80000, 'Sales': 85000}).to_numpy()
edu_add = pd.Series(education).map({'Bachelor': 0, 'Master': 12000, 'PhD': 30000}).to_numpy()
salary = (dept_base + edu_add + 5000 * experience) * rproj.lognormal(0, 0.08, n)
salary[rproj.choice(n, 2, replace=False)] *= 2.2          # two senior executives → high outliers

employees = pd.DataFrame({
    'Employee_ID': [f'E{101 + i}' for i in range(n)],
    'Age': age, 'Gender': gender, 'Department': department, 'Education': education,
    'Experience': experience, 'Salary': salary.round(-2),
})
employees.head(10)
````

## Step 1 — Central tendency

````python
sal = employees['Salary']
central = pd.Series({'Mean': sal.mean(), 'Median': sal.median(), 'Mode (rounded to 10k)': (sal.round(-4)).mode()[0]})
print(central.round(0))
````

## Step 2 — Dispersion

````python
q1, q3 = np.percentile(sal, [25, 75])
dispersion = pd.Series({
    'Range': sal.max() - sal.min(),
    'Variance (sample)': sal.var(),
    'Std (sample)': sal.std(),
    'Coefficient of variation': sal.std() / sal.mean(),       # std relative to the mean
    'Q1': q1, 'Q3': q3, 'IQR': q3 - q1,
    'MAD (median abs. deviation)': stats.median_abs_deviation(sal),
})
dispersion.map('{:,.2f}'.format)      # readable formatting instead of scientific notation
````

## Step 3 — Distribution, skewness and outliers

````python
fig, axes = plt.subplots(1, 3, figsize=(17, 4))
sns.histplot(sal, bins=20, kde=True, ax=axes[0])
axes[0].axvline(sal.mean(), color='red', label='Mean'); axes[0].axvline(sal.median(), color='green', ls='--', label='Median')
axes[0].legend(); axes[0].set_title(f'Salary distribution (skew {sal.skew():.2f}, kurtosis {sal.kurt():.2f})')
sns.boxplot(x=sal, ax=axes[1]); axes[1].set_title('Boxplot')
stats.probplot(sal, dist='norm', plot=axes[2]); axes[2].set_title('Q-Q plot vs normal')
plt.tight_layout(); plt.show()

low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
outliers = employees[(sal < low) | (sal > high)]
print(f'IQR fences: ({low:,.0f}, {high:,.0f})')
print(f'Shapiro-Wilk p = {stats.shapiro(sal).pvalue:.5f}')
outliers[['Employee_ID', 'Department', 'Education', 'Experience', 'Salary']]
````

## Step 4 — Correlation

````python
num = employees[['Age', 'Experience', 'Salary']]
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
sns.heatmap(num.corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, ax=axes[0])
axes[0].set_title('Pearson correlation')
sns.scatterplot(data=employees, x='Experience', y='Salary', hue='Department', ax=axes[1])
axes[1].set_title('Experience vs salary'); plt.tight_layout(); plt.show()

print('Pearson :\n', num.corr().round(3), '\n')
print('Spearman:\n', num.corr(method='spearman').round(3))
r, p = stats.pearsonr(employees['Experience'], employees['Salary'])
print(f'\nExperience–Salary: r = {r:.3f}, p = {p:.2g}')
````

## Step 5 — Confidence intervals

````python
def mean_ci(x, level=0.95):
    return stats.t.interval(level, df=len(x) - 1, loc=np.mean(x), scale=stats.sem(x))

lo, hi = mean_ci(sal)
print(f'Average salary: {sal.mean():,.0f}  (95% CI {lo:,.0f} – {hi:,.0f})')

# Bootstrap CI for the MEDIAN (no formula needed: resample with replacement many times)
boot_medians = [np.median(rproj.choice(sal, len(sal), replace=True)) for _ in range(5000)]
b_lo, b_hi = np.percentile(boot_medians, [2.5, 97.5])
print(f'Median salary : {sal.median():,.0f}  (95% bootstrap CI {b_lo:,.0f} – {b_hi:,.0f})')

dept_ci = employees.groupby('Department')['Salary'].apply(
    lambda x: pd.Series({'mean': x.mean(), 'ci_low': mean_ci(x)[0], 'ci_high': mean_ci(x)[1], 'n': len(x)})).unstack()
dept_ci.round(0)
````

(**Bootstrap** = resample your own data with replacement thousands of times to see how much a statistic varies — useful for medians and other statistics without a simple formula.)

## Step 6 — Hypothesis test: do men and women earn different averages?
- H0: μ_male = μ_female
- H1: μ_male ≠ μ_female

````python
male = employees.loc[employees['Gender'] == 'Male', 'Salary']
female = employees.loc[employees['Gender'] == 'Female', 'Salary']

welch = stats.ttest_ind(male, female, equal_var=False)
mw = stats.mannwhitneyu(male, female)
print(f'Mean male = {male.mean():,.0f} (n={len(male)}), mean female = {female.mean():,.0f} (n={len(female)})')
print(f'Welch t-test : t = {welch.statistic:.2f}, p = {welch.pvalue:.4f}')
print(f'Mann-Whitney : p = {mw.pvalue:.4f}  (robust check, because salary is skewed)')
print('95% CI for the difference (male − female):', np.round(welch.confidence_interval(0.95), 0))
````

## Step 7 — Department differences: ANOVA + post-hoc

````python
groups = {d: g['Salary'].to_numpy() for d, g in employees.groupby('Department')}
print('Levene (equal variances) p =', round(stats.levene(*groups.values()).pvalue, 4))

f, p = stats.f_oneway(*groups.values())
k = stats.kruskal(*groups.values())
print(f'One-way ANOVA : F = {f:.2f}, p = {p:.2g}')
print(f'Kruskal-Wallis: H = {k.statistic:.2f}, p = {k.pvalue:.2g}  (non-parametric check)')

names = list(groups)
tk = stats.tukey_hsd(*groups.values())
pairs = [{'Pair': f'{names[i]} vs {names[j]}', 'Mean diff': groups[names[i]].mean() - groups[names[j]].mean(),
          'Adjusted p': tk.pvalue[i, j]} for i in range(len(names)) for j in range(i + 1, len(names))]
print()
print(pd.DataFrame(pairs).round(4).to_string(index=False))

plt.figure(figsize=(8, 4))
sns.boxplot(data=employees, x='Department', y='Salary', order=['IT', 'Finance', 'Sales', 'HR'])
plt.title(f'Salary by department (ANOVA p = {p:.2g})'); plt.show()
````

## Step 8 — Categorical relationship: department vs education (chi-square)

````python
ct = pd.crosstab(employees['Department'], employees['Education'])
chi2, p_chi, dof, expected = stats.chi2_contingency(ct)
v = np.sqrt(chi2 / (ct.to_numpy().sum() * (min(ct.shape) - 1)))
print(ct, '\n')
print(f'χ² = {chi2:.2f}, df = {dof}, p = {p_chi:.4f}, Cramér\'s V = {v:.2f}')
print('Cells with expected count < 5:', int((expected < 5).sum()), 'of', expected.size)
````

## Step 9 — Final statistical interpretation (in simple English)

1. **Typical salary.** The mean (≈ 146k) is higher than the median (≈ 142k) and the skewness is clearly positive (≈ 1.7) → salaries are **right-skewed**. The **median** is the better "typical salary". Three very high earners (≈ 240k–300k) are flagged as **outliers** by the 1.5×IQR rule; two of them are senior executives. They are real people, so we keep them — but we report the median and IQR alongside the mean and std.

2. **Spread.** The std is ≈ 37k (coefficient of variation ≈ 0.26 → salaries typically differ from the mean by about a quarter of it). The middle 50% of employees earn between ≈ 122k and 162k (IQR ≈ 40k).

3. **Distribution.** Positive skew, high kurtosis (heavy right tail), a Q-Q plot that curves away at the top, and a Shapiro-Wilk p < 0.001 → salaries are **not normal**. With n = 80 the CLT still makes tests on **means** reasonable, but we back every test up with a **non-parametric** version.

4. **Relationships.** Experience is **moderately positively correlated** with salary (Pearson r ≈ 0.45, p < 0.001; Spearman ≈ 0.62 — higher because Spearman is less affected by the executive outliers). Age is also correlated with salary, but mostly because age and experience are strongly related (r ≈ 0.87). Correlation is not causation, and department and education also drive salary.

5. **Average salary.** We are 95% confident that the company's true average salary is between ≈ 137k and 154k. The bootstrap 95% CI for the **median** is ≈ 135k – 150k. (These CIs describe how reliable the *method* is; see Part 14.)

6. **Gender.** Women's average salary in this sample is slightly higher, but the Welch t-test (p ≈ 0.27) and Mann-Whitney test (p ≈ 0.82) show **no evidence** of a difference. The 95% CI for the difference (≈ −26k to +8k) is **wide** and includes 0 → a real difference cannot be ruled out, and it could go either way. A serious pay-equity study would need more data and should control for department, education and experience.

7. **Departments.** ANOVA (F ≈ 10.5, p < 0.001) and Kruskal-Wallis (p < 0.001) agree: average salaries are **not equal** across departments. Tukey's HSD shows **IT pays significantly more than HR and Sales**, and **Finance pays more than Sales**. The other pairs are not significant — partly because HR has only 6 employees, so its mean is very uncertain (its CI is very wide). Levene's test (p ≈ 0.96) supports the equal-variance assumption.

8. **Department vs education.** The chi-square test (p ≈ 0.001, Cramér's V ≈ 0.37) suggests a **moderate association**: IT and Finance employ more Master's/PhD holders, while Sales and HR are mostly Bachelor's. **But** 6 of 12 cells have expected counts below 5, so this result is only **approximate** — merging Master + PhD into one "Postgraduate" group would make it more reliable.

### Overall conclusion
Salaries in this company depend mainly on **department**, **education** and **experience**. They are right-skewed, with a few very high earners, so **medians, IQRs and robust tests** matter. The data gives **no evidence of a gender pay difference**, but the sample is too small to rule one out.

# Part 26 — Statistics Cheat Sheet

| Concept | Meaning | When to use | Python |
|---|---|---|---|
| **Mean** | Sum ÷ count | Symmetric data, totals | `df['x'].mean()`, `np.mean(x)` |
| **Median** | Middle value | Skewed data, outliers, ordinal | `df['x'].median()`, `np.median(x)` |
| **Mode** | Most frequent value | Categorical data, peaks | `df['x'].mode()`, `stats.mode(x)` |
| **Variance** | Average squared distance from the mean | Maths behind std, ANOVA | `s.var()` (ddof=1), `np.var(x, ddof=1)` |
| **Standard deviation** | Typical distance from the mean (original units) | Spread of symmetric data | `s.std()`, `np.std(x, ddof=1)` |
| **IQR** | Q3 − Q1 (middle 50%) | Spread of skewed data, outlier fences | `np.percentile(x, 75) - np.percentile(x, 25)`, `stats.iqr(x)` |
| **Percentile** | Value below which p% of data falls | Thresholds, rankings, quantile bins | `np.percentile(x, p)`, `s.quantile(p/100)` |
| **Z-score** | (x − mean) / std: distance in std units | Compare scales, outliers, standardizing | `stats.zscore(x)`, `StandardScaler` |
| **Correlation** | Strength/direction of a relationship (−1..1) | EDA, feature selection | `df.corr()`, `stats.pearsonr`, `stats.spearmanr` |
| **Sampling** | Choosing part of the population | Surveys, experiments, train/test split | `df.sample()`, `groupby().sample()`, `train_test_split(stratify=)` |
| **Standard error** | Std of a statistic across samples: s/√n | Precision of an estimate | `stats.sem(x)` |
| **CLT** | Sample means are ≈ normal for large n | Justifies CIs & tests for means | (simulation with `np.random`) |
| **Confidence interval** | Range from a method that captures the true value X% of the time | Report any estimate with its uncertainty | `stats.t.interval(0.95, df, loc, scale)`, `res.confidence_interval()` |
| **p-value** | P(data this extreme **given H0 true**) | Evidence against H0 | returned by every test (`.pvalue`) |
| **Hypothesis testing** | H0 vs H1, reject if p ≤ α | Decisions from data, A/B tests | `stats.ttest_*`, `stats.chi2_contingency`, `stats.f_oneway` |
| **Type I error (α)** | False positive: reject a true H0 | Set α before testing | α = 0.05 (convention) |
| **Type II error (β)** | False negative: miss a real effect | Power planning (power = 1 − β) | simulation / power analysis |
| **t-test** | Compare means (1 sample, 2 independent, paired) | Numerical outcome, 1–2 groups | `stats.ttest_1samp`, `stats.ttest_ind(equal_var=False)`, `stats.ttest_rel` |
| **Chi-square** | Are two categorical variables independent? | Contingency tables | `stats.chi2_contingency(pd.crosstab(a, b))` |
| **ANOVA** | Compare means of 3+ groups | Numerical outcome, 3+ groups | `stats.f_oneway`, post-hoc `stats.tukey_hsd` |
| *Non-parametric* | Rank-based versions | Skewed/ordinal data, outliers | `stats.mannwhitneyu`, `stats.wilcoxon`, `stats.kruskal` |
| *Assumption checks* | Normality, equal variance | Before parametric tests | `stats.shapiro`, `stats.probplot`, `stats.levene` |

### Test chooser

```text
Numerical outcome:  1 group vs value → one-sample t
                    2 independent   → Welch t        | skewed/ordinal → Mann-Whitney U
                    2 paired        → paired t       | skewed/ordinal → Wilcoxon
                    3+ groups       → ANOVA + Tukey  | skewed/ordinal → Kruskal-Wallis
Categorical vs categorical         → Chi-square (+ Cramér's V)
Two numerical variables            → Pearson (linear) / Spearman (monotonic)
```

### Golden rules
1. **Plot first**, then compute.
2. Use the **median/IQR** for skewed data.
3. **Std ≠ standard error.** SE shrinks with √n.
4. A 95% CI describes the **method's** success rate, not a probability about one interval.
5. The p-value assumes H0 is true; it is **not** the probability that H0 is true.
6. **Fail to reject ≠ accept.** Small samples miss real effects (low power).
7. Report **effect size + CI**, not only p.
8. Check **assumptions**; use Welch or non-parametric tests when in doubt.
9. Many tests → expect false positives (**multiple testing**).
10. **Correlation ≠ causation**; bias can't be fixed by a bigger sample.
