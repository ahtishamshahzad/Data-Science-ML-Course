````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')              # keep the learning output clean
sns.set_theme(style='whitegrid')               # a clean default look for all plots
pd.set_option('display.width', 120)
rng = np.random.default_rng(42)                # fixed seed → the same data every time you run
print('matplotlib', plt.matplotlib.__version__, '| seaborn', sns.__version__, '| pandas', pd.__version__)
````

# 1. Introduction to Data Visualization

### What is data visualization?
Turning numbers into **pictures** (charts) so that patterns, comparisons, trends and problems become visible quickly.

### Why does it matter?
- Humans spot **shapes, colours and positions** far faster than they read numbers.
- It reveals things summary statistics hide: **outliers, skew, clusters, curved relationships** (four datasets can share the same mean, std and correlation but look completely different — Anscombe's quartet, Topic 12).
- It **communicates** results to non-technical people.

### Visualization vs EDA vs reporting

| | **Exploratory visualization (EDA)** | **Reporting / explanatory visualization** |
|---|---|---|
| Audience | You (the analyst) | Managers, clients, the public |
| Goal | **Discover** — ask questions, find problems | **Explain** one clear message |
| Number of charts | Many, quick and rough | Few, polished |
| Style | Speed over beauty | Clear titles, labels, annotations, consistent colours |

EDA (Topic 7) is the **process** of investigating data; visualization is one of its main **tools**. Reporting uses visualization to **tell a story** that has already been found.

### Choosing the right chart — the key question: *"What do I want to show?"*

| Goal | Chart |
|---|---|
| Compare categories | Bar chart |
| Show a trend over time | Line chart |
| Show a distribution | Histogram, KDE, boxplot, violin, ECDF |
| Show a relationship between two numbers | Scatterplot (+ regression line) |
| Show correlation among many numbers | Heatmap |
| Show parts of a whole | Stacked bar (or pie, only for 2–4 parts) |
(Full decision guide in Section 21.)

### Avoiding misleading visualizations (details in Section 22)
- Bar charts should start at **zero**.
- Label axes with **units**; give every chart a **title**.
- Don't use **3-D** effects or **too many colours**.
- Don't imply **causation** from a correlation.
- Choose bins and scales honestly.

# 2. Matplotlib Fundamentals

```python
import matplotlib.pyplot as plt
```
Matplotlib is the foundation of plotting in Python; Seaborn and Pandas plotting are built on top of it.

### The anatomy of a chart
```text
┌────────────────── Figure (the whole window / image) ──────────────────┐
│   ┌──────────────── Axes (one plot area) ──────────────────────┐      │
│   │  Title                                                     │      │
│   │  Y-axis (label, ticks)   ── lines / bars / points ──       │      │
│   │                          Legend      Grid                  │      │
│   │  X-axis (label, ticks)                                     │      │
│   └────────────────────────────────────────────────────────────┘      │
└───────────────────────────────────────────────────────────────────────┘
```

| Term | Meaning |
|---|---|
| **Figure** | The whole canvas; can hold one or many plots |
| **Axes** | One plot area with its own x-axis and y-axis (⚠️ not the same as "axis") |
| **Title** | What the chart is about |
| **X-axis / Y-axis** | Horizontal / vertical scale, with **labels** (what + units) and **ticks** |
| **Legend** | Explains which colour/marker is which series |
| **Grid** | Light helper lines to read values |
| **Figure size** | `figsize=(width, height)` in inches |
| **Subplots** | Several Axes in one Figure |

### Two ways to use Matplotlib
1. **pyplot style** (`plt.plot`, `plt.title` …) — quick, acts on the "current" plot.
2. **Object-oriented style** (`fig, ax = plt.subplots()`, `ax.plot`, `ax.set_title` …) — explicit; better for several subplots. **Recommended** for anything beyond one quick chart.

````python
months = np.arange(1, 13)
sales_2024 = np.array([120, 135, 150, 160, 158, 170, 185, 190, 176, 200, 230, 260])
sales_2025 = np.array([130, 142, 165, 172, 180, 188, 196, 210, 205, 225, 250, 290])

plt.figure(figsize=(9, 4.5))                                  # 1. create a figure (9 × 4.5 inches)
plt.plot(months, sales_2024, marker='o', label='2024')        # 2. draw a line with circle markers
plt.plot(months, sales_2025, marker='s', label='2025')        #    second line → will get a different colour
plt.title('Monthly Sales (thousand PKR)')                     # 3. title
plt.xlabel('Month')                                           # 4. x-axis label
plt.ylabel('Sales (thousand PKR)')                            # 5. y-axis label (include units!)
plt.xticks(months)                                            #    show every month on the x-axis
plt.legend()                                                  # 6. legend uses the label= of each line
plt.grid(True, alpha=0.3)                                     # 7. light grid
plt.savefig('monthly_sales.png', dpi=150, bbox_inches='tight')   # 8. save BEFORE plt.show()
plt.show()                                                    # 9. display
````

### Expected output & interpretation
Two rising lines; 2025 sits above 2024 in every month → sales grew year over year, with the biggest jump in Nov–Dec (seasonal peak). A file `monthly_sales.png` is saved next to the notebook.

### Saving figures — key options
- `dpi=150` (or 300 for print) → resolution.
- `bbox_inches='tight'` → trims extra white space, keeps labels from being cut off.
- Formats by extension: `.png` (web/slides), `.svg`/`.pdf` (sharp vector graphics).
- ⚠️ Call `savefig` **before** `plt.show()` — after `show()` the figure may already be closed and you save an empty image.

````python
# Object-oriented style + subplots: 1 row × 2 columns
fig, axes = plt.subplots(1, 2, figsize=(12, 4))          # fig = the whole figure, axes = array of 2 Axes

axes[0].plot(months, sales_2024, color='tab:blue', marker='o')
axes[0].set_title('2024'); axes[0].set_xlabel('Month'); axes[0].set_ylabel('Sales')

axes[1].bar(months, sales_2025 - sales_2024, color='tab:green')
axes[1].set_title('Growth vs previous year'); axes[1].set_xlabel('Month'); axes[1].set_ylabel('Extra sales')

fig.suptitle('Object-oriented Matplotlib: one Figure, two Axes', fontsize=13)
plt.tight_layout()                                         # stop titles/labels from overlapping
plt.show()
````

### Common mistakes
- Forgetting axis labels/units or a title.
- Calling `savefig` after `show` (empty file).
- Mixing pyplot and object-oriented calls in a confusing way (e.g. `plt.title` when you have several Axes → it only affects the last one).
- Forgetting `plt.tight_layout()` → overlapping text in subplots.

# 3. Line Charts

### What?
Points connected by lines, usually with **time** on the x-axis.

### Why?
The connecting line emphasises **change and continuity** → perfect for **trends** (rising, falling, seasonal patterns, sudden jumps).

### When?
Time series (daily visits, monthly sales, yearly temperature); comparing a few series over the same time period.

### When not?
- **Unordered categories** (cities, products): a line between "Lahore" and "Karachi" suggests a continuous change that doesn't exist → use a **bar chart**.
- Too many lines (> ~5–6) → "spaghetti chart"; use small multiples (Section 20) or highlight one line.

````python
dates = pd.date_range('2025-01-01', periods=120, freq='D')
visits = pd.DataFrame({
    'Date': dates,
    'Website': (1000 + np.arange(120) * 4 + 150 * np.sin(np.arange(120) * 2 * np.pi / 7) + rng.normal(0, 60, 120)).round(),
    'App': (800 + np.arange(120) * 7 + rng.normal(0, 60, 120)).round(),
})

fig, ax = plt.subplots(figsize=(11, 4.5))
ax.plot(visits['Date'], visits['Website'], label='Website', linewidth=1.5)
ax.plot(visits['Date'], visits['App'], label='App', linewidth=1.5)
ax.set_title('Daily visits: website vs app (Jan–Apr 2025)')
ax.set_xlabel('Date'); ax.set_ylabel('Visits per day')
ax.legend(); plt.show()
````

### Interpretation
- Both channels grow; the **app grows faster** and overtakes the website in the second half of the period (from roughly day 60–70).
- The website line has a regular **weekly wave** (7-day seasonality) — something a table would hide.

### Common mistakes
Unsorted x-values (lines zig-zag back and forth — sort by date first) · a line chart for categories · a y-axis range that exaggerates tiny changes without saying so.

````python
# ❌ When NOT to use a line chart: categories have no order
city_sales = pd.Series({'Lahore': 420, 'Karachi': 510, 'Islamabad': 300, 'Multan': 180})
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
axes[0].plot(city_sales.index, city_sales.values, marker='o', color='tab:red')
axes[0].set_title('❌ Line chart suggests a "trend" between cities')
axes[1].bar(city_sales.index, city_sales.values)
axes[1].set_title('✅ Bar chart compares categories')
plt.tight_layout(); plt.show()
````

# 4. Bar Charts

### What?
Rectangles whose **length** represents a value for each **category**.

| Type | Use when |
|---|---|
| **Vertical bar** | Few categories with short names |
| **Horizontal bar** | Many categories or **long names**; ranking (sorted) |
| **Grouped bar** | Compare **sub-groups** side by side (e.g. each city, male vs female) |
| **Stacked bar** | Show the **total** and how it's **composed** of parts |

### Why?
Humans compare **lengths from a common baseline** very accurately → bar charts are the best tool for comparing categories.

### When not?
Continuous data (use a histogram); time trends with many points (use a line); when the axis doesn't start at 0 (bars would lie — Section 22).

````python
dept_salary = employees.groupby('Department')['Salary'].mean().sort_values(ascending=False)
dept_gender = employees.groupby(['Department', 'Gender'])['Salary'].mean().unstack()     # rows = dept, cols = gender
dept_city = pd.crosstab(employees['Department'], employees['City'])                      # headcount per dept & city

fig, axes = plt.subplots(2, 2, figsize=(14, 9))

axes[0, 0].bar(dept_salary.index, dept_salary.values, color='steelblue')
axes[0, 0].set_title('Vertical bar: average salary by department'); axes[0, 0].set_ylabel('Salary (PKR)')

dept_salary.sort_values().plot(kind='barh', ax=axes[0, 1], color='tab:orange')         # pandas plotting
axes[0, 1].set_title('Horizontal bar (sorted): easier ranking'); axes[0, 1].set_xlabel('Salary (PKR)')

dept_gender.plot(kind='bar', ax=axes[1, 0], rot=0)                                       # one bar per gender per dept
axes[1, 0].set_title('Grouped bar: salary by department & gender'); axes[1, 0].set_ylabel('Salary (PKR)')

dept_city.plot(kind='bar', stacked=True, ax=axes[1, 1], rot=0)
axes[1, 1].set_title('Stacked bar: headcount by department, split by city'); axes[1, 1].set_ylabel('Employees')
plt.tight_layout(); plt.show()
````

### Interpretation
- IT and Finance pay the most on average; HR the least.
- Grouped bars show male and female averages are close in IT and Finance; in HR and Sales there is a gap of about 10,000. With only ~20–50 people per group, check whether such a gap is meaningful (Topic 14) before drawing conclusions.
- The stacked bar shows **Sales and IT are the largest departments**. Lahore contributes the biggest share in Sales and Finance, Karachi in HR (and IT is split almost equally between Lahore and Karachi).

### Common mistakes
- Not starting bars at 0.
- Unsorted bars when ranking matters.
- Stacked bars with many segments: only the bottom segment is easy to compare (the others float) → use grouped bars or small multiples.
- Using a bar chart of **means** without showing spread (Seaborn's `barplot` adds error bars — Section 18).

# 5. Histograms

### What?
A histogram splits a numeric variable into **bins** (intervals) and draws a bar showing **how many values** fall into each bin → it shows the **distribution**.

```python
plt.hist(data, bins=20)          # Matplotlib
sns.histplot(data, bins=20)      # Seaborn (can add kde=True, hue=...)
```

### Key ideas
| Term | Meaning |
|---|---|
| **Bins** | The intervals; their number/width changes the picture |
| **Distribution** | Where values are concentrated and how they spread |
| **Normal distribution** | Symmetric bell shape (heights, measurement errors) |
| **Skewness** | Lopsided: **right-skewed** = long tail to the right (income); left-skewed = long tail to the left |
| **Outliers** | Isolated bars far from the rest |

### When? / When not?
- ✅ One numeric variable; checking shape, skew, outliers before modelling.
- ❌ Categories (use a countplot/bar chart); comparing many groups on one histogram (use KDE, boxplots or facets); very small samples (bins are unreliable).

````python
normal_data = rng.normal(170, 10, 1000)                          # heights → bell-shaped
right_skewed = rng.lognormal(10.8, 0.6, 1000)                    # incomes → long right tail
with_outliers = np.concatenate([rng.normal(50, 5, 500), [95, 98, 102]])

fig, axes = plt.subplots(1, 3, figsize=(17, 4))
axes[0].hist(normal_data, bins=30, color='steelblue', edgecolor='white')
axes[0].set_title('Normal (symmetric)'); axes[0].set_xlabel('Height (cm)'); axes[0].set_ylabel('Count')
sns.histplot(right_skewed, bins=40, kde=True, ax=axes[1], color='tab:orange')
axes[1].set_title(f'Right-skewed (skew = {pd.Series(right_skewed).skew():.2f})'); axes[1].set_xlabel('Income')
sns.histplot(with_outliers, bins=40, ax=axes[2], color='tab:green')
axes[2].set_title('Outliers appear as isolated bars'); axes[2].set_xlabel('Delivery time (min)')
plt.tight_layout(); plt.show()
````

````python
# How the number of bins changes the story
fig, axes = plt.subplots(1, 4, figsize=(19, 3.8))
for ax, b in zip(axes, [3, 10, 30, 200]):
    ax.hist(employees['Salary'], bins=b, color='slateblue', edgecolor='white')
    ax.set_title(f'bins = {b}'); ax.set_xlabel('Salary')
plt.tight_layout(); plt.show()
````

### Interpretation of the bins experiment
- **3 bins:** far too coarse — the shape is hidden.
- **10–30 bins:** the shape is clear.
- **200 bins:** too many — random noise creates spiky gaps.

Rules of thumb: start with Seaborn's automatic bins (`bins='auto'`), then try a few values. Always compare groups using the **same bins**.

### Common mistakes
Too few/too many bins · different bin widths when comparing groups · reading a single histogram of mixed groups (two humps may mean two groups — use `hue=`).

# 6. KDE Plots (Kernel Density Estimation)

### What? (simple language)
Imagine placing a small smooth **"hill"** on top of every data point and adding all the hills together. The result is a **smooth curve** estimating the distribution. The area under the curve is 1.

```python
sns.kdeplot(data=df, x='Salary', hue='Department', fill=True)
```

### KDE vs histogram
| | Histogram | KDE |
|---|---|---|
| Look | Bars (blocky) | Smooth curve |
| Depends on | Number/width of **bins** | **Bandwidth** (width of each hill) |
| Comparing several groups | Messy (overlapping bars) | Clean (overlapping curves) |
| Shows actual counts | ✅ | ❌ (density) |

### Bandwidth
- Small bandwidth → narrow hills → **wiggly** curve (follows noise).
- Large bandwidth → wide hills → **over-smoothed** curve (can hide two humps).
In Seaborn: `bw_adjust=0.5` (less smoothing) … `bw_adjust=2` (more smoothing).

### When can KDE be misleading?
- **Bounded data**: KDE can show density **below 0** for prices or ages, or beyond 100% — values that are impossible. Use `cut=0` or `clip=(0, None)`.
- **Small samples**: a smooth curve looks more certain than the data allows.
- **Discrete data** (counts 0, 1, 2 …): the curve invents values in between.
- Wrong bandwidth hides or invents peaks.

````python
fig, axes = plt.subplots(1, 3, figsize=(18, 4))

sns.histplot(employees['Salary'], bins=25, stat='density', ax=axes[0], color='lightgray')
sns.kdeplot(employees['Salary'], ax=axes[0], color='red', linewidth=2)
axes[0].set_title('Histogram (density) with KDE on top')

for bw in [0.3, 1, 3]:
    sns.kdeplot(employees['Salary'], bw_adjust=bw, ax=axes[1], label=f'bw_adjust={bw}')
axes[1].set_title('Bandwidth: small = wiggly, large = over-smooth'); axes[1].legend()

sns.kdeplot(data=employees, x='Salary', hue='Department', fill=True, common_norm=False, alpha=0.3, ax=axes[2])
axes[2].set_title('Comparing groups with KDE (common_norm=False)')
plt.tight_layout(); plt.show()
````

````python
# KDE misleading example: waiting times can't be negative
waits = rng.exponential(3, 200)
fig, axes = plt.subplots(1, 2, figsize=(12, 3.5))
sns.kdeplot(waits, ax=axes[0], fill=True); axes[0].axvline(0, color='red', ls='--')
axes[0].set_title('❌ Default KDE shows density below 0 minutes')
sns.kdeplot(waits, ax=axes[1], fill=True, clip=(0, None)); axes[1].set_title('✅ clip=(0, None) respects the boundary')
plt.tight_layout(); plt.show()
````

### Interpretation
IT and Finance salary curves sit to the right of HR and Sales → higher typical salaries. `common_norm=False` makes each department's curve integrate to 1, so small and large departments can be compared fairly.

### Common mistakes
Believing the curve outside the data range · comparing groups with `common_norm=True` (default) when group sizes differ · using KDE for small or discrete data.

# 7. Boxplots

### What?
A compact summary of a distribution using **five numbers** plus outliers.

```text
        outlier        whisker          box            whisker
           •     |───────────────[   Q1  │ median │  Q3   ]───────────────|
                 min within fence                          max within fence
```

| Part | Meaning |
|---|---|
| **Median** (line in the box) | Middle value (50th percentile) |
| **Q1** (box bottom/left) | 25th percentile |
| **Q3** (box top/right) | 75th percentile |
| **IQR** = Q3 − Q1 | Spread of the middle 50% (box length) |
| **Whiskers** | Extend to the most extreme points within **1.5 × IQR** of the box |
| **Outliers** | Points beyond the whiskers, drawn individually |

### Numerical example
Data: `12, 15, 16, 18, 19, 20, 22, 25, 60`
- Median = **19**, Q1 = **16**, Q3 = **22** (NumPy's default method), IQR = **6**
- Fences: 16 − 9 = **7** and 22 + 9 = **31**
- 60 > 31 → **outlier**; the upper whisker ends at 25 (largest value ≤ 31).

### When? / When not?
- ✅ Comparing distributions of **several groups** side by side; spotting outliers and skew quickly.
- ❌ Bimodal data (a box hides two humps → use violin/KDE); very small samples (quartiles unstable); audiences unfamiliar with boxplots (explain them!).

````python
vals = np.array([12, 15, 16, 18, 19, 20, 22, 25, 60])
q1, med, q3 = np.percentile(vals, [25, 50, 75])
iqr = q3 - q1
print(f'Q1 = {q1}, median = {med}, Q3 = {q3}, IQR = {iqr}, fences = [{q1 - 1.5 * iqr}, {q3 + 1.5 * iqr}]')
print('Outliers:', vals[(vals < q1 - 1.5 * iqr) | (vals > q3 + 1.5 * iqr)])

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.boxplot(x=vals, ax=axes[0], color='lightblue')
for v, lab in [(q1, 'Q1'), (med, 'Median'), (q3, 'Q3')]:
    axes[0].annotate(lab, (v, -0.45), ha='center')
axes[0].set_title('The numerical example: 60 is an outlier')

sns.boxplot(data=employees, x='Department', y='Salary', order=['IT', 'Finance', 'Sales', 'HR'], ax=axes[1])
axes[1].set_title('Salary by department')
plt.tight_layout(); plt.show()
````

### Interpretation
- The medians confirm IT > Finance > Sales > HR.
- Box heights (IQR) are similar → similar spread within departments.
- Dots beyond the whiskers are unusually high/low salaries to investigate (not automatically errors).

### Common mistakes
Thinking whiskers show min/max (they stop at 1.5 × IQR) · deleting every "outlier" (in skewed data many are normal) · hiding sample size (a box from 5 people looks as solid as one from 5,000 — add a stripplot or counts).

# 8. Violin Plots

### What?
A **boxplot + mirrored KDE**: the width at each height shows how many values are there.

```python
sns.violinplot(data=df, x='Department', y='Salary', inner='quartile')
```

### Boxplot vs violin
| | Boxplot | Violin |
|---|---|---|
| Shows | 5-number summary + outliers | Whole distribution **shape** (+ quartiles inside) |
| Reveals two humps (bimodal)? | ❌ No | ✅ Yes |
| Simplicity | Very simple | Needs a bit more explanation |
| Small samples | OK | KDE can mislead |

### When useful?
Comparing groups whose **shapes** differ (bimodal, skewed). Avoid for small samples or non-technical audiences who haven't seen violins before.

````python
# Bimodal example: delivery times of two couriers mixed together in one city
bimodal = pd.DataFrame({
    'City': ['Lahore'] * 400 + ['Karachi'] * 400,
    'Delivery_min': np.concatenate([rng.normal(30, 4, 400),                                        # one courier
                                    np.concatenate([rng.normal(20, 3, 200), rng.normal(45, 3, 200)])]),  # two couriers
})
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
sns.boxplot(data=bimodal, x='City', y='Delivery_min', ax=axes[0]); axes[0].set_title('Boxplot: both cities look "normal"')
sns.violinplot(data=bimodal, x='City', y='Delivery_min', inner='quartile', ax=axes[1]); axes[1].set_title('Violin: Karachi has TWO groups')
plt.tight_layout(); plt.show()
````

### Interpretation
The boxplots suggest Karachi simply has a wider spread. The violin shows **two separate peaks** (≈ 20 and ≈ 45 minutes) → probably two different couriers or areas. That's an insight a boxplot hides.

### Common mistakes
Over-interpreting the smooth tails · violins for tiny groups · forgetting `inner='quartile'`/`'box'`, which gives the reader reference points.

# 9. Scatterplots

### What?
One **dot per row**, with one numeric variable on x and another on y.

```python
plt.scatter(df['Age'], df['Salary'])
sns.scatterplot(data=df, x='Age', y='Salary')
```

### What can you see?
| Pattern | Meaning |
|---|---|
| Dots rise left → right | **Positive** relationship |
| Dots fall left → right | **Negative** relationship |
| Round cloud | **No** (linear) relationship |
| Separate groups of dots | **Clusters** (sub-populations) |
| Isolated dots far away | **Outliers** |
| Curved band | **Non-linear** relationship |

### When? / When not?
- ✅ Two numeric variables; checking relationships before modelling.
- ❌ Huge datasets → **overplotting** (a solid blob) → use `alpha`, smaller markers, sampling, `hexbin` or 2-D KDE; categorical x-values (use strip/box plots).

````python
x = rng.uniform(0, 10, 150)
patterns = {
    'Positive': 2 * x + rng.normal(0, 2, 150),
    'Negative': -2 * x + rng.normal(0, 2, 150),
    'No relationship': rng.normal(10, 5, 150),
}
fig, axes = plt.subplots(1, 5, figsize=(24, 4))
for ax, (name, y) in zip(axes, patterns.items()):
    ax.scatter(x, y, alpha=0.6, s=18)
    ax.set_title(f'{name} (r = {np.corrcoef(x, y)[0, 1]:.2f})')

clusters = np.vstack([rng.normal([2, 2], 0.5, (60, 2)), rng.normal([7, 7], 0.6, (60, 2)), rng.normal([2, 8], 0.5, (60, 2))])
axes[3].scatter(clusters[:, 0], clusters[:, 1], s=18, alpha=0.7); axes[3].set_title('Clusters')

ox, oy = rng.normal(5, 1, 100), rng.normal(5, 1, 100)
axes[4].scatter(ox, oy, s=18, alpha=0.7); axes[4].scatter([10, 0.5], [1, 11], color='red', s=60)
axes[4].set_title('Outliers (red)')
plt.tight_layout(); plt.show()
````

````python
# Overplotting and how to fix it
big_x = rng.normal(0, 1, 20000); big_y = big_x + rng.normal(0, 1, 20000)
fig, axes = plt.subplots(1, 3, figsize=(17, 4))
axes[0].scatter(big_x, big_y, s=10); axes[0].set_title('❌ 20,000 points: a solid blob')
axes[1].scatter(big_x, big_y, s=2, alpha=0.1); axes[1].set_title('✅ small + transparent (alpha=0.1)')
hb = axes[2].hexbin(big_x, big_y, gridsize=40, cmap='viridis'); plt.colorbar(hb, ax=axes[2], label='count')
axes[2].set_title('✅ hexbin: colour = number of points')
plt.tight_layout(); plt.show()
````

### Common mistakes
Overplotting · forgetting that correlation ≠ causation · using a scatterplot when one variable is categorical · unlabeled axes.

# 10. Hue, Size and Style (Seaborn)

### What?
Seaborn can encode **extra variables** in one scatterplot by changing the **colour**, **size** and **marker shape** of the dots.

```python
sns.scatterplot(
    data=df,          # the DataFrame
    x="Age",          # column for the x-axis
    y="Salary",       # column for the y-axis
    hue="Gender",     # colour of each dot depends on Gender
    size="Experience",# dot size depends on Experience (bigger = more years)
    style="Department"# marker shape depends on Department (circle, cross, square …)
)
```

| Parameter | Encodes | Best for |
|---|---|---|
| `hue` | **Colour** | A categorical variable with few levels (or a numeric one → colour gradient) |
| `size` | **Marker size** | A numeric variable (bigger = more) |
| `style` | **Marker shape** | A categorical variable with few levels (≤ 4–5) |

### When? / When not?
- ✅ When a third variable might explain groups in the scatter (e.g. do men and women follow the same age–salary trend?).
- ❌ Using all three at once on a busy dataset → hard to read. Add one encoding at a time; prefer **facets** (Section 20) when the chart gets crowded.

````python
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))
sns.scatterplot(data=employees, x='Age', y='Salary', hue='Department', ax=axes[0], alpha=0.8)
axes[0].set_title('hue only: colour = department (clear)')

sns.scatterplot(data=employees, x='Age', y='Salary', hue='Gender', size='Experience', style='Department',
                sizes=(15, 150), alpha=0.7, ax=axes[1])
axes[1].set_title('hue + size + style: more information, harder to read')
axes[1].legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=8)   # move the big legend outside
plt.tight_layout(); plt.show()
````

### Interpretation
- Left: salary rises with age for every department, but the IT/Finance dots sit **higher** than the HR/Sales dots at the same age → department shifts the whole line up or down.
- Right: bigger dots (more experience) are on the right (older) → age and experience move together. Gender colours are mixed → no obvious gender gap.

### Common mistakes
Too many encodings · `hue` with dozens of categories (colours become indistinguishable) · a huge legend covering the data (move it outside).

# 12. Heatmaps

### What?
A grid of coloured cells where **colour intensity** represents a value. Most common use: the **correlation matrix**.

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
```

| Element | Meaning |
|---|---|
| Colour intensity | Strength of the value (dark red = strong positive, dark blue = strong negative, white ≈ 0) |
| `annot=True` | Writes the number in each cell |
| `vmin=-1, vmax=1`, `center=0` | Fix the colour scale so 0 is neutral and colours are comparable |
| Diagonal | Always 1 (each variable with itself) |

### Correlation interpretation (Topic 12)
|r| < 0.2 very weak · 0.2–0.4 weak · 0.4–0.6 moderate · 0.6–0.8 strong · > 0.8 very strong. Correlation ≠ causation.

### When not?
Many variables (> ~20) → unreadable annotations; categorical data (correlation needs numbers); as the only check (always look at scatterplots too).

````python
num_cols = ['Age', 'Experience', 'Salary', 'Performance']
corr = employees[num_cols].corr()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, square=True, ax=axes[0])
axes[0].set_title('Correlation heatmap')

mask = np.triu(np.ones_like(corr, dtype=bool))                  # hide the upper triangle (mirror image)
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, square=True, ax=axes[1])
axes[1].set_title('Lower triangle only (no repeated values)')
plt.tight_layout(); plt.show()
````

### Interpretation
- Age ↔ Experience ≈ 0.99: almost the same information (multicollinearity, Topic 12).
- Experience ↔ Salary: strong positive — more experience, higher pay.
- Performance ≈ 0 with everything: unrelated in this data.

A heatmap is also useful for **pivot tables** (e.g. average salary by department × city) — see the final project.

### Common mistakes
No `vmin/vmax` (colours rescale and a weak 0.3 looks dark) · rainbow colour maps for correlations (use a diverging map) · too many decimals.

# 14. Pie Charts

### What?
A circle divided into slices showing **parts of a whole** (percentages that add to 100%).

### When appropriate?
- **Few categories** (2–4) with clearly different sizes.
- The message is simply "**X is about half / a quarter of the total**".

### Why they're hard to compare
Humans judge **angles and areas** poorly. Slices of 23% and 27% look almost identical; comparing slices across two pies is even harder.

### Alternatives
Sorted **bar chart** (best), **100% stacked bar**, or a simple table of percentages.

````python
payment_share = pd.Series({'Card': 34, 'Cash': 31, 'Wallet': 20, 'Bank transfer': 15})
many = pd.Series({'A': 18, 'B': 16, 'C': 15, 'D': 14, 'E': 13, 'F': 12, 'G': 12})

fig, axes = plt.subplots(1, 3, figsize=(18, 4.8))
axes[0].pie(payment_share, labels=payment_share.index, autopct='%1.0f%%', startangle=90)
axes[0].set_title('OK: 4 slices, labelled with %')
axes[1].pie(many, labels=many.index, startangle=90)
axes[1].set_title('❌ 7 similar slices: which is biggest?')
many.sort_values().plot(kind='barh', ax=axes[2], color='steelblue')
axes[2].set_title('✅ Sorted bar chart: ranking is obvious'); axes[2].set_xlabel('Share (%)')
plt.tight_layout(); plt.show()
````

### Common mistakes
Too many slices · no percentages · 3-D or exploded pies (distort sizes) · parts that don't add up to a meaningful whole · comparing two pies side by side.

# 15. Area Charts

### What?
A line chart with the area below the line **filled**. A **stacked area chart** stacks several series so the top edge shows the **total** and each band shows a **component**.

### When?
Time series where both the **total** and the **composition** matter (e.g. revenue by channel over months; energy by source).

### When not?
- Many series (> ~5) or series that cross often → middle bands are hard to read (their baseline moves).
- Comparing exact values of the upper bands → use lines or small multiples.
- Negative values (stacking breaks).

````python
months = pd.date_range('2025-01-01', periods=12, freq='MS')
channels = pd.DataFrame({
    'Online': np.linspace(50, 140, 12) + rng.normal(0, 5, 12),
    'Store': np.linspace(100, 90, 12) + rng.normal(0, 5, 12),
    'Wholesale': np.full(12, 40) + rng.normal(0, 4, 12),
}, index=months).round(1)

fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
axes[0].fill_between(channels.index, channels['Online'], alpha=0.4); axes[0].plot(channels.index, channels['Online'])
axes[0].set_title('Area chart: online revenue'); axes[0].set_ylabel('Revenue (thousand PKR)')
channels.plot(kind='area', stacked=True, ax=axes[1], alpha=0.8)        # pandas stacked area chart
axes[1].set_title('Stacked area: total and composition by channel'); axes[1].set_ylabel('Revenue (thousand PKR)')
plt.tight_layout(); plt.show()
````

### Interpretation
Total revenue grows, and the growth comes almost entirely from **online**; store revenue slowly declines; wholesale is flat.

### Common mistakes
Stacking unrelated series · too many bands · unsorted dates · forgetting that only the bottom band has a flat baseline.

# 16. ECDF (Empirical Cumulative Distribution Function)

### What?
For every value x, the ECDF shows the **share of data ≤ x**. It rises from 0 to 1 as a staircase.

```python
sns.ecdfplot(data=df, x='Salary', hue='Department')
```

### What does it tell us?
- **Percentiles directly:** where the curve crosses 0.5 is the **median**; 0.9 → the 90th percentile.
- "What share of employees earn **below 100,000**?" → read the y-value at x = 100,000.
- Comparing groups: a curve further **right** means larger values.

### Why use it instead of a histogram?
No bins and no bandwidth to choose → no artificial shape; exact and honest, great for **comparing groups**.

### When not?
Audiences unfamiliar with cumulative plots; when you want to see "humps" (use histogram/KDE).

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
sns.ecdfplot(data=employees, x='Salary', ax=axes[0])
share_below = (employees['Salary'] <= 100_000).mean()
axes[0].axvline(100_000, color='red', ls='--'); axes[0].axhline(share_below, color='red', ls='--')
axes[0].set_title(f'{share_below:.0%} of employees earn ≤ 100,000'); axes[0].set_ylabel('Proportion ≤ x')

sns.ecdfplot(data=employees, x='Salary', hue='Department', ax=axes[1])
axes[1].axhline(0.5, color='gray', ls=':'); axes[1].set_title('ECDF by department (dotted line = medians)')
plt.tight_layout(); plt.show()
````

### Interpretation
The IT and Finance curves are furthest right → higher salaries at every percentile. Where each curve crosses the dotted 0.5 line is that department's median.

# 17. Regression Plots

### What?
A scatterplot **plus a fitted trend line** (and a shaded 95% confidence band).

```python
sns.regplot(data=df, x='Experience', y='Salary')          # draws on one Axes
sns.lmplot(data=df, x='Experience', y='Salary', hue='Department', col='Gender')   # figure-level: facets & hue
```

| | `regplot` | `lmplot` |
|---|---|---|
| Level | Axes-level (use inside `plt.subplots`) | Figure-level (creates its own figure) |
| hue / facets | No | ✅ `hue`, `col`, `row` |

Options: `order=2` (curve), `lowess=True` (flexible smooth line), `logistic=True` (0/1 targets), `ci=None` (no band), `scatter_kws={'alpha': 0.3}`.

### Correlation vs regression
| | Correlation | Regression |
|---|---|---|
| Answers | How **strongly** do x and y move together? | **How much** does y change per unit of x? Predict y from x |
| Symmetric? | Yes (x↔y same) | No (y depends on x) |
| Output | r between −1 and 1 | A line: y = a + b·x |

### When not?
Curved relationships with a straight line (use `order=2`/`lowess`); treating the line as proof of causation; extrapolating beyond the data range.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
sns.regplot(data=employees, x='Experience', y='Salary', scatter_kws={'alpha': 0.3, 's': 12},
            line_kws={'color': 'red'}, ax=axes[0])
slope, intercept = np.polyfit(employees['Experience'], employees['Salary'], 1)
r = employees['Experience'].corr(employees['Salary'])
axes[0].set_title(f'Salary ≈ {intercept:,.0f} + {slope:,.0f} × Experience   (r = {r:.2f})')

curve = pd.DataFrame({'x': np.linspace(0, 10, 80)}); curve['y'] = (curve['x'] - 5) ** 2 + rng.normal(0, 2, 80)
sns.regplot(data=curve, x='x', y='y', ax=axes[1], scatter_kws={'s': 12}, label='straight line')
sns.regplot(data=curve, x='x', y='y', order=2, scatter=False, ax=axes[1], color='green', label='order=2')
axes[1].set_title('A straight line misses a curve; order=2 fits'); axes[1].legend()
plt.tight_layout(); plt.show()
````

````python
g = sns.lmplot(data=employees, x='Experience', y='Salary', hue='Department', col='Gender',
               height=4, scatter_kws={'s': 10, 'alpha': 0.4})
g.figure.suptitle('lmplot: one regression line per department, one panel per gender', y=1.04)
plt.show()
````

### Interpretation
Each year of experience adds roughly the slope value to salary. The department lines are **parallel but shifted** → similar raise per year, different starting levels. Male and female panels look alike.

### Common mistakes
Fitting straight lines to curved data · ignoring the confidence band · saying "experience causes salary" (it's an association).

# 18. Categorical Visualization

| Plot | Shows | Use when | Avoid when |
|---|---|---|---|
| `countplot` | Number of rows per category | Frequencies, class balance | You need averages |
| `barplot` | **Mean** (or other estimator) per category + error bar (95% CI) | Comparing averages | Distributions differ a lot (the mean hides them) |
| `boxplot` | 5-number summary + outliers | Comparing spreads & medians of several groups | Bimodal data |
| `violinplot` | Full distribution shape | Shapes differ between groups | Small samples |
| `stripplot` | **Every point** (with jitter) | Small/medium data; show sample size | Thousands of points (overlap) |
| `swarmplot` | Every point, **no overlap** | Small data (< ~1,000 points) | Large data (slow, points overflow) |

A good combination: **boxplot + stripplot** (summary + raw data).

````python
order = ['IT', 'Finance', 'Sales', 'HR']
fig, axes = plt.subplots(2, 3, figsize=(18, 9))
sns.countplot(data=employees, x='Department', order=order, ax=axes[0, 0]); axes[0, 0].set_title('countplot: how many?')
sns.barplot(data=employees, x='Department', y='Salary', order=order, estimator='mean', errorbar=('ci', 95), ax=axes[0, 1])
axes[0, 1].set_title('barplot: mean salary ± 95% CI')
sns.boxplot(data=employees, x='Department', y='Salary', order=order, ax=axes[0, 2]); axes[0, 2].set_title('boxplot: median & spread')
sns.violinplot(data=employees, x='Department', y='Salary', order=order, inner='quartile', ax=axes[1, 0]); axes[1, 0].set_title('violinplot: shape')
sns.stripplot(data=employees, x='Department', y='Salary', order=order, alpha=0.5, jitter=0.25, ax=axes[1, 1]); axes[1, 1].set_title('stripplot: every point (jittered)')
sns.boxplot(data=employees, x='Department', y='Salary', order=order, color='white', ax=axes[1, 2])
sns.swarmplot(data=employees.sample(150, random_state=0), x='Department', y='Salary', order=order, size=3, ax=axes[1, 2])
axes[1, 2].set_title('boxplot + swarmplot (150-row sample)')
plt.tight_layout(); plt.show()
````

### Interpretation
All six views agree that IT and Finance pay more. `barplot`'s error bars show how precise each mean is (HR's bar is widest — it has the fewest people). The strip/swarm plots reveal the actual number of employees behind each summary.

### Common mistakes
Showing only means (barplot) for skewed data · swarmplot on thousands of points · forgetting `order=` (alphabetical order is rarely the most useful).

# 19. Time-Series Visualization

### Steps
1. **Date conversion:** `pd.to_datetime()` so Pandas/Matplotlib understand dates.
2. **Sort** by date and (often) set it as the index.
3. **Line plot** the raw series.
4. **Aggregate**: daily → weekly/monthly with `resample('W')`/`resample('MS')` (MS = month start) and `.sum()`/`.mean()`.
5. **Rolling average** (`.rolling(7).mean()`) smooths noise to reveal the **trend**.

### When not?
Irregular, very sparse data (a line implies continuity — use markers); comparing categories (bar chart).

````python
raw_dates = pd.date_range('2024-01-01', '2025-06-30', freq='D')
days = np.arange(len(raw_dates))
orders_ts = pd.DataFrame({
    'order_date': raw_dates.strftime('%d/%m/%Y'),                               # dates stored as TEXT, as in many CSV files
    'orders': (200 + 0.25 * days                                                 # upward trend
               + 40 * np.sin(2 * np.pi * days / 7)                              # weekly pattern
               + 60 * np.sin(2 * np.pi * (days - 300) / 365)                    # yearly season
               + rng.normal(0, 25, len(days))).round(),
})
print(orders_ts.dtypes)

orders_ts['order_date'] = pd.to_datetime(orders_ts['order_date'], format='%d/%m/%Y')   # 1. text → datetime
orders_ts = orders_ts.sort_values('order_date').set_index('order_date')                 # 2. sort & index
print('\nAfter conversion:', orders_ts.index.dtype)
````

````python
monthly = orders_ts['orders'].resample('MS').sum()                  # 4. monthly totals
rolling7 = orders_ts['orders'].rolling(7).mean()                     # 5. 7-day rolling average
rolling30 = orders_ts['orders'].rolling(30).mean()

fig, axes = plt.subplots(2, 1, figsize=(13, 8))
axes[0].plot(orders_ts.index, orders_ts['orders'], color='lightgray', label='daily (noisy)')
axes[0].plot(rolling7.index, rolling7, label='7-day rolling mean')
axes[0].plot(rolling30.index, rolling30, color='red', linewidth=2, label='30-day rolling mean (trend)')
axes[0].set_title('Daily orders with rolling averages'); axes[0].set_ylabel('Orders per day'); axes[0].legend()

axes[1].bar(monthly.index, monthly.values, width=20, color='steelblue')
axes[1].set_title('Monthly total orders'); axes[1].set_ylabel('Orders per month')
plt.tight_layout(); plt.show()
````

### Interpretation
- The raw daily line is noisy (weekly waves). The **30-day rolling mean** shows the long-term **upward trend** plus a yearly seasonal wave (higher around autumn/winter).
- Monthly totals make month-to-month comparison easy (note: months with more days have slightly higher totals — use `.mean()` for a per-day comparison).

### Common mistakes
Leaving dates as strings (wrong order, e.g. "10/01" before "2/01") · not sorting · comparing monthly **sums** of months with different lengths · a rolling window that's too long (hides real changes) or too short (still noisy).

# 20. Advanced Visualization

| Tool | What it does |
|---|---|
| **FacetGrid** | Same plot repeated for each subgroup ("small multiples") — `col=`, `row=`, `hue=` |
| **catplot** | Figure-level categorical plots (`kind='box'`, `'bar'`, `'violin'`, `'strip'` …) with facets |
| **relplot** | Figure-level scatter/line plots with facets |
| **jointplot** | Scatter in the centre + distributions on the margins |
| **pairplot** | All pairwise relationships (Section 13) |
| **subplots** | Full manual control of several Axes |
| **annotations** | Arrows/text pointing to important points (`ax.annotate`) |
| **custom ticks** | Readable tick labels (`ax.set_xticks`, `FuncFormatter`, rotation) |
| **log scale** | `ax.set_yscale('log')` for data spanning many orders of magnitude |

**Axes-level** functions (`scatterplot`, `boxplot`, `histplot`…) draw into one Axes you give them (`ax=`). **Figure-level** functions (`relplot`, `catplot`, `displot`, `lmplot`, `pairplot`, `jointplot`) create their **own** figure with facets — don't pass `ax=` to them.

````python
# FacetGrid: one histogram per department
g = sns.FacetGrid(employees, col='Department', col_order=['IT', 'Finance', 'Sales', 'HR'], height=3, sharey=False)
g.map_dataframe(sns.histplot, x='Salary', bins=15)
g.set_titles('{col_name}'); g.figure.suptitle('FacetGrid: salary distribution per department', y=1.05)
plt.show()

# catplot: boxplots split by department (x) and city (columns)
g = sns.catplot(data=employees, x='Department', y='Salary', col='City', kind='box', height=3.5, order=['IT', 'Finance', 'Sales', 'HR'])
g.figure.suptitle('catplot(kind="box") faceted by city', y=1.05); plt.show()
````

````python
# relplot: scatter faceted by gender, coloured by department
g = sns.relplot(data=employees, x='Experience', y='Salary', hue='Department', col='Gender', height=3.8, s=15, alpha=0.7)
g.figure.suptitle('relplot: experience vs salary by gender', y=1.05); plt.show()

# jointplot: relationship + both distributions
g = sns.jointplot(data=employees, x='Experience', y='Salary', kind='reg', height=5, scatter_kws={'s': 10, 'alpha': 0.4})
g.figure.suptitle('jointplot(kind="reg")', y=1.02); plt.show()
````

````python
from matplotlib.ticker import FuncFormatter

fig, axes = plt.subplots(1, 3, figsize=(19, 4.5))

# 1. Annotations: point out the highest earner
top = employees.loc[employees['Salary'].idxmax()]
axes[0].scatter(employees['Experience'], employees['Salary'], s=10, alpha=0.5)
axes[0].annotate(f"Highest salary: {top['Salary']:,.0f}", xy=(top['Experience'], top['Salary']),
                 xytext=(top['Experience'] - 25, top['Salary'] + 5000), arrowprops={'arrowstyle': '->', 'color': 'red'})
axes[0].set_title('Annotation with an arrow')

# 2. Custom tick labels: show salaries as "90k" and months by name
axes[1].bar(range(1, 13), sales_2025)
axes[1].set_xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'], rotation=45)
axes[1].yaxis.set_major_formatter(FuncFormatter(lambda v, _: f'{v:.0f}k'))
axes[1].set_title('Custom tick labels (month names, "k" suffix)')

# 3. Log scale: company sizes span 5 orders of magnitude
companies = pd.Series({'Startup': 12, 'Small': 150, 'Medium': 2_300, 'Large': 45_000, 'Giant': 1_200_000})
axes[2].bar(companies.index, companies.values, color='tab:purple'); axes[2].set_yscale('log')
axes[2].set_title('Log scale: 12 and 1,200,000 both visible'); axes[2].set_ylabel('Employees (log scale)')
plt.tight_layout(); plt.show()
````

### When not?
- Facets with dozens of tiny panels (use fewer categories).
- Log scales for audiences who don't know them (label clearly!) or for data with zeros/negatives.
- Annotations that clutter instead of highlight (annotate 1–3 key points).

# 21. Choosing the Correct Chart — Decision Guide

| Question | Recommended chart | Alternatives |
|---|---|---|
| Compare categories (values) | **Bar chart** (sorted; horizontal for long names) | Dot plot |
| How often does each category occur? | **Countplot** | Bar chart of `value_counts()` |
| Distribution of one numeric variable | **Histogram** / **KDE** | ECDF, boxplot |
| Distributions of several groups | **Boxplot** / **violin** | Overlaid KDE, ECDF, FacetGrid of histograms |
| Outliers | **Boxplot** | Scatter, strip plot |
| Relationship between two numbers | **Scatterplot** | regplot (+ trend), hexbin (big data), jointplot |
| Relationship + a third variable | Scatter with **hue/size/style** | relplot with facets |
| Correlation among many numbers | **Heatmap** | Pairplot (few variables) |
| Trend over time | **Line chart** (+ rolling mean) | Area chart, bar chart for monthly totals |
| Composition over time | **Stacked area** / stacked bar | Line per component |
| Parts of a whole (2–4 parts) | **Bar chart** or 100% stacked bar | Pie (only if few, very different parts) |
| Percentiles / "share below x" | **ECDF** | Boxplot |
| Data spanning orders of magnitude | Any chart on a **log scale** | — |

# 22. Misleading Visualizations

A chart can be technically correct and still **mislead**. Recognise these tricks — and never use them.

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
brands = ['Brand A', 'Brand B']; share = [50.5, 49.5]
axes[0].bar(brands, share, color=['tab:red', 'tab:blue']); axes[0].set_ylim(49, 51)
axes[0].set_title('❌ Truncated y-axis: A looks 3× bigger')
axes[1].bar(brands, share, color=['tab:red', 'tab:blue']); axes[1].set_ylim(0, 60)
axes[1].set_title('✅ Axis from 0: they are almost equal')
plt.tight_layout(); plt.show()
````

**1. Truncated y-axis:** bars encode values by **length from zero**. Starting at 49 turns a 1-point difference into a huge visual gap. (Line charts may start above zero, but say so clearly.)

````python
cities = ['Lahore', 'Karachi', 'Islamabad', 'Multan', 'Quetta', 'Peshawar', 'Sialkot', 'Faisalabad', 'Hyderabad', 'Gujranwala']
vals = rng.integers(20, 100, len(cities))
fig, axes = plt.subplots(1, 2, figsize=(15, 4))
axes[0].bar(cities, vals, color=sns.color_palette('hls', len(cities))); axes[0].tick_params(axis='x', rotation=45)
axes[0].set_title('❌ Rainbow colours carry no meaning')
colors = ['tab:red' if c == 'Lahore' else 'lightgray' for c in cities]
axes[1].bar(cities, vals, color=colors); axes[1].tick_params(axis='x', rotation=45)
axes[1].set_title('✅ One highlight colour tells the story (Lahore)')
plt.tight_layout(); plt.show()
````

**2. Too many colours:** colour should **mean** something. Use one neutral colour plus one highlight.

````python
from mpl_toolkits.mplot3d import Axes3D   # noqa: F401  (enables 3-D axes)

vals3 = [30, 25, 25, 20]
labels3 = ['North', 'South', 'East', 'West']
fig = plt.figure(figsize=(13, 4))
ax3d = fig.add_subplot(1, 2, 1, projection='3d')
ax3d.bar3d(range(4), [0] * 4, [0] * 4, 0.6, 0.6, vals3, color='tab:orange')
ax3d.set_xticks(range(4), labels3); ax3d.view_init(elev=25, azim=-35)
ax3d.set_title('❌ 3-D bars: perspective distorts heights')
ax2d = fig.add_subplot(1, 2, 2)
ax2d.bar(labels3, vals3, color='tab:orange'); ax2d.set_title('✅ 2-D bars: easy to compare')
plt.tight_layout(); plt.show()
````

**3. 3-D charts:** perspective makes front bars look bigger and hides values behind others. Use 2-D.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
many_products = pd.Series(rng.integers(5, 100, 40), index=[f'Product {i}' for i in range(40)])
many_products.plot(kind='bar', ax=axes[0]); axes[0].set_title('❌ 40 categories: unreadable'); axes[0].tick_params(axis='x', labelsize=6)
top10 = many_products.nlargest(10).sort_values()
top10.plot(kind='barh', ax=axes[1], color='steelblue'); axes[1].set_title('✅ Top 10 only (group the rest as "Other" if needed)')
plt.tight_layout(); plt.show()
````

**4. Wrong chart type** (Section 3: a line through categories; Section 14: pies with many slices) and **5. Too many categories** (above): show the top N, group the rest, or facet.

````python
mixture = np.concatenate([rng.normal(20, 3, 300), rng.normal(35, 3, 300)])    # two groups
fig, axes = plt.subplots(1, 2, figsize=(13, 3.8))
axes[0].hist(mixture, bins=4, color='gray', edgecolor='white'); axes[0].set_title('❌ bins=4: looks like one group')
axes[1].hist(mixture, bins=40, color='steelblue', edgecolor='white'); axes[1].set_title('✅ bins=40: two clear groups')
plt.tight_layout(); plt.show()
````

**6. Misleading bin sizes:** too few bins can hide important structure (two groups); unequal bin widths can make bars look bigger than they are.

````python
years = np.arange(2010, 2025)
ice_cream = 100 + 8 * (years - 2010) + rng.normal(0, 6, len(years))
drownings = 40 + 3 * (years - 2010) + rng.normal(0, 3, len(years))
fig, ax1 = plt.subplots(figsize=(9, 4))
ax1.plot(years, ice_cream, color='tab:orange', marker='o', label='Ice cream sales'); ax1.set_ylabel('Ice cream sales', color='tab:orange')
ax2 = ax1.twinx(); ax2.plot(years, drownings, color='tab:blue', marker='s', label='Drownings'); ax2.set_ylabel('Drownings', color='tab:blue')
ax1.set_title(f'r = {np.corrcoef(ice_cream, drownings)[0, 1]:.2f} … but ice cream does NOT cause drowning\n(both grow with population & hotter summers — a confounder)')
plt.tight_layout(); plt.show()
````

**7. Correlation mistaken for causation:** two series that both trend over time (or share a hidden cause) look strongly related. A chart shows **association**; causation needs experiments or careful causal analysis. (Also beware dual y-axes: they can make any two series look aligned.)

### Checklist for honest charts
✅ Bars start at 0 · ✅ axis labels + units · ✅ clear title stating the message · ✅ meaningful colours · ✅ no 3-D · ✅ ≤ ~7 categories or top-N · ✅ sensible bins · ✅ source & sample size · ✅ association ≠ causation.
