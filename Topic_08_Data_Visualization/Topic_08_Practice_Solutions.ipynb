# Practice Problems

Run the next cell first — it creates the `shop` dataset used by most problems. Each problem has an **empty code cell** right after it. **Solutions are in the separate Solutions section.**

````python
rp = np.random.default_rng(7)
n = 400
shop = pd.DataFrame({
    'Date': pd.to_datetime('2025-01-01') + pd.to_timedelta(rp.integers(0, 180, n), unit='D'),
    'Store': rp.choice(['North', 'South', 'East', 'West'], n, p=[0.35, 0.25, 0.25, 0.15]),
    'Category': rp.choice(['Electronics', 'Clothing', 'Grocery', 'Books'], n, p=[0.2, 0.3, 0.35, 0.15]),
    'Customer_Age': rp.integers(18, 70, n),
    'Membership': rp.choice(['None', 'Silver', 'Gold'], n, p=[0.55, 0.3, 0.15]),
})
base_price = shop['Category'].map({'Electronics': 25000, 'Clothing': 3500, 'Grocery': 1200, 'Books': 900})
shop['Units'] = rp.integers(1, 6, n)
shop['Amount'] = (base_price * rp.lognormal(0, 0.35, n) * shop['Units']).round(0)
shop['Satisfaction'] = np.clip(rp.normal(3.8, 0.8, n) + 0.3 * (shop['Membership'] == 'Gold'), 1, 5).round(1)
shop = shop.sort_values('Date').reset_index(drop=True)
shop.head()
````

# Solutions

⚠️ Try the problems first!

### Solution 1

````python
daily = shop.groupby('Date')['Amount'].sum()
fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(daily.index, daily.values)
ax.set_title('Total amount per day'); ax.set_xlabel('Date'); ax.set_ylabel('Amount (PKR)'); ax.grid(alpha=0.3)
plt.show()
````

**Explanation:** Grouping by date gives one value per day. The line is very noisy because a single electronics purchase can dominate a day — a rolling mean (Problem 15) helps.

### ✍️ Practice Problem 2

**Line chart with multiple lines:** Plot the **weekly** total `Amount` for each `Store` as separate lines on one chart, with a legend.

*(Solution in the **Practice Solutions** section at the end.)*

````python
# ✍️ Practice Problem 2 — write your solution here
````

### Solution 2

````python
weekly = shop.set_index('Date').groupby('Store')['Amount'].resample('W').sum().unstack(0)   # rows = weeks, columns = stores
fig, ax = plt.subplots(figsize=(11, 4))
for store in weekly.columns:
    ax.plot(weekly.index, weekly[store], marker='o', markersize=3, label=store)
ax.set_title('Weekly amount by store'); ax.set_ylabel('Amount (PKR)'); ax.legend(); plt.show()
````

**Explanation:** `resample('W')` inside the groupby aggregates each store by week. Four lines are still readable; with many stores use small multiples.

### ✍️ Practice Problem 8

**Boxplot:** Draw a boxplot of `Amount` by `Category` and compute Q1, median, Q3 and the IQR outlier fences for the `Clothing` category with NumPy.

*(Solution in the **Practice Solutions** section at the end.)*

````python
# ✍️ Practice Problem 8 — write your solution here
````

### Solution 8

````python
fig, ax = plt.subplots(figsize=(9, 4))
sns.boxplot(data=shop, x='Category', y='Amount', ax=ax); ax.set_yscale('log'); ax.set_title('Amount by category (log scale)')
plt.show()
clothing = shop.loc[shop['Category'] == 'Clothing', 'Amount']
q1, med, q3 = np.percentile(clothing, [25, 50, 75]); iqr = q3 - q1
print(f'Clothing: Q1 = {q1:,.0f}, median = {med:,.0f}, Q3 = {q3:,.0f}, IQR = {iqr:,.0f}')
print(f'Fences: [{q1 - 1.5 * iqr:,.0f}, {q3 + 1.5 * iqr:,.0f}] → outliers: {((clothing < q1 - 1.5 * iqr) | (clothing > q3 + 1.5 * iqr)).sum()}')
````

**Explanation:** The log y-axis lets all four categories fit on one chart. The quartiles quantify the box for Clothing.

### ✍️ Practice Problem 15

**Time series with rolling mean:** Plot daily total `Amount` together with its 7-day and 30-day rolling averages, and a monthly bar chart below it.

*(Solution in the **Practice Solutions** section at the end.)*

````python
# ✍️ Practice Problem 15 — write your solution here
````

### Solution 15

````python
daily15 = shop.groupby('Date')['Amount'].sum().asfreq('D', fill_value=0)     # fill days without sales with 0
fig, axes = plt.subplots(2, 1, figsize=(12, 7))
axes[0].plot(daily15.index, daily15, color='lightgray', label='daily')
axes[0].plot(daily15.rolling(7).mean(), label='7-day mean'); axes[0].plot(daily15.rolling(30).mean(), color='red', label='30-day mean')
axes[0].legend(); axes[0].set_title('Daily amount with rolling averages')
monthly15 = daily15.resample('MS').sum()
axes[1].bar(monthly15.index, monthly15.values, width=20); axes[1].set_title('Monthly total')
plt.tight_layout(); plt.show()
````

**Explanation:** `asfreq('D', fill_value=0)` inserts missing days so the rolling windows are true 7/30-day windows.

### ✍️ Practice Problem 16

**Advanced facets + misleading fix:** (a) Use `sns.catplot` to show `Amount` boxplots by `Category`, faceted by `Store` (log y-axis). (b) Recreate a truncated-axis bar chart of average `Satisfaction` by `Membership` (y from 3.5 to 4.5), then fix it.

*(Solution in the **Practice Solutions** section at the end.)*

````python
# ✍️ Practice Problem 16 — write your solution here
````

### Solution 16

````python
g = sns.catplot(data=shop, x='Category', y='Amount', col='Store', kind='box', col_wrap=2, height=3)
g.set(yscale='log'); g.figure.suptitle('Amount by category, per store', y=1.02); plt.show()

sat = shop.groupby('Membership')['Satisfaction'].mean().reindex(['None', 'Silver', 'Gold'])
fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
axes[0].bar(sat.index, sat.values, color='tab:red'); axes[0].set_ylim(3.5, 4.5); axes[0].set_title('❌ Truncated axis exaggerates')
axes[1].bar(sat.index, sat.values, color='tab:blue'); axes[1].set_ylim(0, 5); axes[1].set_title('✅ Full scale 0–5')
plt.tight_layout(); plt.show()
````

**Explanation:** Faceting keeps each store's chart simple. The truncated axis makes Gold look far better; on the full 0–5 scale the difference is real but modest.

## The dataset (600 orders)

````python
rs = np.random.default_rng(2024)
n = 600
catalog = pd.DataFrame({
    'Product':  ['Laptop', 'Smartphone', 'Headphones', 'T-Shirt', 'Jeans', 'Sneakers', 'Novel', 'Textbook', 'Face Cream', 'Perfume'],
    'Category': ['Electronics'] * 3 + ['Fashion'] * 3 + ['Books'] * 2 + ['Beauty'] * 2,
    'Base_Price': [150000, 90000, 8000, 1500, 3500, 7000, 900, 2500, 1800, 6000],
})
# More orders later in the year (growth) and in Nov–Dec (sale season)
day_weights = np.linspace(1, 1.8, 366) * np.where(np.arange(366) >= 305, 1.6, 1.0)
dates = pd.to_datetime('2024-01-01') + pd.to_timedelta(rs.choice(366, n, p=day_weights / day_weights.sum()), unit='D')

prod_idx = rs.choice(len(catalog), n, p=[0.05, 0.08, 0.12, 0.14, 0.12, 0.09, 0.1, 0.08, 0.12, 0.10])
sales = catalog.iloc[prod_idx].reset_index(drop=True)
sales['Date'] = dates
sales['City'] = rs.choice(['Karachi', 'Lahore', 'Islamabad', 'Faisalabad', 'Multan'], n, p=[0.32, 0.28, 0.18, 0.12, 0.10])
sales['Gender'] = np.where(sales['Category'] == 'Beauty', rs.choice(['Female', 'Male'], n, p=[0.75, 0.25]),
                           rs.choice(['Female', 'Male'], n, p=[0.45, 0.55]))
sales['Age'] = np.clip(rs.normal(33, 9, n), 18, 65).round().astype(int)
sales['Price'] = (sales['Base_Price'] * rs.uniform(0.9, 1.1, n)).round(-1)
sales['Discount'] = np.where(sales['Date'].dt.month >= 11, rs.choice([0.1, 0.2, 0.3], n), rs.choice([0.0, 0.05, 0.1], n))
cheap = sales['Base_Price'] < 5000
sales['Quantity'] = np.where(cheap, rs.integers(1, 6, n), rs.choice([1, 1, 1, 2], n)) + (sales['Discount'] >= 0.2) * rs.integers(0, 2, n)
sales['Revenue'] = (sales['Price'] * sales['Quantity'] * (1 - sales['Discount'])).round(0)
sales['Rating'] = np.clip(rs.normal(4.1, 0.6, n) + sales['Category'].map({'Books': 0.3, 'Beauty': 0.1, 'Fashion': -0.1, 'Electronics': -0.2}), 1, 5).round(1)
sales['Payment_Method'] = np.where(sales['Price'] > 50000, rs.choice(['Card', 'Bank Transfer', 'Cash on Delivery'], n, p=[0.6, 0.3, 0.1]),
                                   rs.choice(['Card', 'Cash on Delivery', 'Wallet'], n, p=[0.35, 0.45, 0.2]))
sales = sales.drop(columns='Base_Price').sort_values('Date').reset_index(drop=True)
sales = sales[['Date', 'Product', 'Category', 'City', 'Gender', 'Age', 'Price', 'Quantity', 'Discount', 'Revenue', 'Rating', 'Payment_Method']]
print(sales.shape)
sales.head()
````

````python
print(sales.dtypes)
print('\nTotal revenue: PKR', f"{sales['Revenue'].sum():,.0f}", '| orders:', len(sales))
sales.describe().round(2)
````

### 1. Sales trend over the year (daily revenue + 30-day rolling average)

````python
daily_rev = sales.groupby('Date')['Revenue'].sum().asfreq('D', fill_value=0)
fig, ax = plt.subplots(figsize=(13, 4.5))
ax.plot(daily_rev.index, daily_rev / 1e3, color='lightgray', label='Daily revenue')
ax.plot(daily_rev.rolling(30).mean() / 1e3, color='tab:red', linewidth=2.5, label='30-day rolling average')
ax.set_title('Sales trend 2024'); ax.set_ylabel('Revenue (thousand PKR)'); ax.legend(); plt.show()
````

1. **What it shows:** Daily revenue and its 30-day rolling average across 2024.
2. **Why this chart:** A **line chart** is the standard way to show change over time; the rolling average removes day-to-day noise from a few big electronics orders.
3. **Insight:** The rolling average trends **upward** through the year, with a clear jump in **November–December** (the discount season). The daily line is very spiky because a single laptop or smartphone order can be worth more than a whole day of small orders.

### 2. Monthly sales

````python
monthly_rev = sales.set_index('Date')['Revenue'].resample('MS').sum()
fig, ax = plt.subplots(figsize=(12, 4))
bars = ax.bar(monthly_rev.index.strftime('%b'), monthly_rev / 1e6, color=['tab:red' if m >= 11 else 'steelblue' for m in monthly_rev.index.month])
ax.bar_label(bars, fmt='%.1f'); ax.set_title('Monthly revenue (million PKR) — sale season highlighted'); ax.set_ylabel('Revenue (million PKR)')
plt.show()
````

1. **What it shows:** Total revenue per month.
2. **Why this chart:** Monthly totals are **discrete periods** → a bar chart makes month-to-month comparison easy; highlight colour marks the sale season.
3. **Insight:** November and December (red) are the two strongest months and bring in about 28% of annual revenue. Other months jump around (e.g. a weak February) because revenue depends on how many big electronics orders happen to fall in each month.

### 3. Product sales (revenue and units)

````python
prod = sales.groupby('Product').agg(Revenue=('Revenue', 'sum'), Units=('Quantity', 'sum')).sort_values('Revenue')
fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
(prod['Revenue'] / 1e6).plot(kind='barh', ax=axes[0], color='steelblue'); axes[0].set_title('Revenue by product (million PKR)')
prod['Units'].sort_values().plot(kind='barh', ax=axes[1], color='tab:orange'); axes[1].set_title('Units sold by product')
plt.tight_layout(); plt.show()
````

1. **What it shows:** Revenue and units sold for each of the 10 products.
2. **Why this chart:** Sorted **horizontal bars** rank many categories with readable names; showing revenue **and** units side by side avoids a one-sided view.
3. **Insight:** **Laptops and smartphones** earn most of the revenue despite selling few units; cheap items (T-shirts, novels, face cream) sell many units but add little revenue.

### 4. Category sales

````python
cat = sales.groupby('Category')['Revenue'].sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(cat.index, cat / 1e6, color='teal'); ax.bar_label(bars, fmt='%.1f')
ax.set_title('Revenue by category (million PKR)'); ax.set_ylabel('Revenue (million PKR)'); plt.show()
print((cat / cat.sum() * 100).round(1).astype(str) + ' %')
````

1. **What it shows:** Total revenue per product category.
2. **Why this chart:** Only four categories → a simple **vertical bar chart** (a pie would work but bars compare more precisely).
3. **Insight:** **Electronics** brings in about 81% of revenue; Books is the smallest category (about 3–4%).

### 5. City sales

````python
city = sales.groupby('City').agg(Revenue=('Revenue', 'sum'), Orders=('Revenue', 'size')).sort_values('Revenue')
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
(city['Revenue'] / 1e6).plot(kind='barh', ax=axes[0], color='slateblue'); axes[0].set_title('Revenue by city (million PKR)')
(city['Revenue'] / city['Orders'] / 1e3).plot(kind='barh', ax=axes[1], color='gray'); axes[1].set_title('Average order value (thousand PKR)')
plt.tight_layout(); plt.show()
````

1. **What it shows:** Total revenue and average order value per city.
2. **Why this chart:** Sorted bars rank cities; adding **average order value** separates "more customers" from "bigger baskets".
3. **Insight:** **Karachi** earns the most revenue, while **Lahore** has the **most orders** but a much lower average order value (≈ 17.6k vs ≈ 27.4k in Karachi). Lahore customers buy more cheap items; Karachi customers buy more electronics. Same rank in a chart, very different business stories.

### 6. Revenue distribution

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(sales['Revenue'], bins=40, ax=axes[0]); axes[0].set_title(f"Revenue per order (skew = {sales['Revenue'].skew():.1f})")
sns.histplot(sales['Revenue'], bins=40, log_scale=True, ax=axes[1]); axes[1].set_title('Same data on a log scale')
plt.tight_layout(); plt.show()
````

1. **What it shows:** How order revenues are distributed.
2. **Why this chart:** A **histogram** shows the shape; the **log scale** is needed because values range from a few hundred to several hundred thousand rupees.
3. **Insight:** Revenue per order is **extremely right-skewed**: most orders are small, a few electronics orders are huge. The log view shows separate humps for cheap and expensive products → report medians, not just means.

### 7. Age distribution

````python
fig, ax = plt.subplots(figsize=(9, 4))
sns.histplot(sales['Age'], bins=20, kde=True, ax=ax)
ax.axvline(sales['Age'].median(), color='red', ls='--', label=f"median = {sales['Age'].median():.0f}")
ax.set_title('Customer age distribution'); ax.legend(); plt.show()
````

1. **What it shows:** Ages of the customers who placed orders.
2. **Why this chart:** **Histogram + KDE** shows both counts and the smooth shape of one numeric variable.
3. **Insight:** The middle half of customers are **28–40 years old** (median ≈ 34) — marketing should target young professionals.

### 8. Rating distribution

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(sales['Rating'], bins=np.arange(1, 5.2, 0.2), ax=axes[0]); axes[0].set_title('Ratings')
sns.boxplot(data=sales, x='Category', y='Rating', order=['Books', 'Beauty', 'Fashion', 'Electronics'], ax=axes[1]); axes[1].set_title('Rating by category')
plt.tight_layout(); plt.show()
````

1. **What it shows:** Overall ratings and ratings per category.
2. **Why this chart:** A histogram for the overall shape; **boxplots** to compare the distribution of several groups.
3. **Insight:** About 87% of ratings are between 3.5 and 5, with a small pile-up at the maximum 5.0. **Books** get the highest ratings (median 4.5) and **Electronics** the lowest (median 3.9) — a possible product-quality or delivery issue to investigate.

### 9. Gender comparison

````python
gender_cat = sales.groupby(['Category', 'Gender'])['Revenue'].sum().unstack() / 1e6
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.countplot(data=sales, x='Category', hue='Gender', ax=axes[0]); axes[0].set_title('Orders by category and gender')
gender_cat.plot(kind='bar', ax=axes[1], rot=0); axes[1].set_title('Revenue by category and gender (million PKR)')
plt.tight_layout(); plt.show()
````

1. **What it shows:** Orders and revenue split by gender within each category.
2. **Why this chart:** **Countplot with hue** and **grouped bars** compare two categorical variables side by side.
3. **Insight:** **Beauty** is bought mostly by women (about 70% of orders). Men place more orders in **Electronics** and **Fashion**; Books is roughly balanced. Overall the split is fairly even.

### 10. Discount vs revenue

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.boxplot(data=sales, x='Discount', y='Revenue', ax=axes[0]); axes[0].set_yscale('log'); axes[0].set_title('Revenue per order by discount level (log)')
disc = sales.groupby('Discount').agg(Orders=('Revenue', 'size'), Avg_Quantity=('Quantity', 'mean'))
disc['Avg_Quantity'].plot(kind='bar', ax=axes[1], rot=0, color='tab:green'); axes[1].set_title('Average quantity per order by discount')
plt.tight_layout(); plt.show()
print(disc.round(2))
````

1. **What it shows:** How revenue per order and quantity change with the discount level.
2. **Why this chart:** Discount has only a few levels, so **boxplots per level** (not a scatter) show the full revenue spread; a bar chart shows average quantity.
3. **Insight:** Higher discounts (20–30%, only in Nov–Dec) come with **larger quantities per order**, but revenue per order depends far more on **which product** was bought than on the discount. Discount is not a clean driver of revenue by itself.

### 11. Price vs quantity

````python
fig, ax = plt.subplots(figsize=(9, 4.5))
sns.stripplot(data=sales, x='Quantity', y='Price', hue='Category', alpha=0.6, jitter=0.25, ax=ax)
ax.set_yscale('log'); ax.set_title('Price vs quantity per order (log price)'); ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left'); plt.show()
print('Correlation price–quantity:', round(sales['Price'].corr(sales['Quantity']), 2))
````

1. **What it shows:** The unit price of each order against how many units were bought.
2. **Why this chart:** Two numeric variables → a scatter-type plot; quantity takes few values, so a jittered **stripplot** avoids overlapping dots; log scale for prices.
3. **Insight:** **Negative relationship:** expensive items (electronics) are bought one at a time; cheap items are bought in larger quantities.

### 12. Correlation heatmap

````python
num = sales[['Age', 'Price', 'Quantity', 'Discount', 'Revenue', 'Rating']]
fig, ax = plt.subplots(figsize=(7, 5.5))
sns.heatmap(num.corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, square=True, ax=ax)
ax.set_title('Correlation between numeric variables'); plt.show()
````

1. **What it shows:** Pairwise correlations between all numeric columns.
2. **Why this chart:** A **heatmap** summarises many correlations at once; the diverging colour map makes positive/negative obvious.
3. **Insight:** Price ↔ Revenue is very strongly positive (revenue is driven by what is bought). Price ↔ Quantity is negative. Rating correlates slightly negatively with price (electronics are rated lower). Age barely correlates with anything. Remember: correlation ≠ causation.

### 13. Customer / product analysis — which age groups buy which categories?

````python
sales['Age_Group'] = pd.cut(sales['Age'], bins=[17, 25, 35, 45, 65], labels=['18–25', '26–35', '36–45', '46+'])
age_cat = pd.crosstab(sales['Age_Group'], sales['Category'], normalize='index') * 100
fig, ax = plt.subplots(figsize=(9, 4))
age_cat.plot(kind='bar', stacked=True, ax=ax, rot=0); ax.set_ylabel('% of orders')
ax.set_title('Category mix by age group (100% stacked)'); ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left'); plt.show()
````

1. **What it shows:** The share of each category in each age group's orders.
2. **Why this chart:** A **100% stacked bar** compares composition across groups of different sizes.
3. **Insight:** The category mix is broadly similar across age groups: differences are a few percentage points (e.g. customers aged 46+ buy fewer books and a bit more electronics). There is no strong age effect → age alone is not a very useful segmentation for category promotions here.

### 14. Multiple-category comparison — revenue by city × category

````python
pivot = sales.pivot_table(index='City', columns='Category', values='Revenue', aggfunc='sum') / 1e6
fig, ax = plt.subplots(figsize=(9, 4.5))
sns.heatmap(pivot, annot=True, fmt='.1f', cmap='YlGnBu', ax=ax); ax.set_title('Revenue by city and category (million PKR)'); plt.show()
````

1. **What it shows:** Revenue for every city–category combination.
2. **Why this chart:** Two categorical dimensions and one value → an **annotated heatmap** of a pivot table.
3. **Insight:** Electronics leads in every city; Karachi–Electronics and Lahore–Electronics are the biggest cells, so those segments deserve the most stock and marketing.

### 15. Payment methods

````python
pay = pd.crosstab(sales['Category'], sales['Payment_Method'], normalize='index') * 100
fig, axes = plt.subplots(1, 2, figsize=(15, 4))
order_pm = sales['Payment_Method'].value_counts().index
sns.countplot(data=sales, y='Payment_Method', order=order_pm, ax=axes[0], color='steelblue'); axes[0].set_title('Orders by payment method')
pay.plot(kind='barh', stacked=True, ax=axes[1]); axes[1].set_xlabel('% of orders'); axes[1].set_title('Payment mix by category')
axes[1].legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout(); plt.show()
````

1. **What it shows:** How customers pay, overall and per category.
2. **Why this chart:** **Countplot** for frequency; **100% stacked bars** for the mix within each category.
3. **Insight:** **Cash on Delivery** and **Card** are almost tied as the most common methods (COD dominates cheap categories like Books and Fashion). Electronics buyers mostly pay by **card or bank transfer** (≈ 68%). Make sure card payments work flawlessly for high-value orders.

### 16. Revenue distribution by city (violin, log scale)

````python
fig, ax = plt.subplots(figsize=(11, 4.5))
sns.violinplot(data=sales, x='City', y='Revenue', log_scale=True, inner='quartile', cut=0, ax=ax)
ax.set_title('Order revenue distribution by city (log scale)'); plt.show()
````

1. **What it shows:** The full shape of order values in each city.
2. **Why this chart:** **Violin plots** reveal multi-peaked shapes (cheap vs expensive products) that boxplots would hide.
3. **Insight:** Every city has the same two-peak pattern — many small orders plus a cluster of large electronics orders. Cities differ mainly in **volume**, not in the kind of orders.

### 17. Dashboard overview (subplots)

````python
fig, axes = plt.subplots(2, 2, figsize=(15, 9))
axes[0, 0].plot(monthly_rev.index.strftime('%b'), monthly_rev / 1e6, marker='o'); axes[0, 0].set_title('Monthly revenue (M PKR)')
(cat / 1e6).plot(kind='bar', ax=axes[0, 1], rot=0, color='teal'); axes[0, 1].set_title('Revenue by category (M PKR)')
(city['Revenue'] / 1e6).plot(kind='barh', ax=axes[1, 0], color='slateblue'); axes[1, 0].set_title('Revenue by city (M PKR)')
sns.histplot(sales['Rating'], bins=20, ax=axes[1, 1]); axes[1, 1].set_title('Ratings')
fig.suptitle('Sales Dashboard 2024', fontsize=16, fontweight='bold'); plt.tight_layout()
plt.savefig('sales_dashboard_2024.png', dpi=120, bbox_inches='tight'); plt.show()
````

# Visualization Cheat Sheet

| Goal | Chart | Matplotlib | Seaborn | Pandas |
|---|---|---|---|---|
| Trend over time | Line | `ax.plot(x, y)` | `sns.lineplot` | `df.plot()` |
| Compare categories | Bar | `ax.bar` / `ax.barh` | `sns.barplot` (mean ± CI) | `df.plot(kind='bar')` |
| Category frequency | Count | — | `sns.countplot` | `s.value_counts().plot(kind='bar')` |
| Grouped / stacked | Bar | several `ax.bar` | `hue=` | `df.plot(kind='bar', stacked=True)` |
| Distribution | Histogram | `ax.hist(x, bins=)` | `sns.histplot` (`kde=True`, `log_scale=True`) | `s.plot(kind='hist')` |
| Smooth distribution | KDE | — | `sns.kdeplot` (`bw_adjust`, `clip`, `common_norm`) | `s.plot(kind='kde')` |
| Percentiles | ECDF | — | `sns.ecdfplot` | — |
| Spread & outliers | Boxplot | `ax.boxplot` | `sns.boxplot` | `df.boxplot()` |
| Distribution shape by group | Violin | `ax.violinplot` | `sns.violinplot(inner='quartile')` | — |
| Raw points by group | Strip / swarm | — | `sns.stripplot`, `sns.swarmplot` | — |
| Relationship | Scatter | `ax.scatter` | `sns.scatterplot(hue, size, style)` | `df.plot(kind='scatter')` |
| Trend line | Regression | `np.polyfit` + `ax.plot` | `sns.regplot`, `sns.lmplot` | — |
| Big-data relationship | Hexbin | `ax.hexbin` | `sns.jointplot(kind='hex')` | `df.plot(kind='hexbin')` |
| Correlation / 2-D table | Heatmap | `ax.imshow` | `sns.heatmap(annot=True)` | — |
| All pairs | Pairplot | — | `sns.pairplot(hue, corner=True)` | `pd.plotting.scatter_matrix` |
| Parts of a whole | Pie (2–4 parts) | `ax.pie(autopct=)` | — | `s.plot(kind='pie')` |
| Composition over time | Stacked area | `ax.stackplot` / `fill_between` | — | `df.plot(kind='area')` |
| Small multiples | Facets | `plt.subplots` | `FacetGrid`, `relplot`, `catplot`, `displot` | `df.plot(subplots=True)` |

### Essential Matplotlib commands
```python
fig, ax = plt.subplots(nrows, ncols, figsize=(w, h))   # figure + axes
ax.set_title(); ax.set_xlabel(); ax.set_ylabel()       # labels
ax.legend(); ax.grid(alpha=0.3)                        # legend & grid
ax.set_xlim(); ax.set_ylim(); ax.set_yscale('log')     # limits & scales
ax.set_xticks(ticks, labels, rotation=45)             # custom ticks
ax.annotate('text', xy=(x, y), xytext=(x2, y2), arrowprops={'arrowstyle': '->'})
ax.bar_label(ax.containers[0])                         # values on bars
plt.tight_layout(); plt.savefig('file.png', dpi=150, bbox_inches='tight'); plt.show()
```

### Golden rules
1. Start with the **question**, then pick the chart (Section 21).
2. Always add a **title, axis labels and units**.
3. **Bars start at zero**; no 3-D; no rainbow colours without meaning.
4. **Sort** categories; show the **top N** when there are many.
5. Check distributions with **several bin sizes**; use **log scales** for heavily skewed data.
6. Show **raw data or spread** (box/violin/strip), not just averages.
7. Use **facets** instead of cramming many series into one chart.
8. **Correlation ≠ causation.**
9. **Save** figures with `savefig` **before** `show()`.
