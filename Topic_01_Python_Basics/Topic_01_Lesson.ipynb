# 4. Variable Naming Rules

Valid variable names can contain:

- Letters
- Numbers
- Underscores

But a variable cannot start with a number.

Good examples:

```python
first_name = "Ali"
age = 25
student_1 = "Ahmed"
```

Avoid names such as:

```python
1name = "Ali"
first-name = "Ali"
```

Python is case-sensitive:

```python
name
Name
NAME
```

These are different variables.

Python supports:

| Operator | Meaning |
|---|---|

`*args` allows a function to receive a variable number of positional arguments.

````python
def calculate_sum(*numbers):
    print(numbers)
    return sum(numbers)

print(calculate_sum(1, 2, 3))
print(calculate_sum(10, 20, 30, 40))
````

`**kwargs` allows a function to receive a variable number of keyword arguments.

````python
def show_info(**info):
    print(info)
    for key, value in info.items():
        print(key, value)

show_info(name="Ali", age=25, city="Lahore")
````

<a id="sec-55"></a>
# 55. Exception Handling

Errors can occur while a program is running.

Python provides `try` and `except` to handle errors.

````python
try:
    number = int("abc")
except ValueError:
    print("Invalid number")
````

<a id="sec-56"></a>
# 56. try / except / else / finally

- `try` contains code that may fail.
- `except` handles the error.
- `else` runs if no error occurs.
- `finally` runs regardless of the result.

````python
try:
    number = int("100")
except ValueError:
    print("Invalid number")
else:
    print("Conversion successful")
finally:
    print("Finished")
````

# 60. Object-Oriented Programming Basics

Python supports object-oriented programming.

A class is a blueprint for creating objects.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")

person = Person("Ali", 25)
person.introduce()
````

- `len()`

- `max()`
- `sorted()`
- `round()`
- `abs()`
- `any()`

````python
numbers = [10, 20, 30, 40, 50]

print("Length:", len(numbers))
print("Sum:", sum(numbers))
print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
print("Sorted:", sorted(numbers, reverse=True))
print("Rounded:", round(10.5678, 2))
print("Absolute:", abs(-50))
````

<a id="sec-62"></a>
# 62. any() and all()

`any()` returns `True` if at least one condition is true.

`all()` returns `True` if every condition is true.

````python
values = [True, False, True]

print("Any:", any(values))
print("All:", all(values))
````

# 63. Membership Operators

Use `in` and `not in` to check whether a value exists in a collection.

````python
names = ["Ali", "Sara", "John"]

print("Ali" in names)
print("David" in names)
print("David" not in names)
````

<a id="sec-64"></a>
# 64. Identity Operators

`is` checks whether two variables refer to the same object.

`==` checks whether two values are equal.

These are not always the same concept.

````python
x = None

print(x is None)
print(x == None)
````

<a id="sec-65"></a>
# 65. Nested Data Structures

Python collections can contain other collections.

This concept is especially useful when working with JSON and API data.

````python
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "John", "score": 78}
]

for student in students:
    print(student["name"], student["score"])
````

# 66. Working with JSON-Like Data

Python dictionaries and lists often represent JSON-style data.

````python
user = {
    "id": 1,
    "name": "Ali",
    "skills": ["Python", "Pandas", "SQL"],
    "address": {
        "city": "Lahore",
        "country": "Pakistan"
    }
}

print(user["name"])
print(user["skills"][0])
print(user["address"]["city"])
````
