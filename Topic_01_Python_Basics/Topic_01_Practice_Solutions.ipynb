Create a function called `calculate_average()`.

````python
def calculate_average(list):
  sum =0
  for x in list:
   sum += x
  return sum / len(list)

print(calculate_average([1,2,3,4,5]))
````

## Solution 8 — `calculate_average()`

````python
def calculate_average(values):
    if len(values) == 0:          # avoid ZeroDivisionError
        return None
    return sum(values) / len(values)

print(calculate_average([10, 20, 30, 45]))
print(calculate_average([]))
````

# 75. Practice Problem 9 — List Comprehension

Given:

```python
numbers = range(1, 21)
```

Create:

1. A list of squares.
2. A list containing only even numbers.
3. A list containing only numbers greater than 10.
4. A list containing the cubes of odd numbers.

````python
numbers = range(1, 21)
square = [ x ** 2 for x in numbers  ]
odd =[x  for x in numbers if x %2 !=0]
even =[x  for x in numbers if x %2 ==0]
grater_than_10 =[x  for x in numbers if x > 10]
cubes_of_odds = [num**3 for num in numbers if num % 2 != 0]


print(square)
print(odd)
print(even)
print(grater_than_10)
print(cubes_of_odds)
````

## Solution 9 — List comprehensions

````python
numbers = range(1, 21)
squares = [n ** 2 for n in numbers]
evens = [n for n in numbers if n % 2 == 0]
greater_than_10 = [n for n in numbers if n > 10]
odd_cubes = [n ** 3 for n in numbers if n % 2 == 1]
print(squares, evens, greater_than_10, odd_cubes, sep="\n")
````

In a real script you would use `input()`. Here we loop over some sample answers so the cell runs without waiting for typing.

# 79. Mini Project — Student Grade Analyzer

Build a small Python program using only the concepts learned in this notebook.

Create a list of student dictionaries.

Example structure:

```python
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "John", "score": 72}
]
```

Your program should:

1. Print every student's name.
2. Print every student's score.
3. Calculate the average score.
4. Find the highest score.
5. Find the lowest score.
6. Assign a grade to each student.
7. Count how many students passed.
8. Count how many students failed.
9. Print the student with the highest score.

## ✅ Solution — Mini Project — Student Grade Analyzer

````python
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "John", "score": 72},
    {"name": "Ayesha", "score": 48},
    {"name": "Usman", "score": 66},
]

def get_grade(score):
    if score >= 90: return "A"
    if score >= 80: return "B"
    if score >= 70: return "C"
    if score >= 60: return "D"
    return "F"

scores = [s["score"] for s in students]
for s in students:                                            # 1, 2, 6
    print(f"{s['name']:<7} {s['score']:>3}  grade {get_grade(s['score'])}")

print("Average :", sum(scores) / len(scores))                  # 3
print("Highest :", max(scores))                                # 4
print("Lowest  :", min(scores))                                # 5
passed = [s for s in students if s["score"] >= 50]
print("Passed  :", len(passed))                                # 7
print("Failed  :", len(students) - len(passed))                # 8
best = max(students, key=lambda s: s["score"])
print("Top student:", best["name"])                            # 9
````

# 81. Final Challenge

Without using Pandas or NumPy, create a small employee dataset using Python lists and dictionaries.

Each employee should have:

- Name
- Age
- Department
- Salary
- Experience

Then write Python code to:

1. Print all employees.
2. Find the employee with the highest salary.
3. Find the employee with the lowest salary.
4. Calculate average salary.
5. Calculate average age.
6. Find employees with salary greater than 100000.
7. Find employees with more than 5 years of experience.
8. Sort employees by salary.
9. Create a list containing only employee names.
10. Create a dictionary mapping employee names to salaries.

Try to solve the entire challenge yourself.

## ✅ Solution — Final Challenge — Employee dataset with plain Python

````python
employees = [
    {"name": "Ali",    "age": 28, "department": "IT",      "salary": 95000,  "experience": 4},
    {"name": "Sara",   "age": 35, "department": "Finance", "salary": 130000, "experience": 10},
    {"name": "John",   "age": 24, "department": "HR",      "salary": 60000,  "experience": 1},
    {"name": "Ayesha", "age": 41, "department": "IT",      "salary": 150000, "experience": 15},
    {"name": "Usman",  "age": 31, "department": "Sales",   "salary": 85000,  "experience": 6},
]

for e in employees:                                                        # 1
    print(e)


print("Highest salary:", max(employees, key=lambda e: e["salary"])["name"])   # 2
print("Lowest salary :", min(employees, key=lambda e: e["salary"])["name"])   # 3
print("Avg salary    :", sum(e["salary"] for e in employees) / len(employees))  # 4
print("Avg age       :", sum(e["age"] for e in employees) / len(employees))     # 5
print("> 100000      :", [e["name"] for e in employees if e["salary"] > 100000])  # 6
print("> 5 yrs exp   :", [e["name"] for e in employees if e["experience"] > 5])   # 7
by_salary = sorted(employees, key=lambda e: e["salary"])                   # 8
print("Sorted        :", [(e["name"], e["salary"]) for e in by_salary])
names = [e["name"] for e in employees]                                     # 9
print("Names         :", names)
salary_map = {e["name"]: e["salary"] for e in employees}                   # 10
print("Name->salary  :", salary_map)
````
